```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import Descriptors

# Core literature dataset schema
def load_and_enrich_dataset(csv_path: str) -> pd.DataFrame:
    df = pd.read_csv(csv_path)

    # Calculate RDKit molecular descriptors for drug SMILES
    def get_descriptors(smiles):
        mol = Chem.MolFromSmiles(smiles)
        if mol:
            return pd.Series([
                Descriptors.MolWt(mol),
                Descriptors.MolLogP(mol),
                Descriptors.TPSA(mol),
                Descriptors.NumHDonors(mol),
                Descriptors.NumHAcceptors(mol)
            ])
        return pd.Series([np.nan] * 5)

    desc_cols = ['drug_mw', 'drug_logp', 'drug_tpsa', 'drug_hbd', 'drug_hba']
    df[desc_cols] = df['drug_smiles'].apply(get_descriptors)

    return df.dropna(subset=['release_pct', 'time_hours', 'study_doi'])
```

```python
import sys
import subprocess

# Auto-install missing dependencies
required_packages = ['rdkit', 'xgboost', 'shap', 'scipy', 'sklearn', 'pandas', 'numpy']
for pkg in required_packages:
    try:
        __import__(pkg)
    except ImportError:
        subprocess.check_call([sys.executable, "-m", "pip", "install", pkg if pkg != 'sklearn' else 'scikit-learn'])

import pandas as pd
import numpy as np
import xgboost as xgb
from rdkit import Chem
from rdkit.Chem import Descriptors
from scipy.optimize import curve_fit, differential_evolution
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_squared_error, r2_score
import shap

# 1. Generate Synthesized Dataset Matching Literature Structure
def generate_sample_dataset(n_samples=500):
    np.random.seed(42)
    studies = [f"10.1016/j.jconrel.2024.{i:03d}" for i in range(1, 15)]
    smiles_list = [
        'CC(C)CC1C(=O)NC(C(=O)N1)Cc2ccccc2', # Mock peptide-like drug
        'CC1=CN(C(=O)NC1=O)C2CC(C(O2)CO)O',   # Zidovudine derivative
        'CC(C)C1=C(C(=C(C=C1)C(=O)O)O)O'     # Small molecule organic acid
    ]

    data = []
    for _ in range(n_samples):
        doi = np.random.choice(studies)
        smiles = np.random.choice(smiles_list)
        polymer_mw = np.random.uniform(10000, 100000)
        polymer_ratio = np.random.uniform(0.2, 0.8)
        drug_loading = np.random.uniform(2.0, 25.0)
        ph = np.random.choice([5.5, 6.8, 7.4])
        time_hours = np.random.uniform(0.5, 72.0)

        # Synthetic physics-informed kinetic relationship
        release_pct = 100 * (1 - np.exp(-0.03 * time_hours * (100000 / polymer_mw)))
        release_pct += np.random.normal(0, 2)
        release_pct = np.clip(release_pct, 0.0, 100.0)

        data.append({
            'study_doi': doi,
            'drug_smiles': smiles,
            'polymer_mw': polymer_mw,
            'polymer_ratio': polymer_ratio,
            'drug_loading_pct': drug_loading,
            'ph': ph,
            'time_hours': time_hours,
            'release_pct': release_pct
        })
    return pd.DataFrame(data)

df = generate_sample_dataset()

# 2. Extract RDKit Descriptors
def extract_rdkit_features(df_input):
    def parse_smiles(s):
        m = Chem.MolFromSmiles(s)
        return pd.Series([
            Descriptors.MolWt(m), Descriptors.MolLogP(m),
            Descriptors.TPSA(m), Descriptors.NumHDonors(m), Descriptors.NumHAcceptors(m)
        ]) if m else pd.Series([np.nan]*5)

    cols = ['drug_mw', 'drug_logp', 'drug_tpsa', 'drug_hbd', 'drug_hba']
    df_input[cols] = df_input['drug_smiles'].apply(parse_smiles)
    return df_input.dropna()

df = extract_rdkit_features(df)

# 3. Train Study-Aware XGBoost Model
features = ['drug_mw', 'drug_logp', 'drug_tpsa', 'polymer_mw', 'polymer_ratio', 'drug_loading_pct', 'ph', 'time_hours']
X = df[features]
y = df['release_pct']
groups = df['study_doi']

gkf = GroupKFold(n_splits=3)
trained_model = None

for fold, (train_idx, val_idx) in enumerate(gkf.split(X, y, groups)):
    X_tr, y_tr = X.iloc[train_idx], y.iloc[train_idx]
    X_va, y_va = X.iloc[val_idx], y.iloc[val_idx]

    model = xgb.XGBRegressor(n_estimators=100, max_depth=4, learning_rate=0.05, random_state=42)
    model.fit(X_tr, y_tr)
    trained_model = model  # Retain for downstream tasks

    preds = model.predict(X_va)
    print(f"Fold {fold+1} Study-Out RMSE: {np.sqrt(mean_squared_error(y_va, preds)):.2f}%")

# 4. Inverse Design Execution
def optimization_target(params):
    p_mw, p_ratio, d_loading = params
    target_times = [4.0, 12.0, 24.0, 48.0]
    target_profile = [15.0, 35.0, 60.0, 85.0]  # Desired zero-order/sustained profile

    predicted_profile = []
    for t in target_times:
        sample_eval = pd.DataFrame([{
            'drug_mw': 300.0, 'drug_logp': 2.1, 'drug_tpsa': 65.0,
            'polymer_mw': p_mw, 'polymer_ratio': p_ratio,
            'drug_loading_pct': d_loading, 'ph': 7.4, 'time_hours': t
        }])[features]
        predicted_profile.append(trained_model.predict(sample_eval)[0])

    return np.sqrt(np.mean((np.array(target_profile) - np.array(predicted_profile))**2))

bounds = [(10000, 100000), (0.1, 0.9), (1.0, 20.0)]
res = differential_evolution(optimization_target, bounds, seed=42)

print("\n--- Optimized Inverse Formulation Parameters ---")
print(f"Target Poly MW: {res.x[0]:.0f} Da")
print(f"Target Poly Ratio: {res.x[1]:.2f}")
print(f"Target Drug Loading: {res.x[2]:.2f} wt%")
print(f"Profile Optimization Error (RMSE): {res.fun:.2f}%")
```

```python
from scipy.optimize import curve_fit

def korsmeyer_peppas(t, k, n):
    return k * np.power(t, n)

def zero_order(t, k0):
    return k0 * t

def fit_formulation_kinetics(df_formulation: pd.DataFrame) -> dict:
    """Fits kinetic models on time-series data up to 60% cumulative release."""
    kp_sub = df_formulation[df_formulation['release_pct'] <= 60.0]

    if len(kp_sub) < 3:
        return {'k_kp': np.nan, 'n_exponent': np.nan, 'mechanism': 'Insufficient Data'}

    try:
        popt, _ = curve_fit(
            korsmeyer_peppas,
            kp_sub['time_hours'],
            kp_sub['release_pct'] / 100.0,
            bounds=(0, [10.0, 2.0]),
            maxfev=2000
        )
        k_kp, n = popt[0], popt[1]

        # Determine transport mechanism based on sphere geometry defaults
        if n <= 0.43:
            mech = 'Fickian Diffusion'
        elif 0.43 < n < 0.85:
            mech = 'Anomalous (Non-Fickian) Transport'
        else:
            mech = 'Case-II Polymer Erosion'

        return {'k_kp': k_kp, 'n_exponent': n, 'mechanism': mech}
    except Exception:
        return {'k_kp': np.nan, 'n_exponent': np.nan, 'mechanism': 'Fit Failed'}
```

```python
import xgboost as xgb
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_squared_error, r2_score

def train_study_aware_xgboost(df: pd.DataFrame):
    features = [
        'drug_mw', 'drug_logp', 'drug_tpsa', 'polymer_mw',
        'polymer_ratio', 'drug_loading_pct', 'ph', 'time_hours'
    ]
    target = 'release_pct'
    groups = df['study_doi']  # Prevents leakage across identical publication conditions

    gkf = GroupKFold(n_splits=5)
    models = []
    metrics = []

    for fold, (train_idx, val_idx) in enumerate(gkf.split(df, df[target], groups)):
        X_tr, y_tr = df.iloc[train_idx][features], df.iloc[train_idx][target]
        X_va, y_va = df.iloc[val_idx][features], df.iloc[val_idx][target]

        model = xgb.XGBRegressor(
            n_estimators=300,
            max_depth=5,
            learning_rate=0.03,
            subsample=0.8,
            colsample_bytree=0.8,
            random_state=42
        )

        model.fit(
            X_tr, y_tr,
            eval_set=[(X_va, y_va)],
            verbose=False
        )

        preds = model.predict(X_va)
        rmse = np.sqrt(mean_squared_error(y_va, preds))
        r2 = r2_score(y_va, preds)

        metrics.append((rmse, r2))
        models.append(model)
        print(f"Fold {fold+1} - Out-of-Study RMSE: {rmse:.3f}, R2: {r2:.3f}")

    return models, metrics, features
```

```python
import shap

def explain_model(model, X_train: pd.DataFrame):
    explainer = shap.TreeExplainer(model)
    shap_values = explainer(X_train)

    # Global feature importance summary plot
    shap.summary_plot(shap_values, X_train)
    return explainer, shap_values
```

```python
from scipy.optimize import differential_evolution

def run_inverse_design(
    target_times: list,
    target_release: list,
    drug_properties: dict,
    trained_model,
    feature_names: list
) -> dict:
    """
    Finds optimal formulation parameters (polymer Mw, polymer ratio, loading wt%)
    matching a specified target profile.
    """

    def objective(candidate_params):
        # Candidate layout: [polymer_mw, polymer_ratio, drug_loading_pct, ph]
        p_mw, p_ratio, d_loading, ph = candidate_params

        predicted_profile = []
        for t in target_times:
            row = pd.DataFrame([{
                'drug_mw': drug_properties['mw'],
                'drug_logp': drug_properties['logp'],
                'drug_tpsa': drug_properties['tpsa'],
                'polymer_mw': p_mw,
                'polymer_ratio': p_ratio,
                'drug_loading_pct': d_loading,
                'ph': ph,
                'time_hours': t
            }])[feature_names]

            pred = trained_model.predict(row)[0]
            predicted_profile.append(pred)

        # Root Mean Squared Error against target release profile
        loss = np.sqrt(np.mean((np.array(target_release) - np.array(predicted_profile))**2))
        return loss

    # Search boundaries: [Polymer Mw, Polymer Ratio, Loading %, pH]
    bounds = [(5000, 150000), (0.1, 1.0), (1.0, 30.0), (6.0, 7.4)]

    result = differential_evolution(objective, bounds, seed=42, maxiter=200)

    return {
        'optimal_polymer_mw': result.x[0],
        'optimal_polymer_ratio': result.x[1],
        'optimal_drug_loading_pct': result.x[2],
        'optimal_ph': result.x[3],
        'profile_rmse': result.fun
    }
```

```python
# Estimate prediction uncertainty using standard deviation across an ensemble of models
def evaluate_candidate_confidence(ensemble_models, candidate_features):
    predictions = [model.predict(candidate_features)[0] for model in ensemble_models]
    mean_pred = np.mean(predictions)
    uncertainty_std = np.std(predictions)  # High std indicates high model uncertainty / OOD
    return mean_pred, uncertainty_std
```

```python
def update_and_retrain_pipeline(existing_df, lab_results_df, model):
    """Appends newly synthesized lab data to literature data and retrains."""
    updated_df = pd.concat([existing_df, lab_results_df], ignore_index=True)

    # Retrain on combined dataset (weighting high-quality lab data if necessary)
    X_updated = updated_df[features]
    y_updated = updated_df['release_pct']

    model.fit(X_updated, y_updated)
    return model, updated_df
```

```python
import numpy as np
from pymoo.core.problem import ElementwiseProblem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize
from pymoo.operators.sampling.rnd import FloatRandomSampling
from pymoo.operators.crossover.sbx import SBX
from pymoo.operators.mutation.pm import PM

class FormulationOptimizationProblem(ElementwiseProblem):
    def __init__(self, trained_model, target_times, target_release, drug_props, features):
        super().__init__(
            n_var=3,          # [Polymer Mw, Polymer Ratio, Drug Loading %]
            n_obj=2,          # [Min Profile Error, Max Drug Loading]
            n_ieq_constr=1,   # Constraint: Hansen Distance Ra <= 5.0
            xl=np.array([10000.0, 0.1, 1.0]),  # Lower bounds
            xu=np.array([120000.0, 0.9, 30.0]) # Upper bounds
        )
        self.model = trained_model
        self.target_times = target_times
        self.target_release = target_release
        self.drug_props = drug_props
        self.features = features

    def _evaluate(self, x, out, *args, **kwargs):
        p_mw, p_ratio, d_loading = x

        # Predict profile across target time points
        preds = []
        for t in self.target_times:
            row = pd.DataFrame([{
                'drug_mw': self.drug_props['mw'],
                'drug_logp': self.drug_props['logp'],
                'drug_tpsa': self.drug_props['tpsa'],
                'polymer_mw': p_mw,
                'polymer_ratio': p_ratio,
                'drug_loading_pct': d_loading,
                'ph': 7.4,
                'time_hours': t
            }])[self.features]
            preds.append(self.model.predict(row)[0])

        # Objective 1: Minimize Profile Trajectory RMSE
        f1 = np.sqrt(np.mean((np.array(self.target_release) - np.array(preds))**2))

        # Objective 2: Maximize Drug Loading (Minimize negative loading)
        f2 = -d_loading

        # Constraint: Polymer Ratio must remain <= 0.8 for processability
        g1 = p_ratio - 0.8  # Constraint <= 0

        out["F"] = [f1, f2]
        out["G"] = [g1]

# Instantiate and run NSGA-II
problem = FormulationOptimizationProblem(
    trained_model=trained_model,
    target_times=[4.0, 12.0, 24.0, 48.0],
    target_release=[15.0, 35.0, 60.0, 85.0],
    drug_props={'mw': 350.0, 'logp': 2.5, 'tpsa': 50.0},
    features=features
)

algorithm = NSGA2(
    pop_size=40,
    sampling=FloatRandomSampling(),
    crossover=SBX(prob=0.9, eta=15),
    mutation=PM(eta=20),
    eliminate_duplicates=True
)

res = minimize(problem, algorithm, ('n_gen', 50), seed=42, verbose=False)

# Display Pareto Optimal Frontier Solutions
print("--- Pareto Optimal Solutions ---")
for i, (X_val, F_val) in enumerate(zip(res.X, res.F)):
    print(f"Candidate {i+1}: Poly Mw = {X_val[0]:.0f} Da | Poly Ratio = {X_val[1]:.2f} | Loading = {X_val[2]:.1f}% | RMSE = {F_val[0]:.2f}%")
```

```python
import json
from datetime import datetime

def generate_synthesis_protocol(
    candidate_id: int,
    candidate_params: dict,
    drug_name: str = "Model Small Molecule",
    batch_mass_mg: float = 100.0,
    preparation_method: str = "Nanoprecipitation"
) -> str:
    """
    Generates a structured, reproducible wet-lab protocol report for candidate polymer-drug formulations.
    """
    poly_mw = candidate_params.get("polymer_mw", 45000)
    poly_ratio = candidate_params.get("polymer_ratio", 0.5)
    drug_loading_pct = candidate_params.get("drug_loading_pct", 10.0)
    rmse = candidate_params.get("rmse", 1.5)

    # Reagent Stoichiometry Calculations
    drug_mass_mg = batch_mass_mg * (drug_loading_pct / 100.0)
    total_polymer_mass_mg = batch_mass_mg - drug_mass_mg
    polymer_a_mass_mg = total_polymer_mass_mg * poly_ratio
    polymer_b_mass_mg = total_polymer_mass_mg * (1.0 - poly_ratio)

    # Solvent & Aqueous Phase Estimates (Standard 10 mg/mL organic phase rules)
    organic_solvent_vol_ml = total_polymer_mass_mg / 10.0  # Acetone or DMSO
    aqueous_phase_vol_ml = organic_solvent_vol_ml * 10.0    # 10:1 aqueous to organic ratio (0.5% w/v PVA)

    protocol_report = f"""
================================================================================
                     WET-LAB SYNTHESIS PROTOCOL REPORT
================================================================================
Generated Date: {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}
Candidate ID  : #{candidate_id}
Drug Model    : {drug_name}
Target Mass   : {batch_mass_mg:.1f} mg
Prep Method   : {preparation_method}
Model Confidence Error (RMSE): {rmse:.2f}%
--------------------------------------------------------------------------------

**REAGENT STOICHIOMETRY**
* Target Drug ({drug_name}): {drug_mass_mg:.2f} mg
* Primary Polymer (Mw ~ {poly_mw:.0f} Da): {polymer_a_mass_mg:.2f} mg
* Co-Polymer / Blend Component: {polymer_b_mass_mg:.2f} mg
* Organic Phase Solvent (Acetone or Acetone:DMSO 9:1): {organic_solvent_vol_ml:.2f} mL
* Aqueous Phase (0.5% w/v PVA in Milli-Q Water): {aqueous_phase_vol_ml:.2f} mL

**PREPARATION STEPS**
1. Dissolve {drug_mass_mg:.2f} mg {drug_name}, {polymer_a_mass_mg:.2f} mg primary polymer, and {polymer_b_mass_mg:.2f} mg co-polymer in {organic_solvent_vol_ml:.2f} mL organic solvent. Vortex for 2 minutes until clear.
2. Prepare {aqueous_phase_vol_ml:.2f} mL of 0.5% w/v Poly(vinyl alcohol) (PVA) solution under magnetic stirring at 500 RPM.
3. Inject the organic phase dropwise (1.0 mL/min via syringe pump) into the stirring aqueous phase.
4. Stir open-container for 4 hours at ambient temperature to allow complete solvent evaporation.
5. Collect nanoparticles via centrifugation at 15,000 RPM (4°C) for 20 minutes.
6. Wash pellet twice with Milli-Q water and lyophilize for 24 hours.

**QUALITY CONTROL CHECKPOINTS**
* Dynamic Light Scattering (DLS): Measure size distribution (PDI < 0.2) and Zeta Potential.
* HPLC Quantitation: Measure supernatant to verify Encapsulation Efficiency (EE%).
* In Vitro Dissolution: Sample release at t = [4h, 12h, 24h, 48h] in PBS pH 7.4 at 37°C.
================================================================================
"""
    return protocol_report


# Run Example Generation
selected_candidate = {
    "polymer_mw": 42500,
    "polymer_ratio": 0.65,
    "drug_loading_pct": 12.5,
    "rmse": 1.42
}

report_output = generate_synthesis_protocol(
    candidate_id=101,
    candidate_params=selected_candidate,
    drug_name="Paclitaxel Analog",
    batch_mass_mg=50.0
)

print(report_output)
```

```python
import os
from datetime import datetime
from reportlab.lib.pagesizes import letter
from reportlab.lib import colors
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle

def export_protocol_pdf(
    candidate_id: int,
    candidate_params: dict,
    output_filename: str = "lab_protocol.pdf",
    drug_name: str = "Paclitaxel Analog",
    batch_mass_mg: float = 100.0
):
    """
    Exports inverse-designed formulation details into a publication-quality PDF protocol.
    """
    poly_mw = candidate_params.get("polymer_mw", 45000)
    poly_ratio = candidate_params.get("polymer_ratio", 0.5)
    drug_loading_pct = candidate_params.get("drug_loading_pct", 10.0)
    rmse = candidate_params.get("rmse", 1.5)

    # Calculations
    drug_mass = batch_mass_mg * (drug_loading_pct / 100.0)
    total_poly = batch_mass_mg - drug_mass
    poly_a = total_poly * poly_ratio
    poly_b = total_poly * (1.0 - poly_ratio)
    org_solvent_ml = total_poly / 10.0
    aq_vol_ml = org_solvent_ml * 10.0

    # Document setup
    doc = SimpleDocTemplate(output_filename, pagesize=letter, rightMargin=36, leftMargin=36, topMargin=36, bottomMargin=36)
    styles = getSampleStyleSheet()

    title_style = ParagraphStyle('TitleStyle', parent=styles['Heading1'], fontSize=18, leading=22, textColor=colors.HexColor('#1A365D'))
    h2_style = ParagraphStyle('H2Style', parent=styles['Heading2'], fontSize=12, leading=16, textColor=colors.HexColor('#2B6CB0'))
    body_style = ParagraphStyle('BodyStyle', parent=styles['Normal'], fontSize=10, leading=14)

    story = []

    # Title Header
    story.append(Paragraph("AI-Guided Polymeric Formulation Protocol", title_style))
    story.append(Paragraph(f"<b>Generated:</b> {datetime.now().strftime('%Y-%m-%d %H:%M:%S')} | <b>Candidate ID:</b> #{candidate_id}", body_style))
    story.append(Spacer(1, 12))

    # Parameter Table
    table_data = [
        ["Parameter", "Target Value", "Unit"],
        ["Target Drug", drug_name, "-"],
        ["Total Batch Mass", f"{batch_mass_mg:.1f}", "mg"],
        ["Drug Mass", f"{drug_mass:.2f}", "mg"],
        ["Primary Polymer (Mw ~ " + f"{poly_mw:.0f}" + " Da)", f"{poly_a:.2f}", "mg"],
        ["Secondary Polymer / Blend", f"{poly_b:.2f}", "mg"],
        ["Organic Solvent Volume", f"{org_solvent_ml:.2f}", "mL"],
        ["Aqueous Phase (0.5% PVA)", f"{aq_vol_ml:.2f}", "mL"],
        ["Model Fit Error (RMSE)", f"{rmse:.2f}", "%"]
    ]

    t = Table(table_data, colWidths=[220, 150, 100])
    t.setStyle(TableStyle([
        ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor('#2B6CB0')),
        ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
        ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
        ('BOTTOMPADDING', (0, 0), (-1, 0), 6),
        ('GRID', (0, 0), (-1, -1), 0.5, colors.HexColor('#CBD5E0')),
        ('ROWBACKGROUNDS', (0, 1), (-1, -1), [colors.white, colors.HexColor('#F7FAFC')])
    ]))
    story.append(t)
    story.append(Spacer(1, 16))

    # Protocol Instructions
    story.append(Paragraph("Standard Preparation Steps (Nanoprecipitation)", h2_style))
    steps = [
        f"1. Dissolve {drug_mass:.2f} mg {drug_name}, {poly_a:.2f} mg primary polymer, and {poly_b:.2f} mg secondary polymer in {org_solvent_ml:.2f} mL organic solvent (vortex for 2 minutes).",
        f"2. Prepare {aq_vol_ml:.2f} mL of 0.5% w/v PVA solution under magnetic stirring (500 RPM).",
        "3. Inject organic phase dropwise (1.0 mL/min) into stirring aqueous phase.",
        "4. Stir open-container for 4 hours at room temperature for solvent evaporation.",
        "5. Centrifuge at 15,000 RPM (4°C) for 20 minutes, wash twice with Milli-Q water, and lyophilize for 24 hours."
    ]
    for step in steps:
        story.append(Paragraph(step, body_style))
        story.append(Spacer(1, 4))

    doc.build(story)
    return output_filename

# Run export test
pdf_path = export_protocol_pdf(
    candidate_id=101,
    candidate_params={"polymer_mw": 42500, "polymer_ratio": 0.65, "drug_loading_pct": 12.5, "rmse": 1.42}
)
print(f"Successfully generated PDF report: {pdf_path}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

class PINNReleaseModel(nn.Module):
    def __init__(self, input_dim: int):
        super(PINNReleaseModel, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.Tanh(),  # Smooth activation required for continuous autograd derivatives
            nn.Linear(64, 64),
            nn.Tanh(),
            nn.Linear(64, 1),
            nn.Sigmoid() # Scale initial output to [0, 1] range (multiplied by 100 for %)
        )

    def forward(self, x):
        return self.net(x) * 100.0


def pinn_loss_function(
    model: nn.Module,
    X_data: torch.Tensor,
    y_data: torch.Tensor,
    k_kp: torch.Tensor,
    n_exp: torch.Tensor,
    time_col_idx: int = 0,
    lambda_b: float = 1.0,
    lambda_m: float = 10.0,
    lambda_k: float = 0.1
):
    """
    Computes data loss combined with physics-informed penalty terms.
    X_data must have requires_grad=True for the time feature column.
    """
    # 1. Data Loss (MSE)
    y_pred = model(X_data)
    loss_data = nn.MSELoss()(y_pred, y_data)

    # 2. Compute Derivative dQ/dt via Automatic Differentiation
    t_tensor = X_data[:, time_col_idx:time_col_idx+1]
    dQ_dt = torch.autograd.grad(
        outputs=y_pred,
        inputs=X_data,
        grad_outputs=torch.ones_like(y_pred),
        create_graph=True,
        retain_graph=True
    )[0][:, time_col_idx:time_col_idx+1]

    # 3. Monotonicity Penalty: Penalize negative release rates (dQ/dt < 0)
    loss_mono = torch.mean(torch.relu(-dQ_dt)**2)

    # 4. Boundary Penalty: Q(t=0) should be 0
    X_zero = X_data.clone()
    X_zero[:, time_col_idx] = 0.0
    loss_boundary = torch.mean(model(X_zero)**2)

    # 5. Kinetic Differential Penalty (Korsmeyer-Peppas residual for Q <= 60%)
    # dQ/dt - n * K * t^(n-1) = 0
    t_safe = torch.clamp(t_tensor, min=1e-3)
    target_dq_dt = n_exp * k_kp * torch.pow(t_safe, n_exp - 1.0) * 100.0

    mask = (y_pred <= 60.0).float()
    loss_kinetics = torch.mean(mask * (dQ_dt - target_dq_dt)**2)

    # Total Multi-Objective Loss
    total_loss = loss_data + (lambda_b * loss_boundary) + (lambda_m * loss_mono) + (lambda_k * loss_kinetics)
    return total_loss


# --- Example Training Iteration ---
# Features: [time_hours, drug_mw, drug_logp, polymer_mw, drug_loading]
X_sample = torch.tensor([
    [4.0, 350.0, 2.1, 45000.0, 10.0],
    [12.0, 350.0, 2.1, 45000.0, 10.0],
    [24.0, 350.0, 2.1, 45000.0, 10.0]
], dtype=torch.float32, requires_grad=True)

y_sample = torch.tensor([[18.0], [42.0], [68.0]], dtype=torch.float32)
k_kp = torch.tensor([[0.05]], dtype=torch.float32)
n_exp = torch.tensor([[0.6]], dtype=torch.float32)

pinn = PINNReleaseModel(input_dim=5)
optimizer = optim.Adam(pinn.parameters(), lr=0.01)

for epoch in range(100):
    optimizer.zero_grad()
    loss = pinn_loss_function(pinn, X_sample, y_sample, k_kp, n_exp)
    loss.backward()
    optimizer.step()

print(f"PINN Optimized Final Loss: {loss.item():.4f}")
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold, GroupKFold
from sklearn.ensemble import RandomForestRegressor
import xgboost as xgb
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

def run_ablation_study(df: pd.DataFrame, features: list, target: str = 'release_pct', group_col: str = 'study_doi'):
    X = df[features]
    y = df[target]
    groups = df[group_col]

    results = []

    # Experiment 1: Naive Random Forest (Standard KFold - Demonstrates Data Leakage)
    kf = KFold(n_splits=5, shuffle=True, random_state=42)
    rmse_naive_rf, r2_naive_rf = [], []
    for train_idx, val_idx in kf.split(X):
        rf = RandomForestRegressor(n_estimators=100, random_state=42)
        rf.fit(X.iloc[train_idx], y.iloc[train_idx])
        preds = rf.predict(X.iloc[val_idx])
        rmse_naive_rf.append(np.sqrt(mean_squared_error(y.iloc[val_idx], preds)))
        r2_naive_rf.append(r2_score(y.iloc[val_idx], preds))

    results.append({
        'Model Pipeline': 'Naive RF (Random K-Fold)',
        'Validation Strategy': 'Random Split (Leaky)',
        'RMSE (%)': np.mean(rmse_naive_rf),
        'R2 Score': np.mean(r2_naive_rf),
        'Data Leakage Risk': 'High'
    })

    # Experiment 2: Naive XGBoost (Standard KFold)
    rmse_naive_xgb, r2_naive_xgb = [], []
    for train_idx, val_idx in kf.split(X):
        model = xgb.XGBRegressor(n_estimators=100, max_depth=4, learning_rate=0.05, random_state=42)
        model.fit(X.iloc[train_idx], y.iloc[train_idx])
        preds = model.predict(X.iloc[val_idx])
        rmse_naive_xgb.append(np.sqrt(mean_squared_error(y.iloc[val_idx], preds)))
        r2_naive_xgb.append(r2_score(y.iloc[val_idx], preds))

    results.append({
        'Model Pipeline': 'Naive XGBoost (Random K-Fold)',
        'Validation Strategy': 'Random Split (Leaky)',
        'RMSE (%)': np.mean(rmse_naive_xgb),
        'R2 Score': np.mean(r2_naive_xgb),
        'Data Leakage Risk': 'High'
    })

    # Experiment 3: Study-Aware XGBoost (GroupKFold)
    gkf = GroupKFold(n_splits=5)
    rmse_group_xgb, r2_group_xgb = [], []
    for train_idx, val_idx in gkf.split(X, y, groups):
        model = xgb.XGBRegressor(n_estimators=100, max_depth=4, learning_rate=0.05, random_state=42)
        model.fit(X.iloc[train_idx], y.iloc[train_idx])
        preds = model.predict(X.iloc[val_idx])
        rmse_group_xgb.append(np.sqrt(mean_squared_error(y.iloc[val_idx], preds)))
        r2_group_xgb.append(r2_score(y.iloc[val_idx], preds))

    results.append({
        'Model Pipeline': 'Study-Aware XGBoost (GroupKFold)',
        'Validation Strategy': 'Leave-Publication-Out',
        'RMSE (%)': np.mean(rmse_group_xgb),
        'R2 Score': np.mean(r2_group_xgb),
        'Data Leakage Risk': 'None'
    })

    ablation_df = pd.DataFrame(results)
    return ablation_df

# Example execution with preloaded dataset
ablation_results = run_ablation_study(df, features)
print(ablation_results.to_string(index=False))
```

```python
import numpy as np
import pandas as pd

def evaluate_experimental_validation(
    time_points: list,
    q_experimental: list,
    q_predicted: list
) -> dict:
    """
    Evaluates AI predictions against newly measured wet-lab dissolution experiments.
    Calculates RMSE, MAE, Max Error, and FDA f2 similarity factor.
    """
    R_t = np.array(q_experimental, dtype=np.float64)  # Reference (Experimental)
    T_t = np.array(q_predicted, dtype=np.float64)     # Test (Predicted)
    n = len(time_points)

    # 1. Standard Error Metrics
    rmse = np.sqrt(np.mean((R_t - T_t)**2))
    mae = np.mean(np.abs(R_t - T_t))
    max_err = np.max(np.abs(R_t - T_t))

    # 2. FDA f2 Similarity Factor Calculation
    mean_squared_diff = np.mean((R_t - T_t)**2)
    f2 = 50.0 * np.log10( (1.0 + mean_squared_diff)**(-0.5) * 100.0 )

    # FDA Guideline Check (f2 >= 50 indicates profile similarity)
    is_equivalent = f2 >= 50.0

    validation_summary = {
        'n_timepoints': n,
        'rmse_pct': round(rmse, 2),
        'mae_pct': round(mae, 2),
        'max_error_pct': round(max_err, 2),
        'f2_similarity_factor': round(f2, 2),
        'fda_bioequivalent': is_equivalent
    }

    # Generate comparison dataframe
    comp_df = pd.DataFrame({
        'Time (h)': time_points,
        'Experimental Release (%)': R_t,
        'Predicted Release (%)': T_t,
        'Absolute Error (%)': np.abs(R_t - T_t)
    })

    return validation_summary, comp_df

# Example wet-lab experiment vs. AI model prediction
time_points = [2.0, 4.0, 8.0, 12.0, 24.0, 48.0]
exp_lab_data = [8.2, 16.5, 31.0, 45.2, 68.0, 89.1]  # Measured in lab via HPLC/UV-Vis
ai_predictions = [7.5, 15.8, 32.1, 44.0, 66.5, 87.0] # Output from inverse engine candidate

summary, detailed_table = evaluate_experimental_validation(time_points, exp_lab_data, ai_predictions)

print("--- Wet-Lab Validation Metrics ---")
for metric, value in summary.items():
    print(f"{metric}: {value}")

print("\n--- Point-by-Point Residual Matrix ---")
print(detailed_table.to_string(index=False))
```

```python
import numpy as np

def calculate_hansen_distance(drug_hsp: tuple, polymer_hsp: tuple) -> float:
    """
    Calculates Hansen interaction distance Ra.
    A value of Ra <= 5.0 MPa^(0.5) generally indicates high miscibility/compatibility.
    """
    d_d1, d_p1, d_h1 = drug_hsp
    d_d2, d_p2, d_h2 = polymer_hsp

    R_a = np.sqrt(4 * (d_d1 - d_d2)**2 + (d_p1 - d_p2)**2 + (d_h1 - d_h2)**2)
    return R_a

# Example HSP Values [Dispersion, Polar, Hydrogen bonding] in MPa^0.5
# Paclitaxel vs PLGA
drug_hsp = (18.5, 5.2, 8.1)
polymer_hsp = (18.6, 4.3, 8.0)

R_a = calculate_hansen_distance(drug_hsp, polymer_hsp)
print(f"Hansen Distance (Ra): {R_a:.2f} MPa^0.5")  # Ra < 5 -> High compatibility
```

```python
from opentrons import protocol_api

metadata = {
    'protocolName': 'AI-Guided Polymer Nanoparticle Synthesis',
    'author': 'PolyRelease AI Pipeline',
    'description': 'Automated formulation of candidate polymer-drug systems',
    'apiLevel': '2.14'
}

def run(protocol: protocol_api.ProtocolContext):
    # Labware setup
    tiprack = protocol.load_labware('opentrons_96_tiprack_300ul', '1')
    reservoir = protocol.load_labware('nest_12_reservoir_15ml', '2')
    plate = protocol.load_labware('nest_96_wellplate_200ul_flat', '3')
    p300 = protocol.load_instrument('p300_single_gen2', 'left', tip_racks=[tiprack])

    # Reagent Locations in Reservoir
    solvent_stock = reservoir['A1']   # Polymer + Drug in Acetone
    pva_stock = reservoir['A2']       # 0.5% w/v PVA Aqueous Solution

    # Candidate Parameters (Dispense volumes in uL derived from inverse design)
    candidates = [
        {'well': 'A1', 'organic_vol': 20.0, 'aqueous_vol': 180.0},
        {'well': 'A2', 'organic_vol': 15.0, 'aqueous_vol': 185.0},
        {'well': 'A3', 'organic_vol': 25.0, 'aqueous_vol': 175.0},
    ]

    # Nanoprecipitation Protocol Execution
    for c in candidates:
        target_well = plate[c['well']]

        # 1. Transfer Aqueous Phase
        p300.pick_up_tip()
        p300.transfer(c['aqueous_vol'], pva_stock, target_well, new_tip='never')

        # 2. Inject Organic Phase Dropwise with Mixing
        p300.transfer(
            c['organic_vol'],
            solvent_stock,
            target_well.bottom(2),
            mix_after=(3, 50),
            new_tip='never'
        )
        p300.drop_tip()

    protocol.comment("Synthesis protocol complete. Proceed to solvent evaporation.")
```

```python
# ============================================================
# CODE 1 — LITERATURE ACQUISITION
# REAL LITERATURE ONLY — NO SYNTHETIC DATA
# ============================================================

import os
import re
import time
import requests
import pandas as pd
from urllib.parse import quote

# ============================================================
# PROJECT DIRECTORIES
# ============================================================

PROJECT_ROOT = "/content/AI_Polymer_Drug_Release"

DIRS = [
    "data/raw",
    "data/raw/metadata",
    "data/raw/pmc_xml",
    "data/raw/tables",
    "data/candidates",
    "data/validated",
    "data/final",
    "data/analysis",
    "results",
    "logs"
]

for d in DIRS:
    os.makedirs(
        os.path.join(PROJECT_ROOT, d),
        exist_ok=True
    )

print("=" * 70)
print("AI POLYMER DRUG RELEASE")
print("CODE 1 — LITERATURE ACQUISITION")
print("=" * 70)

print("\nProject root:")
print(PROJECT_ROOT)


# ============================================================
# SEARCH CONFIGURATION
# ============================================================

SEARCH_QUERIES = [

    '"drug release" AND polymer AND nanoparticle',

    '"drug release" AND polymeric nanoparticle',

    '"drug release profile" AND polymer',

    '"controlled drug release" AND polymer',

    '"sustained drug release" AND polymer',

    '"cumulative drug release" AND polymer',

    '"in vitro drug release" AND polymer',

    '"drug delivery" AND polymer AND release',

    '"polymeric drug delivery" AND release',

    '"nanocomposite" AND "drug release"',

    '"polymer matrix" AND "drug release"',

    '"polymeric nanoparticles" AND "release kinetics"',

    '"PLGA" AND "drug release"',

    '"PLA" AND "drug release"',

    '"PCL" AND "drug release"',

    '"PEG" AND "drug release"',

    '"chitosan" AND "drug release"'
]


# ============================================================
# EUROPE PMC SEARCH
# ============================================================

BASE_URL = (
    "https://www.ebi.ac.uk/europepmc/webservices/rest/"
)

all_records = []

MAX_RESULTS_PER_QUERY = 200


for qnum, query in enumerate(
    SEARCH_QUERIES,
    start=1
):

    print("\n" + "-" * 70)

    print(
        f"Query {qnum}/{len(SEARCH_QUERIES)}"
    )

    print(query)

    params = {
        "query": query,
        "format": "json",
        "pageSize": MAX_RESULTS_PER_QUERY,
        "resultType": "core"
    }

    try:

        response = requests.get(
            BASE_URL + "search",
            params=params,
            timeout=60
        )

        response.raise_for_status()

        data = response.json()

        results = data.get(
            "resultList",
            {}
        ).get(
            "result",
            []
        )

        print(
            "Records returned:",
            len(results)
        )

        for r in results:

            all_records.append({
                "query_number": qnum,
                "search_query": query,
                "pmid": r.get("pmid"),
                "pmcid": r.get("pmcid"),
                "doi": r.get("doi"),
                "title": r.get("title"),
                "journal": r.get("journalTitle"),
                "year": r.get("pubYear"),
                "abstract": r.get("abstractText"),
                "author_string": r.get("authorString"),
                "is_open_access":
                    r.get("isOpenAccess"),
                "source":
                    r.get("source")
            })

    except Exception as e:

        print(
            "SEARCH ERROR:",
            repr(e)
        )

    time.sleep(0.5)


# ============================================================
# CREATE METADATA DATASET
# ============================================================

if not all_records:

    raise RuntimeError(
        "No literature records were retrieved."
    )

metadata = pd.DataFrame(
    all_records
)


# ============================================================
# NORMALIZE IDENTIFIERS
# ============================================================

for col in [
    "pmid",
    "pmcid",
    "doi",
    "title"
]:

    if col in metadata.columns:

        metadata[col] = (
            metadata[col]
            .astype("string")
            .str.strip()
        )


# ============================================================
# REMOVE DUPLICATES
# ============================================================

before = len(metadata)

metadata["dedup_key"] = (
    metadata["pmcid"]
    .fillna("")
)

metadata.loc[
    metadata["dedup_key"] == "",
    "dedup_key"
] = (
    metadata["doi"]
    .fillna("")
)

metadata.loc[
    metadata["dedup_key"] == "",
    "dedup_key"
] = (
    metadata["pmid"]
    .fillna("")
)

metadata = metadata[
    metadata["dedup_key"] != ""
].drop_duplicates(
    subset=["dedup_key"]
).reset_index(
    drop=True
)


# ============================================================
# PRIORITIZE PMC ARTICLES
# ============================================================

metadata["has_pmc"] = (
    metadata["pmcid"]
    .notna()
    &
    (
        metadata["pmcid"]
        .astype(str)
        .str.len() > 0
    )
)

metadata = metadata.sort_values(
    by=[
        "has_pmc",
        "year"
    ],
    ascending=[
        False,
        False
    ]
).reset_index(
    drop=True
)


# ============================================================
# SAVE
# ============================================================

metadata_file = os.path.join(
    PROJECT_ROOT,
    "data",
    "raw",
    "metadata",
    "01_literature_metadata.csv"
)

metadata.to_csv(
    metadata_file,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CODE 1 COMPLETE")
print("=" * 70)

print(
    "Raw search records:",
    before
)

print(
    "Unique literature records:",
    len(metadata)
)

print(
    "Records with PMC:",
    metadata["has_pmc"].sum()
)

print(
    "Records without PMC:",
    (
        ~metadata["has_pmc"]
    ).sum()
)

print(
    "Unique PMCID:",
    metadata["pmcid"].nunique()
)

print(
    "Unique DOI:",
    metadata["doi"].dropna().nunique()
)

print(
    "\nSaved:"
)

print(metadata_file)

print("\nIMPORTANT:")
print(
    "No release observations have been created."
)

print(
    "No values have been interpolated."
)

print(
    "No synthetic observations have been generated."
)
```

```python
# ============================================================
# CODE 2 — PMC FULL-TEXT ACQUISITION
# REAL LITERATURE ONLY
# ============================================================

import os
import time
import requests
import pandas as pd
from concurrent.futures import ThreadPoolExecutor, as_completed

PROJECT_ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    PROJECT_ROOT,
    "data",
    "raw",
    "metadata",
    "01_literature_metadata.csv"
)

XML_DIR = os.path.join(
    PROJECT_ROOT,
    "data",
    "raw",
    "pmc_xml"
)

LOG_DIR = os.path.join(
    PROJECT_ROOT,
    "logs"
)

os.makedirs(XML_DIR, exist_ok=True)
os.makedirs(LOG_DIR, exist_ok=True)

print("=" * 70)
print("CODE 2 — PMC FULL-TEXT ACQUISITION")
print("=" * 70)

# ------------------------------------------------------------
# LOAD METADATA
# ------------------------------------------------------------

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"Metadata file not found:\n{INPUT_FILE}"
    )

metadata = pd.read_csv(INPUT_FILE)

print(
    "Metadata records:",
    len(metadata)
)

# ------------------------------------------------------------
# SELECT PMC ARTICLES
# ------------------------------------------------------------

metadata["pmcid"] = (
    metadata["pmcid"]
    .astype("string")
    .str.strip()
)

pmc = metadata[
    metadata["pmcid"].notna()
    &
    (metadata["pmcid"] != "")
].copy()

pmc = pmc.drop_duplicates(
    subset=["pmcid"]
).reset_index(drop=True)

print(
    "Unique PMC articles:",
    len(pmc)
)

# ------------------------------------------------------------
# DOWNLOAD FUNCTION
# ------------------------------------------------------------

BASE_URL = (
    "https://www.ebi.ac.uk/europepmc/webservices/rest/"
)

session = requests.Session()

HEADERS = {
    "User-Agent":
    "AI-Polymer-Drug-Release-Literature-Research/1.0"
}


def download_article(pmcid):

    pmcid = str(pmcid).strip()

    output_file = os.path.join(
        XML_DIR,
        f"{pmcid}.xml"
    )

    # Already downloaded
    if os.path.exists(output_file):

        try:

            if os.path.getsize(output_file) > 1000:

                return {
                    "pmcid": pmcid,
                    "status": "ALREADY_EXISTS",
                    "file": output_file,
                    "size_bytes":
                        os.path.getsize(output_file)
                }

        except Exception:
            pass

    url = (
        BASE_URL
        + f"{pmcid}/fullTextXML"
    )

    for attempt in range(1, 4):

        try:

            r = session.get(
                url,
                headers=HEADERS,
                timeout=60
            )

            if r.status_code == 200:

                content = r.content

                # Basic XML validity check
                if (
                    len(content) > 1000
                    and b"<article" in content
                ):

                    with open(
                        output_file,
                        "wb"
                    ) as f:

                        f.write(content)

                    return {
                        "pmcid": pmcid,
                        "status": "DOWNLOADED",
                        "file": output_file,
                        "size_bytes":
                            len(content)
                    }

                return {
                    "pmcid": pmcid,
                    "status": "INVALID_XML",
                    "file": "",
                    "size_bytes": len(content)
                }

            elif r.status_code == 404:

                return {
                    "pmcid": pmcid,
                    "status": "NOT_FOUND",
                    "file": "",
                    "size_bytes": 0
                }

            else:

                if attempt < 3:
                    time.sleep(
                        2 ** attempt
                    )

        except Exception as e:

            if attempt == 3:

                return {
                    "pmcid": pmcid,
                    "status": "ERROR",
                    "file": "",
                    "size_bytes": 0,
                    "error": str(e)
                }

            time.sleep(
                2 ** attempt
            )

    return {
        "pmcid": pmcid,
        "status": "FAILED",
        "file": "",
        "size_bytes": 0
    }


# ------------------------------------------------------------
# DOWNLOAD ARTICLES
# ------------------------------------------------------------

results = []

MAX_WORKERS = 4

print(
    f"\nDownloading {len(pmc)} PMC articles..."
)

with ThreadPoolExecutor(
    max_workers=MAX_WORKERS
) as executor:

    futures = {
        executor.submit(
            download_article,
            pmcid
        ): pmcid
        for pmcid in pmc["pmcid"]
    }

    completed = 0

    for future in as_completed(futures):

        completed += 1

        try:
            result = future.result()

        except Exception as e:

            result = {
                "pmcid":
                    futures[future],
                "status": "ERROR",
                "file": "",
                "size_bytes": 0,
                "error": str(e)
            }

        results.append(result)

        if (
            completed % 50 == 0
            or completed == len(pmc)
        ):

            print(
                f"Progress: "
                f"{completed}/{len(pmc)}"
            )


# ------------------------------------------------------------
# SAVE DOWNLOAD LOG
# ------------------------------------------------------------

download_log = pd.DataFrame(
    results
)

DOWNLOAD_LOG_FILE = os.path.join(
    LOG_DIR,
    "02_pmc_download_log.csv"
)

download_log.to_csv(
    DOWNLOAD_LOG_FILE,
    index=False
)


# ------------------------------------------------------------
# MERGE METADATA + DOWNLOAD STATUS
# ------------------------------------------------------------

pmc_status = pmc.merge(
    download_log,
    on="pmcid",
    how="left"
)

STATUS_FILE = os.path.join(
    LOG_DIR,
    "02_pmc_acquisition_status.csv"
)

pmc_status.to_csv(
    STATUS_FILE,
    index=False
)


# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CODE 2 COMPLETE")
print("=" * 70)

print(
    "\nDownload status:"
)

print(
    download_log["status"]
    .value_counts()
    .to_string()
)

successful = (
    download_log["status"]
    .isin([
        "DOWNLOADED",
        "ALREADY_EXISTS"
    ])
    .sum()
)

print(
    "\nUsable full-text XML:",
    successful
)

print(
    "Total PMC candidates:",
    len(pmc)
)

print(
    "\nDownload log:"
)

print(DOWNLOAD_LOG_FILE)

print(
    "\nAcquisition status:"
)

print(STATUS_FILE)

print("\nIMPORTANT:")
print(
    "These XML files are source documents only."
)

print(
    "No release observations have been extracted."
)

print(
    "No values have been generated or interpolated."
)
```

```python
# ============================================================
# CODE 3 — XML TABLE EXTRACTION
# REAL LITERATURE ONLY
# ============================================================

import os
import re
import glob
import pandas as pd
import xml.etree.ElementTree as ET

PROJECT_ROOT = "/content/AI_Polymer_Drug_Release"

XML_DIR = os.path.join(
    PROJECT_ROOT,
    "data",
    "raw",
    "pmc_xml"
)

OUTPUT_DIR = os.path.join(
    PROJECT_ROOT,
    "data",
    "raw",
    "tables"
)

LOG_DIR = os.path.join(
    PROJECT_ROOT,
    "logs"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(LOG_DIR, exist_ok=True)

print("=" * 70)
print("CODE 3 — XML TABLE EXTRACTION")
print("=" * 70)


# ============================================================
# FIND XML FILES
# ============================================================

xml_files = sorted(
    glob.glob(
        os.path.join(
            XML_DIR,
            "*.xml"
        )
    )
)

print(
    "XML files found:",
    len(xml_files)
)

if not xml_files:

    raise FileNotFoundError(
        "No XML files found in:\n"
        + XML_DIR
    )


# ============================================================
# TEXT CLEANING
# ============================================================

def clean_text(text):

    if text is None:
        return ""

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


def element_text(element):

    if element is None:
        return ""

    return clean_text(
        " ".join(
            element.itertext()
        )
    )


# ============================================================
# TABLE EXTRACTION
# ============================================================

all_tables = []

article_summary = []

failed_files = []


for i, xml_file in enumerate(
    xml_files,
    start=1
):

    pmcid = os.path.splitext(
        os.path.basename(xml_file)
    )[0]

    try:

        tree = ET.parse(
            xml_file
        )

        root = tree.getroot()

        # ----------------------------------------------------
        # Article title
        # ----------------------------------------------------

        title_node = root.find(
            ".//article-title"
        )

        title = element_text(
            title_node
        )

        # ----------------------------------------------------
        # Abstract
        # ----------------------------------------------------

        abstract_nodes = root.findall(
            ".//abstract"
        )

        abstract = " ".join(
            element_text(x)
            for x in abstract_nodes
        )

        # ----------------------------------------------------
        # Tables
        # ----------------------------------------------------

        table_wraps = root.findall(
            ".//table-wrap"
        )

        n_tables = 0

        for table_index, table_wrap in enumerate(
            table_wraps
        ):

            n_tables += 1

            # ------------------------------------------------
            # Label
            # ------------------------------------------------

            label_node = table_wrap.find(
                "./label"
            )

            table_number = element_text(
                label_node
            )

            # ------------------------------------------------
            # Caption
            # ------------------------------------------------

            caption_node = table_wrap.find(
                "./caption"
            )

            caption = element_text(
                caption_node
            )

            # ------------------------------------------------
            # Table body
            # ------------------------------------------------

            table_node = table_wrap.find(
                "./table"
            )

            if table_node is None:

                table_node = table_wrap.find(
                    ".//table"
                )

            if table_node is None:
                continue

            # ------------------------------------------------
            # Header extraction
            # ------------------------------------------------

            header_cells = []

            for th in table_node.findall(
                ".//thead//th"
            ):

                header_cells.append(
                    element_text(th)
                )

            # ------------------------------------------------
            # Rows
            # ------------------------------------------------

            rows = table_node.findall(
                ".//tbody//tr"
            )

            if not rows:

                rows = table_node.findall(
                    ".//tr"
                )

            for row_index, row in enumerate(
                rows
            ):

                cells = row.findall(
                    "./th"
                ) + row.findall(
                    "./td"
                )

                if not cells:
                    continue

                row_values = [
                    element_text(c)
                    for c in cells
                ]

                row_text = " | ".join(
                    row_values
                )

                # --------------------------------------------
                # Cell-level records
                # --------------------------------------------

                for column_index, cell_text in enumerate(
                    row_values
                ):

                    all_tables.append({

                        "pmcid":
                            pmcid,

                        "table_index":
                            table_index,

                        "table_number":
                            table_number,

                        "table_caption":
                            caption,

                        "header_text":
                            " | ".join(
                                header_cells
                            ),

                        "row_index":
                            row_index,

                        "column_index":
                            column_index,

                        "raw_cell":
                            cell_text,

                        "row_text":
                            row_text,

                        "full_table_text":
                            clean_text(
                                element_text(
                                    table_node
                                )
                            ),

                        "article_title":
                            title,

                        "abstract":
                            abstract

                    })

        article_summary.append({

            "pmcid":
                pmcid,

            "title":
                title,

            "n_tables":
                n_tables,

            "status":
                "SUCCESS"

        })

    except Exception as e:

        failed_files.append({

            "pmcid":
                pmcid,

            "file":
                xml_file,

            "error":
                str(e)

        })

        article_summary.append({

            "pmcid":
                pmcid,

            "title":
                "",

            "n_tables":
                0,

            "status":
                "ERROR"

        })

    # --------------------------------------------------------
    # Progress
    # --------------------------------------------------------

    if (
        i % 100 == 0
        or i == len(xml_files)
    ):

        print(
            f"Processed "
            f"{i}/{len(xml_files)} articles"
        )


# ============================================================
# SAVE TABLE CELL DATASET
# ============================================================

tables_df = pd.DataFrame(
    all_tables
)

tables_file = os.path.join(
    OUTPUT_DIR,
    "03_extracted_table_cells.csv"
)

tables_df.to_csv(
    tables_file,
    index=False
)


# ============================================================
# SAVE ARTICLE SUMMARY
# ============================================================

article_summary_df = pd.DataFrame(
    article_summary
)

summary_file = os.path.join(
    OUTPUT_DIR,
    "03_article_table_summary.csv"
)

article_summary_df.to_csv(
    summary_file,
    index=False
)


# ============================================================
# SAVE FAILURE LOG
# ============================================================

failure_df = pd.DataFrame(
    failed_files
)

failure_file = os.path.join(
    LOG_DIR,
    "03_xml_table_extraction_failures.csv"
)

failure_df.to_csv(
    failure_file,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CODE 3 COMPLETE")
print("=" * 70)

print(
    "Articles processed:",
    len(xml_files)
)

print(
    "Articles with tables:",
    (
        article_summary_df["n_tables"] > 0
    ).sum()
)

print(
    "Total tables:",
    article_summary_df["n_tables"].sum()
)

print(
    "Extracted table-cell records:",
    len(tables_df)
)

print(
    "Articles with extraction errors:",
    len(failure_df)
)

print(
    "\nSaved:"
)

print(tables_file)
print(summary_file)
print(failure_file)

print("\nIMPORTANT:")
print(
    "This stage only extracts source tables."
)

print(
    "It does NOT interpret numbers as release values."
)

print(
    "It does NOT interpolate or synthesize observations."
)
```

```python
# ============================================================
# CODE 4 — DRUG-RELEASE TABLE IDENTIFICATION
# REAL LITERATURE ONLY
# ============================================================

import os
import re
import pandas as pd
import numpy as np

PROJECT_ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    PROJECT_ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUTPUT_DIR = os.path.join(
    PROJECT_ROOT,
    "data",
    "candidates"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

print("=" * 70)
print("CODE 4 — DRUG-RELEASE TABLE IDENTIFICATION")
print("=" * 70)

# ============================================================
# LOAD
# ============================================================

if not os.path.exists(INPUT_FILE):

    raise FileNotFoundError(
        f"Required Code 3 file not found:\n{INPUT_FILE}"
    )

df = pd.read_csv(
    INPUT_FILE
)

print(
    "Table-cell records:",
    len(df)
)

# ============================================================
# NORMALIZE TEXT
# ============================================================

TEXT_COLUMNS = [
    "table_caption",
    "header_text",
    "row_text",
    "full_table_text",
    "article_title",
    "abstract"
]

for col in TEXT_COLUMNS:

    if col not in df.columns:

        df[col] = ""

    df[col] = (
        df[col]
        .fillna("")
        .astype(str)
    )


df["combined_text"] = (
    df[TEXT_COLUMNS]
    .agg(" ".join, axis=1)
    .str.lower()
)

# ============================================================
# KEYWORD GROUPS
# ============================================================

TIME_TERMS = [
    r"\btime\b",
    r"\bhour\b",
    r"\bhours\b",
    r"\bhr\b",
    r"\bh\b",
    r"\bday\b",
    r"\bdays\b",
    r"\bweek\b",
    r"\bweeks\b",
    r"\bmin\b",
    r"\bminute\b",
    r"\bminutes\b"
]

RELEASE_TERMS = [
    r"\brelease\b",
    r"\breleased\b",
    r"\brelease profile\b",
    r"\brelease kinetics\b",
    r"\bcumulative release\b",
    r"\bdrug release\b",
    r"\bdrug released\b",
    r"\bpercent released\b",
    r"%\s*release",
    r"\breleased percentage\b"
]

DRUG_TERMS = [
    r"\bdrug\b",
    r"\bdrug-loaded\b",
    r"\bdrug loaded\b",
    r"\bactive\b",
    r"\bactive pharmaceutical\b",
    r"\bAPI\b",
    r"\bencapsulated\b",
    r"\bencapsulation\b"
]

POLYMER_TERMS = [
    r"\bpolymer\b",
    r"\bpolymeric\b",
    r"\bPLGA\b",
    r"\bPLA\b",
    r"\bPCL\b",
    r"\bPEG\b",
    r"\bPVA\b",
    r"\bchitosan\b",
    r"\bcellulose\b",
    r"\bpolylactic\b",
    r"\bpolycaprolactone\b"
]

PERCENT_TERMS = [
    r"%"
]

# ============================================================
# SCORING FUNCTION
# ============================================================

def count_matches(text, patterns):

    count = 0

    for pattern in patterns:

        try:

            count += len(
                re.findall(
                    pattern,
                    text,
                    flags=re.I
                )
            )

        except Exception:
            pass

    return count


df["time_score"] = (
    df["combined_text"]
    .apply(
        lambda x:
        count_matches(
            x,
            TIME_TERMS
        )
    )
)

df["release_score"] = (
    df["combined_text"]
    .apply(
        lambda x:
        count_matches(
            x,
            RELEASE_TERMS
        )
    )
)

df["drug_score"] = (
    df["combined_text"]
    .apply(
        lambda x:
        count_matches(
            x,
            DRUG_TERMS
        )
    )
)

df["polymer_score"] = (
    df["combined_text"]
    .apply(
        lambda x:
        count_matches(
            x,
            POLYMER_TERMS
        )
    )
)

df["percent_score"] = (
    df["combined_text"]
    .apply(
        lambda x:
        count_matches(
            x,
            PERCENT_TERMS
        )
    )
)

# ============================================================
# TABLE-LEVEL AGGREGATION
# ============================================================

GROUP_COLS = [
    "pmcid",
    "table_index"
]

table_summary = (
    df.groupby(
        GROUP_COLS,
        dropna=False
    )
    .agg(
        table_number=(
            "table_number",
            "first"
        ),

        table_caption=(
            "table_caption",
            "first"
        ),

        header_text=(
            "header_text",
            "first"
        ),

        article_title=(
            "article_title",
            "first"
        ),

        time_score=(
            "time_score",
            "max"
        ),

        release_score=(
            "release_score",
            "max"
        ),

        drug_score=(
            "drug_score",
            "max"
        ),

        polymer_score=(
            "polymer_score",
            "max"
        ),

        percent_score=(
            "percent_score",
            "max"
        ),

        n_cells=(
            "raw_cell",
            "count"
        )
    )
    .reset_index()
)

# ============================================================
# TABLE-LEVEL SCORE
# ============================================================

table_summary["release_table_score"] = (

    3.0 *
    (table_summary["release_score"] > 0)

    +

    3.0 *
    (table_summary["time_score"] > 0)

    +

    1.5 *
    (table_summary["percent_score"] > 0)

    +

    1.0 *
    (table_summary["drug_score"] > 0)

    +

    1.0 *
    (table_summary["polymer_score"] > 0)
)

# ============================================================
# CLASSIFICATION
# ============================================================

def classify_table(row):

    score = row[
        "release_table_score"
    ]

    has_release = (
        row["release_score"] > 0
    )

    has_time = (
        row["time_score"] > 0
    )

    has_percent = (
        row["percent_score"] > 0
    )

    if (
        score >= 6
        and has_release
        and has_time
        and has_percent
    ):

        return "HIGH_PRIORITY"

    if (
        score >= 4
        and (
            has_release
            or has_time
        )
    ):

        return "MEDIUM_PRIORITY"

    if score >= 2:

        return "LOW_PRIORITY"

    return "NOT_RELEASE_TABLE"


table_summary["priority"] = (
    table_summary.apply(
        classify_table,
        axis=1
    )
)

# ============================================================
# SAVE TABLE-LEVEL RESULTS
# ============================================================

summary_file = os.path.join(
    OUTPUT_DIR,
    "04_release_table_summary.csv"
)

table_summary.to_csv(
    summary_file,
    index=False
)

# ============================================================
# SELECT CANDIDATE TABLES
# ============================================================

candidate_tables = table_summary[
    table_summary["priority"].isin([
        "HIGH_PRIORITY",
        "MEDIUM_PRIORITY"
    ])
].copy()

candidate_tables = candidate_tables.sort_values(
    by="release_table_score",
    ascending=False
)

candidate_table_file = os.path.join(
    OUTPUT_DIR,
    "04_release_table_candidates.csv"
)

candidate_tables.to_csv(
    candidate_table_file,
    index=False
)

# ============================================================
# RETAIN ALL CELLS FROM CANDIDATE TABLES
# ============================================================

candidate_keys = candidate_tables[
    GROUP_COLS
].drop_duplicates()

candidate_cells = df.merge(
    candidate_keys,
    on=GROUP_COLS,
    how="inner"
)

candidate_cells_file = os.path.join(
    OUTPUT_DIR,
    "04_release_table_candidate_cells.csv"
)

candidate_cells.to_csv(
    candidate_cells_file,
    index=False
)

# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CODE 4 COMPLETE")
print("=" * 70)

print(
    "Unique source tables:",
    len(table_summary)
)

print(
    "High-priority release tables:",
    (
        table_summary["priority"]
        == "HIGH_PRIORITY"
    ).sum()
)

print(
    "Medium-priority release tables:",
    (
        table_summary["priority"]
        == "MEDIUM_PRIORITY"
    ).sum()
)

print(
    "Low-priority tables:",
    (
        table_summary["priority"]
        == "LOW_PRIORITY"
    ).sum()
)

print(
    "Non-release tables:",
    (
        table_summary["priority"]
        == "NOT_RELEASE_TABLE"
    ).sum()
)

print(
    "\nCandidate source tables:",
    len(candidate_tables)
)

print(
    "Candidate table-cell records:",
    len(candidate_cells)
)

print(
    "\nPriority distribution:"
)

print(
    table_summary[
        "priority"
    ].value_counts()
)

print(
    "\nTop candidate tables:"
)

display(
    candidate_tables[
        [
            "pmcid",
            "table_index",
            "table_number",
            "table_caption",
            "release_table_score",
            "priority",
            "n_cells"
        ]
    ].head(30)
)

print(
    "\nSaved:"
)

print(summary_file)
print(candidate_table_file)
print(candidate_cells_file)

print("\nIMPORTANT:")
print(
    "These are candidate tables, NOT validated observations."
)

print(
    "No release values have been extracted yet."
)
```

```python
# ============================================================
# CODE 4 FAST — DRUG-RELEASE TABLE IDENTIFICATION
# ============================================================

import os
import re
import pandas as pd
import numpy as np

PROJECT_ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    PROJECT_ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUTPUT_DIR = os.path.join(
    PROJECT_ROOT,
    "data",
    "candidates"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("=" * 70)
print("CODE 4 FAST — DRUG-RELEASE TABLE IDENTIFICATION")
print("=" * 70)

# ============================================================
# LOAD ONLY REQUIRED COLUMNS
# ============================================================

usecols = [
    "pmcid",
    "table_index",
    "table_number",
    "table_caption",
    "header_text",
    "row_text",
    "full_table_text",
    "article_title"
]

df = pd.read_csv(
    INPUT_FILE,
    usecols=usecols,
    low_memory=False
)

print(
    "Table-cell records loaded:",
    f"{len(df):,}"
)

# ============================================================
# NORMALIZE
# ============================================================

for col in usecols:

    df[col] = (
        df[col]
        .fillna("")
        .astype(str)
    )

# ============================================================
# CREATE ONE RECORD PER TABLE
# ============================================================

table_df = (
    df.groupby(
        ["pmcid", "table_index"],
        sort=False
    )
    .agg({
        "table_number": "first",
        "table_caption": "first",
        "header_text": "first",
        "full_table_text": "first",
        "article_title": "first",
        "row_text": lambda x: " | ".join(
            x.astype(str).unique()
        )
    })
    .reset_index()
)

print(
    "Unique source tables:",
    f"{len(table_df):,}"
)

# ============================================================
# COMBINE TABLE TEXT
# ============================================================

table_df["search_text"] = (
    table_df["table_caption"]
    + " "
    + table_df["header_text"]
    + " "
    + table_df["row_text"]
    + " "
    + table_df["full_table_text"]
).str.lower()

# ============================================================
# FAST KEYWORD SCORING
# ============================================================

release_pattern = (
    r"\brelease\b|"
    r"\breleased\b|"
    r"\brelease profile\b|"
    r"\brelease kinetics\b|"
    r"\bcumulative release\b|"
    r"\bdrug release\b|"
    r"\bdrug released\b|"
    r"\bpercent released\b|"
    r"%\s*release"
)

time_pattern = (
    r"\btime\b|"
    r"\bhour\b|"
    r"\bhours\b|"
    r"\bhr\b|"
    r"\bday\b|"
    r"\bdays\b|"
    r"\bweek\b|"
    r"\bweeks\b|"
    r"\bminute\b|"
    r"\bminutes\b|"
    r"\bmin\b"
)

drug_pattern = (
    r"\bdrug\b|"
    r"\bdrug-loaded\b|"
    r"\bdrug loaded\b|"
    r"\bencapsulated\b|"
    r"\bencapsulation\b|"
    r"\bactive pharmaceutical\b"
)

polymer_pattern = (
    r"\bpolymer\b|"
    r"\bpolymeric\b|"
    r"\bplga\b|"
    r"\bpla\b|"
    r"\bpcl\b|"
    r"\bpeg\b|"
    r"\bpva\b|"
    r"\bchitosan\b|"
    r"\bcellulose\b"
)

percent_pattern = r"%"

# ============================================================
# BOOLEAN FEATURES
# ============================================================

print("Scoring tables...")

text = table_df["search_text"]

table_df["has_release"] = (
    text.str.contains(
        release_pattern,
        regex=True,
        na=False
    )
)

table_df["has_time"] = (
    text.str.contains(
        time_pattern,
        regex=True,
        na=False
    )
)

table_df["has_drug"] = (
    text.str.contains(
        drug_pattern,
        regex=True,
        na=False
    )
)

table_df["has_polymer"] = (
    text.str.contains(
        polymer_pattern,
        regex=True,
        na=False
    )
)

table_df["has_percent"] = (
    text.str.contains(
        percent_pattern,
        regex=True,
        na=False
    )
)

# ============================================================
# SCORE
# ============================================================

table_df["release_table_score"] = (
    table_df["has_release"].astype(int) * 3
    +
    table_df["has_time"].astype(int) * 3
    +
    table_df["has_percent"].astype(int) * 1.5
    +
    table_df["has_drug"].astype(int) * 1
    +
    table_df["has_polymer"].astype(int) * 1
)

# ============================================================
# PRIORITY
# ============================================================

conditions = [

    (
        (table_df["release_table_score"] >= 6)
        &
        table_df["has_release"]
        &
        table_df["has_time"]
        &
        table_df["has_percent"]
    ),

    (
        (table_df["release_table_score"] >= 4)
        &
        (
            table_df["has_release"]
            |
            table_df["has_time"]
        )
    ),

    (
        table_df["release_table_score"] >= 2
    )
]

choices = [
    "HIGH_PRIORITY",
    "MEDIUM_PRIORITY",
    "LOW_PRIORITY"
]

table_df["priority"] = np.select(
    conditions,
    choices,
    default="NOT_RELEASE_TABLE"
)

# ============================================================
# SAVE TABLE SUMMARY
# ============================================================

summary_file = os.path.join(
    OUTPUT_DIR,
    "04_release_table_summary.csv"
)

table_df.to_csv(
    summary_file,
    index=False
)

# ============================================================
# SELECT CANDIDATE TABLES
# ============================================================

candidate_tables = (
    table_df[
        table_df["priority"].isin(
            [
                "HIGH_PRIORITY",
                "MEDIUM_PRIORITY"
            ]
        )
    ]
    .sort_values(
        "release_table_score",
        ascending=False
    )
    .copy()
)

candidate_file = os.path.join(
    OUTPUT_DIR,
    "04_release_table_candidates.csv"
)

candidate_tables.to_csv(
    candidate_file,
    index=False
)

# ============================================================
# RECOVER ALL CELLS FROM CANDIDATE TABLES
# ============================================================

keys = candidate_tables[
    ["pmcid", "table_index"]
].drop_duplicates()

candidate_cells = df.merge(
    keys,
    on=["pmcid", "table_index"],
    how="inner"
)

candidate_cells_file = os.path.join(
    OUTPUT_DIR,
    "04_release_table_candidate_cells.csv"
)

candidate_cells.to_csv(
    candidate_cells_file,
    index=False
)

# ============================================================
# RESULTS
# ============================================================

print("\n" + "=" * 70)
print("CODE 4 FAST COMPLETE")
print("=" * 70)

print(
    "Source tables:",
    f"{len(table_df):,}"
)

print(
    "HIGH priority:",
    f"{(table_df['priority'] == 'HIGH_PRIORITY').sum():,}"
)

print(
    "MEDIUM priority:",
    f"{(table_df['priority'] == 'MEDIUM_PRIORITY').sum():,}"
)

print(
    "LOW priority:",
    f"{(table_df['priority'] == 'LOW_PRIORITY').sum():,}"
)

print(
    "Not release tables:",
    f"{(table_df['priority'] == 'NOT_RELEASE_TABLE').sum():,}"
)

print(
    "\nCandidate tables:",
    f"{len(candidate_tables):,}"
)

print(
    "Candidate cell records:",
    f"{len(candidate_cells):,}"
)

print("\nPriority distribution:")
print(
    table_df["priority"].value_counts()
)

print("\nTop 20 candidate tables:")

display(
    candidate_tables[
        [
            "pmcid",
            "table_index",
            "table_number",
            "table_caption",
            "header_text",
            "release_table_score",
            "priority"
        ]
    ].head(20)
)

print("\nSaved:")
print(summary_file)
print(candidate_file)
print(candidate_cells_file)

print("\nIMPORTANT:")
print(
    "These remain CANDIDATE TABLES only."
)

print(
    "No observations have been created."
)

print(
    "No interpolation, duplication, or synthetic data."
)

print(
    "DO NOT RUN ML."
)
```

```python
# ============================================================
# CODE 5 — PRIMARY RELEASE TABLE VALIDATION
# ============================================================

import os
import re
import pandas as pd
import numpy as np

PROJECT_ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    PROJECT_ROOT,
    "data",
    "candidates",
    "04_release_table_candidates.csv"
)

OUTPUT_DIR = os.path.join(
    PROJECT_ROOT,
    "data",
    "candidates"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("=" * 70)
print("CODE 5 — PRIMARY RELEASE TABLE VALIDATION")
print("=" * 70)

df = pd.read_csv(
    INPUT_FILE,
    low_memory=False
)

print(
    "Candidate tables:",
    len(df)
)

# ============================================================
# NORMALIZE
# ============================================================

TEXT_COLS = [
    "table_caption",
    "header_text",
    "article_title"
]

for col in TEXT_COLS:

    if col not in df.columns:
        df[col] = ""

    df[col] = (
        df[col]
        .fillna("")
        .astype(str)
    )

df["search_text"] = (
    df[TEXT_COLS]
    .agg(" ".join, axis=1)
    .str.lower()
)

# ============================================================
# PRIMARY EXPERIMENTAL RELEASE SIGNALS
# ============================================================

primary_terms = [
    r"cumulative drug release",
    r"drug release",
    r"drug released",
    r"cumulative release",
    r"release percentage",
    r"release \(%\)",
    r"release \(%\)",
    r"released \(%\)",
    r"drug released \(%\)",
    r"release profile",
    r"release kinetics"
]

review_terms = [
    r"selected studies",
    r"comparative",
    r"comparison",
    r"overview",
    r"review",
    r"recent studies",
    r"literature",
    r"representative",
    r"summary of studies",
    r"studies on",
    r"reported studies"
]

time_header_terms = [
    r"time\s*\(",
    r"time\s*\[",
    r"time\s+h",
    r"hours?",
    r"\bhr\b",
    r"\bh\b",
    r"days?",
    r"minutes?"
]

release_header_terms = [
    r"release",
    r"released",
    r"cumulative"
]

# ============================================================
# FEATURE EXTRACTION
# ============================================================

def has_any(text, patterns):

    return any(
        re.search(
            p,
            text,
            flags=re.I
        )
        for p in patterns
    )


df["primary_release_signal"] = (
    df["search_text"]
    .apply(
        lambda x:
        has_any(
            x,
            primary_terms
        )
    )
)

df["review_signal"] = (
    df["search_text"]
    .apply(
        lambda x:
        has_any(
            x,
            review_terms
        )
    )
)

df["time_header_signal"] = (
    df["header_text"]
    .apply(
        lambda x:
        has_any(
            x,
            time_header_terms
        )
    )
)

df["release_header_signal"] = (
    df["header_text"]
    .apply(
        lambda x:
        has_any(
            x,
            release_header_terms
        )
    )
)

# ============================================================
# NUMERIC STRUCTURE
# ============================================================

# Number of numeric-looking values in header/caption.
number_pattern = r"(?<![A-Za-z])[-+]?\d+(?:\.\d+)?"

df["numeric_count_text"] = (
    df["search_text"]
    .str.count(number_pattern)
)

# Percent indication
df["has_percent"] = (
    df["search_text"]
    .str.contains(
        "%",
        regex=False,
        na=False
    )
)

# ============================================================
# PRIMARY EVIDENCE SCORE
# ============================================================

df["primary_evidence_score"] = 0

df.loc[
    df["primary_release_signal"],
    "primary_evidence_score"
] += 4

df.loc[
    df["time_header_signal"],
    "primary_evidence_score"
] += 3

df.loc[
    df["release_header_signal"],
    "primary_evidence_score"
] += 3

df.loc[
    df["has_percent"],
    "primary_evidence_score"
] += 1

df.loc[
    df["review_signal"],
    "primary_evidence_score"
] -= 5

# ============================================================
# CLASSIFICATION
# ============================================================

def classify(row):

    score = row["primary_evidence_score"]

    primary = row["primary_release_signal"]
    time = row["time_header_signal"]
    release = row["release_header_signal"]
    review = row["review_signal"]

    # Strongest condition:
    # release + time + numerical percentage context
    if (
        primary
        and time
        and release
        and row["has_percent"]
        and score >= 7
        and not review
    ):
        return "PRIMARY_RELEASE_TABLE"

    # Possible primary table requiring structural inspection
    if (
        primary
        and time
        and release
        and score >= 5
    ):
        return "POSSIBLE_PRIMARY_TABLE"

    # Tables with release language but likely review/summary
    if (
        primary
        and review
    ):
        return "REVIEW_OR_COMPARISON_TABLE"

    if (
        primary
        and not time
    ):
        return "RELEASE_WITHOUT_TIME"

    return "OTHER"

df["classification"] = (
    df.apply(
        classify,
        axis=1
    )
)

# ============================================================
# SAVE FULL VALIDATION TABLE
# ============================================================

validation_file = os.path.join(
    OUTPUT_DIR,
    "05_release_table_validation.csv"
)

df.to_csv(
    validation_file,
    index=False
)

# ============================================================
# PRIMARY CANDIDATES
# ============================================================

primary = df[
    df["classification"].isin(
        [
            "PRIMARY_RELEASE_TABLE",
            "POSSIBLE_PRIMARY_TABLE"
        ]
    )
].copy()

primary = primary.sort_values(
    [
        "classification",
        "primary_evidence_score"
    ],
    ascending=[True, False]
)

primary_file = os.path.join(
    OUTPUT_DIR,
    "05_primary_release_table_candidates.csv"
)

primary.to_csv(
    primary_file,
    index=False
)

# ============================================================
# REVIEW TABLES
# ============================================================

review_tables = df[
    df["classification"]
    == "REVIEW_OR_COMPARISON_TABLE"
].copy()

review_file = os.path.join(
    OUTPUT_DIR,
    "05_review_comparison_tables.csv"
)

review_tables.to_csv(
    review_file,
    index=False
)

# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CODE 5 COMPLETE")
print("=" * 70)

print(
    "Candidate tables:",
    len(df)
)

print("\nClassification:")

print(
    df["classification"]
    .value_counts()
    .to_string()
)

print(
    "\nPRIMARY_RELEASE_TABLE:",
    (
        df["classification"]
        == "PRIMARY_RELEASE_TABLE"
    ).sum()
)

print(
    "POSSIBLE_PRIMARY_TABLE:",
    (
        df["classification"]
        == "POSSIBLE_PRIMARY_TABLE"
    ).sum()
)

print(
    "\nPrimary extraction candidates:",
    len(primary)
)

print("\nTop candidates:")

display(
    primary[
        [
            "pmcid",
            "table_index",
            "table_number",
            "table_caption",
            "header_text",
            "primary_evidence_score",
            "classification"
        ]
    ].head(30)
)

print("\nSaved:")
print(validation_file)
print(primary_file)
print(review_file)

print("\nIMPORTANT:")
print(
    "No observations have been created."
)

print(
    "No interpolation has been performed."
)

print(
    "No synthetic data have been generated."
)

print(
    "Do NOT run ML."
)
```

```python
# ============================================================
# CODE 5 — EXTRACT CANDIDATE DRUG-RELEASE OBSERVATIONS
# ============================================================

import os
import re
import glob
import json
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup

ROOT = "/content/AI_Polymer_Drug_Release"

XML_DIR = os.path.join(ROOT, "data", "raw", "fulltext_xml")
OUT_DIR = os.path.join(ROOT, "data", "raw", "extraction")
os.makedirs(OUT_DIR, exist_ok=True)

OUTPUT = os.path.join(
    OUT_DIR,
    "candidate_release_observations.csv"
)

# ------------------------------------------------------------
# Find all XML files
# ------------------------------------------------------------

xml_files = glob.glob(
    os.path.join(XML_DIR, "**", "*.xml"),
    recursive=True
)

print("=" * 70)
print("CODE 5 — LITERATURE RELEASE EXTRACTION")
print("=" * 70)

print("XML files found:", len(xml_files))


# ------------------------------------------------------------
# Keywords indicating drug-release tables/text
# ------------------------------------------------------------

RELEASE_TERMS = [
    "drug release",
    "release profile",
    "release percentage",
    "release (%)",
    "cumulative release",
    "cumulative drug release",
    "drug released",
    "release amount",
    "in vitro release",
    "in-vitro release",
    "release kinetics"
]

TIME_TERMS = [
    "time",
    "hour",
    "hours",
    "hr",
    "hrs",
    "day",
    "days",
    "min",
    "minute",
    "minutes"
]


# ------------------------------------------------------------
# Helper functions
# ------------------------------------------------------------

def clean_text(x):
    if x is None:
        return ""
    x = str(x)
    x = re.sub(r"\s+", " ", x)
    return x.strip()


def normalize_header(x):
    x = clean_text(x).lower()
    x = x.replace("%", "percent")
    x = re.sub(r"[^a-z0-9]+", " ", x)
    return x.strip()


def numeric_value(x):
    """
    Extract first numeric value from a cell.
    Handles:
      25
      25.4
      25 ± 1.2
      25%
      <25
      >25
    """
    x = clean_text(x)

    match = re.search(
        r"[-+]?\d+(?:\.\d+)?",
        x
    )

    if match:
        try:
            return float(match.group())
        except Exception:
            return np.nan

    return np.nan


def extract_time_value(x):
    """
    Extract numeric time from a cell.
    """
    x = clean_text(x).lower()

    value = numeric_value(x)

    if np.isnan(value):
        return np.nan

    # Convert minutes to hours
    if "min" in x or "minute" in x:
        return value / 60.0

    # Days to hours
    if "day" in x:
        return value * 24.0

    # Otherwise assume hours
    return value


def looks_like_release_header(x):
    x = normalize_header(x)

    terms = [
        "release",
        "released",
        "cumulative release",
        "drug release",
        "release percent",
        "release percentage"
    ]

    return any(term in x for term in terms)


def looks_like_time_header(x):
    x = normalize_header(x)

    terms = [
        "time",
        "hour",
        "hours",
        "hr",
        "hrs",
        "day",
        "days",
        "minute",
        "minutes",
        "min"
    ]

    return any(term in x for term in terms)


# ------------------------------------------------------------
# Parse individual article
# ------------------------------------------------------------

def extract_article_tables(xml_path):

    observations = []

    try:
        with open(
            xml_path,
            "r",
            encoding="utf-8",
            errors="ignore"
        ) as f:
            xml = f.read()

        soup = BeautifulSoup(xml, "xml")

    except Exception as e:
        return observations

    # --------------------------------------------------------
    # DOI
    # --------------------------------------------------------

    doi = ""

    for article_id in soup.find_all("article-id"):

        id_type = article_id.get("pub-id-type", "")

        if id_type.lower() == "doi":
            doi = clean_text(article_id.get_text())
            break

    # --------------------------------------------------------
    # PMCID
    # --------------------------------------------------------

    pmcid = ""

    for article_id in soup.find_all("article-id"):

        id_type = article_id.get("pub-id-type", "")

        if id_type.lower() == "pmc":
            pmcid = clean_text(article_id.get_text())
            break

    # --------------------------------------------------------
    # Article title
    # --------------------------------------------------------

    title_node = soup.find("article-title")

    title = ""

    if title_node:
        title = clean_text(
            title_node.get_text(" ", strip=True)
        )

    # --------------------------------------------------------
    # Examine tables
    # --------------------------------------------------------

    tables = soup.find_all("table-wrap")

    for table_index, table in enumerate(tables):

        table_text = clean_text(
            table.get_text(" ", strip=True)
        ).lower()

        # Skip tables that have no release-related terminology
        if not any(
            term in table_text
            for term in RELEASE_TERMS
        ):
            continue

        caption_node = table.find("caption")

        caption = ""

        if caption_node:
            caption = clean_text(
                caption_node.get_text(
                    " ",
                    strip=True
                )
            )

        # ----------------------------------------------------
        # Extract rows
        # ----------------------------------------------------

        rows = table.find_all("tr")

        if not rows:
            continue

        parsed_rows = []

        for row in rows:

            cells = row.find_all(
                ["th", "td"]
            )

            values = [
                clean_text(
                    cell.get_text(
                        " ",
                        strip=True
                    )
                )
                for cell in cells
            ]

            if values:
                parsed_rows.append(values)

        if len(parsed_rows) < 2:
            continue

        # ----------------------------------------------------
        # Detect header row
        # ----------------------------------------------------

        header_idx = None

        for i, row in enumerate(parsed_rows[:5]):

            row_text = " ".join(
                normalize_header(x)
                for x in row
            )

            has_time = any(
                looks_like_time_header(x)
                for x in row
            )

            has_release = any(
                looks_like_release_header(x)
                for x in row
            )

            if has_time or has_release:
                header_idx = i
                break

        if header_idx is None:
            continue

        headers = parsed_rows[header_idx]

        # ----------------------------------------------------
        # Identify columns
        # ----------------------------------------------------

        time_cols = []
        release_cols = []

        for j, header in enumerate(headers):

            if looks_like_time_header(header):
                time_cols.append(j)

            if looks_like_release_header(header):
                release_cols.append(j)

        if not time_cols or not release_cols:
            continue

        # Use first matching columns initially.
        # Validation will occur later.
        time_col = time_cols[0]
        release_col = release_cols[0]

        # ----------------------------------------------------
        # Extract observations
        # ----------------------------------------------------

        for row_index, row in enumerate(
            parsed_rows[header_idx + 1:],
            start=header_idx + 1
        ):

            if (
                time_col >= len(row)
                or release_col >= len(row)
            ):
                continue

            time_raw = row[time_col]
            release_raw = row[release_col]

            time_hours = extract_time_value(
                time_raw
            )

            release_pct = numeric_value(
                release_raw
            )

            # Basic sanity checks only
            if np.isnan(time_hours):
                continue

            if np.isnan(release_pct):
                continue

            if time_hours < 0:
                continue

            if release_pct < 0 or release_pct > 100:
                continue

            observations.append({

                "study_doi": doi,

                "pmcid": pmcid,

                "article_title": title,

                "table_index": table_index,

                "table_row": row_index,

                "table_caption": caption,

                "time_raw": time_raw,

                "release_raw": release_raw,

                "time_hours": time_hours,

                "release_pct": release_pct,

                "source_xml": os.path.basename(
                    xml_path
                ),

                "extraction_status":
                    "candidate"

            })

    return observations


# ============================================================
# RUN EXTRACTION
# ============================================================

all_observations = []

for i, xml_path in enumerate(xml_files):

    obs = extract_article_tables(xml_path)

    all_observations.extend(obs)

    if (i + 1) % 100 == 0:

        print(
            f"Processed {i + 1}/{len(xml_files)} | "
            f"Candidates: {len(all_observations)}"
        )


# ============================================================
# SAVE
# ============================================================

candidate_df = pd.DataFrame(
    all_observations
)

if candidate_df.empty:

    print("\nNO CANDIDATE OBSERVATIONS FOUND.")

else:

    candidate_df.to_csv(
        OUTPUT,
        index=False
    )

    print("\n" + "=" * 70)
    print("CODE 5 COMPLETE")
    print("=" * 70)

    print(
        "Candidate observations:",
        len(candidate_df)
    )

    print(
        "Unique studies:",
        candidate_df["study_doi"].replace(
            "",
            np.nan
        ).nunique()
    )

    print(
        "Unique articles:",
        candidate_df["source_xml"].nunique()
    )

    print(
        "Output:",
        OUTPUT
    )

    print("\nFirst observations:")

    display(
        candidate_df.head(20)
    )
```

```python
# ============================================================
# CODE 5 — LITERATURE RELEASE EXTRACTION
# CORRECTED XML PATH
# ============================================================

import os
import re
import glob
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup

ROOT = "/content/AI_Polymer_Drug_Release"

XML_DIR = os.path.join(
    ROOT,
    "data",
    "raw",
    "pmc_xml"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "candidates"
)

os.makedirs(OUT_DIR, exist_ok=True)

OUTPUT = os.path.join(
    OUT_DIR,
    "05_candidate_release_observations.csv"
)

print("=" * 70)
print("CODE 5 — LITERATURE RELEASE EXTRACTION")
print("=" * 70)

# ============================================================
# FIND XML FILES
# ============================================================

xml_files = glob.glob(
    os.path.join(
        XML_DIR,
        "*.xml"
    )
)

print(
    "XML files found:",
    len(xml_files)
)

if len(xml_files) == 0:
    raise FileNotFoundError(
        f"No XML files found in:\n{XML_DIR}"
    )

# ============================================================
# PATTERNS
# ============================================================

RELEASE_TERMS = [
    "drug release",
    "release profile",
    "release percentage",
    "release (%)",
    "release %",
    "cumulative release",
    "cumulative drug release",
    "drug released",
    "release amount",
    "in vitro release",
    "in-vitro release",
    "release kinetics"
]

def clean_text(x):

    if x is None:
        return ""

    x = str(x)

    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


def normalize_header(x):

    x = clean_text(x).lower()

    x = x.replace(
        "%",
        "percent"
    )

    x = re.sub(
        r"[^a-z0-9]+",
        " ",
        x
    )

    return x.strip()


def numeric_value(x):

    x = clean_text(x)

    match = re.search(
        r"[-+]?\d+(?:\.\d+)?",
        x
    )

    if match:

        try:
            return float(
                match.group()
            )

        except:
            return np.nan

    return np.nan


def extract_time_value(x):

    x = clean_text(x).lower()

    value = numeric_value(x)

    if np.isnan(value):
        return np.nan

    # minutes → hours
    if (
        "min" in x
        or "minute" in x
    ):
        return value / 60.0

    # days → hours
    if "day" in x:
        return value * 24.0

    # otherwise assume hours
    return value


def is_time_header(x):

    x = normalize_header(x)

    patterns = [
        r"\btime\b",
        r"\bhour\b",
        r"\bhours\b",
        r"\bhr\b",
        r"\bhrs\b",
        r"\bday\b",
        r"\bdays\b",
        r"\bminute\b",
        r"\bminutes\b",
        r"\bmin\b"
    ]

    return any(
        re.search(
            p,
            x
        )
        for p in patterns
    )


def is_release_header(x):

    x = normalize_header(x)

    patterns = [
        r"\brelease\b",
        r"\breleased\b",
        r"\bcumulative release\b",
        r"\bdrug release\b",
        r"\brelease percent\b",
        r"\brelease percentage\b"
    ]

    return any(
        re.search(
            p,
            x
        )
        for p in patterns
    )


# ============================================================
# ARTICLE EXTRACTION
# ============================================================

def extract_article(xml_path):

    observations = []

    try:

        with open(
            xml_path,
            "r",
            encoding="utf-8",
            errors="ignore"
        ) as f:

            xml = f.read()

        soup = BeautifulSoup(
            xml,
            "xml"
        )

    except Exception:

        return observations

    # --------------------------------------------------------
    # PMCID
    # --------------------------------------------------------

    pmcid = ""

    for node in soup.find_all(
        "article-id"
    ):

        if (
            node.get(
                "pub-id-type",
                ""
            ).lower()
            == "pmc"
        ):

            pmcid = clean_text(
                node.get_text()
            )

            break

    # --------------------------------------------------------
    # PMID
    # --------------------------------------------------------

    pmid = ""

    for node in soup.find_all(
        "article-id"
    ):

        if (
            node.get(
                "pub-id-type",
                ""
            ).lower()
            == "pmid"
        ):

            pmid = clean_text(
                node.get_text()
            )

            break

    # --------------------------------------------------------
    # DOI
    # --------------------------------------------------------

    doi = ""

    for node in soup.find_all(
        "article-id"
    ):

        if (
            node.get(
                "pub-id-type",
                ""
            ).lower()
            == "doi"
        ):

            doi = clean_text(
                node.get_text()
            )

            break

    # --------------------------------------------------------
    # TITLE
    # --------------------------------------------------------

    title = ""

    title_node = soup.find(
        "article-title"
    )

    if title_node:

        title = clean_text(
            title_node.get_text(
                " ",
                strip=True
            )
        )

    # --------------------------------------------------------
    # TABLES
    # --------------------------------------------------------

    tables = soup.find_all(
        "table-wrap"
    )

    for table_index, table in enumerate(
        tables
    ):

        table_text = clean_text(
            table.get_text(
                " ",
                strip=True
            )
        )

        table_lower = table_text.lower()

        # --------------------------------------------
        # RELEASE TABLE FILTER
        # --------------------------------------------

        if not any(
            term in table_lower
            for term in RELEASE_TERMS
        ):
            continue

        # --------------------------------------------
        # CAPTION
        # --------------------------------------------

        caption = ""

        caption_node = table.find(
            "caption"
        )

        if caption_node:

            caption = clean_text(
                caption_node.get_text(
                    " ",
                    strip=True
                )
            )

        # --------------------------------------------
        # TABLE LABEL
        # --------------------------------------------

        label = ""

        label_node = table.find(
            "label"
        )

        if label_node:

            label = clean_text(
                label_node.get_text()
            )

        # --------------------------------------------
        # ROWS
        # --------------------------------------------

        rows = table.find_all(
            "tr"
        )

        if len(rows) < 2:
            continue

        parsed_rows = []

        for row in rows:

            cells = row.find_all(
                [
                    "th",
                    "td"
                ]
            )

            values = [
                clean_text(
                    c.get_text(
                        " ",
                        strip=True
                    )
                )
                for c in cells
            ]

            if values:
                parsed_rows.append(
                    values
                )

        if len(parsed_rows) < 2:
            continue

        # --------------------------------------------
        # FIND HEADER
        # --------------------------------------------

        header_idx = None

        for i, row in enumerate(
            parsed_rows[:8]
        ):

            time_signal = any(
                is_time_header(x)
                for x in row
            )

            release_signal = any(
                is_release_header(x)
                for x in row
            )

            if (
                time_signal
                and release_signal
            ):

                header_idx = i
                break

        # --------------------------------------------
        # IF NO COMBINED HEADER,
        # TRY TIME + RELEASE ACROSS HEADER ROWS
        # --------------------------------------------

        if header_idx is None:

            for i, row in enumerate(
                parsed_rows[:8]
            ):

                if any(
                    is_time_header(x)
                    for x in row
                ):

                    header_idx = i
                    break

        if header_idx is None:
            continue

        headers = parsed_rows[
            header_idx
        ]

        # --------------------------------------------
        # IDENTIFY COLUMNS
        # --------------------------------------------

        time_cols = [
            j
            for j, h in enumerate(headers)
            if is_time_header(h)
        ]

        release_cols = [
            j
            for j, h in enumerate(headers)
            if is_release_header(h)
        ]

        if not time_cols:
            continue

        if not release_cols:
            continue

        # --------------------------------------------
        # EXTRACT PAIRS
        # --------------------------------------------

        time_col = time_cols[0]
        release_col = release_cols[0]

        for row_index, row in enumerate(
            parsed_rows[
                header_idx + 1:
            ],
            start=header_idx + 1
        ):

            if (
                time_col >= len(row)
                or release_col >= len(row)
            ):
                continue

            time_raw = row[
                time_col
            ]

            release_raw = row[
                release_col
            ]

            time_h = extract_time_value(
                time_raw
            )

            release_pct = numeric_value(
                release_raw
            )

            # ----------------------------------------
            # VALIDITY
            # ----------------------------------------

            if np.isnan(time_h):
                continue

            if np.isnan(release_pct):
                continue

            if time_h < 0:
                continue

            if (
                release_pct < 0
                or release_pct > 100
            ):
                continue

            observations.append({

                "observation_candidate_id":
                    f"{pmcid}_T{table_index}_R{row_index}",

                "pmcid":
                    pmcid,

                "pmid":
                    pmid,

                "doi":
                    doi,

                "title":
                    title,

                "table_index":
                    table_index,

                "table_label":
                    label,

                "table_caption":
                    caption,

                "row_index":
                    row_index,

                "time_raw":
                    time_raw,

                "release_raw":
                    release_raw,

                "release_time_h":
                    time_h,

                "release_percent":
                    release_pct,

                "source_xml":
                    os.path.basename(
                        xml_path
                    ),

                "source_type":
                    "PMC_FULL_TEXT_TABLE",

                "extraction_status":
                    "CANDIDATE"

            })

    return observations


# ============================================================
# PROCESS ARTICLES
# ============================================================

all_obs = []

for i, xml_path in enumerate(
    xml_files
):

    obs = extract_article(
        xml_path
    )

    all_obs.extend(
        obs
    )

    if (
        (i + 1) % 100 == 0
        or i + 1 == len(xml_files)
    ):

        print(
            f"Processed {i+1:,}/{len(xml_files):,} | "
            f"Candidate observations: {len(all_obs):,}"
        )


# ============================================================
# DATAFRAME
# ============================================================

candidate_df = pd.DataFrame(
    all_obs
)

if candidate_df.empty:

    print(
        "\nNO CANDIDATE OBSERVATIONS FOUND."
    )

else:

    # Remove exact duplicate candidates
    candidate_df = (
        candidate_df
        .drop_duplicates(
            subset=[
                "pmcid",
                "table_index",
                "row_index",
                "release_time_h",
                "release_percent"
            ]
        )
        .reset_index(drop=True)
    )

    candidate_df.to_csv(
        OUTPUT,
        index=False
    )

    print("\n" + "=" * 70)
    print("CODE 5 COMPLETE")
    print("=" * 70)

    print(
        "XML articles:",
        len(xml_files)
    )

    print(
        "Candidate observations:",
        f"{len(candidate_df):,}"
    )

    print(
        "Unique PMCID:",
        candidate_df["pmcid"].nunique()
    )

    print(
        "Unique DOI:",
        candidate_df["doi"].replace(
            "",
            np.nan
        ).nunique()
    )

    print(
        "Unique source tables:",
        candidate_df[
            ["pmcid", "table_index"]
        ].drop_duplicates().shape[0]
    )

    print("\nRelease statistics:")

    print(
        candidate_df[
            "release_percent"
        ].describe()
    )

    print("\nTime statistics:")

    print(
        candidate_df[
            "release_time_h"
        ].describe()
    )

    print("\nPreview:")

    display(
        candidate_df.head(20)
    )

    print(
        "\nSaved:"
    )

    print(
        OUTPUT
    )

    print(
        "\nIMPORTANT:"
    )

    print(
        "These are CANDIDATE observations."
    )

    print(
        "They have NOT yet passed final validation."
    )

    print(
        "Do NOT train ML on this file."
    )
```

```python
# ============================================================
# CODE 5C — CORRECTED STRUCTURE-AWARE RELEASE EXTRACTION
# ============================================================

import os
import re
import glob
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup

ROOT = "/content/AI_Polymer_Drug_Release"

XML_DIR = os.path.join(
    ROOT,
    "data",
    "raw",
    "pmc_xml"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "candidates"
)

os.makedirs(
    OUT_DIR,
    exist_ok=True
)

OUTPUT = os.path.join(
    OUT_DIR,
    "05C_structured_release_candidates.csv"
)

# ============================================================
# HELPERS
# ============================================================

def clean(x):

    if x is None:
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


def number(x):

    x = clean(x)

    # Remove common uncertainty notation
    x = re.sub(
        r"±.*",
        "",
        x
    )

    m = re.search(
        r"[-+]?\d+(?:\.\d+)?",
        x
    )

    if not m:
        return np.nan

    try:
        return float(
            m.group()
        )
    except:
        return np.nan


def time_to_hours(x):

    x = clean(x).lower()

    value = number(x)

    if np.isnan(value):
        return np.nan

    if (
        "minute" in x
        or re.search(r"\bmin\b", x)
    ):
        return value / 60

    if (
        "day" in x
        or re.search(r"\bd\b", x)
    ):
        return value * 24

    # Explicit hour units
    if (
        "hour" in x
        or re.search(r"\bhr\b", x)
        or re.search(r"\bhrs\b", x)
        or re.search(r"\bh\b", x)
    ):
        return value

    # Bare numerical time values are NOT automatically
    # accepted here.
    return np.nan


def is_time_header(x):

    x = clean(x).lower()

    patterns = [
        r"\btime\b",
        r"\btime\s*\(.*h.*\)",
        r"\btime\s*\(.*hr.*\)",
        r"\btime\s*\(.*min.*\)",
        r"\bhours?\b",
        r"\bhrs?\b",
        r"\bdays?\b",
        r"\bminutes?\b",
        r"\bmins?\b"
    ]

    return any(
        re.search(
            p,
            x
        )
        for p in patterns
    )


def is_release_header(x):

    x = clean(x).lower()

    # Release must be explicit.
    patterns = [
        r"\brelease\b",
        r"\breleased\b",
        r"\bcumulative\s+release\b",
        r"\bdrug\s+release\b",
        r"\bdrug\s+released\b",
        r"\brelease\s*\(%",
        r"\brelease\s+percent",
        r"\brelease\s+percentage"
    ]

    return any(
        re.search(
            p,
            x
        )
        for p in patterns
    )


def contains_release_unit(x):

    x = clean(x).lower()

    return (
        "%" in x
        or "percent" in x
    )


# ============================================================
# FIND XML FILES
# ============================================================

xml_files = glob.glob(
    os.path.join(
        XML_DIR,
        "*.xml"
    )
)

print("=" * 70)
print("CODE 5C — CORRECTED STRUCTURE-AWARE EXTRACTION")
print("=" * 70)

print(
    "XML files:",
    len(xml_files)
)

# ============================================================
# EXTRACTION
# ============================================================

records = []

articles_with_release_tables = 0
tables_examined = 0
tables_accepted = 0

for article_no, xml_path in enumerate(
    xml_files,
    start=1
):

    filename = os.path.basename(
        xml_path
    )

    # --------------------------------------------------------
    # CRITICAL FIX:
    # PMCID comes from filename
    # --------------------------------------------------------

    pmcid = os.path.splitext(
        filename
    )[0]

    try:

        with open(
            xml_path,
            "r",
            encoding="utf-8",
            errors="ignore"
        ) as f:

            xml = f.read()

        soup = BeautifulSoup(
            xml,
            "xml"
        )

    except Exception:
        continue

    # --------------------------------------------------------
    # PMID / DOI
    # --------------------------------------------------------

    pmid = ""
    doi = ""

    for node in soup.find_all(
        "article-id"
    ):

        pub_type = clean(
            node.get(
                "pub-id-type",
                ""
            )
        ).lower()

        value = clean(
            node.get_text()
        )

        if pub_type == "pmid":
            pmid = value

        elif pub_type == "doi":
            doi = value

    # --------------------------------------------------------
    # TITLE
    # --------------------------------------------------------

    title = ""

    node = soup.find(
        "article-title"
    )

    if node:

        title = clean(
            node.get_text(
                " ",
                strip=True
            )
        )

    # --------------------------------------------------------
    # TABLES
    # --------------------------------------------------------

    tables = soup.find_all(
        "table-wrap"
    )

    article_has_release_table = False

    for table_index, table in enumerate(
        tables
    ):

        tables_examined += 1

        caption_node = table.find(
            "caption"
        )

        caption = ""

        if caption_node:

            caption = clean(
                caption_node.get_text(
                    " ",
                    strip=True
                )
            )

        label_node = table.find(
            "label"
        )

        label = ""

        if label_node:

            label = clean(
                label_node.get_text()
            )

        rows = table.find_all(
            "tr"
        )

        if len(rows) < 2:
            continue

        parsed = []

        for row in rows:

            cells = row.find_all(
                [
                    "th",
                    "td"
                ]
            )

            values = [
                clean(
                    c.get_text(
                        " ",
                        strip=True
                    )
                )
                for c in cells
            ]

            if values:

                parsed.append(
                    values
                )

        if len(parsed) < 2:
            continue

        # ----------------------------------------------------
        # SEARCH FIRST 10 ROWS FOR A REAL HEADER
        # ----------------------------------------------------

        header_idx = None
        time_col = None
        release_col = None

        for hidx, header in enumerate(
            parsed[:10]
        ):

            time_candidates = [
                i
                for i, cell in enumerate(
                    header
                )
                if is_time_header(cell)
            ]

            release_candidates = [
                i
                for i, cell in enumerate(
                    header
                )
                if is_release_header(cell)
            ]

            if (
                time_candidates
                and release_candidates
            ):

                header_idx = hidx

                time_col = time_candidates[0]

                release_col = release_candidates[0]

                break

        # ----------------------------------------------------
        # REJECT TABLE IF STRUCTURE IS AMBIGUOUS
        # ----------------------------------------------------

        if (
            header_idx is None
            or time_col is None
            or release_col is None
        ):
            continue

        article_has_release_table = True
        tables_accepted += 1

        # ----------------------------------------------------
        # EXTRACT ONLY ROW-ALIGNED VALUES
        # ----------------------------------------------------

        for row_index, row in enumerate(
            parsed[
                header_idx + 1:
            ],
            start=header_idx + 1
        ):

            if (
                time_col >= len(row)
                or release_col >= len(row)
            ):
                continue

            time_raw = row[
                time_col
            ]

            release_raw = row[
                release_col
            ]

            # ------------------------------------------------
            # REQUIRE EXPLICIT TIME UNIT
            # ------------------------------------------------

            time_h = time_to_hours(
                time_raw
            )

            if np.isnan(time_h):
                continue

            # ------------------------------------------------
            # RELEASE VALUE
            # ------------------------------------------------

            release_pct = number(
                release_raw
            )

            if np.isnan(release_pct):
                continue

            # ------------------------------------------------
            # RELEASE RANGE
            # ------------------------------------------------

            if (
                release_pct < 0
                or release_pct > 100
            ):
                continue

            # ------------------------------------------------
            # RECORD
            # ------------------------------------------------

            records.append({

                "candidate_id":
                    f"{pmcid}_T{table_index}_R{row_index}",

                "pmcid":
                    pmcid,

                "pmid":
                    pmid,

                "doi":
                    doi,

                "title":
                    title,

                "table_index":
                    table_index,

                "table_label":
                    label,

                "table_caption":
                    caption,

                "row_index":
                    row_index,

                "time_raw":
                    time_raw,

                "release_raw":
                    release_raw,

                "release_time_h":
                    time_h,

                "release_percent":
                    release_pct,

                "source_xml":
                    filename,

                "source_type":
                    "PMC_XML_TABLE",

                "extraction_status":
                    "CANDIDATE",

                "validation_status":
                    "NOT_YET_VALIDATED"

            })

    if article_has_release_table:

        articles_with_release_tables += 1

    if (
        article_no % 100 == 0
        or article_no == len(xml_files)
    ):

        print(
            f"Processed "
            f"{article_no:,}/{len(xml_files):,} | "
            f"Candidates: {len(records):,}"
        )

# ============================================================
# DATAFRAME
# ============================================================

df = pd.DataFrame(
    records
)

if df.empty:

    print("\nNO STRUCTURED OBSERVATIONS FOUND.")

else:

    # --------------------------------------------------------
    # Remove exact duplicates
    # --------------------------------------------------------

    df = df.drop_duplicates(
        subset=[
            "pmcid",
            "table_index",
            "row_index",
            "release_time_h",
            "release_percent"
        ]
    ).reset_index(
        drop=True
    )

    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    df.to_csv(
        OUTPUT,
        index=False
    )

    # ========================================================
    # SUMMARY
    # ========================================================

    print("\n" + "=" * 70)
    print("CODE 5C COMPLETE")
    print("=" * 70)

    print(
        "XML articles processed:",
        len(xml_files)
    )

    print(
        "Articles containing accepted release tables:",
        articles_with_release_tables
    )

    print(
        "Tables examined:",
        tables_examined
    )

    print(
        "Tables accepted:",
        tables_accepted
    )

    print(
        "Candidate observations:",
        len(df)
    )

    print(
        "Unique PMCID:",
        df["pmcid"].nunique()
    )

    print(
        "Unique DOI:",
        df["doi"].replace(
            "",
            np.nan
        ).nunique()
    )

    print(
        "Unique source tables:",
        df[
            [
                "pmcid",
                "table_index"
            ]
        ].drop_duplicates().shape[0]
    )

    print("\nRelease statistics:")

    print(
        df[
            "release_percent"
        ].describe()
    )

    print("\nTime statistics:")

    print(
        df[
            "release_time_h"
        ].describe()
    )

    print("\nPreview:")

    display(
        df[
            [
                "candidate_id",
                "pmcid",
                "doi",
                "title",
                "table_index",
                "table_caption",
                "time_raw",
                "release_raw",
                "release_time_h",
                "release_percent"
            ]
        ].head(20)
    )

    print(
        "\nSaved:"
    )

    print(
        OUTPUT
    )

    print("\nIMPORTANT:")
    print(
        "These are candidate observations only."
    )
    print(
        "They are NOT yet validated."
    )
    print(
        "Do NOT run ML."
    )
```

```python
# ============================================================
# CODE 5D — STRICT SOURCE-TRACEABLE DRUG-RELEASE EXTRACTION
# ============================================================

import os
import re
import glob
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup

ROOT = "/content/AI_Polymer_Drug_Release"

XML_DIR = os.path.join(
    ROOT,
    "data",
    "raw",
    "pmc_xml"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "candidates"
)

os.makedirs(OUT_DIR, exist_ok=True)

OUTPUT = os.path.join(
    OUT_DIR,
    "05D_strict_release_candidates.csv"
)

print("=" * 70)
print("CODE 5D — STRICT SOURCE-TRACEABLE RELEASE EXTRACTION")
print("=" * 70)

# ============================================================
# BASIC HELPERS
# ============================================================

def clean(x):

    if x is None:
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


def first_number(x):

    x = clean(x)

    m = re.search(
        r"(?<![A-Za-z])[-+]?\d+(?:\.\d+)?",
        x
    )

    if not m:
        return np.nan

    try:
        return float(m.group())
    except:
        return np.nan


def explicit_time(x):

    """
    Convert only explicitly unit-labelled time.
    Bare numbers are NOT accepted as time.
    """

    x = clean(x).lower()

    # Reject negative times
    if re.search(r"-\s*\d", x):
        return np.nan

    m = re.search(
        r"(?<![\w.])(\d+(?:\.\d+)?)\s*"
        r"(minutes?|mins?|min|hours?|hrs?|hr|h|days?|day)\b",
        x
    )

    if not m:
        return np.nan

    value = float(m.group(1))
    unit = m.group(2)

    if unit.startswith("min"):
        return value / 60.0

    if unit.startswith("h"):
        return value

    if unit.startswith("d"):
        return value * 24.0

    return np.nan


def explicit_release(x):

    """
    Extract only an explicitly percentage-labelled value.
    """

    x = clean(x).lower()

    matches = re.findall(
        r"(?<![\w.])(\d+(?:\.\d+)?)\s*%",
        x
    )

    if not matches:
        return np.nan

    # Use the first explicit percentage
    value = float(matches[0])

    if value < 0 or value > 100:
        return np.nan

    return value


def has_release_word(x):

    x = clean(x).lower()

    return bool(
        re.search(
            r"\b(?:drug\s+)?"
            r"(?:cumulative\s+)?"
            r"release(?:d)?\b",
            x
        )
    )


def has_time_word(x):

    x = clean(x).lower()

    return bool(
        re.search(
            r"\b(?:time|hour|hours|hr|hrs|"
            r"day|days|minute|minutes|min)\b",
            x
        )
    )


def time_header(x):

    x = clean(x).lower()

    return bool(
        re.search(
            r"\btime\b"
            r"|time\s*\("
            r"|\bhours?\b"
            r"|\bhrs?\b"
            r"|\bhr\b"
            r"|\bminutes?\b"
            r"|\bmins?\b"
            r"|\bdays?\b"
            r"|\bday\b",
            x
        )
    )


def release_header(x):

    x = clean(x).lower()

    return bool(
        re.search(
            r"\brelease\b"
            r"|\breleased\b"
            r"|\bcumulative\s+release\b"
            r"|\bdrug\s+release\b"
            r"|\brelease\s*\(%"
            r"|\brelease\s+percent",
            x
        )
    )


# ============================================================
# FIND XML
# ============================================================

xml_files = glob.glob(
    os.path.join(
        XML_DIR,
        "*.xml"
    )
)

print(
    "XML files:",
    len(xml_files)
)

if not xml_files:

    raise FileNotFoundError(
        XML_DIR
    )


# ============================================================
# EXTRACTION
# ============================================================

records = []

stats = {
    "articles": 0,
    "tables": 0,
    "release_tables": 0,
    "structured_tables": 0,
    "same_cell_candidates": 0,
    "column_candidates": 0
}


for article_no, xml_path in enumerate(
    xml_files,
    start=1
):

    stats["articles"] += 1

    filename = os.path.basename(
        xml_path
    )

    # --------------------------------------------------------
    # PMCID FROM FILENAME
    # --------------------------------------------------------

    pmcid = os.path.splitext(
        filename
    )[0]

    try:

        with open(
            xml_path,
            "r",
            encoding="utf-8",
            errors="ignore"
        ) as f:

            xml = f.read()

        soup = BeautifulSoup(
            xml,
            "xml"
        )

    except Exception:

        continue


    # --------------------------------------------------------
    # PMID / DOI
    # --------------------------------------------------------

    pmid = ""
    doi = ""

    for node in soup.find_all(
        "article-id"
    ):

        typ = clean(
            node.get(
                "pub-id-type",
                ""
            )
        ).lower()

        value = clean(
            node.get_text()
        )

        if typ == "pmid":
            pmid = value

        elif typ == "doi":
            doi = value


    # --------------------------------------------------------
    # TITLE
    # --------------------------------------------------------

    title = ""

    node = soup.find(
        "article-title"
    )

    if node:

        title = clean(
            node.get_text(
                " ",
                strip=True
            )
        )


    # ========================================================
    # TABLES
    # ========================================================

    tables = soup.find_all(
        "table-wrap"
    )

    for table_index, table in enumerate(
        tables
    ):

        stats["tables"] += 1

        caption = ""

        caption_node = table.find(
            "caption"
        )

        if caption_node:

            caption = clean(
                caption_node.get_text(
                    " ",
                    strip=True
                )
            )

        label = ""

        label_node = table.find(
            "label"
        )

        if label_node:

            label = clean(
                label_node.get_text()
            )


        # ----------------------------------------------------
        # TABLE TEXT
        # ----------------------------------------------------

        table_text = clean(
            table.get_text(
                " ",
                strip=True
            )
        ).lower()

        if not has_release_word(
            table_text
        ):

            continue

        stats[
            "release_tables"
        ] += 1


        # ----------------------------------------------------
        # PARSE ROWS
        # ----------------------------------------------------

        rows = table.find_all(
            "tr"
        )

        parsed = []

        for row in rows:

            cells = row.find_all(
                [
                    "th",
                    "td"
                ]
            )

            vals = [
                clean(
                    c.get_text(
                        " ",
                        strip=True
                    )
                )
                for c in cells
            ]

            if vals:
                parsed.append(vals)


        if len(parsed) < 2:
            continue


        # ====================================================
        # METHOD A:
        # STRUCTURED TIME + RELEASE COLUMNS
        # ====================================================

        header_idx = None
        time_col = None
        release_col = None

        for i, row in enumerate(
            parsed[:10]
        ):

            tcols = [
                j
                for j, cell in enumerate(row)
                if time_header(cell)
            ]

            rcols = [
                j
                for j, cell in enumerate(row)
                if release_header(cell)
            ]

            if (
                tcols
                and rcols
            ):

                header_idx = i
                time_col = tcols[0]
                release_col = rcols[0]

                break


        if (
            header_idx is not None
            and time_col is not None
            and release_col is not None
        ):

            stats[
                "structured_tables"
            ] += 1

            stats[
                "column_candidates"
            ] += 1


            for row_index, row in enumerate(
                parsed[
                    header_idx + 1:
                ],
                start=header_idx + 1
            ):

                if (
                    time_col >= len(row)
                    or release_col >= len(row)
                ):
                    continue

                time_raw = row[
                    time_col
                ]

                release_raw = row[
                    release_col
                ]


                # --------------------------------------------
                # TIME
                # --------------------------------------------

                time_h = explicit_time(
                    time_raw
                )

                if np.isnan(time_h):
                    continue

                if time_h < 0:
                    continue


                # --------------------------------------------
                # RELEASE
                # --------------------------------------------

                release_pct = explicit_release(
                    release_raw
                )

                # A plain number is accepted ONLY when
                # the release header itself explicitly
                # indicates percentage/release.
                if np.isnan(release_pct):

                    if (
                        release_header(
                            row[release_col]
                        )
                        and "%" in
                        clean(
                            parsed[
                                header_idx
                            ][release_col]
                        )
                    ):

                        release_pct = (
                            first_number(
                                release_raw
                            )
                        )

                    else:
                        continue


                if np.isnan(
                    release_pct
                ):
                    continue

                if (
                    release_pct < 0
                    or release_pct > 100
                ):
                    continue


                records.append({

                    "candidate_id":
                        f"{pmcid}_T{table_index}_R{row_index}",

                    "pmcid":
                        pmcid,

                    "pmid":
                        pmid,

                    "doi":
                        doi,

                    "title":
                        title,

                    "table_index":
                        table_index,

                    "table_label":
                        label,

                    "table_caption":
                        caption,

                    "row_index":
                        row_index,

                    "time_raw":
                        time_raw,

                    "release_raw":
                        release_raw,

                    "release_time_h":
                        time_h,

                    "release_percent":
                        release_pct,

                    "extraction_method":
                        "STRUCTURED_TIME_RELEASE_COLUMNS",

                    "source_xml":
                        filename,

                    "validation_status":
                        "CANDIDATE"

                })


        # ====================================================
        # METHOD B:
        # SAME-CELL EXPLICIT RELEASE STATEMENT
        # ====================================================

        for row_index, row in enumerate(
            parsed
        ):

            row_text = " | ".join(
                row
            )

            # Must explicitly mention release
            if not has_release_word(
                row_text
            ):
                continue

            # Must contain BOTH time and percentage
            if not has_time_word(
                row_text
            ):
                continue

            release_pct = explicit_release(
                row_text
            )

            time_h = explicit_time(
                row_text
            )

            if np.isnan(
                release_pct
            ):
                continue

            if np.isnan(
                time_h
            ):
                continue

            if time_h < 0:
                continue

            # ------------------------------------------------
            # Avoid duplicates with structured method
            # ------------------------------------------------

            duplicate = any(
                r["pmcid"] == pmcid
                and r["table_index"] == table_index
                and r["row_index"] == row_index
                for r in records
            )

            if duplicate:
                continue

            stats[
                "same_cell_candidates"
            ] += 1

            records.append({

                "candidate_id":
                    f"{pmcid}_T{table_index}_R{row_index}",

                "pmcid":
                    pmcid,

                "pmid":
                    pmid,

                "doi":
                    doi,

                "title":
                    title,

                "table_index":
                    table_index,

                "table_label":
                    label,

                "table_caption":
                    caption,

                "row_index":
                    row_index,

                "time_raw":
                    row_text,

                "release_raw":
                    row_text,

                "release_time_h":
                    time_h,

                "release_percent":
                    release_pct,

                "extraction_method":
                    "EXPLICIT_SAME_ROW_RELEASE_TIME",

                "source_xml":
                    filename,

                "validation_status":
                    "CANDIDATE"

            })


    if (
        article_no % 100 == 0
        or article_no == len(xml_files)
    ):

        print(
            f"Processed {article_no:,}/{len(xml_files):,} | "
            f"Candidates: {len(records):,}"
        )


# ============================================================
# DATAFRAME
# ============================================================

df = pd.DataFrame(
    records
)

if df.empty:

    print(
        "\nNo candidates extracted."
    )

else:

    # --------------------------------------------------------
    # REMOVE DUPLICATES
    # --------------------------------------------------------

    df = df.drop_duplicates(
        subset=[
            "pmcid",
            "table_index",
            "row_index",
            "release_time_h",
            "release_percent"
        ]
    ).reset_index(
        drop=True
    )


    # --------------------------------------------------------
    # HARD VALIDATION
    # --------------------------------------------------------

    df = df[
        df["release_time_h"].between(
            0,
            10000
        )
    ]

    df = df[
        df["release_percent"].between(
            0,
            100
        )
    ]

    df = df.reset_index(
        drop=True
    )


    # --------------------------------------------------------
    # SAVE
    # --------------------------------------------------------

    df.to_csv(
        OUTPUT,
        index=False
    )


    # ========================================================
    # SUMMARY
    # ========================================================

    print("\n" + "=" * 70)
    print("CODE 5D COMPLETE")
    print("=" * 70)

    print(
        "Articles processed:",
        stats["articles"]
    )

    print(
        "Tables examined:",
        stats["tables"]
    )

    print(
        "Release-related tables:",
        stats["release_tables"]
    )

    print(
        "Structured tables:",
        stats["structured_tables"]
    )

    print(
        "Column-based candidates:",
        stats["column_candidates"]
    )

    print(
        "Same-row candidates:",
        stats["same_cell_candidates"]
    )

    print(
        "Final candidates:",
        len(df)
    )

    print(
        "Unique PMCID:",
        df["pmcid"].nunique()
    )

    print(
        "Unique DOI:",
        df["doi"].replace(
            "",
            np.nan
        ).nunique()
    )

    print(
        "Unique source tables:",
        df[
            [
                "pmcid",
                "table_index"
            ]
        ].drop_duplicates().shape[0]
    )

    print("\nExtraction methods:")

    print(
        df[
            "extraction_method"
        ].value_counts()
    )

    print("\nRelease statistics:")

    print(
        df[
            "release_percent"
        ].describe()
    )

    print("\nTime statistics:")

    print(
        df[
            "release_time_h"
        ].describe()
    )

    print("\nPreview:")

    display(
        df[
            [
                "candidate_id",
                "pmcid",
                "doi",
                "title",
                "table_index",
                "table_caption",
                "time_raw",
                "release_raw",
                "release_time_h",
                "release_percent",
                "extraction_method"
            ]
        ].head(30)
    )

    print(
        "\nSaved:"
    )

    print(
        OUTPUT
    )

    print(
        "\nIMPORTANT:"
    )

    print(
        "Candidate observations only."
    )

    print(
        "No synthetic values."
    )

    print(
        "No interpolation."
    )

    print(
        "No ML."
    )
```

```python
# ============================================================
# CODE 5E — SEMANTIC RELEASE VALIDATION & FALSE-POSITIVE REJECTION
# ============================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 5E — SEMANTIC RELEASE VALIDATION & FALSE-POSITIVE REJECTION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

input_file = os.path.join(
    ROOT,
    "data",
    "candidates",
    "05D_strict_release_candidates.csv"
)

output_dir = os.path.join(
    ROOT,
    "data",
    "validation"
)

os.makedirs(output_dir, exist_ok=True)

if not os.path.exists(input_file):
    raise FileNotFoundError(
        f"\nCode 5D output not found:\n{input_file}\n\n"
        "Run Code 5D first."
    )

df = pd.read_csv(input_file)

print(f"Input candidates: {len(df):,}")
print("Columns:")
print(list(df.columns))


# ============================================================
# COLUMN DETECTION
# ============================================================

def find_col(columns, candidates):
    lower_map = {str(c).lower(): c for c in columns}

    for x in candidates:
        if x.lower() in lower_map:
            return lower_map[x.lower()]

    for c in columns:
        cl = str(c).lower()
        for x in candidates:
            if x.lower() in cl:
                return c

    return None


text_columns = [
    c for c in [
        find_col(
            df.columns,
            ["title", "article_title"]
        ),
        find_col(
            df.columns,
            ["table_caption", "caption"]
        ),
        find_col(
            df.columns,
            ["time_raw", "release_time_raw"]
        ),
        find_col(
            df.columns,
            ["release_raw", "release_text"]
        ),
        find_col(
            df.columns,
            ["row_text", "full_table_text"]
        )
    ]
    if c is not None
]

text_columns = list(dict.fromkeys(text_columns))

print("\nText columns used for semantic validation:")
print(text_columns)


# ============================================================
# COMBINE SOURCE TEXT
# ============================================================

def combine_text(row):

    parts = []

    for c in text_columns:
        value = row.get(c, "")

        if pd.notna(value):
            parts.append(str(value))

    return " | ".join(parts).lower()


df["_source_text"] = df.apply(combine_text, axis=1)


# ============================================================
# EXCLUSION RULES
# ============================================================

# These terms commonly indicate a percentage that is NOT
# cumulative drug release.

non_release_patterns = {

    "ENTRAPMENT_EFFICIENCY":
        r"\b(entrapment efficiency|encapsulation efficiency)\b",

    "DRUG_LOADING":
        r"\b(drug loading|loading efficiency|loading capacity)\b",

    "DEE":
        r"\bdee\b",

    "CMAX":
        r"\bcmax\b",

    "AUC":
        r"\bauc\b",

    "IC50":
        r"\bic50\b",

    "HALF_LIFE":
        r"\b(half[- ]life|t1/2|t½)\b",

    "PARTICLE_SIZE":
        r"\b(particle size|diameter|nm\b|nanometer)\b",

    "PDI":
        r"\b(pdi|polydispersity index)\b",

    "YIELD":
        r"\b(yield|production yield)\b",

    "RECOVERY":
        r"\brecovery\b",

    "BIOAVAILABILITY":
        r"\bbioavailability\b",

    "DEGRADATION":
        r"\bdegradation\b",

    "SWELLING":
        r"\bswelling\b",

    "PERMEABILITY":
        r"\bpermeability\b",

    "CELL_VIABILITY":
        r"\b(cell viability|viability)\b",

    "INHIBITION":
        r"\b(inhibition|inhibition zone)\b",

    "TOXICITY":
        r"\btoxicity\b",

    "CITATION_YEAR_CONTEXT":
        r"\b(19|20)\d{2}\b"
}


# ============================================================
# RELEASE-SPECIFIC POSITIVE SIGNALS
# ============================================================

release_patterns = [

    r"\bcumulative drug release\b",
    r"\bcumulative release\b",
    r"\bdrug release\b",
    r"\bdrug released\b",
    r"\brelease percentage\b",
    r"\brelease \(%\)",
    r"\breleased \(%\)",
    r"\brelease of\b",
    r"\breleased\b",
    r"\bin vitro release\b",
    r"\brelease profile\b",
    r"\brelease kinetics\b",
    r"\bdrug was released\b",
    r"\bdrug release study\b"
]


# ============================================================
# TIME VALIDATION
# ============================================================

def suspicious_time(text, time_value):

    if pd.isna(time_value):
        return True

    try:
        t = float(time_value)
    except:
        return True

    # Negative time is impossible.
    if t < 0:
        return True

    # Very large values can be legitimate, so don't reject
    # automatically. We only flag obvious year-like values.
    if 1900 <= t <= 2100:

        # If the source contains a year context, reject.
        return True

    return False


# ============================================================
# RELEASE VALUE VALIDATION
# ============================================================

def extract_percentages(text):

    if not text:
        return []

    # Match numbers followed by %
    vals = re.findall(
        r"(?<![\w.])(\d+(?:\.\d+)?)\s*%",
        text
    )

    result = []

    for v in vals:
        try:
            x = float(v)

            if 0 <= x <= 100:
                result.append(x)

        except:
            pass

    return result


# ============================================================
# MEAN ± SD PROTECTION
# ============================================================

def detect_sd_only_error(text, release_value):

    if not text:
        return False

    # Example:
    # 78.32 ± 0.80%
    # The release value should be 78.32, not 0.80.

    pattern = re.search(
        r"(\d+(?:\.\d+)?)\s*[±+/-]\s*"
        r"(\d+(?:\.\d+)?)\s*%",
        text
    )

    if not pattern:
        return False

    mean_value = float(pattern.group(1))
    sd_value = float(pattern.group(2))

    try:
        rv = float(release_value)
    except:
        return False

    # If extracted value equals SD but not mean,
    # flag it for manual correction/rejection.
    if abs(rv - sd_value) < 1e-9 and abs(rv - mean_value) > 1e-9:
        return True

    return False


# ============================================================
# YEAR / TIME CONTEXT
# ============================================================

def year_as_time_error(text, time_value):

    try:
        t = float(time_value)
    except:
        return True

    if not (1900 <= t <= 2100):
        return False

    # A year should never be interpreted as release time.
    return True


# ============================================================
# VALIDATION LOOP
# ============================================================

results = []

for idx, row in df.iterrows():

    text = row["_source_text"]

    try:
        time_value = float(row["release_time_h"])
    except:
        time_value = np.nan

    try:
        release_value = float(row["release_percent"])
    except:
        release_value = np.nan

    reasons = []
    warnings = []

    # --------------------------------------------------------
    # Basic numerical checks
    # --------------------------------------------------------

    if pd.isna(time_value):
        reasons.append("MISSING_TIME")

    elif time_value < 0:
        reasons.append("NEGATIVE_TIME")

    if pd.isna(release_value):
        reasons.append("MISSING_RELEASE")

    elif release_value < 0 or release_value > 100:
        reasons.append("RELEASE_OUTSIDE_0_100")


    # --------------------------------------------------------
    # Year detection
    # --------------------------------------------------------

    if year_as_time_error(text, time_value):
        reasons.append("YEAR_MISINTERPRETED_AS_TIME")


    # --------------------------------------------------------
    # False-positive semantic indicators
    # --------------------------------------------------------

    matched_non_release = []

    for label, pattern in non_release_patterns.items():

        if re.search(pattern, text, flags=re.I):

            matched_non_release.append(label)

    # We don't automatically reject every row containing
    # "particle size", etc., because a legitimate release
    # table may contain multiple characteristics.
    #
    # Strong exclusion terms are treated as rejection signals.

    strong_exclusions = {
        "ENTRAPMENT_EFFICIENCY",
        "DRUG_LOADING",
        "DEE",
        "CMAX",
        "AUC",
        "IC50",
        "HALF_LIFE",
        "INHIBITION",
        "CELL_VIABILITY"
    }

    strong_hits = [
        x for x in matched_non_release
        if x in strong_exclusions
    ]

    if strong_hits:

        reasons.append(
            "NON_RELEASE_OUTCOME:" + ",".join(strong_hits)
        )


    # --------------------------------------------------------
    # Release-specific signal
    # --------------------------------------------------------

    release_signal = any(
        re.search(pattern, text, flags=re.I)
        for pattern in release_patterns
    )

    if not release_signal:

        warnings.append(
            "NO_EXPLICIT_RELEASE_SEMANTIC_SIGNAL"
        )


    # --------------------------------------------------------
    # SD / mean error
    # --------------------------------------------------------

    if detect_sd_only_error(text, release_value):

        reasons.append(
            "EXTRACTED_SD_INSTEAD_OF_RELEASE_MEAN"
        )


    # --------------------------------------------------------
    # Explicit "within X time" without release %
    # --------------------------------------------------------

    if re.search(
        r"\brelease(?:d)?\s+(?:within|in)\s+"
        r"\d+(?:\.\d+)?\s*(?:min|minutes|h|hr|hours|day|days)",
        text,
        flags=re.I
    ):

        if not re.search(
            r"\d+(?:\.\d+)?\s*%",
            text
        ):

            reasons.append(
                "RELEASE_TIME_WITHOUT_QUANTITATIVE_RELEASE_PERCENT"
            )


    # --------------------------------------------------------
    # Final decision
    # --------------------------------------------------------

    if len(reasons) > 0:

        decision = "REJECT"

    elif not release_signal:

        decision = "MANUAL_REVIEW"

    else:

        decision = "PASS_CANDIDATE"


    results.append({

        "candidate_id":
            row.get("candidate_id", idx),

        "pmcid":
            row.get("pmcid", ""),

        "doi":
            row.get("doi", ""),

        "title":
            row.get("title", ""),

        "table_index":
            row.get("table_index", ""),

        "table_caption":
            row.get("table_caption", ""),

        "release_time_h":
            time_value,

        "release_percent":
            release_value,

        "release_signal":
            release_signal,

        "non_release_flags":
            ";".join(matched_non_release),

        "validation_decision":
            decision,

        "rejection_reason":
            ";".join(reasons),

        "validation_warning":
            ";".join(warnings),

        "source_trace":
            row.get("candidate_id", idx)

    })


validated = pd.DataFrame(results)


# ============================================================
# SAVE AUDIT FILES
# ============================================================

all_file = os.path.join(
    output_dir,
    "05E_semantic_validation_audit.csv"
)

passed_file = os.path.join(
    output_dir,
    "05E_pass_candidates.csv"
)

manual_file = os.path.join(
    output_dir,
    "05E_manual_review_candidates.csv"
)

rejected_file = os.path.join(
    output_dir,
    "05E_rejected_candidates.csv"
)

validated.to_csv(all_file, index=False)

passed = validated[
    validated["validation_decision"] == "PASS_CANDIDATE"
].copy()

manual = validated[
    validated["validation_decision"] == "MANUAL_REVIEW"
].copy()

rejected = validated[
    validated["validation_decision"] == "REJECT"
].copy()

passed.to_csv(passed_file, index=False)
manual.to_csv(manual_file, index=False)
rejected.to_csv(rejected_file, index=False)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CODE 5E COMPLETE")
print("=" * 70)

print(f"Input candidates:          {len(df):,}")
print(f"PASS candidates:           {len(passed):,}")
print(f"MANUAL REVIEW:             {len(manual):,}")
print(f"REJECTED:                  {len(rejected):,}")

print("\nValidation distribution:")
print(
    validated["validation_decision"]
    .value_counts()
)

print("\nTop rejection reasons:")

reasons = (
    validated.loc[
        validated["rejection_reason"].astype(str).str.len() > 0,
        "rejection_reason"
    ]
    .str.split(";")
    .explode()
    .value_counts()
)

print(reasons.head(20))


print("\nSource diversity among PASS candidates:")

print(
    "Unique PMCID:",
    passed["pmcid"].nunique()
)

print(
    "Unique DOI:",
    passed["doi"].replace(
        {"": np.nan}
    ).dropna().nunique()
)

print(
    "Unique source tables:",
    passed[
        ["pmcid", "table_index"]
    ]
    .drop_duplicates()
    .shape[0]
)


print("\nSaved:")
print(all_file)
print(passed_file)
print(manual_file)
print(rejected_file)

print("\nIMPORTANT:")
print("PASS candidates are still not the final analytical dataset.")
print("They require source-level validation before ML or statistical modeling.")
print("No synthetic observations were generated.")
print("No interpolation was performed.")
print("No observations were duplicated.")
print("No ML was performed.")
```

```python
# ============================================================
# CODE 5F — SOURCE-LEVEL RELEASE VALIDATION
# ============================================================

import os
import re
import glob
import pandas as pd
import numpy as np
import xml.etree.ElementTree as ET

print("=" * 70)
print("CODE 5F — SOURCE-LEVEL RELEASE VALIDATION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

PASS_FILE = os.path.join(
    ROOT, "data", "validation", "05E_pass_candidates.csv"
)

MANUAL_FILE = os.path.join(
    ROOT, "data", "validation", "05E_manual_review_candidates.csv"
)

XML_DIR = os.path.join(
    ROOT, "data", "raw", "pmc_xml"
)

OUT_DIR = os.path.join(
    ROOT, "data", "validation"
)

os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(PASS_FILE):
    raise FileNotFoundError(
        f"PASS file not found:\n{PASS_FILE}\n\n"
        "Run Code 5E first."
    )

if not os.path.exists(XML_DIR):
    raise FileNotFoundError(
        f"XML directory not found:\n{XML_DIR}"
    )

passed = pd.read_csv(PASS_FILE)

if os.path.exists(MANUAL_FILE):
    manual = pd.read_csv(MANUAL_FILE)
else:
    manual = pd.DataFrame()

print(f"PASS candidates:   {len(passed):,}")
print(f"Manual candidates: {len(manual):,}")


# ============================================================
# XML HELPERS
# ============================================================

def local_name(tag):
    """Remove XML namespace."""
    if "}" in tag:
        return tag.split("}", 1)[1]
    return tag


def clean_text(element):

    if element is None:
        return ""

    text = " ".join(
        " ".join(element.itertext()).split()
    )

    return text.strip()


def find_xml_for_pmcid(pmcid):

    if pd.isna(pmcid):
        return None

    pmcid = str(pmcid).strip()

    if not pmcid:
        return None

    direct = os.path.join(
        XML_DIR,
        f"{pmcid}.xml"
    )

    if os.path.exists(direct):
        return direct

    matches = glob.glob(
        os.path.join(XML_DIR, f"*{pmcid}*.xml")
    )

    if matches:
        return matches[0]

    return None


# ============================================================
# TABLE EXTRACTION FROM ORIGINAL XML
# ============================================================

def get_tables(xml_path):

    try:
        tree = ET.parse(xml_path)
        root = tree.getroot()
    except Exception:
        return []

    tables = []

    for elem in root.iter():

        if local_name(elem.tag) != "table-wrap":
            continue

        caption = ""

        for child in elem.iter():
            if local_name(child.tag) == "caption":
                caption = clean_text(child)
                break

        table_element = None

        for child in elem.iter():
            if local_name(child.tag) == "table":
                table_element = child
                break

        if table_element is None:
            continue

        rows = []

        for tr in table_element.iter():

            if local_name(tr.tag) != "tr":
                continue

            cells = []

            for cell in tr:

                name = local_name(cell.tag)

                if name not in ("td", "th"):
                    continue

                cells.append(clean_text(cell))

            if cells:
                rows.append(cells)

        tables.append({
            "caption": caption,
            "rows": rows
        })

    return tables


# ============================================================
# RELEASE SEMANTIC SIGNALS
# ============================================================

release_terms = [
    r"\bcumulative drug release\b",
    r"\bcumulative release\b",
    r"\bdrug release\b",
    r"\bdrug released\b",
    r"\brelease profile\b",
    r"\brelease kinetics\b",
    r"\bin vitro release\b",
    r"\bamount released\b",
    r"\bpercentage released\b",
    r"\bpercent released\b",
    r"\brelease \(%\)",
    r"\breleased \(%\)"
]

non_release_terms = [
    r"\bentrapment efficiency\b",
    r"\bencapsulation efficiency\b",
    r"\bdrug loading\b",
    r"\bloading efficiency\b",
    r"\bcell viability\b",
    r"\binhibition zone\b",
    r"\bhalf[- ]life\b",
    r"\bauc\b",
    r"\bcmax\b",
    r"\bic50\b",
    r"\bparticle size\b",
    r"\bpolydispersity\b",
    r"\bpdi\b"
]


def has_release_signal(text):

    text = str(text).lower()

    return any(
        re.search(p, text)
        for p in release_terms
    )


def has_non_release_signal(text):

    text = str(text).lower()

    return any(
        re.search(p, text)
        for p in non_release_terms
    )


# ============================================================
# NUMERIC EXTRACTION
# ============================================================

def percent_values(text):

    if not text:
        return []

    matches = re.findall(
        r"(?<![\w.])"
        r"(\d+(?:\.\d+)?)"
        r"\s*%",
        str(text)
    )

    vals = []

    for x in matches:

        try:
            x = float(x)

            if 0 <= x <= 100:
                vals.append(x)

        except:
            pass

    return vals


def numeric_values(text):

    if not text:
        return []

    matches = re.findall(
        r"(?<![\w.])"
        r"(\d+(?:\.\d+)?)",
        str(text)
    )

    vals = []

    for x in matches:

        try:
            vals.append(float(x))
        except:
            pass

    return vals


# ============================================================
# LOCATE ORIGINAL TABLE
# ============================================================

def locate_table(xml_path, table_index):

    tables = get_tables(xml_path)

    try:
        ti = int(table_index)
    except:
        ti = None

    if ti is not None and 0 <= ti < len(tables):

        return tables[ti], ti

    # fallback: if exact index unavailable, search all tables
    return None, None


# ============================================================
# VALIDATE ONE CANDIDATE
# ============================================================

def validate_candidate(row):

    pmcid = row.get("pmcid", "")
    table_index = row.get("table_index", "")

    xml_path = find_xml_for_pmcid(pmcid)

    base = {
        "candidate_id":
            row.get("candidate_id", ""),

        "pmcid":
            pmcid,

        "doi":
            row.get("doi", ""),

        "title":
            row.get("title", ""),

        "table_index":
            table_index,

        "candidate_time_h":
            row.get("release_time_h", np.nan),

        "candidate_release_percent":
            row.get("release_percent", np.nan),

        "xml_found":
            bool(xml_path),

        "validation_status":
            "AMBIGUOUS",

        "validation_reason":
            "",

        "source_table_caption":
            "",

        "source_table_text":
            "",

        "source_release_signal":
            False,

        "source_non_release_signal":
            False,

        "source_percent_values":
            "",

        "source_numeric_values":
            ""
    }

    if xml_path is None:

        base["validation_reason"] = (
            "SOURCE_XML_NOT_FOUND"
        )

        return base

    table, actual_index = locate_table(
        xml_path,
        table_index
    )

    if table is None:

        base["validation_reason"] = (
            "SOURCE_TABLE_NOT_FOUND"
        )

        return base

    caption = table["caption"]

    rows = table["rows"]

    table_text = " | ".join(
        " | ".join(r)
        for r in rows
    )

    full_text = (
        caption + " | " + table_text
    )

    base["source_table_caption"] = caption
    base["source_table_text"] = table_text

    base["source_release_signal"] = (
        has_release_signal(full_text)
    )

    base["source_non_release_signal"] = (
        has_non_release_signal(full_text)
    )

    pcts = percent_values(full_text)
    nums = numeric_values(full_text)

    base["source_percent_values"] = ";".join(
        map(str, pcts)
    )

    base["source_numeric_values"] = ";".join(
        map(str, nums)
    )

    # --------------------------------------------------------
    # Numerical sanity
    # --------------------------------------------------------

    try:
        candidate_time = float(
            row.get("release_time_h", np.nan)
        )
    except:
        candidate_time = np.nan

    try:
        candidate_release = float(
            row.get("release_percent", np.nan)
        )
    except:
        candidate_release = np.nan

    if pd.isna(candidate_time):

        base["validation_status"] = "REJECT"
        base["validation_reason"] = "MISSING_TIME"

        return base

    if candidate_time < 0:

        base["validation_status"] = "REJECT"
        base["validation_reason"] = "NEGATIVE_TIME"

        return base

    if pd.isna(candidate_release):

        base["validation_status"] = "REJECT"
        base["validation_reason"] = "MISSING_RELEASE"

        return base

    if not (0 <= candidate_release <= 100):

        base["validation_status"] = "REJECT"
        base["validation_reason"] = (
            "RELEASE_OUTSIDE_0_100"
        )

        return base

    # --------------------------------------------------------
    # Strong source-level release evidence
    # --------------------------------------------------------

    release_signal = base["source_release_signal"]
    non_release_signal = base["source_non_release_signal"]

    # Check whether candidate value actually appears in
    # the original source table as a percentage.
    value_match = False

    for p in pcts:

        if abs(
            float(p) - candidate_release
        ) < 1e-6:

            value_match = True
            break

    # --------------------------------------------------------
    # Decision logic
    # --------------------------------------------------------

    if not release_signal:

        base["validation_status"] = "MANUAL_REVIEW"

        base["validation_reason"] = (
            "NO_CLEAR_SOURCE_RELEASE_SIGNAL"
        )

        return base

    if non_release_signal and not release_signal:

        base["validation_status"] = "REJECT"

        base["validation_reason"] = (
            "SOURCE_TABLE_APPEARS_NON_RELEASE"
        )

        return base

    if not value_match:

        base["validation_status"] = "MANUAL_REVIEW"

        base["validation_reason"] = (
            "CANDIDATE_RELEASE_VALUE_NOT_FOUND_AS_PERCENT_IN_SOURCE_TABLE"
        )

        return base

    # --------------------------------------------------------
    # Strongest automated pass
    # --------------------------------------------------------

    base["validation_status"] = "VALID_RELEASE"

    base["validation_reason"] = (
        "SOURCE_TABLE_CONTAINS_RELEASE_SIGNAL_AND_EXACT_PERCENT_VALUE"
    )

    return base


# ============================================================
# RUN SOURCE VALIDATION
# ============================================================

results = []

for i, (_, row) in enumerate(
    passed.iterrows(),
    start=1
):

    results.append(
        validate_candidate(row)
    )

    if i % 25 == 0 or i == len(passed):

        print(
            f"Validated {i:,}/{len(passed):,}"
        )


validated = pd.DataFrame(results)


# ============================================================
# SAVE
# ============================================================

audit_file = os.path.join(
    OUT_DIR,
    "05F_source_validation_audit.csv"
)

valid_file = os.path.join(
    OUT_DIR,
    "05F_valid_release_observations.csv"
)

review_file = os.path.join(
    OUT_DIR,
    "05F_manual_source_review.csv"
)

reject_file = os.path.join(
    OUT_DIR,
    "05F_source_rejected.csv"
)

validated.to_csv(
    audit_file,
    index=False
)

valid = validated[
    validated["validation_status"]
    == "VALID_RELEASE"
].copy()

review = validated[
    validated["validation_status"]
    == "MANUAL_REVIEW"
].copy()

reject = validated[
    validated["validation_status"]
    == "REJECT"
].copy()

valid.to_csv(
    valid_file,
    index=False
)

review.to_csv(
    review_file,
    index=False
)

reject.to_csv(
    reject_file,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CODE 5F COMPLETE")
print("=" * 70)

print(
    f"Input PASS candidates:       {len(passed):,}"
)

print(
    f"VALID_RELEASE:               {len(valid):,}"
)

print(
    f"MANUAL_SOURCE_REVIEW:        {len(review):,}"
)

print(
    f"SOURCE_REJECTED:             {len(reject):,}"
)

print("\nValidation distribution:")

print(
    validated["validation_status"]
    .value_counts()
)

print("\nSource diversity of VALID_RELEASE:")

print(
    "Unique PMCID:",
    valid["pmcid"].nunique()
)

print(
    "Unique DOI:",
    valid["doi"]
    .replace({"": np.nan})
    .dropna()
    .nunique()
)

print(
    "Unique source tables:",
    valid[
        ["pmcid", "table_index"]
    ]
    .drop_duplicates()
    .shape[0]
)

print("\nSaved:")
print(audit_file)
print(valid_file)
print(review_file)
print(reject_file)

print("\nIMPORTANT:")
print("VALID_RELEASE observations are source-traceable.")
print("No synthetic observations were generated.")
print("No interpolation was performed.")
print("No duplication was performed.")
print("No ML was performed.")
print("=" * 70)
```

```python
# ============================================================
# CODE 5G — PRIMARY-SOURCE & EXPERIMENTAL-UNIT INDEPENDENCE AUDIT
# ============================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 5G — PRIMARY-SOURCE & EXPERIMENTAL-UNIT INDEPENDENCE AUDIT")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05F_valid_release_observations.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"\nRequired file not found:\n{INPUT_FILE}\n\n"
        "Run Code 5F first."
    )

df = pd.read_csv(INPUT_FILE)

print(f"Validated observations loaded: {len(df):,}")


# ============================================================
# COLUMN CHECK
# ============================================================

required = [
    "candidate_id",
    "pmcid",
    "doi",
    "title",
    "table_index",
    "candidate_time_h",
    "candidate_release_percent"
]

missing = [
    c for c in required
    if c not in df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}"
    )


# ============================================================
# NORMALIZATION HELPERS
# ============================================================

def norm_text(x):

    if pd.isna(x):
        return ""

    x = str(x).lower()

    x = re.sub(
        r"[^a-z0-9\s]",
        " ",
        x
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    ).strip()

    return x


def norm_doi(x):

    if pd.isna(x):
        return ""

    x = str(x).strip().lower()

    x = x.replace(
        "https://doi.org/",
        ""
    )

    x = x.replace(
        "http://doi.org/",
        ""
    )

    x = x.replace(
        "doi:",
        ""
    )

    return x.strip()


df["doi_norm"] = df["doi"].apply(norm_doi)
df["title_norm"] = df["title"].apply(norm_text)
df["pmcid_norm"] = df["pmcid"].apply(norm_text)


# ============================================================
# PRIMARY / SECONDARY SOURCE CLASSIFICATION
# ============================================================

review_terms = [
    r"\breview\b",
    r"\bsystematic review\b",
    r"\bscoping review\b",
    r"\bnarrative review\b",
    r"\brecent advances\b",
    r"\bstate of the art\b",
    r"\boverview\b",
    r"\bcomprehensive review\b",
    r"\bcritical review\b",
    r"\bprogress in\b",
    r"\bperspectives\b"
]

experimental_terms = [
    r"\bin vitro\b",
    r"\bexperimental\b",
    r"\bfabricat",
    r"\bsynthesi",
    r"\bprepared\b",
    r"\bformulat",
    r"\bdrug release study\b",
    r"\brelease profile\b",
    r"\bcharacterization\b",
    r"\bmethodology\b"
]


def source_classification(row):

    text = (
        str(row.get("title", "")) +
        " " +
        str(row.get("source_table_caption", "")) +
        " " +
        str(row.get("source_table_text", ""))
    ).lower()

    review_hits = sum(
        bool(re.search(p, text))
        for p in review_terms
    )

    experimental_hits = sum(
        bool(re.search(p, text))
        for p in experimental_terms
    )

    # Strong review language
    if review_hits >= 1 and experimental_hits == 0:

        return "SECONDARY_REVIEW_LIKELY"

    # Strong experimental evidence
    if experimental_hits >= 2:

        return "PRIMARY_EXPERIMENT_LIKELY"

    # Review title itself
    title = str(
        row.get("title", "")
    ).lower()

    if any(
        re.search(p, title)
        for p in review_terms
    ):

        return "SECONDARY_REVIEW_LIKELY"

    return "UNCERTAIN"


df["source_class"] = df.apply(
    source_classification,
    axis=1
)


# ============================================================
# ARTICLE-LEVEL GROUPING
# ============================================================

article_summary = (
    df.groupby(
        ["pmcid_norm", "doi_norm"],
        dropna=False
    )
    .agg(
        observations=(
            "candidate_id",
            "count"
        ),
        source_tables=(
            "table_index",
            "nunique"
        ),
        title=(
            "title",
            "first"
        ),
        source_classes=(
            "source_class",
            lambda x: ";".join(
                sorted(set(x))
            )
        )
    )
    .reset_index()
)


# ============================================================
# SAME ARTICLE / TABLE / TIME DUPLICATION
# ============================================================

df["time_round"] = pd.to_numeric(
    df["candidate_time_h"],
    errors="coerce"
).round(6)

df["release_round"] = pd.to_numeric(
    df["candidate_release_percent"],
    errors="coerce"
).round(6)


# Exact repeated observation
df["exact_observation_duplicate"] = (
    df.duplicated(
        subset=[
            "pmcid_norm",
            "table_index",
            "time_round",
            "release_round"
        ],
        keep=False
    )
)


# Same source article + same time
df["same_article_same_time"] = (
    df.duplicated(
        subset=[
            "pmcid_norm",
            "time_round"
        ],
        keep=False
    )
)


# ============================================================
# FORMULATION-UNIT PROXY
# ============================================================

# At this stage we do NOT claim these are independent
# formulations. We construct a conservative proxy using
# article + table + time.

df["experimental_unit_proxy"] = (
    df["pmcid_norm"].astype(str)
    + "__T"
    + df["table_index"].astype(str)
    + "__TIME"
    + df["time_round"].astype(str)
)


unit_counts = (
    df["experimental_unit_proxy"]
    .value_counts()
)

df["unit_proxy_repeated"] = (
    df["experimental_unit_proxy"]
    .map(unit_counts)
    > 1
)


# ============================================================
# SOURCE-LEVEL CLASSIFICATION
# ============================================================

def independence_status(row):

    if row["exact_observation_duplicate"]:

        return "DUPLICATE"

    if row["source_class"] == "SECONDARY_REVIEW_LIKELY":

        return "SECONDARY_SOURCE"

    if row["source_class"] == "UNCERTAIN":

        return "MANUAL_SOURCE_CHECK"

    if row["same_article_same_time"]:

        return "SAME_ARTICLE_SAME_TIME_CHECK"

    return "PRIMARY_CANDIDATE"


df["independence_status"] = df.apply(
    independence_status,
    axis=1
)


# ============================================================
# PRIMARY CANDIDATE DATASET
# ============================================================

primary_candidates = df[
    df["independence_status"]
    == "PRIMARY_CANDIDATE"
].copy()


secondary = df[
    df["independence_status"]
    == "SECONDARY_SOURCE"
].copy()


manual = df[
    df["independence_status"]
    == "MANUAL_SOURCE_CHECK"
].copy()


duplicates = df[
    df["independence_status"]
    == "DUPLICATE"
].copy()


same_time = df[
    df["independence_status"]
    == "SAME_ARTICLE_SAME_TIME_CHECK"
].copy()


# ============================================================
# SAVE FILES
# ============================================================

audit_file = os.path.join(
    OUT_DIR,
    "05G_independence_audit.csv"
)

primary_file = os.path.join(
    OUT_DIR,
    "05G_primary_source_candidates.csv"
)

secondary_file = os.path.join(
    OUT_DIR,
    "05G_secondary_review_candidates.csv"
)

manual_file = os.path.join(
    OUT_DIR,
    "05G_manual_source_check.csv"
)

article_file = os.path.join(
    OUT_DIR,
    "05G_article_level_summary.csv"
)

df.to_csv(
    audit_file,
    index=False
)

primary_candidates.to_csv(
    primary_file,
    index=False
)

secondary.to_csv(
    secondary_file,
    index=False
)

manual.to_csv(
    manual_file,
    index=False
)

article_summary.to_csv(
    article_file,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CODE 5G COMPLETE")
print("=" * 70)

print(
    f"Validated input observations:     {len(df):,}"
)

print(
    f"PRIMARY_CANDIDATE:                 {len(primary_candidates):,}"
)

print(
    f"SECONDARY_SOURCE:                  {len(secondary):,}"
)

print(
    f"MANUAL_SOURCE_CHECK:               {len(manual):,}"
)

print(
    f"DUPLICATES:                        {len(duplicates):,}"
)

print(
    f"SAME_ARTICLE_SAME_TIME_CHECK:     {len(same_time):,}"
)


# ============================================================
# SOURCE DIVERSITY
# ============================================================

print("\nSOURCE DIVERSITY")

print(
    "All validated PMCIDs:",
    df["pmcid_norm"].nunique()
)

print(
    "Primary candidate PMCIDs:",
    primary_candidates["pmcid_norm"].nunique()
)

print(
    "Primary candidate DOIs:",
    primary_candidates["doi_norm"]
    .replace({"": np.nan})
    .dropna()
    .nunique()
)

print(
    "Primary candidate source tables:",
    primary_candidates[
        ["pmcid_norm", "table_index"]
    ]
    .drop_duplicates()
    .shape[0]
)


# ============================================================
# SOURCE CLASSIFICATION
# ============================================================

print("\nSOURCE CLASSIFICATION")

print(
    df["source_class"]
    .value_counts()
)


# ============================================================
# ARTICLE OBSERVATION DISTRIBUTION
# ============================================================

print("\nOBSERVATIONS PER ARTICLE")

print(
    article_summary["observations"]
    .describe()
)


# ============================================================
# TOP ARTICLES BY OBSERVATION COUNT
# ============================================================

print("\nTOP SOURCE ARTICLES")

print(
    article_summary
    .sort_values(
        "observations",
        ascending=False
    )
    .head(15)[
        [
            "pmcid_norm",
            "doi_norm",
            "observations",
            "source_tables",
            "source_classes"
        ]
    ]
    .to_string(index=False)
)


print("\nSaved:")
print(audit_file)
print(primary_file)
print(secondary_file)
print(manual_file)
print(article_file)

print("\nIMPORTANT:")
print("This stage does NOT claim every time point is an independent experiment.")
print("It does NOT merge observations across formulations.")
print("It does NOT generate observations.")
print("It does NOT interpolate observations.")
print("It does NOT run machine learning.")
print("=" * 70)
```

```python
# ============================================================
# CODE 5H — FORMULATION / EXPERIMENTAL-UNIT RESOLUTION
# ============================================================

import os
import re
import hashlib
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 5H — FORMULATION / EXPERIMENTAL-UNIT RESOLUTION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05G_independence_audit.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"\nRequired file not found:\n{INPUT_FILE}\n\n"
        "Run Code 5G first."
    )

df = pd.read_csv(INPUT_FILE)

print(f"Input observations: {len(df):,}")


# ============================================================
# BASIC COLUMN CHECK
# ============================================================

required = [
    "candidate_id",
    "pmcid",
    "doi",
    "title",
    "table_index",
    "candidate_time_h",
    "candidate_release_percent",
    "source_table_caption",
    "source_table_text"
]

missing = [
    c for c in required
    if c not in df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns:\n{missing}"
    )


# ============================================================
# TEXT NORMALIZATION
# ============================================================

def clean_text(x):

    if pd.isna(x):
        return ""

    x = str(x)

    x = x.replace("\n", " ")
    x = x.replace("\t", " ")

    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


def normalize_for_id(x):

    x = clean_text(x).lower()

    x = re.sub(
        r"[^a-z0-9.%/\-+]+",
        " ",
        x
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


# ============================================================
# SOURCE TEXT
# ============================================================

df["source_context"] = (
    df["source_table_caption"].fillna("").astype(str)
    + " | "
    + df["source_table_text"].fillna("").astype(str)
)

df["source_context"] = df[
    "source_context"
].apply(clean_text)


# ============================================================
# DRUG EXTRACTION
# ============================================================

drug_patterns = {

    "dexamethasone":
        r"\bdexamethasone\b",

    "paclitaxel":
        r"\bpaclitaxel\b",

    "curcumin":
        r"\bcurcumin\b",

    "doxorubicin":
        r"\bdoxorubicin\b",

    "ibuprofen":
        r"\bibuprofen\b",

    "5-fluorouracil":
        r"\b5[\s\-]?fluorouracil\b|\b5fu\b",

    "gemcitabine":
        r"\bgemcitabine\b",

    "metformin":
        r"\bmetformin\b",

    "rifampicin":
        r"\brifampicin\b|\brifampin\b",

    "vancomycin":
        r"\bvancomycin\b",

    "insulin":
        r"\binsulin\b",

    "dox":
        r"\bDOX\b",

    "other_drug":
        r"\b(drug|payload|therapeutic|active ingredient)\b"
}


def extract_drug(text):

    text = str(text)

    for name, pattern in drug_patterns.items():

        if re.search(
            pattern,
            text,
            flags=re.I
        ):

            return name

    return "NOT_IDENTIFIED"


df["drug_extracted"] = df[
    "source_context"
].apply(extract_drug)


# ============================================================
# POLYMER EXTRACTION
# ============================================================

polymer_patterns = {

    "PLGA":
        r"\bPLGA\b|poly\(lactic[\s\-]?co[\s\-]?glycolic acid\)",

    "PLA":
        r"\bPLA\b|polylactic acid|poly\(lactic acid\)",

    "PCL":
        r"\bPCL\b|polycaprolactone",

    "PEG":
        r"\bPEG\b|polyethylene glycol",

    "PEO":
        r"\bPEO\b|polyethylene oxide",

    "chitosan":
        r"\bchitosan\b",

    "alginate":
        r"\balginate\b",

    "gelatin":
        r"\bgelatin\b",

    "PVA":
        r"\bPVA\b|polyvinyl alcohol",

    "poloxamer":
        r"\bpoloxamer\b",

    "polyanhydride":
        r"\bpolyanhydride\b",

    "polyurethane":
        r"\bpolyurethane\b",

    "polystyrene":
        r"\bpolystyrene\b",

    "cellulose":
        r"\bcellulose\b"
}


def extract_polymers(text):

    found = []

    for name, pattern in polymer_patterns.items():

        if re.search(
            pattern,
            str(text),
            flags=re.I
        ):

            found.append(name)

    return ";".join(found) if found else "NOT_IDENTIFIED"


df["polymer_extracted"] = df[
    "source_context"
].apply(extract_polymers)


# ============================================================
# NANOMATERIAL / CARRIER EXTRACTION
# ============================================================

carrier_patterns = {

    "nanoparticle":
        r"\bnanoparticle[s]?\b",

    "nanocarrier":
        r"\bnanocarrier[s]?\b",

    "micelle":
        r"\bmicelle[s]?\b",

    "liposome":
        r"\bliposome[s]?\b",

    "hydrogel":
        r"\bhydrogel[s]?\b",

    "microsphere":
        r"\bmicrosphere[s]?\b",

    "microparticle":
        r"\bmicroparticle[s]?\b",

    "nanofiber":
        r"\bnanofiber[s]?\b",

    "nanofibrous":
        r"\bnanofibrous\b",

    "film":
        r"\bfilm[s]?\b",

    "implant":
        r"\bimplant[s]?\b",

    "beads":
        r"\bbead[s]?\b"
}


def extract_carriers(text):

    found = []

    for name, pattern in carrier_patterns.items():

        if re.search(
            pattern,
            str(text),
            flags=re.I
        ):

            found.append(name)

    return ";".join(found) if found else "NOT_IDENTIFIED"


df["carrier_extracted"] = df[
    "source_context"
].apply(extract_carriers)


# ============================================================
# pH EXTRACTION
# ============================================================

def extract_ph(text):

    matches = re.findall(
        r"\bpH\s*[:=]?\s*"
        r"(\d+(?:\.\d+)?)",
        str(text),
        flags=re.I
    )

    vals = []

    for x in matches:

        try:
            value = float(x)

            if 0 <= value <= 14:
                vals.append(value)

        except:
            pass

    vals = sorted(set(vals))

    return ";".join(
        str(x) for x in vals
    ) if vals else "NOT_IDENTIFIED"


df["ph_extracted"] = df[
    "source_context"
].apply(extract_ph)


# ============================================================
# TEMPERATURE EXTRACTION
# ============================================================

def extract_temperature(text):

    matches = re.findall(
        r"(\d+(?:\.\d+)?)\s*"
        r"(?:°\s*C|degrees?\s*C|deg\s*C)",
        str(text),
        flags=re.I
    )

    vals = []

    for x in matches:

        try:

            value = float(x)

            if -20 <= value <= 150:
                vals.append(value)

        except:
            pass

    vals = sorted(set(vals))

    return ";".join(
        str(x) for x in vals
    ) if vals else "NOT_IDENTIFIED"


df["temperature_C_extracted"] = df[
    "source_context"
].apply(extract_temperature)


# ============================================================
# RELEASE MEDIUM EXTRACTION
# ============================================================

medium_patterns = {

    "PBS":
        r"\bPBS\b|phosphate buffered saline",

    "water":
        r"\bdistilled water\b|\bdeionized water\b|\bwater\b",

    "saline":
        r"\bsaline\b",

    "SGF":
        r"\bSGF\b|simulated gastric fluid",

    "SIF":
        r"\bSIF\b|simulated intestinal fluid",

    "buffer":
        r"\bbuffer\b",

    "dialysis":
        r"\bdialysis\b"
}


def extract_medium(text):

    found = []

    for name, pattern in medium_patterns.items():

        if re.search(
            pattern,
            str(text),
            flags=re.I
        ):

            found.append(name)

    return ";".join(found) if found else "NOT_IDENTIFIED"


df["release_medium_extracted"] = df[
    "source_context"
].apply(extract_medium)


# ============================================================
# FORMULATION / GROUP IDENTIFIER
# ============================================================

def extract_group_identifier(text):

    text = str(text)

    patterns = [

        # Formulation labels
        r"\bF\d+\b",
        r"\bF[\-\s]?\d+\b",

        # Sample / formulation groups
        r"\bformulation\s*\d+\b",
        r"\bformulation\s*[A-Z]\b",
        r"\bgroup\s*[A-Z]\b",
        r"\bgroup\s*\d+\b",

        # Sample identifiers
        r"\bsample\s*[A-Z]\b",
        r"\bsample\s*\d+\b",

        # Batch identifiers
        r"\bbatch\s*[A-Z0-9\-]+\b"
    ]

    found = []

    for pattern in patterns:

        matches = re.findall(
            pattern,
            text,
            flags=re.I
        )

        found.extend(matches)

    found = [
        clean_text(x)
        for x in found
    ]

    found = sorted(
        set(found)
    )

    return ";".join(found) if found else "NOT_IDENTIFIED"


df["group_identifier_extracted"] = df[
    "source_context"
].apply(extract_group_identifier)


# ============================================================
# CONCENTRATION / LOADING SIGNAL
# ============================================================

def extract_concentration_signals(text):

    text = str(text)

    patterns = [

        r"\d+(?:\.\d+)?\s*%\s*(?:w/w|w/v|v/v)?",

        r"\d+(?:\.\d+)?\s*mg\s*/\s*mL",

        r"\d+(?:\.\d+)?\s*mg/mL",

        r"\d+(?:\.\d+)?\s*mg",

        r"\d+(?:\.\d+)?\s*µg",

        r"\d+(?:\.\d+)?\s*ug"
    ]

    found = []

    for pattern in patterns:

        matches = re.findall(
            pattern,
            text,
            flags=re.I
        )

        found.extend(matches)

    found = [
        clean_text(x)
        for x in found
    ]

    found = sorted(
        set(found)
    )

    return ";".join(found) if found else "NOT_IDENTIFIED"


df["concentration_signals"] = df[
    "source_context"
].apply(extract_concentration_signals)


# ============================================================
# CONDITION SIGNATURE
# ============================================================

def make_condition_signature(row):

    values = [

        row["drug_extracted"],

        row["polymer_extracted"],

        row["carrier_extracted"],

        row["ph_extracted"],

        row["temperature_C_extracted"],

        row["release_medium_extracted"],

        row["group_identifier_extracted"],

        row["concentration_signals"]
    ]

    values = [
        normalize_for_id(x)
        for x in values
    ]

    return "||".join(values)


df["condition_signature"] = df.apply(
    make_condition_signature,
    axis=1
)


# ============================================================
# EXPERIMENTAL UNIT PROXY
# ============================================================

def make_unit_proxy(row):

    pmcid = normalize_for_id(
        row["pmcid"]
    )

    signature = normalize_for_id(
        row["condition_signature"]
    )

    table = normalize_for_id(
        row["table_index"]
    )

    return (
        pmcid
        + "||"
        + table
        + "||"
        + signature
    )


df["experimental_unit_proxy"] = df.apply(
    make_unit_proxy,
    axis=1
)


# ============================================================
# HASHED EXPERIMENTAL UNIT ID
# ============================================================

def make_hash(text):

    return hashlib.sha256(
        str(text).encode("utf-8")
    ).hexdigest()[:16]


df["experimental_unit_id"] = (
    "EU-"
    + df[
        "experimental_unit_proxy"
    ].apply(make_hash)
)


# ============================================================
# SAME ARTICLE / TIME ANALYSIS
# ============================================================

df["time_numeric"] = pd.to_numeric(
    df["candidate_time_h"],
    errors="coerce"
)

df["release_numeric"] = pd.to_numeric(
    df["candidate_release_percent"],
    errors="coerce"
)


# Number of observations in each experimental unit
unit_counts = (
    df["experimental_unit_id"]
    .value_counts()
)

df["observations_in_unit"] = (
    df["experimental_unit_id"]
    .map(unit_counts)
)


# Multiple time points from same unit are EXPECTED
df["multi_timepoint_unit"] = (
    df["observations_in_unit"] > 1
)


# ============================================================
# SAME ARTICLE + SAME TIME + DIFFERENT CONDITION
# ============================================================

group_cols = [
    "pmcid",
    "time_numeric"
]

article_time = (
    df.groupby(group_cols)
    .agg(
        observations=(
            "candidate_id",
            "count"
        ),
        unique_units=(
            "experimental_unit_id",
            "nunique"
        )
    )
    .reset_index()
)

article_time["potentially_independent_same_time"] = (
    article_time["unique_units"] > 1
)


# ============================================================
# CLASSIFY EXPERIMENTAL UNIT
# ============================================================

def unit_resolution(row):

    if row["group_identifier_extracted"] != "NOT_IDENTIFIED":
        return "FORMULATION_IDENTIFIER_FOUND"

    if row["drug_extracted"] != "NOT_IDENTIFIED":
        if row["polymer_extracted"] != "NOT_IDENTIFIED":
            return "DRUG_POLYMER_UNIT"

    if row["condition_signature"]:
        return "CONDITION_BASED_UNIT"

    return "UNRESOLVED_UNIT"


df["unit_resolution"] = df.apply(
    unit_resolution,
    axis=1
)


# ============================================================
# FINAL CONSERVATIVE STATUS
# ============================================================

def final_unit_status(row):

    if row["unit_resolution"] == "UNRESOLVED_UNIT":
        return "MANUAL_UNIT_REVIEW"

    if (
        row["source_class"]
        == "SECONDARY_REVIEW_LIKELY"
    ):
        return "SECONDARY_SOURCE"

    return "PRIMARY_UNIT_CANDIDATE"


df["unit_status"] = df.apply(
    final_unit_status,
    axis=1
)


# ============================================================
# SAVE
# ============================================================

audit_file = os.path.join(
    OUT_DIR,
    "05H_formulation_unit_audit.csv"
)

unit_summary_file = os.path.join(
    OUT_DIR,
    "05H_experimental_unit_summary.csv"
)

same_time_file = os.path.join(
    OUT_DIR,
    "05H_same_article_same_time_analysis.csv"
)

primary_units_file = os.path.join(
    OUT_DIR,
    "05H_primary_unit_candidates.csv"
)

manual_units_file = os.path.join(
    OUT_DIR,
    "05H_manual_unit_review.csv"
)

df.to_csv(
    audit_file,
    index=False
)

unit_summary = (
    df.groupby(
        "experimental_unit_id"
    )
    .agg(
        pmcid=(
            "pmcid",
            "first"
        ),
        doi=(
            "doi",
            "first"
        ),
        title=(
            "title",
            "first"
        ),
        drug=(
            "drug_extracted",
            "first"
        ),
        polymer=(
            "polymer_extracted",
            "first"
        ),
        carrier=(
            "carrier_extracted",
            "first"
        ),
        pH=(
            "ph_extracted",
            "first"
        ),
        temperature_C=(
            "temperature_C_extracted",
            "first"
        ),
        release_medium=(
            "release_medium_extracted",
            "first"
        ),
        group_identifier=(
            "group_identifier_extracted",
            "first"
        ),
        observations=(
            "candidate_id",
            "count"
        ),
        min_time_h=(
            "time_numeric",
            "min"
        ),
        max_time_h=(
            "time_numeric",
            "max"
        ),
        unit_resolution=(
            "unit_resolution",
            "first"
        ),
        source_class=(
            "source_class",
            "first"
        )
    )
    .reset_index()
)

unit_summary.to_csv(
    unit_summary_file,
    index=False
)

article_time.to_csv(
    same_time_file,
    index=False
)

primary_units = df[
    df["unit_status"]
    == "PRIMARY_UNIT_CANDIDATE"
].copy()

manual_units = df[
    df["unit_status"]
    == "MANUAL_UNIT_REVIEW"
].copy()

primary_units.to_csv(
    primary_units_file,
    index=False
)

manual_units.to_csv(
    manual_units_file,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CODE 5H COMPLETE")
print("=" * 70)

print(
    f"Input validated observations:      {len(df):,}"
)

print(
    f"Experimental-unit proxies:         "
    f"{df['experimental_unit_id'].nunique():,}"
)

print(
    f"PRIMARY_UNIT_CANDIDATE:             "
    f"{len(primary_units):,}"
)

print(
    f"SECONDARY_SOURCE:                   "
    f"{(df['unit_status'] == 'SECONDARY_SOURCE').sum():,}"
)

print(
    f"MANUAL_UNIT_REVIEW:                 "
    f"{len(manual_units):,}"
)

print("\nUnit-resolution distribution:")

print(
    df["unit_resolution"]
    .value_counts()
)


print("\nExperimental units with multiple time points:")

print(
    unit_summary[
        "observations"
    ]
    .value_counts()
    .sort_index()
)


print("\nSame-article / same-time groups:")

print(
    f"Total groups: "
    f"{len(article_time):,}"
)

print(
    f"Groups with >1 experimental unit: "
    f"{article_time['potentially_independent_same_time'].sum():,}"
)


print("\nDrug extraction:")

print(
    df["drug_extracted"]
    .value_counts()
    .head(15)
)


print("\nPolymer extraction:")

print(
    df["polymer_extracted"]
    .value_counts()
    .head(15)
)


print("\nCarrier extraction:")

print(
    df["carrier_extracted"]
    .value_counts()
    .head(15)
)


print("\nSaved:")
print(audit_file)
print(unit_summary_file)
print(same_time_file)
print(primary_units_file)
print(manual_units_file)


print("\nIMPORTANT:")
print("Experimental-unit IDs are conservative PROXIES.")
print("They are not treated as ground truth.")
print("Multiple time points from one formulation remain allowed.")
print("No observations were generated.")
print("No interpolation was performed.")
print("No observations were duplicated.")
print("No machine learning was performed.")

print("=" * 70)
```

```python
# ============================================================
# CODE 5I — ROW-LEVEL FORMULATION RESOLUTION
# ============================================================

import os
import re
import pandas as pd
import numpy as np
import hashlib

print("=" * 70)
print("CODE 5I — ROW-LEVEL FORMULATION RESOLUTION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05H_formulation_unit_audit.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"\nRequired file not found:\n{INPUT_FILE}\n\n"
        "Run Code 5H first."
    )

df = pd.read_csv(INPUT_FILE)

print(f"Input observations: {len(df):,}")


# ============================================================
# REQUIRED COLUMNS
# ============================================================

required = [
    "candidate_id",
    "pmcid",
    "doi",
    "title",
    "table_index",
    "candidate_time_h",
    "candidate_release_percent",
    "source_table_caption",
    "source_table_text"
]

missing = [
    c for c in required
    if c not in df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}"
    )


# ============================================================
# TEXT CLEANING
# ============================================================

def clean_text(x):

    if pd.isna(x):
        return ""

    x = str(x)

    x = x.replace("\n", " ")
    x = x.replace("\t", " ")

    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


def norm(x):

    x = clean_text(x).lower()

    x = re.sub(
        r"[^a-z0-9.%/\-+ ]",
        " ",
        x
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


# ============================================================
# ROW CONTEXT
# ============================================================

# Code 5F preserved the complete source table text.
# We attempt to reconstruct the relevant source row using:
#
# 1. explicit row_index when available
# 2. candidate release value
# 3. candidate time
#
# This prevents using every polymer/carrier in an entire table
# as metadata for every observation.


def split_source_rows(text):

    text = clean_text(text)

    if not text:
        return []

    # Common separators generated by XML table extraction
    parts = re.split(
        r"\s*\|\s*|\s+;\s+",
        text
    )

    parts = [
        clean_text(x)
        for x in parts
        if clean_text(x)
    ]

    return parts


# ============================================================
# DRUG PATTERNS
# ============================================================

drug_patterns = {

    "dexamethasone":
        r"\bdexamethasone\b",

    "paclitaxel":
        r"\bpaclitaxel\b",

    "curcumin":
        r"\bcurcumin\b",

    "doxorubicin":
        r"\bdoxorubicin\b",

    "ibuprofen":
        r"\bibuprofen\b",

    "metformin":
        r"\bmetformin\b",

    "gemcitabine":
        r"\bgemcitabine\b",

    "rifampicin":
        r"\brifampicin\b|\brifampin\b",

    "vancomycin":
        r"\bvancomycin\b",

    "insulin":
        r"\binsulin\b",

    "5-fluorouracil":
        r"\b5[\s\-]?fluorouracil\b|\b5fu\b",

    "dox":
        r"\bDOX\b"
}


def extract_drugs(text):

    found = []

    for name, pattern in drug_patterns.items():

        if re.search(
            pattern,
            str(text),
            flags=re.I
        ):
            found.append(name)

    return ";".join(found) if found else "NOT_IDENTIFIED"


# ============================================================
# POLYMER PATTERNS
# ============================================================

polymer_patterns = {

    "PLGA":
        r"\bPLGA\b|poly\(lactic[\s\-]?co[\s\-]?glycolic acid\)",

    "PLA":
        r"\bPLA\b|polylactic acid|poly\(lactic acid\)",

    "PCL":
        r"\bPCL\b|polycaprolactone",

    "PEG":
        r"\bPEG\b|polyethylene glycol",

    "PEO":
        r"\bPEO\b|polyethylene oxide",

    "chitosan":
        r"\bchitosan\b",

    "alginate":
        r"\balginate\b",

    "gelatin":
        r"\bgelatin\b",

    "PVA":
        r"\bPVA\b|polyvinyl alcohol",

    "poloxamer":
        r"\bpoloxamer\b",

    "polyurethane":
        r"\bpolyurethane\b",

    "cellulose":
        r"\bcellulose\b",

    "polyanhydride":
        r"\bpolyanhydride\b",

    "polystyrene":
        r"\bpolystyrene\b"
}


def extract_polymers(text):

    found = []

    for name, pattern in polymer_patterns.items():

        if re.search(
            pattern,
            str(text),
            flags=re.I
        ):
            found.append(name)

    return ";".join(found) if found else "NOT_IDENTIFIED"


# ============================================================
# CARRIER PATTERNS
# ============================================================

carrier_patterns = {

    "nanoparticle":
        r"\bnanoparticle[s]?\b",

    "nanocarrier":
        r"\bnanocarrier[s]?\b",

    "micelle":
        r"\bmicelle[s]?\b",

    "liposome":
        r"\bliposome[s]?\b",

    "hydrogel":
        r"\bhydrogel[s]?\b",

    "microsphere":
        r"\bmicrosphere[s]?\b",

    "microparticle":
        r"\bmicroparticle[s]?\b",

    "nanofiber":
        r"\bnanofiber[s]?\b",

    "film":
        r"\bfilm[s]?\b",

    "implant":
        r"\bimplant[s]?\b",

    "beads":
        r"\bbead[s]?\b"
}


def extract_carriers(text):

    found = []

    for name, pattern in carrier_patterns.items():

        if re.search(
            pattern,
            str(text),
            flags=re.I
        ):
            found.append(name)

    return ";".join(found) if found else "NOT_IDENTIFIED"


# ============================================================
# pH
# ============================================================

def extract_ph(text):

    matches = re.findall(
        r"\bpH\s*[:=]?\s*(\d+(?:\.\d+)?)",
        str(text),
        flags=re.I
    )

    vals = []

    for x in matches:

        try:

            v = float(x)

            if 0 <= v <= 14:
                vals.append(v)

        except:
            pass

    vals = sorted(set(vals))

    return ";".join(
        str(x) for x in vals
    ) if vals else "NOT_IDENTIFIED"


# ============================================================
# TEMPERATURE
# ============================================================

def extract_temperature(text):

    matches = re.findall(
        r"(\d+(?:\.\d+)?)\s*"
        r"(?:°\s*C|degrees?\s*C|deg\s*C)",
        str(text),
        flags=re.I
    )

    vals = []

    for x in matches:

        try:

            v = float(x)

            if -20 <= v <= 150:
                vals.append(v)

        except:
            pass

    vals = sorted(set(vals))

    return ";".join(
        str(x) for x in vals
    ) if vals else "NOT_IDENTIFIED"


# ============================================================
# MEDIUM
# ============================================================

medium_patterns = {

    "PBS":
        r"\bPBS\b|phosphate buffered saline",

    "water":
        r"\bdistilled water\b|\bdeionized water\b",

    "saline":
        r"\bsaline\b",

    "SGF":
        r"\bSGF\b|simulated gastric fluid",

    "SIF":
        r"\bSIF\b|simulated intestinal fluid",

    "buffer":
        r"\bbuffer\b",

    "dialysis":
        r"\bdialysis\b"
}


def extract_medium(text):

    found = []

    for name, pattern in medium_patterns.items():

        if re.search(
            pattern,
            str(text),
            flags=re.I
        ):
            found.append(name)

    return ";".join(found) if found else "NOT_IDENTIFIED"


# ============================================================
# FORMULATION IDENTIFIERS
# ============================================================

def extract_formulation_ids(text):

    patterns = [

        r"\bF[\-\s]?\d+\b",

        r"\bformulation\s*\d+\b",

        r"\bformulation\s*[A-Z]\b",

        r"\bgroup\s*[A-Z]\b",

        r"\bgroup\s*\d+\b",

        r"\bsample\s*[A-Z]\b",

        r"\bsample\s*\d+\b",

        r"\bbatch\s*[A-Z0-9\-]+\b"
    ]

    found = []

    for pattern in patterns:

        matches = re.findall(
            pattern,
            str(text),
            flags=re.I
        )

        found.extend(matches)

    found = sorted(
        set(
            clean_text(x)
            for x in found
        )
    )

    return ";".join(found) if found else "NOT_IDENTIFIED"


# ============================================================
# FIND RELEVANT ROW
# ============================================================

def find_relevant_row(row):

    table_text = clean_text(
        row["source_table_text"]
    )

    if not table_text:
        return ""

    candidate_time = row[
        "candidate_time_h"
    ]

    candidate_release = row[
        "candidate_release_percent"
    ]

    time_patterns = []

    try:

        t = float(candidate_time)

        # Exact numeric representation
        time_patterns.extend([
            rf"\b{re.escape(str(t))}\b",
            rf"\b{re.escape(str(int(t)))}\b"
        ])

    except:
        pass

    try:

        r = float(candidate_release)

        time_patterns.extend([
            rf"\b{re.escape(str(r))}\s*%",
            rf"\b{re.escape(str(int(r)))}\s*%"
        ])

    except:
        pass

    # First try splitting into row-like chunks.
    chunks = re.split(
        r"\n+|(?=\b(?:F|Formulation|Group|Sample)\s*[-]?\s*\d+)",
        table_text,
        flags=re.I
    )

    chunks = [
        clean_text(x)
        for x in chunks
        if clean_text(x)
    ]

    # Find a chunk containing candidate release/time.
    scored = []

    for chunk in chunks:

        score = 0

        try:

            r = float(candidate_release)

            if (
                re.search(
                    rf"\b{re.escape(str(r))}\s*%",
                    chunk
                )
                or
                re.search(
                    rf"\b{re.escape(str(int(r)))}\s*%",
                    chunk
                )
            ):
                score += 4

        except:
            pass

        try:

            t = float(candidate_time)

            if (
                re.search(
                    rf"\b{re.escape(str(t))}\b",
                    chunk
                )
                or
                re.search(
                    rf"\b{re.escape(str(int(t)))}\b",
                    chunk
                )
            ):
                score += 3

        except:
            pass

        if extract_drugs(chunk) != "NOT_IDENTIFIED":
            score += 1

        if extract_polymers(chunk) != "NOT_IDENTIFIED":
            score += 1

        if extract_formulation_ids(chunk) != "NOT_IDENTIFIED":
            score += 2

        scored.append(
            (score, chunk)
        )

    if scored:

        scored.sort(
            key=lambda x: x[0],
            reverse=True
        )

        if scored[0][0] > 0:
            return scored[0][1]

    # Fallback: use full source table only when
    # no row-level match can be established.
    return ""


# ============================================================
# ROW-LEVEL EXTRACTION
# ============================================================

records = []

for i, (_, row) in enumerate(
    df.iterrows(),
    start=1
):

    relevant_row = find_relevant_row(row)

    record = row.to_dict()

    record["resolved_source_row"] = (
        relevant_row
    )

    record["row_resolution_status"] = (
        "ROW_RESOLVED"
        if relevant_row
        else "ROW_UNRESOLVED"
    )

    if relevant_row:

        context = relevant_row

    else:

        context = ""

    record["row_drug"] = (
        extract_drugs(context)
    )

    record["row_polymer"] = (
        extract_polymers(context)
    )

    record["row_carrier"] = (
        extract_carriers(context)
    )

    record["row_pH"] = (
        extract_ph(context)
    )

    record["row_temperature_C"] = (
        extract_temperature(context)
    )

    record["row_release_medium"] = (
        extract_medium(context)
    )

    record["row_formulation_id"] = (
        extract_formulation_ids(context)
    )

    records.append(record)

    if (
        i % 25 == 0
        or i == len(df)
    ):

        print(
            f"Processed {i:,}/{len(df):,}"
        )


resolved = pd.DataFrame(records)


# ============================================================
# FALLBACK FLAGS
# ============================================================

resolved["row_metadata_count"] = (
    (
        resolved["row_drug"]
        != "NOT_IDENTIFIED"
    ).astype(int)
    +
    (
        resolved["row_polymer"]
        != "NOT_IDENTIFIED"
    ).astype(int)
    +
    (
        resolved["row_carrier"]
        != "NOT_IDENTIFIED"
    ).astype(int)
)


def metadata_quality(row):

    if row[
        "row_resolution_status"
    ] == "ROW_UNRESOLVED":

        return "UNRESOLVED"

    if row[
        "row_metadata_count"
    ] >= 2:

        return "GOOD_ROW_CONTEXT"

    if row[
        "row_metadata_count"
    ] == 1:

        return "PARTIAL_ROW_CONTEXT"

    return "WEAK_ROW_CONTEXT"


resolved["row_metadata_quality"] = resolved.apply(
    metadata_quality,
    axis=1
)


# ============================================================
# EXPERIMENTAL UNIT SIGNATURE
# ============================================================

def make_signature(row):

    values = [

        row["pmcid"],

        row["row_formulation_id"],

        row["row_drug"],

        row["row_polymer"],

        row["row_carrier"],

        row["row_pH"],

        row["row_temperature_C"],

        row["row_release_medium"]
    ]

    values = [
        norm(x)
        for x in values
    ]

    return "||".join(values)


resolved["row_experimental_signature"] = (
    resolved.apply(
        make_signature,
        axis=1
    )
)


# ============================================================
# HASHED UNIT ID
# ============================================================

def hash_id(x):

    return hashlib.sha256(
        str(x).encode(
            "utf-8"
        )
    ).hexdigest()[:16]


resolved["row_experimental_unit_id"] = (
    "EU5I-"
    + resolved[
        "row_experimental_signature"
    ].apply(hash_id)
)


# ============================================================
# QUALITY STATUS
# ============================================================

def quality_status(row):

    if (
        row["row_metadata_quality"]
        == "GOOD_ROW_CONTEXT"
    ):
        return "ROW_LEVEL_CANDIDATE"

    if (
        row["row_metadata_quality"]
        == "PARTIAL_ROW_CONTEXT"
    ):
        return "MANUAL_METADATA_REVIEW"

    return "MANUAL_METADATA_REVIEW"


resolved["row_level_status"] = (
    resolved.apply(
        quality_status,
        axis=1
    )
)


# ============================================================
# SAVE
# ============================================================

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "05I_row_level_formulation_audit.csv"
)

CANDIDATE_FILE = os.path.join(
    OUT_DIR,
    "05I_row_level_candidates.csv"
)

MANUAL_FILE = os.path.join(
    OUT_DIR,
    "05I_row_level_manual_review.csv"
)

resolved.to_csv(
    AUDIT_FILE,
    index=False
)

row_candidates = resolved[
    resolved["row_level_status"]
    == "ROW_LEVEL_CANDIDATE"
].copy()

manual = resolved[
    resolved["row_level_status"]
    == "MANUAL_METADATA_REVIEW"
].copy()

row_candidates.to_csv(
    CANDIDATE_FILE,
    index=False
)

manual.to_csv(
    MANUAL_FILE,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CODE 5I COMPLETE")
print("=" * 70)

print(
    f"Input observations:              {len(df):,}"
)

print(
    f"ROW_LEVEL_CANDIDATE:             "
    f"{len(row_candidates):,}"
)

print(
    f"MANUAL_METADATA_REVIEW:          "
    f"{len(manual):,}"
)

print("\nRow resolution:")

print(
    resolved[
        "row_resolution_status"
    ].value_counts()
)

print("\nMetadata quality:")

print(
    resolved[
        "row_metadata_quality"
    ].value_counts()
)

print("\nDrug — row-level:")

print(
    resolved[
        "row_drug"
    ].value_counts()
    .head(20)
)

print("\nPolymer — row-level:")

print(
    resolved[
        "row_polymer"
    ].value_counts()
    .head(20)
)

print("\nCarrier — row-level:")

print(
    resolved[
        "row_carrier"
    ].value_counts()
    .head(20)
)

print("\nFormulation identifiers:")

print(
    (
        resolved[
            "row_formulation_id"
        ]
        != "NOT_IDENTIFIED"
    ).sum()
)

print("\nExperimental-unit proxies:")

print(
    resolved[
        "row_experimental_unit_id"
    ].nunique()
)

print("\nSaved:")
print(AUDIT_FILE)
print(CANDIDATE_FILE)
print(MANUAL_FILE)

print("\nIMPORTANT:")
print("This is a row-level extraction aid.")
print("It does NOT establish experimental independence automatically.")
print("It does NOT generate observations.")
print("It does NOT interpolate values.")
print("It does NOT duplicate observations.")
print("It does NOT run machine learning.")

print("=" * 70)
```

```python
# ============================================================
# CODE 5J — EXACT XML ROW RECOVERY
# ============================================================

import os
import re
import glob
import pandas as pd
import numpy as np
import xml.etree.ElementTree as ET

print("=" * 70)
print("CODE 5J — EXACT XML ROW RECOVERY")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05I_row_level_formulation_audit.csv"
)

XML_DIR = os.path.join(
    ROOT,
    "data",
    "raw",
    "pmc_xml"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"Required file not found:\n{INPUT_FILE}\n\n"
        "Run Code 5I first."
    )

if not os.path.exists(XML_DIR):
    raise FileNotFoundError(
        f"XML directory not found:\n{XML_DIR}"
    )

df = pd.read_csv(INPUT_FILE)

print(f"Input observations: {len(df):,}")


# ============================================================
# XML HELPERS
# ============================================================

def local_name(tag):

    if "}" in tag:
        return tag.split("}", 1)[1]

    return tag


def clean_text(element):

    if element is None:
        return ""

    text = " ".join(
        " ".join(element.itertext()).split()
    )

    return text.strip()


def find_xml(pmcid):

    if pd.isna(pmcid):
        return None

    pmcid = str(pmcid).strip()

    direct = os.path.join(
        XML_DIR,
        f"{pmcid}.xml"
    )

    if os.path.exists(direct):
        return direct

    matches = glob.glob(
        os.path.join(
            XML_DIR,
            f"*{pmcid}*.xml"
        )
    )

    if matches:
        return matches[0]

    return None


# ============================================================
# GET TABLES FROM XML
# ============================================================

def get_table_elements(root):

    tables = []

    for elem in root.iter():

        if local_name(elem.tag) == "table-wrap":

            table = None

            for child in elem.iter():

                if local_name(child.tag) == "table":
                    table = child
                    break

            if table is not None:
                tables.append(elem)

    return tables


# ============================================================
# GET ROWS FROM TABLE
# ============================================================

def get_rows(table_wrap):

    table_element = None

    for child in table_wrap.iter():

        if local_name(child.tag) == "table":
            table_element = child
            break

    if table_element is None:
        return []

    rows = []

    for tr in table_element.iter():

        if local_name(tr.tag) != "tr":
            continue

        cells = []

        for cell in tr:

            if local_name(cell.tag) not in (
                "td",
                "th"
            ):
                continue

            cells.append(
                clean_text(cell)
            )

        if cells:
            rows.append(cells)

    return rows


# ============================================================
# EXTRACT EXACT SOURCE ROW
# ============================================================

def recover_exact_row(pmcid, table_index, row_index):

    xml_path = find_xml(pmcid)

    if xml_path is None:

        return {
            "status": "XML_NOT_FOUND",
            "row_text": "",
            "cells": ""
        }

    try:

        tree = ET.parse(xml_path)
        root = tree.getroot()

    except Exception as e:

        return {
            "status": "XML_PARSE_ERROR",
            "row_text": "",
            "cells": ""
        }

    tables = get_table_elements(root)

    try:
        ti = int(table_index)
    except:
        return {
            "status": "INVALID_TABLE_INDEX",
            "row_text": "",
            "cells": ""
        }

    if ti < 0 or ti >= len(tables):

        return {
            "status": "TABLE_INDEX_NOT_FOUND",
            "row_text": "",
            "cells": ""
        }

    rows = get_rows(
        tables[ti]
    )

    try:
        ri = int(row_index)
    except:
        return {
            "status": "INVALID_ROW_INDEX",
            "row_text": "",
            "cells": ""
        }

    # --------------------------------------------------------
    # IMPORTANT:
    # XML extraction may include header rows.
    # Therefore first try exact row index.
    # --------------------------------------------------------

    if 0 <= ri < len(rows):

        cells = rows[ri]

        return {
            "status": "EXACT_ROW",
            "row_text": " | ".join(cells),
            "cells": " || ".join(cells)
        }

    return {
        "status": "ROW_INDEX_NOT_FOUND",
        "row_text": "",
        "cells": ""
    }


# ============================================================
# PROCESS ALL OBSERVATIONS
# ============================================================

recovered = []

for i, (_, row) in enumerate(
    df.iterrows(),
    start=1
):

    result = recover_exact_row(
        row["pmcid"],
        row["table_index"],
        row.get("row_index", np.nan)
    )

    record = row.to_dict()

    record["exact_xml_row_status"] = (
        result["status"]
    )

    record["exact_xml_row_text"] = (
        result["row_text"]
    )

    record["exact_xml_cells"] = (
        result["cells"]
    )

    recovered.append(record)

    if (
        i % 25 == 0
        or i == len(df)
    ):

        print(
            f"Recovered {i:,}/{len(df):,}"
        )


out = pd.DataFrame(
    recovered
)


# ============================================================
# ROW-LEVEL TERM EXTRACTION
# ============================================================

drug_patterns = {

    "dexamethasone":
        r"\bdexamethasone\b",

    "paclitaxel":
        r"\bpaclitaxel\b",

    "curcumin":
        r"\bcurcumin\b",

    "doxorubicin":
        r"\bdoxorubicin\b",

    "ibuprofen":
        r"\bibuprofen\b",

    "metformin":
        r"\bmetformin\b",

    "gemcitabine":
        r"\bgemcitabine\b",

    "rifampicin":
        r"\brifampicin\b|\brifampin\b",

    "vancomycin":
        r"\bvancomycin\b",

    "insulin":
        r"\binsulin\b",

    "5-fluorouracil":
        r"\b5[\s\-]?fluorouracil\b|\b5fu\b",

    "DOX":
        r"\bDOX\b"
}


polymer_patterns = {

    "PLGA":
        r"\bPLGA\b|poly\(lactic[\s\-]?co[\s\-]?glycolic acid\)",

    "PLA":
        r"\bPLA\b|polylactic acid|poly\(lactic acid\)",

    "PCL":
        r"\bPCL\b|polycaprolactone",

    "PEG":
        r"\bPEG\b|polyethylene glycol",

    "PEO":
        r"\bPEO\b|polyethylene oxide",

    "chitosan":
        r"\bchitosan\b",

    "alginate":
        r"\balginate\b",

    "gelatin":
        r"\bgelatin\b",

    "PVA":
        r"\bPVA\b|polyvinyl alcohol",

    "poloxamer":
        r"\bpoloxamer\b",

    "polyurethane":
        r"\bpolyurethane\b",

    "cellulose":
        r"\bcellulose\b"
}


carrier_patterns = {

    "nanoparticle":
        r"\bnanoparticle[s]?\b",

    "nanocarrier":
        r"\bnanocarrier[s]?\b",

    "micelle":
        r"\bmicelle[s]?\b",

    "liposome":
        r"\bliposome[s]?\b",

    "hydrogel":
        r"\bhydrogel[s]?\b",

    "microsphere":
        r"\bmicrosphere[s]?\b",

    "microparticle":
        r"\bmicroparticle[s]?\b",

    "nanofiber":
        r"\bnanofiber[s]?\b",

    "film":
        r"\bfilm[s]?\b",

    "implant":
        r"\bimplant[s]?\b",

    "beads":
        r"\bbead[s]?\b"
}


def extract_terms(text, patterns):

    found = []

    for name, pattern in patterns.items():

        if re.search(
            pattern,
            str(text),
            flags=re.I
        ):
            found.append(name)

    return (
        ";".join(found)
        if found
        else "NOT_IDENTIFIED"
    )


# ============================================================
# EXACT ROW METADATA
# ============================================================

out["xml_row_drug"] = out[
    "exact_xml_row_text"
].apply(
    lambda x:
        extract_terms(
            x,
            drug_patterns
        )
)

out["xml_row_polymer"] = out[
    "exact_xml_row_text"
].apply(
    lambda x:
        extract_terms(
            x,
            polymer_patterns
        )
)

out["xml_row_carrier"] = out[
    "exact_xml_row_text"
].apply(
    lambda x:
        extract_terms(
            x,
            carrier_patterns
        )
)


# ============================================================
# FORMULATION IDENTIFIERS
# ============================================================

def formulation_id(text):

    patterns = [

        r"\bF[\-\s]?\d+\b",

        r"\bformulation\s*\d+\b",

        r"\bformulation\s*[A-Z]\b",

        r"\bgroup\s*[A-Z]\b",

        r"\bgroup\s*\d+\b",

        r"\bsample\s*[A-Z]\b",

        r"\bsample\s*\d+\b",

        r"\bbatch\s*[A-Z0-9\-]+\b"
    ]

    found = []

    for p in patterns:

        found.extend(
            re.findall(
                p,
                str(text),
                flags=re.I
            )
        )

    found = sorted(
        set(
            x.strip()
            for x in found
        )
    )

    return (
        ";".join(found)
        if found
        else "NOT_IDENTIFIED"
    )


out["xml_row_formulation_id"] = out[
    "exact_xml_row_text"
].apply(
    formulation_id
)


# ============================================================
# pH
# ============================================================

def extract_ph(text):

    matches = re.findall(
        r"\bpH\s*[:=]?\s*"
        r"(\d+(?:\.\d+)?)",
        str(text),
        flags=re.I
    )

    vals = []

    for x in matches:

        try:

            v = float(x)

            if 0 <= v <= 14:
                vals.append(v)

        except:
            pass

    vals = sorted(
        set(vals)
    )

    return (
        ";".join(
            str(v)
            for v in vals
        )
        if vals
        else "NOT_IDENTIFIED"
    )


out["xml_row_pH"] = out[
    "exact_xml_row_text"
].apply(
    extract_ph
)


# ============================================================
# QUALITY CHECK
# ============================================================

def row_quality(row):

    if row[
        "exact_xml_row_status"
    ] != "EXACT_ROW":

        return "NO_EXACT_ROW"

    metadata = [

        row["xml_row_drug"],
        row["xml_row_polymer"],
        row["xml_row_carrier"]
    ]

    identified = sum(
        x != "NOT_IDENTIFIED"
        for x in metadata
    )

    if identified >= 2:
        return "STRONG"

    if identified == 1:
        return "PARTIAL"

    return "WEAK"


out["xml_row_quality"] = out.apply(
    row_quality,
    axis=1
)


# ============================================================
# SAVE
# ============================================================

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "05J_exact_xml_row_audit.csv"
)

STRONG_FILE = os.path.join(
    OUT_DIR,
    "05J_strong_row_metadata_candidates.csv"
)

REVIEW_FILE = os.path.join(
    OUT_DIR,
    "05J_row_metadata_manual_review.csv"
)

out.to_csv(
    AUDIT_FILE,
    index=False
)

strong = out[
    out["xml_row_quality"]
    == "STRONG"
].copy()

review = out[
    out["xml_row_quality"]
    != "STRONG"
].copy()

strong.to_csv(
    STRONG_FILE,
    index=False
)

review.to_csv(
    REVIEW_FILE,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CODE 5J COMPLETE")
print("=" * 70)

print(
    f"Input observations:       {len(out):,}"
)

print(
    "\nExact XML row status:"
)

print(
    out[
        "exact_xml_row_status"
    ].value_counts()
)

print(
    "\nRow metadata quality:"
)

print(
    out[
        "xml_row_quality"
    ].value_counts()
)

print(
    "\nStrong row metadata candidates:",
    len(strong)
)

print(
    "\nDrug extracted from exact row:"
)

print(
    out[
        "xml_row_drug"
    ].value_counts()
    .head(20)
)

print(
    "\nPolymer extracted from exact row:"
)

print(
    out[
        "xml_row_polymer"
    ].value_counts()
    .head(20)
)

print(
    "\nCarrier extracted from exact row:"
)

print(
    out[
        "xml_row_carrier"
    ].value_counts()
    .head(20)
)

print(
    "\nFormulation identifiers found:",
    (
        out[
            "xml_row_formulation_id"
        ]
        != "NOT_IDENTIFIED"
    ).sum()
)

print("\nSaved:")
print(AUDIT_FILE)
print(STRONG_FILE)
print(REVIEW_FILE)

print("\nIMPORTANT:")
print("Metadata are extracted from the exact XML source row.")
print("No synthetic observations were generated.")
print("No interpolation was performed.")
print("No observations were duplicated.")
print("No ML was performed.")
print("=" * 70)
```

```python
# ============================================================
# CODE 5J-R — STRUCTURAL KEY RECOVERY FROM CODE 3
# ============================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 5J-R — STRUCTURAL KEY RECOVERY FROM CODE 3")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

VALIDATED_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05F_valid_release_observations.csv"
)

TABLE_CELLS_FILE = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

os.makedirs(OUT_DIR, exist_ok=True)


# ============================================================
# CHECK FILES
# ============================================================

if not os.path.exists(VALIDATED_FILE):

    raise FileNotFoundError(
        f"Validated dataset not found:\n{VALIDATED_FILE}\n\n"
        "Run Code 5F first."
    )

if not os.path.exists(TABLE_CELLS_FILE):

    raise FileNotFoundError(
        f"Code 3 table-cell file not found:\n{TABLE_CELLS_FILE}\n\n"
        "Run Code 3 first."
    )


# ============================================================
# LOAD VALIDATED OBSERVATIONS
# ============================================================

validated = pd.read_csv(
    VALIDATED_FILE
)

print(
    f"Validated observations: "
    f"{len(validated):,}"
)


# ============================================================
# LOAD CODE 3 STRUCTURAL TABLE CELLS
# ============================================================

cells = pd.read_csv(
    TABLE_CELLS_FILE,
    low_memory=False
)

print(
    f"Code 3 table-cell records: "
    f"{len(cells):,}"
)

print(
    "\nCode 3 columns:"
)

print(
    list(cells.columns)
)


# ============================================================
# NORMALIZE COLUMN NAMES
# ============================================================

cells.columns = [
    str(c).strip()
    for c in cells.columns
]

validated.columns = [
    str(c).strip()
    for c in validated.columns
]


# ============================================================
# IDENTIFY STRUCTURAL COLUMNS
# ============================================================

def find_column(df, candidates):

    lower_map = {
        str(c).lower(): c
        for c in df.columns
    }

    for candidate in candidates:

        if candidate.lower() in lower_map:
            return lower_map[
                candidate.lower()
            ]

    return None


pmcid_col = find_column(
    cells,
    [
        "pmcid",
        "PMCID",
        "article_pmcid"
    ]
)

table_col = find_column(
    cells,
    [
        "table_index",
        "table_idx",
        "table"
    ]
)

row_col = find_column(
    cells,
    [
        "row_index",
        "row_idx",
        "row"
    ]
)

col_col = find_column(
    cells,
    [
        "column_index",
        "column_idx",
        "col_index",
        "column"
    ]
)

text_col = find_column(
    cells,
    [
        "raw_cell",
        "cell_text",
        "text",
        "cell"
    ]
)


print("\nDetected structural columns:")

print(
    "PMCID:",
    pmcid_col
)

print(
    "Table:",
    table_col
)

print(
    "Row:",
    row_col
)

print(
    "Column:",
    col_col
)

print(
    "Cell text:",
    text_col
)


if any(
    x is None
    for x in [
        pmcid_col,
        table_col,
        row_col
    ]
):

    raise ValueError(
        "\nCould not identify the required structural "
        "columns in the Code 3 file."
    )


# ============================================================
# EXTRACT TABLE/ROW FROM ORIGINAL CANDIDATE ID
# ============================================================

def parse_candidate_id(candidate_id):

    if pd.isna(candidate_id):

        return (
            None,
            None,
            None
        )

    s = str(
        candidate_id
    ).strip()

    # Expected:
    # PMC123456_T1_R3
    #
    # Also tolerate:
    # PMC123456_T1_R3_x
    # _T1_R3

    match = re.search(
        r"(PMC\d+).*?"
        r"_T(\d+)"
        r"_R(\d+)",
        s,
        flags=re.I
    )

    if not match:

        return (
            None,
            None,
            None
        )

    pmcid = match.group(1)

    table_index = int(
        match.group(2)
    )

    row_index = int(
        match.group(3)
    )

    return (
        pmcid,
        table_index,
        row_index
    )


parsed = validated[
    "candidate_id"
].apply(
    parse_candidate_id
)

validated[
    "recovered_pmcid"
] = parsed.apply(
    lambda x: x[0]
)

validated[
    "recovered_table_index"
] = parsed.apply(
    lambda x: x[1]
)

validated[
    "recovered_row_index"
] = parsed.apply(
    lambda x: x[2]
)


# ============================================================
# PARSING AUDIT
# ============================================================

print("\nCandidate-ID structural recovery:")

print(
    "Parsed PMCID:",
    validated[
        "recovered_pmcid"
    ].notna().sum()
)

print(
    "Parsed table index:",
    validated[
        "recovered_table_index"
    ].notna().sum()
)

print(
    "Parsed row index:",
    validated[
        "recovered_row_index"
    ].notna().sum()
)


failed_parse = validated[
    validated[
        "recovered_row_index"
    ].isna()
].copy()


if len(failed_parse) > 0:

    print(
        "\nWARNING:",
        len(failed_parse),
        "candidate IDs could not be parsed."
    )

    print(
        failed_parse[
            [
                "candidate_id",
                "pmcid"
            ]
        ].head(20).to_string(
            index=False
        )
    )


# ============================================================
# NORMALIZE STRUCTURAL KEYS
# ============================================================

cells[
    "_pmcid_key"
] = cells[
    pmcid_col
].astype(str).str.strip()

cells[
    "_table_key"
] = pd.to_numeric(
    cells[
        table_col
    ],
    errors="coerce"
)

cells[
    "_row_key"
] = pd.to_numeric(
    cells[
        row_col
    ],
    errors="coerce"
)

validated[
    "_pmcid_key"
] = validated[
    "recovered_pmcid"
].astype(str).str.strip()

validated[
    "_table_key"
] = pd.to_numeric(
    validated[
        "recovered_table_index"
    ],
    errors="coerce"
)

validated[
    "_row_key"
] = pd.to_numeric(
    validated[
        "recovered_row_index"
    ],
    errors="coerce"
)


# ============================================================
# SHOW STRUCTURAL RANGE
# ============================================================

print("\nCode 3 structural ranges:")

print(
    "Unique PMCIDs:",
    cells[
        "_pmcid_key"
    ].nunique()
)

print(
    "Unique tables:",
    cells[
        [
            "_pmcid_key",
            "_table_key"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)

print(
    "Unique rows:",
    cells[
        [
            "_pmcid_key",
            "_table_key",
            "_row_key"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


# ============================================================
# RECOVER EXACT SOURCE ROW
# ============================================================

# We don't merge directly on cells because each row has
# multiple cells.
#
# First create one record per:
# PMCID + table + row
#
# and preserve the individual cells as ordered text.

group_cols = [
    "_pmcid_key",
    "_table_key",
    "_row_key"
]


def combine_cells(group):

    g = group.copy()

    if col_col is not None:

        g["_col_sort"] = pd.to_numeric(
            g[col_col],
            errors="coerce"
        )

        g = g.sort_values(
            "_col_sort",
            kind="stable"
        )

    else:

        g = g.copy()

    if text_col is not None:

        values = (
            g[text_col]
            .fillna("")
            .astype(str)
            .tolist()
        )

    else:

        values = []

    values = [
        x.strip()
        for x in values
        if x.strip()
    ]

    return " | ".join(
        values
    )


row_table = (
    cells
    .groupby(
        group_cols,
        dropna=False,
        sort=False
    )
    .apply(
        combine_cells
    )
    .reset_index(
        name="recovered_source_row"
    )
)


print(
    "\nUnique structural rows in Code 3:",
    len(row_table):,
)
```

```python
# ============================================================
# CODE 5J-R — STRUCTURAL KEY RECOVERY FROM CODE 3
# ============================================================

import os
import re
import glob
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 5J-R — STRUCTURAL KEY RECOVERY FROM CODE 3")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

VALIDATED_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05F_valid_release_observations.csv"
)

TABLE_CELLS_FILE = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

os.makedirs(OUT_DIR, exist_ok=True)


# ============================================================
# CHECK FILES
# ============================================================

if not os.path.exists(VALIDATED_FILE):
    raise FileNotFoundError(
        f"\nValidated dataset not found:\n{VALIDATED_FILE}\n\n"
        "Run Code 5F first."
    )

if not os.path.exists(TABLE_CELLS_FILE):
    raise FileNotFoundError(
        f"\nCode 3 table-cell file not found:\n"
        f"{TABLE_CELLS_FILE}\n\n"
        "Run Code 3 first."
    )


# ============================================================
# LOAD DATA
# ============================================================

validated = pd.read_csv(
    VALIDATED_FILE,
    low_memory=False
)

cells = pd.read_csv(
    TABLE_CELLS_FILE,
    low_memory=False
)

print(
    f"Validated observations: {len(validated):,}"
)

print(
    f"Code 3 table-cell records: {len(cells):,}"
)

print("\nCode 3 columns:")
print(list(cells.columns))


# ============================================================
# NORMALIZE COLUMN NAMES
# ============================================================

validated.columns = [
    str(c).strip()
    for c in validated.columns
]

cells.columns = [
    str(c).strip()
    for c in cells.columns
]


# ============================================================
# COLUMN DETECTION
# ============================================================

def find_column(df, candidates):

    lower_map = {
        str(c).strip().lower(): c
        for c in df.columns
    }

    for candidate in candidates:

        key = candidate.lower()

        if key in lower_map:
            return lower_map[key]

    return None


pmcid_col = find_column(
    cells,
    [
        "pmcid",
        "article_pmcid",
        "PMCID"
    ]
)

table_col = find_column(
    cells,
    [
        "table_index",
        "table_idx"
    ]
)

row_col = find_column(
    cells,
    [
        "row_index",
        "row_idx"
    ]
)

col_col = find_column(
    cells,
    [
        "column_index",
        "column_idx",
        "col_index"
    ]
)

text_col = find_column(
    cells,
    [
        "raw_cell",
        "cell_text",
        "text",
        "cell"
    ]
)


print("\nDetected structural columns:")
print("PMCID      :", pmcid_col)
print("Table      :", table_col)
print("Row        :", row_col)
print("Column     :", col_col)
print("Cell text  :", text_col)


required_structural = [
    pmcid_col,
    table_col,
    row_col,
    text_col
]

if any(
    x is None
    for x in required_structural
):

    raise ValueError(
        "\nCould not identify all required Code 3 "
        "structural columns."
    )


# ============================================================
# SHOW EXAMPLE CODE 3 STRUCTURE
# ============================================================

print("\nExample Code 3 records:")

print(
    cells[
        [
            pmcid_col,
            table_col,
            row_col,
            col_col,
            text_col
        ]
    ].head(10).to_string(
        index=False
    )
)


# ============================================================
# PARSE ORIGINAL CANDIDATE ID
# ============================================================

def parse_candidate_id(candidate_id):

    if pd.isna(candidate_id):
        return None, None, None

    s = str(
        candidate_id
    ).strip()

    # Expected:
    # PMC123456_T1_R3
    #
    # Also tolerate:
    # _T1_R3
    # PMC123456_T1_R3_x

    match = re.search(
        r"(PMC\d+).*?_T(\d+)_R(\d+)",
        s,
        flags=re.I
    )

    if not match:
        return None, None, None

    pmcid = match.group(1)

    table_index = int(
        match.group(2)
    )

    row_index = int(
        match.group(3)
    )

    return (
        pmcid,
        table_index,
        row_index
    )


# ============================================================
# FIND CANDIDATE-ID COLUMN
# ============================================================

candidate_id_col = find_column(
    validated,
    [
        "candidate_id",
        "observation_candidate_id",
        "pairing_candidate_id"
    ]
)

if candidate_id_col is None:

    raise ValueError(
        "\nNo candidate ID column found in "
        "validated dataset."
    )


print(
    "\nCandidate ID column:",
    candidate_id_col
)


# ============================================================
# RECOVER STRUCTURAL COORDINATES
# ============================================================

parsed = validated[
    candidate_id_col
].apply(
    parse_candidate_id
)

validated[
    "recovered_pmcid"
] = parsed.apply(
    lambda x: x[0]
)

validated[
    "recovered_table_index"
] = parsed.apply(
    lambda x: x[1]
)

validated[
    "recovered_row_index"
] = parsed.apply(
    lambda x: x[2]
)


print("\nCandidate-ID recovery:")

print(
    "PMCID recovered:",
    validated[
        "recovered_pmcid"
    ].notna().sum()
)

print(
    "Table index recovered:",
    validated[
        "recovered_table_index"
    ].notna().sum()
)

print(
    "Row index recovered:",
    validated[
        "recovered_row_index"
    ].notna().sum()
)


# ============================================================
# FALLBACK: USE PMCID COLUMN IF CANDIDATE ID LACKS PMCID
# ============================================================

if "pmcid" in validated.columns:

    mask = (
        validated[
            "recovered_pmcid"
        ].isna()
    )

    validated.loc[
        mask,
        "recovered_pmcid"
    ] = validated.loc[
        mask,
        "pmcid"
    ]


# ============================================================
# NORMALIZE STRUCTURAL KEYS
# ============================================================

cells[
    "_pmcid_key"
] = cells[
    pmcid_col
].astype(str).str.strip().str.upper()

cells[
    "_table_key"
] = pd.to_numeric(
    cells[
        table_col
    ],
    errors="coerce"
)

cells[
    "_row_key"
] = pd.to_numeric(
    cells[
        row_col
    ],
    errors="coerce"
)

validated[
    "_pmcid_key"
] = validated[
    "recovered_pmcid"
].astype(str).str.strip().str.upper()

validated[
    "_table_key"
] = pd.to_numeric(
    validated[
        "recovered_table_index"
    ],
    errors="coerce"
)

validated[
    "_row_key"
] = pd.to_numeric(
    validated[
        "recovered_row_index"
    ],
    errors="coerce"
)


# ============================================================
# IMPORTANT:
# CHECK WHETHER CODE 3 USES ZERO-BASED OR ONE-BASED ROWS
# ============================================================

print("\nStructural row-index diagnostics:")

print(
    "Minimum Code 3 row index:",
    cells[
        "_row_key"
    ].min()
)

print(
    "Maximum Code 3 row index:",
    cells[
        "_row_key"
    ].max()
)

print(
    "Minimum Code 3 table index:",
    cells[
        "_table_key"
    ].min()
)

print(
    "Maximum Code 3 table index:",
    cells[
        "_table_key"
    ].max()
)


# ============================================================
# CREATE ROW-LEVEL TABLE FROM CODE 3
# ============================================================

def combine_cells(group):

    g = group.copy()

    if col_col is not None:

        g["_col_sort"] = pd.to_numeric(
            g[col_col],
            errors="coerce"
        )

        g = g.sort_values(
            "_col_sort",
            kind="stable"
        )

    values = (
        g[text_col]
        .fillna("")
        .astype(str)
        .tolist()
    )

    values = [
        x.strip()
        for x in values
        if x.strip()
    ]

    return " | ".join(
        values
    )


group_cols = [
    "_pmcid_key",
    "_table_key",
    "_row_key"
]


row_table = (
    cells
    .groupby(
        group_cols,
        dropna=False,
        sort=False
    )
    .apply(
        combine_cells
    )
    .reset_index(
        name="recovered_source_row"
    )
)


print(
    "\nUnique structural rows in Code 3:",
    f"{len(row_table):,}"
)


# ============================================================
# DIRECT STRUCTURAL JOIN
# ============================================================

recovered = validated.merge(
    row_table,
    how="left",
    on=[
        "_pmcid_key",
        "_table_key",
        "_row_key"
    ]
)


recovered[
    "structural_row_recovered"
] = recovered[
    "recovered_source_row"
].notna()


print(
    "\nDirect structural recovery:"
)

print(
    recovered[
        "structural_row_recovered"
    ].value_counts()
)


# ============================================================
# SECOND ATTEMPT:
# TEST ROW INDEX +1 AND -1
#
# This is ONLY for locating the original source row.
# It does NOT alter the observation's actual time/release.
# ============================================================

unrecovered = (
    ~recovered[
        "structural_row_recovered"
    ]
)

if unrecovered.any():

    print(
        "\nAttempting row-index offset recovery..."
    )

    # +1
    plus_one = validated.copy()

    plus_one[
        "_row_key_offset"
    ] = plus_one[
        "_row_key"
    ] + 1

    plus_one = plus_one.merge(
        row_table,
        how="left",
        left_on=[
            "_pmcid_key",
            "_table_key",
            "_row_key_offset"
        ],
        right_on=[
            "_pmcid_key",
            "_table_key",
            "_row_key"
        ],
        suffixes=(
            "",
            "_source"
        )
    )

    plus_one_found = (
        plus_one[
            "recovered_source_row"
        ].notna()
    )

    # -1
    minus_one = validated.copy()

    minus_one[
        "_row_key_offset"
    ] = minus_one[
        "_row_key"
    ] - 1

    minus_one = minus_one.merge(
        row_table,
        how="left",
        left_on=[
            "_pmcid_key",
            "_table_key",
            "_row_key_offset"
        ],
        right_on=[
            "_pmcid_key",
            "_table_key",
            "_row_key"
        ],
        suffixes=(
            "",
            "_source"
        )
    )

    minus_one_found = (
        minus_one[
            "recovered_source_row"
        ].notna()
    )

    print(
        "Potential +1 row recoveries:",
        plus_one_found.sum()
    )

    print(
        "Potential -1 row recoveries:",
        minus_one_found.sum()
    )

else:

    plus_one = None
    minus_one = None


# ============================================================
# DO NOT AUTOMATICALLY ACCEPT OFFSET RECOVERY
#
# We only use exact direct matches for now.
# Offset candidates are saved separately for audit.
# ============================================================

offset_audit = []

for i in range(
    len(validated)
):

    direct = (
        recovered.iloc[i][
            "structural_row_recovered"
        ]
    )

    if direct:
        status = "DIRECT_MATCH"
    else:

        p = (
            plus_one.iloc[i][
                "recovered_source_row"
            ]
            if plus_one is not None
            else np.nan
        )

        m = (
            minus_one.iloc[i][
                "recovered_source_row"
            ]
            if minus_one is not None
            else np.nan
        )

        if pd.notna(p) and pd.notna(m):

            status = (
                "OFFSET_AMBIGUOUS"
            )

        elif pd.notna(p):

            status = (
                "OFFSET_PLUS_ONE_CANDIDATE"
            )

        elif pd.notna(m):

            status = (
                "OFFSET_MINUS_ONE_CANDIDATE"
            )

        else:

            status = "NO_MATCH"

    offset_audit.append(
        status
    )


recovered[
    "structural_recovery_status"
] = offset_audit


# ============================================================
# EXACT ROW METADATA EXTRACTION
# ============================================================

drug_patterns = {

    "dexamethasone":
        r"\bdexamethasone\b",

    "paclitaxel":
        r"\bpaclitaxel\b",

    "curcumin":
        r"\bcurcumin\b",

    "doxorubicin":
        r"\bdoxorubicin\b",

    "ibuprofen":
        r"\bibuprofen\b",

    "metformin":
        r"\bmetformin\b",

    "gemcitabine":
        r"\bgemcitabine\b",

    "rifampicin":
        r"\brifampicin\b|\brifampin\b",

    "vancomycin":
        r"\bvancomycin\b",

    "insulin":
        r"\binsulin\b",

    "5-fluorouracil":
        r"\b5[\s\-]?fluorouracil\b|\b5fu\b",

    "DOX":
        r"\bDOX\b"
}


polymer_patterns = {

    "PLGA":
        r"\bPLGA\b|poly\(lactic[\s\-]?co[\s\-]?glycolic acid\)",

    "PLA":
        r"\bPLA\b|polylactic acid|poly\(lactic acid\)",

    "PCL":
        r"\bPCL\b|polycaprolactone",

    "PEG":
        r"\bPEG\b|polyethylene glycol",

    "PEO":
        r"\bPEO\b|polyethylene oxide",

    "chitosan":
        r"\bchitosan\b",

    "alginate":
        r"\balginate\b",

    "gelatin":
        r"\bgelatin\b",

    "PVA":
        r"\bPVA\b|polyvinyl alcohol",

    "poloxamer":
        r"\bpoloxamer\b",

    "polyurethane":
        r"\bpolyurethane\b",

    "cellulose":
        r"\bcellulose\b"
}


carrier_patterns = {

    "nanoparticle":
        r"\bnanoparticle[s]?\b",

    "nanocarrier":
        r"\bnanocarrier[s]?\b",

    "micelle":
        r"\bmicelle[s]?\b",

    "liposome":
        r"\bliposome[s]?\b",

    "hydrogel":
        r"\bhydrogel[s]?\b",

    "microsphere":
        r"\bmicrosphere[s]?\b",

    "microparticle":
        r"\bmicroparticle[s]?\b",

    "nanofiber":
        r"\bnanofiber[s]?\b",

    "film":
        r"\bfilm[s]?\b",

    "implant":
        r"\bimplant[s]?\b",

    "beads":
        r"\bbead[s]?\b"
}


def extract_terms(
    text,
    patterns
):

    if pd.isna(text):
        return "NOT_IDENTIFIED"

    text = str(text)

    found = []

    for name, pattern in patterns.items():

        if re.search(
            pattern,
            text,
            flags=re.I
        ):
            found.append(name)

    return (
        ";".join(found)
        if found
        else "NOT_IDENTIFIED"
    )


recovered[
    "xml_row_drug"
] = recovered[
    "recovered_source_row"
].apply(
    lambda x:
        extract_terms(
            x,
            drug_patterns
        )
)

recovered[
    "xml_row_polymer"
] = recovered[
    "recovered_source_row"
].apply(
    lambda x:
        extract_terms(
            x,
            polymer_patterns
        )
)

recovered[
    "xml_row_carrier"
] = recovered[
    "recovered_source_row"
].apply(
    lambda x:
        extract_terms(
            x,
            carrier_patterns
        )
)


# ============================================================
# FORMULATION IDENTIFIER
# ============================================================

def extract_formulation_id(text):

    if pd.isna(text):
        return "NOT_IDENTIFIED"

    patterns = [

        r"\bF[\-\s]?\d+\b",

        r"\bformulation\s*\d+\b",

        r"\bformulation\s*[A-Z]\b",

        r"\bgroup\s*[A-Z]\b",

        r"\bgroup\s*\d+\b",

        r"\bsample\s*[A-Z]\b",

        r"\bsample\s*\d+\b",

        r"\bbatch\s*[A-Z0-9\-]+\b"
    ]

    found = []

    for pattern in patterns:

        found.extend(
            re.findall(
                pattern,
                str(text),
                flags=re.I
            )
        )

    found = sorted(
        set(
            x.strip()
            for x in found
        )
    )

    return (
        ";".join(found)
        if found
        else "NOT_IDENTIFIED"
    )


recovered[
    "xml_row_formulation_id"
] = recovered[
    "recovered_source_row"
].apply(
    extract_formulation_id
)


# ============================================================
# ROW METADATA QUALITY
# ============================================================

def metadata_quality(row):

    if not row[
        "structural_row_recovered"
    ]:

        return "NO_STRUCTURAL_ROW"

    values = [

        row["xml_row_drug"],

        row["xml_row_polymer"],

        row["xml_row_carrier"]
    ]

    identified = sum(
        x != "NOT_IDENTIFIED"
        for x in values
    )

    if identified >= 2:
        return "STRONG"

    if identified == 1:
        return "PARTIAL"

    return "WEAK"


recovered[
    "xml_row_quality"
] = recovered.apply(
    metadata_quality,
    axis=1
)


# ============================================================
# SAVE FILES
# ============================================================

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "05JR_structural_row_recovery_audit.csv"
)

STRONG_FILE = os.path.join(
    OUT_DIR,
    "05JR_strong_row_metadata.csv"
)

REVIEW_FILE = os.path.join(
    OUT_DIR,
    "05JR_row_metadata_review.csv"
)

OFFSET_FILE = os.path.join(
    OUT_DIR,
    "05JR_offset_recovery_candidates.csv"
)


recovered.to_csv(
    AUDIT_FILE,
    index=False
)


strong = recovered[
    recovered[
        "xml_row_quality"
    ] == "STRONG"
].copy()


review = recovered[
    recovered[
        "xml_row_quality"
    ] != "STRONG"
].copy()


offset_candidates = recovered[
    recovered[
        "structural_recovery_status"
    ].isin(
        [
            "OFFSET_PLUS_ONE_CANDIDATE",
            "OFFSET_MINUS_ONE_CANDIDATE",
            "OFFSET_AMBIGUOUS"
        ]
    )
].copy()


strong.to_csv(
    STRONG_FILE,
    index=False
)

review.to_csv(
    REVIEW_FILE,
    index=False
)

offset_candidates.to_csv(
    OFFSET_FILE,
    index=False
)


# ============================================================
# FINAL REPORT
# ============================================================

print("\n" + "=" * 70)
print("CODE 5J-R COMPLETE")
print("=" * 70)

print(
    f"Validated observations: "
    f"{len(recovered):,}"
)

print(
    "\nStructural recovery status:"
)

print(
    recovered[
        "structural_recovery_status"
    ].value_counts()
)

print(
    "\nExact direct structural matches:",
    (
        recovered[
            "structural_recovery_status"
        ] == "DIRECT_MATCH"
    ).sum()
)

print(
    "\nOffset recovery candidates:",
    len(offset_candidates)
)

print(
    "\nRow metadata quality:"
)

print(
    recovered[
        "xml_row_quality"
    ].value_counts()
)

print(
    "\nStrong exact-row metadata:",
    len(strong)
)

print(
    "\nDrug from exact structural row:"
)

print(
    recovered[
        "xml_row_drug"
    ].value_counts()
    .head(20)
)

print(
    "\nPolymer from exact structural row:"
)

print(
    recovered[
        "xml_row_polymer"
    ].value_counts()
    .head(20)
)

print(
    "\nCarrier from exact structural row:"
)

print(
    recovered[
        "xml_row_carrier"
    ].value_counts()
    .head(20)
)

print(
    "\nFormulation identifiers:",
    (
        recovered[
            "xml_row_formulation_id"
        ]
        != "NOT_IDENTIFIED"
    ).sum()
)

print("\nSaved:")
print(AUDIT_FILE)
print(STRONG_FILE)
print(REVIEW_FILE)
print(OFFSET_FILE)

print("\nIMPORTANT:")
print(
    "Direct structural matches use the original Code 3 "
    "PMCID/table/row coordinates."
)
print(
    "Offset matches are saved for audit only and are "
    "NOT automatically accepted."
)
print(
    "No synthetic observations were generated."
)
print(
    "No interpolation was performed."
)
print(
    "No observations were duplicated."
)
print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ============================================================
# CODE 5J-R CONTINUATION — STRUCTURAL ROW MATCH + AUDIT
# ============================================================

print("=" * 70)
print("CODE 5J-R CONTINUATION — STRUCTURAL ROW MATCH")
print("=" * 70)


# ============================================================
# CHECK REQUIRED OBJECTS
# ============================================================

required_objects = [
    "validated",
    "cells",
    "row_table",
    "pmcid_col",
    "table_col",
    "row_col",
    "col_col",
    "text_col"
]

missing_objects = [
    x for x in required_objects
    if x not in globals()
]

if missing_objects:

    raise RuntimeError(
        "\nRequired objects are missing:\n"
        + "\n".join(missing_objects)
        + "\n\n"
        "If the previous 5J-R cell was restarted or its "
        "runtime state was lost, rerun the complete 5J-R cell."
    )


# ============================================================
# NORMALIZE KEYS
# ============================================================

cells["_pmcid_key"] = (
    cells[pmcid_col]
    .astype(str)
    .str.strip()
    .str.upper()
)

cells["_table_key"] = pd.to_numeric(
    cells[table_col],
    errors="coerce"
)

cells["_row_key"] = pd.to_numeric(
    cells[row_col],
    errors="coerce"
)

validated["_pmcid_key"] = (
    validated["recovered_pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

validated["_table_key"] = pd.to_numeric(
    validated["recovered_table_index"],
    errors="coerce"
)

validated["_row_key"] = pd.to_numeric(
    validated["recovered_row_index"],
    errors="coerce"
)


# ============================================================
# DIRECT MATCH
# ============================================================

recovered = validated.merge(
    row_table,
    how="left",
    on=[
        "_pmcid_key",
        "_table_key",
        "_row_key"
    ]
)

recovered["structural_row_recovered"] = (
    recovered["recovered_source_row"]
    .notna()
)

print("\nDIRECT STRUCTURAL RECOVERY")

print(
    recovered[
        "structural_row_recovered"
    ].value_counts()
)


# ============================================================
# OFFSET DIAGNOSTIC
#
# Some extraction pipelines use different row numbering
# conventions. We test +/-1 ONLY as a diagnostic.
# We do NOT automatically accept these.
# ============================================================

plus = validated.copy()

plus["_row_test"] = (
    plus["_row_key"] + 1
)

plus = plus.merge(
    row_table,
    how="left",
    left_on=[
        "_pmcid_key",
        "_table_key",
        "_row_test"
    ],
    right_on=[
        "_pmcid_key",
        "_table_key",
        "_row_key"
    ],
    suffixes=(
        "",
        "_source"
    )
)

minus = validated.copy()

minus["_row_test"] = (
    minus["_row_key"] - 1
)

minus = minus.merge(
    row_table,
    how="left",
    left_on=[
        "_pmcid_key",
        "_table_key",
        "_row_test"
    ],
    right_on=[
        "_pmcid_key",
        "_table_key",
        "_row_key"
    ],
    suffixes=(
        "",
        "_source"
    )
)

plus_found = (
    plus["recovered_source_row"]
    .notna()
)

minus_found = (
    minus["recovered_source_row"]
    .notna()
)

print(
    "\nOFFSET DIAGNOSTIC"
)

print(
    "Potential +1 matches:",
    int(plus_found.sum())
)

print(
    "Potential -1 matches:",
    int(minus_found.sum())
)


# ============================================================
# CLASSIFY STRUCTURAL RECOVERY
# ============================================================

statuses = []

for i in range(
    len(recovered)
):

    direct = bool(
        recovered.iloc[i][
            "structural_row_recovered"
        ]
    )

    p = bool(
        plus_found.iloc[i]
    )

    m = bool(
        minus_found.iloc[i]
    )

    if direct:

        statuses.append(
            "DIRECT_MATCH"
        )

    elif p and m:

        statuses.append(
            "OFFSET_AMBIGUOUS"
        )

    elif p:

        statuses.append(
            "OFFSET_PLUS_ONE_CANDIDATE"
        )

    elif m:

        statuses.append(
            "OFFSET_MINUS_ONE_CANDIDATE"
        )

    else:

        statuses.append(
            "NO_MATCH"
        )


recovered[
    "structural_recovery_status"
] = statuses


print(
    "\nSTRUCTURAL RECOVERY STATUS"
)

print(
    recovered[
        "structural_recovery_status"
    ].value_counts()
)


# ============================================================
# EXACT ROW METADATA
# ============================================================

drug_patterns = {

    "dexamethasone":
        r"\bdexamethasone\b",

    "paclitaxel":
        r"\bpaclitaxel\b",

    "curcumin":
        r"\bcurcumin\b",

    "doxorubicin":
        r"\bdoxorubicin\b",

    "ibuprofen":
        r"\bibuprofen\b",

    "metformin":
        r"\bmetformin\b",

    "gemcitabine":
        r"\bgemcitabine\b",

    "rifampicin":
        r"\brifampicin\b|\brifampin\b",

    "vancomycin":
        r"\bvancomycin\b",

    "insulin":
        r"\binsulin\b",

    "5-fluorouracil":
        r"\b5[\s\-]?fluorouracil\b|\b5fu\b",

    "DOX":
        r"\bDOX\b"
}


polymer_patterns = {

    "PLGA":
        r"\bPLGA\b|poly\(lactic[\s\-]?co[\s\-]?glycolic acid\)",

    "PLA":
        r"\bPLA\b|polylactic acid|poly\(lactic acid\)",

    "PCL":
        r"\bPCL\b|polycaprolactone",

    "PEG":
        r"\bPEG\b|polyethylene glycol",

    "PEO":
        r"\bPEO\b|polyethylene oxide",

    "chitosan":
        r"\bchitosan\b",

    "alginate":
        r"\balginate\b",

    "gelatin":
        r"\bgelatin\b",

    "PVA":
        r"\bPVA\b|polyvinyl alcohol",

    "poloxamer":
        r"\bpoloxamer\b",

    "polyurethane":
        r"\bpolyurethane\b",

    "cellulose":
        r"\bcellulose\b"
}


carrier_patterns = {

    "nanoparticle":
        r"\bnanoparticle[s]?\b",

    "nanocarrier":
        r"\bnanocarrier[s]?\b",

    "micelle":
        r"\bmicelle[s]?\b",

    "liposome":
        r"\bliposome[s]?\b",

    "hydrogel":
        r"\bhydrogel[s]?\b",

    "microsphere":
        r"\bmicrosphere[s]?\b",

    "microparticle":
        r"\bmicroparticle[s]?\b",

    "nanofiber":
        r"\bnanofiber[s]?\b",

    "film":
        r"\bfilm[s]?\b",

    "implant":
        r"\bimplant[s]?\b",

    "beads":
        r"\bbead[s]?\b"
}


def extract_terms(text, patterns):

    if pd.isna(text):
        return "NOT_IDENTIFIED"

    found = []

    for name, pattern in patterns.items():

        if re.search(
            pattern,
            str(text),
            flags=re.I
        ):
            found.append(name)

    if found:
        return ";".join(found)

    return "NOT_IDENTIFIED"


recovered["xml_row_drug"] = (
    recovered["recovered_source_row"]
    .apply(
        lambda x:
        extract_terms(
            x,
            drug_patterns
        )
    )
)

recovered["xml_row_polymer"] = (
    recovered["recovered_source_row"]
    .apply(
        lambda x:
        extract_terms(
            x,
            polymer_patterns
        )
    )
)

recovered["xml_row_carrier"] = (
    recovered["recovered_source_row"]
    .apply(
        lambda x:
        extract_terms(
            x,
            carrier_patterns
        )
    )
)


# ============================================================
# FORMULATION IDENTIFIER
# ============================================================

def extract_formulation_id(text):

    if pd.isna(text):
        return "NOT_IDENTIFIED"

    patterns = [

        r"\bF[\-\s]?\d+\b",

        r"\bformulation\s*\d+\b",

        r"\bformulation\s*[A-Z]\b",

        r"\bgroup\s*[A-Z]\b",

        r"\bgroup\s*\d+\b",

        r"\bsample\s*[A-Z]\b",

        r"\bsample\s*\d+\b",

        r"\bbatch\s*[A-Z0-9\-]+\b"
    ]

    found = []

    for pattern in patterns:

        found.extend(
            re.findall(
                pattern,
                str(text),
                flags=re.I
            )
        )

    found = sorted(
        set(
            x.strip()
            for x in found
        )
    )

    return (
        ";".join(found)
        if found
        else "NOT_IDENTIFIED"
    )


recovered["xml_row_formulation_id"] = (
    recovered["recovered_source_row"]
    .apply(
        extract_formulation_id
    )
)


# ============================================================
# METADATA QUALITY
# ============================================================

def metadata_quality(row):

    if row[
        "structural_recovery_status"
    ] != "DIRECT_MATCH":

        return "NOT_EXACT"

    values = [
        row["xml_row_drug"],
        row["xml_row_polymer"],
        row["xml_row_carrier"]
    ]

    identified = sum(
        x != "NOT_IDENTIFIED"
        for x in values
    )

    if identified >= 2:
        return "STRONG"

    if identified == 1:
        return "PARTIAL"

    return "WEAK"


recovered["xml_row_quality"] = (
    recovered.apply(
        metadata_quality,
        axis=1
    )
)


# ============================================================
# SAVE
# ============================================================

AUDIT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05JR_structural_row_recovery_audit.csv"
)

STRONG_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05JR_strong_row_metadata.csv"
)

REVIEW_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05JR_row_metadata_review.csv"
)

OFFSET_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05JR_offset_recovery_candidates.csv"
)


recovered.to_csv(
    AUDIT_FILE,
    index=False
)

strong = recovered[
    recovered[
        "xml_row_quality"
    ] == "STRONG"
].copy()

review = recovered[
    recovered[
        "xml_row_quality"
    ] != "STRONG"
].copy()

offset_candidates = recovered[
    recovered[
        "structural_recovery_status"
    ].isin(
        [
            "OFFSET_PLUS_ONE_CANDIDATE",
            "OFFSET_MINUS_ONE_CANDIDATE",
            "OFFSET_AMBIGUOUS"
        ]
    )
].copy()

strong.to_csv(
    STRONG_FILE,
    index=False
)

review.to_csv(
    REVIEW_FILE,
    index=False
)

offset_candidates.to_csv(
    OFFSET_FILE,
    index=False
)


# ============================================================
# FINAL REPORT
# ============================================================

print("\n" + "=" * 70)
print("CODE 5J-R COMPLETE")
print("=" * 70)

print(
    f"Validated observations: "
    f"{len(recovered):,}"
)

print(
    "\nSTRUCTURAL RECOVERY:"
)

print(
    recovered[
        "structural_recovery_status"
    ].value_counts()
)

print(
    "\nDirect exact-row matches:",
    int(
        (
            recovered[
                "structural_recovery_status"
            ]
            == "DIRECT_MATCH"
        ).sum()
    )
)

print(
    "\nOffset candidates:",
    len(offset_candidates)
)

print(
    "\nROW METADATA QUALITY:"
)

print(
    recovered[
        "xml_row_quality"
    ].value_counts()
)

print(
    "\nDrug from exact structural row:"
)

print(
    recovered[
        "xml_row_drug"
    ].value_counts()
    .head(20)
)

print(
    "\nPolymer from exact structural row:"
)

print(
    recovered[
        "xml_row_polymer"
    ].value_counts()
    .head(20)
)

print(
    "\nCarrier from exact structural row:"
)

print(
    recovered[
        "xml_row_carrier"
    ].value_counts()
    .head(20)
)

print(
    "\nFormulation identifiers:",
    int(
        (
            recovered[
                "xml_row_formulation_id"
            ]
            != "NOT_IDENTIFIED"
        ).sum()
    )
)

print("\nSaved:")
print(AUDIT_FILE)
print(STRONG_FILE)
print(REVIEW_FILE)
print(OFFSET_FILE)

print("\nIMPORTANT:")
print(
    "Direct matches are the only rows currently "
    "considered structurally recovered."
)
print(
    "Offset matches are diagnostic only."
)
print(
    "No synthetic observations were generated."
)
print(
    "No interpolation was performed."
)
print(
    "No observations were duplicated."
)
print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ============================================================
# CODE 5K — FORMULATION CONTEXT RESOLUTION
# ============================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 5K — FORMULATION CONTEXT RESOLUTION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05JR_structural_row_recovery_audit.csv"
)

CODE3_FILE = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"\nRequired 5J-R file not found:\n{INPUT_FILE}\n"
        "\nRun Code 5J-R first."
    )

if not os.path.exists(CODE3_FILE):
    raise FileNotFoundError(
        f"\nRequired Code 3 file not found:\n{CODE3_FILE}"
    )


# ============================================================
# LOAD
# ============================================================

obs = pd.read_csv(
    INPUT_FILE,
    low_memory=False
)

cells = pd.read_csv(
    CODE3_FILE,
    low_memory=False
)

print(
    f"5J-R observations: {len(obs):,}"
)

print(
    f"Code 3 cells: {len(cells):,}"
)


# ============================================================
# NORMALIZE COLUMN NAMES
# ============================================================

obs.columns = [
    str(c).strip()
    for c in obs.columns
]

cells.columns = [
    str(c).strip()
    for c in cells.columns
]


# ============================================================
# CHECK REQUIRED COLUMNS
# ============================================================

required_obs = [
    "recovered_pmcid",
    "recovered_table_index",
    "recovered_row_index",
    "structural_recovery_status",
    "recovered_source_row"
]

missing_obs = [
    c for c in required_obs
    if c not in obs.columns
]

if missing_obs:
    raise ValueError(
        f"Missing 5J-R columns: {missing_obs}"
    )


required_cells = [
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell",
    "row_text",
    "full_table_text"
]

missing_cells = [
    c for c in required_cells
    if c not in cells.columns
]

if missing_cells:
    raise ValueError(
        f"Missing Code 3 columns: {missing_cells}"
    )


# ============================================================
# NORMALIZE KEYS
# ============================================================

obs["_pmcid"] = (
    obs["recovered_pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

obs["_table"] = pd.to_numeric(
    obs["recovered_table_index"],
    errors="coerce"
)

obs["_row"] = pd.to_numeric(
    obs["recovered_row_index"],
    errors="coerce"
)

cells["_pmcid"] = (
    cells["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

cells["_table"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

cells["_row"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

cells["_col"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)


# ============================================================
# TEXT NORMALIZATION
# ============================================================

def clean(x):

    if pd.isna(x):
        return ""

    x = str(x)

    x = x.replace(
        "\n",
        " "
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


# ============================================================
# TERM DICTIONARIES
# ============================================================

drug_patterns = {

    "dexamethasone":
        r"\bdexamethasone\b",

    "paclitaxel":
        r"\bpaclitaxel\b",

    "curcumin":
        r"\bcurcumin\b",

    "doxorubicin":
        r"\bdoxorubicin\b",

    "ibuprofen":
        r"\bibuprofen\b",

    "metformin":
        r"\bmetformin\b",

    "gemcitabine":
        r"\bgemcitabine\b",

    "rifampicin":
        r"\brifampicin\b|\brifampin\b",

    "vancomycin":
        r"\bvancomycin\b",

    "insulin":
        r"\binsulin\b",

    "5-fluorouracil":
        r"\b5[\s\-]?fluorouracil\b|\b5fu\b",

    "DOX":
        r"\bDOX\b"
}


polymer_patterns = {

    "PLGA":
        r"\bPLGA\b|poly\(lactic[\s\-]?co[\s\-]?glycolic acid\)",

    "PLA":
        r"\bPLA\b|polylactic acid|poly\(lactic acid\)",

    "PCL":
        r"\bPCL\b|polycaprolactone",

    "PEG":
        r"\bPEG\b|polyethylene glycol",

    "PEO":
        r"\bPEO\b|polyethylene oxide",

    "chitosan":
        r"\bchitosan\b",

    "alginate":
        r"\balginate\b",

    "gelatin":
        r"\bgelatin\b",

    "PVA":
        r"\bPVA\b|polyvinyl alcohol",

    "poloxamer":
        r"\bpoloxamer\b",

    "polyurethane":
        r"\bpolyurethane\b",

    "cellulose":
        r"\bcellulose\b"
}


carrier_patterns = {

    "nanoparticle":
        r"\bnanoparticle[s]?\b",

    "nanocarrier":
        r"\bnanocarrier[s]?\b",

    "micelle":
        r"\bmicelle[s]?\b",

    "liposome":
        r"\bliposome[s]?\b",

    "hydrogel":
        r"\bhydrogel[s]?\b",

    "microsphere":
        r"\bmicrosphere[s]?\b",

    "microparticle":
        r"\bmicroparticle[s]?\b",

    "nanofiber":
        r"\bnanofiber[s]?\b",

    "film":
        r"\bfilm[s]?\b",

    "implant":
        r"\bimplant[s]?\b",

    "beads":
        r"\bbead[s]?\b"
}


def extract_terms(
    text,
    patterns
):

    text = clean(text)

    if not text:
        return "NOT_IDENTIFIED"

    found = []

    for name, pattern in patterns.items():

        if re.search(
            pattern,
            text,
            flags=re.I
        ):
            found.append(name)

    return (
        ";".join(found)
        if found
        else "NOT_IDENTIFIED"
    )


# ============================================================
# FORMULATION IDENTIFIERS
# ============================================================

def formulation_terms(text):

    text = clean(text)

    if not text:
        return []

    patterns = [

        r"\bF[\-\s]?\d+\b",

        r"\bformulation\s*\d+\b",

        r"\bformulation\s*[A-Z]\b",

        r"\bgroup\s*[A-Z]\b",

        r"\bgroup\s*\d+\b",

        r"\bsample\s*[A-Z]\b",

        r"\bsample\s*\d+\b",

        r"\bbatch\s*[A-Z0-9\-]+\b",

        r"\btest\s*[A-Z]\b"
    ]

    found = []

    for pattern in patterns:

        found.extend(
            re.findall(
                pattern,
                text,
                flags=re.I
            )
        )

    return sorted(
        set(
            x.strip()
            for x in found
        )
    )


# ============================================================
# GET TABLE ROWS
# ============================================================

# For each observation, retrieve all rows from the same
# PMCID + table.
#
# We DO NOT use the whole article.
# We DO NOT use unrelated tables.
#
# Context is restricted to the source table.


def get_source_table(
    pmcid,
    table_index
):

    subset = cells[
        (
            cells["_pmcid"] == pmcid
        )
        &
        (
            cells["_table"] == table_index
        )
    ].copy()

    if len(subset) == 0:
        return subset

    return subset.sort_values(
        [
            "_row",
            "_col"
        ],
        kind="stable"
    )


# ============================================================
# BUILD ROW REPRESENTATIONS
# ============================================================

def build_rows(table):

    if len(table) == 0:
        return pd.DataFrame()

    rows = []

    for row_index, group in table.groupby(
        "_row",
        sort=True
    ):

        group = group.sort_values(
            "_col"
        )

        cells_text = [
            clean(x)
            for x in group[
                "raw_cell"
            ].tolist()
        ]

        cells_text = [
            x
            for x in cells_text
            if x
        ]

        rows.append(
            {
                "row_index": row_index,

                "row_text":
                    " | ".join(
                        cells_text
                    ),

                "row_drug":
                    extract_terms(
                        " | ".join(
                            cells_text
                        ),
                        drug_patterns
                    ),

                "row_polymer":
                    extract_terms(
                        " | ".join(
                            cells_text
                        ),
                        polymer_patterns
                    ),

                "row_carrier":
                    extract_terms(
                        " | ".join(
                            cells_text
                        ),
                        carrier_patterns
                    ),

                "formulation_ids":
                    ";".join(
                        formulation_terms(
                            " | ".join(
                                cells_text
                            )
                        )
                    )
            }
        )

    return pd.DataFrame(rows)


# ============================================================
# IDENTIFY HEADER / CONTEXT ROWS
# ============================================================

def looks_like_header(text):

    text = clean(text).lower()

    header_terms = [
        "drug",
        "polymer",
        "carrier",
        "formulation",
        "composition",
        "material",
        "delivery system",
        "release",
        "time",
        "hours",
        "duration",
        "concentration"
    ]

    hits = sum(
        term in text
        for term in header_terms
    )

    return hits >= 2


# ============================================================
# RESOLVE FORMULATION CONTEXT
# ============================================================

results = []

for i, obs_row in obs.iterrows():

    pmcid = obs_row["_pmcid"]

    table_index = obs_row["_table"]

    release_row = obs_row["_row"]

    recovery_status = (
        obs_row[
            "structural_recovery_status"
        ]
    )

    table = get_source_table(
        pmcid,
        table_index
    )

    row_df = build_rows(
        table
    )

    result = obs_row.to_dict()

    result[
        "context_resolution_status"
    ] = "UNRESOLVED"

    result[
        "context_rows_examined"
    ] = len(row_df)

    result[
        "context_source_table"
    ] = table_index

    result[
        "context_formulation_id"
    ] = "NOT_IDENTIFIED"

    result[
        "context_drug"
    ] = "NOT_IDENTIFIED"

    result[
        "context_polymer"
    ] = "NOT_IDENTIFIED"

    result[
        "context_carrier"
    ] = "NOT_IDENTIFIED"

    result[
        "context_source_rows"
    ] = ""

    # --------------------------------------------------------
    # Only direct matches are automatically trusted.
    # Offset cases remain manual.
    # --------------------------------------------------------

    if recovery_status != "DIRECT_MATCH":

        result[
            "context_resolution_status"
        ] = "MANUAL_STRUCTURAL_REVIEW"

        results.append(
            result
        )

        continue

    if len(row_df) == 0:

        result[
            "context_resolution_status"
        ] = "NO_TABLE_CONTEXT"

        results.append(
            result
        )

        continue

    # --------------------------------------------------------
    # Locate exact release row
    # --------------------------------------------------------

    exact = row_df[
        row_df[
            "row_index"
        ] == release_row
    ]

    if len(exact) == 0:

        result[
            "context_resolution_status"
        ] = "EXACT_ROW_NOT_FOUND"

        results.append(
            result
        )

        continue

    exact_row = exact.iloc[0]

    # --------------------------------------------------------
    # Start with exact row information
    # --------------------------------------------------------

    drug = exact_row[
        "row_drug"
    ]

    polymer = exact_row[
        "row_polymer"
    ]

    carrier = exact_row[
        "row_carrier"
    ]

    form_ids = exact_row[
        "formulation_ids"
    ]

    source_rows = [
        str(
            int(release_row)
        )
    ]

    # --------------------------------------------------------
    # Search nearby rows ONLY when the exact row does not
    # provide the relevant formulation metadata.
    #
    # Maximum context window = 3 rows before/after.
    # --------------------------------------------------------

    candidate_context = row_df[
        (
            row_df["row_index"]
            >= release_row - 3
        )
        &
        (
            row_df["row_index"]
            <= release_row + 3
        )
        &
        (
            row_df["row_index"]
            != release_row
        )
    ].copy()

    # --------------------------------------------------------
    # Score neighboring rows.
    # --------------------------------------------------------

    scored = []

    for _, context_row in candidate_context.iterrows():

        distance = abs(
            context_row[
                "row_index"
            ] - release_row
        )

        score = 0

        if (
            context_row[
                "row_drug"
            ]
            != "NOT_IDENTIFIED"
        ):
            score += 2

        if (
            context_row[
                "row_polymer"
            ]
            != "NOT_IDENTIFIED"
        ):
            score += 2

        if (
            context_row[
                "row_carrier"
            ]
            != "NOT_IDENTIFIED"
        ):
            score += 1

        if (
            context_row[
                "formulation_ids"
            ]
        ):
            score += 4

        if looks_like_header(
            context_row[
                "row_text"
            ]
        ):
            score -= 3

        # Closer rows get higher priority
        score -= distance * 0.5

        scored.append(
            (
                score,
                context_row
            )
        )

    scored.sort(
        key=lambda x: x[0],
        reverse=True
    )

    # --------------------------------------------------------
    # Add context only if it is reasonably strong.
    # --------------------------------------------------------

    used_context = []

    for score, context_row in scored:

        if score < 2:
            continue

        used_context.append(
            context_row
        )

        if len(
            used_context
        ) >= 2:
            break

    # --------------------------------------------------------
    # Fill missing fields.
    # --------------------------------------------------------

    for context_row in used_context:

        context_text = (
            context_row[
                "row_text"
            ]
        )

        source_rows.append(
            str(
                int(
                    context_row[
                        "row_index"
                    ]
                )
            )
        )

        if (
            drug
            == "NOT_IDENTIFIED"
        ):

            drug = (
                context_row[
                    "row_drug"
                ]
            )

        if (
            polymer
            == "NOT_IDENTIFIED"
        ):

            polymer = (
                context_row[
                    "row_polymer"
                ]
            )

        if (
            carrier
            == "NOT_IDENTIFIED"
        ):

            carrier = (
                context_row[
                    "row_carrier"
                ]
            )

        if not form_ids:

            form_ids = (
                context_row[
                    "formulation_ids"
                ]
            )

    # --------------------------------------------------------
    # Determine confidence.
    # --------------------------------------------------------

    identified = sum(
        x != "NOT_IDENTIFIED"
        for x in [
            drug,
            polymer,
            carrier
        ]
    )

    if (
        identified >= 2
        and
        form_ids
    ):

        status = (
            "STRONG_CONTEXT"
        )

    elif identified >= 2:

        status = (
            "GOOD_CONTEXT"
        )

    elif identified == 1:

        status = (
            "PARTIAL_CONTEXT"
        )

    else:

        status = (
            "UNRESOLVED"
        )

    result[
        "context_resolution_status"
    ] = status

    result[
        "context_formulation_id"
    ] = (
        form_ids
        if form_ids
        else "NOT_IDENTIFIED"
    )

    result[
        "context_drug"
    ] = drug

    result[
        "context_polymer"
    ] = polymer

    result[
        "context_carrier"
    ] = carrier

    result[
        "context_source_rows"
    ] = ";".join(
        sorted(
            set(
                source_rows
            )
        )
    )

    results.append(
        result
    )


resolved = pd.DataFrame(
    results
)


# ============================================================
# QUALITY FLAGS
# ============================================================

def formulation_quality(row):

    status = row[
        "context_resolution_status"
    ]

    if status == "STRONG_CONTEXT":
        return "HIGH"

    if status == "GOOD_CONTEXT":
        return "MODERATE"

    if status == "PARTIAL_CONTEXT":
        return "LOW"

    return "MANUAL"


resolved[
    "formulation_context_quality"
] = resolved.apply(
    formulation_quality,
    axis=1
)


# ============================================================
# CONSERVATIVE ML-ELIGIBILITY FLAG
# ============================================================

resolved[
    "ml_metadata_ready"
] = (
    resolved[
        "formulation_context_quality"
    ]
    .isin(
        [
            "HIGH",
            "MODERATE"
        ]
    )
)


# ============================================================
# SAVE
# ============================================================

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "05K_formulation_context_audit.csv"
)

READY_FILE = os.path.join(
    OUT_DIR,
    "05K_formulation_context_ready.csv"
)

MANUAL_FILE = os.path.join(
    OUT_DIR,
    "05K_formulation_context_manual_review.csv"
)

resolved.to_csv(
    AUDIT_FILE,
    index=False
)

ready = resolved[
    resolved[
        "ml_metadata_ready"
    ]
].copy()

manual = resolved[
    ~resolved[
        "ml_metadata_ready"
    ]
].copy()

ready.to_csv(
    READY_FILE,
    index=False
)

manual.to_csv(
    MANUAL_FILE,
    index=False
)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CODE 5K COMPLETE")
print("=" * 70)

print(
    f"Input observations: "
    f"{len(resolved):,}"
)

print(
    "\nContext resolution:"
)

print(
    resolved[
        "context_resolution_status"
    ].value_counts()
)

print(
    "\nFormulation-context quality:"
)

print(
    resolved[
        "formulation_context_quality"
    ].value_counts()
)

print(
    "\nML metadata ready:",
    int(
        resolved[
            "ml_metadata_ready"
        ].sum()
    )
)

print(
    "\nDrug:"
)

print(
    resolved[
        "context_drug"
    ].value_counts()
    .head(20)
)

print(
    "\nPolymer:"
)

print(
    resolved[
        "context_polymer"
    ].value_counts()
    .head(20)
)

print(
    "\nCarrier:"
)

print(
    resolved[
        "context_carrier"
    ].value_counts()
    .head(20)
)

print(
    "\nFormulation IDs:"
)

print(
    (
        resolved[
            "context_formulation_id"
        ]
        != "NOT_IDENTIFIED"
    ).sum()
)

print("\nSaved:")
print(AUDIT_FILE)
print(READY_FILE)
print(MANUAL_FILE)

print("\nIMPORTANT:")
print(
    "Context is restricted to the same source table."
)
print(
    "Exact release rows are kept separate from nearby "
    "context rows."
)
print(
    "Offset structural matches are NOT automatically accepted."
)
print(
    "No synthetic observations were generated."
)
print(
    "No interpolation was performed."
)
print(
    "No observations were duplicated."
)
print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ============================================================
# CODE 5L — COLUMN-AWARE FORMULATION METADATA RECOVERY
# ============================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 5L — COLUMN-AWARE FORMULATION METADATA RECOVERY")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "05K_formulation_context_audit.csv"
)

CODE3_FILE = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"\nRequired Code 5K file not found:\n{INPUT_FILE}\n"
        "\nRun Code 5K first."
    )

if not os.path.exists(CODE3_FILE):
    raise FileNotFoundError(
        f"\nRequired Code 3 file not found:\n{CODE3_FILE}"
    )


# ============================================================
# LOAD
# ============================================================

obs = pd.read_csv(
    INPUT_FILE,
    low_memory=False
)

cells = pd.read_csv(
    CODE3_FILE,
    low_memory=False
)

print(
    f"5K observations: {len(obs):,}"
)

print(
    f"Code 3 cells: {len(cells):,}"
)


# ============================================================
# NORMALIZE
# ============================================================

cells["_pmcid"] = (
    cells["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

cells["_table"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

cells["_row"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

cells["_col"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)


obs["_pmcid"] = (
    obs["recovered_pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

obs["_table"] = pd.to_numeric(
    obs["recovered_table_index"],
    errors="coerce"
)

obs["_row"] = pd.to_numeric(
    obs["recovered_row_index"],
    errors="coerce"
)


# ============================================================
# TEXT CLEANING
# ============================================================

def clean(x):

    if pd.isna(x):
        return ""

    x = str(x)
    x = x.replace("\n", " ")
    x = re.sub(r"\s+", " ", x)

    return x.strip()


# ============================================================
# PATTERNS
# ============================================================

DRUG_PATTERNS = {
    "dexamethasone":
        r"\bdexamethasone\b",

    "paclitaxel":
        r"\bpaclitaxel\b",

    "curcumin":
        r"\bcurcumin\b",

    "doxorubicin":
        r"\bdoxorubicin\b",

    "DOX":
        r"\bDOX\b",

    "ibuprofen":
        r"\bibuprofen\b",

    "metformin":
        r"\bmetformin\b",

    "gemcitabine":
        r"\bgemcitabine\b",

    "rifampicin":
        r"\brifampicin\b|\brifampin\b",

    "vancomycin":
        r"\bvancomycin\b",

    "insulin":
        r"\binsulin\b",

    "5-fluorouracil":
        r"\b5[\s\-]?fluorouracil\b|\b5FU\b"
}


POLYMER_PATTERNS = {
    "PLGA":
        r"\bPLGA\b|poly\(lactic[\s\-]?co[\s\-]?glycolic acid\)",

    "PLA":
        r"\bPLA\b|polylactic acid|poly\(lactic acid\)",

    "PCL":
        r"\bPCL\b|polycaprolactone",

    "PEG":
        r"\bPEG\b|polyethylene glycol",

    "PEO":
        r"\bPEO\b|polyethylene oxide",

    "chitosan":
        r"\bchitosan\b",

    "alginate":
        r"\balginate\b",

    "gelatin":
        r"\bgelatin\b",

    "PVA":
        r"\bPVA\b|polyvinyl alcohol",

    "poloxamer":
        r"\bpoloxamer\b",

    "polyurethane":
        r"\bpolyurethane\b",

    "cellulose":
        r"\bcellulose\b"
}


CARRIER_PATTERNS = {
    "nanoparticle":
        r"\bnanoparticle[s]?\b",

    "nanocarrier":
        r"\bnanocarrier[s]?\b",

    "micelle":
        r"\bmicelle[s]?\b",

    "liposome":
        r"\bliposome[s]?\b",

    "hydrogel":
        r"\bhydrogel[s]?\b",

    "microsphere":
        r"\bmicrosphere[s]?\b",

    "microparticle":
        r"\bmicroparticle[s]?\b",

    "nanofiber":
        r"\bnanofiber[s]?\b",

    "film":
        r"\bfilm[s]?\b",

    "implant":
        r"\bimplant[s]?\b",

    "beads":
        r"\bbead[s]?\b"
}


def find_terms(text, patterns):

    text = clean(text)

    if not text:
        return []

    found = []

    for name, pattern in patterns.items():

        if re.search(
            pattern,
            text,
            flags=re.I
        ):
            found.append(name)

    return found


# ============================================================
# HEADER CLASSIFICATION
# ============================================================

def classify_header(header):

    h = clean(header).lower()

    if not h:
        return []

    classes = []

    # Drug/payload
    if any(
        x in h
        for x in [
            "drug",
            "payload",
            "active",
            "therapeutic",
            "compound",
            "loaded drug",
            "drug name"
        ]
    ):
        classes.append("DRUG")

    # Polymer/material
    if any(
        x in h
        for x in [
            "polymer",
            "material",
            "matrix",
            "polymeric",
            "polymer type",
            "polymer composition"
        ]
    ):
        classes.append("POLYMER")

    # Carrier
    if any(
        x in h
        for x in [
            "carrier",
            "delivery system",
            "delivery vehicle",
            "nanoparticle",
            "nanocarrier",
            "formulation type",
            "dosage form"
        ]
    ):
        classes.append("CARRIER")

    # Formulation
    if any(
        x in h
        for x in [
            "formulation",
            "sample",
            "group",
            "batch",
            "code",
            "system"
        ]
    ):
        classes.append("FORMULATION")

    # Release/time
    if any(
        x in h
        for x in [
            "release",
            "cumulative",
            "time",
            "hour",
            "duration",
            "day"
        ]
    ):
        classes.append("RELEASE")

    return classes


# ============================================================
# BUILD TABLE STRUCTURE
# ============================================================

def build_table_structure(
    pmcid,
    table_index
):

    t = cells[
        (cells["_pmcid"] == pmcid)
        &
        (cells["_table"] == table_index)
    ].copy()

    if len(t) == 0:
        return None

    t = t.sort_values(
        ["_row", "_col"]
    )

    # --------------------------------------------------------
    # Header text supplied by Code 3
    # --------------------------------------------------------

    header_values = []

    if "header_text" in t.columns:

        for x in t["header_text"].dropna():

            x = clean(x)

            if x:
                header_values.append(x)

    header_text = " | ".join(
        list(dict.fromkeys(header_values))
    )

    # --------------------------------------------------------
    # Infer column headers from first few rows if possible.
    # --------------------------------------------------------

    row_numbers = sorted(
        t["_row"].dropna().unique()
    )

    header_rows = []

    for r in row_numbers[:4]:

        row = t[
            t["_row"] == r
        ].sort_values("_col")

        text = " | ".join(
            clean(x)
            for x in row["raw_cell"]
            if clean(x)
        )

        if not text:
            continue

        header_rows.append(
            {
                "row": r,
                "text": text
            }
        )

    # --------------------------------------------------------
    # Column map.
    # --------------------------------------------------------

    column_map = {}

    for col, group in t.groupby(
        "_col",
        sort=True
    ):

        texts = []

        if "header_text" in group.columns:

            texts.extend(
                clean(x)
                for x in group[
                    "header_text"
                ].dropna()
                if clean(x)
            )

        # Add first rows as contextual candidates
        for r in row_numbers[:3]:

            vals = group[
                group["_row"] == r
            ]["raw_cell"]

            texts.extend(
                clean(x)
                for x in vals
                if clean(x)
            )

        combined = " | ".join(
            list(
                dict.fromkeys(texts)
            )
        )

        column_map[int(col)] = {
            "header": combined,
            "classes":
                classify_header(combined)
        }

    return {
        "table": t,
        "header_text": header_text,
        "header_rows": header_rows,
        "column_map": column_map
    }


# ============================================================
# FIND FORMULATION CONTEXT
# ============================================================

def get_column_value(
    table,
    row,
    col
):

    x = table[
        (table["_row"] == row)
        &
        (table["_col"] == col)
    ]

    if len(x) == 0:
        return ""

    return clean(
        x.iloc[0]["raw_cell"]
    )


def collect_column_metadata(
    structure,
    release_row
):

    table = structure["table"]
    column_map = structure["column_map"]

    found = {
        "drug": [],
        "polymer": [],
        "carrier": [],
        "formulation": [],
        "release": []
    }

    source_columns = {
        "drug": [],
        "polymer": [],
        "carrier": [],
        "formulation": []
    }

    # --------------------------------------------------------
    # Examine every column in the exact release row.
    # --------------------------------------------------------

    release_cells = table[
        table["_row"] == release_row
    ].sort_values("_col")

    for _, cell in release_cells.iterrows():

        col = int(
            cell["_col"]
        )

        value = clean(
            cell["raw_cell"]
        )

        if not value:
            continue

        info = column_map.get(
            col,
            {
                "header": "",
                "classes": []
            }
        )

        header = info["header"]
        classes = info["classes"]

        # ----------------------------------------------
        # If header says DRUG
        # ----------------------------------------------

        if "DRUG" in classes:

            terms = find_terms(
                value,
                DRUG_PATTERNS
            )

            if terms:

                found["drug"].extend(
                    terms
                )

                source_columns[
                    "drug"
                ].append(
                    f"{col}:{header}"
                )

        # ----------------------------------------------
        # POLYMER
        # ----------------------------------------------

        if "POLYMER" in classes:

            terms = find_terms(
                value,
                POLYMER_PATTERNS
            )

            if terms:

                found["polymer"].extend(
                    terms
                )

                source_columns[
                    "polymer"
                ].append(
                    f"{col}:{header}"
                )

        # ----------------------------------------------
        # CARRIER
        # ----------------------------------------------

        if "CARRIER" in classes:

            terms = find_terms(
                value,
                CARRIER_PATTERNS
            )

            if terms:

                found["carrier"].extend(
                    terms
                )

                source_columns[
                    "carrier"
                ].append(
                    f"{col}:{header}"
                )

        # ----------------------------------------------
        # FORMULATION
        # ----------------------------------------------

        if "FORMULATION" in classes:

            found_ids = re.findall(
                r"\b(?:F|Formulation|Group|Sample|Batch|Code|Test)[\s\-]*[A-Za-z0-9]+\b",
                value,
                flags=re.I
            )

            if found_ids:

                found[
                    "formulation"
                ].extend(
                    found_ids
                )

                source_columns[
                    "formulation"
                ].append(
                    f"{col}:{header}"
                )

    return (
        found,
        source_columns
    )


# ============================================================
# FALLBACK: HEADER ITSELF
# ============================================================

def recover_from_header(
    structure
):

    header = structure[
        "header_text"
    ]

    return {
        "drug":
            find_terms(
                header,
                DRUG_PATTERNS
            ),

        "polymer":
            find_terms(
                header,
                POLYMER_PATTERNS
            ),

        "carrier":
            find_terms(
                header,
                CARRIER_PATTERNS
            )
    }


# ============================================================
# PROCESS OBSERVATIONS
# ============================================================

output = []

for i, row in obs.iterrows():

    result = row.to_dict()

    status = row[
        "structural_recovery_status"
    ]

    pmcid = row["_pmcid"]
    table_index = row["_table"]
    release_row = row["_row"]

    result[
        "column_metadata_status"
    ] = "NOT_ATTEMPTED"

    result[
        "column_drug"
    ] = "NOT_IDENTIFIED"

    result[
        "column_polymer"
    ] = "NOT_IDENTIFIED"

    result[
        "column_carrier"
    ] = "NOT_IDENTIFIED"

    result[
        "column_formulation"
    ] = "NOT_IDENTIFIED"

    result[
        "metadata_source_columns"
    ] = ""

    result[
        "metadata_header_text"
    ] = ""

    # --------------------------------------------------------
    # Do not automatically process unresolved structural rows.
    # --------------------------------------------------------

    if status != "DIRECT_MATCH":

        result[
            "column_metadata_status"
        ] = "MANUAL_STRUCTURAL_REVIEW"

        output.append(
            result
        )

        continue

    structure = build_table_structure(
        pmcid,
        table_index
    )

    if structure is None:

        result[
            "column_metadata_status"
        ] = "NO_TABLE"

        output.append(
            result
        )

        continue

    result[
        "metadata_header_text"
    ] = structure[
        "header_text"
    ]

    found, source_columns = (
        collect_column_metadata(
            structure,
            release_row
        )
    )

    # --------------------------------------------------------
    # Header fallback
    # --------------------------------------------------------

    header_found = recover_from_header(
        structure
    )

    for key in [
        "drug",
        "polymer",
        "carrier"
    ]:

        if not found[key]:

            found[key] = (
                header_found[key]
            )

    # --------------------------------------------------------
    # Clean values
    # --------------------------------------------------------

    for key in [
        "drug",
        "polymer",
        "carrier",
        "formulation"
    ]:

        vals = list(
            dict.fromkeys(
                found[key]
            )
        )

        if vals:

            result[
                f"column_{key}"
            ] = ";".join(
                vals
            )

    source_parts = []

    for key, vals in source_columns.items():

        for v in vals:

            source_parts.append(
                f"{key.upper()}={v}"
            )

    result[
        "metadata_source_columns"
    ] = " || ".join(
        source_parts
    )

    # --------------------------------------------------------
    # Quality determination
    # --------------------------------------------------------

    identified = sum(
        result[
            f"column_{x}"
        ] != "NOT_IDENTIFIED"
        for x in [
            "drug",
            "polymer",
            "carrier"
        ]
    )

    formulation_found = (
        result[
            "column_formulation"
        ]
        != "NOT_IDENTIFIED"
    )

    if (
        identified >= 2
        and formulation_found
    ):

        result[
            "column_metadata_status"
        ] = "STRONG_COLUMN_CONTEXT"

    elif identified >= 2:

        result[
            "column_metadata_status"
        ] = "GOOD_COLUMN_CONTEXT"

    elif identified == 1:

        result[
            "column_metadata_status"
        ] = "PARTIAL_COLUMN_CONTEXT"

    else:

        result[
            "column_metadata_status"
        ] = "UNRESOLVED"

    output.append(
        result
    )


result_df = pd.DataFrame(
    output
)


# ============================================================
# FINAL CONSERVATIVE FLAGS
# ============================================================

result_df[
    "column_metadata_quality"
] = "MANUAL"

result_df.loc[
    result_df[
        "column_metadata_status"
    ] == "STRONG_COLUMN_CONTEXT",
    "column_metadata_quality"
] = "HIGH"

result_df.loc[
    result_df[
        "column_metadata_status"
    ] == "GOOD_COLUMN_CONTEXT",
    "column_metadata_quality"
] = "MODERATE"

result_df.loc[
    result_df[
        "column_metadata_status"
    ] == "PARTIAL_COLUMN_CONTEXT",
    "column_metadata_quality"
] = "LOW"


# ============================================================
# IMPORTANT:
# Combine 5K and 5L conservatively.
#
# We do NOT overwrite 5K values with weaker 5L values.
# 5L is used only to fill missing fields.
# ============================================================

for field in [
    "drug",
    "polymer",
    "carrier"
]:

    k_col = f"context_{field}"
    l_col = f"column_{field}"

    final_col = f"resolved_{field}"

    result_df[final_col] = (
        result_df[k_col]
        if k_col in result_df.columns
        else "NOT_IDENTIFIED"
    )

    if k_col in result_df.columns:

        missing_mask = (
            result_df[k_col]
            .isna()
            |
            (
                result_df[k_col]
                .astype(str)
                .eq("NOT_IDENTIFIED")
            )
        )

        result_df.loc[
            missing_mask,
            final_col
        ] = result_df.loc[
            missing_mask,
            l_col
        ]


# ============================================================
# RESOLVED FORMULATION
# ============================================================

result_df[
    "resolved_formulation"
] = result_df.get(
    "context_formulation_id",
    "NOT_IDENTIFIED"
)

if "column_formulation" in result_df.columns:

    mask = (
        result_df[
            "resolved_formulation"
        ]
        .isna()
        |
        result_df[
            "resolved_formulation"
        ]
        .astype(str)
        .eq("NOT_IDENTIFIED")
    )

    result_df.loc[
        mask,
        "resolved_formulation"
    ] = result_df.loc[
        mask,
        "column_formulation"
    ]


# ============================================================
# FINAL METADATA COMPLETENESS
# ============================================================

def completeness(row):

    fields = [
        row["resolved_drug"],
        row["resolved_polymer"],
        row["resolved_carrier"]
    ]

    n = sum(
        x != "NOT_IDENTIFIED"
        for x in fields
    )

    if n == 3:
        return "COMPLETE"

    if n == 2:
        return "SUBSTANTIALLY_COMPLETE"

    if n == 1:
        return "INCOMPLETE"

    return "UNRESOLVED"


result_df[
    "metadata_completeness"
] = result_df.apply(
    completeness,
    axis=1
)


# ============================================================
# SOURCE TRACEABILITY FLAG
# ============================================================

result_df[
    "source_traceable"
] = (
    result_df[
        "structural_recovery_status"
    ]
    == "DIRECT_MATCH"
)


# ============================================================
# DO NOT CALL THIS THE FINAL ML DATASET
# ============================================================

result_df[
    "candidate_for_final_validation"
] = (
    result_df[
        "source_traceable"
    ]
    &
    result_df[
        "metadata_completeness"
    ].isin(
        [
            "COMPLETE",
            "SUBSTANTIALLY_COMPLETE"
        ]
    )
)


# ============================================================
# SAVE
# ============================================================

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "05L_column_metadata_audit.csv"
)

COMPLETE_FILE = os.path.join(
    OUT_DIR,
    "05L_metadata_complete_candidates.csv"
)

REVIEW_FILE = os.path.join(
    OUT_DIR,
    "05L_metadata_manual_review.csv"
)

result_df.to_csv(
    AUDIT_FILE,
    index=False
)

complete = result_df[
    result_df[
        "candidate_for_final_validation"
    ]
].copy()

manual = result_df[
    ~result_df[
        "candidate_for_final_validation"
    ]
].copy()

complete.to_csv(
    COMPLETE_FILE,
    index=False
)

manual.to_csv(
    REVIEW_FILE,
    index=False
)


# ============================================================
# REPORT
# ============================================================

print("\n" + "=" * 70)
print("CODE 5L COMPLETE")
print("=" * 70)

print(
    f"Input observations: "
    f"{len(result_df):,}"
)

print(
    "\nColumn metadata status:"
)

print(
    result_df[
        "column_metadata_status"
    ].value_counts()
)

print(
    "\nColumn metadata quality:"
)

print(
    result_df[
        "column_metadata_quality"
    ].value_counts()
)

print(
    "\nMetadata completeness:"
)

print(
    result_df[
        "metadata_completeness"
    ].value_counts()
)

print(
    "\nResolved drugs:"
)

print(
    result_df[
        "resolved_drug"
    ].value_counts()
    .head(20)
)

print(
    "\nResolved polymers:"
)

print(
    result_df[
        "resolved_polymer"
    ].value_counts()
    .head(20)
)

print(
    "\nResolved carriers:"
)

print(
    result_df[
        "resolved_carrier"
    ].value_counts()
    .head(20)
)

print(
    "\nResolved formulation IDs:",
    int(
        (
            result_df[
                "resolved_formulation"
            ]
            .astype(str)
            .ne("NOT_IDENTIFIED")
        ).sum()
    )
)

print(
    "\nCandidate for final validation:",
    int(
        result_df[
            "candidate_for_final_validation"
        ].sum()
    )
)

print("\nSaved:")
print(AUDIT_FILE)
print(COMPLETE_FILE)
print(REVIEW_FILE)

print("\nIMPORTANT:")
print(
    "Metadata were recovered only from the same source table."
)
print(
    "Column/header context is used only to fill missing metadata."
)
print(
    "No observations were generated."
)
print(
    "No interpolation was performed."
)
print(
    "No observations were duplicated."
)
print(
    "Offset structural matches remain manual."
)
print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ============================================================
# CODE 6A — HIGH-PRECISION TIME–RELEASE EXTRACTION
# FROM CANDIDATE RELEASE TABLES
#
# PURPOSE:
# Recover substantially more genuine literature time–release
# observations from the 1,407 candidate tables.
#
# IMPORTANT:
# - No synthetic observations
# - No interpolation
# - No duplication
# - No ML
# - Candidate observations only
# - Every observation retains PMCID/table/row/column provenance
# ============================================================

import os
import re
import math
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6A — HIGH-PRECISION TIME–RELEASE EXTRACTION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

CELLS_FILE = os.path.join(
    ROOT,
    "data",
    "candidates",
    "04_release_table_candidate_cells.csv"
)

TABLE_FILE = os.path.join(
    ROOT,
    "data",
    "candidates",
    "04_release_table_candidates.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "candidates"
)

os.makedirs(
    OUT_DIR,
    exist_ok=True
)

if not os.path.exists(CELLS_FILE):
    raise FileNotFoundError(
        f"\nRequired file not found:\n{CELLS_FILE}\n"
        "\nRun Code 4 FAST first."
    )

if not os.path.exists(TABLE_FILE):
    print(
        "\nWARNING: table summary file not found."
    )
    print(
        "Continuing with candidate-cell file."
    )


# ============================================================
# LOAD
# ============================================================

cells = pd.read_csv(
    CELLS_FILE,
    low_memory=False
)

print(
    f"Candidate cell records loaded: {len(cells):,}"
)


# ============================================================
# NORMALIZE COLUMN NAMES
# ============================================================

cells.columns = [
    str(c).strip()
    for c in cells.columns
]

print(
    "Columns:",
    list(cells.columns)
)


# ============================================================
# DETECT STRUCTURAL COLUMNS
# ============================================================

def detect_column(
    columns,
    candidates
):

    lower_map = {
        str(c).lower().strip(): c
        for c in columns
    }

    for candidate in candidates:

        if candidate.lower() in lower_map:

            return lower_map[
                candidate.lower()
            ]

    for c in columns:

        lc = str(c).lower()

        for candidate in candidates:

            if candidate.lower() in lc:

                return c

    return None


pmcid_col = detect_column(
    cells.columns,
    [
        "pmcid",
        "source_pmcid"
    ]
)

table_col = detect_column(
    cells.columns,
    [
        "table_index",
        "table_idx"
    ]
)

row_col = detect_column(
    cells.columns,
    [
        "row_index",
        "row_idx"
    ]
)

column_col = detect_column(
    cells.columns,
    [
        "column_index",
        "col_index"
    ]
)

raw_col = detect_column(
    cells.columns,
    [
        "raw_cell",
        "cell_text"
    ]
)

row_text_col = detect_column(
    cells.columns,
    [
        "row_text"
    ]
)

header_col = detect_column(
    cells.columns,
    [
        "header_text"
    ])

caption_col = detect_column(
    cells.columns,
    [
        "table_caption",
        "caption"
    ]
)

title_col = detect_column(
    cells.columns,
    [
        "article_title",
        "title"
    ]
)

doi_col = detect_column(
    cells.columns,
    [
        "doi"
    ]
)

pmid_col = detect_column(
    cells.columns,
    [
        "pmid"
    ]
)


required = {
    "PMCID": pmcid_col,
    "table": table_col,
    "row": row_col,
    "column": column_col,
    "raw cell": raw_col
}

missing = [
    k
    for k, v in required.items()
    if v is None
]

if missing:

    raise ValueError(
        "Could not detect required columns:\n"
        + "\n".join(missing)
    )

print("\nDetected structural columns:")

for k, v in required.items():

    print(
        f"{k:12s}: {v}"
    )


# ============================================================
# NORMALIZE STRUCTURE
# ============================================================

cells["_pmcid"] = (
    cells[pmcid_col]
    .astype(str)
    .str.strip()
    .str.upper()
)

cells["_table"] = pd.to_numeric(
    cells[table_col],
    errors="coerce"
)

cells["_row"] = pd.to_numeric(
    cells[row_col],
    errors="coerce"
)

cells["_column"] = pd.to_numeric(
    cells[column_col],
    errors="coerce"
)


# ============================================================
# TEXT CLEANING
# ============================================================

def clean(x):

    if pd.isna(x):
        return ""

    x = str(x)

    x = x.replace(
        "\n",
        " "
    )

    x = x.replace(
        "\xa0",
        " "
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


# ============================================================
# HEADER / SEMANTIC KEYWORDS
# ============================================================

TIME_HEADER_TERMS = [
    "time",
    "hours",
    "hour",
    "hr",
    "hrs",
    "day",
    "days",
    "duration",
    "sampling time",
    "release time",
    "time point",
    "timepoint"
]

RELEASE_HEADER_TERMS = [
    "release",
    "drug release",
    "cumulative release",
    "cumulative drug release",
    "released",
    "release (%)",
    "release %",
    "released (%)",
    "cumulative (%)",
    "cumulative drug"
]

NON_RELEASE_TERMS = [
    "loading",
    "drug loading",
    "encapsulation",
    "entrapment",
    "entrapment efficiency",
    "efficiency",
    "yield",
    "inhibition",
    "inhibition zone",
    "cell viability",
    "viability",
    "cytotoxicity",
    "auc",
    "area under curve",
    "cmax",
    "tmax",
    "half-life",
    "half life",
    "pharmacokinetic",
    "bioavailability",
    "particle size",
    "diameter",
    "size",
    "zeta potential",
    "polydispersity",
    "pdi",
    "contact angle",
    "roughness",
    "weight loss",
    "swelling",
    "degradation",
    "absorbance",
    "concentration",
    "dose"
]

REFERENCE_TERMS = [
    "study",
    "reference",
    "ref.",
    "author",
    "year",
    "citation"
]


# ============================================================
# HEADER CLASSIFICATION
# ============================================================

def header_score(
    text,
    terms
):

    text = clean(text).lower()

    score = 0

    for term in terms:

        if term in text:

            score += 1

    return score


def classify_header(text):

    text = clean(text)

    lower = text.lower()

    time_score = header_score(
        lower,
        TIME_HEADER_TERMS
    )

    release_score = header_score(
        lower,
        RELEASE_HEADER_TERMS
    )

    nonrelease_score = header_score(
        lower,
        NON_RELEASE_TERMS
    )

    return {
        "time_score":
            time_score,

        "release_score":
            release_score,

        "nonrelease_score":
            nonrelease_score
    }


# ============================================================
# NUMERIC EXTRACTION
# ============================================================

NUMBER_RE = re.compile(
    r"""
    (?<![A-Za-z])
    [-+]?
    (?:\d+(?:\.\d*)?|\.\d+)
    (?:[eE][-+]?\d+)?
    %?
    (?![A-Za-z])
    """,
    re.VERBOSE
)


def numeric_values(text):

    text = clean(text)

    vals = []

    for m in NUMBER_RE.finditer(text):

        raw = m.group(
            0
        )

        try:

            number = float(
                raw.replace(
                    "%",
                    ""
                )
            )

            vals.append(
                {
                    "raw": raw,
                    "value": number,
                    "percent":
                        "%" in raw
                }
            )

        except Exception:

            pass

    return vals


# ============================================================
# TIME PARSER
# ============================================================

def parse_time(
    text,
    header=""
):

    text = clean(text)

    if not text:

        return None, None

    lower = text.lower()

    # --------------------------------------------------------
    # Explicit units
    # --------------------------------------------------------

    patterns = [

        (
            r"(\d+(?:\.\d+)?)\s*(?:minutes?|mins?|min)\b",
            1 / 60,
            "MIN"
        ),

        (
            r"(\d+(?:\.\d+)?)\s*(?:hours?|hrs?|hr|h)\b",
            1,
            "H"
        ),

        (
            r"(\d+(?:\.\d+)?)\s*(?:days?|d)\b",
            24,
            "DAY"
        ),

        (
            r"(\d+(?:\.\d+)?)\s*(?:weeks?|wk|wks?)\b",
            168,
            "WEEK"
        )
    ]

    for pattern, multiplier, unit in patterns:

        m = re.search(
            pattern,
            lower
        )

        if m:

            value = float(
                m.group(1)
            )

            return (
                value * multiplier,
                unit
            )

    # --------------------------------------------------------
    # Numeric-only values.
    #
    # ONLY accept these if the column/header strongly indicates
    # time. This prevents 5.26, 8.4, 50 etc. from becoming time
    # merely because they are numbers.
    # --------------------------------------------------------

    hscore = classify_header(
        header
    )

    if (
        hscore["time_score"] > 0
        and
        hscore["nonrelease_score"] == 0
    ):

        vals = numeric_values(
            text
        )

        if len(vals) == 1:

            value = vals[0][
                "value"
            ]

            if (
                0 <= value <= 5000
            ):

                return (
                    value,
                    "HEADER_TIME"
                )

    return None, None


# ============================================================
# RELEASE PARSER
# ============================================================

def parse_release(
    text,
    header=""
):

    text = clean(text)

    if not text:

        return None, None

    lower = text.lower()

    # --------------------------------------------------------
    # Explicit percent
    # --------------------------------------------------------

    percent_matches = re.findall(
        r"([-+]?\d+(?:\.\d+)?)\s*%",
        text
    )

    if len(percent_matches) == 1:

        value = float(
            percent_matches[0]
        )

        if 0 <= value <= 100:

            return (
                value,
                "EXPLICIT_PERCENT"
            )

    # --------------------------------------------------------
    # Explicit release language + percentage
    # --------------------------------------------------------

    m = re.search(
        r"""
        (?:release|
           released|
           cumulative\s+release|
           cumulative\s+drug\s+release)
        [^%\d]{0,50}
        (\d+(?:\.\d+)?)
        \s*%
        """,
        lower,
        flags=re.I | re.X
    )

    if m:

        value = float(
            m.group(1)
        )

        if 0 <= value <= 100:

            return (
                value,
                "RELEASE_PERCENT"
            )

    # --------------------------------------------------------
    # Numeric-only release value.
    #
    # ACCEPT ONLY if the column/header is release-specific.
    # --------------------------------------------------------

    hscore = classify_header(
        header
    )

    if (
        hscore["release_score"] > 0
        and
        hscore["nonrelease_score"] == 0
    ):

        vals = numeric_values(
            text
        )

        # Exactly one numeric value
        if len(vals) == 1:

            value = vals[0][
                "value"
            ]

            if 0 <= value <= 100:

                return (
                    value,
                    "HEADER_RELEASE"
                )

    return None, None


# ============================================================
# BUILD TABLE STRUCTURE
# ============================================================

def table_context(
    group
):

    # --------------------------------------------------------
    # Header text
    # --------------------------------------------------------

    header_values = []

    if header_col:

        for x in group[
            header_col
        ].dropna():

            x = clean(x)

            if x:

                header_values.append(
                    x
                )

    header_text = " | ".join(
        list(
            dict.fromkeys(
                header_values
            )
        )
    )

    # --------------------------------------------------------
    # Caption
    # --------------------------------------------------------

    caption = ""

    if caption_col:

        vals = [
            clean(x)
            for x in group[
                caption_col
            ].dropna()
            if clean(x)
        ]

        if vals:

            caption = vals[0]

    # --------------------------------------------------------
    # Article title
    # --------------------------------------------------------

    title = ""

    if title_col:

        vals = [
            clean(x)
            for x in group[
                title_col
            ].dropna()
            if clean(x)
        ]

        if vals:

            title = vals[0]

    return (
        header_text,
        caption,
        title
    )


# ============================================================
# DETECT TIME / RELEASE COLUMNS
# ============================================================

def detect_columns(
    group
):

    column_info = {}

    for col, cg in group.groupby(
        "_column",
        sort=True
    ):

        texts = []

        # Header text
        if header_col:

            texts.extend(
                clean(x)
                for x in cg[
                    header_col
                ].dropna()
                if clean(x)
            )

        # Raw cells in first few rows
        first_rows = sorted(
            cg[
                "_row"
            ].dropna()
            .unique()
        )[:4]

        for r in first_rows:

            vals = cg[
                cg["_row"] == r
            ][raw_col]

            texts.extend(
                clean(x)
                for x in vals
                if clean(x)
            )

        header = " | ".join(
            list(
                dict.fromkeys(
                    texts
                )
            )
        )

        scores = classify_header(
            header
        )

        column_info[
            int(col)
        ] = {
            "header":
                header,

            "time_score":
                scores[
                    "time_score"
                ],

            "release_score":
                scores[
                    "release_score"
                ],

            "nonrelease_score":
                scores[
                    "nonrelease_score"
                ]
        }

    return column_info


# ============================================================
# TABLE-LEVEL EXTRACTION
# ============================================================

observations = []

table_audit = []

grouped = cells.groupby(
    ["_pmcid", "_table"],
    sort=False
)

total_tables = len(
    grouped
)

print(
    f"\nCandidate source tables: "
    f"{total_tables:,}"
)

for table_counter, (
    key,
    group
) in enumerate(
    grouped,
    start=1
):

    pmcid, table_index = key

    group = group.copy()

    group = group.sort_values(
        [
            "_row",
            "_column"
        ]
    )

    header_text, caption, title = (
        table_context(
            group
        )
    )

    col_info = detect_columns(
        group
    )

    # --------------------------------------------------------
    # Candidate time/release columns
    # --------------------------------------------------------

    time_cols = [
        c
        for c, info in col_info.items()
        if (
            info["time_score"] > 0
            and
            info["nonrelease_score"] == 0
        )
    ]

    release_cols = [
        c
        for c, info in col_info.items()
        if (
            info["release_score"] > 0
            and
            info["nonrelease_score"] == 0
        )
    ]

    # --------------------------------------------------------
    # Table-level audit
    # --------------------------------------------------------

    table_audit.append(
        {
            "pmcid":
                pmcid,

            "table_index":
                table_index,

            "header_text":
                header_text,

            "table_caption":
                caption,

            "time_columns":
                ";".join(
                    map(
                        str,
                        time_cols
                    )
                ),

            "release_columns":
                ";".join(
                    map(
                        str,
                        release_cols
                    )
                ),

            "time_column_count":
                len(time_cols),

            "release_column_count":
                len(release_cols)
        }
    )

    # --------------------------------------------------------
    # Require a recognizable release column OR explicit
    # release language in cells.
    # --------------------------------------------------------

    rows = sorted(
        group[
            "_row"
        ].dropna()
        .unique()
    )

    for row_index in rows:

        row = group[
            group["_row"] == row_index
        ].sort_values(
            "_column"
        )

        # ----------------------------------------------------
        # First: column-paired extraction.
        # ----------------------------------------------------

        for tcol in time_cols:

            time_cell = row[
                row["_column"] == tcol
            ]

            if len(time_cell) == 0:
                continue

            time_raw = clean(
                time_cell.iloc[0][
                    raw_col
                ]
            )

            time_header = col_info[
                tcol
            ]["header"]

            time_h, time_method = (
                parse_time(
                    time_raw,
                    time_header
                )
            )

            if time_h is None:
                continue

            for rcol in release_cols:

                release_cell = row[
                    row["_column"] == rcol
                ]

                if len(release_cell) == 0:
                    continue

                release_raw = clean(
                    release_cell.iloc[0][
                        raw_col
                    ]
                )

                release_header = (
                    col_info[
                        rcol
                    ]["header"]
                )

                release_pct, release_method = (
                    parse_release(
                        release_raw,
                        release_header
                    )
                )

                if release_pct is None:
                    continue

                # --------------------------------------------
                # Hard validity checks
                # --------------------------------------------

                if not (
                    0 <= release_pct <= 100
                ):
                    continue

                if not (
                    0 <= time_h <= 5000
                ):
                    continue

                # --------------------------------------------
                # Avoid identical column pairing
                # --------------------------------------------

                if tcol == rcol:
                    continue

                observations.append(
                    {
                        "candidate_observation_id":
                            f"6A_{len(observations)+1:07d}",

                        "pmcid":
                            pmcid,

                        "table_index":
                            table_index,

                        "row_index":
                            row_index,

                        "time_column":
                            tcol,

                        "release_column":
                            rcol,

                        "time_raw":
                            time_raw,

                        "release_raw":
                            release_raw,

                        "release_time_h":
                            time_h,

                        "release_percent":
                            release_pct,

                        "time_extraction_method":
                            time_method,

                        "release_extraction_method":
                            release_method,

                        "table_header":
                            header_text,

                        "table_caption":
                            caption,

                        "article_title":
                            title,

                        "source_type":
                            "CANDIDATE_TABLE_COLUMN_PAIR",

                        "validation_status":
                            "CANDIDATE"
                    }
                )

        # ----------------------------------------------------
        # Second: explicit same-row release statements.
        #
        # This handles cells such as:
        #
        # "80% after 24 h"
        #
        # BUT only explicit release language is accepted.
        # ----------------------------------------------------

        for _, cell in row.iterrows():

            text = clean(
                cell[raw_col]
            )

            if not text:
                continue

            lower = text.lower()

            if (
                "release" not in lower
                and
                "released" not in lower
            ):
                continue

            # Explicit release percent
            release_match = re.search(
                r"""
                (?:release|
                   released|
                   cumulative\s+release|
                   cumulative\s+drug\s+release)
                [^%\d]{0,80}
                (\d+(?:\.\d+)?)
                \s*%
                """,
                text,
                flags=re.I | re.X
            )

            if not release_match:
                continue

            release_pct = float(
                release_match.group(1)
            )

            if not (
                0 <= release_pct <= 100
            ):
                continue

            # Explicit time
            time_match = re.search(
                r"""
                (\d+(?:\.\d+)?)
                \s*
                (minutes?|mins?|min|
                 hours?|hrs?|hr|h|
                 days?|day|d|
                 weeks?|week|wk)
                \b
                """,
                text,
                flags=re.I | re.X
            )

            if not time_match:
                continue

            time_value = float(
                time_match.group(1)
            )

            unit = (
                time_match.group(2)
                .lower()
            )

            if unit.startswith(
                ("min",)
            ):

                time_h = (
                    time_value / 60
                )

            elif unit.startswith(
                ("h", "hr")
            ):

                time_h = time_value

            elif unit.startswith(
                ("d", "day")
            ):

                time_h = (
                    time_value * 24
                )

            elif unit.startswith(
                ("w", "wk")
            ):

                time_h = (
                    time_value * 168
                )

            else:

                continue

            if not (
                0 <= time_h <= 5000
            ):
                continue

            observations.append(
                {
                    "candidate_observation_id":
                        f"6A_{len(observations)+1:07d}",

                    "pmcid":
                        pmcid,

                    "table_index":
                        table_index,

                    "row_index":
                        row_index,

                    "time_column":
                        np.nan,

                    "release_column":
                        int(
                            cell["_column"]
                        ),

                    "time_raw":
                        time_match.group(0),

                    "release_raw":
                        release_match.group(0),

                    "release_time_h":
                        time_h,

                    "release_percent":
                        release_pct,

                    "time_extraction_method":
                        "EXPLICIT_TIME_IN_RELEASE_TEXT",

                    "release_extraction_method":
                        "EXPLICIT_RELEASE_PERCENT",

                    "table_header":
                        header_text,

                    "table_caption":
                        caption,

                    "article_title":
                        title,

                    "source_type":
                        "CANDIDATE_TABLE_EXPLICIT",

                    "validation_status":
                        "CANDIDATE"
                }
            )

    if (
        table_counter % 100 == 0
        or
        table_counter == total_tables
    ):

        print(
            f"Processed "
            f"{table_counter:,}/"
            f"{total_tables:,} tables | "
            f"Candidate observations: "
            f"{len(observations):,}"
        )


# ============================================================
# CREATE DATAFRAMES
# ============================================================

obs_df = pd.DataFrame(
    observations
)

audit_df = pd.DataFrame(
    table_audit
)


# ============================================================
# REMOVE EXACT DUPLICATES ONLY
#
# This does NOT create observations.
# It removes accidental duplicate extraction of the exact
# same source cell pair.
# ============================================================

if len(obs_df) > 0:

    before = len(
        obs_df
    )

    obs_df = obs_df.drop_duplicates(
        subset=[
            "pmcid",
            "table_index",
            "row_index",
            "time_column",
            "release_column",
            "time_raw",
            "release_raw",
            "release_time_h",
            "release_percent"
        ]
    ).copy()

    removed_duplicates = (
        before - len(obs_df)
    )

else:

    removed_duplicates = 0


# ============================================================
# ADD SOURCE IDENTIFIERS
# ============================================================

if len(obs_df) > 0:

    obs_df["source_key"] = (
        obs_df["pmcid"].astype(str)
        + "_T"
        + obs_df["table_index"].astype(str)
    )

    obs_df["source_row_key"] = (
        obs_df["source_key"]
        + "_R"
        + obs_df["row_index"].astype(str)
    )


# ============================================================
# SAVE
# ============================================================

OBS_FILE = os.path.join(
    OUT_DIR,
    "06A_high_precision_release_candidates.csv"
)

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "06A_candidate_table_column_audit.csv"
)

if len(obs_df) > 0:

    obs_df.to_csv(
        OBS_FILE,
        index=False
    )

audit_df.to_csv(
    AUDIT_FILE,
    index=False
)


# ============================================================
# REPORT
# ============================================================

print("\n" + "=" * 70)
print("CODE 6A COMPLETE")
print("=" * 70)

print(
    f"Candidate source tables: "
    f"{total_tables:,}"
)

print(
    f"Candidate observations before exact-duplicate removal: "
    f"{len(observations):,}"
)

print(
    f"Exact duplicate extractions removed: "
    f"{removed_duplicates:,}"
)

print(
    f"Final candidate observations: "
    f"{len(obs_df):,}"
)

if len(obs_df) > 0:

    print(
        "\nUnique PMCIDs:",
        obs_df[
            "pmcid"
        ].nunique()
    )

    print(
        "Unique source tables:",
        obs_df[
            [
                "pmcid",
                "table_index"
            ]
        ]
        .drop_duplicates()
        .shape[0]
    )

    print(
        "Unique source rows:",
        obs_df[
            [
                "pmcid",
                "table_index",
                "row_index"
            ]
        ]
        .drop_duplicates()
        .shape[0]
    )

    print(
        "\nRelease statistics:"
    )

    print(
        obs_df[
            "release_percent"
        ].describe()
    )

    print(
        "\nTime statistics:"
    )

    print(
        obs_df[
            "release_time_h"
        ].describe()
    )

    print(
        "\nExtraction methods:"
    )

    print(
        obs_df[
            "time_extraction_method"
        ].value_counts()
    )

    print(
        "\nRelease extraction methods:"
    )

    print(
        obs_df[
            "release_extraction_method"
        ].value_counts()
    )

    print(
        "\nPreview:"
    )

    print(
        obs_df[
            [
                "candidate_observation_id",
                "pmcid",
                "table_index",
                "row_index",
                "time_raw",
                "release_raw",
                "release_time_h",
                "release_percent",
                "time_extraction_method",
                "release_extraction_method"
            ]
        ].head(20).to_string(
            index=False
        )
    )

print("\nSaved:")

print(
    OBS_FILE
)

print(
    AUDIT_FILE
)

print("\nIMPORTANT:")

print(
    "These are CANDIDATE observations only."
)

print(
    "They have NOT passed semantic or source validation."
)

print(
    "No synthetic observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were intentionally duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ============================================================
# CODE 6A-R — HIGH-PRECISION TIME–RELEASE EXTRACTION
# USING CODE 4 TABLE CANDIDATES + CODE 3 STRUCTURAL CELLS
# ============================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6A-R — HIGH-PRECISION TIME–RELEASE EXTRACTION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

CODE4_FILE = os.path.join(
    ROOT,
    "data",
    "candidates",
    "04_release_table_candidates.csv"
)

CODE3_FILE = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "candidates"
)

os.makedirs(OUT_DIR, exist_ok=True)

# ------------------------------------------------------------
# CHECK FILES
# ------------------------------------------------------------

if not os.path.exists(CODE4_FILE):
    raise FileNotFoundError(
        f"Code 4 candidate-table file not found:\n{CODE4_FILE}"
    )

if not os.path.exists(CODE3_FILE):
    raise FileNotFoundError(
        f"Code 3 structural-cell file not found:\n{CODE3_FILE}"
    )

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

candidate_tables = pd.read_csv(
    CODE4_FILE,
    low_memory=False
)

cells = pd.read_csv(
    CODE3_FILE,
    low_memory=False
)

print(
    f"Code 4 candidate tables: "
    f"{len(candidate_tables):,}"
)

print(
    f"Code 3 cell records: "
    f"{len(cells):,}"
)

print("\nCode 3 columns:")
print(list(cells.columns))

# ------------------------------------------------------------
# REQUIRED STRUCTURAL COLUMNS
# ------------------------------------------------------------

required = [
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell"
]

missing = [
    x for x in required
    if x not in cells.columns
]

if missing:
    raise ValueError(
        "Code 3 is missing required structural columns:\n"
        + "\n".join(missing)
    )

# ------------------------------------------------------------
# NORMALIZE KEYS
# ------------------------------------------------------------

candidate_tables["_pmcid"] = (
    candidate_tables["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

candidate_tables["_table"] = pd.to_numeric(
    candidate_tables["table_index"],
    errors="coerce"
)

cells["_pmcid"] = (
    cells["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

cells["_table"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

cells["_row"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

cells["_column"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)

# ------------------------------------------------------------
# RESTRICT CODE 3 TO CODE 4 CANDIDATE TABLES
# ------------------------------------------------------------

candidate_keys = candidate_tables[
    ["_pmcid", "_table"]
].drop_duplicates()

print(
    f"\nUnique candidate source tables: "
    f"{len(candidate_keys):,}"
)

cells = cells.merge(
    candidate_keys,
    on=["_pmcid", "_table"],
    how="inner"
)

print(
    f"Structural cells belonging to candidate tables: "
    f"{len(cells):,}"
)

# ------------------------------------------------------------
# TEXT CLEANING
# ------------------------------------------------------------

def clean(x):

    if pd.isna(x):
        return ""

    x = str(x)
    x = x.replace("\n", " ")
    x = x.replace("\xa0", " ")
    x = re.sub(r"\s+", " ", x)

    return x.strip()


# ============================================================
# SEMANTIC TERMS
# ============================================================

TIME_TERMS = [
    "time",
    "time point",
    "timepoint",
    "hour",
    "hours",
    "hr",
    "hrs",
    "day",
    "days",
    "duration",
    "release time"
]

RELEASE_TERMS = [
    "release",
    "drug release",
    "cumulative release",
    "cumulative drug release",
    "released",
    "release (%)",
    "release %",
    "released (%)",
    "cumulative (%)"
]

NON_RELEASE_TERMS = [
    "loading",
    "drug loading",
    "encapsulation",
    "encapsulation efficiency",
    "entrapment",
    "entrapment efficiency",
    "inhibition",
    "inhibition zone",
    "cell viability",
    "viability",
    "cytotoxicity",
    "auc",
    "area under curve",
    "cmax",
    "tmax",
    "half-life",
    "half life",
    "pharmacokinetic",
    "bioavailability",
    "particle size",
    "diameter",
    "zeta potential",
    "polydispersity",
    "pdi",
    "contact angle",
    "roughness",
    "concentration",
    "dose",
    "yield"
]


def contains_term(
    text,
    terms
):

    text = clean(text).lower()

    return any(
        term in text
        for term in terms
    )


# ============================================================
# HEADER INFORMATION
# ============================================================

def get_header_for_column(
    group,
    column_index
):

    sub = group[
        group["_column"] == column_index
    ]

    texts = []

    if "header_text" in sub.columns:

        for x in sub[
            "header_text"
        ].dropna():

            x = clean(x)

            if x:
                texts.append(x)

    # Add first few cells from this column.
    # This helps when XML header extraction is incomplete.

    rows = sorted(
        sub["_row"]
        .dropna()
        .unique()
    )[:3]

    for r in rows:

        vals = sub[
            sub["_row"] == r
        ]["raw_cell"]

        for x in vals:

            x = clean(x)

            if x:
                texts.append(x)

    return " | ".join(
        list(
            dict.fromkeys(texts)
        )
    )


# ============================================================
# TIME PARSER
# ============================================================

def parse_time(
    text,
    header
):

    text = clean(text)
    header = clean(header)

    if not text:
        return None, None

    lower = text.lower()

    # Explicit units only
    patterns = [

        (
            r"(\d+(?:\.\d+)?)\s*(?:minutes?|mins?|min)\b",
            lambda x: x / 60,
            "MINUTES"
        ),

        (
            r"(\d+(?:\.\d+)?)\s*(?:hours?|hrs?|hr|h)\b",
            lambda x: x,
            "HOURS"
        ),

        (
            r"(\d+(?:\.\d+)?)\s*(?:days?|day|d)\b",
            lambda x: x * 24,
            "DAYS"
        ),

        (
            r"(\d+(?:\.\d+)?)\s*(?:weeks?|week|wk|wks?)\b",
            lambda x: x * 168,
            "WEEKS"
        )
    ]

    for pattern, converter, method in patterns:

        m = re.search(
            pattern,
            lower
        )

        if m:

            value = float(
                m.group(1)
            )

            hours = converter(
                value
            )

            if 0 <= hours <= 5000:

                return hours, method

    # Numeric-only time is allowed ONLY when header clearly
    # identifies the column as time.

    header_lower = header.lower()

    if (
        any(
            term in header_lower
            for term in TIME_TERMS
        )
        and
        not contains_term(
            header,
            NON_RELEASE_TERMS
        )
    ):

        nums = re.findall(
            r"[-+]?\d+(?:\.\d+)?",
            text
        )

        if len(nums) == 1:

            value = float(
                nums[0]
            )

            if 0 <= value <= 5000:

                return (
                    value,
                    "HEADER_TIME_NUMERIC"
                )

    return None, None


# ============================================================
# RELEASE PARSER
# ============================================================

def parse_release(
    text,
    header
):

    text = clean(text)
    header = clean(header)

    if not text:
        return None, None

    # Never treat obvious non-release outcomes as release.

    if contains_term(
        text,
        NON_RELEASE_TERMS
    ):

        # Explicit release wording can override some generic
        # terms, but not clearly unrelated outcomes.

        lower = text.lower()

        if (
            "release" not in lower
            and
            "released" not in lower
        ):

            return None, None

    # --------------------------------------------------------
    # Explicit percentage
    # --------------------------------------------------------

    matches = re.findall(
        r"([-+]?\d+(?:\.\d+)?)\s*%",
        text
    )

    if len(matches) == 1:

        value = float(
            matches[0]
        )

        if 0 <= value <= 100:

            return (
                value,
                "EXPLICIT_PERCENT"
            )

    # --------------------------------------------------------
    # Release language + number + %
    # --------------------------------------------------------

    m = re.search(
        r"""
        (?:cumulative\s+drug\s+release|
           cumulative\s+release|
           drug\s+release|
           release|
           released)
        [^%\d]{0,100}
        (\d+(?:\.\d+)?)
        \s*%
        """,
        text,
        flags=re.I | re.X
    )

    if m:

        value = float(
            m.group(1)
        )

        if 0 <= value <= 100:

            return (
                value,
                "EXPLICIT_RELEASE_PERCENT"
            )

    # --------------------------------------------------------
    # Numeric-only release:
    # only accept when the column header is explicitly release.
    # --------------------------------------------------------

    header_lower = header.lower()

    if (
        any(
            term in header_lower
            for term in RELEASE_TERMS
        )
        and
        not contains_term(
            header,
            NON_RELEASE_TERMS
        )
    ):

        nums = re.findall(
            r"[-+]?\d+(?:\.\d+)?",
            text
        )

        if len(nums) == 1:

            value = float(
                nums[0]
            )

            if 0 <= value <= 100:

                return (
                    value,
                    "RELEASE_HEADER_NUMERIC"
                )

    return None, None


# ============================================================
# TABLE-LEVEL EXTRACTION
# ============================================================

observations = []
table_audit = []

groups = cells.groupby(
    ["_pmcid", "_table"],
    sort=False
)

total_tables = len(
    groups
)

print(
    f"\nProcessing candidate tables: "
    f"{total_tables:,}"
)

for counter, (
    key,
    group
) in enumerate(
    groups,
    start=1
):

    pmcid, table_index = key

    group = group.sort_values(
        ["_row", "_column"]
    ).copy()

    # --------------------------------------------------------
    # Column headers
    # --------------------------------------------------------

    columns = sorted(
        group["_column"]
        .dropna()
        .unique()
    )

    column_headers = {}

    for col in columns:

        column_headers[
            int(col)
        ] = get_header_for_column(
            group,
            col
        )

    # --------------------------------------------------------
    # Identify release/time columns.
    #
    # Strict requirement:
    # release column must have release wording.
    # --------------------------------------------------------

    time_columns = []
    release_columns = []

    for col, header in column_headers.items():

        h = header.lower()

        if (
            any(
                term in h
                for term in TIME_TERMS
            )
            and
            not contains_term(
                h,
                NON_RELEASE_TERMS
            )
        ):

            time_columns.append(
                col
            )

        if (
            any(
                term in h
                for term in RELEASE_TERMS
            )
            and
            not contains_term(
                h,
                NON_RELEASE_TERMS
            )
        ):

            release_columns.append(
                col
            )

    table_audit.append(
        {
            "pmcid":
                pmcid,

            "table_index":
                table_index,

            "time_columns":
                ";".join(
                    map(
                        str,
                        time_columns
                    )
                ),

            "release_columns":
                ";".join(
                    map(
                        str,
                        release_columns
                    )
                ),

            "time_column_count":
                len(time_columns),

            "release_column_count":
                len(release_columns)
        }
    )

    # --------------------------------------------------------
    # ROW-LEVEL PAIRING
    # --------------------------------------------------------

    for row_index, row_group in group.groupby(
        "_row",
        sort=True
    ):

        row_group = row_group.sort_values(
            "_column"
        )

        # ----------------------------------------------------
        # Normal column pairing
        # ----------------------------------------------------

        for tcol in time_columns:

            tcell = row_group[
                row_group["_column"] == tcol
            ]

            if len(tcell) == 0:
                continue

            time_raw = clean(
                tcell.iloc[0][
                    "raw_cell"
                ]
            )

            time_header = (
                column_headers[
                    int(tcol)
                ]
            )

            time_h, time_method = (
                parse_time(
                    time_raw,
                    time_header
                )
            )

            if time_h is None:
                continue

            for rcol in release_columns:

                if rcol == tcol:
                    continue

                rcell = row_group[
                    row_group["_column"] == rcol
                ]

                if len(rcell) == 0:
                    continue

                release_raw = clean(
                    rcell.iloc[0][
                        "raw_cell"
                    ]
                )

                release_header = (
                    column_headers[
                        int(rcol)
                    ]
                )

                release_pct, release_method = (
                    parse_release(
                        release_raw,
                        release_header
                    )
                )

                if release_pct is None:
                    continue

                if not (
                    0 <= release_pct <= 100
                ):
                    continue

                if not (
                    0 <= time_h <= 5000
                ):
                    continue

                observations.append(
                    {
                        "candidate_id":
                            f"6A-{len(observations)+1:07d}",

                        "pmcid":
                            pmcid,

                        "table_index":
                            table_index,

                        "row_index":
                            row_index,

                        "time_column":
                            tcol,

                        "release_column":
                            rcol,

                        "time_raw":
                            time_raw,

                        "release_raw":
                            release_raw,

                        "release_time_h":
                            time_h,

                        "release_percent":
                            release_pct,

                        "time_method":
                            time_method,

                        "release_method":
                            release_method,

                        "time_header":
                            time_header,

                        "release_header":
                            release_header,

                        "source_type":
                            "STRUCTURED_COLUMN_PAIR",

                        "validation_status":
                            "CANDIDATE"
                    }
                )

        # ----------------------------------------------------
        # EXPLICIT RELEASE STATEMENTS
        #
        # Example:
        # "80% release after 24 h"
        #
        # These are accepted even when columns are not cleanly
        # identified, but only with BOTH explicit release % and
        # explicit time unit.
        # ----------------------------------------------------

        for _, cell in row_group.iterrows():

            text = clean(
                cell["raw_cell"]
            )

            if not text:
                continue

            if (
                "release" not in text.lower()
                and
                "released" not in text.lower()
            ):

                continue

            rel = re.search(
                r"""
                (?:cumulative\s+drug\s+release|
                   cumulative\s+release|
                   drug\s+release|
                   release|
                   released)
                [^%\d]{0,100}
                (\d+(?:\.\d+)?)
                \s*%
                """,
                text,
                flags=re.I | re.X
            )

            tm = re.search(
                r"""
                (\d+(?:\.\d+)?)
                \s*
                (minutes?|mins?|min|
                 hours?|hrs?|hr|h|
                 days?|day|d|
                 weeks?|week|wk)
                \b
                """,
                text,
                flags=re.I | re.X
            )

            if not rel or not tm:
                continue

            release_pct = float(
                rel.group(1)
            )

            value = float(
                tm.group(1)
            )

            unit = tm.group(2).lower()

            if unit.startswith(
                "min"
            ):

                time_h = value / 60

            elif (
                unit.startswith("h")
                or
                unit.startswith("hr")
            ):

                time_h = value

            elif (
                unit.startswith("d")
            ):

                time_h = value * 24

            elif (
                unit.startswith("w")
            ):

                time_h = value * 168

            else:

                continue

            if not (
                0 <= release_pct <= 100
            ):
                continue

            if not (
                0 <= time_h <= 5000
            ):
                continue

            observations.append(
                {
                    "candidate_id":
                        f"6A-{len(observations)+1:07d}",

                    "pmcid":
                        pmcid,

                    "table_index":
                        table_index,

                    "row_index":
                        row_index,

                    "time_column":
                        np.nan,

                    "release_column":
                        int(
                            cell["_column"]
                        ),

                    "time_raw":
                        tm.group(0),

                    "release_raw":
                        rel.group(0),

                    "release_time_h":
                        time_h,

                    "release_percent":
                        release_pct,

                    "time_method":
                        "EXPLICIT_TIME_UNIT",

                    "release_method":
                        "EXPLICIT_RELEASE_PERCENT",

                    "time_header":
                        "",

                    "release_header":
                        column_headers.get(
                            int(
                                cell["_column"]
                            ),
                            ""
                        ),

                    "source_type":
                        "EXPLICIT_RELEASE_STATEMENT",

                    "validation_status":
                        "CANDIDATE"
                }
            )

    if (
        counter % 100 == 0
        or
        counter == total_tables
    ):

        print(
            f"Processed "
            f"{counter:,}/"
            f"{total_tables:,} | "
            f"Candidates: "
            f"{len(observations):,}"
        )


# ============================================================
# DATAFRAME
# ============================================================

result = pd.DataFrame(
    observations
)

audit = pd.DataFrame(
    table_audit
)


# ============================================================
# EXACT SOURCE DUPLICATE REMOVAL
# ============================================================

if len(result) > 0:

    before = len(result)

    result = result.drop_duplicates(
        subset=[
            "pmcid",
            "table_index",
            "row_index",
            "time_column",
            "release_column",
            "time_raw",
            "release_raw",
            "release_time_h",
            "release_percent"
        ]
    ).copy()

    duplicates_removed = (
        before - len(result)
    )

else:

    duplicates_removed = 0


# ============================================================
# BASIC SANITY FLAGS
# ============================================================

if len(result) > 0:

    result["release_valid"] = (
        result["release_percent"]
        .between(
            0,
            100,
            inclusive="both"
        )
    )

    result["time_valid"] = (
        result["release_time_h"]
        .between(
            0,
            5000,
            inclusive="both"
        )
    )

    result["candidate_valid_numeric"] = (
        result["release_valid"]
        &
        result["time_valid"]
    )

else:

    result["candidate_valid_numeric"] = []


# ============================================================
# SAVE
# ============================================================

OBS_FILE = os.path.join(
    OUT_DIR,
    "06A_high_precision_release_candidates.csv"
)

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "06A_candidate_table_column_audit.csv"
)

result.to_csv(
    OBS_FILE,
    index=False
)

audit.to_csv(
    AUDIT_FILE,
    index=False
)


# ============================================================
# REPORT
# ============================================================

print("\n" + "=" * 70)
print("CODE 6A-R COMPLETE")
print("=" * 70)

print(
    "Candidate source tables:",
    total_tables
)

print(
    "Candidate observations before duplicate removal:",
    len(observations)
)

print(
    "Exact duplicate extractions removed:",
    duplicates_removed
)

print(
    "Final candidate observations:",
    len(result)
)

if len(result) > 0:

    print(
        "\nUnique PMCID:",
        result["pmcid"].nunique()
    )

    print(
        "Unique source tables:",
        result[
            ["pmcid", "table_index"]
        ].drop_duplicates().shape[0]
    )

    print(
        "Unique source rows:",
        result[
            [
                "pmcid",
                "table_index",
                "row_index"
            ]
        ].drop_duplicates().shape[0]
    )

    print(
        "\nTime extraction methods:"
    )

    print(
        result[
            "time_method"
        ].value_counts()
    )

    print(
        "\nRelease extraction methods:"
    )

    print(
        result[
            "release_method"
        ].value_counts()
    )

    print(
        "\nRelease statistics:"
    )

    print(
        result[
            "release_percent"
        ].describe()
    )

    print(
        "\nTime statistics:"
    )

    print(
        result[
            "release_time_h"
        ].describe()
    )

    print(
        "\nPreview:"
    )

    print(
        result[
            [
                "candidate_id",
                "pmcid",
                "table_index",
                "row_index",
                "time_raw",
                "release_raw",
                "release_time_h",
                "release_percent",
                "time_method",
                "release_method"
            ]
        ]
        .head(25)
        .to_string(
            index=False
        )
    )

print("\nSaved:")
print(OBS_FILE)
print(AUDIT_FILE)

print("\nIMPORTANT:")
print(
    "These are candidate observations only."
)
print(
    "They have NOT passed semantic validation."
)
print(
    "They have NOT passed source-level validation."
)
print(
    "No synthetic observations were generated."
)
print(
    "No interpolation was performed."
)
print(
    "No intentional duplication was performed."
)
print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ============================================================
# CODE 6B — STRUCTURAL TIME–RELEASE PAIRING REPAIR
# ============================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6B — STRUCTURAL TIME–RELEASE PAIRING REPAIR")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

CODE3_FILE = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

CODE4_FILE = os.path.join(
    ROOT,
    "data",
    "candidates",
    "04_release_table_candidates.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "candidates"
)

os.makedirs(
    OUT_DIR,
    exist_ok=True
)

if not os.path.exists(CODE3_FILE):
    raise FileNotFoundError(
        f"Code 3 file not found:\n{CODE3_FILE}"
    )

if not os.path.exists(CODE4_FILE):
    raise FileNotFoundError(
        f"Code 4 file not found:\n{CODE4_FILE}"
    )


# ============================================================
# LOAD
# ============================================================

cells = pd.read_csv(
    CODE3_FILE,
    low_memory=False
)

candidate_tables = pd.read_csv(
    CODE4_FILE,
    low_memory=False
)

print(
    f"Code 3 cells: {len(cells):,}"
)

print(
    f"Code 4 candidate tables: {len(candidate_tables):,}"
)


# ============================================================
# NORMALIZE STRUCTURAL KEYS
# ============================================================

cells["_pmcid"] = (
    cells["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

cells["_table"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

cells["_row"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

cells["_col"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)

candidate_tables["_pmcid"] = (
    candidate_tables["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

candidate_tables["_table"] = pd.to_numeric(
    candidate_tables["table_index"],
    errors="coerce"
)

candidate_keys = candidate_tables[
    ["_pmcid", "_table"]
].drop_duplicates()

cells = cells.merge(
    candidate_keys,
    on=["_pmcid", "_table"],
    how="inner"
)

print(
    f"Candidate-table structural cells: "
    f"{len(cells):,}"
)


# ============================================================
# TEXT HELPERS
# ============================================================

def clean(x):

    if pd.isna(x):
        return ""

    x = str(x)
    x = x.replace("\n", " ")
    x = x.replace("\xa0", " ")

    return re.sub(
        r"\s+",
        " ",
        x
    ).strip()


def lower(x):

    return clean(x).lower()


# ============================================================
# STRONG HEADER VOCABULARY
# ============================================================

TIME_STRONG = [
    "time (h)",
    "time (hr)",
    "time (hrs)",
    "time (hour)",
    "time (hours)",
    "time (day)",
    "time (days)",
    "time point",
    "timepoint",
    "release time",
    "sampling time",
    "hours",
    "hour",
    "hrs",
    "days",
    "day"
]

RELEASE_STRONG = [
    "drug release (%)",
    "drug release %",
    "cumulative drug release (%)",
    "cumulative drug release %",
    "cumulative release (%)",
    "cumulative release %",
    "release (%)",
    "release %",
    "released (%)",
    "released %",
    "cumulative (%)"
]

BAD_HEADER = [
    "pH",
    "particle size",
    "diameter",
    "size",
    "zeta",
    "polydispersity",
    "pdi",
    "loading",
    "encapsulation",
    "entrapment",
    "efficiency",
    "yield",
    "dose",
    "concentration",
    "cmax",
    "tmax",
    "auc",
    "half-life",
    "half life",
    "viability",
    "inhibition",
    "absorbance",
    "roughness",
    "contact angle",
    "weight loss",
    "swelling",
    "degradation"
]


# ============================================================
# HEADER SCORING
# ============================================================

def score_time_header(text):

    t = lower(text)

    score = 0

    for term in TIME_STRONG:

        if term.lower() in t:
            score += 5

    # Generic "time" gets only weak support.
    if (
        "time" in t
        and score == 0
    ):
        score += 1

    for term in BAD_HEADER:

        if term.lower() in t:
            score -= 5

    return score


def score_release_header(text):

    t = lower(text)

    score = 0

    for term in RELEASE_STRONG:

        if term.lower() in t:
            score += 5

    # Generic release wording is weaker.
    if (
        "release" in t
        and score == 0
    ):
        score += 1

    for term in BAD_HEADER:

        if term.lower() in t:
            score -= 5

    return score


# ============================================================
# GET HEADER FOR EACH COLUMN
# ============================================================

def column_header(
    group,
    col
):

    sub = group[
        group["_col"] == col
    ]

    texts = []

    if "header_text" in sub.columns:

        for x in sub[
            "header_text"
        ].dropna():

            x = clean(x)

            if x:
                texts.append(x)

    # Include first few row values ONLY as context.
    # They do NOT themselves define the header.

    result = " | ".join(
        list(
            dict.fromkeys(
                texts
            )
        )
    )

    return result


# ============================================================
# NUMERIC PARSERS
# ============================================================

def parse_numeric(
    text
):

    text = clean(text)

    # Do not accept pH as a numeric observation.
    if re.search(
        r"\bpH\s*=",
        text,
        flags=re.I
    ):
        return None

    # Diameter symbol.
    if "⌀" in text or "Ø" in text:
        return None

    # ± values are usually mean ± SD.
    # They are not release values by themselves.
    if "±" in text:
        return None

    # Concentration-like values.
    if re.search(
        r"\b(?:mg|µg|ug|ng|mol|mmol|mg/ml|mg/mL)\b",
        text,
        flags=re.I
    ):
        return None

    nums = re.findall(
        r"[-+]?\d+(?:\.\d+)?",
        text
    )

    if len(nums) != 1:
        return None

    try:

        return float(
            nums[0]
        )

    except:

        return None


# ============================================================
# TIME PARSER
# ============================================================

def parse_time(
    text,
    header
):

    text = clean(text)
    h = lower(header)

    # Explicit unit
    m = re.search(
        r"(\d+(?:\.\d+)?)\s*"
        r"(minutes?|mins?|min|"
        r"hours?|hrs?|hr|h|"
        r"days?|day|d|"
        r"weeks?|week|wk)",
        text,
        flags=re.I
    )

    if m:

        value = float(
            m.group(1)
        )

        unit = m.group(2).lower()

        if unit.startswith("min"):
            hours = value / 60

        elif (
            unit.startswith("h")
            or unit.startswith("hr")
        ):
            hours = value

        elif unit.startswith("d"):
            hours = value * 24

        elif unit.startswith("w"):
            hours = value * 168

        else:
            return None

        if 0 <= hours <= 5000:
            return hours

        return None

    # Numeric-only cell.
    value = parse_numeric(
        text
    )

    if value is None:
        return None

    # Numeric-only time is accepted ONLY for a strongly
    # identified time header.
    strong = any(
        term.lower() in h
        for term in TIME_STRONG
    )

    if not strong:
        return None

    if 0 <= value <= 5000:
        return value

    return None


# ============================================================
# RELEASE PARSER
# ============================================================

def parse_release(
    text,
    header
):

    text = clean(text)
    h = lower(header)

    # Never use pH / diameter / SD / concentration.
    if re.search(
        r"\bpH\s*=",
        text,
        flags=re.I
    ):
        return None

    if "⌀" in text or "Ø" in text:
        return None

    if "±" in text:
        return None

    if re.search(
        r"\b(?:mg|µg|ug|ng|mol|mmol)\b",
        text,
        flags=re.I
    ):
        return None

    # Explicit percent
    m = re.search(
        r"(\d+(?:\.\d+)?)\s*%",
        text
    )

    if m:

        value = float(
            m.group(1)
        )

        if 0 <= value <= 100:
            return value

        return None

    # Numeric-only release.
    value = parse_numeric(
        text
    )

    if value is None:
        return None

    strong = any(
        term.lower() in h
        for term in RELEASE_STRONG
    )

    if not strong:
        return None

    if 0 <= value <= 100:
        return value

    return None


# ============================================================
# PROCESS TABLES
# ============================================================

results = []
audit = []

groups = cells.groupby(
    ["_pmcid", "_table"],
    sort=False
)

total = len(groups)

print(
    f"\nCandidate tables to inspect: {total:,}"
)

for i, (
    key,
    group
) in enumerate(
    groups,
    start=1
):

    pmcid, table_index = key

    group = group.sort_values(
        ["_row", "_col"]
    ).copy()

    # --------------------------------------------------------
    # Determine actual column headers.
    # --------------------------------------------------------

    columns = sorted(
        group["_col"]
        .dropna()
        .unique()
    )

    headers = {}

    for col in columns:

        headers[int(col)] = (
            column_header(
                group,
                col
            )
        )

    # --------------------------------------------------------
    # Score columns.
    # --------------------------------------------------------

    time_scores = {
        col:
            score_time_header(
                header
            )
        for col, header
        in headers.items()
    }

    release_scores = {
        col:
            score_release_header(
                header
            )
        for col, header
        in headers.items()
    }

    # --------------------------------------------------------
    # ONLY strongly identified columns.
    # --------------------------------------------------------

    time_candidates = [
        col
        for col, score
        in time_scores.items()
        if score >= 5
    ]

    release_candidates = [
        col
        for col, score
        in release_scores.items()
        if score >= 5
    ]

    # --------------------------------------------------------
    # Record table audit.
    # --------------------------------------------------------

    audit.append(
        {
            "pmcid":
                pmcid,

            "table_index":
                table_index,

            "time_candidate_columns":
                ";".join(
                    map(
                        str,
                        time_candidates
                    )
                ),

            "release_candidate_columns":
                ";".join(
                    map(
                        str,
                        release_candidates
                    )
                ),

            "time_candidate_count":
                len(time_candidates),

            "release_candidate_count":
                len(release_candidates),

            "table_status":
                (
                    "PAIRABLE"
                    if (
                        len(time_candidates) >= 1
                        and
                        len(release_candidates) >= 1
                    )
                    else
                    "AMBIGUOUS"
                )
        }
    )

    # --------------------------------------------------------
    # IMPORTANT:
    # If there is no strong release column OR no strong time
    # column, do NOT manufacture pairs.
    # --------------------------------------------------------

    if (
        len(time_candidates) == 0
        or
        len(release_candidates) == 0
    ):

        continue

    # --------------------------------------------------------
    # If multiple strong time/release columns exist,
    # pair only structurally adjacent / corresponding
    # columns rather than Cartesian-product pairing.
    # --------------------------------------------------------

    pairs = []

    for tcol in time_candidates:

        # Closest release column.
        release_col = min(
            release_candidates,
            key=lambda x:
                abs(
                    int(x) - int(tcol)
                )
        )

        pairs.append(
            (
                tcol,
                release_col
            )
        )

    pairs = list(
        dict.fromkeys(
            pairs
        )
    )

    # --------------------------------------------------------
    # Row-level extraction
    # --------------------------------------------------------

    for row_index, row_group in group.groupby(
        "_row",
        sort=True
    ):

        row_group = row_group.sort_values(
            "_col"
        )

        for tcol, rcol in pairs:

            tcell = row_group[
                row_group["_col"] == tcol
            ]

            rcell = row_group[
                row_group["_col"] == rcol
            ]

            if (
                len(tcell) == 0
                or
                len(rcell) == 0
            ):
                continue

            time_raw = clean(
                tcell.iloc[0][
                    "raw_cell"
                ]
            )

            release_raw = clean(
                rcell.iloc[0][
                    "raw_cell"
                ]
            )

            time_h = parse_time(
                time_raw,
                headers[int(tcol)]
            )

            release_pct = parse_release(
                release_raw,
                headers[int(rcol)]
            )

            if time_h is None:
                continue

            if release_pct is None:
                continue

            results.append(
                {
                    "candidate_id":
                        f"6B-{len(results)+1:07d}",

                    "pmcid":
                        pmcid,

                    "table_index":
                        table_index,

                    "row_index":
                        row_index,

                    "time_column":
                        int(tcol),

                    "release_column":
                        int(rcol),

                    "time_raw":
                        time_raw,

                    "release_raw":
                        release_raw,

                    "release_time_h":
                        time_h,

                    "release_percent":
                        release_pct,

                    "time_header":
                        headers[int(tcol)],

                    "release_header":
                        headers[int(rcol)],

                    "source_type":
                        "STRICT_STRUCTURAL_PAIR",

                    "validation_status":
                        "CANDIDATE"
                }
            )

    if (
        i % 100 == 0
        or
        i == total
    ):

        print(
            f"Processed {i:,}/{total:,} | "
            f"Candidates: {len(results):,}"
        )


# ============================================================
# DATAFRAME
# ============================================================

result = pd.DataFrame(
    results
)

audit_df = pd.DataFrame(
    audit
)


# ============================================================
# EXACT SOURCE DUPLICATE REMOVAL
# ============================================================

if len(result) > 0:

    before = len(result)

    result = result.drop_duplicates(
        subset=[
            "pmcid",
            "table_index",
            "row_index",
            "time_column",
            "release_column",
            "time_raw",
            "release_raw",
            "release_time_h",
            "release_percent"
        ]
    )

    duplicates_removed = (
        before - len(result)
    )

else:

    duplicates_removed = 0


# ============================================================
# SANITY FILTER
# ============================================================

if len(result) > 0:

    result = result[
        result["release_percent"].between(
            0,
            100
        )
    ]

    result = result[
        result["release_time_h"].between(
            0,
            5000
        )
    ]


# ============================================================
# SAVE
# ============================================================

RESULT_FILE = os.path.join(
    OUT_DIR,
    "06B_strict_structural_release_candidates.csv"
)

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "06B_structural_pairing_audit.csv"
)

result.to_csv(
    RESULT_FILE,
    index=False
)

audit_df.to_csv(
    AUDIT_FILE,
    index=False
)


# ============================================================
# REPORT
# ============================================================

print("\n" + "=" * 70)
print("CODE 6B COMPLETE")
print("=" * 70)

print(
    f"Candidate tables examined: {total:,}"
)

print(
    f"Strict candidate observations: "
    f"{len(result):,}"
)

print(
    f"Exact duplicates removed: "
    f"{duplicates_removed:,}"
)

if len(result) > 0:

    print(
        "\nUnique PMCIDs:",
        result["pmcid"].nunique()
    )

    print(
        "Unique source tables:",
        result[
            [
                "pmcid",
                "table_index"
            ]
        ]
        .drop_duplicates()
        .shape[0]
    )

    print(
        "Unique source rows:",
        result[
            [
                "pmcid",
                "table_index",
                "row_index"
            ]
        ]
        .drop_duplicates()
        .shape[0]
    )

    print(
        "\nRelease statistics:"
    )

    print(
        result[
            "release_percent"
        ].describe()
    )

    print(
        "\nTime statistics:"
    )

    print(
        result[
            "release_time_h"
        ].describe()
    )

    print(
        "\nPreview:"
    )

    print(
        result[
            [
                "candidate_id",
                "pmcid",
                "table_index",
                "row_index",
                "time_raw",
                "release_raw",
                "release_time_h",
                "release_percent",
                "time_header",
                "release_header"
            ]
        ]
        .head(25)
        .to_string(
            index=False
        )
    )

print(
    "\nTable pairing status:"
)

print(
    audit_df[
        "table_status"
    ].value_counts()
)

print("\nSaved:")
print(RESULT_FILE)
print(AUDIT_FILE)

print("\nIMPORTANT:")
print(
    "These are candidate observations only."
)
print(
    "Only strongly identified time/release columns were used."
)
print(
    "Ambiguous numeric tables were excluded."
)
print(
    "No synthetic observations were generated."
)
print(
    "No interpolation was performed."
)
print(
    "No observations were intentionally duplicated."
)
print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ============================================================
# CODE 6C — XML HEADER-CELL STRUCTURE RECOVERY
# ============================================================

import os
import re
import pandas as pd
import numpy as np
import xml.etree.ElementTree as ET

print("=" * 70)
print("CODE 6C — XML HEADER-CELL STRUCTURE RECOVERY")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

XML_DIR = os.path.join(
    ROOT,
    "data",
    "raw",
    "pmc_xml"
)

CODE4_FILE = os.path.join(
    ROOT,
    "data",
    "candidates",
    "04_release_table_candidates.csv"
)

CODE3_FILE = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "candidates"
)

os.makedirs(
    OUT_DIR,
    exist_ok=True
)

if not os.path.isdir(XML_DIR):
    raise FileNotFoundError(
        f"PMC XML directory not found:\n{XML_DIR}"
    )

if not os.path.exists(CODE4_FILE):
    raise FileNotFoundError(
        f"Code 4 file not found:\n{CODE4_FILE}"
    )

if not os.path.exists(CODE3_FILE):
    raise FileNotFoundError(
        f"Code 3 file not found:\n{CODE3_FILE}"
    )


# ============================================================
# LOAD
# ============================================================

candidate_tables = pd.read_csv(
    CODE4_FILE,
    low_memory=False
)

cells = pd.read_csv(
    CODE3_FILE,
    low_memory=False
)

print(
    f"Code 4 candidate tables: {len(candidate_tables):,}"
)

print(
    f"Code 3 cells: {len(cells):,}"
)


# ============================================================
# NORMALIZE KEYS
# ============================================================

candidate_tables["_pmcid"] = (
    candidate_tables["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

candidate_tables["_table"] = pd.to_numeric(
    candidate_tables["table_index"],
    errors="coerce"
)

cells["_pmcid"] = (
    cells["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

cells["_table"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

cells["_row"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

cells["_col"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)


candidate_keys = candidate_tables[
    ["_pmcid", "_table"]
].drop_duplicates()


# ============================================================
# TEXT HELPERS
# ============================================================

def clean(x):

    if x is None:
        return ""

    x = str(x)

    x = (
        x.replace("\n", " ")
         .replace("\t", " ")
         .replace("\xa0", " ")
    )

    return re.sub(
        r"\s+",
        " ",
        x
    ).strip()


def local_name(tag):

    if "}" in tag:
        return tag.split("}", 1)[1]

    return tag


# ============================================================
# XML TABLE EXTRACTION
# ============================================================

def get_table_elements(root):

    tables = []

    for elem in root.iter():

        if local_name(
            elem.tag
        ) == "table":

            tables.append(
                elem
            )

    return tables


def cell_text(elem):

    parts = []

    for x in elem.iter():

        if x.text:
            parts.append(
                x.text
            )

    return clean(
        " ".join(parts)
    )


def get_rows(table):

    rows = []

    for elem in table.iter():

        if local_name(
            elem.tag
        ) == "tr":

            rows.append(
                elem
            )

    return rows


def get_cells(row):

    output = []

    for elem in row:

        name = local_name(
            elem.tag
        )

        if name in [
            "td",
            "th"
        ]:

            output.append(
                (
                    name,
                    cell_text(elem)
                )
            )

    return output


# ============================================================
# HEADER IDENTIFICATION
# ============================================================

TIME_TERMS = [
    "time (h)",
    "time (hr)",
    "time (hrs)",
    "time (hour)",
    "time (hours)",
    "time (day)",
    "time (days)",
    "time point",
    "timepoint",
    "release time",
    "sampling time",
    "duration",
    "hours",
    "days"
]

RELEASE_TERMS = [
    "drug release (%)",
    "drug release %",
    "cumulative drug release (%)",
    "cumulative drug release %",
    "cumulative release (%)",
    "cumulative release %",
    "release (%)",
    "release %",
    "released (%)",
    "released %",
    "maximum drug release (%)",
    "maximum release (%)"
]

BAD_TERMS = [
    "particle size",
    "diameter",
    "zeta potential",
    "pdi",
    "polydispersity",
    "loading",
    "encapsulation efficiency",
    "entrapment efficiency",
    "inhibition",
    "cell viability",
    "cmax",
    "tmax",
    "auc",
    "half-life",
    "dose",
    "concentration",
    "ph",
    "contact angle"
]


def score_time_header(
    text
):

    t = clean(text).lower()

    score = 0

    for term in TIME_TERMS:

        if term.lower() in t:
            score += 10

    for term in BAD_TERMS:

        if term.lower() in t:
            score -= 10

    return score


def score_release_header(
    text
):

    t = clean(text).lower()

    score = 0

    for term in RELEASE_TERMS:

        if term.lower() in t:
            score += 10

    for term in BAD_TERMS:

        if term.lower() in t:
            score -= 10

    return score


# ============================================================
# FIND REAL XML HEADER CELLS
# ============================================================

def recover_headers(
    table
):

    rows = get_rows(
        table
    )

    header_rows = []

    # --------------------------------------------------------
    # First priority: explicit <thead>
    # --------------------------------------------------------

    for elem in table.iter():

        if local_name(
            elem.tag
        ) == "thead":

            for child in elem.iter():

                if local_name(
                    child.tag
                ) == "tr":

                    header_rows.append(
                        child
                    )

    # --------------------------------------------------------
    # Second priority: rows containing <th>
    # --------------------------------------------------------

    if not header_rows:

        for row in rows:

            cells_here = get_cells(
                row
            )

            if any(
                kind == "th"
                for kind, text
                in cells_here
            ):

                header_rows.append(
                    row
                )

    # --------------------------------------------------------
    # If no <th>, inspect first rows cautiously.
    # --------------------------------------------------------

    if not header_rows and rows:

        for row in rows[:3]:

            cells_here = get_cells(
                row
            )

            texts = [
                text
                for kind, text
                in cells_here
            ]

            combined = " | ".join(
                texts
            ).lower()

            if (
                "release" in combined
                or
                "time" in combined
            ):

                header_rows.append(
                    row
                )

                break

    # --------------------------------------------------------
    # Convert header rows into positional cells.
    # --------------------------------------------------------

    header_cells = []

    for row in header_rows:

        cells_here = get_cells(
            row
        )

        for col, (
            kind,
            text
        ) in enumerate(
            cells_here
        ):

            if text:

                header_cells.append(
                    {
                        "column_index":
                            col,

                        "header":
                            clean(text),

                        "cell_type":
                            kind
                    }
                )

    return header_cells


# ============================================================
# BUILD XML INDEX
# ============================================================

candidate_pmcs = set(
    candidate_tables[
        "_pmcid"
    ].dropna()
)

xml_files = {}

for pmcid in candidate_pmcs:

    path = os.path.join(
        XML_DIR,
        f"{pmcid}.xml"
    )

    if os.path.exists(path):

        xml_files[
            pmcid
        ] = path

print(
    f"\nCandidate PMCIDs with XML: "
    f"{len(xml_files):,}"
)


# ============================================================
# PROCESS CANDIDATE TABLES
# ============================================================

header_records = []
pair_results = []
table_audit = []

for n, row in candidate_tables.iterrows():

    pmcid = row["_pmcid"]
    table_index = int(
        row["_table"]
    )

    xml_path = xml_files.get(
        pmcid
    )

    if xml_path is None:

        continue

    try:

        tree = ET.parse(
            xml_path
        )

        root = tree.getroot()

    except Exception:

        continue

    xml_tables = get_table_elements(
        root
    )

    if (
        table_index < 0
        or
        table_index >= len(xml_tables)
    ):

        continue

    table = xml_tables[
        table_index
    ]

    headers = recover_headers(
        table
    )

    if not headers:

        table_audit.append(
            {
                "pmcid":
                    pmcid,

                "table_index":
                    table_index,

                "header_recovery":
                    "NO_HEADER_FOUND",

                "time_columns":
                    "",

                "release_columns":
                    ""
            }
        )

        continue

    time_cols = []
    release_cols = []

    for h in headers:

        tscore = score_time_header(
            h["header"]
        )

        rscore = score_release_header(
            h["header"]
        )

        if tscore >= 10:

            time_cols.append(
                h["column_index"]
            )

        if rscore >= 10:

            release_cols.append(
                h["column_index"]
            )

        header_records.append(
            {
                "pmcid":
                    pmcid,

                "table_index":
                    table_index,

                "column_index":
                    h["column_index"],

                "header":
                    h["header"],

                "time_score":
                    tscore,

                "release_score":
                    rscore
            }
        )

    table_audit.append(
        {
            "pmcid":
                pmcid,

            "table_index":
                table_index,

            "header_recovery":
                "RECOVERED",

            "time_columns":
                ";".join(
                    map(
                        str,
                        time_cols
                    )
                ),

            "release_columns":
                ";".join(
                    map(
                        str,
                        release_cols
                    )
                )
        }
    )


# ============================================================
# SAVE HEADER RECOVERY
# ============================================================

headers_df = pd.DataFrame(
    header_records
)

audit_df = pd.DataFrame(
    table_audit
)

HEADER_FILE = os.path.join(
    OUT_DIR,
    "06C_xml_header_recovery.csv"
)

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "06C_xml_header_recovery_audit.csv"
)

headers_df.to_csv(
    HEADER_FILE,
    index=False
)

audit_df.to_csv(
    AUDIT_FILE,
    index=False
)


# ============================================================
# REPORT
# ============================================================

print(
    "\n" + "=" * 70
)

print(
    "CODE 6C COMPLETE"
)

print(
    "=" * 70
)

print(
    f"Candidate tables: "
    f"{len(candidate_tables):,}"
)

print(
    f"Tables with recovered XML headers: "
    f"{len(audit_df):,}"
)

if len(audit_df) > 0:

    print(
        "\nHeader recovery:"
    )

    print(
        audit_df[
            "header_recovery"
        ].value_counts()
    )

    print(
        "\nTables with time columns:"
    )

    print(
        (
            audit_df[
                "time_columns"
            ]
            .fillna("")
            .str.len()
            > 0
        ).sum()
    )

    print(
        "Tables with release columns:"
    )

    print(
        (
            audit_df[
                "release_columns"
            ]
            .fillna("")
            .str.len()
            > 0
        ).sum()
    )

print(
    "\nRecovered header examples:"
)

if len(headers_df) > 0:

    print(
        headers_df[
            [
                "pmcid",
                "table_index",
                "column_index",
                "header",
                "time_score",
                "release_score"
            ]
        ]
        .head(40)
        .to_string(
            index=False
        )
    )

print(
    "\nSaved:"
)

print(
    HEADER_FILE
)

print(
    AUDIT_FILE
)

print(
    "\nIMPORTANT:"
)

print(
    "This stage only recovers actual XML header-cell structure."
)

print(
    "It does NOT create release observations."
)

print(
    "It does NOT interpolate values."
)

print(
    "It does NOT duplicate observations."
)

print(
    "It does NOT run machine learning."
)

print(
    "=" * 70
)
```

```python
# ============================================================
# CODE 6C-R — IMPROVED XML HEADER SEMANTIC CLASSIFICATION
# ============================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6C-R — IMPROVED XML HEADER SEMANTIC CLASSIFICATION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT,
    "data",
    "candidates",
    "06C_xml_header_recovery.csv"
)

OUTPUT_FILE = os.path.join(
    ROOT,
    "data",
    "candidates",
    "06CR_xml_header_classification.csv"
)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"Required file not found:\n{INPUT_FILE}"
    )

df = pd.read_csv(
    INPUT_FILE,
    low_memory=False
)

print(
    f"Headers loaded: {len(df):,}"
)


# ============================================================
# NORMALIZATION
# ============================================================

def clean(x):

    if pd.isna(x):
        return ""

    x = str(x)

    x = (
        x.replace("\n", " ")
         .replace("\t", " ")
         .replace("\xa0", " ")
    )

    return re.sub(
        r"\s+",
        " ",
        x
    ).strip()


df["header_clean"] = (
    df["header"]
    .apply(clean)
)

df["header_lower"] = (
    df["header_clean"]
    .str.lower()
)


# ============================================================
# STRONG RELEASE TERMS
# ============================================================

RELEASE_STRONG = [

    r"\bdrug\s+release\b",

    r"\bcumulative\s+drug\s+release\b",

    r"\bcumulative\s+release\b",

    r"\btotal\s+drug\s+released\b",

    r"\btotal\s+drug\s+release\b",

    r"\bdrug\s+released\b",

    r"\bamount\s+released\b",

    r"\bamount\s+of\s+drug\s+released\b",

    r"\bmaximum\s+drug\s+release\b",

    r"\bmaximum\s+release\b",

    r"\brelease\s*\(%\)",

    r"\brelease\s*%",

    r"\breleased\s*\(%\)",

    r"\breleased\s*%",

    r"\bcumulative\s*\(%\)"
]


# ============================================================
# STRONG TIME TERMS
# ============================================================

TIME_STRONG = [

    r"\btime\s*\(\s*h(?:r|rs|ours?)?\s*\)",

    r"\btime\s*\(\s*days?\s*\)",

    r"\btime\s*\(\s*min(?:ute)?s?\s*\)",

    r"\btime\s*\(\s*weeks?\s*\)",

    r"\bhours?\b",

    r"\bhrs?\b",

    r"\bdays?\b",

    r"\btime\s+point\b",

    r"\btimepoint\b",

    r"\bsampling\s+time\b",

    r"\brelease\s+time\b",

    r"\btime\s+for\s+maximum\b",

    r"\btime\s+to\s+maximum\b",

    r"\btime\s+at\s+maximum\b"
]


# ============================================================
# DANGEROUS / NON-RELEASE TERMS
# ============================================================

BAD_RELEASE = [

    r"\bparticle\s+size\b",

    r"\bdiameter\b",

    r"\bzeta\s+potential\b",

    r"\bpolydispersity\b",

    r"\bpdi\b",

    r"\bencapsulation\b",

    r"\bentrapment\b",

    r"\bloading\b",

    r"\bcell\s+viability\b",

    r"\bviability\b",

    r"\binhibition\b",

    r"\bcmax\b",

    r"\btmax\b",

    r"\bauc\b",

    r"\bhalf[- ]life\b",

    r"\bdose\b",

    r"\bconcentration\b",

    r"\bcontact\s+angle\b",

    r"\broughness\b",

    r"\bpH\b",

    r"\byield\b"
]


# ============================================================
# TIME EXCLUSIONS
# ============================================================

BAD_TIME = [

    r"\bparticle\s+size\b",

    r"\bdiameter\b",

    r"\bsize\b",

    r"\bconcentration\b",

    r"\bdose\b",

    r"\bloading\b",

    r"\btemperature\b",

    r"\bph\b",

    r"\bcmax\b",

    r"\bauc\b",

    r"\bhalf[- ]life\b"
]


def matches_any(
    text,
    patterns
):

    for pattern in patterns:

        if re.search(
            pattern,
            text,
            flags=re.I
        ):

            return True

    return False


# ============================================================
# CLASSIFICATION
# ============================================================

def classify_release(
    text
):

    if not text:

        return (
            "NOT_RELEASE",
            0
        )

    if matches_any(
        text,
        BAD_RELEASE
    ):

        return (
            "NOT_RELEASE",
            -20
        )

    if matches_any(
        text,
        RELEASE_STRONG
    ):

        return (
            "RELEASE",
            10
        )

    # Generic "release" is only weak evidence.
    if re.search(
        r"\brelease\b",
        text,
        flags=re.I
    ):

        return (
            "RELEASE_WEAK",
            2
        )

    return (
        "NOT_RELEASE",
        0
    )


def classify_time(
    text
):

    if not text:

        return (
            "NOT_TIME",
            0
        )

    if matches_any(
        text,
        BAD_TIME
    ):

        return (
            "NOT_TIME",
            -20
        )

    # --------------------------------------------------------
    # Time-to-maximum release
    # --------------------------------------------------------

    if re.search(
        r"\btime\s+(?:for|to|at)\s+maximum\b",
        text,
        flags=re.I
    ):

        return (
            "TIME_TO_MAX_RELEASE",
            10
        )

    # --------------------------------------------------------
    # Explicit time axis
    # --------------------------------------------------------

    if matches_any(
        text,
        TIME_STRONG
    ):

        return (
            "TIME",
            10
        )

    # --------------------------------------------------------
    # Duration needs caution.
    # --------------------------------------------------------

    if re.search(
        r"\bduration\b",
        text,
        flags=re.I
    ):

        if re.search(
            r"\brelease\b",
            text,
            flags=re.I
        ):

            return (
                "RELEASE_DURATION_REVIEW",
                5
            )

        return (
            "DURATION_OTHER",
            0
        )

    return (
        "NOT_TIME",
        0
    )


release_class = []
release_score = []

time_class = []
time_score = []

for text in df[
    "header_clean"
]:

    rc, rs = classify_release(
        text
    )

    tc, ts = classify_time(
        text
    )

    release_class.append(rc)
    release_score.append(rs)

    time_class.append(tc)
    time_score.append(ts)


df[
    "release_class"
] = release_class

df[
    "release_semantic_score"
] = release_score

df[
    "time_class"
] = time_class

df[
    "time_semantic_score"
] = time_score


# ============================================================
# FINAL ROLE
# ============================================================

def determine_role(
    row
):

    rc = row[
        "release_class"
    ]

    tc = row[
        "time_class"
    ]

    if (
        rc == "RELEASE"
        and
        tc in [
            "TIME",
            "TIME_TO_MAX_RELEASE"
        ]
    ):

        return "TIME_AND_RELEASE_HEADER"

    if rc == "RELEASE":

        return "RELEASE_HEADER"

    if tc in [
        "TIME",
        "TIME_TO_MAX_RELEASE"
    ]:

        return "TIME_HEADER"

    if tc == "RELEASE_DURATION_REVIEW":

        return "RELEASE_DURATION_REVIEW"

    if rc == "RELEASE_WEAK":

        return "RELEASE_WEAK"

    return "OTHER"


df[
    "header_role"
] = df.apply(
    determine_role,
    axis=1
)


# ============================================================
# SAVE
# ============================================================

df.to_csv(
    OUTPUT_FILE,
    index=False
)


# ============================================================
# REPORT
# ============================================================

print(
    "\n" + "=" * 70
)

print(
    "CODE 6C-R COMPLETE"
)

print(
    "=" * 70
)

print(
    "\nHeader roles:"
)

print(
    df[
        "header_role"
    ].value_counts()
)

print(
    "\nRelease classifications:"
)

print(
    df[
        "release_class"
    ].value_counts()
)

print(
    "\nTime classifications:"
)

print(
    df[
        "time_class"
    ].value_counts()
)

print(
    "\nRELEASE HEADERS:"
)

release_df = df[
    df["release_class"].isin(
        [
            "RELEASE",
            "RELEASE_WEAK"
        ]
    )
]

if len(release_df) > 0:

    print(
        release_df[
            [
                "pmcid",
                "table_index",
                "column_index",
                "header_clean",
                "release_class",
                "release_semantic_score"
            ]
        ]
        .drop_duplicates()
        .head(100)
        .to_string(
            index=False
        )
    )

print(
    "\nTIME HEADERS:"
)

time_df = df[
    df["time_class"].isin(
        [
            "TIME",
            "TIME_TO_MAX_RELEASE",
            "RELEASE_DURATION_REVIEW"
        ]
    )
]

if len(time_df) > 0:

    print(
        time_df[
            [
                "pmcid",
                "table_index",
                "column_index",
                "header_clean",
                "time_class",
                "time_semantic_score"
            ]
        ]
        .drop_duplicates()
        .head(100)
        .to_string(
            index=False
        )
    )

print(
    "\nSaved:"
)

print(
    OUTPUT_FILE
)

print(
    "\nIMPORTANT:"
)

print(
    "This stage only classifies recovered XML headers."
)

print(
    "It does NOT create observations."
)

print(
    "It does NOT pair time and release values."
)

print(
    "It does NOT interpolate or synthesize data."
)

print(
    "It does NOT run ML."
)

print(
    "=" * 70
)
```

```python
# ======================================================================
# CODE 6C-R2 — STRICT ANALYTICAL HEADER CLASSIFICATION
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6C-R2 — STRICT ANALYTICAL HEADER CLASSIFICATION")
print("=" * 70)

# ----------------------------------------------------------------------
# PATHS
# ----------------------------------------------------------------------

input_file = (
    "/content/AI_Polymer_Drug_Release/data/candidates/"
    "06CR_xml_header_classification.csv"
)

output_file = (
    "/content/AI_Polymer_Drug_Release/data/candidates/"
    "06CR2_strict_header_classification.csv"
)

if not os.path.exists(input_file):
    raise FileNotFoundError(
        f"Code 6C-R output not found:\n{input_file}"
    )

df = pd.read_csv(input_file)

print(f"Headers loaded: {len(df):,}")
print(f"Columns: {list(df.columns)}")

# ----------------------------------------------------------------------
# COLUMN DETECTION
# ----------------------------------------------------------------------

def detect_col(columns, candidates):
    lower = {str(c).lower(): c for c in columns}

    for x in candidates:
        if x.lower() in lower:
            return lower[x.lower()]

    for c in columns:
        cl = str(c).lower()
        for x in candidates:
            if x.lower() in cl:
                return c

    return None


pmcid_col = detect_col(
    df.columns,
    ["pmcid"]
)

table_col = detect_col(
    df.columns,
    ["table_index"]
)

column_col = detect_col(
    df.columns,
    ["column_index"]
)

header_col = detect_col(
    df.columns,
    ["header_clean", "header", "header_text"]
)

if header_col is None:
    raise ValueError(
        "Could not detect header column."
    )

print("\nDetected columns:")
print("PMCID   :", pmcid_col)
print("Table   :", table_col)
print("Column  :", column_col)
print("Header  :", header_col)

# ----------------------------------------------------------------------
# NORMALIZATION
# ----------------------------------------------------------------------

df["header_clean_r2"] = (
    df[header_col]
    .fillna("")
    .astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
    .str.lower()
)

# ----------------------------------------------------------------------
# STRONG RELEASE-VALUE PATTERNS
# ----------------------------------------------------------------------

release_value_patterns = [
    r"\bcumulative\b.*\brelease\b.*%",
    r"\bcumulative\b.*\breleased\b.*%",
    r"\bdrug\s+release\b.*%",
    r"\bdrug\s+released\b.*%",
    r"\brelease\b.*\(%\)",
    r"\brelease\s*\(%\)",
    r"\b%.*release\b",
    r"\bpercent(?:age)?\b.*\brelease\b",
    r"\brelease\b.*\bpercent(?:age)?\b",
    r"\bamount\b.*\brelease\b",
    r"\btotal\s+drug\s+released\b",
    r"\btotal\s+drug\s+release\b",
    r"\bmaximum\s+drug\s+release\b",
    r"\bmaximum\s+release\b.*%",
    r"\bdrug\s+release\s+after\b.*%",
    r"\bdo[xc]?\s+release\b.*%",
    r"\brelease\s+of\b.*%",
]

# ----------------------------------------------------------------------
# STRONG TIME-SAMPLING PATTERNS
# ----------------------------------------------------------------------

time_patterns = [
    r"^time\s*\(\s*h",
    r"^time\s*\(\s*hr",
    r"^time\s*\(\s*hours",
    r"^time\s*\(\s*min",
    r"^time\s*\(\s*minutes",
    r"^time\s*\(\s*days",
    r"^time\s*\(\s*day",
    r"^time\s*\(\s*weeks",
    r"^time\s*\(\s*week",
    r"^time\s+point$",
    r"^time$",
    r"^time\s*\[",
    r"^days?$",
    r"^hours?$",
    r"^hrs?\.?$",
]

# ----------------------------------------------------------------------
# TIME-TO-EVENT PATTERNS
# ----------------------------------------------------------------------

time_event_patterns = [
    r"time.*maximum.*release",
    r"time.*maximum.*drug.*release",
    r"time.*max.*release",
    r"release.*time",
]

# ----------------------------------------------------------------------
# NON-RELEASE / NON-SAMPLING TIME EXCLUSIONS
# ----------------------------------------------------------------------

non_release_time_patterns = [
    r"sonication",
    r"stirring",
    r"mixing",
    r"centrifug",
    r"incubation",
    r"incubated",
    r"storage",
    r"retention",
    r"residence",
    r"flow\s+rate",
    r"hausner",
    r"disintegration",
    r"dissolution\s+time",
    r"drying",
    r"freeze[- ]drying",
    r"lyophiliz",
    r"fabrication",
    r"preparation",
    r"reaction\s+time",
    r"crosslinking\s+time",
    r"gelation\s+time",
    r"processing\s+time",
    r"lag\s+time",
    r"half[- ]life",
]

# ----------------------------------------------------------------------
# RELEASE SEMANTIC EXCLUSIONS
# ----------------------------------------------------------------------

non_value_release_patterns = [
    r"\brelease\s+mechanism\b",
    r"\brelease\s+kinetic",
    r"\brelease\s+rate\b",
    r"\brelease\s+exponent\b",
    r"\brelease\s+constant\b",
    r"\brate\s+constant\b",
    r"\bmechanism.*release\b",
    r"\bmodel.*release\b",
    r"\brelease\s+conditions?\b",
    r"\brelease\s+duration\b",
    r"\bduration.*release\b",
    r"\brelease\s+strategy\b",
    r"\brelease\s+configuration\b",
    r"\brelease\s+behavior\b",
    r"\brelease\s+property\b",
]

# ----------------------------------------------------------------------
# CLASSIFICATION
# ----------------------------------------------------------------------

def matches_any(text, patterns):
    return any(re.search(p, text, flags=re.I) for p in patterns)


def classify_header(text):

    t = text.strip()

    if not t:
        return (
            "OTHER",
            "NOT_RELEASE",
            "NOT_TIME",
            0,
            0
        )

    # --------------------------------------------------------------
    # HARD EXCLUSIONS
    # --------------------------------------------------------------

    if matches_any(t, non_release_time_patterns):
        return (
            "OTHER",
            "NOT_RELEASE",
            "NOT_TIME",
            0,
            0
        )

    # --------------------------------------------------------------
    # STRONG RELEASE VALUE
    # --------------------------------------------------------------

    strong_release = matches_any(
        t,
        release_value_patterns
    )

    release_excluded = matches_any(
        t,
        non_value_release_patterns
    )

    # --------------------------------------------------------------
    # TIME-TO-RELEASE EVENT
    # --------------------------------------------------------------

    time_event = matches_any(
        t,
        time_event_patterns
    )

    # Explicit maximum-release headers are analytically useful.
    if (
        "maximum" in t
        and "release" in t
        and (
            "%" in t
            or "percent" in t
        )
    ):
        return (
            "TIME_AND_RELEASE_EVENT",
            "RELEASE",
            "TIME",
            10,
            10
        )

    # --------------------------------------------------------------
    # STRONG RELEASE COLUMN
    # --------------------------------------------------------------

    if strong_release and not release_excluded:
        return (
            "RELEASE_VALUE_COLUMN",
            "RELEASE",
            "NOT_TIME",
            10,
            0
        )

    # --------------------------------------------------------------
    # NON-NUMERICAL RELEASE SEMANTICS
    # --------------------------------------------------------------

    if release_excluded:
        return (
            "NON_ANALYTICAL_RELEASE",
            "RELEASE_WEAK",
            "NOT_TIME",
            2,
            0
        )

    # --------------------------------------------------------------
    # TIME SAMPLING
    # --------------------------------------------------------------

    strong_time = matches_any(
        t,
        time_patterns
    )

    if strong_time:
        return (
            "TIME_SAMPLING_COLUMN",
            "NOT_RELEASE",
            "TIME",
            0,
            10
        )

    # --------------------------------------------------------------
    # DURATION / RELEASE TIME — REVIEW
    # --------------------------------------------------------------

    if (
        "release" in t
        and (
            "duration" in t
            or "time" in t
        )
    ):
        return (
            "RELEASE_TIME_EVENT_REVIEW",
            "RELEASE_WEAK",
            "DURATION_OTHER",
            2,
            5
        )

    # --------------------------------------------------------------
    # OTHER
    # --------------------------------------------------------------

    return (
        "OTHER",
        "NOT_RELEASE",
        "NOT_TIME",
        0,
        0
    )


results = df["header_clean_r2"].apply(
    classify_header
)

df[
    [
        "header_role_r2",
        "release_class_r2",
        "time_class_r2",
        "release_score_r2",
        "time_score_r2"
    ]
] = pd.DataFrame(
    results.tolist(),
    index=df.index
)

# ----------------------------------------------------------------------
# ADD ORIGINAL HEADER FOR AUDIT
# ----------------------------------------------------------------------

df["header_original"] = df[header_col]

# ----------------------------------------------------------------------
# SAVE
# ----------------------------------------------------------------------

os.makedirs(
    os.path.dirname(output_file),
    exist_ok=True
)

df.to_csv(
    output_file,
    index=False
)

# ----------------------------------------------------------------------
# REPORT
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("CODE 6C-R2 COMPLETE")
print("=" * 70)

print("\nHeader roles:")
print(
    df["header_role_r2"]
    .value_counts()
)

print("\nRelease classifications:")
print(
    df["release_class_r2"]
    .value_counts()
)

print("\nTime classifications:")
print(
    df["time_class_r2"]
    .value_counts()
)

print("\n" + "=" * 70)
print("RELEASE VALUE HEADERS")
print("=" * 70)

release_view = df[
    df["header_role_r2"].isin(
        [
            "RELEASE_VALUE_COLUMN",
            "TIME_AND_RELEASE_EVENT"
        ]
    )
][
    [
        c for c in
        [
            pmcid_col,
            table_col,
            column_col,
            "header_original",
            "header_role_r2",
            "release_score_r2",
            "time_score_r2"
        ]
        if c is not None
    ]
]

print(
    release_view.head(100).to_string(
        index=False
    )
)

print("\n" + "=" * 70)
print("TIME SAMPLING HEADERS")
print("=" * 70)

time_view = df[
    df["header_role_r2"].isin(
        [
            "TIME_SAMPLING_COLUMN",
            "TIME_AND_RELEASE_EVENT"
        ]
    )
][
    [
        c for c in
        [
            pmcid_col,
            table_col,
            column_col,
            "header_original",
            "header_role_r2",
            "release_score_r2",
            "time_score_r2"
        ]
        if c is not None
    ]
]

print(
    time_view.head(100).to_string(
        index=False
    )
)

print("\nSaved:")
print(output_file)

print("\nIMPORTANT:")
print("This stage ONLY classifies actual XML headers.")
print("It does NOT create observations.")
print("It does NOT pair time and release values.")
print("It does NOT interpolate or synthesize data.")
print("It does NOT duplicate observations.")
print("It does NOT run ML.")

print("=" * 70)
```

```python
# ======================================================================
# CODE 6D — STRICT XML STRUCTURAL TIME–RELEASE PAIRING
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6D — STRICT XML STRUCTURAL TIME–RELEASE PAIRING")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

HEADER_FILE = os.path.join(
    ROOT,
    "data",
    "candidates",
    "06CR2_strict_header_classification.csv"
)

CELLS_FILE = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT_FILE = os.path.join(
    ROOT,
    "data",
    "candidates",
    "06D_strict_xml_time_release_candidates.csv"
)

AUDIT_FILE = os.path.join(
    ROOT,
    "data",
    "candidates",
    "06D_structural_pairing_audit.csv"
)

if not os.path.exists(HEADER_FILE):
    raise FileNotFoundError(
        f"Missing Code 6C-R2 output:\n{HEADER_FILE}"
    )

if not os.path.exists(CELLS_FILE):
    raise FileNotFoundError(
        f"Missing Code 3 table-cell file:\n{CELLS_FILE}"
    )


# ======================================================================
# LOAD
# ======================================================================

headers = pd.read_csv(
    HEADER_FILE,
    low_memory=False
)

cells = pd.read_csv(
    CELLS_FILE,
    low_memory=False
)

print(
    f"Classified XML headers: {len(headers):,}"
)

print(
    f"Code 3 cells: {len(cells):,}"
)


# ======================================================================
# NORMALIZE STRUCTURAL KEYS
# ======================================================================

for df in [headers, cells]:

    df["_pmcid"] = (
        df["pmcid"]
        .astype(str)
        .str.strip()
        .str.upper()
    )

    df["_table"] = pd.to_numeric(
        df["table_index"],
        errors="coerce"
    )

    df["_column"] = pd.to_numeric(
        df["column_index"],
        errors="coerce"
    )

cells["_row"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)


# ======================================================================
# HEADER ROLES
# ======================================================================

release_headers = headers[
    headers["header_role_r2"].isin(
        [
            "RELEASE_VALUE_COLUMN",
            "TIME_AND_RELEASE_EVENT"
        ]
    )
].copy()

time_headers = headers[
    headers["header_role_r2"].isin(
        [
            "TIME_SAMPLING_COLUMN",
            "TIME_AND_RELEASE_EVENT"
        ]
    )
].copy()


print(
    f"Release-value headers: {len(release_headers):,}"
)

print(
    f"Time headers: {len(time_headers):,}"
)


# ======================================================================
# HELPER FUNCTIONS
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
        .replace("\xa0", " ")
        .replace("\n", " ")
        .replace("\t", " ")
    ).strip()


def parse_numeric(text):

    if text is None:
        return None

    s = clean(text)

    if not s:
        return None

    # Remove common footnote markers.
    s = re.sub(
        r"[\[\]\(\)†‡*]+",
        " ",
        s
    )

    # Unicode minus.
    s = (
        s.replace("−", "-")
         .replace("–", "-")
         .replace("—", "-")
    )

    # Percentage sign is acceptable.
    s = s.replace("%", " ")

    # Handle inequalities conservatively.
    s = re.sub(
        r"^(?:<|>)\s*",
        "",
        s
    )

    # Remove ± SD after the first numeric value.
    s = re.split(
        r"\s*(?:±|\+/-)\s*",
        s
    )[0]

    # Only accept cells that are essentially numeric.
    if not re.fullmatch(
        r"[-+]?\d*\.?\d+(?:[eE][-+]?\d+)?",
        s.strip()
    ):
        return None

    try:
        return float(
            s.strip()
        )
    except Exception:
        return None


# ======================================================================
# TIME UNIT CONVERSION
# ======================================================================

def time_to_hours(
    value,
    header
):

    if value is None:
        return None

    h = clean(header).lower()

    # minutes
    if (
        "min" in h
        and "time" in h
    ):
        return value / 60.0

    # days
    if (
        "day" in h
        or h.strip() == "days"
        or h.strip() == "day"
    ):
        return value * 24.0

    # weeks
    if "week" in h:
        return value * 168.0

    # hours
    if (
        "hour" in h
        or "hr" in h
        or "(h" in h
        or "[h" in h
        or h.strip() == "time"
    ):
        return value

    # Unknown unit:
    # Do NOT guess.
    return None


# ======================================================================
# RELEASE VALIDITY
# ======================================================================

def valid_release(
    value
):

    if value is None:
        return False

    # Percentage release should normally be 0–100.
    return (
        value >= 0
        and value <= 100
    )


# ======================================================================
# INDEX CELLS
# ======================================================================

cells = cells[
    cells["_pmcid"].notna()
    &
    cells["_table"].notna()
    &
    cells["_row"].notna()
    &
    cells["_column"].notna()
].copy()

cells["_cell_text"] = (
    cells["raw_cell"]
    .apply(clean)
)


# ======================================================================
# CREATE HEADER MAP
# ======================================================================

header_map = {}

for _, row in headers.iterrows():

    key = (
        row["_pmcid"],
        int(row["_table"]),
        int(row["_column"])
    )

    header_map[key] = {
        "header":
            clean(row["header_original"]),

        "role":
            row["header_role_r2"],

        "release_score":
            row["release_score_r2"],

        "time_score":
            row["time_score_r2"]
    }


# ======================================================================
# FIND STRUCTURALLY PAIRABLE TABLES
# ======================================================================

table_groups = {}

for key, meta in header_map.items():

    pmcid, table, column = key

    group_key = (
        pmcid,
        table
    )

    if group_key not in table_groups:

        table_groups[group_key] = []

    table_groups[
        group_key
    ].append(
        (
            column,
            meta
        )
    )


pairable_tables = {}

for key, items in table_groups.items():

    release_cols = []
    time_cols = []

    for column, meta in items:

        role = meta["role"]

        if role in [
            "RELEASE_VALUE_COLUMN",
            "TIME_AND_RELEASE_EVENT"
        ]:

            release_cols.append(
                (
                    column,
                    meta
                )
            )

        if role in [
            "TIME_SAMPLING_COLUMN",
            "TIME_AND_RELEASE_EVENT"
        ]:

            time_cols.append(
                (
                    column,
                    meta
                )
            )

    # Need at least one of each.
    if release_cols and time_cols:

        pairable_tables[key] = {
            "release_cols":
                release_cols,

            "time_cols":
                time_cols
        }


print(
    f"Structurally pairable tables: "
    f"{len(pairable_tables):,}"
)


# ======================================================================
# EXTRACT ROW-LEVEL PAIRS
# ======================================================================

observations = []

audit = []

obs_counter = 0

for table_key, structure in pairable_tables.items():

    pmcid, table_index = table_key

    release_cols = structure[
        "release_cols"
    ]

    time_cols = structure[
        "time_cols"
    ]

    table_cells = cells[
        (
            cells["_pmcid"] == pmcid
        )
        &
        (
            cells["_table"] ==
            table_index
        )
    ].copy()

    if table_cells.empty:
        continue

    # --------------------------------------------------------------
    # Map each row / column to cell value.
    # --------------------------------------------------------------

    for row_index, row_cells in table_cells.groupby(
        "_row"
    ):

        cell_map = {}

        for _, cell in row_cells.iterrows():

            col = int(
                cell["_column"]
            )

            cell_map[col] = {
                "text":
                    clean(
                        cell["raw_cell"]
                    ),

                "row_text":
                    clean(
                        cell.get(
                            "row_text",
                            ""
                        )
                    )
            }

        # ----------------------------------------------------------
        # Every release/time column combination is considered,
        # but only when both cells are in the SAME XML ROW.
        # ----------------------------------------------------------

        for release_col, release_meta in release_cols:

            if release_col not in cell_map:
                continue

            release_text = cell_map[
                release_col
            ]["text"]

            release_value = parse_numeric(
                release_text
            )

            if not valid_release(
                release_value
            ):
                continue

            for time_col, time_meta in time_cols:

                if time_col not in cell_map:
                    continue

                # Do not pair a column with itself.
                if release_col == time_col:
                    continue

                time_text = cell_map[
                    time_col
                ]["text"]

                time_value_raw = parse_numeric(
                    time_text
                )

                if time_value_raw is None:
                    continue

                time_header = time_meta[
                    "header"
                ]

                time_hours = time_to_hours(
                    time_value_raw,
                    time_header
                )

                # Unknown time unit = manual review,
                # not an analytical observation.
                if time_hours is None:
                    continue

                # Reject impossible negative time.
                if time_hours < 0:
                    continue

                obs_counter += 1

                observations.append(
                    {
                        "observation_id":
                            f"6D-{obs_counter:07d}",

                        "pmcid":
                            pmcid,

                        "table_index":
                            int(table_index),

                        "row_index":
                            int(row_index),

                        "time_column":
                            int(time_col),

                        "release_column":
                            int(release_col),

                        "time_raw":
                            time_text,

                        "release_raw":
                            release_text,

                        "release_time_h":
                            time_hours,

                        "release_percent":
                            release_value,

                        "time_header":
                            time_header,

                        "release_header":
                            release_meta[
                                "header"
                            ],

                        "row_text":
                            cell_map[
                                release_col
                            ]["row_text"],

                        "pairing_status":
                            "STRICT_SAME_ROW",

                        "source_trace":
                            (
                                f"{pmcid}:"
                                f"table_{table_index}:"
                                f"row_{row_index}:"
                                f"time_col_{time_col}:"
                                f"release_col_{release_col}"
                            )
                    }
                )


# ======================================================================
# DATAFRAME
# ======================================================================

obs_df = pd.DataFrame(
    observations
)

if obs_df.empty:

    print(
        "\nNo strict structural observations were produced."
    )

else:

    # Exact duplicate protection.
    before = len(
        obs_df
    )

    obs_df = obs_df.drop_duplicates(
        subset=[
            "pmcid",
            "table_index",
            "row_index",
            "time_column",
            "release_column",
            "release_time_h",
            "release_percent"
        ]
    ).copy()

    duplicates_removed = (
        before - len(obs_df)
    )

    # --------------------------------------------------------------
    # Basic audit
    # --------------------------------------------------------------

    for (
        pmcid,
        table_index
    ), group in obs_df.groupby(
        [
            "pmcid",
            "table_index"
        ]
    ):

        audit.append(
            {
                "pmcid":
                    pmcid,

                "table_index":
                    table_index,

                "candidate_observations":
                    len(group),

                "unique_rows":
                    group[
                        "row_index"
                    ].nunique(),

                "unique_time_columns":
                    group[
                        "time_column"
                    ].nunique(),

                "unique_release_columns":
                    group[
                        "release_column"
                    ].nunique()
            }
        )

# ======================================================================
# SAVE
# ======================================================================

obs_df.to_csv(
    OUT_FILE,
    index=False
)

pd.DataFrame(
    audit
).to_csv(
    AUDIT_FILE,
    index=False
)


# ======================================================================
# REPORT
# ======================================================================

print(
    "\n" + "=" * 70
)

print(
    "CODE 6D COMPLETE"
)

print(
    "=" * 70
)

print(
    f"Structurally pairable tables: "
    f"{len(pairable_tables):,}"
)

print(
    f"Strict observations: "
    f"{len(obs_df):,}"
)

print(
    f"Exact duplicate extractions removed: "
    f"{duplicates_removed if not obs_df.empty else 0}"
)

if not obs_df.empty:

    print(
        f"Unique PMCIDs: "
        f"{obs_df['pmcid'].nunique():,}"
    )

    print(
        f"Unique source tables: "
        f"{obs_df[['pmcid','table_index']].drop_duplicates().shape[0]:,}"
    )

    print(
        f"Unique source rows: "
        f"{obs_df[['pmcid','table_index','row_index']].drop_duplicates().shape[0]:,}"
    )

    print(
        "\nRelease statistics:"
    )

    print(
        obs_df[
            "release_percent"
        ].describe()
    )

    print(
        "\nTime statistics (hours):"
    )

    print(
        obs_df[
            "release_time_h"
        ].describe()
    )

    print(
        "\nPreview:"
    )

    print(
        obs_df[
            [
                "observation_id",
                "pmcid",
                "table_index",
                "row_index",
                "time_raw",
                "release_raw",
                "release_time_h",
                "release_percent",
                "time_header",
                "release_header"
            ]
        ]
        .head(30)
        .to_string(
            index=False
        )
    )

print(
    "\nSaved:"
)

print(
    OUT_FILE
)

print(
    AUDIT_FILE
)

print(
    "\nIMPORTANT:"
)

print(
    "These are CANDIDATE observations only."
)

print(
    "They have NOT passed semantic validation."
)

print(
    "They have NOT passed source-level validation."
)

print(
    "Only same-row XML structural pairings were used."
)

print(
    "No synthetic observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No intentional duplication was performed."
)

print(
    "No machine learning was performed."
)

print(
    "=" * 70
)
```

```python
# ======================================================================
# CODE 6E — SEMANTIC VALIDATION OF STRICT TIME–RELEASE PAIRS
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6E — SEMANTIC VALIDATION OF STRICT TIME–RELEASE PAIRS")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT,
    "data",
    "candidates",
    "06D_strict_xml_time_release_candidates.csv"
)

OUT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "06E_semantic_time_release_validation.csv"
)

PASS_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "06E_pass_candidates.csv"
)

REVIEW_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "06E_manual_review.csv"
)

REJECT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "06E_rejected.csv"
)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"Missing Code 6D output:\n{INPUT_FILE}"
    )

os.makedirs(
    os.path.dirname(OUT_FILE),
    exist_ok=True
)

df = pd.read_csv(
    INPUT_FILE,
    low_memory=False
)

print(
    f"Input observations: {len(df):,}"
)

print(
    f"Columns: {list(df.columns)}"
)


# ======================================================================
# TEXT NORMALIZATION
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
        .replace("\xa0", " ")
        .replace("\n", " ")
        .replace("\t", " ")
    ).strip()


for col in [
    "time_header",
    "release_header",
    "row_text",
    "time_raw",
    "release_raw"
]:

    if col in df.columns:
        df[col + "_clean"] = (
            df[col]
            .apply(clean)
        )


# ======================================================================
# SEMANTIC TERM SETS
# ======================================================================

RELEASE_POSITIVE = [
    "drug release",
    "drug released",
    "drug release (%)",
    "cumulative drug release",
    "cumulative release",
    "percent release",
    "percentage of drug release",
    "drug release amount",
    "amount release",
    "release amount",
    "release (%)",
    "released (%)",
    "total drug release",
    "total drug released",
    "drug release degree",
    "release efficiency"
]

NON_RELEASE = [
    "cell viability",
    "viability",
    "cytotoxicity",
    "inhibition",
    "inhibitory",
    "auc",
    "area under curve",
    "cmax",
    "half-life",
    "half life",
    "entrapment efficiency",
    "encapsulation efficiency",
    "drug loading",
    "loading efficiency",
    "loading capacity",
    "particle size",
    "diameter",
    "zeta potential",
    "contact angle",
    "degradation",
    "swelling",
    "permeability",
    "penetration",
    "in vitro activity",
    "in vivo activity"
]

TIME_EVENT = [
    "maximum drug release",
    "max drug release",
    "time for maximum",
    "time to maximum",
    "time of maximum",
    "maximum % drug release"
]


# ======================================================================
# CLASSIFICATION FUNCTIONS
# ======================================================================

def contains_any(
    text,
    terms
):

    text = text.lower()

    return any(
        term in text
        for term in terms
    )


def classify_release_header(
    header
):

    h = header.lower()

    # Explicitly reject obvious non-release outcomes.
    if contains_any(
        h,
        NON_RELEASE
    ):
        return "NON_RELEASE"

    # Event-type maximum release.
    if contains_any(
        h,
        TIME_EVENT
    ):
        return "MAX_RELEASE_EVENT"

    # Strong ordinary release header.
    if contains_any(
        h,
        RELEASE_POSITIVE
    ):
        return "TIMECOURSE_RELEASE"

    return "REVIEW"


def validate_row_semantics(
    row
):

    release_header = row[
        "release_header_clean"
    ]

    time_header = row[
        "time_header_clean"
    ]

    row_text = row[
        "row_text_clean"
    ]

    release_raw = row[
        "release_raw_clean"
    ]

    # --------------------------------------------------------------
    # Header classification
    # --------------------------------------------------------------

    release_class = classify_release_header(
        release_header
    )

    # --------------------------------------------------------------
    # Release numeric validation
    # --------------------------------------------------------------

    value = pd.to_numeric(
        row["release_percent"],
        errors="coerce"
    )

    time = pd.to_numeric(
        row["release_time_h"],
        errors="coerce"
    )

    if pd.isna(value):

        return (
            "REJECT",
            "NON_NUMERIC_RELEASE",
            release_class
        )

    if pd.isna(time):

        return (
            "REJECT",
            "NON_NUMERIC_TIME",
            release_class
        )

    if value < 0 or value > 100:

        return (
            "REJECT",
            "RELEASE_OUTSIDE_0_100",
            release_class
        )

    if time < 0:

        return (
            "REJECT",
            "NEGATIVE_TIME",
            release_class
        )

    # --------------------------------------------------------------
    # Header semantics
    # --------------------------------------------------------------

    if release_class == "NON_RELEASE":

        return (
            "REJECT",
            "NON_RELEASE_HEADER",
            release_class
        )

    if release_class == "REVIEW":

        return (
            "MANUAL_REVIEW",
            "AMBIGUOUS_RELEASE_HEADER",
            release_class
        )

    # --------------------------------------------------------------
    # Row-level false-positive protection
    # --------------------------------------------------------------

    row_lower = row_text.lower()

    strong_nonrelease = [
        "cell viability",
        "entrapment efficiency",
        "drug loading",
        "loading efficiency",
        "half-life",
        "half life",
        "auc",
        "cmax",
        "particle size",
        "zeta potential",
        "contact angle",
        "inhibition zone"
    ]

    # If row text is clearly describing a different outcome,
    # reject even if the header scoring was permissive.
    if any(
        term in row_lower
        for term in strong_nonrelease
    ):

        # Allow only if release wording is also explicit.
        if not any(
            term in row_lower
            for term in [
                "drug release",
                "drug released",
                "cumulative release"
            ]
        ):

            return (
                "REJECT",
                "ROW_NON_RELEASE_OUTCOME",
                release_class
            )

    # --------------------------------------------------------------
    # Time header validation
    # --------------------------------------------------------------

    time_lower = time_header.lower()

    if not (
        "time" in time_lower
        or "day" in time_lower
        or "hour" in time_lower
        or "hr" in time_lower
        or "min" in time_lower
        or "week" in time_lower
    ):

        return (
            "MANUAL_REVIEW",
            "AMBIGUOUS_TIME_HEADER",
            release_class
        )

    # --------------------------------------------------------------
    # Zero-release values
    # --------------------------------------------------------------

    # Zero is not automatically rejected.
    # It can be a legitimate t=0 observation.
    if value == 0:

        if time == 0:

            return (
                "PASS_CANDIDATE",
                "VALID_T0_RELEASE",
                release_class
            )

        return (
            "MANUAL_REVIEW",
            "ZERO_RELEASE_NONZERO_TIME",
            release_class
        )

    # --------------------------------------------------------------
    # Everything structurally and semantically consistent.
    # --------------------------------------------------------------

    return (
        "PASS_CANDIDATE",
        "SEMANTICALLY_VALID",
        release_class
    )


# ======================================================================
# APPLY VALIDATION
# ======================================================================

results = df.apply(
    validate_row_semantics,
    axis=1,
    result_type="expand"
)

results.columns = [
    "validation_decision",
    "validation_reason",
    "observation_type"
]

df = pd.concat(
    [
        df.reset_index(drop=True),
        results.reset_index(drop=True)
    ],
    axis=1
)


# ======================================================================
# SOURCE TRACE
# ======================================================================

df["source_trace"] = (
    df["pmcid"].astype(str)
    + ":table_"
    + df["table_index"].astype(str)
    + ":row_"
    + df["row_index"].astype(str)
    + ":time_col_"
    + df["time_column"].astype(str)
    + ":release_col_"
    + df["release_column"].astype(str)
)


# ======================================================================
# EXACT DUPLICATE AUDIT
# ======================================================================

duplicate_key = [
    "pmcid",
    "table_index",
    "row_index",
    "time_column",
    "release_column",
    "release_time_h",
    "release_percent"
]

df["exact_duplicate"] = (
    df.duplicated(
        subset=duplicate_key,
        keep=False
    )
)


# ======================================================================
# SAVE DATASETS
# ======================================================================

df.to_csv(
    OUT_FILE,
    index=False
)

passes = df[
    df["validation_decision"]
    == "PASS_CANDIDATE"
].copy()

manual = df[
    df["validation_decision"]
    == "MANUAL_REVIEW"
].copy()

rejected = df[
    df["validation_decision"]
    == "REJECT"
].copy()

passes.to_csv(
    PASS_FILE,
    index=False
)

manual.to_csv(
    REVIEW_FILE,
    index=False
)

rejected.to_csv(
    REJECT_FILE,
    index=False
)


# ======================================================================
# REPORT
# ======================================================================

print(
    "\n" + "=" * 70
)

print(
    "CODE 6E COMPLETE"
)

print(
    "=" * 70
)

print(
    f"Input observations: "
    f"{len(df):,}"
)

print(
    f"PASS candidates: "
    f"{len(passes):,}"
)

print(
    f"MANUAL REVIEW: "
    f"{len(manual):,}"
)

print(
    f"REJECTED: "
    f"{len(rejected):,}"
)


print(
    "\nObservation type:"
)

print(
    df[
        "observation_type"
    ].value_counts()
)


print(
    "\nValidation reasons:"
)

print(
    df[
        "validation_reason"
    ].value_counts()
)


print(
    "\nSource diversity among PASS candidates:"
)

print(
    f"Unique PMCID: "
    f"{passes['pmcid'].nunique():,}"
)

print(
    f"Unique source tables: "
    f"{passes[['pmcid','table_index']].drop_duplicates().shape[0]:,}"
)

print(
    f"Unique source rows: "
    f"{passes[['pmcid','table_index','row_index']].drop_duplicates().shape[0]:,}"
)


print(
    "\nPASS candidate release statistics:"
)

if not passes.empty:

    print(
        passes[
            "release_percent"
        ].describe()
    )

print(
    "\nPASS candidate time statistics:"
)

if not passes.empty:

    print(
        passes[
            "release_time_h"
        ].describe()
    )


print(
    "\nSaved:"
)

print(
    OUT_FILE
)

print(
    PASS_FILE
)

print(
    REVIEW_FILE
)

print(
    REJECT_FILE
)


print(
    "\nIMPORTANT:"
)

print(
    "PASS candidates are not yet the final analytical dataset."
)

print(
    "Maximum-release events are retained as a separate observation type."
)

print(
    "No interpolation was performed."
)

print(
    "No synthetic observations were generated."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print(
    "=" * 70
)
```

```python
# ======================================================================
# CODE 6F — SOURCE-LEVEL VALIDATION OF STRICT TIME–RELEASE OBSERVATIONS
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6F — SOURCE-LEVEL VALIDATION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "06E_pass_candidates.csv"
)

CODE3_FILE = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "06F_source_validation_audit.csv"
)

VALID_FILE = os.path.join(
    OUT_DIR,
    "06F_valid_source_release.csv"
)

REVIEW_FILE = os.path.join(
    OUT_DIR,
    "06F_manual_source_review.csv"
)

REJECT_FILE = os.path.join(
    OUT_DIR,
    "06F_source_rejected.csv"
)

os.makedirs(
    OUT_DIR,
    exist_ok=True
)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"Missing Code 6E file:\n{INPUT_FILE}"
    )

if not os.path.exists(CODE3_FILE):
    raise FileNotFoundError(
        f"Missing Code 3 file:\n{CODE3_FILE}"
    )


# ======================================================================
# LOAD
# ======================================================================

obs = pd.read_csv(
    INPUT_FILE,
    low_memory=False
)

cells = pd.read_csv(
    CODE3_FILE,
    low_memory=False
)

print(
    f"Code 6E observations: {len(obs):,}"
)

print(
    f"Code 3 cells: {len(cells):,}"
)

print(
    f"Code 3 columns: {list(cells.columns)}"
)


# ======================================================================
# NORMALIZATION
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
        .replace("\xa0", " ")
        .replace("\n", " ")
        .replace("\t", " ")
    ).strip()


obs["_pmcid"] = (
    obs["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

cells["_pmcid"] = (
    cells["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

obs["_table"] = pd.to_numeric(
    obs["table_index"],
    errors="coerce"
)

obs["_row"] = pd.to_numeric(
    obs["row_index"],
    errors="coerce"
)

obs["_time_col"] = pd.to_numeric(
    obs["time_column"],
    errors="coerce"
)

obs["_release_col"] = pd.to_numeric(
    obs["release_column"],
    errors="coerce"
)

cells["_table"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

cells["_row"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

cells["_column"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)

cells["_cell"] = (
    cells["raw_cell"]
    .apply(clean)
)


# ======================================================================
# NUMERIC NORMALIZATION
# ======================================================================

def first_numeric(text):

    s = clean(text)

    if not s:
        return None

    s = (
        s.replace("−", "-")
         .replace("–", "-")
         .replace("—", "-")
    )

    # Remove common uncertainty after primary value.
    s = re.split(
        r"\s*(?:±|\+/-)\s*",
        s
    )[0]

    # Remove percentage sign.
    s = s.replace("%", "")

    # Remove leading inequality.
    s = re.sub(
        r"^[<>]\s*",
        "",
        s
    )

    match = re.search(
        r"[-+]?\d+(?:\.\d+)?",
        s
    )

    if not match:
        return None

    try:
        return float(
            match.group()
        )
    except Exception:
        return None


# ======================================================================
# SOURCE MATCHING
# ======================================================================

valid_records = []
review_records = []
reject_records = []

audit_records = []

for idx, row in obs.iterrows():

    pmcid = row["_pmcid"]
    table = row["_table"]
    source_row = row["_row"]
    time_col = row["_time_col"]
    release_col = row["_release_col"]

    expected_time = pd.to_numeric(
        row["release_time_h"],
        errors="coerce"
    )

    expected_release = pd.to_numeric(
        row["release_percent"],
        errors="coerce"
    )

    # --------------------------------------------------------------
    # Find exact source table.
    # --------------------------------------------------------------

    table_cells = cells[
        (cells["_pmcid"] == pmcid)
        &
        (cells["_table"] == table)
    ].copy()

    if table_cells.empty:

        record = row.to_dict()

        record.update({
            "source_validation":
                "SOURCE_REJECTED",

            "source_reason":
                "SOURCE_TABLE_NOT_FOUND",

            "matched_time_cell":
                "",

            "matched_release_cell":
                "",

            "source_row_text":
                ""
        })

        reject_records.append(
            record
        )

        audit_records.append(
            record
        )

        continue


    # --------------------------------------------------------------
    # Find exact XML row.
    # --------------------------------------------------------------

    row_cells = table_cells[
        table_cells["_row"] == source_row
    ].copy()

    if row_cells.empty:

        record = row.to_dict()

        record.update({
            "source_validation":
                "MANUAL_SOURCE_REVIEW",

            "source_reason":
                "EXACT_SOURCE_ROW_NOT_FOUND",

            "matched_time_cell":
                "",

            "matched_release_cell":
                "",

            "source_row_text":
                ""
        })

        review_records.append(
            record
        )

        audit_records.append(
            record
        )

        continue


    # --------------------------------------------------------------
    # Exact time column.
    # --------------------------------------------------------------

    time_match = row_cells[
        row_cells["_column"] == time_col
    ]

    release_match = row_cells[
        row_cells["_column"] == release_col
    ]


    if time_match.empty:

        record = row.to_dict()

        record.update({
            "source_validation":
                "MANUAL_SOURCE_REVIEW",

            "source_reason":
                "TIME_COLUMN_NOT_FOUND_IN_EXACT_ROW",

            "matched_time_cell":
                "",

            "matched_release_cell":
                "",

            "source_row_text":
                clean(
                    row_cells.iloc[0].get(
                        "row_text",
                        ""
                    )
                )
        })

        review_records.append(
            record
        )

        audit_records.append(
            record
        )

        continue


    if release_match.empty:

        record = row.to_dict()

        record.update({
            "source_validation":
                "MANUAL_SOURCE_REVIEW",

            "source_reason":
                "RELEASE_COLUMN_NOT_FOUND_IN_EXACT_ROW",

            "matched_time_cell":
                clean(
                    time_match.iloc[0]["_cell"]
                ),

            "matched_release_cell":
                "",

            "source_row_text":
                clean(
                    row_cells.iloc[0].get(
                        "row_text",
                        ""
                    )
                )
        })

        review_records.append(
            record
        )

        audit_records.append(
            record
        )

        continue


    # --------------------------------------------------------------
    # Extract exact source values.
    # --------------------------------------------------------------

    source_time_text = clean(
        time_match.iloc[0]["_cell"]
    )

    source_release_text = clean(
        release_match.iloc[0]["_cell"]
    )

    source_time_value = first_numeric(
        source_time_text
    )

    source_release_value = first_numeric(
        source_release_text
    )

    source_row_text = clean(
        row_cells.iloc[0].get(
            "row_text",
            ""
        )
    )


    # --------------------------------------------------------------
    # Check exact numeric agreement.
    # --------------------------------------------------------------

    time_ok = False
    release_ok = False

    if (
        source_time_value is not None
        and not pd.isna(expected_time)
    ):

        # The expected time is already converted to hours.
        # Convert source cell according to the original
        # time header when possible.

        header = clean(
            row["time_header"]
        ).lower()

        if (
            "min" in header
            and "time" in header
        ):

            source_time_hours = (
                source_time_value / 60.0
            )

        elif (
            "day" in header
        ):

            source_time_hours = (
                source_time_value * 24.0
            )

        elif (
            "week" in header
        ):

            source_time_hours = (
                source_time_value * 168.0
            )

        else:

            source_time_hours = (
                source_time_value
            )

        time_ok = np.isclose(
            source_time_hours,
            expected_time,
            atol=1e-6,
            rtol=1e-6
        )

    if (
        source_release_value is not None
        and not pd.isna(expected_release)
    ):

        release_ok = np.isclose(
            source_release_value,
            expected_release,
            atol=1e-6,
            rtol=1e-6
        )


    # --------------------------------------------------------------
    # Final source decision.
    # --------------------------------------------------------------

    record = row.to_dict()

    record.update({

        "source_validation":
            None,

        "source_reason":
            None,

        "matched_time_cell":
            source_time_text,

        "matched_release_cell":
            source_release_text,

        "source_time_numeric":
            source_time_value,

        "source_release_numeric":
            source_release_value,

        "source_time_matches":
            time_ok,

        "source_release_matches":
            release_ok,

        "source_row_text":
            source_row_text
    })


    if time_ok and release_ok:

        record[
            "source_validation"
        ] = "VALID_SOURCE_RELEASE"

        record[
            "source_reason"
        ] = "EXACT_XML_CELL_MATCH"

        valid_records.append(
            record
        )

    elif (
        source_time_value is not None
        and source_release_value is not None
    ):

        record[
            "source_validation"
        ] = "MANUAL_SOURCE_REVIEW"

        record[
            "source_reason"
        ] = "SOURCE_VALUE_MISMATCH"

        review_records.append(
            record
        )

    else:

        record[
            "source_validation"
        ] = "SOURCE_REJECTED"

        record[
            "source_reason"
        ] = "NON_NUMERIC_SOURCE_CELL"

        reject_records.append(
            record
        )

    audit_records.append(
        record
    )


# ======================================================================
# DATAFRAMES
# ======================================================================

audit = pd.DataFrame(
    audit_records
)

valid = pd.DataFrame(
    valid_records
)

review = pd.DataFrame(
    review_records
)

rejected = pd.DataFrame(
    reject_records
)


# ======================================================================
# EXACT SOURCE DUPLICATE AUDIT
# ======================================================================

duplicate_columns = [
    "pmcid",
    "table_index",
    "row_index",
    "time_column",
    "release_column",
    "release_time_h",
    "release_percent"
]

if not valid.empty:

    valid["source_duplicate"] = (
        valid.duplicated(
            subset=duplicate_columns,
            keep=False
        )
    )

else:

    valid["source_duplicate"] = []


# ======================================================================
# SAVE
# ======================================================================

audit.to_csv(
    AUDIT_FILE,
    index=False
)

valid.to_csv(
    VALID_FILE,
    index=False
)

review.to_csv(
    REVIEW_FILE,
    index=False
)

rejected.to_csv(
    REJECT_FILE,
    index=False
)


# ======================================================================
# REPORT
# ======================================================================

print(
    "\n" + "=" * 70
)

print(
    "CODE 6F COMPLETE"
)

print(
    "=" * 70
)

print(
    f"Input observations: "
    f"{len(obs):,}"
)

print(
    f"VALID_SOURCE_RELEASE: "
    f"{len(valid):,}"
)

print(
    f"MANUAL_SOURCE_REVIEW: "
    f"{len(review):,}"
)

print(
    f"SOURCE_REJECTED: "
    f"{len(rejected):,}"
)


print(
    "\nSource-validation reasons:"
)

if not audit.empty:

    print(
        audit[
            "source_reason"
        ].value_counts()
    )


print(
    "\nSource diversity among VALID_SOURCE_RELEASE:"
)

if not valid.empty:

    print(
        f"Unique PMCID: "
        f"{valid['pmcid'].nunique():,}"
    )

    print(
        f"Unique source tables: "
        f"{valid[['pmcid','table_index']]"
        ".drop_duplicates().shape[0]:,}"
    )

    print(
        f"Unique source rows: "
        f"{valid[['pmcid','table_index','row_index']]"
        ".drop_duplicates().shape[0]:,}"
    )

    print(
        f"Exact source duplicates: "
        f"{valid['source_duplicate'].sum():,}"
    )


print(
    "\nObservation types:"
)

if not valid.empty:

    print(
        valid[
            "observation_type"
        ].value_counts()
    )


print(
    "\nSaved:"
)

print(
    AUDIT_FILE
)

print(
    VALID_FILE
)

print(
    REVIEW_FILE
)

print(
    REJECT_FILE
)


print(
    "\nIMPORTANT:"
)

print(
    "VALID_SOURCE_RELEASE observations are traceable "
    "to exact Code 3 XML cells."
)

print(
    "Manual-review observations are NOT included "
    "in the validated dataset."
)

print(
    "No synthetic observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print(
    "=" * 70
)
```

```python
# ======================================================================
# CODE 6F — SOURCE-LEVEL VALIDATION OF STRICT TIME–RELEASE OBSERVATIONS
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6F — SOURCE-LEVEL VALIDATION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT, "data", "validation",
    "06E_pass_candidates.csv"
)

CODE3_FILE = os.path.join(
    ROOT, "data", "raw", "tables",
    "03_extracted_table_cells.csv"
)

OUT_DIR = os.path.join(
    ROOT, "data", "validation"
)

AUDIT_FILE = os.path.join(
    OUT_DIR, "06F_source_validation_audit.csv"
)

VALID_FILE = os.path.join(
    OUT_DIR, "06F_valid_source_release.csv"
)

REVIEW_FILE = os.path.join(
    OUT_DIR, "06F_manual_source_review.csv"
)

REJECT_FILE = os.path.join(
    OUT_DIR, "06F_source_rejected.csv"
)

os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"Missing Code 6E file:\n{INPUT_FILE}"
    )

if not os.path.exists(CODE3_FILE):
    raise FileNotFoundError(
        f"Missing Code 3 file:\n{CODE3_FILE}"
    )


# ======================================================================
# LOAD
# ======================================================================

obs = pd.read_csv(
    INPUT_FILE,
    low_memory=False
)

cells = pd.read_csv(
    CODE3_FILE,
    low_memory=False
)

print(f"Code 6E observations: {len(obs):,}")
print(f"Code 3 cells: {len(cells):,}")


# ======================================================================
# NORMALIZATION
# ======================================================================

def clean(x):
    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
        .replace("\xa0", " ")
        .replace("\n", " ")
        .replace("\t", " ")
    ).strip()


obs["_pmcid"] = (
    obs["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

cells["_pmcid"] = (
    cells["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

obs["_table"] = pd.to_numeric(
    obs["table_index"],
    errors="coerce"
)

obs["_row"] = pd.to_numeric(
    obs["row_index"],
    errors="coerce"
)

obs["_time_col"] = pd.to_numeric(
    obs["time_column"],
    errors="coerce"
)

obs["_release_col"] = pd.to_numeric(
    obs["release_column"],
    errors="coerce"
)

cells["_table"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

cells["_row"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

cells["_column"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)

cells["_cell"] = (
    cells["raw_cell"]
    .apply(clean)
)


# ======================================================================
# NUMERIC EXTRACTION
# ======================================================================

def first_numeric(text):

    s = clean(text)

    if not s:
        return None

    s = (
        s.replace("−", "-")
         .replace("–", "-")
         .replace("—", "-")
    )

    # Keep the primary value when cell contains ± SD
    s = re.split(
        r"\s*(?:±|\+/-)\s*",
        s
    )[0]

    s = s.replace("%", "")

    s = re.sub(
        r"^[<>]\s*",
        "",
        s
    )

    match = re.search(
        r"[-+]?\d+(?:\.\d+)?",
        s
    )

    if not match:
        return None

    try:
        return float(match.group())
    except Exception:
        return None


# ======================================================================
# SOURCE VALIDATION
# ======================================================================

valid_records = []
review_records = []
reject_records = []
audit_records = []

total = len(obs)

for idx, row in obs.iterrows():

    pmcid = row["_pmcid"]
    table = row["_table"]
    source_row = row["_row"]
    time_col = row["_time_col"]
    release_col = row["_release_col"]

    expected_time = pd.to_numeric(
        row["release_time_h"],
        errors="coerce"
    )

    expected_release = pd.to_numeric(
        row["release_percent"],
        errors="coerce"
    )

    # --------------------------------------------------------------
    # Exact source table
    # --------------------------------------------------------------

    table_cells = cells[
        (cells["_pmcid"] == pmcid)
        &
        (cells["_table"] == table)
    ].copy()

    if table_cells.empty:

        record = row.to_dict()

        record.update({
            "source_validation":
                "SOURCE_REJECTED",

            "source_reason":
                "SOURCE_TABLE_NOT_FOUND",

            "matched_time_cell":
                "",

            "matched_release_cell":
                "",

            "source_time_numeric":
                np.nan,

            "source_release_numeric":
                np.nan,

            "source_time_matches":
                False,

            "source_release_matches":
                False,

            "source_row_text":
                ""
        })

        reject_records.append(record)
        audit_records.append(record)

        continue


    # --------------------------------------------------------------
    # Exact source row
    # --------------------------------------------------------------

    row_cells = table_cells[
        table_cells["_row"] == source_row
    ].copy()

    if row_cells.empty:

        record = row.to_dict()

        record.update({
            "source_validation":
                "MANUAL_SOURCE_REVIEW",

            "source_reason":
                "EXACT_SOURCE_ROW_NOT_FOUND",

            "matched_time_cell":
                "",

            "matched_release_cell":
                "",

            "source_time_numeric":
                np.nan,

            "source_release_numeric":
                np.nan,

            "source_time_matches":
                False,

            "source_release_matches":
                False,

            "source_row_text":
                ""
        })

        review_records.append(record)
        audit_records.append(record)

        continue


    # --------------------------------------------------------------
    # Exact time column
    # --------------------------------------------------------------

    time_match = row_cells[
        row_cells["_column"] == time_col
    ]

    release_match = row_cells[
        row_cells["_column"] == release_col
    ]


    if time_match.empty:

        record = row.to_dict()

        record.update({
            "source_validation":
                "MANUAL_SOURCE_REVIEW",

            "source_reason":
                "TIME_COLUMN_NOT_FOUND_IN_EXACT_ROW",

            "matched_time_cell":
                "",

            "matched_release_cell":
                "",

            "source_time_numeric":
                np.nan,

            "source_release_numeric":
                np.nan,

            "source_time_matches":
                False,

            "source_release_matches":
                False,

            "source_row_text":
                clean(
                    row_cells.iloc[0].get(
                        "row_text",
                        ""
                    )
                )
        })

        review_records.append(record)
        audit_records.append(record)

        continue


    if release_match.empty:

        record = row.to_dict()

        record.update({
            "source_validation":
                "MANUAL_SOURCE_REVIEW",

            "source_reason":
                "RELEASE_COLUMN_NOT_FOUND_IN_EXACT_ROW",

            "matched_time_cell":
                clean(
                    time_match.iloc[0]["_cell"]
                ),

            "matched_release_cell":
                "",

            "source_time_numeric":
                np.nan,

            "source_release_numeric":
                np.nan,

            "source_time_matches":
                False,

            "source_release_matches":
                False,

            "source_row_text":
                clean(
                    row_cells.iloc[0].get(
                        "row_text",
                        ""
                    )
                )
        })

        review_records.append(record)
        audit_records.append(record)

        continue


    # --------------------------------------------------------------
    # Exact XML cell values
    # --------------------------------------------------------------

    source_time_text = clean(
        time_match.iloc[0]["_cell"]
    )

    source_release_text = clean(
        release_match.iloc[0]["_cell"]
    )

    source_time_value = first_numeric(
        source_time_text
    )

    source_release_value = first_numeric(
        source_release_text
    )

    source_row_text = clean(
        row_cells.iloc[0].get(
            "row_text",
            ""
        )
    )


    # --------------------------------------------------------------
    # Time conversion
    # --------------------------------------------------------------

    time_ok = False
    release_ok = False

    if (
        source_time_value is not None
        and not pd.isna(expected_time)
    ):

        header = clean(
            row["time_header"]
        ).lower()

        if "min" in header:
            source_time_hours = (
                source_time_value / 60.0
            )

        elif "day" in header:
            source_time_hours = (
                source_time_value * 24.0
            )

        elif "week" in header:
            source_time_hours = (
                source_time_value * 168.0
            )

        else:
            source_time_hours = (
                source_time_value
            )

        time_ok = np.isclose(
            source_time_hours,
            expected_time,
            atol=1e-6,
            rtol=1e-6
        )


    # --------------------------------------------------------------
    # Release comparison
    # --------------------------------------------------------------

    if (
        source_release_value is not None
        and not pd.isna(expected_release)
    ):

        release_ok = np.isclose(
            source_release_value,
            expected_release,
            atol=1e-6,
            rtol=1e-6
        )


    # --------------------------------------------------------------
    # Build record
    # --------------------------------------------------------------

    record = row.to_dict()

    record.update({
        "matched_time_cell":
            source_time_text,

        "matched_release_cell":
            source_release_text,

        "source_time_numeric":
            source_time_value,

        "source_release_numeric":
            source_release_value,

        "source_time_matches":
            time_ok,

        "source_release_matches":
            release_ok,

        "source_row_text":
            source_row_text
    })


    # --------------------------------------------------------------
    # Decision
    # --------------------------------------------------------------

    if time_ok and release_ok:

        record["source_validation"] = (
            "VALID_SOURCE_RELEASE"
        )

        record["source_reason"] = (
            "EXACT_XML_CELL_MATCH"
        )

        valid_records.append(record)

    elif (
        source_time_value is not None
        and source_release_value is not None
    ):

        record["source_validation"] = (
            "MANUAL_SOURCE_REVIEW"
        )

        record["source_reason"] = (
            "SOURCE_VALUE_MISMATCH"
        )

        review_records.append(record)

    else:

        record["source_validation"] = (
            "SOURCE_REJECTED"
        )

        record["source_reason"] = (
            "NON_NUMERIC_SOURCE_CELL"
        )

        reject_records.append(record)

    audit_records.append(record)

    if (idx + 1) % 25 == 0:
        print(
            f"Validated {idx + 1}/{total}"
        )


# ======================================================================
# DATAFRAMES
# ======================================================================

audit = pd.DataFrame(audit_records)
valid = pd.DataFrame(valid_records)
review = pd.DataFrame(review_records)
rejected = pd.DataFrame(reject_records)


# ======================================================================
# DUPLICATE AUDIT
# ======================================================================

duplicate_columns = [
    "pmcid",
    "table_index",
    "row_index",
    "time_column",
    "release_column",
    "release_time_h",
    "release_percent"
]

if not valid.empty:

    valid["source_duplicate"] = (
        valid.duplicated(
            subset=duplicate_columns,
            keep=False
        )
    )

else:

    valid["source_duplicate"] = pd.Series(
        dtype=bool
    )


# ======================================================================
# SAVE
# ======================================================================

audit.to_csv(
    AUDIT_FILE,
    index=False
)

valid.to_csv(
    VALID_FILE,
    index=False
)

review.to_csv(
    REVIEW_FILE,
    index=False
)

rejected.to_csv(
    REJECT_FILE,
    index=False
)


# ======================================================================
# FINAL REPORT
# ======================================================================

print("\n" + "=" * 70)
print("CODE 6F COMPLETE")
print("=" * 70)

print(
    f"Input observations:       {len(obs):,}"
)

print(
    f"VALID_SOURCE_RELEASE:     {len(valid):,}"
)

print(
    f"MANUAL_SOURCE_REVIEW:     {len(review):,}"
)

print(
    f"SOURCE_REJECTED:          {len(rejected):,}"
)


print("\nSource-validation reasons:")

if not audit.empty:
    print(
        audit[
            "source_reason"
        ].value_counts()
    )


print("\nSource diversity among VALID_SOURCE_RELEASE:")

if not valid.empty:

    unique_pmcids = (
        valid["pmcid"]
        .nunique()
    )

    unique_tables = (
        valid[
            ["pmcid", "table_index"]
        ]
        .drop_duplicates()
        .shape[0]
    )

    unique_rows = (
        valid[
            ["pmcid", "table_index", "row_index"]
        ]
        .drop_duplicates()
        .shape[0]
    )

    duplicate_count = int(
        valid["source_duplicate"].sum()
    )

    print(
        f"Unique PMCID:             {unique_pmcids:,}"
    )

    print(
        f"Unique source tables:     {unique_tables:,}"
    )

    print(
        f"Unique source rows:       {unique_rows:,}"
    )

    print(
        f"Exact source duplicates:  {duplicate_count:,}"
    )


print("\nObservation types:")

if not valid.empty:
    print(
        valid[
            "observation_type"
        ].value_counts()
    )


print("\nSaved:")
print(AUDIT_FILE)
print(VALID_FILE)
print(REVIEW_FILE)
print(REJECT_FILE)


print("\nIMPORTANT:")
print(
    "VALID_SOURCE_RELEASE observations are traceable "
    "to exact Code 3 XML cells."
)

print(
    "Manual-review observations are NOT included "
    "in the validated dataset."
)

print("No synthetic observations were generated.")
print("No interpolation was performed.")
print("No observations were duplicated.")
print("No machine learning was performed.")

print("=" * 70)
```

```python
# ======================================================================
# CODE 6G — FORMULATION & EXPERIMENTAL-UNIT RESOLUTION
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6G — FORMULATION & EXPERIMENTAL-UNIT RESOLUTION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "06F_valid_source_release.csv"
)

CODE3_FILE = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "06G_formulation_unit_audit.csv"
)

READY_FILE = os.path.join(
    OUT_DIR,
    "06G_formulation_ready_candidates.csv"
)

REVIEW_FILE = os.path.join(
    OUT_DIR,
    "06G_manual_formulation_review.csv"
)

os.makedirs(
    OUT_DIR,
    exist_ok=True
)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"Missing Code 6F output:\n{INPUT_FILE}"
    )

if not os.path.exists(CODE3_FILE):
    raise FileNotFoundError(
        f"Missing Code 3 file:\n{CODE3_FILE}"
    )


# ======================================================================
# LOAD
# ======================================================================

obs = pd.read_csv(
    INPUT_FILE,
    low_memory=False
)

cells = pd.read_csv(
    CODE3_FILE,
    low_memory=False
)

print(
    f"Validated 6F observations: {len(obs):,}"
)

print(
    f"Code 3 cells: {len(cells):,}"
)


# ======================================================================
# NORMALIZATION
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
        .replace("\xa0", " ")
        .replace("\n", " ")
        .replace("\t", " ")
    ).strip()


def norm(x):

    return clean(x).lower()


for df in [obs, cells]:

    df["_pmcid"] = (
        df["pmcid"]
        .astype(str)
        .str.strip()
        .str.upper()
    )

    df["_table"] = pd.to_numeric(
        df["table_index"],
        errors="coerce"
    )

    df["_row"] = pd.to_numeric(
        df["row_index"],
        errors="coerce"
    )


cells["_column"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)

cells["_cell"] = (
    cells["raw_cell"]
    .apply(clean)
)

cells["_row_text"] = (
    cells["row_text"]
    .apply(clean)
)


# ======================================================================
# KEYWORD DICTIONARIES
# ======================================================================

DRUGS = [
    "dexamethasone",
    "curcumin",
    "paclitaxel",
    "doxorubicin",
    "dox",
    "metformin",
    "ibuprofen",
    "gemcitabine",
    "5-fluorouracil",
    "5 fu",
    "rifampicin",
    "vancomycin",
    "insulin",
    "drug"
]

POLYMERS = [
    "plga",
    "pla",
    "pcl",
    "peg",
    "peo",
    "pva",
    "pvp",
    "chitosan",
    "alginate",
    "gelatin",
    "cellulose",
    "poloxamer",
    "polyurethane",
    "hyaluronic acid",
    "hyaluronan"
]

CARRIERS = [
    "nanoparticle",
    "nanoparticles",
    "nanocarrier",
    "nanocarriers",
    "liposome",
    "liposomes",
    "hydrogel",
    "hydrogels",
    "micelle",
    "micelles",
    "microsphere",
    "microspheres",
    "microparticle",
    "microparticles",
    "nanofiber",
    "nanofibers",
    "film",
    "films",
    "implant",
    "implants",
    "bead",
    "beads",
    "scaffold",
    "scaffolds"
]

CONDITION_TERMS = [
    "ph",
    "pbs",
    "buffer",
    "acidic",
    "neutral",
    "physiological",
    "37",
    "25",
    "media",
    "medium",
    "water",
    "saline",
    "sink condition",
    "release medium"
]

FORMULATION_TERMS = [
    "formulation",
    "formula",
    "form",
    "sample",
    "group",
    "batch",
    "composition",
    "polymer",
    "drug",
    "carrier",
    "nanoparticle",
    "hydrogel",
    "microsphere",
    "microparticle",
    "film",
    "implant"
]


# ======================================================================
# EXTRACTION HELPERS
# ======================================================================

def find_terms(text, terms):

    text = norm(text)

    found = []

    for term in terms:

        if term in text:

            found.append(term)

    return found


def unique_join(values):

    out = []

    for value in values:

        if not value:
            continue

        for item in str(value).split(";"):

            item = item.strip()

            if item and item not in out:

                out.append(item)

    return ";".join(out)


def make_context(table_cells, target_row):

    """
    Conservative context:

    - exact release row
    - immediately adjacent rows
    - table header information
    """

    rows = sorted(
        table_cells["_row"]
        .dropna()
        .unique()
    )

    nearby = [
        r for r in rows
        if abs(r - target_row) <= 2
    ]

    context = table_cells[
        table_cells["_row"].isin(
            nearby
        )
    ].copy()

    text_parts = []

    for _, r in context.iterrows():

        text_parts.append(
            clean(r["_cell"])
        )

        if r["_row_text"]:

            text_parts.append(
                clean(r["_row_text"])
            )

    return " | ".join(
        [x for x in text_parts if x]
    )


# ======================================================================
# PROCESS
# ======================================================================

audit_records = []
ready_records = []
review_records = []

for i, row in obs.iterrows():

    pmcid = row["_pmcid"]
    table = row["_table"]
    target_row = row["_row"]

    table_cells = cells[
        (cells["_pmcid"] == pmcid)
        &
        (cells["_table"] == table)
    ].copy()

    if table_cells.empty:

        record = row.to_dict()

        record.update({
            "formulation_resolution":
                "MANUAL_REVIEW",

            "resolution_reason":
                "SOURCE_TABLE_NOT_FOUND",

            "drug":
                "NOT_IDENTIFIED",

            "polymer":
                "NOT_IDENTIFIED",

            "carrier":
                "NOT_IDENTIFIED",

            "condition":
                "NOT_IDENTIFIED",

            "formulation_identifier":
                "NOT_IDENTIFIED",

            "experimental_unit_proxy":
                "NOT_IDENTIFIED",

            "context_text":
                ""
        })

        review_records.append(record)
        audit_records.append(record)

        continue


    # --------------------------------------------------------------
    # Exact row
    # --------------------------------------------------------------

    exact_row = table_cells[
        table_cells["_row"] == target_row
    ].copy()

    # --------------------------------------------------------------
    # Conservative local context
    # --------------------------------------------------------------

    context_text = make_context(
        table_cells,
        target_row
    )

    exact_text = " | ".join(
        exact_row["_cell"]
        .tolist()
    )

    exact_row_text = " | ".join(
        exact_row["_row_text"]
        .tolist()
    )

    combined = " | ".join(
        [
            exact_text,
            exact_row_text,
            context_text
        ]
    )

    # --------------------------------------------------------------
    # Extract metadata
    # --------------------------------------------------------------

    drugs = find_terms(
        combined,
        DRUGS
    )

    polymers = find_terms(
        combined,
        POLYMERS
    )

    carriers = find_terms(
        combined,
        CARRIERS
    )

    conditions = find_terms(
        combined,
        CONDITION_TERMS
    )


    # --------------------------------------------------------------
    # Remove generic "drug" from identified drug list
    # when a specific drug is available.
    # --------------------------------------------------------------

    if len(drugs) > 1 and "drug" in drugs:

        drugs = [
            x for x in drugs
            if x != "drug"
        ]


    # --------------------------------------------------------------
    # Formulation identifier detection
    # --------------------------------------------------------------

    formulation_identifier = ""

    # First preference: explicit labels in exact row.
    identifier_patterns = [
        r"\bF\d+\b",
        r"\bF-\d+\b",
        r"\bF_\d+\b",
        r"\bformulation\s*\d+\b",
        r"\bformulation\s*[A-Za-z]\b",
        r"\bgroup\s*\d+\b",
        r"\bgroup\s*[A-Za-z]\b",
        r"\bsample\s*\d+\b",
        r"\bsample\s*[A-Za-z]\b",
        r"\bbatch\s*\d+\b",
        r"\bS\d+\b"
    ]

    for pattern in identifier_patterns:

        match = re.search(
            pattern,
            exact_text,
            flags=re.I
        )

        if match:

            formulation_identifier = (
                match.group(0)
            )

            break


    # --------------------------------------------------------------
    # If no explicit identifier, use exact row as a proxy only
    # when it contains formulation-specific information.
    # --------------------------------------------------------------

    if not formulation_identifier:

        if (
            drugs
            or polymers
            or carriers
        ):

            formulation_identifier = (
                "ROW_PROXY_"
                + str(int(target_row))
            )


    # --------------------------------------------------------------
    # Experimental unit proxy
    # --------------------------------------------------------------

    if formulation_identifier:

        experimental_unit_proxy = (
            f"{pmcid}_T{int(table)}_"
            f"{formulation_identifier}"
        )

    else:

        experimental_unit_proxy = (
            f"{pmcid}_T{int(table)}_"
            f"ROW{int(target_row)}"
        )


    # --------------------------------------------------------------
    # Metadata quality
    # --------------------------------------------------------------

    identified_fields = sum(
        [
            bool(drugs),
            bool(polymers),
            bool(carriers),
            bool(conditions),
            bool(formulation_identifier)
        ]
    )

    if identified_fields >= 4:

        quality = "HIGH"

    elif identified_fields >= 3:

        quality = "GOOD"

    elif identified_fields >= 1:

        quality = "PARTIAL"

    else:

        quality = "WEAK"


    # --------------------------------------------------------------
    # Conservative resolution decision
    # --------------------------------------------------------------

    # A row is ready for further validation only if it has
    # meaningful formulation information.
    #
    # We do NOT call this "independent".
    #

    if (
        bool(formulation_identifier)
        and
        (
            bool(drugs)
            or bool(polymers)
            or bool(carriers)
        )
    ):

        status = "FORMULATION_CONTEXT_READY"

    else:

        status = "MANUAL_REVIEW"


    record = row.to_dict()

    record.update({

        "formulation_resolution":
            status,

        "resolution_reason":
            (
                "FORMULATION_CONTEXT_FOUND"
                if status ==
                "FORMULATION_CONTEXT_READY"
                else
                "INSUFFICIENT_FORMULATION_CONTEXT"
            ),

        "drug":
            unique_join(drugs)
            if drugs
            else "NOT_IDENTIFIED",

        "polymer":
            unique_join(polymers)
            if polymers
            else "NOT_IDENTIFIED",

        "carrier":
            unique_join(carriers)
            if carriers
            else "NOT_IDENTIFIED",

        "condition":
            unique_join(conditions)
            if conditions
            else "NOT_IDENTIFIED",

        "formulation_identifier":
            formulation_identifier
            if formulation_identifier
            else "NOT_IDENTIFIED",

        "experimental_unit_proxy":
            experimental_unit_proxy,

        "formulation_context_quality":
            quality,

        "exact_row_text":
            exact_text,

        "source_row_text":
            exact_row_text,

        "context_text":
            context_text
    })


    audit_records.append(record)

    if status == "FORMULATION_CONTEXT_READY":

        ready_records.append(record)

    else:

        review_records.append(record)


    if (i + 1) % 10 == 0:

        print(
            f"Processed {i + 1}/{len(obs)}"
        )


# ======================================================================
# DATAFRAMES
# ======================================================================

audit = pd.DataFrame(
    audit_records
)

ready = pd.DataFrame(
    ready_records
)

review = pd.DataFrame(
    review_records
)


# ======================================================================
# IMPORTANT: SAME ARTICLE / SAME TIME AUDIT
# ======================================================================

if not ready.empty:

    same_time_groups = (
        ready.groupby(
            [
                "pmcid",
                "table_index",
                "release_time_h"
            ]
        )
        .size()
        .reset_index(
            name="observations"
        )
    )

    same_time_groups[
        "multiple_observations"
    ] = (
        same_time_groups[
            "observations"
        ] > 1
    )

else:

    same_time_groups = pd.DataFrame()


# ======================================================================
# SAVE
# ======================================================================

audit.to_csv(
    AUDIT_FILE,
    index=False
)

ready.to_csv(
    READY_FILE,
    index=False
)

review.to_csv(
    REVIEW_FILE,
    index=False
)


# ======================================================================
# REPORT
# ======================================================================

print(
    "\n" + "=" * 70
)

print(
    "CODE 6G COMPLETE"
)

print(
    "=" * 70
)

print(
    f"Input observations:              {len(obs):,}"
)

print(
    f"FORMULATION_CONTEXT_READY:       {len(ready):,}"
)

print(
    f"MANUAL_REVIEW:                   {len(review):,}"
)


print(
    "\nFormulation-context quality:"
)

if not ready.empty:

    print(
        ready[
            "formulation_context_quality"
        ].value_counts()
    )


print(
    "\nDrug identification:"
)

if not ready.empty:

    print(
        ready[
            "drug"
        ].value_counts()
    )


print(
    "\nPolymer identification:"
)

if not ready.empty:

    print(
        ready[
            "polymer"
        ].value_counts()
    )


print(
    "\nCarrier identification:"
)

if not ready.empty:

    print(
        ready[
            "carrier"
        ].value_counts()
    )


print(
    "\nUnique formulation identifiers:"
)

if not ready.empty:

    print(
        ready[
            "formulation_identifier"
        ].nunique()
    )


print(
    "\nExperimental-unit proxies:"
)

if not ready.empty:

    print(
        ready[
            "experimental_unit_proxy"
        ].nunique()
    )


print(
    "\nSame-article / same-time groups:"
)

if not same_time_groups.empty:

    print(
        same_time_groups[
            "multiple_observations"
        ].value_counts()
    )


print(
    "\nSaved:"
)

print(
    AUDIT_FILE
)

print(
    READY_FILE
)

print(
    REVIEW_FILE
)


print(
    "\nIMPORTANT:"
)

print(
    "Formulation identifiers are conservative proxies."
)

print(
    "They do NOT establish experimental independence."
)

print(
    "Multiple time points from the same formulation remain valid."
)

print(
    "Different release conditions are kept separate."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print(
    "=" * 70
)
```

```python
# ======================================================================
# CODE 6H — TARGETED FORMULATION METADATA RECOVERY
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6H — TARGETED FORMULATION METADATA RECOVERY")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    ROOT,
    "data",
    "validation",
    "06G_formulation_unit_audit.csv"
)

CODE3_FILE = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "06H_targeted_metadata_audit.csv"
)

RECOVERED_FILE = os.path.join(
    OUT_DIR,
    "06H_recovered_metadata_candidates.csv"
)

REVIEW_FILE = os.path.join(
    OUT_DIR,
    "06H_remaining_manual_review.csv"
)

os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"Missing Code 6G file:\n{INPUT_FILE}"
    )

if not os.path.exists(CODE3_FILE):
    raise FileNotFoundError(
        f"Missing Code 3 file:\n{CODE3_FILE}"
    )

obs = pd.read_csv(INPUT_FILE, low_memory=False)
cells = pd.read_csv(CODE3_FILE, low_memory=False)

print(f"Code 6G observations: {len(obs):,}")
print(f"Code 3 cells: {len(cells):,}")


# ======================================================================
# NORMALIZE
# ======================================================================

def clean(x):
    if pd.isna(x):
        return ""
    return re.sub(
        r"\s+",
        " ",
        str(x)
        .replace("\xa0", " ")
        .replace("\n", " ")
        .replace("\t", " ")
    ).strip()


def norm(x):
    return clean(x).lower()


obs["_pmcid"] = (
    obs["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

cells["_pmcid"] = (
    cells["pmcid"]
    .astype(str)
    .str.strip()
    .str.upper()
)

obs["_table"] = pd.to_numeric(
    obs["table_index"],
    errors="coerce"
)

obs["_row"] = pd.to_numeric(
    obs["row_index"],
    errors="coerce"
)

cells["_table"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

cells["_row"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

cells["_column"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)

cells["_cell"] = cells["raw_cell"].apply(clean)
cells["_row_text"] = cells["row_text"].apply(clean)
cells["_header"] = cells["header_text"].apply(clean)


# ======================================================================
# TARGET ONLY MANUAL OBSERVATIONS
# ======================================================================

if "formulation_resolution" in obs.columns:

    target = obs[
        obs["formulation_resolution"]
        .astype(str)
        .eq("MANUAL_REVIEW")
    ].copy()

else:

    target = obs.copy()

print(
    f"Manual-review observations targeted: {len(target):,}"
)


# ======================================================================
# TERM SETS
# ======================================================================

DRUGS = [
    "dexamethasone",
    "curcumin",
    "paclitaxel",
    "doxorubicin",
    "dox",
    "metformin",
    "ibuprofen",
    "gemcitabine",
    "5-fluorouracil",
    "rifampicin",
    "vancomycin",
    "insulin"
]

POLYMERS = [
    "plga",
    "pla",
    "pcl",
    "peg",
    "peo",
    "pva",
    "pvp",
    "chitosan",
    "alginate",
    "gelatin",
    "cellulose",
    "poloxamer",
    "polyurethane"
]

CARRIERS = [
    "nanoparticle",
    "nanocarrier",
    "liposome",
    "hydrogel",
    "micelle",
    "microsphere",
    "microparticle",
    "nanofiber",
    "film",
    "implant",
    "bead",
    "scaffold"
]

CONDITIONS = [
    "ph",
    "pbs",
    "buffer",
    "acidic",
    "neutral",
    "saline",
    "water",
    "release medium",
    "medium"
]


def find_terms(text, terms):

    text = norm(text)

    found = []

    for term in terms:

        if term in text and term not in found:
            found.append(term)

    return found


def join_unique(values):

    result = []

    for value in values:

        if not value:
            continue

        for x in str(value).split(";"):

            x = x.strip()

            if x and x not in result:
                result.append(x)

    return ";".join(result)


# ======================================================================
# FORMULATION / GROUP IDENTIFIERS
# ======================================================================

ID_PATTERNS = [

    r"\bF[\s\-_]*\d+\b",
    r"\bF[\s\-_]*[A-Z]\b",

    r"\bformulation[\s\-_]*\d+\b",
    r"\bformulation[\s\-_]*[A-Z]\b",

    r"\bformula[\s\-_]*\d+\b",
    r"\bformula[\s\-_]*[A-Z]\b",

    r"\bgroup[\s\-_]*\d+\b",
    r"\bgroup[\s\-_]*[A-Z]\b",

    r"\bsample[\s\-_]*\d+\b",
    r"\bsample[\s\-_]*[A-Z]\b",

    r"\bbatch[\s\-_]*\d+\b",
    r"\bbatch[\s\-_]*[A-Z]\b",

    r"\bS[\s\-_]*\d+\b"
]


def find_identifier(text):

    for pattern in ID_PATTERNS:

        m = re.search(
            pattern,
            text,
            flags=re.I
        )

        if m:
            return m.group(0)

    return ""


# ======================================================================
# CONDITION EXTRACTION
# ======================================================================

def extract_conditions(text):

    found = []

    # pH values
    ph_matches = re.findall(
        r"\bpH\s*[=:]?\s*([0-9]+(?:\.[0-9]+)?)",
        text,
        flags=re.I
    )

    for x in ph_matches:

        value = f"pH {x}"

        if value not in found:
            found.append(value)

    # Explicit temperature
    temp_matches = re.findall(
        r"([0-9]+(?:\.[0-9]+)?)\s*°?\s*C",
        text,
        flags=re.I
    )

    for x in temp_matches:

        value = f"{x} C"

        if value not in found:
            found.append(value)

    # Medium terminology
    lower = text.lower()

    for term in [
        "pbs",
        "phosphate buffer",
        "saline",
        "water",
        "acidic",
        "neutral",
        "release medium"
    ]:

        if term in lower and term not in found:
            found.append(term)

    return ";".join(found)


# ======================================================================
# CONTEXT RECOVERY
# ======================================================================

records = []

for counter, (_, ob) in enumerate(
    target.iterrows(),
    start=1
):

    pmcid = ob["_pmcid"]
    table = ob["_table"]
    row_idx = ob["_row"]

    table_cells = cells[
        (cells["_pmcid"] == pmcid)
        &
        (cells["_table"] == table)
    ].copy()

    record = ob.to_dict()

    if table_cells.empty:

        record.update({
            "6H_status": "MANUAL_REVIEW",
            "6H_reason": "TABLE_NOT_FOUND",
            "recovered_drug": "NOT_IDENTIFIED",
            "recovered_polymer": "NOT_IDENTIFIED",
            "recovered_carrier": "NOT_IDENTIFIED",
            "recovered_condition": "NOT_IDENTIFIED",
            "recovered_formulation_id": "NOT_IDENTIFIED",
            "metadata_evidence": ""
        })

        records.append(record)
        continue


    # --------------------------------------------------------------
    # Exact row
    # --------------------------------------------------------------

    exact = table_cells[
        table_cells["_row"] == row_idx
    ].copy()


    # --------------------------------------------------------------
    # Same-table rows
    # --------------------------------------------------------------

    rows = sorted(
        table_cells["_row"]
        .dropna()
        .unique()
    )

    nearby_rows = [
        r for r in rows
        if abs(r - row_idx) <= 3
    ]

    nearby = table_cells[
        table_cells["_row"].isin(
            nearby_rows
        )
    ].copy()


    exact_text = " | ".join(
        exact["_cell"].tolist()
    )

    exact_row_text = " | ".join(
        exact["_row_text"].tolist()
    )

    nearby_text = " | ".join(
        nearby["_cell"].tolist()
    )

    headers = " | ".join(
        table_cells["_header"]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    # Strongest evidence first
    exact_combined = " | ".join([
        exact_text,
        exact_row_text
    ])

    table_combined = " | ".join([
        headers,
        nearby_text
    ])

    # --------------------------------------------------------------
    # Recover metadata
    # --------------------------------------------------------------

    exact_drugs = find_terms(
        exact_combined,
        DRUGS
    )

    exact_polymers = find_terms(
        exact_combined,
        POLYMERS
    )

    exact_carriers = find_terms(
        exact_combined,
        CARRIERS
    )

    exact_condition = extract_conditions(
        exact_combined
    )

    # If absent from exact row, inspect same-table context.
    context_drugs = find_terms(
        table_combined,
        DRUGS
    )

    context_polymers = find_terms(
        table_combined,
        POLYMERS
    )

    context_carriers = find_terms(
        table_combined,
        CARRIERS
    )

    context_condition = extract_conditions(
        table_combined
    )


    # --------------------------------------------------------------
    # Prefer exact-row metadata
    # --------------------------------------------------------------

    recovered_drug = join_unique([
        ";".join(exact_drugs),
        ";".join(context_drugs)
    ])

    recovered_polymer = join_unique([
        ";".join(exact_polymers),
        ";".join(context_polymers)
    ])

    recovered_carrier = join_unique([
        ";".join(exact_carriers),
        ";".join(context_carriers)
    ])

    recovered_condition = join_unique([
        exact_condition,
        context_condition
    ])


    # --------------------------------------------------------------
    # Identifier
    # --------------------------------------------------------------

    formulation_id = find_identifier(
        exact_combined
    )

    if not formulation_id:

        formulation_id = find_identifier(
            table_combined
        )


    # --------------------------------------------------------------
    # Evidence strength
    # --------------------------------------------------------------

    exact_fields = sum([
        bool(exact_drugs),
        bool(exact_polymers),
        bool(exact_carriers),
        bool(exact_condition),
        bool(formulation_id)
    ])

    context_fields = sum([
        bool(context_drugs),
        bool(context_polymers),
        bool(context_carriers),
        bool(context_condition)
    ])


    # --------------------------------------------------------------
    # Conservative status
    # --------------------------------------------------------------

    if (
        formulation_id
        and
        exact_fields >= 2
    ):

        status = "RECOVERED_STRONG"

    elif (
        exact_fields >= 2
        and
        (
            exact_drugs
            or exact_polymers
            or exact_carriers
        )
    ):

        status = "RECOVERED_GOOD"

    elif (
        exact_fields >= 1
        and
        context_fields >= 1
    ):

        status = "PARTIAL_RECOVERY"

    else:

        status = "MANUAL_REVIEW"


    reason = {
        "RECOVERED_STRONG":
            "EXACT_ROW_PLUS_FORMULATION_CONTEXT",

        "RECOVERED_GOOD":
            "EXACT_ROW_METADATA",

        "PARTIAL_RECOVERY":
            "EXACT_AND_SAME_TABLE_CONTEXT",

        "MANUAL_REVIEW":
            "INSUFFICIENT_STRUCTURAL_METADATA"
    }[status]


    evidence = " || ".join([
        "HEADERS: " + headers,
        "EXACT_ROW: " + exact_combined,
        "NEARBY: " + nearby_text
    ])


    record.update({

        "6H_status": status,

        "6H_reason": reason,

        "recovered_drug":
            recovered_drug
            if recovered_drug
            else "NOT_IDENTIFIED",

        "recovered_polymer":
            recovered_polymer
            if recovered_polymer
            else "NOT_IDENTIFIED",

        "recovered_carrier":
            recovered_carrier
            if recovered_carrier
            else "NOT_IDENTIFIED",

        "recovered_condition":
            recovered_condition
            if recovered_condition
            else "NOT_IDENTIFIED",

        "recovered_formulation_id":
            formulation_id
            if formulation_id
            else "NOT_IDENTIFIED",

        "exact_metadata_fields":
            exact_fields,

        "context_metadata_fields":
            context_fields,

        "metadata_evidence":
            evidence
    })

    records.append(record)

    if counter % 10 == 0:

        print(
            f"Processed {counter}/{len(target)}"
        )


# ======================================================================
# SAVE
# ======================================================================

audit = pd.DataFrame(records)

recovered = audit[
    audit["6H_status"].isin([
        "RECOVERED_STRONG",
        "RECOVERED_GOOD"
    ])
].copy()

remaining = audit[
    ~audit["6H_status"].isin([
        "RECOVERED_STRONG",
        "RECOVERED_GOOD"
    ])
].copy()


audit.to_csv(
    AUDIT_FILE,
    index=False
)

recovered.to_csv(
    RECOVERED_FILE,
    index=False
)

remaining.to_csv(
    REVIEW_FILE,
    index=False
)


# ======================================================================
# REPORT
# ======================================================================

print("\n" + "=" * 70)
print("CODE 6H COMPLETE")
print("=" * 70)

print(
    f"Manual observations targeted:       {len(target):,}"
)

print(
    f"RECOVERED_STRONG:                   "
    f"{(audit['6H_status'] == 'RECOVERED_STRONG').sum():,}"
)

print(
    f"RECOVERED_GOOD:                     "
    f"{(audit['6H_status'] == 'RECOVERED_GOOD').sum():,}"
)

print(
    f"PARTIAL_RECOVERY:                   "
    f"{(audit['6H_status'] == 'PARTIAL_RECOVERY').sum():,}"
)

print(
    f"REMAINING_MANUAL_REVIEW:            "
    f"{(audit['6H_status'] == 'MANUAL_REVIEW').sum():,}"
)

print("\nStatus distribution:")
print(
    audit["6H_status"].value_counts()
)

print("\nRecovered drugs:")
if not recovered.empty:
    print(
        recovered[
            "recovered_drug"
        ].value_counts()
    )

print("\nRecovered polymers:")
if not recovered.empty:
    print(
        recovered[
            "recovered_polymer"
        ].value_counts()
    )

print("\nRecovered carriers:")
if not recovered.empty:
    print(
        recovered[
            "recovered_carrier"
        ].value_counts()
    )

print("\nRecovered formulation IDs:")
if not recovered.empty:
    print(
        recovered[
            "recovered_formulation_id"
        ].nunique()
    )

print("\nSaved:")
print(AUDIT_FILE)
print(RECOVERED_FILE)
print(REVIEW_FILE)

print("\nIMPORTANT:")
print(
    "Metadata are recovered only from the same source table."
)
print(
    "Exact-row evidence is preferred over nearby context."
)
print(
    "Nearby context does not automatically establish experimental identity."
)
print(
    "No observations were generated."
)
print(
    "No interpolation was performed."
)
print(
    "No observations were duplicated."
)
print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6I — CROSS-BRANCH OVERLAP & DEDUPLICATION AUDIT
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6I — CROSS-BRANCH OVERLAP & DEDUPLICATION AUDIT")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

FILE_5F = os.path.join(
    ROOT,
    "data",
    "validation",
    "05F_valid_release_observations.csv"
)

FILE_6F = os.path.join(
    ROOT,
    "data",
    "validation",
    "06F_valid_source_release.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "06I_cross_branch_overlap_audit.csv"
)

UNIQUE_FILE = os.path.join(
    OUT_DIR,
    "06I_unique_6F_observations.csv"
)

OVERLAP_FILE = os.path.join(
    OUT_DIR,
    "06I_overlapping_observations.csv"
)

os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(FILE_5F):
    raise FileNotFoundError(
        f"Missing 5F dataset:\n{FILE_5F}"
    )

if not os.path.exists(FILE_6F):
    raise FileNotFoundError(
        f"Missing 6F dataset:\n{FILE_6F}"
    )


# ======================================================================
# LOAD
# ======================================================================

old = pd.read_csv(
    FILE_5F,
    low_memory=False
)

new = pd.read_csv(
    FILE_6F,
    low_memory=False
)

print(
    f"5F observations: {len(old):,}"
)

print(
    f"6F observations: {len(new):,}"
)


# ======================================================================
# NORMALIZATION
# ======================================================================

def clean_text(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
        .replace("\xa0", " ")
        .strip()
    ).lower()


def num(x):

    return pd.to_numeric(
        x,
        errors="coerce"
    )


for df in [old, new]:

    df["_pmcid"] = (
        df["pmcid"]
        .astype(str)
        .str.strip()
        .str.upper()
    )

    df["_table"] = num(
        df["table_index"]
    )

    df["_row"] = num(
        df["row_index"]
    )

    df["_time"] = num(
        df["release_time_h"]
    )

    df["_release"] = num(
        df["release_percent"]
    )


# ======================================================================
# EXACT STRUCTURAL KEY
# ======================================================================

def make_key(df):

    return (
        df["_pmcid"].fillna("")
        + "|T"
        + df["_table"].fillna(-999).astype(str)
        + "|R"
        + df["_row"].fillna(-999).astype(str)
        + "|TIME"
        + df["_time"].round(6).fillna(-999).astype(str)
        + "|REL"
        + df["_release"].round(6).fillna(-999).astype(str)
    )


old["exact_key"] = make_key(old)
new["exact_key"] = make_key(new)


old_keys = set(
    old["exact_key"]
)

new_keys = set(
    new["exact_key"]
)


# ======================================================================
# EXACT OVERLAP
# ======================================================================

new["_exact_overlap"] = (
    new["exact_key"]
    .isin(old_keys)
)


exact_overlap = new[
    new["_exact_overlap"]
].copy()

exact_unique = new[
    ~new["_exact_overlap"]
].copy()


# ======================================================================
# SECONDARY KEY
# PMCID + TABLE + TIME + RELEASE
# Useful when row numbering differs between extraction branches.
# ======================================================================

def make_secondary_key(df):

    return (
        df["_pmcid"].fillna("")
        + "|T"
        + df["_table"].fillna(-999).astype(str)
        + "|TIME"
        + df["_time"].round(6).fillna(-999).astype(str)
        + "|REL"
        + df["_release"].round(6).fillna(-999).astype(str)
    )


old["secondary_key"] = make_secondary_key(old)
new["secondary_key"] = make_secondary_key(new)

old_secondary = set(
    old["secondary_key"]
)

new["_secondary_overlap"] = (
    new["secondary_key"]
    .isin(old_secondary)
)


secondary_only_overlap = new[
    (~new["_exact_overlap"])
    &
    (new["_secondary_overlap"])
].copy()


# ======================================================================
# SOURCE TABLE + TIME + RELEASE KEY
# ======================================================================

def make_table_value_key(df):

    return (
        df["_pmcid"].fillna("")
        + "|T"
        + df["_table"].fillna(-999).astype(str)
        + "|TIME"
        + df["_time"].round(6).fillna(-999).astype(str)
        + "|REL"
        + df["_release"].round(6).fillna(-999).astype(str)
    )


# Already equivalent to secondary key.
# Kept explicitly for audit readability.


# ======================================================================
# CLASSIFICATION
# ======================================================================

def classify(row):

    if row["_exact_overlap"]:

        return "EXACT_DUPLICATE"

    if row["_secondary_overlap"]:

        return "SAME_SOURCE_TABLE_TIME_RELEASE"

    return "UNIQUE_TO_6F"


new["overlap_class"] = new.apply(
    classify,
    axis=1
)


# ======================================================================
# ARTICLE-LEVEL OVERLAP
# ======================================================================

old_pmcids = set(
    old["_pmcid"]
)

new["article_seen_in_5F"] = (
    new["_pmcid"]
    .isin(old_pmcids)
)


# ======================================================================
# SAVE AUDIT
# ======================================================================

audit = new.copy()

audit.to_csv(
    AUDIT_FILE,
    index=False
)

exact_overlap.to_csv(
    OVERLAP_FILE,
    index=False
)

exact_unique.to_csv(
    UNIQUE_FILE,
    index=False
)


# ======================================================================
# REPORT
# ======================================================================

print("\n" + "=" * 70)
print("CODE 6I COMPLETE")
print("=" * 70)

print(
    f"5F observations:                       {len(old):,}"
)

print(
    f"6F observations:                       {len(new):,}"
)

print(
    f"Exact duplicates:                      {len(exact_overlap):,}"
)

print(
    f"Same table/time/release overlap:       "
    f"{len(secondary_only_overlap):,}"
)

print(
    f"Unique to 6F:                          {len(exact_unique):,}"
)

print(
    f"6F articles already represented in 5F:"
    f" {new['article_seen_in_5F'].sum():,}"
)

print(
    f"6F articles not represented in 5F:"
    f" {(~new['article_seen_in_5F']).sum():,}"
)


print("\nOverlap classification:")

print(
    new[
        "overlap_class"
    ].value_counts()
)


print("\nUnique PMCID among 6F:")

print(
    new["_pmcid"].nunique()
)


print("\nUnique PMCID among exact-overlap observations:")

if not exact_overlap.empty:

    print(
        exact_overlap[
            "_pmcid"
        ].nunique()
    )

else:

    print(0)


print("\nUnique PMCID among unique-to-6F observations:")

if not exact_unique.empty:

    print(
        exact_unique[
            "_pmcid"
        ].nunique()
    )

else:

    print(0)


print("\nSaved:")

print(AUDIT_FILE)
print(UNIQUE_FILE)
print(OVERLAP_FILE)


print("\nIMPORTANT:")

print(
    "No observations were merged automatically."
)

print(
    "Exact structural matches are treated as duplicates."
)

print(
    "Same-table/time/release matches are flagged separately."
)

print(
    "The 6F branch is NOT appended to 5F until this audit is reviewed."
)

print(
    "No synthetic observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6I-R — CROSS-BRANCH OVERLAP & DEDUPLICATION AUDIT
# ROBUST VERSION — DOES NOT ASSUME row_index EXISTS
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6I-R — CROSS-BRANCH OVERLAP & DEDUPLICATION AUDIT")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

FILE_5F = os.path.join(
    ROOT,
    "data",
    "validation",
    "05F_valid_release_observations.csv"
)

FILE_6F = os.path.join(
    ROOT,
    "data",
    "validation",
    "06F_valid_source_release.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "06IR_cross_branch_overlap_audit.csv"
)

UNIQUE_FILE = os.path.join(
    OUT_DIR,
    "06IR_unique_6F_observations.csv"
)

OVERLAP_FILE = os.path.join(
    OUT_DIR,
    "06IR_overlapping_observations.csv"
)

os.makedirs(
    OUT_DIR,
    exist_ok=True
)


# ======================================================================
# LOAD
# ======================================================================

if not os.path.exists(FILE_5F):
    raise FileNotFoundError(
        f"Missing 5F dataset:\n{FILE_5F}"
    )

if not os.path.exists(FILE_6F):
    raise FileNotFoundError(
        f"Missing 6F dataset:\n{FILE_6F}"
    )

old = pd.read_csv(
    FILE_5F,
    low_memory=False
)

new = pd.read_csv(
    FILE_6F,
    low_memory=False
)

print(
    f"5F observations: {len(old):,}"
)

print(
    f"6F observations: {len(new):,}"
)

print("\n5F columns:")
print(list(old.columns))

print("\n6F columns:")
print(list(new.columns))


# ======================================================================
# HELPER FUNCTIONS
# ======================================================================

def find_column(df, candidates):

    lower_map = {
        str(c).lower().strip(): c
        for c in df.columns
    }

    for candidate in candidates:

        key = candidate.lower().strip()

        if key in lower_map:
            return lower_map[key]

    return None


def numeric_series(df, column):

    if column is None:

        return pd.Series(
            np.nan,
            index=df.index
        )

    return pd.to_numeric(
        df[column],
        errors="coerce"
    )


def clean_text(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
        .replace("\xa0", " ")
        .strip()
    ).lower()


# ======================================================================
# DETECT STRUCTURAL COLUMNS
# ======================================================================

PMCID_5F = find_column(
    old,
    [
        "pmcid",
        "PMCID"
    ]
)

PMCID_6F = find_column(
    new,
    [
        "pmcid",
        "PMCID"
    ]
)

TABLE_5F = find_column(
    old,
    [
        "table_index",
        "table",
        "table_idx"
    ]
)

TABLE_6F = find_column(
    new,
    [
        "table_index",
        "table",
        "table_idx"
    ]
)

ROW_5F = find_column(
    old,
    [
        "row_index",
        "row",
        "source_row",
        "xml_row",
        "row_idx"
    ]
)

ROW_6F = find_column(
    new,
    [
        "row_index",
        "row",
        "source_row",
        "xml_row",
        "row_idx"
    ]
)

TIME_5F = find_column(
    old,
    [
        "release_time_h",
        "time_h",
        "time_hours",
        "time"
    ]
)

TIME_6F = find_column(
    new,
    [
        "release_time_h",
        "time_h",
        "time_hours",
        "time"
    ]
)

REL_5F = find_column(
    old,
    [
        "release_percent",
        "release_pct",
        "release"
    ]
)

REL_6F = find_column(
    new,
    [
        "release_percent",
        "release_pct",
        "release"
    ]
)

print("\nDetected columns:")

print(
    "5F PMCID:",
    PMCID_5F
)

print(
    "6F PMCID:",
    PMCID_6F
)

print(
    "5F table:",
    TABLE_5F
)

print(
    "6F table:",
    TABLE_6F
)

print(
    "5F row:",
    ROW_5F
)

print(
    "6F row:",
    ROW_6F
)

print(
    "5F time:",
    TIME_5F
)

print(
    "6F time:",
    TIME_6F
)

print(
    "5F release:",
    REL_5F
)

print(
    "6F release:",
    REL_6F
)


# ======================================================================
# REQUIRED CORE COLUMNS
# ======================================================================

required = [
    ("5F PMCID", PMCID_5F),
    ("6F PMCID", PMCID_6F),
    ("5F table", TABLE_5F),
    ("6F table", TABLE_6F),
    ("5F time", TIME_5F),
    ("6F time", TIME_6F),
    ("5F release", REL_5F),
    ("6F release", REL_6F)
]

missing = [
    name
    for name, col in required
    if col is None
]

if missing:

    raise ValueError(
        "Could not detect required columns:\n"
        + "\n".join(missing)
    )


# ======================================================================
# NORMALIZED STRUCTURAL FIELDS
# ======================================================================

old["_pmcid_norm"] = (
    old[PMCID_5F]
    .astype(str)
    .str.strip()
    .str.upper()
)

new["_pmcid_norm"] = (
    new[PMCID_6F]
    .astype(str)
    .str.strip()
    .str.upper()
)

old["_table_norm"] = numeric_series(
    old,
    TABLE_5F
)

new["_table_norm"] = numeric_series(
    new,
    TABLE_6F
)

old["_row_norm"] = numeric_series(
    old,
    ROW_5F
)

new["_row_norm"] = numeric_series(
    new,
    ROW_6F
)

old["_time_norm"] = numeric_series(
    old,
    TIME_5F
)

new["_time_norm"] = numeric_series(
    new,
    TIME_6F
)

old["_release_norm"] = numeric_series(
    old,
    REL_5F
)

new["_release_norm"] = numeric_series(
    new,
    REL_6F
)


# ======================================================================
# VALUE-BASED KEY
# ======================================================================
# This key does NOT require row_index.
#
# PMCID + table + time + release
#
# This is the primary cross-branch comparison.
# ======================================================================

def value_key(df):

    return (
        df["_pmcid_norm"].fillna("")
        + "|T"
        + df["_table_norm"]
            .fillna(-999999)
            .round(6)
            .astype(str)
        + "|TIME"
        + df["_time_norm"]
            .fillna(-999999)
            .round(6)
            .astype(str)
        + "|REL"
        + df["_release_norm"]
            .fillna(-999999)
            .round(6)
            .astype(str)
    )


old["_value_key"] = value_key(old)
new["_value_key"] = value_key(new)

old_value_keys = set(
    old["_value_key"]
)

new["_value_key"] = (
    new["_value_key"].astype(str)
)


# ======================================================================
# EXACT VALUE-LEVEL OVERLAP
# ======================================================================

new["_value_overlap"] = (
    new["_value_key"]
    .isin(old_value_keys)
)


# ======================================================================
# STRUCTURAL ROW OVERLAP
# ======================================================================

if ROW_5F is not None and ROW_6F is not None:

    def structural_key(df):

        return (
            df["_pmcid_norm"].fillna("")
            + "|T"
            + df["_table_norm"]
                .fillna(-999999)
                .round(6)
                .astype(str)
            + "|R"
            + df["_row_norm"]
                .fillna(-999999)
                .round(6)
                .astype(str)
            + "|TIME"
            + df["_time_norm"]
                .fillna(-999999)
                .round(6)
                .astype(str)
            + "|REL"
            + df["_release_norm"]
                .fillna(-999999)
                .round(6)
                .astype(str)
        )

    old["_structural_key"] = structural_key(old)
    new["_structural_key"] = structural_key(new)

    old_structural_keys = set(
        old["_structural_key"]
    )

    new["_structural_overlap"] = (
        new["_structural_key"]
        .isin(old_structural_keys)
    )

else:

    new["_structural_key"] = ""
    new["_structural_overlap"] = False


# ======================================================================
# ARTICLE OVERLAP
# ======================================================================

old_pmcids = set(
    old["_pmcid_norm"]
)

new["_article_seen_in_5F"] = (
    new["_pmcid_norm"]
    .isin(old_pmcids)
)


# ======================================================================
# CLASSIFICATION
# ======================================================================

def classify(row):

    if row["_structural_overlap"]:

        return "STRUCTURAL_DUPLICATE"

    if row["_value_overlap"]:

        return "SAME_SOURCE_TABLE_TIME_RELEASE"

    if row["_article_seen_in_5F"]:

        return "NEW_OBSERVATION_FROM_EXISTING_ARTICLE"

    return "UNIQUE_ARTICLE_AND_OBSERVATION"


new["overlap_class"] = new.apply(
    classify,
    axis=1
)


# ======================================================================
# SUBSETS
# ======================================================================

overlap = new[
    new["overlap_class"].isin([
        "STRUCTURAL_DUPLICATE",
        "SAME_SOURCE_TABLE_TIME_RELEASE"
    ])
].copy()

unique = new[
    new["overlap_class"].isin([
        "NEW_OBSERVATION_FROM_EXISTING_ARTICLE",
        "UNIQUE_ARTICLE_AND_OBSERVATION"
    ])
].copy()


# ======================================================================
# SAVE
# ======================================================================

new.to_csv(
    AUDIT_FILE,
    index=False
)

unique.to_csv(
    UNIQUE_FILE,
    index=False
)

overlap.to_csv(
    OVERLAP_FILE,
    index=False
)


# ======================================================================
# REPORT
# ======================================================================

print("\n" + "=" * 70)
print("CODE 6I-R COMPLETE")
print("=" * 70)

print(
    f"5F observations:                         {len(old):,}"
)

print(
    f"6F observations:                         {len(new):,}"
)

print(
    f"Structural duplicates:                   "
    f"{(new['overlap_class'] == 'STRUCTURAL_DUPLICATE').sum():,}"
)

print(
    f"Same table/time/release overlap:         "
    f"{(new['overlap_class'] == 'SAME_SOURCE_TABLE_TIME_RELEASE').sum():,}"
)

print(
    f"New observations from existing articles: "
    f"{(new['overlap_class'] == 'NEW_OBSERVATION_FROM_EXISTING_ARTICLE').sum():,}"
)

print(
    f"Unique article + observation:            "
    f"{(new['overlap_class'] == 'UNIQUE_ARTICLE_AND_OBSERVATION').sum():,}"
)

print(
    f"Total unique-to-6F observations:         "
    f"{len(unique):,}"
)

print(
    f"6F articles already represented in 5F:  "
    f"{new['_article_seen_in_5F'].sum():,}"
)

print(
    f"6F articles not represented in 5F:      "
    f"{(~new['_article_seen_in_5F']).sum():,}"
)


print("\nOverlap classification:")
print(
    new["overlap_class"].value_counts()
)


print("\nUnique PMCID in 5F:")
print(
    old["_pmcid_norm"].nunique()
)


print("\nUnique PMCID in 6F:")
print(
    new["_pmcid_norm"].nunique()
)


print("\nUnique PMCID among unique-to-6F observations:")

if not unique.empty:

    print(
        unique["_pmcid_norm"].nunique()
    )

else:

    print(0)


print("\nSaved:")
print(AUDIT_FILE)
print(UNIQUE_FILE)
print(OVERLAP_FILE)


print("\nIMPORTANT:")
print(
    "No observations were merged."
)

print(
    "No observations were deleted from the original branches."
)

print(
    "This is an audit only."
)

print(
    "Value-level overlap uses PMCID + table + time + release."
)

print(
    "Structural overlap additionally uses row index when available."
)

print(
    "No synthetic observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6I-R2 — CROSS-BRANCH OVERLAP & DEDUPLICATION AUDIT
# SCHEMA-CORRECTED VERSION
# ======================================================================

import os
import pandas as pd
import numpy as np

print("=" * 70)
print("CODE 6I-R2 — CROSS-BRANCH OVERLAP & DEDUPLICATION AUDIT")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

FILE_5F = os.path.join(
    ROOT,
    "data",
    "validation",
    "05F_valid_release_observations.csv"
)

FILE_6F = os.path.join(
    ROOT,
    "data",
    "validation",
    "06F_valid_source_release.csv"
)

OUT_DIR = os.path.join(
    ROOT,
    "data",
    "validation"
)

AUDIT_FILE = os.path.join(
    OUT_DIR,
    "06IR2_cross_branch_overlap_audit.csv"
)

UNIQUE_FILE = os.path.join(
    OUT_DIR,
    "06IR2_unique_6F_observations.csv"
)

OVERLAP_FILE = os.path.join(
    OUT_DIR,
    "06IR2_overlapping_observations.csv"
)

os.makedirs(
    OUT_DIR,
    exist_ok=True
)


# ======================================================================
# LOAD
# ======================================================================

if not os.path.exists(FILE_5F):
    raise FileNotFoundError(
        f"Missing 5F dataset:\n{FILE_5F}"
    )

if not os.path.exists(FILE_6F):
    raise FileNotFoundError(
        f"Missing 6F dataset:\n{FILE_6F}"
    )

old = pd.read_csv(
    FILE_5F,
    low_memory=False
)

new = pd.read_csv(
    FILE_6F,
    low_memory=False
)

print(
    f"5F observations: {len(old):,}"
)

print(
    f"6F observations: {len(new):,}"
)


# ======================================================================
# SCHEMA MAPPING
# ======================================================================

# 5F schema
old_pmcid_col = "pmcid"
old_table_col = "table_index"
old_time_col = "candidate_time_h"
old_release_col = "candidate_release_percent"

# 6F schema
new_pmcid_col = "pmcid"
new_table_col = "table_index"
new_row_col = "row_index"
new_time_col = "release_time_h"
new_release_col = "release_percent"


required_5f = [
    old_pmcid_col,
    old_table_col,
    old_time_col,
    old_release_col
]

required_6f = [
    new_pmcid_col,
    new_table_col,
    new_row_col,
    new_time_col,
    new_release_col
]

missing_5f = [
    c for c in required_5f
    if c not in old.columns
]

missing_6f = [
    c for c in required_6f
    if c not in new.columns
]

if missing_5f:

    raise ValueError(
        "Missing required 5F columns:\n"
        + "\n".join(missing_5f)
    )

if missing_6f:

    raise ValueError(
        "Missing required 6F columns:\n"
        + "\n".join(missing_6f)
    )


print("\nSchema mapping:")
print(
    f"5F time    : {old_time_col}"
)
print(
    f"5F release : {old_release_col}"
)
print(
    f"6F time    : {new_time_col}"
)
print(
    f"6F release : {new_release_col}"
)


# ======================================================================
# NORMALIZED FIELDS
# ======================================================================

old["_pmcid_norm"] = (
    old[old_pmcid_col]
    .astype(str)
    .str.strip()
    .str.upper()
)

new["_pmcid_norm"] = (
    new[new_pmcid_col]
    .astype(str)
    .str.strip()
    .str.upper()
)

old["_table_norm"] = pd.to_numeric(
    old[old_table_col],
    errors="coerce"
)

new["_table_norm"] = pd.to_numeric(
    new[new_table_col],
    errors="coerce"
)

old["_time_norm"] = pd.to_numeric(
    old[old_time_col],
    errors="coerce"
)

new["_time_norm"] = pd.to_numeric(
    new[new_time_col],
    errors="coerce"
)

old["_release_norm"] = pd.to_numeric(
    old[old_release_col],
    errors="coerce"
)

new["_release_norm"] = pd.to_numeric(
    new[new_release_col],
    errors="coerce"
)

new["_row_norm"] = pd.to_numeric(
    new[new_row_col],
    errors="coerce"
)


# ======================================================================
# PRIMARY CROSS-BRANCH KEY
#
# PMCID + TABLE + TIME + RELEASE
#
# This is deliberately independent of row_index because 5F does not
# contain a row_index.
# ======================================================================

def make_value_key(df):

    return (
        df["_pmcid_norm"].fillna("")
        + "|T"
        + df["_table_norm"]
            .fillna(-999999)
            .round(6)
            .astype(str)
        + "|TIME"
        + df["_time_norm"]
            .fillna(-999999)
            .round(6)
            .astype(str)
        + "|REL"
        + df["_release_norm"]
            .fillna(-999999)
            .round(6)
            .astype(str)
    )


old["_value_key"] = make_value_key(old)
new["_value_key"] = make_value_key(new)

old_keys = set(
    old["_value_key"]
)


# ======================================================================
# VALUE-LEVEL OVERLAP
# ======================================================================

new["_value_overlap"] = (
    new["_value_key"]
    .isin(old_keys)
)


# ======================================================================
# ARTICLE-LEVEL OVERLAP
# ======================================================================

old_articles = set(
    old["_pmcid_norm"]
)

new["_article_seen_in_5F"] = (
    new["_pmcid_norm"]
    .isin(old_articles)
)


# ======================================================================
# ROW-LEVEL DIAGNOSTIC
#
# Cannot be a cross-branch duplicate key because 5F has no row_index.
# We therefore report the 6F row but DO NOT use it to claim duplication.
# ======================================================================

new["_row_available_only_in_6F"] = (
    new["_row_norm"].notna()
)


# ======================================================================
# CLASSIFICATION
# ======================================================================

def classify(row):

    if row["_value_overlap"]:

        return "VALUE_LEVEL_OVERLAP"

    if row["_article_seen_in_5F"]:

        return "NEW_OBSERVATION_FROM_EXISTING_ARTICLE"

    return "UNIQUE_ARTICLE_AND_OBSERVATION"


new["overlap_class"] = new.apply(
    classify,
    axis=1
)


# ======================================================================
# SUBSETS
# ======================================================================

overlap = new[
    new["overlap_class"]
    == "VALUE_LEVEL_OVERLAP"
].copy()

unique = new[
    new["overlap_class"]
    != "VALUE_LEVEL_OVERLAP"
].copy()


# ======================================================================
# SAVE
# ======================================================================

new.to_csv(
    AUDIT_FILE,
    index=False
)

unique.to_csv(
    UNIQUE_FILE,
    index=False
)

overlap.to_csv(
    OVERLAP_FILE,
    index=False
)


# ======================================================================
# REPORT
# ======================================================================

print("\n" + "=" * 70)
print("CODE 6I-R2 COMPLETE")
print("=" * 70)

print(
    f"5F observations:                         {len(old):,}"
)

print(
    f"6F observations:                         {len(new):,}"
)

print(
    f"Value-level overlap:                     "
    f"{len(overlap):,}"
)

print(
    f"New observations from existing articles: "
    f"{(
        new['overlap_class']
        == 'NEW_OBSERVATION_FROM_EXISTING_ARTICLE'
    ).sum():,}"
)

print(
    f"Unique article + observation:            "
    f"{(
        new['overlap_class']
        == 'UNIQUE_ARTICLE_AND_OBSERVATION'
    ).sum():,}"
)

print(
    f"Total unique-to-5F/6F comparison:       "
    f"{len(unique):,}"
)


print("\nOverlap classification:")

print(
    new[
        "overlap_class"
    ].value_counts()
)


print("\nUnique PMCIDs:")

print(
    f"5F: {old['_pmcid_norm'].nunique():,}"
)

print(
    f"6F: {new['_pmcid_norm'].nunique():,}"
)

print(
    f"6F PMCIDs already represented in 5F: "
    f"{new['_article_seen_in_5F'].sum():,}"
)

print(
    f"6F PMCIDs not represented in 5F: "
    f"{(~new['_article_seen_in_5F']).sum():,}"
)


# ======================================================================
# OVERLAP PREVIEW
# ======================================================================

if not overlap.empty:

    print("\nFirst overlapping observations:")

    preview_cols = [
        "_pmcid_norm",
        "_table_norm",
        "_time_norm",
        "_release_norm",
        "overlap_class"
    ]

    print(
        overlap[
            preview_cols
        ].head(20).to_string(
            index=False
        )
    )

else:

    print(
        "\nNo value-level overlap detected."
    )


# ======================================================================
# UNIQUE 6F PREVIEW
# ======================================================================

if not unique.empty:

    print(
        "\nFirst observations not already present "
        "by value-level matching:"
    )

    preview_cols = [
        "_pmcid_norm",
        "_table_norm",
        "_row_norm",
        "_time_norm",
        "_release_norm",
        "overlap_class"
    ]

    print(
        unique[
            preview_cols
        ].head(20).to_string(
            index=False
        )
    )


# ======================================================================
# SAVE SUMMARY
# ======================================================================

summary = pd.DataFrame({

    "metric": [
        "5F_observations",
        "6F_observations",
        "value_level_overlap",
        "new_observations_existing_article",
        "unique_article_observations",
        "5F_unique_pmcid",
        "6F_unique_pmcid",
        "6F_articles_seen_in_5F",
        "6F_articles_not_seen_in_5F"
    ],

    "value": [
        len(old),
        len(new),
        len(overlap),

        (
            new["overlap_class"]
            == "NEW_OBSERVATION_FROM_EXISTING_ARTICLE"
        ).sum(),

        (
            new["overlap_class"]
            == "UNIQUE_ARTICLE_AND_OBSERVATION"
        ).sum(),

        old["_pmcid_norm"].nunique(),
        new["_pmcid_norm"].nunique(),

        new["_article_seen_in_5F"].sum(),
        (~new["_article_seen_in_5F"]).sum()
    ]

})

SUMMARY_FILE = os.path.join(
    OUT_DIR,
    "06IR2_cross_branch_summary.csv"
)

summary.to_csv(
    SUMMARY_FILE,
    index=False
)


print("\nSaved:")
print(AUDIT_FILE)
print(UNIQUE_FILE)
print(OVERLAP_FILE)
print(SUMMARY_FILE)


print("\nIMPORTANT:")
print(
    "This is an overlap audit only."
)

print(
    "5F and 6F have different schemas."
)

print(
    "5F has no row_index, so row_index is NOT used "
    "to declare duplication."
)

print(
    "Primary overlap criterion is PMCID + table + time + release."
)

print(
    "No observations were merged."
)

print(
    "No observations were deleted from either original dataset."
)

print(
    "No synthetic observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6J — FINAL 6F FORMULATION / EXPERIMENTAL-UNIT RESOLUTION
# ======================================================================
#
# Purpose:
#   Resolve drug, polymer, carrier, formulation, condition and
#   experimental-unit metadata for the 60 validated 6F observations.
#
# Rules:
#   1. Use ONLY the same source table.
#   2. Prefer exact source-row evidence.
#   3. Use table headers and row context only when structurally justified.
#   4. Do NOT infer metadata from other articles.
#   5. Do NOT infer missing drug/polymer/carrier from general knowledge.
#   6. Keep different release conditions separate.
#   7. Multiple time points from one formulation are retained.
#   8. Do NOT generate observations.
#   9. Do NOT interpolate.
#  10. Do NOT run ML.
# ======================================================================

import os
import re
import numpy as np
import pandas as pd

print("=" * 70)
print("CODE 6J — FINAL 6F FORMULATION / EXPERIMENTAL-UNIT RESOLUTION")
print("=" * 70)

ROOT = "/content/AI_Polymer_Drug_Release"

F6 = os.path.join(
    ROOT,
    "data",
    "validation",
    "06F_valid_source_release.csv"
)

F3 = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT = os.path.join(
    ROOT,
    "data",
    "validation"
)

os.makedirs(OUT, exist_ok=True)

if not os.path.exists(F6):
    raise FileNotFoundError(
        f"6F validated dataset not found:\n{F6}"
    )

if not os.path.exists(F3):
    raise FileNotFoundError(
        f"Code 3 table-cell dataset not found:\n{F3}"
    )

valid = pd.read_csv(F6, low_memory=False)
cells = pd.read_csv(F3, low_memory=False)

print(f"6F observations: {len(valid):,}")
print(f"Code 3 cells:    {len(cells):,}")

# ======================================================================
# CHECK REQUIRED COLUMNS
# ======================================================================

required_6f = [
    "observation_id",
    "pmcid",
    "table_index",
    "row_index",
    "time_column",
    "release_column",
    "time_raw",
    "release_raw",
    "release_time_h",
    "release_percent",
    "time_header",
    "release_header",
    "row_text"
]

required_3 = [
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell",
    "row_text",
    "full_table_text",
    "table_caption",
    "header_text",
    "article_title"
]

missing_6f = [
    c for c in required_6f
    if c not in valid.columns
]

missing_3 = [
    c for c in required_3
    if c not in cells.columns
]

if missing_6f:
    raise ValueError(
        "Missing required 6F columns:\n"
        + "\n".join(missing_6f)
    )

if missing_3:
    raise ValueError(
        "Missing required Code 3 columns:\n"
        + "\n".join(missing_3)
    )

# ======================================================================
# NORMALIZATION
# ======================================================================

def norm_text(x):
    if pd.isna(x):
        return ""
    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


def norm_id(x):
    if pd.isna(x):
        return ""
    return str(x).strip().upper()


valid["_pmcid"] = valid["pmcid"].map(norm_id)
cells["_pmcid"] = cells["pmcid"].map(norm_id)

valid["_table"] = pd.to_numeric(
    valid["table_index"],
    errors="coerce"
)

cells["_table"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

valid["_row"] = pd.to_numeric(
    valid["row_index"],
    errors="coerce"
)

cells["_row"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

cells["_col"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)

for col in [
    "time_header",
    "release_header",
    "row_text",
    "time_raw",
    "release_raw"
]:
    valid[f"_clean_{col}"] = valid[col].map(norm_text)

for col in [
    "raw_cell",
    "row_text",
    "full_table_text",
    "table_caption",
    "header_text",
    "article_title"
]:
    cells[f"_clean_{col}"] = cells[col].map(norm_text)


# ======================================================================
# METADATA VOCABULARY
#
# These are extraction terms, NOT assumptions about the experiment.
# ======================================================================

DRUG_TERMS = {
    "doxorubicin": [
        "doxorubicin", "dox"
    ],
    "curcumin": [
        "curcumin"
    ],
    "paclitaxel": [
        "paclitaxel", "taxol"
    ],
    "dexamethasone": [
        "dexamethasone"
    ],
    "metformin": [
        "metformin"
    ],
    "ibuprofen": [
        "ibuprofen"
    ],
    "gemcitabine": [
        "gemcitabine"
    ],
    "5-fluorouracil": [
        "5-fluorouracil", "5 fu", "5-fu"
    ],
    "rifampicin": [
        "rifampicin", "rifampin"
    ],
    "vancomycin": [
        "vancomycin"
    ],
    "insulin": [
        "insulin"
    ]
}

POLYMER_TERMS = {
    "PLGA": ["plga"],
    "PLA": ["pla", "polylactic acid"],
    "PCL": ["pcl", "polycaprolactone"],
    "PEG": ["peg", "polyethylene glycol"],
    "PVA": ["pva", "polyvinyl alcohol"],
    "PEO": ["peo", "polyethylene oxide"],
    "chitosan": ["chitosan"],
    "alginate": ["alginate"],
    "gelatin": ["gelatin"],
    "cellulose": ["cellulose"],
    "poloxamer": ["poloxamer"],
    "polyurethane": ["polyurethane"]
}

CARRIER_TERMS = {
    "nanoparticle": [
        "nanoparticle",
        "nanoparticles",
        "nanoparticle-based"
    ],
    "nanocarrier": [
        "nanocarrier",
        "nanocarriers"
    ],
    "hydrogel": [
        "hydrogel",
        "hydrogels"
    ],
    "microparticle": [
        "microparticle",
        "microparticles"
    ],
    "microsphere": [
        "microsphere",
        "microspheres"
    ],
    "nanofiber": [
        "nanofiber",
        "nanofibers"
    ],
    "liposome": [
        "liposome",
        "liposomes"
    ],
    "micelle": [
        "micelle",
        "micelles"
    ],
    "film": [
        "film",
        "films"
    ],
    "bead": [
        "bead",
        "beads"
    ],
    "implant": [
        "implant",
        "implants"
    ]
}

CONDITION_TERMS = [
    "ph",
    "pbs",
    "buffer",
    "saline",
    "acidic",
    "neutral",
    "alkaline",
    "intestinal",
    "gastric",
    "water",
    "simulated gastric",
    "simulated intestinal",
    "temperature",
    "37",
    "25",
    "37°c"
]


# ======================================================================
# EXTRACTION HELPERS
# ======================================================================

def find_terms(text, vocabulary):

    text = norm_text(text).lower()

    found = []

    for canonical, terms in vocabulary.items():

        for term in terms:

            pattern = r"(?<![a-z0-9])" + re.escape(term.lower()) + r"(?![a-z0-9])"

            if re.search(pattern, text):
                found.append(canonical)
                break

    return list(dict.fromkeys(found))


def find_condition_context(text):

    text = norm_text(text)

    if not text:
        return ""

    low = text.lower()

    hits = []

    # pH values
    for m in re.finditer(
        r"\bpH\s*[=:]?\s*(\d+(?:\.\d+)?)",
        text,
        flags=re.I
    ):
        hits.append(f"pH {m.group(1)}")

    # temperature
    for m in re.finditer(
        r"(\d+(?:\.\d+)?)\s*°?\s*C\b",
        text,
        flags=re.I
    ):
        hits.append(f"{m.group(1)} C")

    # common release media
    media_patterns = [
        r"PBS",
        r"phosphate buffer",
        r"buffer",
        r"saline",
        r"simulated gastric fluid",
        r"simulated intestinal fluid",
        r"SGF",
        r"SIF"
    ]

    for pat in media_patterns:

        if re.search(
            pat,
            text,
            flags=re.I
        ):
            hits.append(
                re.sub(
                    r"\s+",
                    " ",
                    pat
                )
            )

    return "; ".join(
        dict.fromkeys(hits)
    )


def unique_join(values):

    out = []

    for x in values:

        x = norm_text(x)

        if not x:
            continue

        if x not in out:
            out.append(x)

    return ";".join(out)


# ======================================================================
# BUILD TABLE LOOKUP
# ======================================================================

print("\nBuilding same-table structural lookup...")

table_groups = {}

for key, group in cells.groupby(
    ["_pmcid", "_table"],
    sort=False
):

    table_groups[key] = group.copy()

print(
    f"Unique Code 3 source tables: {len(table_groups):,}"
)


# ======================================================================
# RESOLVE EACH 6F OBSERVATION
# ======================================================================

results = []

for i, obs in valid.iterrows():

    pmcid = obs["_pmcid"]
    table = obs["_table"]
    row = obs["_row"]

    table_key = (
        pmcid,
        table
    )

    table_cells = table_groups.get(
        table_key,
        pd.DataFrame()
    )

    exact_row = pd.DataFrame()

    if not table_cells.empty and pd.notna(row):

        exact_row = table_cells[
            table_cells["_row"] == row
        ].copy()

    # --------------------------------------------------------------
    # Exact row text
    # --------------------------------------------------------------

    exact_row_text = ""

    if not exact_row.empty:

        exact_row_text = unique_join(
            exact_row["_clean_raw_cell"]
        )

    # --------------------------------------------------------------
    # Full table context
    # --------------------------------------------------------------

    full_table_text = ""

    if not table_cells.empty:

        full_table_text = unique_join(
            table_cells["_clean_full_table_text"]
        )

    # --------------------------------------------------------------
    # Header context
    # --------------------------------------------------------------

    header_text = ""

    if not table_cells.empty:

        header_text = unique_join(
            table_cells["_clean_header_text"]
        )

    # Also include the 6F headers because these are validated
    # structural headers.
    header_context = " | ".join([
        norm_text(obs["time_header"]),
        norm_text(obs["release_header"]),
        header_text
    ])

    # --------------------------------------------------------------
    # Article title
    # --------------------------------------------------------------

    article_title = ""

    if not table_cells.empty:

        vals = table_cells[
            "_clean_article_title"
        ].dropna().unique()

        if len(vals):
            article_title = vals[0]

    # --------------------------------------------------------------
    # Row context
    # --------------------------------------------------------------

    row_context = " | ".join([
        exact_row_text,
        norm_text(obs["row_text"])
    ])

    # --------------------------------------------------------------
    # IMPORTANT:
    # Metadata hierarchy:
    #
    # 1. Exact row
    # 2. Validated table headers
    # 3. Same-table context
    #
    # We record evidence rather than silently filling everything.
    # --------------------------------------------------------------

    exact_drugs = find_terms(
        exact_row_text,
        DRUG_TERMS
    )

    exact_polymers = find_terms(
        exact_row_text,
        POLYMER_TERMS
    )

    exact_carriers = find_terms(
        exact_row_text,
        CARRIER_TERMS
    )

    row_drugs = find_terms(
        row_context,
        DRUG_TERMS
    )

    row_polymers = find_terms(
        row_context,
        POLYMER_TERMS
    )

    row_carriers = find_terms(
        row_context,
        CARRIER_TERMS
    )

    table_drugs = find_terms(
        full_table_text,
        DRUG_TERMS
    )

    table_polymers = find_terms(
        full_table_text,
        POLYMER_TERMS
    )

    table_carriers = find_terms(
        full_table_text,
        CARRIER_TERMS
    )

    # --------------------------------------------------------------
    # Resolve only when exact-row evidence exists.
    #
    # Otherwise mark table-level candidates separately.
    # --------------------------------------------------------------

    if exact_drugs:
        resolved_drug = ";".join(exact_drugs)
        drug_source = "EXACT_ROW"

    elif row_drugs:
        resolved_drug = ";".join(row_drugs)
        drug_source = "ROW_CONTEXT"

    else:
        resolved_drug = "NOT_IDENTIFIED"
        drug_source = "UNRESOLVED"

    if exact_polymers:
        resolved_polymer = ";".join(exact_polymers)
        polymer_source = "EXACT_ROW"

    elif row_polymers:
        resolved_polymer = ";".join(row_polymers)
        polymer_source = "ROW_CONTEXT"

    else:
        resolved_polymer = "NOT_IDENTIFIED"
        polymer_source = "UNRESOLVED"

    if exact_carriers:
        resolved_carrier = ";".join(exact_carriers)
        carrier_source = "EXACT_ROW"

    elif row_carriers:
        resolved_carrier = ";".join(row_carriers)
        carrier_source = "ROW_CONTEXT"

    else:
        resolved_carrier = "NOT_IDENTIFIED"
        carrier_source = "UNRESOLVED"

    # --------------------------------------------------------------
    # Condition
    # --------------------------------------------------------------

    condition = find_condition_context(
        row_context
    )

    if not condition:

        condition = find_condition_context(
            header_context
        )

    # --------------------------------------------------------------
    # Formulation ID
    #
    # Conservative deterministic proxy.
    # It is NOT experimental ground truth.
    # --------------------------------------------------------------

    formulation_components = [
        resolved_drug,
        resolved_polymer,
        resolved_carrier,
        condition
    ]

    formulation_components = [
        x for x in formulation_components
        if x and x != "NOT_IDENTIFIED"
    ]

    if formulation_components:

        formulation_id = (
            pmcid
            + "_T"
            + str(int(table))
            + "_"
            + "_".join(
                formulation_components
            )
        )

        formulation_id = re.sub(
            r"[^A-Za-z0-9_.;:-]+",
            "_",
            formulation_id
        )

    else:

        formulation_id = "UNRESOLVED"

    # --------------------------------------------------------------
    # Experimental unit proxy
    #
    # Same formulation + same source table + same condition.
    # Time is deliberately NOT included because multiple time points
    # from the same formulation belong to one release experiment.
    # --------------------------------------------------------------

    if formulation_id != "UNRESOLVED":

        experimental_unit_id = (
            formulation_id
            + "_UNIT"
        )

    else:

        experimental_unit_id = (
            pmcid
            + "_T"
            + str(int(table))
            + "_UNIT_UNRESOLVED"
        )

    # --------------------------------------------------------------
    # Metadata quality
    # --------------------------------------------------------------

    exact_count = sum([
        bool(exact_drugs),
        bool(exact_polymers),
        bool(exact_carriers)
    ])

    row_count = sum([
        bool(row_drugs),
        bool(row_polymers),
        bool(row_carriers)
    ])

    if exact_count >= 2:

        metadata_quality = "HIGH"

    elif exact_count == 1 or row_count >= 2:

        metadata_quality = "GOOD"

    elif row_count == 1:

        metadata_quality = "PARTIAL"

    else:

        metadata_quality = "LOW"

    # --------------------------------------------------------------
    # Final readiness
    #
    # Release values themselves are already source validated in 6F.
    # This readiness flag concerns metadata suitability for modeling.
    # --------------------------------------------------------------

    if (
        resolved_drug != "NOT_IDENTIFIED"
        and resolved_polymer != "NOT_IDENTIFIED"
        and resolved_carrier != "NOT_IDENTIFIED"
    ):

        final_status = "READY_FOR_FINAL_VALIDATION"

    elif (
        resolved_drug != "NOT_IDENTIFIED"
        and resolved_polymer != "NOT_IDENTIFIED"
    ):

        final_status = "PARTIAL_METADATA"

    else:

        final_status = "MANUAL_REVIEW"

    results.append({

        "observation_id":
            obs["observation_id"],

        "pmcid":
            obs["pmcid"],

        "table_index":
            obs["table_index"],

        "row_index":
            obs["row_index"],

        "time_column":
            obs["time_column"],

        "release_column":
            obs["release_column"],

        "time_raw":
            obs["time_raw"],

        "release_raw":
            obs["release_raw"],

        "release_time_h":
            obs["release_time_h"],

        "release_percent":
            obs["release_percent"],

        "time_header":
            obs["time_header"],

        "release_header":
            obs["release_header"],

        "exact_row_text":
            exact_row_text,

        "table_context":
            full_table_text,

        "article_title":
            article_title,

        "drug":
            resolved_drug,

        "polymer":
            resolved_polymer,

        "carrier":
            resolved_carrier,

        "condition":
            condition,

        "drug_source":
            drug_source,

        "polymer_source":
            polymer_source,

        "carrier_source":
            carrier_source,

        "formulation_id":
            formulation_id,

        "experimental_unit_id":
            experimental_unit_id,

        "metadata_quality":
            metadata_quality,

        "final_status":
            final_status,

        "exact_row_found":
            bool(not exact_row.empty),

        "exact_row_cell_count":
            len(exact_row)
    })


    if (i + 1) % 10 == 0:

        print(
            f"Processed {i+1}/{len(valid)}"
        )


# ======================================================================
# OUTPUT
# ======================================================================

resolved = pd.DataFrame(
    results
)

# ======================================================================
# SAME-ARTICLE / SAME-TIME DIAGNOSTIC
# ======================================================================

resolved["_time_key"] = pd.to_numeric(
    resolved["release_time_h"],
    errors="coerce"
).round(8)

resolved["_group_key"] = (
    resolved["pmcid"].astype(str)
    + "|T"
    + resolved["table_index"].astype(str)
    + "|TIME"
    + resolved["_time_key"].astype(str)
)

group_counts = (
    resolved
    .groupby("_group_key")
    .size()
    .rename("observations_at_same_article_table_time")
)

resolved = resolved.merge(
    group_counts,
    left_on="_group_key",
    right_index=True,
    how="left"
)

resolved["same_article_same_time_multiple"] = (
    resolved["observations_at_same_article_table_time"] > 1
)

resolved = resolved.drop(
    columns=[
        "_time_key",
        "_group_key"
    ]
)


# ======================================================================
# FORMULATION-LEVEL TIMEPOINT COUNTS
# ======================================================================

unit_counts = (
    resolved[
        resolved["experimental_unit_id"]
        != "UNRESOLVED"
    ]
    .groupby("experimental_unit_id")
    .size()
    .rename("timepoints_per_experimental_unit")
)

resolved = resolved.merge(
    unit_counts,
    left_on="experimental_unit_id",
    right_index=True,
    how="left"
)

resolved["timepoints_per_experimental_unit"] = (
    resolved[
        "timepoints_per_experimental_unit"
    ].fillna(1)
)


# ======================================================================
# FINAL DATASETS
# ======================================================================

ready = resolved[
    resolved["final_status"]
    == "READY_FOR_FINAL_VALIDATION"
].copy()

partial = resolved[
    resolved["final_status"]
    == "PARTIAL_METADATA"
].copy()

manual = resolved[
    resolved["final_status"]
    == "MANUAL_REVIEW"
].copy()


# ======================================================================
# SAVE
# ======================================================================

AUDIT_FILE = os.path.join(
    OUT,
    "06J_final_formulation_unit_audit.csv"
)

READY_FILE = os.path.join(
    OUT,
    "06J_ready_for_final_validation.csv"
)

PARTIAL_FILE = os.path.join(
    OUT,
    "06J_partial_metadata.csv"
)

MANUAL_FILE = os.path.join(
    OUT,
    "06J_manual_formulation_review.csv"
)

resolved.to_csv(
    AUDIT_FILE,
    index=False
)

ready.to_csv(
    READY_FILE,
    index=False
)

partial.to_csv(
    PARTIAL_FILE,
    index=False
)

manual.to_csv(
    MANUAL_FILE,
    index=False
)


# ======================================================================
# REPORT
# ======================================================================

print("\n" + "=" * 70)
print("CODE 6J COMPLETE")
print("=" * 70)

print(
    f"Input 6F observations:          {len(valid):,}"
)

print(
    f"Resolved observations:           {len(resolved):,}"
)

print(
    f"Ready for final validation:      {len(ready):,}"
)

print(
    f"Partial metadata:                {len(partial):,}"
)

print(
    f"Manual formulation review:       {len(manual):,}"
)

print("\nMetadata quality:")

print(
    resolved[
        "metadata_quality"
    ].value_counts()
)

print("\nFinal status:")

print(
    resolved[
        "final_status"
    ].value_counts()
)

print("\nDrug identification:")

print(
    resolved[
        "drug"
    ].value_counts()
)

print("\nPolymer identification:")

print(
    resolved[
        "polymer"
    ].value_counts()
)

print("\nCarrier identification:")

print(
    resolved[
        "carrier"
    ].value_counts()
)

print(
    "\nUnique formulation IDs: "
    + str(
        resolved[
            resolved["formulation_id"]
            != "UNRESOLVED"
        ]["formulation_id"].nunique()
    )
)

print(
    "Unique experimental-unit proxies: "
    + str(
        resolved[
            resolved["experimental_unit_id"]
            != "UNRESOLVED"
        ]["experimental_unit_id"].nunique()
    )
)

print(
    "\nObservations with multiple time points "
    "within the same experimental-unit proxy:"
)

multi = resolved[
    resolved[
        "timepoints_per_experimental_unit"
    ] > 1
]

print(
    multi[
        [
            "experimental_unit_id",
            "timepoints_per_experimental_unit"
        ]
    ]
    .drop_duplicates()
    .sort_values(
        "timepoints_per_experimental_unit",
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)

print("\nSaved:")
print(AUDIT_FILE)
print(READY_FILE)
print(PARTIAL_FILE)
print(MANUAL_FILE)

print("\nIMPORTANT:")
print(
    "6F release observations remain source-validated."
)

print(
    "Metadata were recovered only from the same source table."
)

print(
    "Exact-row evidence is preferred."
)

print(
    "No metadata were inferred from unrelated articles."
)

print(
    "Different release conditions remain separate."
)

print(
    "Multiple time points from the same formulation are retained."
)

print(
    "Experimental-unit IDs are conservative proxies, not ground truth."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6K — COLUMN-AWARE STRUCTURAL FORMULATION RECOVERY
# ======================================================================
#
# PURPOSE
# -------
# Recover formulation metadata for the 60 source-validated 6F release
# observations using the actual Code 3 row/column structure.
#
# CORE PRINCIPLE
# --------------
# Do NOT infer formulation identity from the whole table.
#
# Instead:
#   1. Locate the exact validated release row.
#   2. Reconstruct the complete row by column.
#   3. Identify actual header/column roles.
#   4. Recover metadata from the appropriate column(s).
#   5. Keep release condition, drug and formulation series separate.
#
# NO:
#   - synthetic observations
#   - interpolation
#   - duplication
#   - cross-article inference
#   - ML
#
# ======================================================================

import os
import re
import numpy as np
import pandas as pd

print("=" * 70)
print("CODE 6K — COLUMN-AWARE STRUCTURAL FORMULATION RECOVERY")
print("=" * 70)


# ======================================================================
# PATHS
# ======================================================================

ROOT = "/content/AI_Polymer_Drug_Release"

F6 = os.path.join(
    ROOT,
    "data",
    "validation",
    "06F_valid_source_release.csv"
)

F3 = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

OUT = os.path.join(
    ROOT,
    "data",
    "validation"
)

os.makedirs(
    OUT,
    exist_ok=True
)

if not os.path.exists(F6):
    raise FileNotFoundError(
        f"Missing 6F file:\n{F6}"
    )

if not os.path.exists(F3):
    raise FileNotFoundError(
        f"Missing Code 3 file:\n{F3}"
    )


# ======================================================================
# LOAD
# ======================================================================

valid = pd.read_csv(
    F6,
    low_memory=False
)

cells = pd.read_csv(
    F3,
    low_memory=False
)

print(
    f"6F observations: {len(valid):,}"
)

print(
    f"Code 3 cells:    {len(cells):,}"
)


# ======================================================================
# REQUIRED COLUMNS
# ======================================================================

required_6f = [
    "observation_id",
    "pmcid",
    "table_index",
    "row_index",
    "time_column",
    "release_column",
    "time_raw",
    "release_raw",
    "release_time_h",
    "release_percent"
]

required_3 = [
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell",
    "row_text",
    "full_table_text",
    "table_caption",
    "header_text",
    "article_title"
]

missing_6f = [
    x for x in required_6f
    if x not in valid.columns
]

missing_3 = [
    x for x in required_3
    if x not in cells.columns
]

if missing_6f:
    raise ValueError(
        "Missing 6F columns:\n"
        + "\n".join(missing_6f)
    )

if missing_3:
    raise ValueError(
        "Missing Code 3 columns:\n"
        + "\n".join(missing_3)
    )


# ======================================================================
# NORMALIZATION
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
        .replace("\xa0", " ")
        .strip()
    )


def lower(x):
    return clean(x).lower()


valid["_pmcid"] = valid["pmcid"].map(
    lambda x: str(x).strip().upper()
)

cells["_pmcid"] = cells["pmcid"].map(
    lambda x: str(x).strip().upper()
)

valid["_table"] = pd.to_numeric(
    valid["table_index"],
    errors="coerce"
)

cells["_table"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

valid["_row"] = pd.to_numeric(
    valid["row_index"],
    errors="coerce"
)

cells["_row"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

cells["_col"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)


# ======================================================================
# VOCABULARIES
# ======================================================================

DRUGS = {
    "doxorubicin": [
        "doxorubicin",
        "dox"
    ],
    "curcumin": [
        "curcumin"
    ],
    "paclitaxel": [
        "paclitaxel",
        "taxol"
    ],
    "dexamethasone": [
        "dexamethasone"
    ],
    "metformin": [
        "metformin"
    ],
    "ibuprofen": [
        "ibuprofen"
    ],
    "5-fluorouracil": [
        "5-fluorouracil",
        "5 fu",
        "5-fu"
    ],
    "gemcitabine": [
        "gemcitabine"
    ],
    "rifampicin": [
        "rifampicin",
        "rifampin"
    ],
    "vancomycin": [
        "vancomycin"
    ],
    "insulin": [
        "insulin"
    ]
}


POLYMERS = {
    "PLGA": [
        "plga"
    ],
    "PLA": [
        "pla",
        "polylactic acid"
    ],
    "PCL": [
        "pcl",
        "polycaprolactone"
    ],
    "PEG": [
        "peg",
        "polyethylene glycol"
    ],
    "PVA": [
        "pva",
        "polyvinyl alcohol"
    ],
    "PEO": [
        "peo",
        "polyethylene oxide"
    ],
    "chitosan": [
        "chitosan"
    ],
    "alginate": [
        "alginate"
    ],
    "gelatin": [
        "gelatin"
    ],
    "cellulose": [
        "cellulose"
    ],
    "poloxamer": [
        "poloxamer"
    ],
    "polyurethane": [
        "polyurethane"
    ]
}


CARRIERS = {
    "nanoparticle": [
        "nanoparticle",
        "nanoparticles"
    ],
    "nanocarrier": [
        "nanocarrier",
        "nanocarriers"
    ],
    "hydrogel": [
        "hydrogel",
        "hydrogels"
    ],
    "microparticle": [
        "microparticle",
        "microparticles"
    ],
    "microsphere": [
        "microsphere",
        "microspheres"
    ],
    "nanofiber": [
        "nanofiber",
        "nanofibers"
    ],
    "liposome": [
        "liposome",
        "liposomes"
    ],
    "micelle": [
        "micelle",
        "micelles"
    ],
    "film": [
        "film",
        "films"
    ],
    "bead": [
        "bead",
        "beads"
    ],
    "implant": [
        "implant",
        "implants"
    ]
}


# ======================================================================
# COLUMN ROLE VOCABULARIES
# ======================================================================

DRUG_HEADER_TERMS = [
    "drug",
    "drug name",
    "drug loaded",
    "drug incorporated",
    "payload",
    "active",
    "active agent",
    "compound",
    "therapeutic",
    "api"
]

POLYMER_HEADER_TERMS = [
    "polymer",
    "polymer type",
    "polymer matrix",
    "material",
    "materials",
    "matrix",
    "composition",
    "polymeric material"
]

CARRIER_HEADER_TERMS = [
    "carrier",
    "delivery system",
    "delivery systems",
    "formulation",
    "formulations",
    "vehicle",
    "nanocarrier",
    "nanoparticle",
    "microparticle",
    "microsphere",
    "hydrogel",
    "film",
    "system",
    "dosage form"
]

CONDITION_HEADER_TERMS = [
    "condition",
    "conditions",
    "ph",
    "medium",
    "media",
    "release medium",
    "release condition",
    "environment",
    "buffer",
    "temperature"
]

TIME_HEADER_TERMS = [
    "time",
    "time (h)",
    "time (hr)",
    "time (hours)",
    "hour",
    "hours",
    "release time"
]

RELEASE_HEADER_TERMS = [
    "release",
    "drug release",
    "cumulative release",
    "release (%)",
    "drug release (%)",
    "cumulative drug release"
]


# ======================================================================
# GENERIC TERM MATCHER
# ======================================================================

def vocabulary_hits(text, vocabulary):

    text = lower(text)

    found = []

    for canonical, terms in vocabulary.items():

        for term in terms:

            if term.lower() in text:

                found.append(canonical)
                break

    return list(dict.fromkeys(found))


def header_score(text, terms):

    text = lower(text)

    score = 0

    for term in terms:

        if term in text:

            score += 1

    return score


# ======================================================================
# BUILD TABLE LOOKUP
# ======================================================================

print(
    "\nBuilding structural table lookup..."
)

table_lookup = {}

for key, group in cells.groupby(
    ["_pmcid", "_table"],
    sort=False
):

    table_lookup[key] = group.copy()

print(
    f"Unique source tables: {len(table_lookup):,}"
)


# ======================================================================
# FIND HEADER-LIKE ROWS
# ======================================================================

def get_table_headers(table):

    if table.empty:
        return {}

    headers = {}

    # Code 3 already contains header_text.
    # Collect unique header text associated with columns.
    for col, group in table.groupby(
        "_col",
        sort=True
    ):

        texts = []

        for x in group["header_text"].dropna():

            x = clean(x)

            if x and x not in texts:

                texts.append(x)

        # Also inspect the first few row_text values because some
        # XML tables encode headers there.
        if not texts:

            for x in group["row_text"].dropna().head(5):

                x = clean(x)

                if x and x not in texts:

                    texts.append(x)

        headers[col] = " | ".join(
            texts
        )

    return headers


# ======================================================================
# COLUMN ROLE CLASSIFICATION
# ======================================================================

def classify_column(header):

    h = lower(header)

    scores = {
        "DRUG": header_score(
            h,
            DRUG_HEADER_TERMS
        ),

        "POLYMER": header_score(
            h,
            POLYMER_HEADER_TERMS
        ),

        "CARRIER": header_score(
            h,
            CARRIER_HEADER_TERMS
        ),

        "CONDITION": header_score(
            h,
            CONDITION_HEADER_TERMS
        ),

        "TIME": header_score(
            h,
            TIME_HEADER_TERMS
        ),

        "RELEASE": header_score(
            h,
            RELEASE_HEADER_TERMS
        )
    }

    best_role = max(
        scores,
        key=scores.get
    )

    best_score = scores[best_role]

    if best_score <= 0:

        return "OTHER", 0

    return best_role, best_score


# ======================================================================
# CONDITION EXTRACTION
# ======================================================================

def extract_condition(text):

    text = clean(text)

    if not text:
        return ""

    found = []

    # pH
    for m in re.finditer(
        r"\bpH\s*[=:]?\s*(\d+(?:\.\d+)?)",
        text,
        flags=re.I
    ):

        found.append(
            f"pH {m.group(1)}"
        )

    # common media
    for term in [
        "PBS",
        "phosphate buffer",
        "simulated gastric fluid",
        "simulated intestinal fluid",
        "SGF",
        "SIF",
        "saline"
    ]:

        if re.search(
            re.escape(term),
            text,
            flags=re.I
        ):

            found.append(term)

    # temperature
    for m in re.finditer(
        r"(\d+(?:\.\d+)?)\s*°?\s*C\b",
        text,
        flags=re.I
    ):

        found.append(
            f"{m.group(1)} C"
        )

    return "; ".join(
        dict.fromkeys(found)
    )


# ======================================================================
# RESOLVE OBSERVATIONS
# ======================================================================

records = []

for n, obs in valid.iterrows():

    pmcid = obs["_pmcid"]
    table_index = obs["_table"]
    row_index = obs["_row"]

    table_key = (
        pmcid,
        table_index
    )

    table = table_lookup.get(
        table_key,
        pd.DataFrame()
    )

    if table.empty:

        records.append({
            "observation_id":
                obs["observation_id"],
            "pmcid":
                obs["pmcid"],
            "table_index":
                obs["table_index"],
            "row_index":
                obs["row_index"],
            "recovery_status":
                "NO_TABLE_FOUND"
        })

        continue

    # --------------------------------------------------------------
    # Exact release row
    # --------------------------------------------------------------

    row = table[
        table["_row"] == row_index
    ].copy()

    exact_row_found = (
        not row.empty
    )

    # --------------------------------------------------------------
    # Headers
    # --------------------------------------------------------------

    headers = get_table_headers(
        table
    )

    column_roles = {}

    for col, header in headers.items():

        role, score = classify_column(
            header
        )

        column_roles[col] = (
            role,
            score,
            header
        )

    # --------------------------------------------------------------
    # Identify validated release/time columns
    # --------------------------------------------------------------

    release_col = pd.to_numeric(
        obs["release_column"],
        errors="coerce"
    )

    time_col = pd.to_numeric(
        obs["time_column"],
        errors="coerce"
    )

    # --------------------------------------------------------------
    # Extract exact row cells
    # --------------------------------------------------------------

    row_values = {}

    if exact_row_found:

        for _, r in row.iterrows():

            col = r["_col"]

            row_values[col] = clean(
                r["raw_cell"]
            )

    # --------------------------------------------------------------
    # Metadata candidates by actual column role
    # --------------------------------------------------------------

    drug_candidates = []
    polymer_candidates = []
    carrier_candidates = []
    condition_candidates = []

    drug_evidence = []
    polymer_evidence = []
    carrier_evidence = []
    condition_evidence = []

    # --------------------------------------------------------------
    # Inspect every actual column in the exact row.
    # --------------------------------------------------------------

    for col, value in row_values.items():

        header = headers.get(
            col,
            ""
        )

        role, score, _ = column_roles.get(
            col,
            ("OTHER", 0, header)
        )

        if not value:
            continue

        # Never interpret the validated time/release columns as
        # formulation metadata.
        if (
            pd.notna(release_col)
            and col == release_col
        ):
            continue

        if (
            pd.notna(time_col)
            and col == time_col
        ):
            continue

        # Drug column
        if role == "DRUG":

            hits = vocabulary_hits(
                value,
                DRUGS
            )

            if hits:

                drug_candidates.extend(
                    hits
                )

                drug_evidence.append(
                    f"COL {int(col)} [{header}]: {value}"
                )

        # Polymer column
        elif role == "POLYMER":

            hits = vocabulary_hits(
                value,
                POLYMERS
            )

            if hits:

                polymer_candidates.extend(
                    hits
                )

                polymer_evidence.append(
                    f"COL {int(col)} [{header}]: {value}"
                )

        # Carrier/formulation column
        elif role == "CARRIER":

            hits = vocabulary_hits(
                value,
                CARRIERS
            )

            if hits:

                carrier_candidates.extend(
                    hits
                )

                carrier_evidence.append(
                    f"COL {int(col)} [{header}]: {value}"
                )

        # Condition column
        elif role == "CONDITION":

            condition = extract_condition(
                value
            )

            if condition:

                condition_candidates.append(
                    condition
                )

                condition_evidence.append(
                    f"COL {int(col)} [{header}]: {value}"
                )

    # --------------------------------------------------------------
    # SECONDARY: inspect header itself for explicit condition
    # --------------------------------------------------------------

    for col, (
        role,
        score,
        header
    ) in column_roles.items():

        if role == "CONDITION":

            c = extract_condition(
                header
            )

            if c:

                condition_candidates.append(
                    c
                )

                condition_evidence.append(
                    f"HEADER COL {int(col)}: {header}"
                )

    # --------------------------------------------------------------
    # De-duplicate
    # --------------------------------------------------------------

    drug_candidates = list(
        dict.fromkeys(
            drug_candidates
        )
    )

    polymer_candidates = list(
        dict.fromkeys(
            polymer_candidates
        )
    )

    carrier_candidates = list(
        dict.fromkeys(
            carrier_candidates
        )
    )

    condition_candidates = list(
        dict.fromkeys(
            condition_candidates
        )
    )

    # --------------------------------------------------------------
    # IMPORTANT:
    #
    # A column-aware hit is stronger than whole-row matching.
    # Multiple values are preserved rather than arbitrarily choosing
    # one.
    # --------------------------------------------------------------

    drug = (
        ";".join(drug_candidates)
        if drug_candidates
        else "NOT_IDENTIFIED"
    )

    polymer = (
        ";".join(polymer_candidates)
        if polymer_candidates
        else "NOT_IDENTIFIED"
    )

    carrier = (
        ";".join(carrier_candidates)
        if carrier_candidates
        else "NOT_IDENTIFIED"
    )

    condition = (
        ";".join(condition_candidates)
        if condition_candidates
        else ""
    )

    # --------------------------------------------------------------
    # Formulation identifier
    #
    # Do NOT include time.
    # Include condition because different pH conditions are distinct
    # release experiments.
    # --------------------------------------------------------------

    components = [
        drug,
        polymer,
        carrier,
        condition
    ]

    components = [
        x for x in components
        if x
        and x != "NOT_IDENTIFIED"
    ]

    if components:

        formulation_id = (
            pmcid
            + "_T"
            + str(int(table_index))
            + "_"
            + "_".join(
                components
            )
        )

        formulation_id = re.sub(
            r"[^A-Za-z0-9_.;-]+",
            "_",
            formulation_id
        )

    else:

        formulation_id = "UNRESOLVED"

    # --------------------------------------------------------------
    # Metadata quality
    # --------------------------------------------------------------

    identified_domains = sum([
        drug != "NOT_IDENTIFIED",
        polymer != "NOT_IDENTIFIED",
        carrier != "NOT_IDENTIFIED",
        bool(condition)
    ])

    if identified_domains >= 3:

        quality = "HIGH"

    elif identified_domains == 2:

        quality = "GOOD"

    elif identified_domains == 1:

        quality = "PARTIAL"

    else:

        quality = "LOW"

    # --------------------------------------------------------------
    # Final status
    #
    # For ML-ready formulation metadata, drug + polymer are mandatory.
    # Carrier is useful but not mandatory if the actual table does not
    # provide one.
    # --------------------------------------------------------------

    if (
        drug != "NOT_IDENTIFIED"
        and polymer != "NOT_IDENTIFIED"
    ):

        status = "READY_FOR_METADATA_REVIEW"

    else:

        status = "MANUAL_REVIEW"

    # --------------------------------------------------------------
    # Store reconstructed row
    # --------------------------------------------------------------

    row_snapshot = " || ".join([
        f"{int(col)}={value}"
        for col, value in sorted(
            row_values.items()
        )
    ])

    header_snapshot = " || ".join([
        f"{int(col)}={header}"
        for col, header in sorted(
            headers.items()
        )
        if header
    ])

    records.append({

        "observation_id":
            obs["observation_id"],

        "pmcid":
            obs["pmcid"],

        "table_index":
            obs["table_index"],

        "row_index":
            obs["row_index"],

        "release_column":
            obs["release_column"],

        "time_column":
            obs["time_column"],

        "release_time_h":
            obs["release_time_h"],

        "release_percent":
            obs["release_percent"],

        "time_raw":
            obs["time_raw"],

        "release_raw":
            obs["release_raw"],

        "drug":
            drug,

        "polymer":
            polymer,

        "carrier":
            carrier,

        "condition":
            condition,

        "drug_evidence":
            " || ".join(
                drug_evidence
            ),

        "polymer_evidence":
            " || ".join(
                polymer_evidence
            ),

        "carrier_evidence":
            " || ".join(
                carrier_evidence
            ),

        "condition_evidence":
            " || ".join(
                condition_evidence
            ),

        "formulation_id":
            formulation_id,

        "metadata_quality":
            quality,

        "recovery_status":
            status,

        "exact_row_found":
            exact_row_found,

        "header_snapshot":
            header_snapshot,

        "row_snapshot":
            row_snapshot,

        "source_trace":
            f"{pmcid}|TABLE={int(table_index)}|ROW={int(row_index)}"
    })

    if (n + 1) % 10 == 0:

        print(
            f"Processed {n+1}/{len(valid)}"
        )


# ======================================================================
# DATAFRAME
# ======================================================================

resolved = pd.DataFrame(
    records
)


# ======================================================================
# EXPERIMENTAL UNIT
#
# Same formulation + same condition.
# Time is NOT part of the unit.
# ======================================================================

resolved["experimental_unit_id"] = (
    resolved["formulation_id"]
    .astype(str)
    + "_UNIT"
)

resolved.loc[
    resolved["formulation_id"]
    == "UNRESOLVED",
    "experimental_unit_id"
] = (
    resolved.loc[
        resolved["formulation_id"]
        == "UNRESOLVED",
        "pmcid"
    ].astype(str)
    + "_T"
    + resolved.loc[
        resolved["formulation_id"]
        == "UNRESOLVED",
        "table_index"
    ].astype(str)
    + "_UNIT_UNRESOLVED"
)


# ======================================================================
# TIMEPOINT COUNT
# ======================================================================

unit_counts = (
    resolved
    .groupby(
        "experimental_unit_id"
    )
    .size()
    .rename(
        "observations_per_unit"
    )
)

resolved = resolved.merge(
    unit_counts,
    left_on="experimental_unit_id",
    right_index=True,
    how="left"
)


# ======================================================================
# SAVE
# ======================================================================

AUDIT = os.path.join(
    OUT,
    "06K_column_aware_formulation_audit.csv"
)

READY = os.path.join(
    OUT,
    "06K_metadata_ready_candidates.csv"
)

MANUAL = os.path.join(
    OUT,
    "06K_metadata_manual_review.csv"
)

resolved.to_csv(
    AUDIT,
    index=False
)

ready = resolved[
    resolved["recovery_status"]
    == "READY_FOR_METADATA_REVIEW"
].copy()

manual = resolved[
    resolved["recovery_status"]
    == "MANUAL_REVIEW"
].copy()

ready.to_csv(
    READY,
    index=False
)

manual.to_csv(
    MANUAL,
    index=False
)


# ======================================================================
# REPORT
# ======================================================================

print("\n" + "=" * 70)
print("CODE 6K COMPLETE")
print("=" * 70)

print(
    f"Input observations:                 {len(valid):,}"
)

print(
    f"Resolved observations:              {len(resolved):,}"
)

print(
    f"Metadata-review candidates:         {len(ready):,}"
)

print(
    f"Manual-review observations:         {len(manual):,}"
)

print("\nMetadata quality:")

print(
    resolved[
        "metadata_quality"
    ].value_counts()
)

print("\nRecovery status:")

print(
    resolved[
        "recovery_status"
    ].value_counts()
)

print("\nDrug identification:")

print(
    resolved[
        "drug"
    ].value_counts()
)

print("\nPolymer identification:")

print(
    resolved[
        "polymer"
    ].value_counts()
)

print("\nCarrier identification:")

print(
    resolved[
        "carrier"
    ].value_counts()
)

print("\nCondition identification:")

condition_known = (
    resolved["condition"]
    .astype(str)
    .str.len()
    > 0
)

print(
    f"Identified: {condition_known.sum():,}"
)

print(
    f"Not identified: {(~condition_known).sum():,}"
)

print(
    "\nUnique formulation IDs: "
    + str(
        resolved[
            resolved["formulation_id"]
            != "UNRESOLVED"
        ]["formulation_id"].nunique()
    )
)

print(
    "Unique experimental units: "
    + str(
        resolved[
            resolved["formulation_id"]
            != "UNRESOLVED"
        ]["experimental_unit_id"].nunique()
    )
)


# ======================================================================
# SAME UNIT / TIMEPOINT DIAGNOSTIC
# ======================================================================

print(
    "\nExperimental units with multiple observations:"
)

multi = (
    resolved[
        resolved["observations_per_unit"] > 1
    ][
        [
            "experimental_unit_id",
            "observations_per_unit"
        ]
    ]
    .drop_duplicates()
    .sort_values(
        "observations_per_unit",
        ascending=False
    )
)

if multi.empty:

    print("None")

else:

    print(
        multi.to_string(
            index=False
        )
    )


# ======================================================================
# CRITICAL DIAGNOSTIC FOR MANUAL REVIEW
# ======================================================================

print(
    "\nManual-review preview:"
)

manual_cols = [
    "observation_id",
    "pmcid",
    "table_index",
    "row_index",
    "release_time_h",
    "release_percent",
    "drug",
    "polymer",
    "carrier",
    "condition",
    "header_snapshot",
    "row_snapshot"
]

print(
    manual[
        manual_cols
    ]
    .head(20)
    .to_string(
        index=False
    )
)


print("\nSaved:")
print(AUDIT)
print(READY)
print(MANUAL)

print("\nIMPORTANT:")
print(
    "Release observations remain exactly those validated by 6F."
)

print(
    "Metadata are recovered from actual row/column structure."
)

print(
    "Whole-table text is NOT used to assign formulation identity."
)

print(
    "Time and release columns are excluded from formulation metadata."
)

print(
    "Different drug/condition columns remain separate."
)

print(
    "Time is NOT included in experimental-unit identity."
)

print(
    "No cross-article metadata inference was performed."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6K-R — RELEASE-HEADER METADATA RECOVERY
# ======================================================================
# Purpose:
# Recover drug and experimental condition directly from the EXACT
# validated release-column header.
#
# Rules:
# - Uses ONLY 6F validated observations.
# - Uses ONLY the exact release-column header.
# - Does NOT use whole-table text.
# - Does NOT infer polymer/carrier unless independently present in the
#   release header.
# - Does NOT alter release_time_h or release_percent.
# - Does NOT merge observations.
# - Does NOT generate/interpolate observations.
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"

INPUT = os.path.join(
    ROOT,
    "data",
    "validation",
    "06F_valid_source_release.csv"
)

OUT_AUDIT = os.path.join(
    ROOT,
    "data",
    "validation",
    "06KR_release_header_metadata_audit.csv"
)

OUT_READY = os.path.join(
    ROOT,
    "data",
    "validation",
    "06KR_metadata_ready_candidates.csv"
)

OUT_MANUAL = os.path.join(
    ROOT,
    "data",
    "validation",
    "06KR_metadata_manual_review.csv"
)

print("=" * 70)
print("CODE 6K-R — RELEASE-HEADER METADATA RECOVERY")
print("=" * 70)

df = pd.read_csv(INPUT)

print("6F observations:", len(df))
print("Columns:", list(df.columns))


# ----------------------------------------------------------------------
# Helpers
# ----------------------------------------------------------------------

def clean_text(x):
    if pd.isna(x):
        return ""
    return re.sub(r"\s+", " ", str(x)).strip()


def normalize_header(x):
    s = clean_text(x)
    s = s.replace("–", "-").replace("—", "-")
    return s


# ----------------------------------------------------------------------
# Drug recognition
# ----------------------------------------------------------------------

DRUG_PATTERNS = {
    "doxorubicin": [
        r"\bdoxorubicin\b",
        r"\bdoxo\b",
        r"\bdox\b",
    ],
    "dexamethasone": [
        r"\bdexamethasone\b",
    ],
    "paclitaxel": [
        r"\bpaclitaxel\b",
    ],
    "curcumin": [
        r"\bcurcumin\b",
    ],
    "metformin": [
        r"\bmetformin\b",
    ],
    "ibuprofen": [
        r"\bibuprofen\b",
    ],
    "5-fluorouracil": [
        r"\b5[- ]fluorouracil\b",
        r"\b5fu\b",
        r"\b5-fu\b",
    ],
    "gemcitabine": [
        r"\bgemcitabine\b",
    ],
    "rifampicin": [
        r"\brifampicin\b",
        r"\brifampin\b",
    ],
    "vancomycin": [
        r"\bvancomycin\b",
    ],
    "insulin": [
        r"\binsulin\b",
    ],
    "dox": [
        r"\bdox\b",
    ],
}


def extract_drug(header):
    s = normalize_header(header).lower()

    hits = []

    # Prefer specific names before generic DOX.
    ordered = [
        "doxorubicin",
        "dexamethasone",
        "paclitaxel",
        "curcumin",
        "metformin",
        "ibuprofen",
        "5-fluorouracil",
        "gemcitabine",
        "rifampicin",
        "vancomycin",
        "insulin",
        "dox",
    ]

    for drug in ordered:
        for pat in DRUG_PATTERNS[drug]:
            if re.search(pat, s):
                hits.append(drug)
                break

    # Remove generic dox if doxorubicin was explicitly identified.
    if "doxorubicin" in hits and "dox" in hits:
        hits.remove("dox")

    hits = list(dict.fromkeys(hits))

    if len(hits) == 1:
        return hits[0], "HEADER_EXACT"

    if len(hits) > 1:
        return ";".join(hits), "HEADER_MULTIPLE"

    return "NOT_IDENTIFIED", "NOT_FOUND"


# ----------------------------------------------------------------------
# Condition recognition
# ----------------------------------------------------------------------

def extract_condition(header):
    s = normalize_header(header)

    conditions = []

    # pH
    ph_matches = re.findall(
        r"\bpH\s*[=:]?\s*([0-9]+(?:\.[0-9]+)?)",
        s,
        flags=re.I
    )

    for ph in ph_matches:
        conditions.append(f"pH {ph}")

    # Temperature
    temp_matches = re.findall(
        r"(?<![A-Za-z])(-?\d+(?:\.\d+)?)\s*°?\s*C\b",
        s,
        flags=re.I
    )

    for temp in temp_matches:
        conditions.append(f"{temp} °C")

    # Common release media
    media_patterns = [
        "PBS",
        "phosphate buffer",
        "phosphate buffered saline",
        "simulated gastric fluid",
        "simulated intestinal fluid",
        "SGF",
        "SIF",
        "water",
        "distilled water",
        "acetate buffer",
    ]

    for media in media_patterns:
        if re.search(r"\b" + re.escape(media) + r"\b", s, flags=re.I):
            conditions.append(media)

    conditions = list(dict.fromkeys(conditions))

    if len(conditions) == 1:
        return conditions[0], "HEADER_EXACT"

    if len(conditions) > 1:
        return ";".join(conditions), "HEADER_MULTIPLE"

    return "NOT_IDENTIFIED", "NOT_FOUND"


# ----------------------------------------------------------------------
# Formulation ID
#
# IMPORTANT:
# Time is NEVER part of the formulation ID.
# Drug + condition are used where available.
# ----------------------------------------------------------------------

def make_formulation_id(row):
    pmcid = clean_text(row["pmcid"])
    table = clean_text(row["table_index"])

    drug = row["header_drug"]
    condition = row["release_condition"]

    parts = [pmcid, f"T{table}"]

    if drug != "NOT_IDENTIFIED":
        parts.append(re.sub(r"[^A-Za-z0-9]+", "_", drug))

    if condition != "NOT_IDENTIFIED":
        parts.append(re.sub(r"[^A-Za-z0-9]+", "_", condition))

    if len(parts) <= 2:
        return f"{pmcid}_T{table}_UNIT_UNRESOLVED"

    return "_".join(parts) + "_UNIT"


# ----------------------------------------------------------------------
# Detect release header
# ----------------------------------------------------------------------

def get_release_header(row):
    """
    Prefer the explicit release_header field created by 6D.
    If unavailable, fall back to release_header_clean.
    """
    for col in [
        "release_header",
        "release_header_clean",
        "release_column"
    ]:
        if col in row.index:
            value = clean_text(row[col])
            if value:
                return value

    return ""


# ----------------------------------------------------------------------
# Process
# ----------------------------------------------------------------------

records = []

for _, row in df.iterrows():

    header = get_release_header(row)

    drug, drug_status = extract_drug(header)
    condition, condition_status = extract_condition(header)

    # Polymer/carrier deliberately remain unresolved here.
    polymer = "NOT_IDENTIFIED"
    carrier = "NOT_IDENTIFIED"

    formulation_status = "MANUAL_REVIEW"

    # Strong metadata if the header identifies a drug and/or condition.
    if drug != "NOT_IDENTIFIED" or condition != "NOT_IDENTIFIED":
        formulation_status = "HEADER_METADATA_RECOVERED"

    rec = row.to_dict()

    rec["release_header_snapshot"] = header
    rec["header_drug"] = drug
    rec["drug_recovery_status"] = drug_status
    rec["release_condition"] = condition
    rec["condition_recovery_status"] = condition_status
    rec["header_polymer"] = polymer
    rec["header_carrier"] = carrier

    rec["formulation_id_header"] = make_formulation_id(rec)

    rec["formulation_metadata_status"] = formulation_status

    records.append(rec)


out = pd.DataFrame(records)


# ----------------------------------------------------------------------
# Experimental-unit proxy
# ----------------------------------------------------------------------

out["experimental_unit_id_header"] = out["formulation_id_header"]


# ----------------------------------------------------------------------
# Quality classification
# ----------------------------------------------------------------------

def metadata_quality(row):

    drug_ok = row["header_drug"] != "NOT_IDENTIFIED"
    cond_ok = row["release_condition"] != "NOT_IDENTIFIED"

    if drug_ok and cond_ok:
        return "HIGH"

    if drug_ok or cond_ok:
        return "GOOD"

    return "LOW"


out["metadata_quality"] = out.apply(metadata_quality, axis=1)


# ----------------------------------------------------------------------
# Ready-for-next-stage criterion
#
# We do NOT require polymer/carrier because many release tables do not
# encode those in the release header.
# ----------------------------------------------------------------------

out["ready_for_metadata_validation"] = (
    out["header_drug"].ne("NOT_IDENTIFIED")
    |
    out["release_condition"].ne("NOT_IDENTIFIED")
)


# ----------------------------------------------------------------------
# Diagnostics
# ----------------------------------------------------------------------

print()
print("======================================================================")
print("CODE 6K-R COMPLETE")
print("======================================================================")

print("Input observations:", len(out))

print()
print("Drug recovery:")
print(out["header_drug"].value_counts(dropna=False))

print()
print("Condition recovery:")
print(out["release_condition"].value_counts(dropna=False))

print()
print("Metadata quality:")
print(out["metadata_quality"].value_counts(dropna=False))

print()
print("Formulation metadata status:")
print(out["formulation_metadata_status"].value_counts(dropna=False))

print()
print(
    "Metadata-ready candidates:",
    int(out["ready_for_metadata_validation"].sum())
)

print(
    "Remaining unresolved:",
    int((~out["ready_for_metadata_validation"]).sum())
)

print()
print("Unique formulation IDs:",
      out["formulation_id_header"].nunique())

print("Unique experimental units:",
      out["experimental_unit_id_header"].nunique())


# ----------------------------------------------------------------------
# Important structural check:
# same PMCID + table + time should remain separate when release headers
# identify different drugs/conditions.
# ----------------------------------------------------------------------

print()
print("Header-separated series check:")

series_cols = [
    "pmcid",
    "table_index",
    "release_header_snapshot"
]

series_counts = (
    out.groupby(series_cols, dropna=False)
       .size()
       .reset_index(name="observations")
)

print(
    "Unique PMCID/table/header series:",
    len(series_counts)
)


# ----------------------------------------------------------------------
# Save
# ----------------------------------------------------------------------

out.to_csv(OUT_AUDIT, index=False)

ready = out[
    out["ready_for_metadata_validation"]
].copy()

manual = out[
    ~out["ready_for_metadata_validation"]
].copy()

ready.to_csv(OUT_READY, index=False)
manual.to_csv(OUT_MANUAL, index=False)

print()
print("Saved:")
print(OUT_AUDIT)
print(OUT_READY)
print(OUT_MANUAL)

print()
print("IMPORTANT:")
print("Release observations remain exactly those validated by 6F.")
print("Drug/condition metadata are recovered only from the exact release header.")
print("Polymer and carrier are NOT inferred from unrelated table text.")
print("Time is NOT included in formulation identity.")
print("Different drug/condition release columns remain separate.")
print("No observations were generated.")
print("No interpolation was performed.")
print("No observations were duplicated.")
print("No machine learning was performed.")
print("=" * 70)
```

```python
# ======================================================================
# CODE 6L — CONSERVATIVE METADATA CONSOLIDATION
# ======================================================================

import os
import pandas as pd
import numpy as np
import re

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")

F6G = os.path.join(VAL, "06G_formulation_unit_audit.csv")
F6H = os.path.join(VAL, "06H_targeted_metadata_audit.csv")
FKR = os.path.join(VAL, "06KR_release_header_metadata_audit.csv")

OUT_AUDIT = os.path.join(
    VAL, "06L_conservative_metadata_consolidation_audit.csv"
)

OUT_READY = os.path.join(
    VAL, "06L_final_metadata_ready_candidates.csv"
)

OUT_MANUAL = os.path.join(
    VAL, "06L_final_metadata_manual_review.csv"
)


print("=" * 70)
print("CODE 6L — CONSERVATIVE METADATA CONSOLIDATION")
print("=" * 70)


# ----------------------------------------------------------------------
# LOAD
# ----------------------------------------------------------------------

g = pd.read_csv(F6G)
h = pd.read_csv(F6H)
k = pd.read_csv(FKR)

print("6G observations:", len(g))
print("6H observations:", len(h))
print("6K-R observations:", len(k))


# ----------------------------------------------------------------------
# NORMALIZATION HELPERS
# ----------------------------------------------------------------------

def clean(x):
    if pd.isna(x):
        return ""
    return re.sub(r"\s+", " ", str(x)).strip()


def norm(x):
    return clean(x).lower()


def first_existing(df, names):
    for c in names:
        if c in df.columns:
            return c
    return None


# ----------------------------------------------------------------------
# IDENTIFY COMMON OBSERVATION KEY
#
# 6G/6H may use observation_id or another inherited identifier.
# 6K-R definitely carries observation_id from 6F.
# ----------------------------------------------------------------------

common_key = None

for candidate in [
    "observation_id",
    "candidate_id",
]:
    if (
        candidate in g.columns
        and candidate in h.columns
        and candidate in k.columns
    ):
        common_key = candidate
        break

if common_key is None:
    raise ValueError(
        "Could not find a common observation identifier across 6G, 6H and 6K-R."
    )

print("Common observation key:", common_key)


# ----------------------------------------------------------------------
# PREVENT MANY-TO-MANY MERGES
# ----------------------------------------------------------------------

for name, df in [("6G", g), ("6H", h), ("6K-R", k)]:

    dup = df[common_key].duplicated().sum()

    print(
        f"{name} duplicate observation IDs:",
        int(dup)
    )

    if dup > 0:
        raise ValueError(
            f"{name} contains duplicated {common_key} values. "
            "Consolidation stopped to prevent many-to-many merging."
        )


# ----------------------------------------------------------------------
# KEEP ONLY METADATA COLUMNS FROM EACH BRANCH
# ----------------------------------------------------------------------

# 6G
g_cols = [
    common_key,
    "pmcid",
    "table_index",
    "row_index",
    "drug",
    "polymer",
    "carrier",
    "formulation_id",
    "experimental_unit_id",
    "formulation_context_quality",
]

g_cols = [c for c in g_cols if c in g.columns]

g_meta = g[g_cols].copy()


# 6H
h_possible = [
    common_key,
    "drug",
    "polymer",
    "carrier",
    "recovered_drug",
    "recovered_polymer",
    "recovered_carrier",
    "recovered_formulation_id",
    "6H_status",
    "metadata_quality",
]

h_cols = [c for c in h_possible if c in h.columns]

h_meta = h[h_cols].copy()


# 6K-R
k_possible = [
    common_key,
    "pmcid",
    "table_index",
    "row_index",
    "release_header_snapshot",
    "header_drug",
    "drug_recovery_status",
    "release_condition",
    "condition_recovery_status",
    "header_polymer",
    "header_carrier",
    "formulation_id_header",
    "experimental_unit_id_header",
    "metadata_quality",
]

k_cols = [c for c in k_possible if c in k.columns]

k_meta = k[k_cols].copy()


# ----------------------------------------------------------------------
# SAFE MERGE
# ----------------------------------------------------------------------

resolved = k_meta.merge(
    g_meta.drop(
        columns=[
            c for c in ["pmcid", "table_index", "row_index"]
            if c in g_meta.columns
        ]
    ),
    on=common_key,
    how="left",
    suffixes=("", "_6G")
)

resolved = resolved.merge(
    h_meta.drop(
        columns=[
            c for c in [
                "drug",
                "polymer",
                "carrier"
            ]
            if c in h_meta.columns
        ]
    ),
    on=common_key,
    how="left",
    suffixes=("", "_6H")
)


# ----------------------------------------------------------------------
# VALUE VALIDITY
# ----------------------------------------------------------------------

def valid_value(x):
    s = norm(x)

    return (
        s != ""
        and s not in [
            "not_identified",
            "nan",
            "none",
            "unresolved",
            "unknown",
        ]
    )


# ----------------------------------------------------------------------
# CONSERVATIVE EVIDENCE SELECTION
#
# Priority:
# 1. 6K exact header
# 2. 6G exact structural/context evidence
# 3. 6H targeted same-table recovery
# ----------------------------------------------------------------------

def choose_metadata(row, fields):
    """
    Return:
      selected value
      evidence source
    """

    for value_col, source in fields:

        if value_col in row.index:

            value = row[value_col]

            if valid_value(value):

                return clean(value), source

    return "NOT_IDENTIFIED", "UNRESOLVED"


# ----------------------------------------------------------------------
# DRUG
# ----------------------------------------------------------------------

drug_priority = [
    ("header_drug", "6K_EXACT_RELEASE_HEADER"),
]

for c in [
    "drug",
    "drug_6G",
    "recovered_drug",
]:

    if c in resolved.columns:
        drug_priority.append(
            (c, "SAME_TABLE_STRUCTURAL_RECOVERY")
        )

if "drug_6H" in resolved.columns:
    drug_priority.append(
        ("drug_6H", "6H_TARGETED_RECOVERY")
    )

resolved[
    "final_drug",
    "drug_evidence"
] = zip(*[
    choose_metadata(row, drug_priority)
    for _, row in resolved.iterrows()
])


# ----------------------------------------------------------------------
# POLYMER
# ----------------------------------------------------------------------

polymer_priority = []

# 6K intentionally does not infer polymer.
# Therefore polymer begins with 6G/6H structural evidence.

for c in [
    "polymer",
    "polymer_6G",
    "recovered_polymer",
    "polymer_6H",
]:

    if c in resolved.columns:

        source = (
            "SAME_TABLE_STRUCTURAL_RECOVERY"
            if "6G" in c or c == "polymer"
            else "6H_TARGETED_RECOVERY"
        )

        polymer_priority.append(
            (c, source)
        )

resolved[
    "final_polymer",
    "polymer_evidence"
] = zip(*[
    choose_metadata(row, polymer_priority)
    for _, row in resolved.iterrows()
])


# ----------------------------------------------------------------------
# CARRIER
# ----------------------------------------------------------------------

carrier_priority = []

for c in [
    "carrier",
    "carrier_6G",
    "recovered_carrier",
    "carrier_6H",
]:

    if c in resolved.columns:

        source = (
            "SAME_TABLE_STRUCTURAL_RECOVERY"
            if "6G" in c or c == "carrier"
            else "6H_TARGETED_RECOVERY"
        )

        carrier_priority.append(
            (c, source)
        )

resolved[
    "final_carrier",
    "carrier_evidence"
] = zip(*[
    choose_metadata(row, carrier_priority)
    for _, row in resolved.iterrows()
])


# ----------------------------------------------------------------------
# RELEASE CONDITION
#
# Only 6K exact header is accepted here.
# ----------------------------------------------------------------------

resolved["final_release_condition"] = resolved.get(
    "release_condition",
    "NOT_IDENTIFIED"
)

resolved["condition_evidence"] = np.where(
    resolved["final_release_condition"].apply(valid_value),
    "6K_EXACT_RELEASE_HEADER",
    "UNRESOLVED"
)


# ----------------------------------------------------------------------
# FORMULATION ID
#
# IMPORTANT:
# Do NOT use time.
# Do NOT combine different release conditions.
# ----------------------------------------------------------------------

def make_final_formulation_id(row):

    pmcid = clean(row.get("pmcid", ""))
    table = clean(row.get("table_index", ""))

    drug = clean(row["final_drug"])
    polymer = clean(row["final_polymer"])
    carrier = clean(row["final_carrier"])
    condition = clean(row["final_release_condition"])

    parts = [pmcid, f"T{table}"]

    if valid_value(drug):
        parts.append(
            re.sub(r"[^A-Za-z0-9]+", "_", drug)
        )

    if valid_value(polymer):
        parts.append(
            re.sub(r"[^A-Za-z0-9]+", "_", polymer)
        )

    if valid_value(carrier):
        parts.append(
            re.sub(r"[^A-Za-z0-9]+", "_", carrier)
        )

    if valid_value(condition):
        parts.append(
            re.sub(r"[^A-Za-z0-9]+", "_", condition)
        )

    if len(parts) <= 2:
        return f"{pmcid}_T{table}_UNRESOLVED"

    return "_".join(parts)


resolved["final_formulation_id"] = resolved.apply(
    make_final_formulation_id,
    axis=1
)


# ----------------------------------------------------------------------
# FINAL METADATA QUALITY
# ----------------------------------------------------------------------

def quality(row):

    drug = valid_value(row["final_drug"])
    polymer = valid_value(row["final_polymer"])
    carrier = valid_value(row["final_carrier"])
    condition = valid_value(row["final_release_condition"])

    # High:
    # drug + polymer + carrier + condition
    if drug and polymer and carrier and condition:
        return "HIGH"

    # Good:
    # drug + at least one formulation descriptor
    if drug and (polymer or carrier):
        return "GOOD"

    # Moderate:
    # exact-header drug and/or condition
    if drug or condition:
        return "MODERATE"

    return "LOW"


resolved["final_metadata_quality"] = resolved.apply(
    quality,
    axis=1
)


# ----------------------------------------------------------------------
# FINAL VALIDATION STATUS
#
# We do NOT require every metadata field.
#
# Ready:
#   exact source-validated release observation AND
#   at least one strong metadata anchor.
#
# Manual:
#   no reliable metadata anchor.
# ----------------------------------------------------------------------

resolved["final_metadata_status"] = np.where(
    (
        resolved["final_drug"].apply(valid_value)
        |
        resolved["final_polymer"].apply(valid_value)
        |
        resolved["final_carrier"].apply(valid_value)
        |
        resolved["final_release_condition"].apply(valid_value)
    ),
    "READY_METADATA_VALIDATION",
    "MANUAL_METADATA_REVIEW"
)


# ----------------------------------------------------------------------
# STRUCTURAL SAFETY CHECK
#
# No loss of 6F observations.
# ----------------------------------------------------------------------

if len(resolved) != len(k):
    raise ValueError(
        "Observation count changed during consolidation. "
        "STOP: possible merge problem."
    )


# ----------------------------------------------------------------------
# DUPLICATION CHECK
# ----------------------------------------------------------------------

if resolved[common_key].duplicated().any():
    raise ValueError(
        "Duplicate observation IDs detected after consolidation."
    )


# ----------------------------------------------------------------------
# REPORT
# ----------------------------------------------------------------------

ready = resolved[
    resolved["final_metadata_status"]
    == "READY_METADATA_VALIDATION"
].copy()

manual = resolved[
    resolved["final_metadata_status"]
    == "MANUAL_METADATA_REVIEW"
].copy()


print()
print("=" * 70)
print("CODE 6L COMPLETE")
print("=" * 70)

print("Input 6K-R observations:", len(k))
print("Consolidated observations:", len(resolved))

print()
print("Final metadata status:")
print(
    resolved["final_metadata_status"].value_counts()
)

print()
print("Final metadata quality:")
print(
    resolved["final_metadata_quality"].value_counts()
)

print()
print("Final drug:")
print(
    resolved["final_drug"].value_counts()
)

print()
print("Final polymer:")
print(
    resolved["final_polymer"].value_counts()
)

print()
print("Final carrier:")
print(
    resolved["final_carrier"].value_counts()
)

print()
print("Final release conditions:")
print(
    resolved["final_release_condition"].value_counts()
)

print()
print("Unique final formulation IDs:",
      resolved["final_formulation_id"].nunique())

print()
print("Ready metadata validation:",
      len(ready))

print("Manual metadata review:",
      len(manual))

print()
print("Observation-count preservation:")
print("6K-R:", len(k))
print("6L  :", len(resolved))

print()
print("Duplicate observation IDs:",
      int(resolved[common_key].duplicated().sum()))


# ----------------------------------------------------------------------
# SAVE
# ----------------------------------------------------------------------

resolved.to_csv(
    OUT_AUDIT,
    index=False
)

ready.to_csv(
    OUT_READY,
    index=False
)

manual.to_csv(
    OUT_MANUAL,
    index=False
)

print()
print("Saved:")
print(OUT_AUDIT)
print(OUT_READY)
print(OUT_MANUAL)

print()
print("IMPORTANT:")
print("6F source-validated release values are unchanged.")
print("6K exact-header metadata has highest priority.")
print("Polymer/carrier are never inferred from unrelated article text.")
print("Different release conditions remain separate.")
print("Time is NOT part of formulation identity.")
print("No observations were generated.")
print("No interpolation was performed.")
print("No observations were duplicated.")
print("No machine learning was performed.")

print("=" * 70)
```

```python
# ======================================================================
# CODE 6L — CONSERVATIVE METADATA CONSOLIDATION — FIXED
# ======================================================================

import os
import pandas as pd
import numpy as np
import re

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")

F6G = os.path.join(
    VAL, "06G_formulation_unit_audit.csv"
)

F6H = os.path.join(
    VAL, "06H_targeted_metadata_audit.csv"
)

FKR = os.path.join(
    VAL, "06KR_release_header_metadata_audit.csv"
)

OUT_AUDIT = os.path.join(
    VAL, "06L_conservative_metadata_consolidation_audit.csv"
)

OUT_READY = os.path.join(
    VAL, "06L_final_metadata_ready_candidates.csv"
)

OUT_MANUAL = os.path.join(
    VAL, "06L_final_metadata_manual_review.csv"
)


print("=" * 70)
print("CODE 6L — CONSERVATIVE METADATA CONSOLIDATION — FIXED")
print("=" * 70)


# ======================================================================
# 1. LOAD DATA
# ======================================================================

g = pd.read_csv(F6G)
h = pd.read_csv(F6H)
k = pd.read_csv(FKR)

print("6G observations:", len(g))
print("6H observations:", len(h))
print("6K-R observations:", len(k))


# ======================================================================
# 2. COMMON OBSERVATION KEY
# ======================================================================

common_key = None

for candidate in [
    "observation_id",
    "candidate_id",
]:

    if (
        candidate in g.columns
        and candidate in h.columns
        and candidate in k.columns
    ):
        common_key = candidate
        break


if common_key is None:
    raise ValueError(
        "Could not find a common observation identifier "
        "across 6G, 6H and 6K-R."
    )


print("Common observation key:", common_key)


# ======================================================================
# 3. CHECK DUPLICATES
# ======================================================================

for name, df in [
    ("6G", g),
    ("6H", h),
    ("6K-R", k)
]:

    dup = int(df[common_key].duplicated().sum())

    print(
        f"{name} duplicate observation IDs:",
        dup
    )

    if dup > 0:
        raise ValueError(
            f"{name} contains duplicate {common_key} values. "
            "STOPPING to prevent many-to-many merging."
        )


# ======================================================================
# 4. NORMALIZATION HELPERS
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


def norm(x):
    return clean(x).lower()


def valid_value(x):

    s = norm(x)

    return (
        s != ""
        and s not in [
            "not_identified",
            "nan",
            "none",
            "unresolved",
            "unknown",
            "manual_review",
        ]
    )


# ======================================================================
# 5. SELECT RELEVANT COLUMNS
# ======================================================================

# -------------------------
# 6G
# -------------------------

g_cols = [
    common_key,
    "pmcid",
    "table_index",
    "row_index",
    "drug",
    "polymer",
    "carrier",
    "formulation_id",
    "experimental_unit_id",
    "formulation_context_quality",
]

g_cols = [
    c for c in g_cols
    if c in g.columns
]

g_meta = g[g_cols].copy()


# -------------------------
# 6H
# -------------------------

h_possible = [
    common_key,
    "drug",
    "polymer",
    "carrier",
    "recovered_drug",
    "recovered_polymer",
    "recovered_carrier",
    "recovered_formulation_id",
    "6H_status",
    "metadata_quality",
]

h_cols = [
    c for c in h_possible
    if c in h.columns
]

h_meta = h[h_cols].copy()


# -------------------------
# 6K-R
# -------------------------

k_possible = [
    common_key,
    "pmcid",
    "table_index",
    "row_index",
    "release_header_snapshot",
    "header_drug",
    "drug_recovery_status",
    "release_condition",
    "condition_recovery_status",
    "header_polymer",
    "header_carrier",
    "formulation_id_header",
    "experimental_unit_id_header",
    "metadata_quality",
]

k_cols = [
    c for c in k_possible
    if c in k.columns
]

k_meta = k[k_cols].copy()


# ======================================================================
# 6. SAFE MERGE
#
# IMPORTANT:
# Start from 6K-R because it contains all 60 source-validated
# observations.
# ======================================================================

resolved = k_meta.copy()


# -------------------------
# Merge 6G
# -------------------------

g_drop = [
    c for c in [
        "pmcid",
        "table_index",
        "row_index"
    ]
    if c in g_meta.columns
]

g_merge = g_meta.drop(
    columns=g_drop,
    errors="ignore"
)

resolved = resolved.merge(
    g_merge,
    on=common_key,
    how="left",
    suffixes=("", "_6G"),
    validate="one_to_one"
)


# -------------------------
# Merge 6H
# -------------------------

h_drop = [
    c for c in [
        "drug",
        "polymer",
        "carrier"
    ]
    if c in h_meta.columns
]

h_merge = h_meta.drop(
    columns=h_drop,
    errors="ignore"
)

resolved = resolved.merge(
    h_merge,
    on=common_key,
    how="left",
    suffixes=("", "_6H"),
    validate="one_to_one"
)


print()
print("Rows after safe consolidation:", len(resolved))


if len(resolved) != len(k):

    raise ValueError(
        "CRITICAL: Observation count changed during consolidation."
    )


# ======================================================================
# 7. SAFE METADATA SELECTION FUNCTION
# ======================================================================

def choose_metadata(row, priority):

    for value_col, source in priority:

        if value_col not in row.index:
            continue

        value = row[value_col]

        if valid_value(value):

            return (
                clean(value),
                source
            )

    return (
        "NOT_IDENTIFIED",
        "UNRESOLVED"
    )


# ======================================================================
# 8. DRUG RESOLUTION
#
# Priority:
# 1. 6K-R exact release-header drug
# 2. 6G same-table structural metadata
# 3. 6H targeted recovery
# ======================================================================

drug_priority = []

if "header_drug" in resolved.columns:

    drug_priority.append(
        (
            "header_drug",
            "6K_EXACT_RELEASE_HEADER"
        )
    )


for c in [
    "drug",
    "drug_6G",
]:

    if c in resolved.columns:

        drug_priority.append(
            (
                c,
                "6G_STRUCTURAL_RECOVERY"
            )
        )


for c in [
    "recovered_drug",
    "drug_6H",
]:

    if c in resolved.columns:

        drug_priority.append(
            (
                c,
                "6H_TARGETED_RECOVERY"
            )
        )


drug_results = [
    choose_metadata(
        row,
        drug_priority
    )
    for _, row in resolved.iterrows()
]


# FIXED — no zip assignment
resolved["final_drug"] = [
    x[0]
    for x in drug_results
]

resolved["drug_evidence"] = [
    x[1]
    for x in drug_results
]


# ======================================================================
# 9. POLYMER RESOLUTION
#
# Polymer is NOT inferred from the release header.
# ======================================================================

polymer_priority = []

for c in [
    "polymer",
    "polymer_6G",
]:

    if c in resolved.columns:

        polymer_priority.append(
            (
                c,
                "6G_STRUCTURAL_RECOVERY"
            )
        )


for c in [
    "recovered_polymer",
    "polymer_6H",
]:

    if c in resolved.columns:

        polymer_priority.append(
            (
                c,
                "6H_TARGETED_RECOVERY"
            )
        )


polymer_results = [
    choose_metadata(
        row,
        polymer_priority
    )
    for _, row in resolved.iterrows()
]


# FIXED — no zip assignment
resolved["final_polymer"] = [
    x[0]
    for x in polymer_results
]

resolved["polymer_evidence"] = [
    x[1]
    for x in polymer_results
]


# ======================================================================
# 10. CARRIER RESOLUTION
# ======================================================================

carrier_priority = []

for c in [
    "carrier",
    "carrier_6G",
]:

    if c in resolved.columns:

        carrier_priority.append(
            (
                c,
                "6G_STRUCTURAL_RECOVERY"
            )
        )


for c in [
    "recovered_carrier",
    "carrier_6H",
]:

    if c in resolved.columns:

        carrier_priority.append(
            (
                c,
                "6H_TARGETED_RECOVERY"
            )
        )


carrier_results = [
    choose_metadata(
        row,
        carrier_priority
    )
    for _, row in resolved.iterrows()
]


# FIXED — no zip assignment
resolved["final_carrier"] = [
    x[0]
    for x in carrier_results
]

resolved["carrier_evidence"] = [
    x[1]
    for x in carrier_results
]


# ======================================================================
# 11. RELEASE CONDITION
#
# Only exact 6K-R header evidence is used.
# ======================================================================

if "release_condition" in resolved.columns:

    resolved["final_release_condition"] = (
        resolved["release_condition"]
        .apply(clean)
    )

else:

    resolved["final_release_condition"] = (
        "NOT_IDENTIFIED"
    )


resolved["condition_evidence"] = np.where(
    resolved["final_release_condition"].apply(
        valid_value
    ),
    "6K_EXACT_RELEASE_HEADER",
    "UNRESOLVED"
)


# ======================================================================
# 12. FORMULATION ID
#
# IMPORTANT:
# Time is NOT used as formulation identity.
# Release conditions remain separate.
# ======================================================================

def make_formulation_id(row):

    pmcid = clean(
        row.get("pmcid", "")
    )

    table = clean(
        row.get("table_index", "")
    )

    drug = clean(
        row["final_drug"]
    )

    polymer = clean(
        row["final_polymer"]
    )

    carrier = clean(
        row["final_carrier"]
    )

    condition = clean(
        row["final_release_condition"]
    )


    parts = [
        pmcid,
        f"T{table}"
    ]


    if valid_value(drug):

        parts.append(
            re.sub(
                r"[^A-Za-z0-9]+",
                "_",
                drug
            )
        )


    if valid_value(polymer):

        parts.append(
            re.sub(
                r"[^A-Za-z0-9]+",
                "_",
                polymer
            )
        )


    if valid_value(carrier):

        parts.append(
            re.sub(
                r"[^A-Za-z0-9]+",
                "_",
                carrier
            )
        )


    if valid_value(condition):

        parts.append(
            re.sub(
                r"[^A-Za-z0-9]+",
                "_",
                condition
            )
        )


    if len(parts) <= 2:

        return (
            f"{pmcid}_T{table}_UNRESOLVED"
        )


    return "_".join(parts)


resolved["final_formulation_id"] = (
    resolved.apply(
        make_formulation_id,
        axis=1
    )
)


# ======================================================================
# 13. METADATA QUALITY
# ======================================================================

def metadata_quality(row):

    drug = valid_value(
        row["final_drug"]
    )

    polymer = valid_value(
        row["final_polymer"]
    )

    carrier = valid_value(
        row["final_carrier"]
    )

    condition = valid_value(
        row["final_release_condition"]
    )


    # Highest quality
    if (
        drug
        and polymer
        and carrier
        and condition
    ):

        return "HIGH"


    # Good
    if (
        drug
        and (
            polymer
            or carrier
        )
    ):

        return "GOOD"


    # Moderate
    if (
        drug
        or condition
    ):

        return "MODERATE"


    return "LOW"


resolved["final_metadata_quality"] = (
    resolved.apply(
        metadata_quality,
        axis=1
    )
)


# ======================================================================
# 14. FINAL METADATA STATUS
#
# IMPORTANT:
# This does NOT mean the observation itself is unvalidated.
# All 60 observations remain source validated from 6F.
# This status only describes metadata readiness.
# ======================================================================

resolved["final_metadata_status"] = np.where(

    (
        resolved["final_drug"].apply(
            valid_value
        )

        |

        resolved["final_polymer"].apply(
            valid_value
        )

        |

        resolved["final_carrier"].apply(
            valid_value
        )

        |

        resolved[
            "final_release_condition"
        ].apply(
            valid_value
        )
    ),

    "READY_METADATA_VALIDATION",

    "MANUAL_METADATA_REVIEW"
)


# ======================================================================
# 15. FINAL DATASETS
# ======================================================================

ready = resolved[
    resolved["final_metadata_status"]
    == "READY_METADATA_VALIDATION"
].copy()


manual = resolved[
    resolved["final_metadata_status"]
    == "MANUAL_METADATA_REVIEW"
].copy()


# ======================================================================
# 16. SAFETY CHECKS
# ======================================================================

if len(resolved) != 60:

    raise ValueError(
        f"Expected 60 observations, found {len(resolved)}."
    )


if resolved[common_key].duplicated().any():

    raise ValueError(
        "Duplicate observation IDs detected after consolidation."
    )


# ======================================================================
# 17. REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6L COMPLETE")
print("=" * 70)

print(
    "Input 6K-R observations:",
    len(k)
)

print(
    "Consolidated observations:",
    len(resolved)
)

print()
print("Final metadata status:")

print(
    resolved[
        "final_metadata_status"
    ].value_counts()
)


print()
print("Final metadata quality:")

print(
    resolved[
        "final_metadata_quality"
    ].value_counts()
)


print()
print("Final drug:")

print(
    resolved[
        "final_drug"
    ].value_counts()
)


print()
print("Final polymer:")

print(
    resolved[
        "final_polymer"
    ].value_counts()
)


print()
print("Final carrier:")

print(
    resolved[
        "final_carrier"
    ].value_counts()
)


print()
print("Final release conditions:")

print(
    resolved[
        "final_release_condition"
    ].value_counts()
)


print()
print(
    "Unique final formulation IDs:",
    resolved[
        "final_formulation_id"
    ].nunique()
)


print()
print(
    "Ready for metadata validation:",
    len(ready)
)

print(
    "Manual metadata review:",
    len(manual)
)


print()
print("Observation-count preservation:")

print(
    "6K-R:",
    len(k)
)

print(
    "6L  :",
    len(resolved)
)


print()
print(
    "Duplicate observation IDs:",
    int(
        resolved[
            common_key
        ].duplicated().sum()
    )
)


# ======================================================================
# 18. SAVE
# ======================================================================

resolved.to_csv(
    OUT_AUDIT,
    index=False
)

ready.to_csv(
    OUT_READY,
    index=False
)

manual.to_csv(
    OUT_MANUAL,
    index=False
)


print()
print("Saved:")

print(
    OUT_AUDIT
)

print(
    OUT_READY
)

print(
    OUT_MANUAL
)


# ======================================================================
# 19. DATA-INTEGRITY STATEMENT
# ======================================================================

print()
print("IMPORTANT:")
print(
    "6F source-validated release values are unchanged."
)

print(
    "6K-R exact release-header metadata has highest priority."
)

print(
    "Polymer and carrier are NOT inferred from unrelated article text."
)

print(
    "Different release conditions remain separate."
)

print(
    "Time is NOT part of formulation identity."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6M — TARGETED SAME-TABLE FORMULATION METADATA RECOVERY
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")
RAW = os.path.join(ROOT, "data", "raw", "tables")

F6L = os.path.join(
    VAL,
    "06L_conservative_metadata_consolidation_audit.csv"
)

F6G = os.path.join(
    VAL,
    "06G_formulation_unit_audit.csv"
)

F6K = os.path.join(
    VAL,
    "06KR_release_header_metadata_audit.csv"
)

F_CODE3 = os.path.join(
    RAW,
    "03_extracted_table_cells.csv"
)

OUT_AUDIT = os.path.join(
    VAL,
    "06M_targeted_formulation_recovery_audit.csv"
)

OUT_RECOVERED = os.path.join(
    VAL,
    "06M_recovered_metadata_candidates.csv"
)

OUT_MANUAL = os.path.join(
    VAL,
    "06M_remaining_manual_review.csv"
)


print("=" * 70)
print("CODE 6M — TARGETED SAME-TABLE FORMULATION METADATA RECOVERY")
print("=" * 70)


# ======================================================================
# 1. LOAD
# ======================================================================

df = pd.read_csv(F6L)
g = pd.read_csv(F6G)
k = pd.read_csv(F6K)

print("6L observations:", len(df))
print("6G observations:", len(g))
print("6K-R observations:", len(k))


# ======================================================================
# 2. IDENTIFY UNRESOLVED OBSERVATIONS
# ======================================================================

manual_mask = (
    df["final_metadata_status"]
    == "MANUAL_METADATA_REVIEW"
)

manual = df.loc[
    manual_mask
].copy()

print(
    "Observations requiring targeted recovery:",
    len(manual)
)


# ======================================================================
# 3. LOAD CODE 3 STRUCTURAL TABLE CELLS
# ======================================================================

code3 = pd.read_csv(F_CODE3)

required = [
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell",
    "row_text",
    "full_table_text",
    "header_text",
]

missing = [
    c for c in required
    if c not in code3.columns
]

if missing:
    raise ValueError(
        "Code 3 is missing required columns:\n"
        + "\n".join(missing)
    )


print(
    "Code 3 cell records:",
    f"{len(code3):,}"
)


# ======================================================================
# 4. NORMALIZATION
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


def norm(x):

    return clean(x).lower()


def valid(x):

    s = norm(x)

    return (
        s != ""
        and s not in [
            "not_identified",
            "unresolved",
            "unknown",
            "none",
            "nan",
            "manual_review"
        ]
    )


# ======================================================================
# 5. VOCABULARIES
#
# These are used ONLY for explicit lexical evidence in the same table.
# They do not create inferred metadata.
# ======================================================================

DRUGS = [
    "doxorubicin",
    "dox",
    "dexamethasone",
    "curcumin",
    "paclitaxel",
    "metformin",
    "ibuprofen",
    "5-fluorouracil",
    "5 fluorouracil",
    "5-fu",
    "gemcitabine",
    "rifampicin",
    "vancomycin",
    "insulin",
]


POLYMERS = [
    "plga",
    "pla",
    "pcl",
    "peg",
    "peo",
    "pva",
    "pvp",
    "chitosan",
    "alginate",
    "cellulose",
    "gelatin",
    "polyurethane",
    "poloxamer",
]


CARRIERS = [
    "nanoparticle",
    "nanoparticles",
    "microparticle",
    "microparticles",
    "microsphere",
    "microspheres",
    "hydrogel",
    "hydrogels",
    "nanofiber",
    "nanofibers",
    "film",
    "films",
    "liposome",
    "liposomes",
    "bead",
    "beads",
    "implant",
    "implants",
    "micelle",
    "micelles",
]


# ======================================================================
# 6. LEXICAL EXTRACTION
# ======================================================================

def find_terms(text, vocabulary):

    text_n = norm(text)

    found = []

    for term in vocabulary:

        if term in text_n:

            found.append(term)

    return found


def normalize_terms(values):

    if not values:
        return "NOT_IDENTIFIED"

    # preserve order
    out = []

    for x in values:

        if x not in out:
            out.append(x)

    return ";".join(out)


# ======================================================================
# 7. BUILD SAME-TABLE LOOKUP
#
# ONLY the exact PMCID + table_index associated with the observation
# is examined.
# ======================================================================

group_cols = [
    "pmcid",
    "table_index"
]

table_groups = {
    key: group.copy()
    for key, group in code3.groupby(
        group_cols,
        dropna=False
    )
}

print(
    "Unique Code 3 source tables:",
    len(table_groups)
)


# ======================================================================
# 8. RECOVERY FUNCTION
# ======================================================================

def recover_metadata(row):

    pmcid = clean(
        row["pmcid"]
    )

    try:
        table_index = int(
            float(row["table_index"])
        )
    except Exception:
        return {
            "drug_recovered": "NOT_IDENTIFIED",
            "polymer_recovered": "NOT_IDENTIFIED",
            "carrier_recovered": "NOT_IDENTIFIED",
            "formulation_context": "",
            "recovery_status": "NO_TABLE_KEY",
            "recovery_evidence": "",
        }


    key = (
        pmcid,
        table_index
    )


    if key not in table_groups:

        return {
            "drug_recovered": "NOT_IDENTIFIED",
            "polymer_recovered": "NOT_IDENTIFIED",
            "carrier_recovered": "NOT_IDENTIFIED",
            "formulation_context": "",
            "recovery_status": "NO_TABLE_FOUND",
            "recovery_evidence": "",
        }


    table = table_groups[key].copy()


    # --------------------------------------------------------------
    # Gather actual table text
    # --------------------------------------------------------------

    row_texts = [
        clean(x)
        for x in table["row_text"]
        if clean(x)
    ]

    headers = [
        clean(x)
        for x in table["header_text"]
        if clean(x)
    ]

    cells = [
        clean(x)
        for x in table["raw_cell"]
        if clean(x)
    ]


    combined = " ".join(
        row_texts + headers + cells
    )


    # --------------------------------------------------------------
    # Find explicit lexical evidence
    # --------------------------------------------------------------

    drugs = find_terms(
        combined,
        DRUGS
    )

    polymers = find_terms(
        combined,
        POLYMERS
    )

    carriers = find_terms(
        combined,
        CARRIERS
    )


    # --------------------------------------------------------------
    # Stronger row-level evidence
    #
    # Look specifically for rows containing release-related terms.
    # --------------------------------------------------------------

    release_rows = []

    release_words = [
        "release",
        "drug release",
        "cumulative",
        "released",
        "release (%)",
    ]


    for rt in row_texts:

        rt_n = norm(rt)

        if any(
            word in rt_n
            for word in release_words
        ):

            release_rows.append(rt)


    release_context = " ".join(
        release_rows
    )


    row_drugs = find_terms(
        release_context,
        DRUGS
    )

    row_polymers = find_terms(
        release_context,
        POLYMERS
    )

    row_carriers = find_terms(
        release_context,
        CARRIERS
    )


    # --------------------------------------------------------------
    # Prefer release-row evidence
    # --------------------------------------------------------------

    if row_drugs:
        final_drugs = row_drugs
        drug_strength = "STRONG"
    else:
        final_drugs = drugs
        drug_strength = (
            "MODERATE"
            if drugs
            else "NONE"
        )


    if row_polymers:
        final_polymers = row_polymers
        polymer_strength = "STRONG"
    else:
        final_polymers = polymers
        polymer_strength = (
            "MODERATE"
            if polymers
            else "NONE"
        )


    if row_carriers:
        final_carriers = row_carriers
        carrier_strength = "STRONG"
    else:
        final_carriers = carriers
        carrier_strength = (
            "MODERATE"
            if carriers
            else "NONE"
        )


    # --------------------------------------------------------------
    # IMPORTANT SAFETY RULE
    #
    # If many unrelated drugs/polymers/carriers appear throughout a
    # review table, do NOT call them formulation metadata.
    #
    # More than 3 different drugs is treated as ambiguous.
    # --------------------------------------------------------------

    ambiguous = False

    if len(final_drugs) > 3:
        ambiguous = True

    if len(final_polymers) > 4:
        ambiguous = True

    if len(final_carriers) > 4:
        ambiguous = True


    # --------------------------------------------------------------
    # Status
    # --------------------------------------------------------------

    if ambiguous:

        status = "AMBIGUOUS_TABLE_CONTEXT"

    elif (
        drug_strength == "STRONG"
        and (
            polymer_strength == "STRONG"
            or carrier_strength == "STRONG"
        )
    ):

        status = "RECOVERED_STRONG"

    elif (
        drug_strength in [
            "STRONG",
            "MODERATE"
        ]
        and (
            polymer_strength in [
                "STRONG",
                "MODERATE"
            ]
            or carrier_strength in [
                "STRONG",
                "MODERATE"
            ]
        )
    ):

        status = "RECOVERED_GOOD"

    elif (
        drug_strength != "NONE"
        or polymer_strength != "NONE"
        or carrier_strength != "NONE"
    ):

        status = "PARTIAL_RECOVERY"

    else:

        status = "UNRESOLVED"


    evidence = (
        f"drug={drug_strength};"
        f"polymer={polymer_strength};"
        f"carrier={carrier_strength}"
    )


    return {
        "drug_recovered":
            normalize_terms(final_drugs),

        "polymer_recovered":
            normalize_terms(final_polymers),

        "carrier_recovered":
            normalize_terms(final_carriers),

        "formulation_context":
            clean(release_context)[:2000],

        "recovery_status":
            status,

        "recovery_evidence":
            evidence,
    }


# ======================================================================
# 9. PROCESS ONLY THE 21 UNRESOLVED OBSERVATIONS
# ======================================================================

results = []

for i, (_, row) in enumerate(
    manual.iterrows(),
    start=1
):

    result = recover_metadata(row)

    result["
observation_id"] = row[
        "observation_id"
    ]

    results.append(result)

    print(
        f"Processed {i}/{len(manual)}"
    )


recovery = pd.DataFrame(
    results
)


# ======================================================================
# 10. MERGE RECOVERY BACK
# ======================================================================

df2 = df.merge(
    recovery,
    on="observation_id",
    how="left",
    validate="one_to_one"
)


# ======================================================================
# 11. FINAL METADATA CONSOLIDATION
#
# Never overwrite stronger 6L evidence.
# ======================================================================

def choose_final(
    existing,
    recovered
):

    if valid(existing):

        return clean(existing)

    if valid(recovered):

        return clean(recovered)

    return "NOT_IDENTIFIED"


df2["final_drug_6M"] = [
    choose_final(
        a,
        b
    )
    for a, b in zip(
        df2["final_drug"],
        df2["drug_recovered"]
    )
]


df2["final_polymer_6M"] = [
    choose_final(
        a,
        b
    )
    for a, b in zip(
        df2["final_polymer"],
        df2["polymer_recovered"]
    )
]


df2["final_carrier_6M"] = [
    choose_final(
        a,
        b
    )
    for a, b in zip(
        df2["final_carrier"],
        df2["carrier_recovered"]
    )
]


# ======================================================================
# 12. FINAL STATUS
# ======================================================================

def final_status(row):

    # Existing 6L metadata are already acceptable
    if row["final_metadata_status"] == "READY_METADATA_VALIDATION":
        return "READY_METADATA_VALIDATION"


    # Strong/good new recovery
    if row["recovery_status"] in [
        "RECOVERED_STRONG",
        "RECOVERED_GOOD"
    ]:

        return "READY_METADATA_VALIDATION"


    return "MANUAL_METADATA_REVIEW"


df2["final_6M_status"] = df2.apply(
    final_status,
    axis=1
)


# ======================================================================
# 13. EVIDENCE CLASS
# ======================================================================

def evidence_class(row):

    if row["final_metadata_status"] == "READY_METADATA_VALIDATION":

        if (
            row["drug_evidence"]
            == "6K_EXACT_RELEASE_HEADER"
        ):

            return "EXACT_HEADER_ANCHORED"

        return "6G_OR_6H_STRUCTURAL"


    if row["recovery_status"] == "RECOVERED_STRONG":

        return "STRONG_SAME_TABLE_RECOVERY"


    if row["recovery_status"] == "RECOVERED_GOOD":

        return "GOOD_SAME_TABLE_RECOVERY"


    if row["recovery_status"] == "PARTIAL_RECOVERY":

        return "PARTIAL"

    return "UNRESOLVED"


df2["final_metadata_evidence"] = df2.apply(
    evidence_class,
    axis=1
)


# ======================================================================
# 14. SAFETY CHECKS
# ======================================================================

if len(df2) != 60:

    raise ValueError(
        f"CRITICAL: Expected 60 observations, "
        f"found {len(df2)}."
    )


if df2[
    "observation_id"
].duplicated().any():

    raise ValueError(
        "CRITICAL: Duplicate observation IDs detected."
    )


# ======================================================================
# 15. SPLIT OUTPUTS
# ======================================================================

recovered = df2[
    df2["final_6M_status"]
    == "READY_METADATA_VALIDATION"
].copy()


remaining = df2[
    df2["final_6M_status"]
    == "MANUAL_METADATA_REVIEW"
].copy()


# ======================================================================
# 16. REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6M COMPLETE")
print("=" * 70)

print(
    "Total source-validated observations:",
    len(df2)
)

print(
    "Originally metadata-ready:",
    int(
        (
            df2["final_metadata_status"]
            == "READY_METADATA_VALIDATION"
        ).sum()
    )
)

print(
    "Originally manual:",
    len(manual)
)

print()
print("6M recovery status:")

print(
    recovery[
        "recovery_status"
    ].value_counts()
)


print()
print("Final metadata status:")

print(
    df2[
        "final_6M_status"
    ].value_counts()
)


print()
print("Final drugs:")

print(
    df2[
        "final_drug_6M"
    ].value_counts()
)


print()
print("Final polymers:")

print(
    df2[
        "final_polymer_6M"
    ].value_counts()
)


print()
print("Final carriers:")

print(
    df2[
        "final_carrier_6M"
    ].value_counts()
)


print()
print(
    "Final metadata evidence:"
)

print(
    df2[
        "final_metadata_evidence"
    ].value_counts()
)


print()
print(
    "Ready metadata observations:",
    len(recovered)
)

print(
    "Remaining manual observations:",
    len(remaining)
)


print()
print(
    "Observation preservation:"
)

print(
    "Input :",
    len(df)
)

print(
    "Output:",
    len(df2)
)


# ======================================================================
# 17. SAVE
# ======================================================================

df2.to_csv(
    OUT_AUDIT,
    index=False
)

recovered.to_csv(
    OUT_RECOVERED,
    index=False
)

remaining.to_csv(
    OUT_MANUAL,
    index=False
)


print()
print("Saved:")

print(
    OUT_AUDIT
)

print(
    OUT_RECOVERED
)

print(
    OUT_MANUAL
)


print()
print("IMPORTANT:")
print(
    "Only the 21 unresolved 6L observations were targeted."
)

print(
    "All 60 source-validated release observations are retained."
)

print(
    "Recovery is restricted to the same PMCID + table."
)

print(
    "Exact release observations are never modified."
)

print(
    "No unrelated article text is used to infer formulation identity."
)

print(
    "Ambiguous review tables remain manual review."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6M — TARGETED SAME-TABLE FORMULATION METADATA RECOVERY — FIXED
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")
RAW = os.path.join(ROOT, "data", "raw", "tables")

F6L = os.path.join(
    VAL,
    "06L_conservative_metadata_consolidation_audit.csv"
)

F_CODE3 = os.path.join(
    RAW,
    "03_extracted_table_cells.csv"
)

OUT_AUDIT = os.path.join(
    VAL,
    "06M_targeted_formulation_recovery_audit.csv"
)

OUT_RECOVERED = os.path.join(
    VAL,
    "06M_recovered_metadata_candidates.csv"
)

OUT_MANUAL = os.path.join(
    VAL,
    "06M_remaining_manual_review.csv"
)


print("=" * 70)
print("CODE 6M — TARGETED SAME-TABLE FORMULATION METADATA RECOVERY")
print("=" * 70)


# ======================================================================
# 1. LOAD 6L
# ======================================================================

df = pd.read_csv(F6L)

print("6L observations:", len(df))


if "observation_id" not in df.columns:
    raise ValueError(
        "6L file does not contain observation_id."
    )


# ======================================================================
# 2. IDENTIFY THE 6L MANUAL OBSERVATIONS
# ======================================================================

if "final_metadata_status" not in df.columns:
    raise ValueError(
        "6L file does not contain final_metadata_status."
    )


manual_mask = (
    df["final_metadata_status"]
    == "MANUAL_METADATA_REVIEW"
)

manual = df.loc[
    manual_mask
].copy()

print(
    "Observations requiring targeted recovery:",
    len(manual)
)


# ======================================================================
# 3. LOAD CODE 3
# ======================================================================

code3 = pd.read_csv(F_CODE3)

print(
    "Code 3 cell records:",
    f"{len(code3):,}"
)


required_columns = [
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell",
    "row_text",
    "full_table_text",
    "header_text"
]

missing = [
    c
    for c in required_columns
    if c not in code3.columns
]

if missing:
    raise ValueError(
        "Code 3 is missing required columns:\n"
        + "\n".join(missing)
    )


# ======================================================================
# 4. HELPER FUNCTIONS
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


def norm(x):

    return clean(x).lower()


def valid(x):

    s = norm(x)

    return (
        s != ""
        and s not in [
            "not_identified",
            "unresolved",
            "unknown",
            "none",
            "nan",
            "manual_review"
        ]
    )


# ======================================================================
# 5. VOCABULARIES
# ======================================================================

DRUGS = [
    "doxorubicin",
    "dox",
    "dexamethasone",
    "curcumin",
    "paclitaxel",
    "metformin",
    "ibuprofen",
    "5-fluorouracil",
    "5 fluorouracil",
    "5-fu",
    "gemcitabine",
    "rifampicin",
    "vancomycin",
    "insulin"
]


POLYMERS = [
    "plga",
    "pla",
    "pcl",
    "peg",
    "peo",
    "pva",
    "pvp",
    "chitosan",
    "alginate",
    "cellulose",
    "gelatin",
    "polyurethane",
    "poloxamer"
]


CARRIERS = [
    "nanoparticle",
    "nanoparticles",
    "microparticle",
    "microparticles",
    "microsphere",
    "microspheres",
    "hydrogel",
    "hydrogels",
    "nanofiber",
    "nanofibers",
    "film",
    "films",
    "liposome",
    "liposomes",
    "bead",
    "beads",
    "implant",
    "implants",
    "micelle",
    "micelles"
]


def find_terms(text, vocabulary):

    text_n = norm(text)

    found = []

    for term in vocabulary:

        if term in text_n:

            found.append(term)

    return found


def normalize_terms(values):

    if not values:
        return "NOT_IDENTIFIED"

    output = []

    for value in values:

        if value not in output:

            output.append(value)

    return ";".join(output)


# ======================================================================
# 6. BUILD TABLE LOOKUP
# ======================================================================

table_groups = {}

for key, group in code3.groupby(
    ["pmcid", "table_index"],
    dropna=False
):

    table_groups[key] = group.copy()


print(
    "Unique Code 3 source tables:",
    len(table_groups)
)


# ======================================================================
# 7. RECOVERY FUNCTION
# ======================================================================

def recover_metadata(row):

    pmcid = clean(
        row["pmcid"]
    )

    try:

        table_index = int(
            float(row["table_index"])
        )

    except Exception:

        return {
            "drug_recovered": "NOT_IDENTIFIED",
            "polymer_recovered": "NOT_IDENTIFIED",
            "carrier_recovered": "NOT_IDENTIFIED",
            "formulation_context": "",
            "recovery_status": "NO_TABLE_KEY",
            "recovery_evidence": ""
        }


    key = (
        pmcid,
        table_index
    )


    if key not in table_groups:

        return {
            "drug_recovered": "NOT_IDENTIFIED",
            "polymer_recovered": "NOT_IDENTIFIED",
            "carrier_recovered": "NOT_IDENTIFIED",
            "formulation_context": "",
            "recovery_status": "NO_TABLE_FOUND",
            "recovery_evidence": ""
        }


    table = table_groups[key].copy()


    # --------------------------------------------------------------
    # TABLE TEXT
    # --------------------------------------------------------------

    row_texts = [
        clean(x)
        for x in table["row_text"]
        if clean(x)
    ]

    headers = [
        clean(x)
        for x in table["header_text"]
        if clean(x)
    ]

    cells = [
        clean(x)
        for x in table["raw_cell"]
        if clean(x)
    ]


    combined = " ".join(
        row_texts +
        headers +
        cells
    )


    # --------------------------------------------------------------
    # GENERAL TABLE EVIDENCE
    # --------------------------------------------------------------

    drugs = find_terms(
        combined,
        DRUGS
    )

    polymers = find_terms(
        combined,
        POLYMERS
    )

    carriers = find_terms(
        combined,
        CARRIERS
    )


    # --------------------------------------------------------------
    # RELEASE-RELATED ROWS
    # --------------------------------------------------------------

    release_rows = []

    release_words = [
        "release",
        "drug release",
        "cumulative",
        "released",
        "release (%)"
    ]


    for text in row_texts:

        text_n = norm(text)

        if any(
            word in text_n
            for word in release_words
        ):

            release_rows.append(text)


    release_context = " ".join(
        release_rows
    )


    row_drugs = find_terms(
        release_context,
        DRUGS
    )

    row_polymers = find_terms(
        release_context,
        POLYMERS
    )

    row_carriers = find_terms(
        release_context,
        CARRIERS
    )


    # --------------------------------------------------------------
    # PREFERENCE FOR RELEASE-ROW EVIDENCE
    # --------------------------------------------------------------

    if row_drugs:

        final_drugs = row_drugs
        drug_strength = "STRONG"

    elif drugs:

        final_drugs = drugs
        drug_strength = "MODERATE"

    else:

        final_drugs = []
        drug_strength = "NONE"


    if row_polymers:

        final_polymers = row_polymers
        polymer_strength = "STRONG"

    elif polymers:

        final_polymers = polymers
        polymer_strength = "MODERATE"

    else:

        final_polymers = []
        polymer_strength = "NONE"


    if row_carriers:

        final_carriers = row_carriers
        carrier_strength = "STRONG"

    elif carriers:

        final_carriers = carriers
        carrier_strength = "MODERATE"

    else:

        final_carriers = []
        carrier_strength = "NONE"


    # --------------------------------------------------------------
    # AMBIGUITY PROTECTION
    # --------------------------------------------------------------

    ambiguous = False

    if len(final_drugs) > 3:
        ambiguous = True

    if len(final_polymers) > 4:
        ambiguous = True

    if len(final_carriers) > 4:
        ambiguous = True


    # --------------------------------------------------------------
    # STATUS
    # --------------------------------------------------------------

    if ambiguous:

        status = "AMBIGUOUS_TABLE_CONTEXT"

    elif (
        drug_strength == "STRONG"
        and (
            polymer_strength == "STRONG"
            or carrier_strength == "STRONG"
        )
    ):

        status = "RECOVERED_STRONG"

    elif (
        drug_strength in [
            "STRONG",
            "MODERATE"
        ]
        and (
            polymer_strength in [
                "STRONG",
                "MODERATE"
            ]
            or carrier_strength in [
                "STRONG",
                "MODERATE"
            ]
        )
    ):

        status = "RECOVERED_GOOD"

    elif (
        drug_strength != "NONE"
        or polymer_strength != "NONE"
        or carrier_strength != "NONE"
    ):

        status = "PARTIAL_RECOVERY"

    else:

        status = "UNRESOLVED"


    evidence = (
        "drug="
        + drug_strength
        + ";polymer="
        + polymer_strength
        + ";carrier="
        + carrier_strength
    )


    return {
        "drug_recovered": normalize_terms(
            final_drugs
        ),

        "polymer_recovered": normalize_terms(
            final_polymers
        ),

        "carrier_recovered": normalize_terms(
            final_carriers
        ),

        "formulation_context": clean(
            release_context
        )[:2000],

        "recovery_status": status,

        "recovery_evidence": evidence
    }


# ======================================================================
# 8. PROCESS THE 21 MANUAL OBSERVATIONS
# ======================================================================

results = []

total_manual = len(manual)

for i, (_, row) in enumerate(
    manual.iterrows(),
    start=1
):

    result = recover_metadata(
        row
    )

    # FIXED LINE
    result["observation_id"] = row[
        "observation_id"
    ]

    results.append(
        result
    )

    print(
        f"Processed {i}/{total_manual}"
    )


recovery = pd.DataFrame(
    results
)


# ======================================================================
# 9. MERGE RECOVERY BACK TO ALL 60
# ======================================================================

df2 = df.merge(
    recovery,
    on="observation_id",
    how="left",
    validate="one_to_one"
)


# ======================================================================
# 10. FINAL DRUG
# ======================================================================

def choose_final(
    existing,
    recovered
):

    if valid(existing):

        return clean(existing)

    if valid(recovered):

        return clean(recovered)

    return "NOT_IDENTIFIED"


df2["final_drug_6M"] = [
    choose_final(
        existing,
        recovered
    )
    for existing, recovered
    in zip(
        df2["final_drug"],
        df2["drug_recovered"]
    )
]


# ======================================================================
# 11. FINAL POLYMER
# ======================================================================

df2["final_polymer_6M"] = [
    choose_final(
        existing,
        recovered
    )
    for existing, recovered
    in zip(
        df2["final_polymer"],
        df2["polymer_recovered"]
    )
]


# ======================================================================
# 12. FINAL CARRIER
# ======================================================================

df2["final_carrier_6M"] = [
    choose_final(
        existing,
        recovered
    )
    for existing, recovered
    in zip(
        df2["final_carrier"],
        df2["carrier_recovered"]
    )
]


# ======================================================================
# 13. FINAL STATUS
# ======================================================================

def determine_status(row):

    # Existing 6L-ready observations remain ready.
    if (
        row["final_metadata_status"]
        == "READY_METADATA_VALIDATION"
    ):

        return "READY_METADATA_VALIDATION"


    # Strong recovery
    if row["recovery_status"] in [
        "RECOVERED_STRONG",
        "RECOVERED_GOOD"
    ]:

        return "READY_METADATA_VALIDATION"


    return "MANUAL_METADATA_REVIEW"


df2["final_6M_status"] = df2.apply(
    determine_status,
    axis=1
)


# ======================================================================
# 14. EVIDENCE CLASS
# ======================================================================

def determine_evidence(row):

    if (
        row["final_metadata_status"]
        == "READY_METADATA_VALIDATION"
    ):

        if (
            row["drug_evidence"]
            == "6K_EXACT_RELEASE_HEADER"
        ):

            return "EXACT_HEADER_ANCHORED"

        return "6G_OR_6H_STRUCTURAL"


    if (
        row["recovery_status"]
        == "RECOVERED_STRONG"
    ):

        return "STRONG_SAME_TABLE_RECOVERY"


    if (
        row["recovery_status"]
        == "RECOVERED_GOOD"
    ):

        return "GOOD_SAME_TABLE_RECOVERY"


    if (
        row["recovery_status"]
        == "PARTIAL_RECOVERY"
    ):

        return "PARTIAL"


    return "UNRESOLVED"


df2["final_metadata_evidence"] = df2.apply(
    determine_evidence,
    axis=1
)


# ======================================================================
# 15. SAFETY CHECKS
# ======================================================================

if len(df2) != 60:

    raise ValueError(
        "CRITICAL: Expected 60 observations, "
        f"found {len(df2)}."
    )


if (
    df2["observation_id"]
    .duplicated()
    .any()
):

    raise ValueError(
        "CRITICAL: Duplicate observation IDs detected."
    )


# ======================================================================
# 16. SPLIT
# ======================================================================

recovered = df2[
    df2["final_6M_status"]
    == "READY_METADATA_VALIDATION"
].copy()


remaining = df2[
    df2["final_6M_status"]
    == "MANUAL_METADATA_REVIEW"
].copy()


# ======================================================================
# 17. REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6M COMPLETE")
print("=" * 70)

print(
    "Total source-validated observations:",
    len(df2)
)

print(
    "Originally metadata-ready:",
    int(
        (
            df2["final_metadata_status"]
            == "READY_METADATA_VALIDATION"
        ).sum()
    )
)

print(
    "Originally manual:",
    len(manual)
)


print()
print("6M recovery status:")

print(
    recovery[
        "recovery_status"
    ].value_counts()
)


print()
print("Final metadata status:")

print(
    df2[
        "final_6M_status"
    ].value_counts()
)


print()
print("Final drugs:")

print(
    df2[
        "final_drug_6M"
    ].value_counts()
)


print()
print("Final polymers:")

print(
    df2[
        "final_polymer_6M"
    ].value_counts()
)


print()
print("Final carriers:")

print(
    df2[
        "final_carrier_6M"
    ].value_counts()
)


print()
print("Final metadata evidence:")

print(
    df2[
        "final_metadata_evidence"
    ].value_counts()
)


print()
print(
    "Ready metadata observations:",
    len(recovered)
)

print(
    "Remaining manual observations:",
    len(remaining)
)


print()
print("Observation preservation:")

print(
    "Input :",
    len(df)
)

print(
    "Output:",
    len(df2)
)


# ======================================================================
# 18. SAVE
# ======================================================================

df2.to_csv(
    OUT_AUDIT,
    index=False
)

recovered.to_csv(
    OUT_RECOVERED,
    index=False
)

remaining.to_csv(
    OUT_MANUAL,
    index=False
)


print()
print("Saved:")

print(
    OUT_AUDIT
)

print(
    OUT_RECOVERED
)

print(
    OUT_MANUAL
)


print()
print("IMPORTANT:")
print(
    "Only the 21 unresolved 6L observations were targeted."
)

print(
    "All 60 source-validated release observations are retained."
)

print(
    "Recovery is restricted to the same PMCID + table."
)

print(
    "Exact release observations are never modified."
)

print(
    "No unrelated article text is used to infer formulation identity."
)

print(
    "Ambiguous tables remain manual review."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6N — FINAL ANALYTICAL ELIGIBILITY & QUALITY GATE
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")

INPUT_FILE = os.path.join(
    VAL,
    "06M_targeted_formulation_recovery_audit.csv"
)

OUT_AUDIT = os.path.join(
    VAL,
    "06N_final_analytical_quality_audit.csv"
)

OUT_RELEASE_VALID = os.path.join(
    VAL,
    "06N_release_valid_observations.csv"
)

OUT_ANALYTICAL = os.path.join(
    VAL,
    "06N_analytical_ready_observations.csv"
)

OUT_ML = os.path.join(
    VAL,
    "06N_ml_ready_observations.csv"
)

OUT_MANUAL = os.path.join(
    VAL,
    "06N_manual_review_observations.csv"
)

OUT_EXCLUDED = os.path.join(
    VAL,
    "06N_excluded_observations.csv"
)

print("=" * 70)
print("CODE 6N — FINAL ANALYTICAL ELIGIBILITY & QUALITY GATE")
print("=" * 70)


# ======================================================================
# 1. LOAD 6M
# ======================================================================

df = pd.read_csv(INPUT_FILE)

print("Input observations:", len(df))


if "observation_id" not in df.columns:
    raise ValueError(
        "Required column missing: observation_id"
    )


# ======================================================================
# 2. REQUIRED RELEASE FIELDS
# ======================================================================

required_release = [
    "pmcid",
    "table_index",
    "row_index",
    "release_time_h",
    "release_percent"
]

missing = [
    c for c in required_release
    if c not in df.columns
]

if missing:
    raise ValueError(
        "Required release columns missing:\n"
        + "\n".join(missing)
    )


# ======================================================================
# 3. HELPER FUNCTIONS
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


def norm(x):

    return clean(x).lower()


def is_identified(x):

    s = norm(x)

    return (
        s != ""
        and s not in [
            "not_identified",
            "unresolved",
            "unknown",
            "none",
            "nan",
            "manual_review"
        ]
    )


# ======================================================================
# 4. NUMERIC CLEANING
# ======================================================================

df["release_time_h_numeric"] = pd.to_numeric(
    df["release_time_h"],
    errors="coerce"
)

df["release_percent_numeric"] = pd.to_numeric(
    df["release_percent"],
    errors="coerce"
)


# ======================================================================
# 5. BASIC SOURCE-TRACEABILITY CHECK
# ======================================================================

df["source_trace_valid"] = (
    df["pmcid"].notna()
    & df["table_index"].notna()
    & df["row_index"].notna()
)


# ======================================================================
# 6. TIME VALIDITY
# ======================================================================

df["time_valid"] = (
    df["release_time_h_numeric"].notna()
    & (df["release_time_h_numeric"] >= 0)
)


# ======================================================================
# 7. RELEASE VALIDITY
# ======================================================================

df["release_valid_range"] = (
    df["release_percent_numeric"].notna()
    & (
        df["release_percent_numeric"] >= 0
    )
    & (
        df["release_percent_numeric"] <= 100
    )
)


# ======================================================================
# 8. DUPLICATE OBSERVATION CHECK
# ======================================================================

duplicate_keys = [
    "pmcid",
    "table_index",
    "row_index",
    "release_time_h_numeric",
    "release_percent_numeric"
]

df["exact_value_duplicate"] = (
    df.duplicated(
        subset=duplicate_keys,
        keep=False
    )
)


df["duplicate_observation_id"] = (
    df["observation_id"].duplicated(
        keep=False
    )
)


# ======================================================================
# 9. OBSERVATION TYPE
# ======================================================================

if "observation_type" in df.columns:

    df["analysis_observation_type"] = (
        df["observation_type"]
        .fillna("UNKNOWN")
        .astype(str)
    )

else:

    df["analysis_observation_type"] = "UNKNOWN"


df["is_timecourse"] = (
    df["analysis_observation_type"]
    == "TIMECOURSE_RELEASE"
)


df["is_max_release_event"] = (
    df["analysis_observation_type"]
    == "MAX_RELEASE_EVENT"
)


# ======================================================================
# 10. SOURCE-VALIDATED CHECK
# ======================================================================

if "source_validation" in df.columns:

    df["source_validation_ok"] = (
        df["source_validation"]
        .astype(str)
        .str.upper()
        .isin([
            "VALID_SOURCE_RELEASE",
            "VALID_RELEASE",
            "EXACT_XML_CELL_MATCH"
        ])
    )

else:

    # 6M originates from 6F, which was already source validated.
    df["source_validation_ok"] = True


# ======================================================================
# 11. METADATA FIELDS
# ======================================================================

# Prefer 6M consolidated fields.

drug_col = "final_drug_6M"
polymer_col = "final_polymer_6M"
carrier_col = "final_carrier_6M"

for col in [
    drug_col,
    polymer_col,
    carrier_col
]:

    if col not in df.columns:

        df[col] = "NOT_IDENTIFIED"


df["drug_identified"] = df[
    drug_col
].apply(is_identified)


df["polymer_identified"] = df[
    polymer_col
].apply(is_identified)


df["carrier_identified"] = df[
    carrier_col
].apply(is_identified)


# ======================================================================
# 12. FORMULATION IDENTIFIER
# ======================================================================

formulation_candidates = [
    "final_formulation_id",
    "formulation_id",
    "experimental_unit_id",
    "experimental_unit_proxy"
]

available_formulation_col = None

for col in formulation_candidates:

    if col in df.columns:

        available_formulation_col = col
        break


if available_formulation_col is not None:

    df["formulation_identifier_present"] = (
        df[available_formulation_col]
        .apply(is_identified)
    )

else:

    df["formulation_identifier_present"] = False


# ======================================================================
# 13. RELEASE-CONDITION CHECK
# ======================================================================

condition_candidates = [
    "final_release_condition",
    "release_condition",
    "condition"
]

condition_col = None

for col in condition_candidates:

    if col in df.columns:

        condition_col = col
        break


if condition_col is not None:

    df["release_condition_identified"] = (
        df[condition_col]
        .apply(is_identified)
    )

else:

    df["release_condition_identified"] = False


# ======================================================================
# 14. METADATA QUALITY
# ======================================================================

df["metadata_count"] = (
    df["drug_identified"].astype(int)
    + df["polymer_identified"].astype(int)
    + df["carrier_identified"].astype(int)
)


# ======================================================================
# 15. RELEASE-VALID STATUS
# ======================================================================

df["release_validation_status"] = np.select(

    [
        ~df["source_trace_valid"],

        ~df["source_validation_ok"],

        ~df["time_valid"],

        ~df["release_valid_range"],

        df["duplicate_observation_id"],

    ],

    [
        "SOURCE_TRACE_FAILURE",

        "SOURCE_VALIDATION_FAILURE",

        "INVALID_TIME",

        "INVALID_RELEASE",

        "DUPLICATE_OBSERVATION_ID",
    ],

    default="RELEASE_VALID"
)


# ======================================================================
# 16. ANALYTICAL ELIGIBILITY
# ======================================================================

# Analytical-ready means the release observation itself is valid.
#
# We do NOT require polymer/carrier identification here because doing so
# would incorrectly discard source-valid observations simply because
# literature tables did not provide complete formulation metadata.

df["analytical_status"] = np.select(

    [
        df["release_validation_status"]
        != "RELEASE_VALID",

        df["is_timecourse"]
        & ~df["time_valid"],

    ],

    [
        "NOT_ANALYTICALLY_VALID",

        "NOT_ANALYTICALLY_VALID",
    ],

    default="ANALYTICAL_READY"
)


# ======================================================================
# 17. ML ELIGIBILITY — CONSERVATIVE
# ======================================================================

# ML requires:
#   1. valid source
#   2. valid time
#   3. valid release
#   4. identified drug
#   5. identified polymer OR carrier
#   6. formulation/experimental-unit context
#
# This does NOT imply statistical independence.
# Multiple time points from the same formulation remain clustered.

df["ml_metadata_complete"] = (
    df["drug_identified"]
    & (
        df["polymer_identified"]
        | df["carrier_identified"]
    )
    & df["formulation_identifier_present"]
)


df["ml_status"] = np.select(

    [
        df["release_validation_status"]
        != "RELEASE_VALID",

        ~df["ml_metadata_complete"],

    ],

    [
        "NOT_ML_READY_SOURCE",

        "NOT_ML_READY_METADATA",

    ],

    default="ML_READY"
)


# ======================================================================
# 18. MAXIMUM-RELEASE EVENTS
# ======================================================================

# Keep maximum-release events in the validated analytical dataset,
# but explicitly flag them so they cannot silently be treated as
# ordinary time-course points.

df["modeling_role"] = np.select(

    [
        df["is_max_release_event"],

        df["is_timecourse"],

    ],

    [
        "MAX_RELEASE_EVENT",

        "TIMECOURSE_RELEASE",

    ],

    default="OTHER_RELEASE_OBSERVATION"
)


# ======================================================================
# 19. FINAL ELIGIBILITY
# ======================================================================

df["final_eligibility"] = np.select(

    [
        df["release_validation_status"]
        != "RELEASE_VALID",

        (
            df["release_validation_status"]
            == "RELEASE_VALID"
        )
        & (
            df["ml_status"]
            == "ML_READY"
        ),

        (
            df["release_validation_status"]
            == "RELEASE_VALID"
        )
        & (
            df["analytical_status"]
            == "ANALYTICAL_READY"
        ),

    ],

    [
        "EXCLUDED",

        "ML_READY",

        "ANALYTICAL_READY",

    ],

    default="MANUAL_REVIEW"
)


# ======================================================================
# 20. MANUAL REVIEW REASON
# ======================================================================

def manual_reason(row):

    reasons = []

    if not row["drug_identified"]:
        reasons.append(
            "DRUG_NOT_IDENTIFIED"
        )

    if not row["polymer_identified"]:
        reasons.append(
            "POLYMER_NOT_IDENTIFIED"
        )

    if not row["carrier_identified"]:
        reasons.append(
            "CARRIER_NOT_IDENTIFIED"
        )

    if not row[
        "formulation_identifier_present"
    ]:
        reasons.append(
            "FORMULATION_ID_NOT_IDENTIFIED"
        )

    if not row[
        "release_condition_identified"
    ]:
        reasons.append(
            "RELEASE_CONDITION_NOT_IDENTIFIED"
        )

    if not reasons:

        return "NONE"

    return ";".join(reasons)


df["manual_review_reason"] = df.apply(
    manual_reason,
    axis=1
)


# ======================================================================
# 21. CHECK FOR SAME-ARTICLE SAME-TIME MULTIPLE OBSERVATIONS
# ======================================================================

group_cols = [
    "pmcid",
    "table_index",
    "release_time_h_numeric"
]

group_sizes = (
    df.groupby(
        group_cols,
        dropna=False
    )["observation_id"]
    .transform("count")
)


df["same_article_table_time_n"] = (
    group_sizes
)


df["same_article_table_time_multiple"] = (
    df["same_article_table_time_n"] > 1
)


# ======================================================================
# 22. CHECK MONOTONIC TIME WITHIN EXPERIMENTAL UNIT
# ======================================================================

unit_col = None

for col in [
    "experimental_unit_id",
    "experimental_unit_proxy",
    "final_formulation_id",
    "formulation_id"
]:

    if col in df.columns:

        if df[col].apply(is_identified).any():

            unit_col = col
            break


if unit_col is not None:

    df["unit_time_duplicate"] = (
        df.groupby(
            unit_col,
            dropna=False
        )["release_time_h_numeric"]
        .transform(
            lambda x: x.duplicated(
                keep=False
            )
        )
    )

else:

    df["unit_time_duplicate"] = False


# ======================================================================
# 23. FINAL SAFETY CHECKS
# ======================================================================

if len(df) != 60:

    raise ValueError(
        "CRITICAL: Observation count changed. "
        f"Expected 60, found {len(df)}."
    )


if df[
    "observation_id"
].duplicated().any():

    raise ValueError(
        "CRITICAL: Duplicate observation IDs detected."
    )


if (
    (df["release_percent_numeric"] < 0)
    | (df["release_percent_numeric"] > 100)
).any():

    bad = df[
        (df["release_percent_numeric"] < 0)
        | (df["release_percent_numeric"] > 100)
    ]

    raise ValueError(
        "CRITICAL: Invalid release percentages detected: "
        f"{len(bad)}"
    )


if (
    df["release_time_h_numeric"] < 0
).any():

    bad = df[
        df["release_time_h_numeric"] < 0
    ]

    raise ValueError(
        "CRITICAL: Negative release times detected: "
        f"{len(bad)}"
    )


# ======================================================================
# 24. OUTPUT DATASETS
# ======================================================================

release_valid = df[
    df["release_validation_status"]
    == "RELEASE_VALID"
].copy()


analytical_ready = df[
    df["final_eligibility"]
    .isin([
        "ANALYTICAL_READY",
        "ML_READY"
    ])
].copy()


ml_ready = df[
    df["final_eligibility"]
    == "ML_READY"
].copy()


manual_review = df[
    df["final_eligibility"]
    == "ANALYTICAL_READY"
].copy()


# Important:
# manual metadata observations remain visible separately.
manual_metadata = df[
    df["final_metadata_status"]
    == "MANUAL_METADATA_REVIEW"
].copy()


excluded = df[
    df["final_eligibility"]
    == "EXCLUDED"
].copy()


# ======================================================================
# 25. REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6N COMPLETE")
print("=" * 70)

print(
    "Input observations:",
    len(df)
)

print()
print("Release validation:")
print(
    df[
        "release_validation_status"
    ].value_counts()
)

print()
print("Analytical status:")
print(
    df[
        "analytical_status"
    ].value_counts()
)

print()
print("ML status:")
print(
    df[
        "ml_status"
    ].value_counts()
)

print()
print("Final eligibility:")
print(
    df[
        "final_eligibility"
    ].value_counts()
)

print()
print("Observation types:")
print(
    df[
        "analysis_observation_type"
    ].value_counts()
)

print()
print("Modeling role:")
print(
    df[
        "modeling_role"
    ].value_counts()
)

print()
print("Drug identified:")
print(
    df[
        "drug_identified"
    ].value_counts()
)

print()
print("Polymer identified:")
print(
    df[
        "polymer_identified"
    ].value_counts()
)

print()
print("Carrier identified:")
print(
    df[
        "carrier_identified"
    ].value_counts()
)

print()
print("Formulation identifier:")
print(
    df[
        "formulation_identifier_present"
    ].value_counts()
)

print()
print("Release condition:")
print(
    df[
        "release_condition_identified"
    ].value_counts()
)

print()
print("Same article/table/time groups:")
print(
    df[
        "same_article_table_time_multiple"
    ].value_counts()
)

print()
print("Manual metadata observations:")
print(
    len(manual_metadata)
)

print()
print("Final dataset sizes:")
print(
    "Release-valid:",
    len(release_valid)
)

print(
    "Analytical-ready:",
    len(analytical_ready)
)

print(
    "ML-ready:",
    len(ml_ready)
)

print(
    "Manual review:",
    len(manual_review)
)

print(
    "Excluded:",
    len(excluded)
)

print()
print("Observation preservation:")
print(
    "Input :",
    len(df)
)

print(
    "Output:",
    len(df)
)


# ======================================================================
# 26. SAVE
# ======================================================================

df.to_csv(
    OUT_AUDIT,
    index=False
)

release_valid.to_csv(
    OUT_RELEASE_VALID,
    index=False
)

analytical_ready.to_csv(
    OUT_ANALYTICAL,
    index=False
)

ml_ready.to_csv(
    OUT_ML,
    index=False
)

manual_review.to_csv(
    OUT_MANUAL,
    index=False
)

excluded.to_csv(
    OUT_EXCLUDED,
    index=False
)


print()
print("Saved:")
print(OUT_AUDIT)
print(OUT_RELEASE_VALID)
print(OUT_ANALYTICAL)
print(OUT_ML)
print(OUT_MANUAL)
print(OUT_EXCLUDED)

print()
print("IMPORTANT:")
print(
    "This is an eligibility gate only."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No observations were silently deleted."
)

print(
    "Multiple time points remain retained."
)

print(
    "MAX_RELEASE_EVENT observations remain explicitly labelled."
)

print(
    "ML-ready does NOT imply independent observations."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6N-R — FINAL ANALYTICAL ELIGIBILITY & QUALITY GATE
# RELEASE DATA FROM 6F + METADATA FROM 6M
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")

FILE_6M = os.path.join(
    VAL,
    "06M_targeted_formulation_recovery_audit.csv"
)

FILE_6F = os.path.join(
    VAL,
    "06F_valid_source_release.csv"
)

OUT_AUDIT = os.path.join(
    VAL,
    "06N_final_analytical_quality_audit.csv"
)

OUT_RELEASE_VALID = os.path.join(
    VAL,
    "06N_release_valid_observations.csv"
)

OUT_ANALYTICAL = os.path.join(
    VAL,
    "06N_analytical_ready_observations.csv"
)

OUT_ML = os.path.join(
    VAL,
    "06N_ml_ready_observations.csv"
)

OUT_MANUAL = os.path.join(
    VAL,
    "06N_manual_review_observations.csv"
)

OUT_EXCLUDED = os.path.join(
    VAL,
    "06N_excluded_observations.csv"
)

print("=" * 70)
print("CODE 6N-R — FINAL ANALYTICAL ELIGIBILITY & QUALITY GATE")
print("=" * 70)


# ======================================================================
# 1. LOAD 6M METADATA
# ======================================================================

meta = pd.read_csv(FILE_6M)

print("6M observations:", len(meta))

if "observation_id" not in meta.columns:
    raise ValueError(
        "6M does not contain observation_id."
    )


# ======================================================================
# 2. LOAD 6F SOURCE-VALIDATED RELEASE DATA
# ======================================================================

release = pd.read_csv(FILE_6F)

print(
    "6F source-validated observations:",
    len(release)
)

if "observation_id" not in release.columns:
    raise ValueError(
        "6F does not contain observation_id."
    )


# ======================================================================
# 3. CHECK FOR DUPLICATE IDS
# ======================================================================

if meta["observation_id"].duplicated().any():

    raise ValueError(
        "Duplicate observation_id values detected in 6M."
    )


if release["observation_id"].duplicated().any():

    raise ValueError(
        "Duplicate observation_id values detected in 6F."
    )


# ======================================================================
# 4. REQUIRED 6F RELEASE COLUMNS
# ======================================================================

required_release = [
    "observation_id",
    "pmcid",
    "table_index",
    "row_index",
    "release_time_h",
    "release_percent"
]

missing_release = [
    c
    for c in required_release
    if c not in release.columns
]

if missing_release:

    raise ValueError(
        "6F is missing required release columns:\n"
        + "\n".join(missing_release)
    )


# ======================================================================
# 5. MERGE RELEASE DATA INTO METADATA
# ======================================================================

release_cols = [
    "observation_id",
    "pmcid",
    "table_index",
    "row_index",
    "release_time_h",
    "release_percent"
]

if "observation_type" in release.columns:
    release_cols.append("observation_type")

if "time_raw" in release.columns:
    release_cols.append("time_raw")

if "release_raw" in release.columns:
    release_cols.append("release_raw")

release_small = release[
    release_cols
].copy()


# Prevent metadata columns from being overwritten.

meta_drop = [
    c
    for c in release_small.columns
    if (
        c != "observation_id"
        and c in meta.columns
    )
]

if meta_drop:
    release_small = release_small.drop(
        columns=meta_drop
    )


df = meta.merge(
    release_small,
    on="observation_id",
    how="left",
    validate="one_to_one"
)


# ======================================================================
# 6. CRITICAL OBSERVATION COUNT CHECK
# ======================================================================

if len(df) != 60:

    raise ValueError(
        "CRITICAL: Expected 60 consolidated observations, "
        f"found {len(df)}."
    )


print(
    "Rows after 6F + 6M consolidation:",
    len(df)
)


# ======================================================================
# 7. HELPER FUNCTIONS
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


def norm(x):

    return clean(x).lower()


def identified(x):

    s = norm(x)

    return (
        s != ""
        and s not in [
            "not_identified",
            "unresolved",
            "unknown",
            "none",
            "nan",
            "manual_review"
        ]
    )


# ======================================================================
# 8. NUMERIC RELEASE FIELDS
# ======================================================================

df["release_time_h_numeric"] = pd.to_numeric(
    df["release_time_h"],
    errors="coerce"
)

df["release_percent_numeric"] = pd.to_numeric(
    df["release_percent"],
    errors="coerce"
)


# ======================================================================
# 9. SOURCE TRACEABILITY
# ======================================================================

df["source_trace_valid"] = (
    df["pmcid"].notna()
    & df["table_index"].notna()
    & df["row_index"].notna()
)


# ======================================================================
# 10. TIME VALIDITY
# ======================================================================

df["time_valid"] = (
    df["release_time_h_numeric"].notna()
    & (
        df["release_time_h_numeric"] >= 0
    )
)


# ======================================================================
# 11. RELEASE VALIDITY
# ======================================================================

df["release_valid_range"] = (
    df["release_percent_numeric"].notna()
    & (
        df["release_percent_numeric"] >= 0
    )
    & (
        df["release_percent_numeric"] <= 100
    )
)


# ======================================================================
# 12. DUPLICATE CHECK
# ======================================================================

df["duplicate_observation_id"] = (
    df["observation_id"].duplicated(
        keep=False
    )
)


value_key = [
    "pmcid",
    "table_index",
    "row_index",
    "release_time_h_numeric",
    "release_percent_numeric"
]

df["exact_value_duplicate"] = (
    df.duplicated(
        subset=value_key,
        keep=False
    )
)


# ======================================================================
# 13. OBSERVATION TYPE
# ======================================================================

if "observation_type" not in df.columns:

    df["observation_type"] = "UNKNOWN"


df["observation_type"] = (
    df["observation_type"]
    .fillna("UNKNOWN")
    .astype(str)
)


df["is_timecourse"] = (
    df["observation_type"]
    == "TIMECOURSE_RELEASE"
)


df["is_max_release_event"] = (
    df["observation_type"]
    == "MAX_RELEASE_EVENT"
)


# ======================================================================
# 14. SOURCE VALIDATION
# ======================================================================

df["source_validation_ok"] = True

# 6F itself is the source-validated branch.
# Verify the source-validation field if available.

if "source_validation" in release.columns:

    source_map = release[
        [
            "observation_id",
            "source_validation"
        ]
    ].drop_duplicates(
        "observation_id"
    )

    df = df.merge(
        source_map,
        on="observation_id",
        how="left",
        suffixes=("", "_6F")
    )

    df["source_validation_ok"] = (
        df["source_validation_6F"]
        .astype(str)
        .str.upper()
        .isin([
            "VALID_SOURCE_RELEASE",
            "VALID_RELEASE",
            "EXACT_XML_CELL_MATCH"
        ])
    )


# ======================================================================
# 15. METADATA COLUMNS FROM 6M
# ======================================================================

drug_col = "final_drug_6M"
polymer_col = "final_polymer_6M"
carrier_col = "final_carrier_6M"

for col in [
    drug_col,
    polymer_col,
    carrier_col
]:

    if col not in df.columns:
        df[col] = "NOT_IDENTIFIED"


df["drug_identified"] = (
    df[drug_col]
    .apply(identified)
)


df["polymer_identified"] = (
    df[polymer_col]
    .apply(identified)
)


df["carrier_identified"] = (
    df[carrier_col]
    .apply(identified)
)


# ======================================================================
# 16. FORMULATION / EXPERIMENTAL UNIT
# ======================================================================

formulation_col = None

for col in [
    "final_formulation_id",
    "formulation_id",
    "experimental_unit_id",
    "experimental_unit_proxy",
    "experimental_unit_id_6M"
]:

    if col in df.columns:

        if df[col].apply(identified).any():

            formulation_col = col
            break


if formulation_col is None:

    df["formulation_identifier_present"] = False

else:

    df["formulation_identifier_present"] = (
        df[formulation_col]
        .apply(identified)
    )


# ======================================================================
# 17. RELEASE CONDITION
# ======================================================================

condition_col = None

for col in [
    "final_release_condition",
    "release_condition",
    "condition"
]:

    if col in df.columns:

        condition_col = col
        break


if condition_col is None:

    df["release_condition_identified"] = False

else:

    df["release_condition_identified"] = (
        df[condition_col]
        .apply(identified)
    )


# ======================================================================
# 18. METADATA COUNT
# ======================================================================

df["metadata_count"] = (
    df["drug_identified"].astype(int)
    + df["polymer_identified"].astype(int)
    + df["carrier_identified"].astype(int)
)


# ======================================================================
# 19. RELEASE VALIDATION STATUS
# ======================================================================

df["release_validation_status"] = np.select(

    [
        ~df["source_trace_valid"],

        ~df["source_validation_ok"],

        ~df["time_valid"],

        ~df["release_valid_range"],

        df["duplicate_observation_id"]
    ],

    [
        "SOURCE_TRACE_FAILURE",

        "SOURCE_VALIDATION_FAILURE",

        "INVALID_TIME",

        "INVALID_RELEASE",

        "DUPLICATE_OBSERVATION_ID"
    ],

    default="RELEASE_VALID"
)


# ======================================================================
# 20. ANALYTICAL STATUS
# ======================================================================

df["analytical_status"] = np.where(

    df["release_validation_status"]
    == "RELEASE_VALID",

    "ANALYTICAL_READY",

    "NOT_ANALYTICALLY_VALID"
)


# ======================================================================
# 21. CONSERVATIVE ML METADATA GATE
# ======================================================================

# ML requires:
#
#   valid release
#   identified drug
#   identified polymer OR carrier
#   formulation / experimental-unit context
#
# This does NOT claim statistical independence.

df["ml_metadata_complete"] = (
    df["release_validation_status"]
    .eq("RELEASE_VALID")
    & df["drug_identified"]
    & (
        df["polymer_identified"]
        | df["carrier_identified"]
    )
    & df["formulation_identifier_present"]
)


df["ml_status"] = np.where(

    df["ml_metadata_complete"],

    "ML_READY",

    "NOT_ML_READY_METADATA"
)


# ======================================================================
# 22. MODELING ROLE
# ======================================================================

df["modeling_role"] = np.select(

    [
        df["is_timecourse"],

        df["is_max_release_event"]
    ],

    [
        "TIMECOURSE_RELEASE",

        "MAX_RELEASE_EVENT"
    ],

    default="OTHER_RELEASE_OBSERVATION"
)


# ======================================================================
# 23. FINAL ELIGIBILITY
# ======================================================================

df["final_eligibility"] = np.select(

    [
        df["release_validation_status"]
        != "RELEASE_VALID",

        df["ml_status"]
        == "ML_READY",

        df["analytical_status"]
        == "ANALYTICAL_READY"
    ],

    [
        "EXCLUDED",

        "ML_READY",

        "ANALYTICAL_READY"
    ],

    default="MANUAL_REVIEW"
)


# ======================================================================
# 24. MANUAL REVIEW REASONS
# ======================================================================

def review_reason(row):

    reasons = []

    if not row["drug_identified"]:
        reasons.append(
            "DRUG_NOT_IDENTIFIED"
        )

    if not row["polymer_identified"]:
        reasons.append(
            "POLYMER_NOT_IDENTIFIED"
        )

    if not row["carrier_identified"]:
        reasons.append(
            "CARRIER_NOT_IDENTIFIED"
        )

    if not row[
        "formulation_identifier_present"
    ]:
        reasons.append(
            "FORMULATION_ID_NOT_IDENTIFIED"
        )

    if not row[
        "release_condition_identified"
    ]:
        reasons.append(
            "RELEASE_CONDITION_NOT_IDENTIFIED"
        )

    if not reasons:
        return "NONE"

    return ";".join(reasons)


df["manual_review_reason"] = df.apply(
    review_reason,
    axis=1
)


# ======================================================================
# 25. SAME ARTICLE / TABLE / TIME CHECK
# ======================================================================

group_cols = [
    "pmcid",
    "table_index",
    "release_time_h_numeric"
]

df["same_article_table_time_n"] = (
    df.groupby(
        group_cols,
        dropna=False
    )["observation_id"]
    .transform("count")
)


df["same_article_table_time_multiple"] = (
    df["same_article_table_time_n"] > 1
)


# ======================================================================
# 26. SAFETY CHECKS
# ======================================================================

if len(df) != 60:

    raise ValueError(
        "CRITICAL: Observation count changed."
    )


if df[
    "observation_id"
].duplicated().any():

    raise ValueError(
        "CRITICAL: Duplicate observation IDs."
    )


if (
    df["release_time_h_numeric"] < 0
).any():

    raise ValueError(
        "CRITICAL: Negative release times."
    )


if (
    (
        df["release_percent_numeric"]
        < 0
    )
    |
    (
        df["release_percent_numeric"]
        > 100
    )
).any():

    raise ValueError(
        "CRITICAL: Release percentage outside 0–100."
    )


# ======================================================================
# 27. CREATE OUTPUT DATASETS
# ======================================================================

release_valid = df[
    df["release_validation_status"]
    == "RELEASE_VALID"
].copy()


analytical_ready = df[
    df["final_eligibility"]
    .isin([
        "ANALYTICAL_READY",
        "ML_READY"
    ])
].copy()


ml_ready = df[
    df["final_eligibility"]
    == "ML_READY"
].copy()


manual_review = df[
    df["final_eligibility"]
    == "ANALYTICAL_READY"
].copy()


# Correct manual metadata subset:
manual_metadata = df[
    df["final_metadata_status"]
    == "MANUAL_METADATA_REVIEW"
].copy()


excluded = df[
    df["final_eligibility"]
    == "EXCLUDED"
].copy()


# ======================================================================
# 28. REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6N-R COMPLETE")
print("=" * 70)

print(
    "6M observations:",
    len(meta)
)

print(
    "6F observations:",
    len(release)
)

print(
    "Merged observations:",
    len(df)
)

print()
print("Release validation:")
print(
    df[
        "release_validation_status"
    ].value_counts()
)

print()
print("Analytical status:")
print(
    df[
        "analytical_status"
    ].value_counts()
)

print()
print("ML status:")
print(
    df[
        "ml_status"
    ].value_counts()
)

print()
print("Final eligibility:")
print(
    df[
        "final_eligibility"
    ].value_counts()
)

print()
print("Observation types:")
print(
    df[
        "observation_type"
    ].value_counts()
)

print()
print("Drug identified:")
print(
    df[
        "drug_identified"
    ].value_counts()
)

print()
print("Polymer identified:")
print(
    df[
        "polymer_identified"
    ].value_counts()
)

print()
print("Carrier identified:")
print(
    df[
        "carrier_identified"
    ].value_counts()
)

print()
print("Formulation identifier:")
print(
    df[
        "formulation_identifier_present"
    ].value_counts()
)

print()
print("Release condition:")
print(
    df[
        "release_condition_identified"
    ].value_counts()
)

print()
print("Same article/table/time:")
print(
    df[
        "same_article_table_time_multiple"
    ].value_counts()
)

print()
print("FINAL COUNTS")
print("-" * 40)

print(
    "Release-valid:",
    len(release_valid)
)

print(
    "Analytical-ready:",
    len(analytical_ready)
)

print(
    "ML-ready:",
    len(ml_ready)
)

print(
    "Manual metadata review:",
    len(manual_metadata)
)

print(
    "Excluded:",
    len(excluded)
)

print()
print("Observation preservation:")
print(
    "Input :",
    len(meta)
)

print(
    "Output:",
    len(df)
)


# ======================================================================
# 29. SAVE
# ======================================================================

df.to_csv(
    OUT_AUDIT,
    index=False
)

release_valid.to_csv(
    OUT_RELEASE_VALID,
    index=False
)

analytical_ready.to_csv(
    OUT_ANALYTICAL,
    index=False
)

ml_ready.to_csv(
    OUT_ML,
    index=False
)

manual_metadata.to_csv(
    OUT_MANUAL,
    index=False
)

excluded.to_csv(
    OUT_EXCLUDED,
    index=False
)


print()
print("Saved:")
print(OUT_AUDIT)
print(OUT_RELEASE_VALID)
print(OUT_ANALYTICAL)
print(OUT_ML)
print(OUT_MANUAL)
print(OUT_EXCLUDED)

print()
print("IMPORTANT:")
print(
    "Release values come directly from the source-validated 6F branch."
)

print(
    "Metadata come from the consolidated 6M branch."
)

print(
    "The two branches are joined only by observation_id."
)

print(
    "No release values are regenerated or recalculated."
)

print(
    "No observations are interpolated."
)

print(
    "No observations are duplicated."
)

print(
    "Multiple time points remain retained."
)

print(
    "MAX_RELEASE_EVENT remains separately labelled."
)

print(
    "ML-ready does NOT imply independent observations."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6N-R2 — FINAL ANALYTICAL ELIGIBILITY & QUALITY GATE
# FIXED FOR ACTUAL 6F + 6M SCHEMAS
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")

FILE_6M = os.path.join(
    VAL,
    "06M_targeted_formulation_recovery_audit.csv"
)

FILE_6F = os.path.join(
    VAL,
    "06F_valid_source_release.csv"
)

OUT_AUDIT = os.path.join(
    VAL,
    "06N_final_analytical_quality_audit.csv"
)

OUT_RELEASE_VALID = os.path.join(
    VAL,
    "06N_release_valid_observations.csv"
)

OUT_ANALYTICAL = os.path.join(
    VAL,
    "06N_analytical_ready_observations.csv"
)

OUT_ML = os.path.join(
    VAL,
    "06N_ml_ready_observations.csv"
)

OUT_MANUAL = os.path.join(
    VAL,
    "06N_manual_review_observations.csv"
)

OUT_EXCLUDED = os.path.join(
    VAL,
    "06N_excluded_observations.csv"
)

print("=" * 70)
print("CODE 6N-R2 — FINAL ANALYTICAL ELIGIBILITY & QUALITY GATE")
print("=" * 70)


# ======================================================================
# 1. LOAD 6M METADATA
# ======================================================================

meta = pd.read_csv(FILE_6M)

print("6M observations:", len(meta))

if "observation_id" not in meta.columns:
    raise ValueError(
        "6M does not contain observation_id."
    )


# ======================================================================
# 2. LOAD 6F SOURCE-VALIDATED RELEASE DATA
# ======================================================================

release = pd.read_csv(FILE_6F)

print(
    "6F source-validated observations:",
    len(release)
)

if "observation_id" not in release.columns:
    raise ValueError(
        "6F does not contain observation_id."
    )


# ======================================================================
# 3. CHECK DUPLICATES
# ======================================================================

if meta["observation_id"].duplicated().any():
    raise ValueError(
        "Duplicate observation_id values detected in 6M."
    )

if release["observation_id"].duplicated().any():
    raise ValueError(
        "Duplicate observation_id values detected in 6F."
    )


# ======================================================================
# 4. REQUIRED 6F RELEASE FIELDS
# ======================================================================

required_release = [
    "observation_id",
    "pmcid",
    "table_index",
    "row_index",
    "release_time_h",
    "release_percent"
]

missing = [
    c for c in required_release
    if c not in release.columns
]

if missing:
    raise ValueError(
        "6F is missing required columns:\n"
        + "\n".join(missing)
    )


# ======================================================================
# 5. KEEP ONLY AUTHORITATIVE RELEASE FIELDS FROM 6F
# ======================================================================

release_cols = [
    "observation_id",
    "pmcid",
    "table_index",
    "row_index",
    "release_time_h",
    "release_percent"
]

for optional_col in [
    "observation_type",
    "time_raw",
    "release_raw",
    "source_reason",
    "source_validation"
]:

    if optional_col in release.columns:
        release_cols.append(optional_col)


release_small = release[
    release_cols
].copy()


# ======================================================================
# 6. REMOVE POTENTIAL COLLISION COLUMNS FROM 6M
# ======================================================================

collision_cols = [
    c for c in release_small.columns
    if (
        c != "observation_id"
        and c in meta.columns
    )
]

if collision_cols:

    print(
        "Removing overlapping metadata columns from 6M:",
        collision_cols
    )

    meta_for_merge = meta.drop(
        columns=collision_cols
    )

else:

    meta_for_merge = meta.copy()


# ======================================================================
# 7. MERGE
# ======================================================================

df = meta_for_merge.merge(
    release_small,
    on="observation_id",
    how="left",
    validate="one_to_one"
)

print(
    "Rows after 6F + 6M consolidation:",
    len(df)
)


# ======================================================================
# 8. CRITICAL COUNT CHECK
# ======================================================================

if len(df) != 60:

    raise ValueError(
        "CRITICAL: Expected 60 observations after merge, "
        f"found {len(df)}."
    )


# ======================================================================
# 9. HELPER FUNCTIONS
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


def identified(x):

    s = clean(x).lower()

    return (
        s != ""
        and s not in [
            "not_identified",
            "unresolved",
            "unknown",
            "none",
            "nan",
            "manual_review"
        ]
    )


# ======================================================================
# 10. NUMERIC RELEASE VARIABLES
# ======================================================================

df["release_time_h_numeric"] = pd.to_numeric(
    df["release_time_h"],
    errors="coerce"
)

df["release_percent_numeric"] = pd.to_numeric(
    df["release_percent"],
    errors="coerce"
)


# ======================================================================
# 11. SOURCE TRACEABILITY
# ======================================================================

df["source_trace_valid"] = (
    df["pmcid"].notna()
    & df["table_index"].notna()
    & df["row_index"].notna()
)


# ======================================================================
# 12. TIME VALIDITY
# ======================================================================

df["time_valid"] = (
    df["release_time_h_numeric"].notna()
    & (
        df["release_time_h_numeric"] >= 0
    )
)


# ======================================================================
# 13. RELEASE RANGE
# ======================================================================

df["release_valid_range"] = (
    df["release_percent_numeric"].notna()
    & (
        df["release_percent_numeric"] >= 0
    )
    & (
        df["release_percent_numeric"] <= 100
    )
)


# ======================================================================
# 14. OBSERVATION-ID DUPLICATES
# ======================================================================

df["duplicate_observation_id"] = (
    df["observation_id"].duplicated(
        keep=False
    )
)


# ======================================================================
# 15. EXACT VALUE DUPLICATES
# ======================================================================

value_key = [
    "pmcid",
    "table_index",
    "row_index",
    "release_time_h_numeric",
    "release_percent_numeric"
]

df["exact_value_duplicate"] = (
    df.duplicated(
        subset=value_key,
        keep=False
    )
)


# ======================================================================
# 16. SOURCE VALIDATION
# ======================================================================

# 6F is already the source-validated branch.
# Therefore source validity is TRUE by provenance.
#
# If source_validation is present, use it only as an audit field.
# Do NOT require source_validation_6F.

df["source_validation_ok"] = True

if "source_validation" in df.columns:

    source_text = (
        df["source_validation"]
        .fillna("")
        .astype(str)
        .str.upper()
    )

    known_valid = source_text.isin([
        "VALID_SOURCE_RELEASE",
        "VALID_RELEASE",
        "EXACT_XML_CELL_MATCH"
    ])

    # Only override TRUE where a recognizable value is present.
    recognized = source_text != ""

    df.loc[
        recognized,
        "source_validation_ok"
    ] = known_valid.loc[
        recognized
    ]


# ======================================================================
# 17. OBSERVATION TYPE
# ======================================================================

if "observation_type" not in df.columns:

    df["observation_type"] = "UNKNOWN"


df["observation_type"] = (
    df["observation_type"]
    .fillna("UNKNOWN")
    .astype(str)
)


df["is_timecourse"] = (
    df["observation_type"]
    == "TIMECOURSE_RELEASE"
)


df["is_max_release_event"] = (
    df["observation_type"]
    == "MAX_RELEASE_EVENT"
)


# ======================================================================
# 18. METADATA FIELDS FROM 6M
# ======================================================================

metadata_columns = {
    "drug": "final_drug_6M",
    "polymer": "final_polymer_6M",
    "carrier": "final_carrier_6M"
}


for col in metadata_columns.values():

    if col not in df.columns:

        df[col] = "NOT_IDENTIFIED"


df["drug_identified"] = (
    df["final_drug_6M"]
    .apply(identified)
)

df["polymer_identified"] = (
    df["final_polymer_6M"]
    .apply(identified)
)

df["carrier_identified"] = (
    df["final_carrier_6M"]
    .apply(identified)
)


# ======================================================================
# 19. FORMULATION / EXPERIMENTAL UNIT
# ======================================================================

formulation_col = None

for candidate in [
    "final_formulation_id",
    "formulation_id",
    "experimental_unit_id",
    "experimental_unit_proxy",
    "experimental_unit_id_6M"
]:

    if candidate in df.columns:

        if df[candidate].apply(
            identified
        ).any():

            formulation_col = candidate
            break


if formulation_col is None:

    df["formulation_identifier_present"] = False

else:

    df["formulation_identifier_present"] = (
        df[formulation_col]
        .apply(identified)
    )


# ======================================================================
# 20. RELEASE CONDITION
# ======================================================================

condition_col = None

for candidate in [
    "final_release_condition",
    "release_condition",
    "condition"
]:

    if candidate in df.columns:

        condition_col = candidate
        break


if condition_col is None:

    df["release_condition_identified"] = False

else:

    df["release_condition_identified"] = (
        df[condition_col]
        .apply(identified)
    )


# ======================================================================
# 21. METADATA COMPLETENESS
# ======================================================================

df["metadata_count"] = (
    df["drug_identified"].astype(int)
    + df["polymer_identified"].astype(int)
    + df["carrier_identified"].astype(int)
)


# ======================================================================
# 22. RELEASE VALIDATION STATUS
# ======================================================================

df["release_validation_status"] = np.select(

    [
        ~df["source_trace_valid"],

        ~df["source_validation_ok"],

        ~df["time_valid"],

        ~df["release_valid_range"],

        df["duplicate_observation_id"]
    ],

    [
        "SOURCE_TRACE_FAILURE",

        "SOURCE_VALIDATION_FAILURE",

        "INVALID_TIME",

        "INVALID_RELEASE",

        "DUPLICATE_OBSERVATION_ID"
    ],

    default="RELEASE_VALID"
)


# ======================================================================
# 23. ANALYTICAL STATUS
# ======================================================================

df["analytical_status"] = np.where(

    df["release_validation_status"]
    == "RELEASE_VALID",

    "ANALYTICAL_READY",

    "NOT_ANALYTICALLY_VALID"
)


# ======================================================================
# 24. CONSERVATIVE ML GATE
# ======================================================================

df["ml_metadata_complete"] = (
    df["release_validation_status"]
    .eq("RELEASE_VALID")
    & df["drug_identified"]
    & (
        df["polymer_identified"]
        | df["carrier_identified"]
    )
    & df["formulation_identifier_present"]
)


df["ml_status"] = np.where(

    df["ml_metadata_complete"],

    "ML_READY",

    "NOT_ML_READY_METADATA"
)


# ======================================================================
# 25. MODELING ROLE
# ======================================================================

df["modeling_role"] = np.select(

    [
        df["is_timecourse"],

        df["is_max_release_event"]
    ],

    [
        "TIMECOURSE_RELEASE",

        "MAX_RELEASE_EVENT"
    ],

    default="OTHER_RELEASE_OBSERVATION"
)


# ======================================================================
# 26. FINAL ELIGIBILITY
# ======================================================================

df["final_eligibility"] = np.select(

    [
        df["release_validation_status"]
        != "RELEASE_VALID",

        df["ml_status"]
        == "ML_READY",

        df["analytical_status"]
        == "ANALYTICAL_READY"
    ],

    [
        "EXCLUDED",

        "ML_READY",

        "ANALYTICAL_READY"
    ],

    default="MANUAL_REVIEW"
)


# ======================================================================
# 27. MANUAL REVIEW REASON
# ======================================================================

def review_reason(row):

    reasons = []

    if not row["drug_identified"]:
        reasons.append(
            "DRUG_NOT_IDENTIFIED"
        )

    if not row["polymer_identified"]:
        reasons.append(
            "POLYMER_NOT_IDENTIFIED"
        )

    if not row["carrier_identified"]:
        reasons.append(
            "CARRIER_NOT_IDENTIFIED"
        )

    if not row[
        "formulation_identifier_present"
    ]:
        reasons.append(
            "FORMULATION_ID_NOT_IDENTIFIED"
        )

    if not row[
        "release_condition_identified"
    ]:
        reasons.append(
            "RELEASE_CONDITION_NOT_IDENTIFIED"
        )

    if not reasons:
        return "NONE"

    return ";".join(reasons)


df["manual_review_reason"] = df.apply(
    review_reason,
    axis=1
)


# ======================================================================
# 28. SAME ARTICLE / TABLE / TIME
# ======================================================================

group_cols = [
    "pmcid",
    "table_index",
    "release_time_h_numeric"
]

df["same_article_table_time_n"] = (
    df.groupby(
        group_cols,
        dropna=False
    )["observation_id"]
    .transform("count")
)


df["same_article_table_time_multiple"] = (
    df["same_article_table_time_n"] > 1
)


# ======================================================================
# 29. FINAL SAFETY CHECKS
# ======================================================================

if len(df) != 60:

    raise ValueError(
        "CRITICAL: Observation count changed."
    )


if df[
    "observation_id"
].duplicated().any():

    raise ValueError(
        "CRITICAL: Duplicate observation IDs."
    )


if (
    df["release_time_h_numeric"] < 0
).any():

    raise ValueError(
        "CRITICAL: Negative release times detected."
    )


bad_release = (
    df["release_percent_numeric"] < 0
) | (
    df["release_percent_numeric"] > 100
)

if bad_release.any():

    raise ValueError(
        "CRITICAL: Release percentage outside 0–100."
    )


# ======================================================================
# 30. OUTPUT DATASETS
# ======================================================================

release_valid = df[
    df["release_validation_status"]
    == "RELEASE_VALID"
].copy()


analytical_ready = df[
    df["final_eligibility"].isin([
        "ANALYTICAL_READY",
        "ML_READY"
    ])
].copy()


ml_ready = df[
    df["final_eligibility"]
    == "ML_READY"
].copy()


manual_review = df[
    df["final_metadata_status"]
    == "MANUAL_METADATA_REVIEW"
].copy()


excluded = df[
    df["final_eligibility"]
    == "EXCLUDED"
].copy()


# ======================================================================
# 31. REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6N-R2 COMPLETE")
print("=" * 70)

print(
    "6M observations:",
    len(meta)
)

print(
    "6F source-validated observations:",
    len(release)
)

print(
    "Merged observations:",
    len(df)
)

print()
print("Release validation:")
print(
    df[
        "release_validation_status"
    ].value_counts()
)

print()
print("Analytical status:")
print(
    df[
        "analytical_status"
    ].value_counts()
)

print()
print("ML status:")
print(
    df[
        "ml_status"
    ].value_counts()
)

print()
print("Final eligibility:")
print(
    df[
        "final_eligibility"
    ].value_counts()
)

print()
print("Observation types:")
print(
    df[
        "observation_type"
    ].value_counts()
)

print()
print("Modeling role:")
print(
    df[
        "modeling_role"
    ].value_counts()
)

print()
print("Drug identified:")
print(
    df[
        "drug_identified"
    ].value_counts()
)

print()
print("Polymer identified:")
print(
    df[
        "polymer_identified"
    ].value_counts()
)

print()
print("Carrier identified:")
print(
    df[
        "carrier_identified"
    ].value_counts()
)

print()
print("Formulation identifier:")
print(
    df[
        "formulation_identifier_present"
    ].value_counts()
)

print()
print("Release condition:")
print(
    df[
        "release_condition_identified"
    ].value_counts()
)

print()
print("Same article/table/time:")
print(
    df[
        "same_article_table_time_multiple"
    ].value_counts()
)

print()
print("FINAL COUNTS")
print("-" * 40)

print(
    "Release-valid:",
    len(release_valid)
)

print(
    "Analytical-ready:",
    len(analytical_ready)
)

print(
    "ML-ready:",
    len(ml_ready)
)

print(
    "Manual metadata review:",
    len(manual_review)
)

print(
    "Excluded:",
    len(excluded)
)

print()
print("Observation preservation:")
print(
    "Input:",
    len(meta)
)

print(
    "Output:",
    len(df)
)


# ======================================================================
# 32. SAVE
# ======================================================================

df.to_csv(
    OUT_AUDIT,
    index=False
)

release_valid.to_csv(
    OUT_RELEASE_VALID,
    index=False
)

analytical_ready.to_csv(
    OUT_ANALYTICAL,
    index=False
)

ml_ready.to_csv(
    OUT_ML,
    index=False
)

manual_review.to_csv(
    OUT_MANUAL,
    index=False
)

excluded.to_csv(
    OUT_EXCLUDED,
    index=False
)


print()
print("Saved:")
print(OUT_AUDIT)
print(OUT_RELEASE_VALID)
print(OUT_ANALYTICAL)
print(OUT_ML)
print(OUT_MANUAL)
print(OUT_EXCLUDED)

print()
print("IMPORTANT:")
print(
    "6F remains the authoritative source for release values."
)

print(
    "6M remains the authoritative source for consolidated metadata."
)

print(
    "The branches are joined only by observation_id."
)

print(
    "No release values were regenerated."
)

print(
    "No observations were interpolated."
)

print(
    "No observations were duplicated."
)

print(
    "Multiple time points remain retained."
)

print(
    "MAX_RELEASE_EVENT observations remain separately labelled."
)

print(
    "ML-ready does NOT imply statistical independence."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6O — TARGETED SOURCE-TABLE FORMULATION RECONSTRUCTION
# ======================================================================
# Purpose:
#   Recover missing Drug / Polymer / Carrier / Release-condition /
#   Formulation metadata for the 60 source-validated 6F observations.
#
# Rules:
#   1. Only the exact PMCID + table_index represented by 6F/6N is searched.
#   2. Code 3 structural cells are the primary evidence.
#   3. Same-table headers and row context may be used.
#   4. Article-level text is NOT used to infer formulation identity.
#   5. No release values are changed.
#   6. No observations are generated.
#   7. No interpolation.
#   8. No duplication.
#   9. Ambiguous metadata remain UNRESOLVED.
# ======================================================================

import os
import re
import numpy as np
import pandas as pd

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")
RAW = os.path.join(ROOT, "data", "raw")

FILE_6N = os.path.join(
    VAL,
    "06N_final_analytical_quality_audit.csv"
)

FILE_6F = os.path.join(
    VAL,
    "06F_valid_source_release.csv"
)

FILE_3 = os.path.join(
    RAW,
    "03_table_cell_records.csv"
)

# Try alternate Code 3 locations if necessary
if not os.path.exists(FILE_3):
    alternatives = [
        os.path.join(ROOT, "data", "processed", "03_table_cell_records.csv"),
        os.path.join(ROOT, "data", "candidates", "03_table_cell_records.csv"),
        os.path.join(ROOT, "data", "intermediate", "03_table_cell_records.csv"),
    ]

    for p in alternatives:
        if os.path.exists(p):
            FILE_3 = p
            break


OUT_AUDIT = os.path.join(
    VAL,
    "06O_targeted_source_table_reconstruction_audit.csv"
)

OUT_READY = os.path.join(
    VAL,
    "06O_recovered_metadata_candidates.csv"
)

OUT_MANUAL = os.path.join(
    VAL,
    "06O_remaining_metadata_review.csv"
)


print("=" * 70)
print("CODE 6O — TARGETED SOURCE-TABLE FORMULATION RECONSTRUCTION")
print("=" * 70)


# ======================================================================
# 1. LOAD 6N
# ======================================================================

df = pd.read_csv(FILE_6N)

print("6N observations:", len(df))


if "observation_id" not in df.columns:
    raise ValueError(
        "06N does not contain observation_id."
    )

if len(df) != 60:
    raise ValueError(
        f"Expected 60 observations from 6N, found {len(df)}."
    )


# ======================================================================
# 2. LOAD CODE 3
# ======================================================================

if not os.path.exists(FILE_3):
    raise FileNotFoundError(
        "Could not find Code 3 table-cell file.\n"
        "Expected one of:\n"
        f"{FILE_3}"
    )


cells = pd.read_csv(FILE_3)

print(
    "Code 3 cell records:",
    len(cells)
)


required_code3 = [
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell",
    "row_text",
    "full_table_text",
    "header_text"
]

missing = [
    c for c in required_code3
    if c not in cells.columns
]

if missing:
    raise ValueError(
        "Code 3 is missing required columns:\n"
        + "\n".join(missing)
    )


# ======================================================================
# 3. NORMALIZATION
# ======================================================================

def norm_pmcid(x):
    s = str(x).strip().upper()

    if not s.startswith("PMC"):
        s = "PMC" + s

    return s


def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


cells["pmcid_norm"] = (
    cells["pmcid"]
    .apply(norm_pmcid)
)

df["pmcid_norm"] = (
    df["pmcid"]
    .apply(norm_pmcid)
)


cells["table_index_num"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

df["table_index_num"] = pd.to_numeric(
    df["table_index"],
    errors="coerce"
)


cells["row_index_num"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

df["row_index_num"] = pd.to_numeric(
    df["row_index"],
    errors="coerce"
)


cells["column_index_num"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)


cells["raw_clean"] = (
    cells["raw_cell"]
    .apply(clean)
)

cells["header_clean"] = (
    cells["header_text"]
    .apply(clean)
)

cells["row_clean"] = (
    cells["row_text"]
    .apply(clean)
)

cells["table_clean"] = (
    cells["full_table_text"]
    .apply(clean)
)


# ======================================================================
# 4. IDENTIFY ONLY SOURCE TABLES USED BY 6N
# ======================================================================

source_keys = (
    df[
        [
            "pmcid_norm",
            "table_index_num"
        ]
    ]
    .drop_duplicates()
    .copy()
)

print()
print(
    "Unique source tables represented in 6N:",
    len(source_keys)
)

print(
    "Source PMCIDs represented:",
    source_keys["pmcid_norm"]
    .nunique()
)


# ======================================================================
# 5. FILTER CODE 3 TO ONLY THESE SOURCE TABLES
# ======================================================================

keyed_cells = cells.merge(
    source_keys,
    on=[
        "pmcid_norm",
        "table_index_num"
    ],
    how="inner"
)

print(
    "Code 3 cells belonging to validated source tables:",
    len(keyed_cells)
)


# ======================================================================
# 6. KEYWORD DICTIONARIES
# ======================================================================

drug_terms = {
    "doxorubicin": [
        "doxorubicin",
        "DOX"
    ],
    "5-fluorouracil": [
        "5-fluorouracil",
        "5-fluorouracil",
        "5-fu",
        "5fu"
    ],
    "paclitaxel": [
        "paclitaxel"
    ],
    "curcumin": [
        "curcumin"
    ],
    "dexamethasone": [
        "dexamethasone"
    ],
    "ibuprofen": [
        "ibuprofen"
    ],
    "metformin": [
        "metformin"
    ],
    "gemcitabine": [
        "gemcitabine"
    ],
    "rifampicin": [
        "rifampicin"
    ],
    "vancomycin": [
        "vancomycin"
    ]
}


polymer_terms = {
    "PLGA": [
        "PLGA",
        "poly(lactic-co-glycolic acid)"
    ],
    "PLA": [
        "PLA",
        "polylactic acid",
        "poly(lactic acid)"
    ],
    "PCL": [
        "PCL",
        "polycaprolactone"
    ],
    "PEG": [
        "PEG",
        "polyethylene glycol"
    ],
    "PVA": [
        "PVA",
        "polyvinyl alcohol"
    ],
    "PVP": [
        "PVP",
        "polyvinylpyrrolidone"
    ],
    "PEO": [
        "PEO",
        "polyethylene oxide"
    ],
    "chitosan": [
        "chitosan"
    ],
    "alginate": [
        "alginate",
        "sodium alginate"
    ],
    "cellulose": [
        "cellulose"
    ],
    "gelatin": [
        "gelatin"
    ],
    "poloxamer": [
        "poloxamer"
    ],
    "polyurethane": [
        "polyurethane"
    ]
}


carrier_terms = {
    "nanoparticle": [
        "nanoparticle",
        "nanoparticles"
    ],
    "hydrogel": [
        "hydrogel",
        "hydrogels"
    ],
    "microparticle": [
        "microparticle",
        "microparticles"
    ],
    "microsphere": [
        "microsphere",
        "microspheres"
    ],
    "liposome": [
        "liposome",
        "liposomes"
    ],
    "nanofiber": [
        "nanofiber",
        "nanofibers"
    ],
    "film": [
        "film",
        "films"
    ],
    "bead": [
        "bead",
        "beads"
    ],
    "implant": [
        "implant",
        "implants"
    ],
    "micelle": [
        "micelle",
        "micelles"
    ]
}


condition_patterns = [
    r"\bpH\s*[0-9]+(?:\.[0-9]+)?",
    r"\bpH\s*=\s*[0-9]+(?:\.[0-9]+)?",
    r"\bpH\s*[0-9]+(?:\.[0-9]+)?\s*[-–]\s*[0-9]+(?:\.[0-9]+)?",
]


# ======================================================================
# 7. HELPERS
# ======================================================================

def find_terms(text, dictionary):

    text = clean(text)

    if not text:
        return []

    found = []

    lower = text.lower()

    for canonical, terms in dictionary.items():

        for term in terms:

            if term.lower() in lower:

                found.append(canonical)
                break

    return sorted(set(found))


def find_conditions(text):

    text = clean(text)

    if not text:
        return []

    found = []

    for pattern in condition_patterns:

        matches = re.findall(
            pattern,
            text,
            flags=re.I
        )

        found.extend(matches)

    # Normalize
    output = []

    for x in found:

        x = re.sub(
            r"\s+",
            " ",
            x
        ).strip()

        if x not in output:
            output.append(x)

    return output


def choose_unique(values):

    values = [
        clean(x)
        for x in values
        if clean(x)
    ]

    values = list(
        dict.fromkeys(values)
    )

    if len(values) == 1:
        return values[0]

    if len(values) == 0:
        return "NOT_IDENTIFIED"

    return "AMBIGUOUS:" + "|".join(
        values
    )


# ======================================================================
# 8. BUILD TABLE-LEVEL EVIDENCE
# ======================================================================

table_evidence = {}

for (
    pmcid,
    table_index
), sub in keyed_cells.groupby(
    [
        "pmcid_norm",
        "table_index_num"
    ]
):

    # All structural text from the table
    table_text = " ".join(
        sub[
            [
                "header_clean",
                "row_clean",
                "table_clean"
            ]
        ]
        .fillna("")
        .astype(str)
        .values
        .flatten()
    )

    drugs = find_terms(
        table_text,
        drug_terms
    )

    polymers = find_terms(
        table_text,
        polymer_terms
    )

    carriers = find_terms(
        table_text,
        carrier_terms
    )

    conditions = find_conditions(
        table_text
    )

    table_evidence[
        (
            pmcid,
            table_index
        )
    ] = {
        "drugs": drugs,
        "polymers": polymers,
        "carriers": carriers,
        "conditions": conditions
    }


# ======================================================================
# 9. OBSERVATION-LEVEL STRUCTURAL RECOVERY
# ======================================================================

records = []


for idx, obs in df.iterrows():

    pmcid = obs["pmcid_norm"]
    table = obs["table_index_num"]

    key = (
        pmcid,
        table
    )

    sub = keyed_cells[
        (
            keyed_cells["pmcid_norm"]
            == pmcid
        )
        &
        (
            keyed_cells["table_index_num"]
            == table
        )
    ].copy()

    # --------------------------------------------------------------
    # Locate exact release row
    # --------------------------------------------------------------

    if (
        pd.notna(obs.get("row_index_num"))
    ):

        row_cells = sub[
            sub["row_index_num"]
            == obs["row_index_num"]
        ].copy()

    else:

        row_cells = pd.DataFrame(
            columns=sub.columns
        )


    # --------------------------------------------------------------
    # Exact row text
    # --------------------------------------------------------------

    exact_row_text = ""

    if len(row_cells) > 0:

        exact_row_text = " ".join(
            row_cells[
                [
                    "raw_clean",
                    "row_clean",
                    "header_clean"
                ]
            ]
            .fillna("")
            .astype(str)
            .values
            .flatten()
        )


    # --------------------------------------------------------------
    # Nearby same-table evidence
    # --------------------------------------------------------------

    # This is evidence only.
    # It is NOT automatically accepted as formulation identity.

    same_table_text = " ".join(
        sub[
            [
                "header_clean",
                "row_clean"
            ]
        ]
        .fillna("")
        .astype(str)
        .values
        .flatten()
    )


    # --------------------------------------------------------------
    # Extract exact-row metadata
    # --------------------------------------------------------------

    row_drugs = find_terms(
        exact_row_text,
        drug_terms
    )

    row_polymers = find_terms(
        exact_row_text,
        polymer_terms
    )

    row_carriers = find_terms(
        exact_row_text,
        carrier_terms
    )

    row_conditions = find_conditions(
        exact_row_text
    )


    # --------------------------------------------------------------
    # Extract same-table metadata
    # --------------------------------------------------------------

    table_info = table_evidence.get(
        key,
        {
            "drugs": [],
            "polymers": [],
            "carriers": [],
            "conditions": []
        }
    )


    # --------------------------------------------------------------
    # Conservative decision rules
    # --------------------------------------------------------------

    # Drug
    if len(row_drugs) == 1:

        recovered_drug = row_drugs[0]
        drug_source = "EXACT_ROW"

    elif len(row_drugs) > 1:

        recovered_drug = (
            "AMBIGUOUS:" +
            "|".join(row_drugs)
        )
        drug_source = "EXACT_ROW_AMBIGUOUS"

    else:

        recovered_drug = (
            "NOT_IDENTIFIED"
        )
        drug_source = "UNRESOLVED"


    # Polymer
    if len(row_polymers) == 1:

        recovered_polymer = row_polymers[0]
        polymer_source = "EXACT_ROW"

    elif len(row_polymers) > 1:

        recovered_polymer = (
            "AMBIGUOUS:" +
            "|".join(row_polymers)
        )
        polymer_source = "EXACT_ROW_AMBIGUOUS"

    else:

        recovered_polymer = (
            "NOT_IDENTIFIED"
        )
        polymer_source = "UNRESOLVED"


    # Carrier
    if len(row_carriers) == 1:

        recovered_carrier = row_carriers[0]
        carrier_source = "EXACT_ROW"

    elif len(row_carriers) > 1:

        recovered_carrier = (
            "AMBIGUOUS:" +
            "|".join(row_carriers)
        )
        carrier_source = "EXACT_ROW_AMBIGUOUS"

    else:

        recovered_carrier = (
            "NOT_IDENTIFIED"
        )
        carrier_source = "UNRESOLVED"


    # Release condition
    if len(row_conditions) == 1:

        recovered_condition = (
            row_conditions[0]
        )

        condition_source = "EXACT_ROW"

    elif len(row_conditions) > 1:

        recovered_condition = (
            "AMBIGUOUS:" +
            "|".join(row_conditions)
        )

        condition_source = (
            "EXACT_ROW_AMBIGUOUS"
        )

    else:

        recovered_condition = (
            "NOT_IDENTIFIED"
        )

        condition_source = "UNRESOLVED"


    # --------------------------------------------------------------
    # Table-level evidence is recorded separately.
    #
    # IMPORTANT:
    # It is NOT promoted to final formulation metadata.
    # --------------------------------------------------------------

    table_drugs = table_info["drugs"]
    table_polymers = table_info["polymers"]
    table_carriers = table_info["carriers"]
    table_conditions = table_info["conditions"]


    # --------------------------------------------------------------
    # Strong recovery flag
    # --------------------------------------------------------------

    strong_fields = sum([
        recovered_drug != "NOT_IDENTIFIED"
        and not recovered_drug.startswith(
            "AMBIGUOUS:"
        ),

        recovered_polymer != "NOT_IDENTIFIED"
        and not recovered_polymer.startswith(
            "AMBIGUOUS:"
        ),

        recovered_carrier != "NOT_IDENTIFIED"
        and not recovered_carrier.startswith(
            "AMBIGUOUS:"
        )
    ])


    if strong_fields >= 2:

        recovery_status = (
            "STRONG_ROW_RECOVERY"
        )

    elif strong_fields == 1:

        recovery_status = (
            "PARTIAL_ROW_RECOVERY"
        )

    else:

        recovery_status = (
            "UNRESOLVED"
        )


    records.append({

        "observation_id":
            obs["observation_id"],

        "pmcid":
            pmcid,

        "table_index":
            table,

        "row_index":
            obs.get(
                "row_index_num",
                np.nan
            ),

        "release_time_h":
            obs.get(
                "release_time_h",
                np.nan
            ),

        "release_percent":
            obs.get(
                "release_percent",
                np.nan
            ),

        "observation_type":
            obs.get(
                "observation_type",
                "UNKNOWN"
            ),

        "exact_row_text":
            exact_row_text,

        "recovered_drug":
            recovered_drug,

        "drug_source":
            drug_source,

        "recovered_polymer":
            recovered_polymer,

        "polymer_source":
            polymer_source,

        "recovered_carrier":
            recovered_carrier,

        "carrier_source":
            carrier_source,

        "recovered_release_condition":
            recovered_condition,

        "condition_source":
            condition_source,

        "same_table_drugs":
            ";".join(table_drugs),

        "same_table_polymers":
            ";".join(table_polymers),

        "same_table_carriers":
            ";".join(table_carriers),

        "same_table_conditions":
            ";".join(table_conditions),

        "recovery_status":
            recovery_status,

        "strong_metadata_fields":
            strong_fields,

        "source_table_evidence_available":
            len(sub) > 0

    })


# ======================================================================
# 10. CREATE RESULT
# ======================================================================

result = pd.DataFrame(
    records
)


# ======================================================================
# 11. STRICT READY RULE
# ======================================================================
# For this stage, metadata are considered recovered only when:
#
#   - exact-row evidence exists
#   - no ambiguity exists
#   - at least TWO formulation fields are recovered
#
# We do NOT require every field to be present.
#
# This is intentionally conservative.

result["metadata_ready_6O"] = (
    result["recovery_status"]
    == "STRONG_ROW_RECOVERY"
)


result["manual_review_6O"] = (
    ~result["metadata_ready_6O"]
)


# ======================================================================
# 12. SAFETY CHECKS
# ======================================================================

if len(result) != 60:

    raise ValueError(
        f"Observation count changed: "
        f"{len(result)} instead of 60."
    )


if result[
    "observation_id"
].duplicated().any():

    raise ValueError(
        "Duplicate observation IDs detected."
    )


# ======================================================================
# 13. SUMMARY
# ======================================================================

print()
print("=" * 70)
print("CODE 6O COMPLETE")
print("=" * 70)

print(
    "Input 6N observations:",
    len(df)
)

print(
    "Targeted source tables:",
    len(source_keys)
)

print(
    "Code 3 cells searched:",
    len(keyed_cells)
)

print()
print("Recovery status:")
print(
    result[
        "recovery_status"
    ].value_counts()
)

print()
print("Metadata-ready:")
print(
    result[
        "metadata_ready_6O"
    ].value_counts()
)

print()
print("Drug recovery:")
print(
    result[
        "recovered_drug"
    ].value_counts()
)

print()
print("Polymer recovery:")
print(
    result[
        "recovered_polymer"
    ].value_counts()
)

print()
print("Carrier recovery:")
print(
    result[
        "recovered_carrier"
    ].value_counts()
)

print()
print("Release-condition recovery:")
print(
    result[
        "recovered_release_condition"
    ].value_counts()
)

print()
print("Recovery sources:")
print(
    result[
        "drug_source"
    ].value_counts()
)

print()
print("Strong metadata observations:",
      int(result["metadata_ready_6O"].sum()))

print(
    "Remaining manual review:",
    int(result["manual_review_6O"].sum())
)


# ======================================================================
# 14. SAVE
# ======================================================================

result.to_csv(
    OUT_AUDIT,
    index=False
)

result[
    result["metadata_ready_6O"]
].to_csv(
    OUT_READY,
    index=False
)

result[
    result["manual_review_6O"]
].to_csv(
    OUT_MANUAL,
    index=False
)


print()
print("Saved:")
print(OUT_AUDIT)
print(OUT_READY)
print(OUT_MANUAL)

print()
print("IMPORTANT:")
print(
    "Only the six validated source-article/table groups "
    "were searched."
)

print(
    "Exact-row evidence has priority."
)

print(
    "Same-table evidence is retained as evidence only "
    "and is NOT automatically promoted to formulation identity."
)

print(
    "No unrelated article text was used."
)

print(
    "No release values were changed."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6O — TARGETED SOURCE-TABLE FORMULATION RECONSTRUCTION
# ======================================================================
# Purpose:
#   Recover missing Drug / Polymer / Carrier / Release-condition /
#   Formulation metadata for the 60 source-validated 6F observations.
#
# Rules:
#   1. Only the exact PMCID + table_index represented by 6F/6N is searched.
#   2. Code 3 structural cells are the primary evidence.
#   3. Same-table headers and row context may be used.
#   4. Article-level text is NOT used to infer formulation identity.
#   5. No release values are changed.
#   6. No observations are generated.
#   7. No interpolation.
#   8. No duplication.
#   9. Ambiguous metadata remain UNRESOLVED.
# ======================================================================

import os
import re
import numpy as np
import pandas as pd

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")
RAW = os.path.join(ROOT, "data", "raw")

FILE_6N = os.path.join(
    VAL,
    "06N_final_analytical_quality_audit.csv"
)

FILE_6F = os.path.join(
    VAL,
    "06F_valid_source_release.csv"
)

FILE_3 = os.path.join(
    RAW,
    "03_table_cell_records.csv"
)

# Try alternate Code 3 locations if necessary
if not os.path.exists(FILE_3):
    alternatives = [
        os.path.join(ROOT, "data", "processed", "03_table_cell_records.csv"),
        os.path.join(ROOT, "data", "candidates", "03_table_cell_records.csv"),
        os.path.join(ROOT, "data", "intermediate", "03_table_cell_records.csv"),
    ]

    for p in alternatives:
        if os.path.exists(p):
            FILE_3 = p
            break


OUT_AUDIT = os.path.join(
    VAL,
    "06O_targeted_source_table_reconstruction_audit.csv"
)

OUT_READY = os.path.join(
    VAL,
    "06O_recovered_metadata_candidates.csv"
)

OUT_MANUAL = os.path.join(
    VAL,
    "06O_remaining_metadata_review.csv"
)


print("=" * 70)
print("CODE 6O — TARGETED SOURCE-TABLE FORMULATION RECONSTRUCTION")
print("=" * 70)


# ======================================================================
# 1. LOAD 6N
# ======================================================================

df = pd.read_csv(FILE_6N)

print("6N observations:", len(df))


if "observation_id" not in df.columns:
    raise ValueError(
        "06N does not contain observation_id."
    )

if len(df) != 60:
    raise ValueError(
        f"Expected 60 observations from 6N, found {len(df)}."
    )


# ======================================================================
# 2. LOAD CODE 3
# ======================================================================

if not os.path.exists(FILE_3):
    raise FileNotFoundError(
        "Could not find Code 3 table-cell file.\n"
        "Expected one of:\n"
        f"{FILE_3}"
    )


cells = pd.read_csv(FILE_3)

print(
    "Code 3 cell records:",
    len(cells)
)


required_code3 = [
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell",
    "row_text",
    "full_table_text",
    "header_text"
]

missing = [
    c for c in required_code3
    if c not in cells.columns
]

if missing:
    raise ValueError(
        "Code 3 is missing required columns:\n"
        + "\n".join(missing)
    )


# ======================================================================
# 3. NORMALIZATION
# ======================================================================

def norm_pmcid(x):
    s = str(x).strip().upper()

    if not s.startswith("PMC"):
        s = "PMC" + s

    return s


def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


cells["pmcid_norm"] = (
    cells["pmcid"]
    .apply(norm_pmcid)
)

df["pmcid_norm"] = (
    df["pmcid"]
    .apply(norm_pmcid)
)


cells["table_index_num"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)

df["table_index_num"] = pd.to_numeric(
    df["table_index"],
    errors="coerce"
)


cells["row_index_num"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

df["row_index_num"] = pd.to_numeric(
    df["row_index"],
    errors="coerce"
)


cells["column_index_num"] = pd.to_numeric(
    cells["column_index"],
    errors="coerce"
)


cells["raw_clean"] = (
    cells["raw_cell"]
    .apply(clean)
)

cells["header_clean"] = (
    cells["header_text"]
    .apply(clean)
)

cells["row_clean"] = (
    cells["row_text"]
    .apply(clean)
)

cells["table_clean"] = (
    cells["full_table_text"]
    .apply(clean)
)


# ======================================================================
# 4. IDENTIFY ONLY SOURCE TABLES USED BY 6N
# ======================================================================

source_keys = (
    df[
        [
            "pmcid_norm",
            "table_index_num"
        ]
    ]
    .drop_duplicates()
    .copy()
)

print()
print(
    "Unique source tables represented in 6N:",
    len(source_keys)
)

print(
    "Source PMCIDs represented:",
    source_keys["pmcid_norm"]
    .nunique()
)


# ======================================================================
# 5. FILTER CODE 3 TO ONLY THESE SOURCE TABLES
# ======================================================================

keyed_cells = cells.merge(
    source_keys,
    on=[
        "pmcid_norm",
        "table_index_num"
    ],
    how="inner"
)

print(
    "Code 3 cells belonging to validated source tables:",
    len(keyed_cells)
)


# ======================================================================
# 6. KEYWORD DICTIONARIES
# ======================================================================

drug_terms = {
    "doxorubicin": [
        "doxorubicin",
        "DOX"
    ],
    "5-fluorouracil": [
        "5-fluorouracil",
        "5-fluorouracil",
        "5-fu",
        "5fu"
    ],
    "paclitaxel": [
        "paclitaxel"
    ],
    "curcumin": [
        "curcumin"
    ],
    "dexamethasone": [
        "dexamethasone"
    ],
    "ibuprofen": [
        "ibuprofen"
    ],
    "metformin": [
        "metformin"
    ],
    "gemcitabine": [
        "gemcitabine"
    ],
    "rifampicin": [
        "rifampicin"
    ],
    "vancomycin": [
        "vancomycin"
    ]
}


polymer_terms = {
    "PLGA": [
        "PLGA",
        "poly(lactic-co-glycolic acid)"
    ],
    "PLA": [
        "PLA",
        "polylactic acid",
        "poly(lactic acid)"
    ],
    "PCL": [
        "PCL",
        "polycaprolactone"
    ],
    "PEG": [
        "PEG",
        "polyethylene glycol"
    ],
    "PVA": [
        "PVA",
        "polyvinyl alcohol"
    ],
    "PVP": [
        "PVP",
        "polyvinylpyrrolidone"
    ],
    "PEO": [
        "PEO",
        "polyethylene oxide"
    ],
    "chitosan": [
        "chitosan"
    ],
    "alginate": [
        "alginate",
        "sodium alginate"
    ],
    "cellulose": [
        "cellulose"
    ],
    "gelatin": [
        "gelatin"
    ],
    "poloxamer": [
        "poloxamer"
    ],
    "polyurethane": [
        "polyurethane"
    ]
}


carrier_terms = {
    "nanoparticle": [
        "nanoparticle",
        "nanoparticles"
    ],
    "hydrogel": [
        "hydrogel",
        "hydrogels"
    ],
    "microparticle": [
        "microparticle",
        "microparticles"
    ],
    "microsphere": [
        "microsphere",
        "microspheres"
    ],
    "liposome": [
        "liposome",
        "liposomes"
    ],
    "nanofiber": [
        "nanofiber",
        "nanofibers"
    ],
    "film": [
        "film",
        "films"
    ],
    "bead": [
        "bead",
        "beads"
    ],
    "implant": [
        "implant",
        "implants"
    ],
    "micelle": [
        "micelle",
        "micelles"
    ]
}


condition_patterns = [
    r"\bpH\s*[0-9]+(?:\.[0-9]+)?",
    r"\bpH\s*=\s*[0-9]+(?:\.[0-9]+)?",
    r"\bpH\s*[0-9]+(?:\.[0-9]+)?\s*[-–]\s*[0-9]+(?:\.[0-9]+)?",
]


# ======================================================================
# 7. HELPERS
# ======================================================================

def find_terms(text, dictionary):

    text = clean(text)

    if not text:
        return []

    found = []

    lower = text.lower()

    for canonical, terms in dictionary.items():

        for term in terms:

            if term.lower() in lower:

                found.append(canonical)
                break

    return sorted(set(found))


def find_conditions(text):

    text = clean(text)

    if not text:
        return []

    found = []

    for pattern in condition_patterns:

        matches = re.findall(
            pattern,
            text,
            flags=re.I
        )

        found.extend(matches)

    # Normalize
    output = []

    for x in found:

        x = re.sub(
            r"\s+",
            " ",
            x
        ).strip()

        if x not in output:
            output.append(x)

    return output


def choose_unique(values):

    values = [
        clean(x)
        for x in values
        if clean(x)
    ]

    values = list(
        dict.fromkeys(values)
    )

    if len(values) == 1:
        return values[0]

    if len(values) == 0:
        return "NOT_IDENTIFIED"

    return "AMBIGUOUS:" + "|".join(
        values
    )


# ======================================================================
# 8. BUILD TABLE-LEVEL EVIDENCE
# ======================================================================

table_evidence = {}

for (
    pmcid,
    table_index
), sub in keyed_cells.groupby(
    [
        "pmcid_norm",
        "table_index_num"
    ]
):

    # All structural text from the table
    table_text = " ".join(
        sub[
            [
                "header_clean",
                "row_clean",
                "table_clean"
            ]
        ]
        .fillna("")
        .astype(str)
        .values
        .flatten()
    )

    drugs = find_terms(
        table_text,
        drug_terms
    )

    polymers = find_terms(
        table_text,
        polymer_terms
    )

    carriers = find_terms(
        table_text,
        carrier_terms
    )

    conditions = find_conditions(
        table_text
    )

    table_evidence[
        (
            pmcid,
            table_index
        )
    ] = {
        "drugs": drugs,
        "polymers": polymers,
        "carriers": carriers,
        "conditions": conditions
    }


# ======================================================================
# 9. OBSERVATION-LEVEL STRUCTURAL RECOVERY
# ======================================================================

records = []


for idx, obs in df.iterrows():

    pmcid = obs["pmcid_norm"]
    table = obs["table_index_num"]

    key = (
        pmcid,
        table
    )

    sub = keyed_cells[
        (
            keyed_cells["pmcid_norm"]
            == pmcid
        )
        &
        (
            keyed_cells["table_index_num"]
            == table
        )
    ].copy()

    # --------------------------------------------------------------
    # Locate exact release row
    # --------------------------------------------------------------

    if (
        pd.notna(obs.get("row_index_num"))
    ):

        row_cells = sub[
            sub["row_index_num"]
            == obs["row_index_num"]
        ].copy()

    else:

        row_cells = pd.DataFrame(
            columns=sub.columns
        )


    # --------------------------------------------------------------
    # Exact row text
    # --------------------------------------------------------------

    exact_row_text = ""

    if len(row_cells) > 0:

        exact_row_text = " ".join(
            row_cells[
                [
                    "raw_clean",
                    "row_clean",
                    "header_clean"
                ]
            ]
            .fillna("")
            .astype(str)
            .values
            .flatten()
        )


    # --------------------------------------------------------------
    # Nearby same-table evidence
    # --------------------------------------------------------------

    # This is evidence only.
    # It is NOT automatically accepted as formulation identity.

    same_table_text = " ".join(
        sub[
            [
                "header_clean",
                "row_clean"
            ]
        ]
        .fillna("")
        .astype(str)
        .values
        .flatten()
    )


    # --------------------------------------------------------------
    # Extract exact-row metadata
    # --------------------------------------------------------------

    row_drugs = find_terms(
        exact_row_text,
        drug_terms
    )

    row_polymers = find_terms(
        exact_row_text,
        polymer_terms
    )

    row_carriers = find_terms(
        exact_row_text,
        carrier_terms
    )

    row_conditions = find_conditions(
        exact_row_text
    )


    # --------------------------------------------------------------
    # Extract same-table metadata
    # --------------------------------------------------------------

    table_info = table_evidence.get(
        key,
        {
            "drugs": [],
            "polymers": [],
            "carriers": [],
            "conditions": []
        }
    )


    # --------------------------------------------------------------
    # Conservative decision rules
    # --------------------------------------------------------------

    # Drug
    if len(row_drugs) == 1:

        recovered_drug = row_drugs[0]
        drug_source = "EXACT_ROW"

    elif len(row_drugs) > 1:

        recovered_drug = (
            "AMBIGUOUS:" +
            "|".join(row_drugs)
        )
        drug_source = "EXACT_ROW_AMBIGUOUS"

    else:

        recovered_drug = (
            "NOT_IDENTIFIED"
        )
        drug_source = "UNRESOLVED"


    # Polymer
    if len(row_polymers) == 1:

        recovered_polymer = row_polymers[0]
        polymer_source = "EXACT_ROW"

    elif len(row_polymers) > 1:

        recovered_polymer = (
            "AMBIGUOUS:" +
            "|".join(row_polymers)
        )
        polymer_source = "EXACT_ROW_AMBIGUOUS"

    else:

        recovered_polymer = (
            "NOT_IDENTIFIED"
        )
        polymer_source = "UNRESOLVED"


    # Carrier
    if len(row_carriers) == 1:

        recovered_carrier = row_carriers[0]
        carrier_source = "EXACT_ROW"

    elif len(row_carriers) > 1:

        recovered_carrier = (
            "AMBIGUOUS:" +
            "|".join(row_carriers)
        )
        carrier_source = "EXACT_ROW_AMBIGUOUS"

    else:

        recovered_carrier = (
            "NOT_IDENTIFIED"
        )
        carrier_source = "UNRESOLVED"


    # Release condition
    if len(row_conditions) == 1:

        recovered_condition = (
            row_conditions[0]
        )

        condition_source = "EXACT_ROW"

    elif len(row_conditions) > 1:

        recovered_condition = (
            "AMBIGUOUS:" +
            "|".join(row_conditions)
        )

        condition_source = (
            "EXACT_ROW_AMBIGUOUS"
        )

    else:

        recovered_condition = (
            "NOT_IDENTIFIED"
        )

        condition_source = "UNRESOLVED"


    # --------------------------------------------------------------
    # Table-level evidence is recorded separately.
    #
    # IMPORTANT:
    # It is NOT promoted to final formulation metadata.
    # --------------------------------------------------------------

    table_drugs = table_info["drugs"]
    table_polymers = table_info["polymers"]
    table_carriers = table_info["carriers"]
    table_conditions = table_info["conditions"]


    # --------------------------------------------------------------
    # Strong recovery flag
    # --------------------------------------------------------------

    strong_fields = sum([
        recovered_drug != "NOT_IDENTIFIED"
        and not recovered_drug.startswith(
            "AMBIGUOUS:"
        ),

        recovered_polymer != "NOT_IDENTIFIED"
        and not recovered_polymer.startswith(
            "AMBIGUOUS:"
        ),

        recovered_carrier != "NOT_IDENTIFIED"
        and not recovered_carrier.startswith(
            "AMBIGUOUS:"
        )
    ])


    if strong_fields >= 2:

        recovery_status = (
            "STRONG_ROW_RECOVERY"
        )

    elif strong_fields == 1:

        recovery_status = (
            "PARTIAL_ROW_RECOVERY"
        )

    else:

        recovery_status = (
            "UNRESOLVED"
        )


    records.append({

        "observation_id":
            obs["observation_id"],

        "pmcid":
            pmcid,

        "table_index":
            table,

        "row_index":
            obs.get(
                "row_index_num",
                np.nan
            ),

        "release_time_h":
            obs.get(
                "release_time_h",
                np.nan
            ),

        "release_percent":
            obs.get(
                "release_percent",
                np.nan
            ),

        "observation_type":
            obs.get(
                "observation_type",
                "UNKNOWN"
            ),

        "exact_row_text":
            exact_row_text,

        "recovered_drug":
            recovered_drug,

        "drug_source":
            drug_source,

        "recovered_polymer":
            recovered_polymer,

        "polymer_source":
            polymer_source,

        "recovered_carrier":
            recovered_carrier,

        "carrier_source":
            carrier_source,

        "recovered_release_condition":
            recovered_condition,

        "condition_source":
            condition_source,

        "same_table_drugs":
            ";".join(table_drugs),

        "same_table_polymers":
            ";".join(table_polymers),

        "same_table_carriers":
            ";".join(table_carriers),

        "same_table_conditions":
            ";".join(table_conditions),

        "recovery_status":
            recovery_status,

        "strong_metadata_fields":
            strong_fields,

        "source_table_evidence_available":
            len(sub) > 0

    })


# ======================================================================
# 10. CREATE RESULT
# ======================================================================

result = pd.DataFrame(
    records
)


# ======================================================================
# 11. STRICT READY RULE
# ======================================================================
# For this stage, metadata are considered recovered only when:
#
#   - exact-row evidence exists
#   - no ambiguity exists
#   - at least TWO formulation fields are recovered
#
# We do NOT require every field to be present.
#
# This is intentionally conservative.

result["metadata_ready_6O"] = (
    result["recovery_status"]
    == "STRONG_ROW_RECOVERY"
)


result["manual_review_6O"] = (
    ~result["metadata_ready_6O"]
)


# ======================================================================
# 12. SAFETY CHECKS
# ======================================================================

if len(result) != 60:

    raise ValueError(
        f"Observation count changed: "
        f"{len(result)} instead of 60."
    )


if result[
    "observation_id"
].duplicated().any():

    raise ValueError(
        "Duplicate observation IDs detected."
    )


# ======================================================================
# 13. SUMMARY
# ======================================================================

print()
print("=" * 70)
print("CODE 6O COMPLETE")
print("=" * 70)

print(
    "Input 6N observations:",
    len(df)
)

print(
    "Targeted source tables:",
    len(source_keys)
)

print(
    "Code 3 cells searched:",
    len(keyed_cells)
)

print()
print("Recovery status:")
print(
    result[
        "recovery_status"
    ].value_counts()
)

print()
print("Metadata-ready:")
print(
    result[
        "metadata_ready_6O"
    ].value_counts()
)

print()
print("Drug recovery:")
print(
    result[
        "recovered_drug"
    ].value_counts()
)

print()
print("Polymer recovery:")
print(
    result[
        "recovered_polymer"
    ].value_counts()
)

print()
print("Carrier recovery:")
print(
    result[
        "recovered_carrier"
    ].value_counts()
)

print()
print("Release-condition recovery:")
print(
    result[
        "recovered_release_condition"
    ].value_counts()
)

print()
print("Recovery sources:")
print(
    result[
        "drug_source"
    ].value_counts()
)

print()
print("Strong metadata observations:",
      int(result["metadata_ready_6O"].sum()))

print(
    "Remaining manual review:",
    int(result["manual_review_6O"].sum())
)


# ======================================================================
# 14. SAVE
# ======================================================================

result.to_csv(
    OUT_AUDIT,
    index=False
)

result[
    result["metadata_ready_6O"]
].to_csv(
    OUT_READY,
    index=False
)

result[
    result["manual_review_6O"]
].to_csv(
    OUT_MANUAL,
    index=False
)


print()
print("Saved:")
print(OUT_AUDIT)
print(OUT_READY)
print(OUT_MANUAL)

print()
print("IMPORTANT:")
print(
    "Only the six validated source-article/table groups "
    "were searched."
)

print(
    "Exact-row evidence has priority."
)

print(
    "Same-table evidence is retained as evidence only "
    "and is NOT automatically promoted to formulation identity."
)

print(
    "No unrelated article text was used."
)

print(
    "No release values were changed."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6O — FIXED CODE 3 FILE DISCOVERY
# ======================================================================

import os
import glob
import pandas as pd

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")

FILE_6N = os.path.join(
    VAL,
    "06N_final_analytical_quality_audit.csv"
)

# ----------------------------------------------------------------------
# 1. LOAD 6N
# ----------------------------------------------------------------------

df = pd.read_csv(FILE_6N)

print("=" * 70)
print("CODE 6O — TARGETED SOURCE-TABLE FORMULATION RECONSTRUCTION")
print("=" * 70)

print(
    "6N observations:",
    len(df)
)

if "observation_id" not in df.columns:
    raise ValueError(
        "06N does not contain observation_id."
    )

if len(df) != 60:
    raise ValueError(
        f"Expected 60 observations from 6N, found {len(df)}."
    )


# ----------------------------------------------------------------------
# 2. AUTOMATICALLY DISCOVER CODE 3
# ----------------------------------------------------------------------

print()
print("Searching for Code 3 structural-cell dataset...")


required_code3 = {
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell"
}


candidate_files = []

for pattern in [
    "**/*.csv",
    "**/*.CSV"
]:

    candidate_files.extend(
        glob.glob(
            os.path.join(
                ROOT,
                pattern
            ),
            recursive=True
        )
    )


# Remove duplicates
candidate_files = sorted(
    set(candidate_files)
)


code3_matches = []


for path in candidate_files:

    try:

        # Read header only
        cols = set(
            pd.read_csv(
                path,
                nrows=0
            ).columns
        )

        if required_code3.issubset(cols):

            code3_matches.append(
                path
            )

    except Exception:
        continue


print(
    "Potential Code 3 files found:",
    len(code3_matches)
)


if len(code3_matches) == 0:

    raise FileNotFoundError(
        "\nCould not automatically locate the Code 3 "
        "structural-cell dataset.\n\n"
        "The file must contain these columns:\n"
        + "\n".join(
            sorted(required_code3)
        )
        + "\n\n"
        "Search root:\n"
        + ROOT
    )


# ----------------------------------------------------------------------
# 3. SCORE POSSIBLE FILES
# ----------------------------------------------------------------------

def score_code3_file(path):

    name = os.path.basename(
        path
    ).lower()

    score = 0

    preferred_terms = [
        "03",
        "code3",
        "code_3",
        "table_cell",
        "cell_records",
        "structural",
        "table_cells"
    ]

    for term in preferred_terms:

        if term in name:
            score += 1

    return score


code3_matches = sorted(
    code3_matches,
    key=score_code3_file,
    reverse=True
)


FILE_3 = code3_matches[0]


print()
print("Selected Code 3 file:")
print(FILE_3)

if len(code3_matches) > 1:

    print()
    print("Other compatible files found:")

    for p in code3_matches[1:10]:

        print(
            " -",
            p
        )


# ----------------------------------------------------------------------
# 4. LOAD CODE 3
# ----------------------------------------------------------------------

cells = pd.read_csv(
    FILE_3
)

print()
print(
    "Code 3 cell records:",
    len(cells)
)

print(
    "Code 3 columns:",
    list(cells.columns)
)


# ----------------------------------------------------------------------
# 5. STRICT COLUMN CHECK
# ----------------------------------------------------------------------

required_code3_full = [
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell"
]

missing = [
    c
    for c in required_code3_full
    if c not in cells.columns
]

if missing:

    raise ValueError(
        "Selected file does not contain all required "
        "Code 3 columns:\n"
        + "\n".join(missing)
    )


# ----------------------------------------------------------------------
# 6. OPTIONAL COLUMNS
# ----------------------------------------------------------------------

for optional_col in [
    "table_number",
    "table_caption",
    "header_text",
    "row_text",
    "full_table_text",
    "article_title",
    "abstract"
]:

    if optional_col not in cells.columns:

        cells[optional_col] = ""


# ----------------------------------------------------------------------
# 7. CONTINUE WITH THE ORIGINAL 6O CODE
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("CODE 3 DISCOVERY SUCCESSFUL")
print("=" * 70)

print(
    "Selected Code 3 file:",
    FILE_3
)

print(
    "Code 3 records:",
    len(cells)
)

print(
    "6N observations:",
    len(df)
)

print()
print(
    "Proceeding to targeted same-table reconstruction..."
)
```

```python
# ======================================================================
# CODE 6O — CONTINUATION ONLY
# ======================================================================

print("=" * 70)
print("CODE 6O — STARTING ACTUAL RECONSTRUCTION")
print("=" * 70)

print("6N observations:", len(df))
print("Code 3 records:", len(cells))

# Normalize
def clean(x):
    if pd.isna(x):
        return ""
    return re.sub(r"\s+", " ", str(x)).strip()

cells["pmcid_norm"] = cells["pmcid"].astype(str).str.upper().str.strip()
df["pmcid_norm"] = df["pmcid"].astype(str).str.upper().str.strip()

cells["table_index_num"] = pd.to_numeric(
    cells["table_index"], errors="coerce"
)

df["table_index_num"] = pd.to_numeric(
    df["table_index"], errors="coerce"
)

cells["row_index_num"] = pd.to_numeric(
    cells["row_index"], errors="coerce"
)

df["row_index_num"] = pd.to_numeric(
    df["row_index"], errors="coerce"
)

cells["raw_clean"] = cells["raw_cell"].apply(clean)
cells["header_clean"] = cells["header_text"].apply(clean)
cells["row_clean"] = cells["row_text"].apply(clean)

# Only validated source tables
source_keys = df[
    ["pmcid_norm", "table_index_num"]
].drop_duplicates()

keyed_cells = cells.merge(
    source_keys,
    on=["pmcid_norm", "table_index_num"],
    how="inner"
)

print(
    "Validated source tables:",
    len(source_keys)
)

print(
    "Code 3 cells restricted to validated tables:",
    len(keyed_cells)
)

# Vocabulary
drug_terms = {
    "doxorubicin": ["doxorubicin", "DOX"],
    "5-fluorouracil": ["5-fluorouracil", "5-fu", "5fu"],
    "paclitaxel": ["paclitaxel"],
    "curcumin": ["curcumin"],
    "dexamethasone": ["dexamethasone"],
    "ibuprofen": ["ibuprofen"],
    "metformin": ["metformin"],
    "gemcitabine": ["gemcitabine"],
    "rifampicin": ["rifampicin"],
    "vancomycin": ["vancomycin"]
}

polymer_terms = {
    "PLGA": ["PLGA", "poly(lactic-co-glycolic acid)"],
    "PLA": ["PLA", "polylactic acid", "poly(lactic acid)"],
    "PCL": ["PCL", "polycaprolactone"],
    "PEG": ["PEG", "polyethylene glycol"],
    "PVA": ["PVA", "polyvinyl alcohol"],
    "PVP": ["PVP", "polyvinylpyrrolidone"],
    "PEO": ["PEO", "polyethylene oxide"],
    "chitosan": ["chitosan"],
    "alginate": ["alginate", "sodium alginate"],
    "cellulose": ["cellulose"],
    "gelatin": ["gelatin"],
    "poloxamer": ["poloxamer"],
    "polyurethane": ["polyurethane"]
}

carrier_terms = {
    "nanoparticle": ["nanoparticle", "nanoparticles"],
    "hydrogel": ["hydrogel", "hydrogels"],
    "microparticle": ["microparticle", "microparticles"],
    "microsphere": ["microsphere", "microspheres"],
    "liposome": ["liposome", "liposomes"],
    "nanofiber": ["nanofiber", "nanofibers"],
    "film": ["film", "films"],
    "bead": ["bead", "beads"],
    "implant": ["implant", "implants"],
    "micelle": ["micelle", "micelles"]
}

condition_patterns = [
    r"\bpH\s*=\s*[0-9]+(?:\.[0-9]+)?",
    r"\bpH\s*[0-9]+(?:\.[0-9]+)?"
]

def find_terms(text, dictionary):
    text = clean(text).lower()

    if not text:
        return []

    found = []

    for canonical, terms in dictionary.items():
        if any(term.lower() in text for term in terms):
            found.append(canonical)

    return sorted(set(found))


def find_conditions(text):
    text = clean(text)

    found = []

    for pattern in condition_patterns:
        found.extend(
            re.findall(
                pattern,
                text,
                flags=re.I
            )
        )

    return sorted(set(found))


def resolve(values):

    values = sorted(set(values))

    if len(values) == 1:
        return values[0]

    if len(values) == 0:
        return "NOT_IDENTIFIED"

    return "AMBIGUOUS:" + "|".join(values)


# Process observations
records = []

for i, obs in df.iterrows():

    pmcid = obs["pmcid_norm"]
    table = obs["table_index_num"]
    row = obs["row_index_num"]

    sub = keyed_cells[
        (keyed_cells["pmcid_norm"] == pmcid)
        &
        (keyed_cells["table_index_num"] == table)
    ]

    exact_row = sub[
        sub["row_index_num"] == row
    ]

    if len(exact_row):

        row_text = " ".join(
            exact_row[
                [
                    "raw_clean",
                    "row_clean",
                    "header_clean"
                ]
            ]
            .fillna("")
            .astype(str)
            .values
            .flatten()
        )

    else:
        row_text = ""

    drugs = find_terms(
        row_text,
        drug_terms
    )

    polymers = find_terms(
        row_text,
        polymer_terms
    )

    carriers = find_terms(
        row_text,
        carrier_terms
    )

    conditions = find_conditions(
        row_text
    )

    drug = resolve(drugs)
    polymer = resolve(polymers)
    carrier = resolve(carriers)
    condition = resolve(conditions)

    strong_fields = sum([
        len(drugs) == 1,
        len(polymers) == 1,
        len(carriers) == 1
    ])

    if strong_fields >= 2:
        status = "STRONG_ROW_RECOVERY"
    elif strong_fields == 1:
        status = "PARTIAL_ROW_RECOVERY"
    else:
        status = "UNRESOLVED"

    records.append({
        "observation_id": obs["observation_id"],
        "pmcid": pmcid,
        "table_index": table,
        "row_index": row,
        "release_time_h": obs["release_time_h"],
        "release_percent": obs["release_percent"],
        "observation_type": obs.get(
            "observation_type",
            "UNKNOWN"
        ),
        "exact_row_text": row_text,
        "recovered_drug": drug,
        "recovered_polymer": polymer,
        "recovered_carrier": carrier,
        "recovered_release_condition": condition,
        "drug_source": (
            "EXACT_ROW" if len(drugs) == 1
            else "EXACT_ROW_AMBIGUOUS" if len(drugs) > 1
            else "UNRESOLVED"
        ),
        "polymer_source": (
            "EXACT_ROW" if len(polymers) == 1
            else "EXACT_ROW_AMBIGUOUS" if len(polymers) > 1
            else "UNRESOLVED"
        ),
        "carrier_source": (
            "EXACT_ROW" if len(carriers) == 1
            else "EXACT_ROW_AMBIGUOUS" if len(carriers) > 1
            else "UNRESOLVED"
        ),
        "condition_source": (
            "EXACT_ROW" if len(conditions) == 1
            else "EXACT_ROW_AMBIGUOUS" if len(conditions) > 1
            else "UNRESOLVED"
        ),
        "recovery_status": status,
        "strong_metadata_fields": strong_fields,
        "source_table_cells": len(sub)
    })

    if (i + 1) % 10 == 0:
        print(
            f"Processed {i+1}/{len(df)}"
        )

result = pd.DataFrame(records)

result["metadata_ready_6O"] = (
    result["recovery_status"]
    == "STRONG_ROW_RECOVERY"
)

result["manual_review_6O"] = (
    ~result["metadata_ready_6O"]
)

# Safety
if len(result) != 60:
    raise ValueError(
        f"Observation count changed: {len(result)}"
    )

if result["observation_id"].duplicated().any():
    raise ValueError(
        "Duplicate observation IDs detected."
    )

# Outputs
OUT_AUDIT = os.path.join(
    VAL,
    "06O_targeted_source_table_reconstruction_audit.csv"
)

OUT_READY = os.path.join(
    VAL,
    "06O_recovered_metadata_candidates.csv"
)

OUT_MANUAL = os.path.join(
    VAL,
    "06O_remaining_metadata_review.csv"
)

result.to_csv(
    OUT_AUDIT,
    index=False
)

result[
    result["metadata_ready_6O"]
].to_csv(
    OUT_READY,
    index=False
)

result[
    result["manual_review_6O"]
].to_csv(
    OUT_MANUAL,
    index=False
)

print()
print("=" * 70)
print("CODE 6O COMPLETE")
print("=" * 70)

print("Input observations:", len(df))
print("Output observations:", len(result))

print()
print("Recovery status:")
print(result["recovery_status"].value_counts())

print()
print("Drug recovery:")
print(result["recovered_drug"].value_counts())

print()
print("Polymer recovery:")
print(result["recovered_polymer"].value_counts())

print()
print("Carrier recovery:")
print(result["recovered_carrier"].value_counts())

print()
print("Release condition:")
print(
    result[
        "recovered_release_condition"
    ].value_counts()
)

print()
print(
    "Strong metadata observations:",
    int(result["metadata_ready_6O"].sum())
)

print(
    "Remaining manual review:",
    int(result["manual_review_6O"].sum())
)

print()
print("Saved:")
print(OUT_AUDIT)
print(OUT_READY)
print(OUT_MANUAL)

print()
print("No release values changed.")
print("No observations generated.")
print("No interpolation performed.")
print("No observations duplicated.")
print("No machine learning performed.")

print("=" * 70)
```

```python
# ======================================================================
# CODE 6P — EXPERIMENTAL-UNIT INDEPENDENCE & MANUAL-REVIEW PRIORITIZATION
# ======================================================================

import os
import re
import numpy as np
import pandas as pd

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")

FILE_6N = os.path.join(
    VAL,
    "06N_analytical_ready_observations.csv"
)

FILE_6O = os.path.join(
    VAL,
    "06O_targeted_source_table_reconstruction_audit.csv"
)

print("=" * 70)
print("CODE 6P — EXPERIMENTAL-UNIT INDEPENDENCE & MANUAL-REVIEW")
print("=" * 70)


# ======================================================================
# 1. LOAD
# ======================================================================

df6n = pd.read_csv(FILE_6N)
df6o = pd.read_csv(FILE_6O)

print("6N observations:", len(df6n))
print("6O observations:", len(df6o))


# ======================================================================
# 2. CHECK OBSERVATION IDs
# ======================================================================

if "observation_id" not in df6n.columns:
    raise ValueError(
        "6N missing observation_id"
    )

if "observation_id" not in df6o.columns:
    raise ValueError(
        "6O missing observation_id"
    )

if df6n["observation_id"].duplicated().any():
    raise ValueError(
        "Duplicate observation IDs in 6N"
    )

if df6o["observation_id"].duplicated().any():
    raise ValueError(
        "Duplicate observation IDs in 6O"
    )


# ======================================================================
# 3. SAFE MERGE
# ======================================================================

o_cols = [
    "observation_id",
    "recovered_drug",
    "recovered_polymer",
    "recovered_carrier",
    "recovered_release_condition",
    "recovery_status",
    "strong_metadata_fields",
    "metadata_ready_6O"
]

o_cols = [
    c for c in o_cols
    if c in df6o.columns
]

metadata = df6o[o_cols].copy()

df = df6n.merge(
    metadata,
    on="observation_id",
    how="left",
    validate="one_to_one"
)

print(
    "Merged observations:",
    len(df)
)


# ======================================================================
# 4. NORMALIZATION
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


for col in [
    "pmcid",
    "table_index",
    "row_index",
    "release_time_h",
    "release_percent",
    "observation_type"
]:

    if col in df.columns:
        df[col] = df[col].apply(clean)


# ======================================================================
# 5. CONSERVATIVE METADATA FLAGS
# ======================================================================

def is_identified(x):

    s = clean(x)

    if not s:
        return False

    if s.upper() in [
        "NOT_IDENTIFIED",
        "UNKNOWN",
        "UNRESOLVED"
    ]:
        return False

    if s.upper().startswith("AMBIGUOUS:"):
        return False

    return True


df["drug_identified_6P"] = (
    df["recovered_drug"]
    .apply(is_identified)
)

df["polymer_identified_6P"] = (
    df["recovered_polymer"]
    .apply(is_identified)
)

df["carrier_identified_6P"] = (
    df["recovered_carrier"]
    .apply(is_identified)
)

df["condition_identified_6P"] = (
    df["recovered_release_condition"]
    .apply(is_identified)
)


# ======================================================================
# 6. MANUAL-REVIEW PRIORITY
# ======================================================================

def priority(row):

    score = 0
    reasons = []

    if not row["drug_identified_6P"]:
        score += 3
        reasons.append("DRUG_UNRESOLVED")

    if not row["polymer_identified_6P"]:
        score += 3
        reasons.append("POLYMER_UNRESOLVED")

    if not row["carrier_identified_6P"]:
        score += 2
        reasons.append("CARRIER_UNRESOLVED")

    if not row["condition_identified_6P"]:
        score += 2
        reasons.append("CONDITION_UNRESOLVED")

    if clean(
        row.get(
            "recovery_status",
            ""
        )
    ) == "UNRESOLVED":

        score += 2
        reasons.append("ROW_UNRESOLVED")

    return pd.Series({
        "manual_review_priority_score": score,
        "manual_review_reason":
            ";".join(reasons)
    })


priority_df = df.apply(
    priority,
    axis=1
)

df = pd.concat(
    [
        df,
        priority_df
    ],
    axis=1
)


# ======================================================================
# 7. CONSERVATIVE EXPERIMENTAL-UNIT PROXY
# ======================================================================

def make_unit_proxy(row):

    pmcid = clean(
        row.get("pmcid", "")
    )

    table = clean(
        row.get("table_index", "")
    )

    drug = clean(
        row.get("recovered_drug", "")
    )

    polymer = clean(
        row.get("recovered_polymer", "")
    )

    carrier = clean(
        row.get("recovered_carrier", "")
    )

    condition = clean(
        row.get(
            "recovered_release_condition",
            ""
        )
    )

    # IMPORTANT:
    # Time is deliberately NOT included.
    #
    # Condition IS included because pH 5.5 and pH 7.4
    # represent different release conditions.

    return "|".join([
        pmcid,
        "T" + table,
        drug,
        polymer,
        carrier,
        condition
    ])


df["experimental_unit_proxy_6P"] = (
    df.apply(
        make_unit_proxy,
        axis=1
    )
)


# ======================================================================
# 8. TIME-POINT COUNT PER PROXY
# ======================================================================

unit_counts = (
    df.groupby(
        "experimental_unit_proxy_6P"
    )[
        "release_time_h"
    ]
    .nunique()
    .rename(
        "unique_timepoints"
    )
    .reset_index()
)

df = df.merge(
    unit_counts,
    on="experimental_unit_proxy_6P",
    how="left"
)


# ======================================================================
# 9. OBSERVATIONS PER ARTICLE/TABLE/TIME
# ======================================================================

article_table_time = (
    df.groupby(
        [
            "pmcid",
            "table_index",
            "release_time_h"
        ]
    )
    .size()
    .rename(
        "observations_same_article_table_time"
    )
    .reset_index()
)

df = df.merge(
    article_table_time,
    on=[
        "pmcid",
        "table_index",
        "release_time_h"
    ],
    how="left"
)

df[
    "same_article_table_time_multiple_6P"
] = (
    df[
        "observations_same_article_table_time"
    ] > 1
)


# ======================================================================
# 10. MODELING ELIGIBILITY — CONSERVATIVE
# ======================================================================

df["metadata_complete_for_ml_6P"] = (
    df["drug_identified_6P"]
    &
    df["polymer_identified_6P"]
    &
    df["carrier_identified_6P"]
)

# Condition is required when the source clearly contains
# multiple release conditions.

multiple_conditions = (
    df.groupby(
        [
            "pmcid",
            "table_index"
        ]
    )[
        "recovered_release_condition"
    ]
    .nunique()
    .reset_index(
        name="table_condition_count"
    )
)

df = df.merge(
    multiple_conditions,
    on=[
        "pmcid",
        "table_index"
    ],
    how="left"
)

df["condition_required_6P"] = (
    df["table_condition_count"] > 1
)

df["condition_ok_for_ml_6P"] = (
    (~df["condition_required_6P"])
    |
    df["condition_identified_6P"]
)

df["conservative_ml_metadata_candidate"] = (
    df["metadata_complete_for_ml_6P"]
    &
    df["condition_ok_for_ml_6P"]
)


# ======================================================================
# 11. INDEPENDENCE WARNING
# ======================================================================

df["timecourse_group_6P"] = (
    df[
        [
            "pmcid",
            "table_index",
            "experimental_unit_proxy_6P"
        ]
    ]
    .astype(str)
    .agg(
        "|".join,
        axis=1
    )
)


# ======================================================================
# 12. OUTPUTS
# ======================================================================

AUDIT = os.path.join(
    VAL,
    "06P_experimental_unit_independence_audit.csv"
)

MANUAL = os.path.join(
    VAL,
    "06P_prioritized_manual_review.csv"
)

READY = os.path.join(
    VAL,
    "06P_conservative_ml_metadata_candidates.csv"
)

UNIT_SUMMARY = os.path.join(
    VAL,
    "06P_experimental_unit_summary.csv"
)

df.to_csv(
    AUDIT,
    index=False
)

df.sort_values(
    "manual_review_priority_score",
    ascending=False
).to_csv(
    MANUAL,
    index=False
)

df[
    df["conservative_ml_metadata_candidate"]
].to_csv(
    READY,
    index=False
)

unit_summary = (
    df.groupby(
        "experimental_unit_proxy_6P"
    )
    .agg(
        pmcid=("pmcid", "first"),
        table_index=("table_index", "first"),
        observations=("observation_id", "count"),
        unique_timepoints=("release_time_h", "nunique"),
        drugs=("recovered_drug", "first"),
        polymers=("recovered_polymer", "first"),
        carriers=("recovered_carrier", "first"),
        conditions=(
            "recovered_release_condition",
            "first"
        )
    )
    .reset_index()
)

unit_summary.to_csv(
    UNIT_SUMMARY,
    index=False
)


# ======================================================================
# 13. REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6P COMPLETE")
print("=" * 70)

print(
    "Input observations:",
    len(df)
)

print()
print("Metadata identification:")

print(
    "Drug identified:",
    int(df["drug_identified_6P"].sum())
)

print(
    "Polymer identified:",
    int(df["polymer_identified_6P"].sum())
)

print(
    "Carrier identified:",
    int(df["carrier_identified_6P"].sum())
)

print(
    "Condition identified:",
    int(df["condition_identified_6P"].sum())
)

print()
print(
    "Conservative ML metadata candidates:",
    int(
        df[
            "conservative_ml_metadata_candidate"
        ].sum()
    )
)

print()
print(
    "Unique experimental-unit proxies:",
    df[
        "experimental_unit_proxy_6P"
    ].nunique()
)

print()
print(
    "Observations belonging to multi-timepoint "
    "proxies:",
    int(
        (
            df["unique_timepoints"] > 1
        ).sum()
    )
)

print()
print(
    "Same article/table/time groups >1 observation:",
    int(
        (
            df[
                "same_article_table_time_multiple_6P"
            ]
        ).sum()
    )
)

print()
print("Manual-review priority:")
print(
    df[
        "manual_review_priority_score"
    ].value_counts()
    .sort_index(
        ascending=False
    )
)

print()
print("Observation types:")
print(
    df[
        "observation_type"
    ].value_counts()
)

print()
print("Saved:")
print(AUDIT)
print(MANUAL)
print(READY)
print(UNIT_SUMMARY)

print()
print("IMPORTANT:")
print(
    "Experimental-unit proxies are conservative and "
    "are not treated as proof of independence."
)

print(
    "Multiple time points from one formulation remain "
    "within the same experimental unit."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6P-R — EXPERIMENTAL-UNIT INDEPENDENCE & MANUAL-REVIEW
#            ROBUST SCHEMA VERSION
# ======================================================================

import os
import re
import numpy as np
import pandas as pd

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")

FILE_6N = os.path.join(
    VAL,
    "06N_analytical_ready_observations.csv"
)

FILE_6O = os.path.join(
    VAL,
    "06O_targeted_source_table_reconstruction_audit.csv"
)

print("=" * 70)
print("CODE 6P-R — EXPERIMENTAL-UNIT INDEPENDENCE & MANUAL-REVIEW")
print("=" * 70)


# ======================================================================
# 1. LOAD
# ======================================================================

df6n = pd.read_csv(FILE_6N)
df6o = pd.read_csv(FILE_6O)

print("6N observations:", len(df6n))
print("6O observations:", len(df6o))


# ======================================================================
# 2. DISPLAY SCHEMAS
# ======================================================================

print()
print("6N columns:")
print(list(df6n.columns))

print()
print("6O columns:")
print(list(df6o.columns))


# ======================================================================
# 3. IDENTIFY OBSERVATION ID
# ======================================================================

def detect_column(df, candidates):

    for c in candidates:

        if c in df.columns:
            return c

    return None


id6n = detect_column(
    df6n,
    [
        "observation_id",
        "candidate_id",
        "id"
    ]
)

id6o = detect_column(
    df6o,
    [
        "observation_id",
        "candidate_id",
        "id"
    ]
)

if id6n is None:
    raise ValueError(
        "Could not identify observation ID in 6N."
    )

if id6o is None:
    raise ValueError(
        "Could not identify observation ID in 6O."
    )

print()
print("6N observation ID:", id6n)
print("6O observation ID:", id6o)


# ======================================================================
# 4. STANDARDIZE OBSERVATION ID
# ======================================================================

df6n = df6n.copy()
df6o = df6o.copy()

df6n["__observation_id"] = (
    df6n[id6n]
    .astype(str)
    .str.strip()
)

df6o["__observation_id"] = (
    df6o[id6o]
    .astype(str)
    .str.strip()
)


# ======================================================================
# 5. CHECK DUPLICATES
# ======================================================================

if df6n[
    "__observation_id"
].duplicated().any():

    raise ValueError(
        "Duplicate observation IDs detected in 6N."
    )

if df6o[
    "__observation_id"
].duplicated().any():

    raise ValueError(
        "Duplicate observation IDs detected in 6O."
    )


# ======================================================================
# 6. DETECT 6O METADATA COLUMNS
# ======================================================================

def detect_metadata_column(df, names):

    for name in names:

        if name in df.columns:
            return name

    return None


drug_col = detect_metadata_column(
    df6o,
    [
        "recovered_drug",
        "final_drug_6M",
        "final_drug",
        "context_drug",
        "resolved_drug",
        "xml_row_drug",
        "row_drug",
        "drug"
    ]
)

polymer_col = detect_metadata_column(
    df6o,
    [
        "recovered_polymer",
        "final_polymer_6M",
        "final_polymer",
        "context_polymer",
        "resolved_polymer",
        "xml_row_polymer",
        "row_polymer",
        "polymer"
    ]
)

carrier_col = detect_metadata_column(
    df6o,
    [
        "recovered_carrier",
        "final_carrier_6M",
        "final_carrier",
        "context_carrier",
        "resolved_carrier",
        "xml_row_carrier",
        "row_carrier",
        "carrier"
    ]
)

condition_col = detect_metadata_column(
    df6o,
    [
        "recovered_release_condition",
        "final_release_condition",
        "release_condition",
        "context_release_condition",
        "resolved_release_condition"
    ]
)

status_col = detect_metadata_column(
    df6o,
    [
        "recovery_status",
        "final_6M_status",
        "final_metadata_status",
        "context_resolution_status"
    ]
)


print()
print("Detected metadata columns:")
print("Drug      :", drug_col)
print("Polymer   :", polymer_col)
print("Carrier   :", carrier_col)
print("Condition :", condition_col)
print("Status    :", status_col)


# ======================================================================
# 7. FAIL SAFELY IF REQUIRED METADATA ARE ABSENT
# ======================================================================

if drug_col is None:
    print()
    print(
        "WARNING: No drug column found in 6O."
    )

if polymer_col is None:
    print(
        "WARNING: No polymer column found in 6O."
    )

if carrier_col is None:
    print(
        "WARNING: No carrier column found in 6O."
    )


# ======================================================================
# 8. CREATE STANDARDIZED METADATA COLUMNS
# ======================================================================

def safe_series(df, col):

    if col is None:

        return pd.Series(
            ["NOT_IDENTIFIED"] * len(df),
            index=df.index
        )

    return (
        df[col]
        .fillna("NOT_IDENTIFIED")
        .astype(str)
        .str.strip()
    )


df6o["__drug"] = safe_series(
    df6o,
    drug_col
)

df6o["__polymer"] = safe_series(
    df6o,
    polymer_col
)

df6o["__carrier"] = safe_series(
    df6o,
    carrier_col
)

df6o["__condition"] = safe_series(
    df6o,
    condition_col
)


# ======================================================================
# 9. MERGE
# ======================================================================

metadata_cols = [
    "__observation_id",
    "__drug",
    "__polymer",
    "__carrier",
    "__condition"
]

if status_col is not None:

    metadata_cols.append(
        status_col
    )

metadata = df6o[
    metadata_cols
].copy()

rename_map = {
    "__drug":
        "recovered_drug_6P",
    "__polymer":
        "recovered_polymer_6P",
    "__carrier":
        "recovered_carrier_6P",
    "__condition":
        "recovered_condition_6P"
}

metadata = metadata.rename(
    columns=rename_map
)

df = df6n.merge(
    metadata,
    on="__observation_id",
    how="left",
    validate="one_to_one"
)

print()
print(
    "Rows after consolidation:",
    len(df)
)

if len(df) != 60:

    raise ValueError(
        f"Expected 60 rows after merge, "
        f"found {len(df)}."
    )


# ======================================================================
# 10. STANDARDIZE SOURCE COLUMNS
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


for col in [
    "pmcid",
    "table_index",
    "row_index",
    "release_time_h",
    "release_percent",
    "observation_type"
]:

    if col in df.columns:

        df[col] = df[col].apply(
            clean
        )


# ======================================================================
# 11. IDENTIFICATION FUNCTION
# ======================================================================

def identified(x):

    s = clean(x)

    if not s:
        return False

    upper = s.upper()

    if upper in [
        "NOT_IDENTIFIED",
        "UNKNOWN",
        "UNRESOLVED",
        "NAN",
        "NONE"
    ]:
        return False

    if upper.startswith(
        "AMBIGUOUS:"
    ):
        return False

    return True


df["drug_identified_6P"] = (
    df[
        "recovered_drug_6P"
    ].apply(identified)
)

df["polymer_identified_6P"] = (
    df[
        "recovered_polymer_6P"
    ].apply(identified)
)

df["carrier_identified_6P"] = (
    df[
        "recovered_carrier_6P"
    ].apply(identified)
)

df["condition_identified_6P"] = (
    df[
        "recovered_condition_6P"
    ].apply(identified)
)


# ======================================================================
# 12. MANUAL REVIEW PRIORITY
# ======================================================================

def calculate_priority(row):

    score = 0
    reasons = []

    if not row[
        "drug_identified_6P"
    ]:

        score += 3
        reasons.append(
            "DRUG_UNRESOLVED"
        )

    if not row[
        "polymer_identified_6P"
    ]:

        score += 3
        reasons.append(
            "POLYMER_UNRESOLVED"
        )

    if not row[
        "carrier_identified_6P"
    ]:

        score += 2
        reasons.append(
            "CARRIER_UNRESOLVED"
        )

    if not row[
        "condition_identified_6P"
    ]:

        score += 2
        reasons.append(
            "CONDITION_UNRESOLVED"
        )

    return pd.Series({
        "manual_review_priority_score":
            score,

        "manual_review_reason":
            ";".join(reasons)
    })


priority = df.apply(
    calculate_priority,
    axis=1
)

df = pd.concat(
    [
        df,
        priority
    ],
    axis=1
)


# ======================================================================
# 13. CONSERVATIVE EXPERIMENTAL-UNIT PROXY
# ======================================================================

def make_proxy(row):

    return "|".join([
        clean(row.get("pmcid", "")),
        "T" + clean(
            row.get(
                "table_index",
                ""
            )
        ),
        clean(
            row.get(
                "recovered_drug_6P",
                ""
            )
        ),
        clean(
            row.get(
                "recovered_polymer_6P",
                ""
            )
        ),
        clean(
            row.get(
                "recovered_carrier_6P",
                ""
            )
        ),
        clean(
            row.get(
                "recovered_condition_6P",
                ""
            )
        )
    ])


df[
    "experimental_unit_proxy_6P"
] = df.apply(
    make_proxy,
    axis=1
)


# ======================================================================
# 14. TIMEPOINT COUNTS
# ======================================================================

time_counts = (
    df.groupby(
        "experimental_unit_proxy_6P"
    )[
        "release_time_h"
    ]
    .nunique()
    .rename(
        "unique_timepoints"
    )
    .reset_index()
)

df = df.merge(
    time_counts,
    on="experimental_unit_proxy_6P",
    how="left"
)


# ======================================================================
# 15. SAME ARTICLE/TABLE/TIME
# ======================================================================

group_counts = (
    df.groupby(
        [
            "pmcid",
            "table_index",
            "release_time_h"
        ]
    )
    .size()
    .rename(
        "observations_same_article_table_time"
    )
    .reset_index()
)

df = df.merge(
    group_counts,
    on=[
        "pmcid",
        "table_index",
        "release_time_h"
    ],
    how="left"
)

df[
    "same_article_table_time_multiple_6P"
] = (
    df[
        "observations_same_article_table_time"
    ] > 1
)


# ======================================================================
# 16. CONSERVATIVE ML METADATA GATE
# ======================================================================

df[
    "metadata_complete_for_ml_6P"
] = (
    df[
        "drug_identified_6P"
    ]
    &
    df[
        "polymer_identified_6P"
    ]
    &
    df[
        "carrier_identified_6P"
    ]
)


# ----------------------------------------------------------------------
# Determine whether a table contains multiple explicit conditions
# ----------------------------------------------------------------------

condition_table_counts = (
    df.groupby(
        [
            "pmcid",
            "table_index"
        ]
    )[
        "recovered_condition_6P"
    ]
    .nunique()
    .rename(
        "table_condition_count"
    )
    .reset_index()
)

df = df.merge(
    condition_table_counts,
    on=[
        "pmcid",
        "table_index"
    ],
    how="left"
)

df[
    "condition_required_6P"
] = (
    df[
        "table_condition_count"
    ] > 1
)

df[
    "condition_ok_for_ml_6P"
] = (
    (~df["condition_required_6P"])
    |
    df["condition_identified_6P"]
)

df[
    "conservative_ml_metadata_candidate"
] = (
    df[
        "metadata_complete_for_ml_6P"
    ]
    &
    df[
        "condition_ok_for_ml_6P"
    ]
)


# ======================================================================
# 17. EXPERIMENTAL UNIT SUMMARY
# ======================================================================

unit_summary = (
    df.groupby(
        "experimental_unit_proxy_6P"
    )
    .agg(
        pmcid=("pmcid", "first"),
        table_index=("table_index", "first"),
        observations=(
            "observation_id",
            "count"
        ),
        unique_timepoints=(
            "release_time_h",
            "nunique"
        ),
        drug=(
            "recovered_drug_6P",
            "first"
        ),
        polymer=(
            "recovered_polymer_6P",
            "first"
        ),
        carrier=(
            "recovered_carrier_6P",
            "first"
        ),
        condition=(
            "recovered_condition_6P",
            "first"
        )
    )
    .reset_index()
)


# ======================================================================
# 18. SAVE
# ======================================================================

AUDIT = os.path.join(
    VAL,
    "06P_experimental_unit_independence_audit.csv"
)

MANUAL = os.path.join(
    VAL,
    "06P_prioritized_manual_review.csv"
)

READY = os.path.join(
    VAL,
    "06P_conservative_ml_metadata_candidates.csv"
)

UNIT = os.path.join(
    VAL,
    "06P_experimental_unit_summary.csv"
)

df.to_csv(
    AUDIT,
    index=False
)

df.sort_values(
    "manual_review_priority_score",
    ascending=False
).to_csv(
    MANUAL,
    index=False
)

df[
    df[
        "conservative_ml_metadata_candidate"
    ]
].to_csv(
    READY,
    index=False
)

unit_summary.to_csv(
    UNIT,
    index=False
)


# ======================================================================
# 19. FINAL REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6P-R COMPLETE")
print("=" * 70)

print(
    "Input observations:",
    len(df)
)

print()
print(
    "Drug identified:",
    int(
        df[
            "drug_identified_6P"
        ].sum()
    )
)

print(
    "Polymer identified:",
    int(
        df[
            "polymer_identified_6P"
        ].sum()
    )
)

print(
    "Carrier identified:",
    int(
        df[
            "carrier_identified_6P"
        ].sum()
    )
)

print(
    "Condition identified:",
    int(
        df[
            "condition_identified_6P"
        ].sum()
    )
)

print()
print(
    "Unique experimental-unit proxies:",
    df[
        "experimental_unit_proxy_6P"
    ].nunique()
)

print()
print(
    "Multi-timepoint observations:",
    int(
        (
            df[
                "unique_timepoints"
            ] > 1
        ).sum()
    )
)

print()
print(
    "Same article/table/time multiple:",
    int(
        df[
            "same_article_table_time_multiple_6P"
        ].sum()
    )
)

print()
print(
    "Conservative ML metadata candidates:",
    int(
        df[
            "conservative_ml_metadata_candidate"
        ].sum()
    )
)

print()
print("Manual-review priority distribution:")

print(
    df[
        "manual_review_priority_score"
    ]
    .value_counts()
    .sort_index(
        ascending=False
    )
)

print()
print("Observation types:")

print(
    df[
        "observation_type"
    ].value_counts()
)

print()
print("Saved:")
print(AUDIT)
print(MANUAL)
print(READY)
print(UNIT)

print()
print("IMPORTANT:")
print(
    "No release values were changed."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "Experimental-unit proxies do not prove statistical independence."
)

print(
    "Multiple time points remain repeated measurements "
    "within the same experimental unit."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6Q — EXPERIMENTAL-UNIT CONSOLIDATION & FORMULATION CONSISTENCY
# ======================================================================

import os
import re
import numpy as np
import pandas as pd

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")

FILE_6P = os.path.join(
    VAL,
    "06P_experimental_unit_independence_audit.csv"
)

print("=" * 70)
print("CODE 6Q — EXPERIMENTAL-UNIT CONSOLIDATION & FORMULATION CONSISTENCY")
print("=" * 70)

df = pd.read_csv(FILE_6P)

print("6P observations:", len(df))

if len(df) != 60:
    raise ValueError(
        f"Expected 60 observations, found {len(df)}"
    )

if "observation_id" not in df.columns:
    raise ValueError(
        "observation_id is missing from 6P."
    )

# ======================================================================
# 1. SAFE CLEANING
# ======================================================================

def clean(x):
    if pd.isna(x):
        return ""
    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].apply(clean)


# ======================================================================
# 2. FIND THE UNIT PROXY
# ======================================================================

unit_candidates = [
    "experimental_unit_proxy_6P",
    "experimental_unit_id",
    "experimental_unit_id_header",
    "final_formulation_id",
    "formulation_identifier",
    "formulation_id_header"
]

unit_col = None

for c in unit_candidates:
    if c in df.columns:
        unit_col = c
        break

if unit_col is None:
    raise ValueError(
        "Could not identify an experimental-unit/formulation column."
    )

print("Experimental-unit proxy:", unit_col)


# ======================================================================
# 3. FIND METADATA COLUMNS
# ======================================================================

def find_col(candidates):

    for c in candidates:
        if c in df.columns:
            return c

    return None


drug_col = find_col([
    "recovered_drug_6P",
    "recovered_drug",
    "final_drug",
    "final_drug_6M",
    "drug",
    "drug_extracted"
])

polymer_col = find_col([
    "recovered_polymer_6P",
    "recovered_polymer",
    "final_polymer",
    "final_polymer_6M",
    "polymer",
    "polymer_extracted"
])

carrier_col = find_col([
    "recovered_carrier_6P",
    "recovered_carrier",
    "final_carrier",
    "final_carrier_6M",
    "carrier",
    "carrier_extracted"
])

condition_col = find_col([
    "recovered_condition_6P",
    "recovered_release_condition",
    "final_release_condition",
    "release_condition",
    "condition"
])

print("Drug column     :", drug_col)
print("Polymer column  :", polymer_col)
print("Carrier column  :", carrier_col)
print("Condition column:", condition_col)


# ======================================================================
# 4. STANDARDIZE MISSING VALUES
# ======================================================================

def standardize_value(x):

    s = clean(x)

    if s == "":
        return "NOT_IDENTIFIED"

    if s.upper() in [
        "NOT_IDENTIFIED",
        "UNKNOWN",
        "UNRESOLVED",
        "NONE",
        "NAN"
    ]:
        return "NOT_IDENTIFIED"

    return s


for col in [
    drug_col,
    polymer_col,
    carrier_col,
    condition_col
]:

    if col is not None:

        df[col] = df[col].apply(
            standardize_value
        )


# ======================================================================
# 5. IDENTIFY AMBIGUOUS VALUES
# ======================================================================

def is_ambiguous(x):

    s = clean(x).upper()

    return (
        s.startswith("AMBIGUOUS:")
        or "|" in s
    )


def identified(x):

    s = clean(x)

    return (
        s != ""
        and s.upper()
        not in [
            "NOT_IDENTIFIED",
            "UNKNOWN",
            "UNRESOLVED"
        ]
        and not is_ambiguous(s)
    )


# ======================================================================
# 6. UNIT-LEVEL CONSISTENCY FUNCTION
# ======================================================================

def summarize_field(group, col):

    if col is None:
        return {
            "value": "NOT_AVAILABLE",
            "n_unique": 0,
            "conflict": False,
            "identified": False
        }

    vals = [
        standardize_value(v)
        for v in group[col]
    ]

    vals = sorted(set(vals))

    meaningful = [
        v for v in vals
        if v != "NOT_IDENTIFIED"
    ]

    if len(meaningful) == 0:

        return {
            "value": "NOT_IDENTIFIED",
            "n_unique": 0,
            "conflict": False,
            "identified": False
        }

    if len(meaningful) == 1:

        value = meaningful[0]

        return {
            "value": value,
            "n_unique": 1,
            "conflict": False,
            "identified": identified(value)
        }

    return {
        "value":
            "CONFLICT:" + "|".join(meaningful),
        "n_unique":
            len(meaningful),
        "conflict":
            True,
        "identified":
            False
    }


# ======================================================================
# 7. BUILD UNIT-LEVEL SUMMARY
# ======================================================================

unit_records = []

for unit_id, group in df.groupby(
    unit_col,
    dropna=False
):

    drug = summarize_field(
        group,
        drug_col
    )

    polymer = summarize_field(
        group,
        polymer_col
    )

    carrier = summarize_field(
        group,
        carrier_col
    )

    condition = summarize_field(
        group,
        condition_col
    )

    times = pd.to_numeric(
        group["release_time_h"],
        errors="coerce"
    ).dropna()

    releases = pd.to_numeric(
        group["release_percent"],
        errors="coerce"
    ).dropna()

    n_obs = len(group)

    n_timepoints = times.nunique()

    n_conditions = (
        group[condition_col].nunique()
        if condition_col is not None
        else 0
    )

    # --------------------------------------------------------------
    # Conservative unit status
    # --------------------------------------------------------------

    conflicts = sum([
        drug["conflict"],
        polymer["conflict"],
        carrier["conflict"],
        condition["conflict"]
    ])

    identified_fields = sum([
        drug["identified"],
        polymer["identified"],
        carrier["identified"]
    ])

    if conflicts > 0:

        unit_status = "CONFLICT"

    elif identified_fields >= 2:

        unit_status = "STRONG_UNIT_METADATA"

    elif identified_fields == 1:

        unit_status = "PARTIAL_UNIT_METADATA"

    else:

        unit_status = "UNRESOLVED_UNIT"


    unit_records.append({

        "experimental_unit_proxy":
            unit_id,

        "observations":
            n_obs,

        "unique_timepoints":
            n_timepoints,

        "min_time_h":
            times.min()
            if len(times)
            else np.nan,

        "max_time_h":
            times.max()
            if len(times)
            else np.nan,

        "drug":
            drug["value"],

        "drug_n_unique":
            drug["n_unique"],

        "drug_conflict":
            drug["conflict"],

        "polymer":
            polymer["value"],

        "polymer_n_unique":
            polymer["n_unique"],

        "polymer_conflict":
            polymer["conflict"],

        "carrier":
            carrier["value"],

        "carrier_n_unique":
            carrier["n_unique"],

        "carrier_conflict":
            carrier["conflict"],

        "release_condition":
            condition["value"],

        "condition_n_unique":
            condition["n_unique"],

        "condition_conflict":
            condition["conflict"],

        "identified_metadata_fields":
            identified_fields,

        "unit_status":
            unit_status
    })


units = pd.DataFrame(
    unit_records
)


# ======================================================================
# 8. TIMECOURSE FLAG
# ======================================================================

units["multiple_timepoints"] = (
    units["unique_timepoints"] > 1
)


# ======================================================================
# 9. CONDITION-SPLIT AUDIT
# ======================================================================

condition_counts = (
    df.groupby(
        unit_col
    )[condition_col]
    .nunique()
    if condition_col is not None
    else pd.Series(dtype=int)
)

if condition_col is not None:

    condition_counts = (
        condition_counts
        .rename(
            "condition_values_per_unit"
        )
        .reset_index()
    )

    units = units.merge(
        condition_counts,
        on=unit_col,
        how="left"
    )

else:

    units[
        "condition_values_per_unit"
    ] = 0


units[
    "multiple_release_conditions"
] = (
    units[
        "condition_values_per_unit"
    ] > 1
)


# ======================================================================
# 10. MAP UNIT STATUS BACK TO 60 OBSERVATIONS
# ======================================================================

status_map = units[
    [
        "experimental_unit_proxy",
        "unit_status",
        "multiple_timepoints",
        "multiple_release_conditions"
    ]
].copy()

status_map = status_map.rename(
    columns={
        "experimental_unit_proxy":
            unit_col
    }
)

df6q = df.merge(
    status_map,
    on=unit_col,
    how="left",
    validate="many_to_one"
)


# ======================================================================
# 11. CONSERVATIVE UNIT-LEVEL ML CANDIDATE
# ======================================================================

df6q[
    "unit_level_ml_candidate"
] = (
    df6q["unit_status"]
    == "STRONG_UNIT_METADATA"
)


# ======================================================================
# 12. SAFETY CHECKS
# ======================================================================

if len(df6q) != 60:

    raise ValueError(
        "Observation count changed during unit mapping."
    )

if df6q[
    "observation_id"
].duplicated().any():

    raise ValueError(
        "Duplicate observation IDs detected."
    )


# ======================================================================
# 13. SAVE
# ======================================================================

OUT_AUDIT = os.path.join(
    VAL,
    "06Q_experimental_unit_observation_audit.csv"
)

OUT_UNITS = os.path.join(
    VAL,
    "06Q_experimental_unit_summary.csv"
)

OUT_STRONG = os.path.join(
    VAL,
    "06Q_strong_unit_candidates.csv"
)

OUT_REVIEW = os.path.join(
    VAL,
    "06Q_unit_manual_review.csv"
)

df6q.to_csv(
    OUT_AUDIT,
    index=False
)

units.to_csv(
    OUT_UNITS,
    index=False
)

units[
    units["unit_status"]
    == "STRONG_UNIT_METADATA"
].to_csv(
    OUT_STRONG,
    index=False
)

units[
    units["unit_status"]
    != "STRONG_UNIT_METADATA"
].to_csv(
    OUT_REVIEW,
    index=False
)


# ======================================================================
# 14. REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6Q COMPLETE")
print("=" * 70)

print(
    "Input observations:",
    len(df)
)

print(
    "Unique experimental-unit proxies:",
    len(units)
)

print()
print("Unit status:")
print(
    units[
        "unit_status"
    ].value_counts()
)

print()
print(
    "Units with multiple time points:",
    int(
        units[
            "multiple_timepoints"
        ].sum()
    )
)

print(
    "Units with multiple release conditions:",
    int(
        units[
            "multiple_release_conditions"
        ].sum()
    )
)

print()
print("Unit-level drug identification:")
print(
    units[
        "drug"
    ].value_counts()
)

print()
print("Unit-level polymer identification:")
print(
    units[
        "polymer"
    ].value_counts()
)

print()
print("Unit-level carrier identification:")
print(
    units[
        "carrier"
    ].value_counts()
)

print()
print("Unit-level release conditions:")
print(
    units[
        "release_condition"
    ].value_counts()
)

print()
print(
    "Strong unit metadata candidates:",
    int(
        (
            units["unit_status"]
            == "STRONG_UNIT_METADATA"
        ).sum()
    )
)

print(
    "Partial unit metadata:",
    int(
        (
            units["unit_status"]
            == "PARTIAL_UNIT_METADATA"
        ).sum()
    )
)

print(
    "Conflicting units:",
    int(
        (
            units["unit_status"]
            == "CONFLICT"
        ).sum()
    )
)

print(
    "Unresolved units:",
    int(
        (
            units["unit_status"]
            == "UNRESOLVED_UNIT"
        ).sum()
    )
)

print()
print(
    "Observation preservation:",
    len(df),
    "->",
    len(df6q)
)

print()
print("Saved:")
print(OUT_AUDIT)
print(OUT_UNITS)
print(OUT_STRONG)
print(OUT_REVIEW)

print()
print("IMPORTANT:")
print(
    "Unit consolidation is an audit operation only."
)

print(
    "The original 60 observations remain intact."
)

print(
    "Multiple time points are not counted as independent experiments."
)

print(
    "Different release conditions are not merged."
)

print(
    "Experimental-unit proxies do not prove independence."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6Q-R — EXPERIMENTAL-UNIT CONSOLIDATION & FORMULATION CONSISTENCY
#                    FIXED UNIT-KEY VERSION
# ======================================================================

import os
import re
import numpy as np
import pandas as pd

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")

FILE_6P = os.path.join(
    VAL,
    "06P_experimental_unit_independence_audit.csv"
)

print("=" * 70)
print("CODE 6Q-R — EXPERIMENTAL-UNIT CONSOLIDATION & FORMULATION CONSISTENCY")
print("=" * 70)

# ======================================================================
# 1. LOAD 6P
# ======================================================================

df = pd.read_csv(FILE_6P)

print("6P observations:", len(df))

if len(df) != 60:
    raise ValueError(
        f"Expected 60 observations from 6P, found {len(df)}."
    )

if "observation_id" not in df.columns:
    raise ValueError(
        "observation_id is missing from 6P."
    )


# ======================================================================
# 2. CLEAN TEXT
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


for col in df.columns:

    if df[col].dtype == "object":

        df[col] = df[col].apply(clean)


# ======================================================================
# 3. IDENTIFY EXPERIMENTAL-UNIT PROXY
# ======================================================================

unit_candidates = [
    "experimental_unit_proxy_6P",
    "experimental_unit_id",
    "experimental_unit_id_header",
    "final_formulation_id",
    "formulation_identifier",
    "formulation_id_header"
]

unit_col = None

for c in unit_candidates:

    if c in df.columns:

        unit_col = c
        break

if unit_col is None:

    raise ValueError(
        "Could not identify an experimental-unit/formulation column."
    )

print(
    "Experimental-unit proxy:",
    unit_col
)


# ======================================================================
# 4. FORCE UNIT KEY INTO A CLEAN STRING COLUMN
# ======================================================================

df["__unit_key"] = (
    df[unit_col]
    .fillna("UNRESOLVED_UNIT")
    .astype(str)
    .str.strip()
)

df.loc[
    df["__unit_key"].isin([
        "",
        "nan",
        "None"
    ]),
    "__unit_key"
] = "UNRESOLVED_UNIT"


print(
    "Unique unit proxies:",
    df["__unit_key"].nunique()
)


# ======================================================================
# 5. IDENTIFY METADATA COLUMNS
# ======================================================================

def find_col(candidates):

    for c in candidates:

        if c in df.columns:

            return c

    return None


drug_col = find_col([
    "recovered_drug_6P",
    "recovered_drug",
    "final_drug",
    "final_drug_6M",
    "drug"
])

polymer_col = find_col([
    "recovered_polymer_6P",
    "recovered_polymer",
    "final_polymer",
    "final_polymer_6M",
    "polymer"
])

carrier_col = find_col([
    "recovered_carrier_6P",
    "recovered_carrier",
    "final_carrier",
    "final_carrier_6M",
    "carrier"
])

condition_col = find_col([
    "recovered_condition_6P",
    "recovered_release_condition",
    "final_release_condition",
    "release_condition",
    "condition"
])


print("Drug column     :", drug_col)
print("Polymer column  :", polymer_col)
print("Carrier column  :", carrier_col)
print("Condition column:", condition_col)


# ======================================================================
# 6. STANDARDIZE METADATA
# ======================================================================

def standardize_value(x):

    s = clean(x)

    if s == "":
        return "NOT_IDENTIFIED"

    if s.upper() in [
        "NOT_IDENTIFIED",
        "UNKNOWN",
        "UNRESOLVED",
        "NONE",
        "NAN"
    ]:
        return "NOT_IDENTIFIED"

    return s


for col in [
    drug_col,
    polymer_col,
    carrier_col,
    condition_col
]:

    if col is not None:

        df[col] = df[col].apply(
            standardize_value
        )


# ======================================================================
# 7. FIELD SUMMARY
# ======================================================================

def summarize_field(group, col):

    if col is None:

        return {
            "value": "NOT_AVAILABLE",
            "n_unique": 0,
            "conflict": False,
            "identified": False
        }

    vals = [
        standardize_value(v)
        for v in group[col]
    ]

    vals = sorted(set(vals))

    meaningful = [
        v for v in vals
        if v != "NOT_IDENTIFIED"
    ]

    if len(meaningful) == 0:

        return {
            "value": "NOT_IDENTIFIED",
            "n_unique": 0,
            "conflict": False,
            "identified": False
        }

    if len(meaningful) == 1:

        value = meaningful[0]

        return {
            "value": value,
            "n_unique": 1,
            "conflict": False,
            "identified": True
        }

    return {
        "value":
            "CONFLICT:" + "|".join(meaningful),

        "n_unique":
            len(meaningful),

        "conflict":
            True,

        "identified":
            False
    }


# ======================================================================
# 8. BUILD UNIT SUMMARY
# ======================================================================

unit_records = []

for unit_key, group in df.groupby(
    "__unit_key",
    dropna=False
):

    drug = summarize_field(
        group,
        drug_col
    )

    polymer = summarize_field(
        group,
        polymer_col
    )

    carrier = summarize_field(
        group,
        carrier_col
    )

    condition = summarize_field(
        group,
        condition_col
    )

    times = pd.to_numeric(
        group["release_time_h"],
        errors="coerce"
    ).dropna()

    releases = pd.to_numeric(
        group["release_percent"],
        errors="coerce"
    ).dropna()

    conflicts = sum([
        drug["conflict"],
        polymer["conflict"],
        carrier["conflict"],
        condition["conflict"]
    ])

    identified_fields = sum([
        drug["identified"],
        polymer["identified"],
        carrier["identified"]
    ])

    if conflicts > 0:

        unit_status = "CONFLICT"

    elif identified_fields >= 3:

        unit_status = "STRONG_UNIT_METADATA"

    elif identified_fields >= 1:

        unit_status = "PARTIAL_UNIT_METADATA"

    else:

        unit_status = "UNRESOLVED_UNIT"


    unit_records.append({

        "__unit_key":
            unit_key,

        "observations":
            len(group),

        "unique_timepoints":
            times.nunique(),

        "min_time_h":
            times.min()
            if len(times)
            else np.nan,

        "max_time_h":
            times.max()
            if len(times)
            else np.nan,

        "drug":
            drug["value"],

        "drug_n_unique":
            drug["n_unique"],

        "drug_conflict":
            drug["conflict"],

        "polymer":
            polymer["value"],

        "polymer_n_unique":
            polymer["n_unique"],

        "polymer_conflict":
            polymer["conflict"],

        "carrier":
            carrier["value"],

        "carrier_n_unique":
            carrier["n_unique"],

        "carrier_conflict":
            carrier["conflict"],

        "release_condition":
            condition["value"],

        "condition_n_unique":
            condition["n_unique"],

        "condition_conflict":
            condition["conflict"],

        "identified_metadata_fields":
            identified_fields,

        "unit_status":
            unit_status
    })


units = pd.DataFrame(
    unit_records
)


# ======================================================================
# 9. EXPLICITLY VERIFY UNIT KEY COLUMN
# ======================================================================

if "__unit_key" not in units.columns:

    raise RuntimeError(
        "Internal error: __unit_key was not created in unit summary."
    )


# ======================================================================
# 10. TIMEPOINT / CONDITION AUDIT
# ======================================================================

units[
    "multiple_timepoints"
] = (
    units[
        "unique_timepoints"
    ] > 1
)


if condition_col is not None:

    condition_counts = (
        df.groupby(
            "__unit_key",
            dropna=False
        )[condition_col]
        .nunique()
        .reset_index(
            name="condition_values_per_unit"
        )
    )

else:

    condition_counts = pd.DataFrame({
        "__unit_key":
            units["__unit_key"],

        "condition_values_per_unit":
            0
    })


# ======================================================================
# 11. SAFE MERGE OF CONDITION COUNTS
# ======================================================================

units = units.merge(
    condition_counts,
    on="__unit_key",
    how="left",
    validate="one_to_one"
)

units[
    "condition_values_per_unit"
] = (
    units[
        "condition_values_per_unit"
    ]
    .fillna(0)
    .astype(int)
)

units[
    "multiple_release_conditions"
] = (
    units[
        "condition_values_per_unit"
    ] > 1
)


# ======================================================================
# 12. MAP UNIT INFORMATION BACK TO ALL 60 OBSERVATIONS
# ======================================================================

status_map = units[
    [
        "__unit_key",
        "unit_status",
        "multiple_timepoints",
        "multiple_release_conditions"
    ]
].copy()


df6q = df.merge(
    status_map,
    on="__unit_key",
    how="left",
    validate="many_to_one"
)


# ======================================================================
# 13. CONSERVATIVE ML CANDIDATE
# ======================================================================

df6q[
    "unit_level_ml_candidate"
] = (
    df6q[
        "unit_status"
    ]
    == "STRONG_UNIT_METADATA"
)


# ======================================================================
# 14. SAFETY CHECKS
# ======================================================================

if len(df6q) != 60:

    raise ValueError(
        f"Observation count changed: "
        f"60 -> {len(df6q)}"
    )

if df6q[
    "observation_id"
].duplicated().any():

    raise ValueError(
        "Duplicate observation IDs detected after 6Q merge."
    )


# ======================================================================
# 15. SAVE OUTPUTS
# ======================================================================

OUT_AUDIT = os.path.join(
    VAL,
    "06Q_experimental_unit_observation_audit.csv"
)

OUT_UNITS = os.path.join(
    VAL,
    "06Q_experimental_unit_summary.csv"
)

OUT_STRONG = os.path.join(
    VAL,
    "06Q_strong_unit_candidates.csv"
)

OUT_REVIEW = os.path.join(
    VAL,
    "06Q_unit_manual_review.csv"
)

df6q.to_csv(
    OUT_AUDIT,
    index=False
)

units.to_csv(
    OUT_UNITS,
    index=False
)

units[
    units[
        "unit_status"
    ] == "STRONG_UNIT_METADATA"
].to_csv(
    OUT_STRONG,
    index=False
)

units[
    units[
        "unit_status"
    ] != "STRONG_UNIT_METADATA"
].to_csv(
    OUT_REVIEW,
    index=False
)


# ======================================================================
# 16. REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6Q-R COMPLETE")
print("=" * 70)

print(
    "Input observations:",
    len(df)
)

print(
    "Unique experimental-unit proxies:",
    len(units)
)

print()
print("Unit status:")
print(
    units[
        "unit_status"
    ].value_counts()
)

print()
print(
    "Units with multiple time points:",
    int(
        units[
            "multiple_timepoints"
        ].sum()
    )
)

print(
    "Units with multiple release conditions:",
    int(
        units[
            "multiple_release_conditions"
        ].sum()
    )
)

print()
print("Unit-level drug identification:")
print(
    units[
        "drug"
    ].value_counts()
)

print()
print("Unit-level polymer identification:")
print(
    units[
        "polymer"
    ].value_counts()
)

print()
print("Unit-level carrier identification:")
print(
    units[
        "carrier"
    ].value_counts()
)

print()
print("Unit-level release conditions:")
print(
    units[
        "release_condition"
    ].value_counts()
)

print()
print(
    "Strong unit metadata candidates:",
    int(
        (
            units["unit_status"]
            == "STRONG_UNIT_METADATA"
        ).sum()
    )
)

print(
    "Partial unit metadata:",
    int(
        (
            units["unit_status"]
            == "PARTIAL_UNIT_METADATA"
        ).sum()
    )
)

print(
    "Conflicting units:",
    int(
        (
            units["unit_status"]
            == "CONFLICT"
        ).sum()
    )
)

print(
    "Unresolved units:",
    int(
        (
            units["unit_status"]
            == "UNRESOLVED_UNIT"
        ).sum()
    )
)

print()
print(
    "Observation preservation:",
    len(df),
    "->",
    len(df6q)
)

print()
print("Saved:")
print(OUT_AUDIT)
print(OUT_UNITS)
print(OUT_STRONG)
print(OUT_REVIEW)

print()
print("IMPORTANT:")
print(
    "Unit consolidation is an audit operation only."
)

print(
    "The original 60 source-validated observations remain intact."
)

print(
    "Multiple time points are not counted as independent experiments."
)

print(
    "Different release conditions are not merged."
)

print(
    "Experimental-unit proxies do not prove independence."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6R — TARGETED EXPERIMENTAL-UNIT EVIDENCE REVIEW
# ======================================================================

import os
import re
import numpy as np
import pandas as pd

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")
RAW = os.path.join(ROOT, "data", "raw")

FILE_6Q = os.path.join(
    VAL,
    "06Q_experimental_unit_observation_audit.csv"
)

FILE_UNITS = os.path.join(
    VAL,
    "06Q_experimental_unit_summary.csv"
)

# ----------------------------------------------------------------------
# Locate Code 3 structural-cell file
# ----------------------------------------------------------------------

print("=" * 70)
print("CODE 6R — TARGETED EXPERIMENTAL-UNIT EVIDENCE REVIEW")
print("=" * 70)

df6q = pd.read_csv(FILE_6Q)
units = pd.read_csv(FILE_UNITS)

print("6Q observations:", len(df6q))
print("6Q units:", len(units))

if len(df6q) != 60:
    raise ValueError(
        f"Expected 60 observations from 6Q, found {len(df6q)}"
    )

# ======================================================================
# 1. LOCATE CODE 3
# ======================================================================

candidate_files = [
    os.path.join(
        RAW,
        "03_table_cell_records.csv"
    ),
    os.path.join(
        RAW,
        "tables",
        "03_extracted_table_cells.csv"
    )
]

# Recursive fallback
if not any(os.path.exists(x) for x in candidate_files):

    for root, dirs, files in os.walk(RAW):

        for f in files:

            low = f.lower()

            if (
                "03_" in low
                and "table" in low
                and "cell" in low
                and low.endswith(".csv")
            ):

                candidate_files.append(
                    os.path.join(root, f)
                )


existing = [
    x for x in candidate_files
    if os.path.exists(x)
]

if not existing:

    raise FileNotFoundError(
        "Could not locate Code 3 structural-cell dataset."
    )

FILE_3 = existing[0]

print(
    "Selected Code 3 file:",
    FILE_3
)

cells = pd.read_csv(FILE_3)

print(
    "Code 3 records:",
    len(cells)
)

print(
    "Code 3 columns:",
    list(cells.columns)
)


# ======================================================================
# 2. REQUIRED CODE 3 STRUCTURAL COLUMNS
# ======================================================================

required_3 = [
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell"
]

missing_3 = [
    c for c in required_3
    if c not in cells.columns
]

if missing_3:

    raise ValueError(
        "Code 3 missing required columns:\n"
        + "\n".join(missing_3)
    )


# ======================================================================
# 3. CLEAN STRUCTURAL KEYS
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


for c in [
    "pmcid",
    "table_index",
    "row_index",
    "column_index"
]:

    cells[c] = cells[c].apply(clean)


df6q["pmcid"] = df6q["pmcid"].apply(clean)
df6q["table_index"] = df6q["table_index"].apply(clean)
df6q["row_index"] = df6q["row_index"].apply(clean)


# ======================================================================
# 4. IDENTIFY UNIT KEY
# ======================================================================

unit_col = None

for c in [
    "experimental_unit_proxy_6P",
    "experimental_unit_proxy",
    "experimental_unit_id",
    "final_formulation_id"
]:

    if c in df6q.columns:

        unit_col = c
        break

if unit_col is None:

    raise ValueError(
        "Could not identify experimental-unit proxy in 6Q."
    )

print(
    "Unit key:",
    unit_col
)


# ======================================================================
# 5. TARGET THE 21 UNITS
# ======================================================================

target_units = (
    df6q[
        [
            unit_col,
            "pmcid",
            "table_index"
        ]
    ]
    .drop_duplicates()
)

print(
    "Target unit/table combinations:",
    len(target_units)
)


# ======================================================================
# 6. RESTRICT CODE 3 TO VALIDATED TABLES
# ======================================================================

table_keys = target_units[
    [
        "pmcid",
        "table_index"
    ]
].drop_duplicates()

cells_target = cells.merge(
    table_keys,
    on=[
        "pmcid",
        "table_index"
    ],
    how="inner"
)

print(
    "Code 3 cells in targeted tables:",
    len(cells_target)
)


# ======================================================================
# 7. CREATE FULL ROW TEXT WHEN AVAILABLE
# ======================================================================

if "row_text" not in cells_target.columns:

    cells_target["row_text"] = (
        cells_target
        .groupby(
            [
                "pmcid",
                "table_index",
                "row_index"
            ]
        )["raw_cell"]
        .transform(
            lambda x:
                " | ".join(
                    x.astype(str)
                )
        )
    )


# ======================================================================
# 8. CREATE TABLE-LEVEL TEXT
# ======================================================================

if "full_table_text" not in cells_target.columns:

    cells_target["full_table_text"] = (
        cells_target
        .groupby(
            [
                "pmcid",
                "table_index"
            ]
        )["raw_cell"]
        .transform(
            lambda x:
                " | ".join(
                    x.astype(str)
                )
        )
    )


# ======================================================================
# 9. METADATA VOCABULARIES
# ======================================================================

DRUG_TERMS = {
    "doxorubicin": [
        "doxorubicin",
        "dox"
    ],
    "5-fluorouracil": [
        "5-fluorouracil",
        "5 fluorouracil",
        "5-fu",
        "5 fu"
    ],
    "paclitaxel": [
        "paclitaxel"
    ],
    "curcumin": [
        "curcumin"
    ],
    "metformin": [
        "metformin"
    ],
    "ibuprofen": [
        "ibuprofen"
    ],
    "gemcitabine": [
        "gemcitabine"
    ],
    "rifampicin": [
        "rifampicin"
    ],
    "vancomycin": [
        "vancomycin"
    ]
}

POLYMER_TERMS = [
    "PLGA",
    "PLA",
    "PCL",
    "PEG",
    "PEO",
    "PVA",
    "chitosan",
    "alginate",
    "cellulose",
    "gelatin",
    "poloxamer",
    "polyurethane"
]

CARRIER_TERMS = [
    "nanoparticle",
    "nanoparticles",
    "nanocarrier",
    "microparticle",
    "microparticles",
    "microsphere",
    "microspheres",
    "liposome",
    "liposomes",
    "hydrogel",
    "hydrogels",
    "nanofiber",
    "nanofibers",
    "film",
    "implant",
    "bead",
    "beads",
    "micelle",
    "micelles"
]


# ======================================================================
# 10. EXTRACTION HELPERS
# ======================================================================

def find_drugs(text):

    t = clean(text).lower()

    found = []

    for canonical, terms in DRUG_TERMS.items():

        for term in terms:

            if term.lower() in t:

                found.append(canonical)
                break

    return sorted(set(found))


def find_polymers(text):

    t = clean(text).lower()

    found = []

    for term in POLYMER_TERMS:

        if term.lower() in t:

            found.append(term)

    return sorted(set(found))


def find_carriers(text):

    t = clean(text).lower()

    found = []

    for term in CARRIER_TERMS:

        if term.lower() in t:

            found.append(term)

    return sorted(set(found))


def find_conditions(text):

    t = clean(text).lower()

    found = []

    # Explicit pH expressions
    matches = re.findall(
        r"pH\s*=?\s*(\d+(?:\.\d+)?)",
        t,
        flags=re.I
    )

    for m in matches:

        found.append(
            "pH " + m
        )

    return sorted(set(found))


# ======================================================================
# 11. BUILD TARGETED EVIDENCE
# ======================================================================

evidence_records = []

for _, unit in target_units.iterrows():

    unit_id = unit[unit_col]
    pmcid = unit["pmcid"]
    table = unit["table_index"]

    subset = cells_target[
        (
            cells_target["pmcid"]
            == pmcid
        )
        &
        (
            cells_target["table_index"]
            == table
        )
    ].copy()

    # --------------------------------------------------------------
    # Full table evidence
    # --------------------------------------------------------------

    if len(subset) == 0:

        evidence_records.append({

            "experimental_unit_proxy":
                unit_id,

            "pmcid":
                pmcid,

            "table_index":
                table,

            "evidence_type":
                "NO_SOURCE_CELLS",

            "row_index":
                "",

            "column_index":
                "",

            "raw_cell":
                "",

            "row_text":
                "",

            "table_text":
                "",

            "drug_evidence":
                "",

            "polymer_evidence":
                "",

            "carrier_evidence":
                "",

            "condition_evidence":
                "",

            "evidence_strength":
                "NONE"
        })

        continue


    # --------------------------------------------------------------
    # Get source-table context
    # --------------------------------------------------------------

    for row_idx, row_group in subset.groupby(
        "row_index",
        dropna=False
    ):

        row_group = row_group.sort_values(
            "column_index"
        )

        row_text = " | ".join(
            row_group["raw_cell"]
            .astype(str)
            .tolist()
        )

        table_text = clean(
            row_group["full_table_text"].iloc[0]
        )

        # Headers
        header_text = ""

        if "header_text" in row_group.columns:

            header_text = " | ".join(
                row_group["header_text"]
                .astype(str)
                .tolist()
            )

        combined = " | ".join([
            header_text,
            row_text,
            table_text
        ])

        drugs = find_drugs(
            combined
        )

        polymers = find_polymers(
            combined
        )

        carriers = find_carriers(
            combined
        )

        conditions = find_conditions(
            combined
        )

        # ----------------------------------------------------------
        # Evidence strength
        # ----------------------------------------------------------

        metadata_hits = (
            len(drugs)
            + len(polymers)
            + len(carriers)
        )

        if metadata_hits >= 3:

            strength = "STRONG_CONTEXT"

        elif metadata_hits >= 1:

            strength = "PARTIAL_CONTEXT"

        else:

            strength = "NO_METADATA_HIT"


        evidence_records.append({

            "experimental_unit_proxy":
                unit_id,

            "pmcid":
                pmcid,

            "table_index":
                table,

            "evidence_type":
                "SOURCE_ROW",

            "row_index":
                row_idx,

            "column_index":
                "|".join(
                    row_group[
                        "column_index"
                    ]
                    .astype(str)
                    .tolist()
                ),

            "raw_cell":
                " | ".join(
                    row_group[
                        "raw_cell"
                    ]
                    .astype(str)
                    .tolist()
                ),

            "row_text":
                row_text,

            "table_text":
                table_text,

            "header_text":
                header_text,

            "drug_evidence":
                ";".join(drugs),

            "polymer_evidence":
                ";".join(polymers),

            "carrier_evidence":
                ";".join(carriers),

            "condition_evidence":
                ";".join(conditions),

            "evidence_strength":
                strength
        })


evidence = pd.DataFrame(
    evidence_records
)


# ======================================================================
# 12. UNIT-LEVEL EVIDENCE SUMMARY
# ======================================================================

summary_records = []

for unit_id, group in evidence.groupby(
    "experimental_unit_proxy",
    dropna=False
):

    all_drugs = sorted(
        set(
            x
            for s in group["drug_evidence"]
            for x in clean(s).split(";")
            if x
        )
    )

    all_polymers = sorted(
        set(
            x
            for s in group["polymer_evidence"]
            for x in clean(s).split(";")
            if x
        )
    )

    all_carriers = sorted(
        set(
            x
            for s in group["carrier_evidence"]
            for x in clean(s).split(";")
            if x
        )
    )

    all_conditions = sorted(
        set(
            x
            for s in group["condition_evidence"]
            for x in clean(s).split(";")
            if x
        )
    )

    # --------------------------------------------------------------
    # Remove obvious non-metadata contamination
    # --------------------------------------------------------------

    all_polymers = [
        x for x in all_polymers
        if x.lower()
        not in [
            "nanoparticle",
            "nanoparticles",
            "nanocarrier",
            "microparticle",
            "microparticles",
            "microsphere",
            "microspheres",
            "liposome",
            "liposomes",
            "hydrogel",
            "hydrogels",
            "nanofiber",
            "nanofibers",
            "film",
            "implant",
            "bead",
            "beads",
            "micelle",
            "micelles"
        ]
    ]

    # --------------------------------------------------------------
    # Classify evidence
    # --------------------------------------------------------------

    n_fields = sum([
        len(all_drugs) > 0,
        len(all_polymers) > 0,
        len(all_carriers) > 0
    ])

    if n_fields >= 3:

        unit_evidence_status = (
            "STRONG_MULTI_FIELD_EVIDENCE"
        )

    elif n_fields == 2:

        unit_evidence_status = (
            "GOOD_MULTI_FIELD_EVIDENCE"
        )

    elif n_fields == 1:

        unit_evidence_status = (
            "PARTIAL_EVIDENCE"
        )

    else:

        unit_evidence_status = (
            "NO_DIRECT_METADATA_EVIDENCE"
        )

    # Explicit ambiguity
    if len(all_polymers) > 1:

        polymer_status = "AMBIGUOUS"

    elif len(all_polymers) == 1:

        polymer_status = "SINGLE_CANDIDATE"

    else:

        polymer_status = "UNRESOLVED"


    if len(all_carriers) > 1:

        carrier_status = "AMBIGUOUS"

    elif len(all_carriers) == 1:

        carrier_status = "SINGLE_CANDIDATE"

    else:

        carrier_status = "UNRESOLVED"


    summary_records.append({

        "experimental_unit_proxy":
            unit_id,

        "pmcid":
            group["pmcid"].iloc[0],

        "table_index":
            group["table_index"].iloc[0],

        "source_rows_examined":
            group["row_index"].nunique(),

        "drug_evidence":
            ";".join(all_drugs)
            if all_drugs
            else "NOT_IDENTIFIED",

        "polymer_evidence":
            ";".join(all_polymers)
            if all_polymers
            else "NOT_IDENTIFIED",

        "carrier_evidence":
            ";".join(all_carriers)
            if all_carriers
            else "NOT_IDENTIFIED",

        "condition_evidence":
            ";".join(all_conditions)
            if all_conditions
            else "NOT_IDENTIFIED",

        "polymer_status":
            polymer_status,

        "carrier_status":
            carrier_status,

        "unit_evidence_status":
            unit_evidence_status
    })


unit_evidence_summary = pd.DataFrame(
    summary_records
)


# ======================================================================
# 13. MERGE ORIGINAL 6Q STATUS
# ======================================================================

original_unit_status = units[
    [
        "__unit_key",
        "unit_status",
        "observations",
        "unique_timepoints"
    ]
].copy()

original_unit_status = original_unit_status.rename(
    columns={
        "__unit_key":
            "experimental_unit_proxy"
    }
)

unit_evidence_summary = (
    unit_evidence_summary
    .merge(
        original_unit_status,
        on="experimental_unit_proxy",
        how="left",
        validate="one_to_one"
    )
)


# ======================================================================
# 14. MANUAL REVIEW CLASSIFICATION
# ======================================================================

def final_review_class(row):

    status = row[
        "unit_evidence_status"
    ]

    polymer_status = row[
        "polymer_status"
    ]

    carrier_status = row[
        "carrier_status"
    ]

    if (
        status
        == "STRONG_MULTI_FIELD_EVIDENCE"
        and polymer_status
        == "SINGLE_CANDIDATE"
        and carrier_status
        == "SINGLE_CANDIDATE"
    ):

        return "HIGH_PRIORITY_VALIDATION_CANDIDATE"

    if (
        status
        in [
            "STRONG_MULTI_FIELD_EVIDENCE",
            "GOOD_MULTI_FIELD_EVIDENCE"
        ]
    ):

        return "MANUAL_CONFIRMATION_REQUIRED"

    if (
        polymer_status
        == "AMBIGUOUS"
        or carrier_status
        == "AMBIGUOUS"
    ):

        return "AMBIGUOUS_MANUAL_REVIEW"

    return "LOW_EVIDENCE_MANUAL_REVIEW"


unit_evidence_summary[
    "final_review_class"
] = unit_evidence_summary.apply(
    final_review_class,
    axis=1
)


# ======================================================================
# 15. CONSERVATIVE FINAL CANDIDATE FLAG
# ======================================================================

unit_evidence_summary[
    "candidate_for_metadata_validation"
] = (
    unit_evidence_summary[
        "final_review_class"
    ]
    == "HIGH_PRIORITY_VALIDATION_CANDIDATE"
)


# ======================================================================
# 16. SAVE
# ======================================================================

OUT_EVIDENCE = os.path.join(
    VAL,
    "06R_targeted_unit_source_evidence.csv"
)

OUT_SUMMARY = os.path.join(
    VAL,
    "06R_unit_evidence_summary.csv"
)

OUT_PRIORITY = os.path.join(
    VAL,
    "06R_high_priority_unit_candidates.csv"
)

OUT_MANUAL = os.path.join(
    VAL,
    "06R_manual_unit_review.csv"
)

evidence.to_csv(
    OUT_EVIDENCE,
    index=False
)

unit_evidence_summary.to_csv(
    OUT_SUMMARY,
    index=False
)

unit_evidence_summary[
    unit_evidence_summary[
        "candidate_for_metadata_validation"
    ]
].to_csv(
    OUT_PRIORITY,
    index=False
)

unit_evidence_summary[
    ~unit_evidence_summary[
        "candidate_for_metadata_validation"
    ]
].to_csv(
    OUT_MANUAL,
    index=False
)


# ======================================================================
# 17. REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6R COMPLETE")
print("=" * 70)

print(
    "Input observations:",
    len(df6q)
)

print(
    "Experimental units reviewed:",
    len(unit_evidence_summary)
)

print(
    "Source rows examined:",
    len(evidence)
)

print()
print("Unit evidence status:")
print(
    unit_evidence_summary[
        "unit_evidence_status"
    ].value_counts()
)

print()
print("Polymer evidence status:")
print(
    unit_evidence_summary[
        "polymer_status"
    ].value_counts()
)

print()
print("Carrier evidence status:")
print(
    unit_evidence_summary[
        "carrier_status"
    ].value_counts()
)

print()
print("Final review classification:")
print(
    unit_evidence_summary[
        "final_review_class"
    ].value_counts()
)

print()
print(
    "High-priority metadata validation candidates:",
    int(
        unit_evidence_summary[
            "candidate_for_metadata_validation"
        ].sum()
    )
)

print()
print("Drug evidence:")
print(
    unit_evidence_summary[
        "drug_evidence"
    ].value_counts()
)

print()
print("Polymer evidence:")
print(
    unit_evidence_summary[
        "polymer_evidence"
    ].value_counts()
)

print()
print("Carrier evidence:")
print(
    unit_evidence_summary[
        "carrier_evidence"
    ].value_counts()
)

print()
print("Condition evidence:")
print(
    unit_evidence_summary[
        "condition_evidence"
    ].value_counts()
)

print()
print("Saved:")
print(OUT_EVIDENCE)
print(OUT_SUMMARY)
print(OUT_PRIORITY)
print(OUT_MANUAL)

print()
print("IMPORTANT:")
print(
    "This stage only audits source evidence for the 21 experimental units."
)

print(
    "No release values were modified."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "Metadata are not accepted merely because a term occurs somewhere "
    "in an article."
)

print(
    "Exact row/table evidence is preferred."
)

print(
    "Ambiguous polymer/carrier evidence remains manual review."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6S — PRECISION METADATA EVIDENCE CLEANUP & FALSE-POSITIVE FILTER
# ======================================================================

import os
import re
import numpy as np
import pandas as pd

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")

FILE_6R = os.path.join(
    VAL,
    "06R_targeted_unit_source_evidence.csv"
)

FILE_6R_SUMMARY = os.path.join(
    VAL,
    "06R_unit_evidence_summary.csv"
)

FILE_6Q = os.path.join(
    VAL,
    "06Q_experimental_unit_observation_audit.csv"
)

print("=" * 70)
print("CODE 6S — PRECISION METADATA EVIDENCE CLEANUP & FALSE-POSITIVE FILTER")
print("=" * 70)

evidence = pd.read_csv(FILE_6R)
summary = pd.read_csv(FILE_6R_SUMMARY)
df6q = pd.read_csv(FILE_6Q)

print("6R evidence rows:", len(evidence))
print("6R units:", len(summary))
print("6Q observations:", len(df6q))

if len(df6q) != 60:
    raise ValueError(
        f"Expected 60 observations in 6Q, found {len(df6q)}."
    )


# ======================================================================
# 1. CLEANING
# ======================================================================

def clean(x):

    if pd.isna(x):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(x)
    ).strip()


for col in evidence.columns:

    if evidence[col].dtype == "object":

        evidence[col] = evidence[col].apply(clean)


# ======================================================================
# 2. PRECISION DRUG EXTRACTION
# ======================================================================

DRUG_PATTERNS = {
    "doxorubicin": [
        r"\bdoxorubicin\b",
        r"\bDOX\b"
    ],

    "5-fluorouracil": [
        r"\b5[\s-]?fluorouracil\b",
        r"\b5[\s-]?FU\b"
    ],

    "paclitaxel": [
        r"\bpaclitaxel\b"
    ],

    "curcumin": [
        r"\bcurcumin\b"
    ],

    "metformin": [
        r"\bmetformin\b"
    ],

    "ibuprofen": [
        r"\bibuprofen\b"
    ],

    "gemcitabine": [
        r"\bgemcitabine\b"
    ],

    "rifampicin": [
        r"\brifampicin\b"
    ],

    "vancomycin": [
        r"\bvancomycin\b"
    ]
}


def exact_drugs(text):

    t = clean(text)

    found = []

    for canonical, patterns in DRUG_PATTERNS.items():

        for pattern in patterns:

            if re.search(
                pattern,
                t,
                flags=re.I
            ):

                found.append(
                    canonical
                )

                break

    return sorted(
        set(found)
    )


# ======================================================================
# 3. PRECISE POLYMER EXTRACTION
# ======================================================================

POLYMER_PATTERNS = [
    r"\bPLGA\b",
    r"\bPLA\b",
    r"\bPCL\b",
    r"\bPEG\b",
    r"\bPEO\b",
    r"\bPVA\b",
    r"\bchitosan\b",
    r"\balginate\b",
    r"\bcellulose\b",
    r"\bgelatin\b",
    r"\bpoloxamer\b",
    r"\bpolyurethane\b"
]


def exact_polymers(text):

    t = clean(text)

    found = []

    for pattern in POLYMER_PATTERNS:

        m = re.search(
            pattern,
            t,
            flags=re.I
        )

        if m:

            found.append(
                m.group(0)
            )

    return sorted(
        set(found),
        key=str.lower
    )


# ======================================================================
# 4. PRECISE CARRIER EXTRACTION
# ======================================================================

CARRIER_PATTERNS = [
    r"\bnanoparticles?\b",
    r"\bnanocarrier\b",
    r"\bmicroparticles?\b",
    r"\bmicrospheres?\b",
    r"\bliposomes?\b",
    r"\bhydrogels?\b",
    r"\bnanofibers?\b",
    r"\bfilms?\b",
    r"\bimplants?\b",
    r"\bbeads?\b",
    r"\bmicelles?\b"
]


def exact_carriers(text):

    t = clean(text)

    found = []

    for pattern in CARRIER_PATTERNS:

        m = re.search(
            pattern,
            t,
            flags=re.I
        )

        if m:

            found.append(
                m.group(0)
            )

    return sorted(
        set(found),
        key=str.lower
    )


# ======================================================================
# 5. STRICT pH EXTRACTION
# ======================================================================
#
# IMPORTANT:
# A number is NOT considered pH merely because it is between 0 and 14.
#
# Valid evidence must contain an explicit pH label, such as:
#
#   pH 7.4
#   pH = 7.4
#   pH: 7.4
#
# This prevents values such as 41.5 or 47.3 from becoming pH values.
# ======================================================================

PH_PATTERN = re.compile(
    r"""
    \bpH
    \s*
    (?:=|:)?
    \s*
    (
        \d+(?:\.\d+)?
    )
    \b
    """,
    flags=re.I | re.X
)


def exact_ph(text):

    t = clean(text)

    matches = PH_PATTERN.findall(t)

    values = []

    for m in matches:

        try:

            value = float(m)

        except Exception:

            continue

        if 0 <= value <= 14:

            if value.is_integer():

                values.append(
                    f"pH {int(value)}"
                )

            else:

                values.append(
                    f"pH {value:g}"
                )

    return sorted(
        set(values)
    )


# ======================================================================
# 6. REMOVE NUMERIC pH FALSE POSITIVES
# ======================================================================

def remove_bad_ph(value):

    s = clean(value)

    if not s:

        return ""

    parts = re.split(
        r"[;|]",
        s
    )

    valid = []

    for part in parts:

        part = clean(part)

        m = re.fullmatch(
            r"pH\s*(?:=|:)?\s*(\d+(?:\.\d+)?)",
            part,
            flags=re.I
        )

        if not m:
            continue

        try:

            value = float(
                m.group(1)
            )

        except Exception:

            continue

        if 0 <= value <= 14:

            valid.append(
                f"pH {value:g}"
            )

    return ";".join(
        sorted(
            set(valid)
        )
    )


# ======================================================================
# 7. EVIDENCE-TYPE CONTROL
# ======================================================================
#
# We distinguish:
#
#   SOURCE_ROW      = row-level evidence
#   TABLE_CONTEXT   = table-wide context
#
# Table-wide vocabulary is NOT automatically accepted as formulation
# identity.
# ======================================================================

if "evidence_type" not in evidence.columns:

    evidence["evidence_type"] = (
        "SOURCE_ROW"
    )

evidence[
    "evidence_scope"
] = np.where(
    evidence[
        "evidence_type"
    ].eq("SOURCE_ROW"),
    "ROW_LEVEL",
    "TABLE_LEVEL"
)


# ======================================================================
# 8. RECALCULATE EVIDENCE FROM ACTUAL SOURCE ROW
# ======================================================================

evidence[
    "strict_drug_evidence"
] = evidence[
    "row_text"
].apply(
    exact_drugs
).apply(
    lambda x: ";".join(x)
)


evidence[
    "strict_polymer_evidence"
] = evidence[
    "row_text"
].apply(
    exact_polymers
).apply(
    lambda x: ";".join(x)
)


evidence[
    "strict_carrier_evidence"
] = evidence[
    "row_text"
].apply(
    exact_carriers
).apply(
    lambda x: ";".join(x)
)


evidence[
    "strict_condition_evidence"
] = evidence[
    "row_text"
].apply(
    exact_ph
).apply(
    lambda x: ";".join(x)
)


# ======================================================================
# 9. HEADER-ASSISTED EVIDENCE
# ======================================================================
#
# Header evidence can identify what a column represents, but is not
# sufficient by itself to assign a formulation unless the row contains
# compatible formulation information.
# ======================================================================

if "header_text" in evidence.columns:

    evidence[
        "header_drug_evidence"
    ] = evidence[
        "header_text"
    ].apply(
        exact_drugs
    ).apply(
        lambda x: ";".join(x)
    )

    evidence[
        "header_polymer_evidence"
    ] = evidence[
        "header_text"
    ].apply(
        exact_polymers
    ).apply(
        lambda x: ";".join(x)
    )

    evidence[
        "header_carrier_evidence"
    ] = evidence[
        "header_text"
    ].apply(
        exact_carriers
    ).apply(
        lambda x: ";".join(x)
    )

    evidence[
        "header_condition_evidence"
    ] = evidence[
        "header_text"
    ].apply(
        exact_ph
    ).apply(
        lambda x: ";".join(x)
    )

else:

    evidence[
        "header_drug_evidence"
    ] = ""

    evidence[
        "header_polymer_evidence"
    ] = ""

    evidence[
        "header_carrier_evidence"
    ] = ""

    evidence[
        "header_condition_evidence"
    ] = ""


# ======================================================================
# 10. STRICT CONDITION EVIDENCE
# ======================================================================

evidence[
    "strict_condition_evidence"
] = evidence.apply(
    lambda r: ";".join(
        sorted(
            set(
                (
                    clean(
                        r[
                            "strict_condition_evidence"
                        ]
                    )
                    + ";"
                    + clean(
                        r[
                            "header_condition_evidence"
                        ]
                    )
                )
                .split(";")
            )
            -
            {""}
        )
    ),
    axis=1
)


# ======================================================================
# 11. UNIT-LEVEL CONSOLIDATION
# ======================================================================

unit_records = []

for unit_id, group in evidence.groupby(
    "experimental_unit_proxy",
    dropna=False
):

    # --------------------------------------------------------------
    # Only row-level evidence is allowed to establish metadata.
    # --------------------------------------------------------------

    row_group = group[
        group[
            "evidence_scope"
        ] == "ROW_LEVEL"
    ]

    drug_values = sorted(
        set(
            x
            for s in row_group[
                "strict_drug_evidence"
            ]
            for x in clean(s).split(";")
            if x
        )
    )

    polymer_values = sorted(
        set(
            x
            for s in row_group[
                "strict_polymer_evidence"
            ]
            for x in clean(s).split(";")
            if x
        ),
        key=str.lower
    )

    carrier_values = sorted(
        set(
            x
            for s in row_group[
                "strict_carrier_evidence"
            ]
            for x in clean(s).split(";")
            if x
        ),
        key=str.lower
    )

    condition_values = sorted(
        set(
            x
            for s in row_group[
                "strict_condition_evidence"
            ]
            for x in clean(s).split(";")
            if x
        ),
        key=str.lower
    )

    # --------------------------------------------------------------
    # Filter polymer/carrier terms that are clearly carrier terms.
    # --------------------------------------------------------------

    carrier_lower = {
        x.lower()
        for x in [
            "nanoparticle",
            "nanoparticles",
            "nanocarrier",
            "microparticle",
            "microparticles",
            "microsphere",
            "microspheres",
            "liposome",
            "liposomes",
            "hydrogel",
            "hydrogels",
            "nanofiber",
            "nanofibers",
            "film",
            "films",
            "implant",
            "implants",
            "bead",
            "beads",
            "micelle",
            "micelles"
        ]
    }

    polymer_values = [
        x for x in polymer_values
        if x.lower()
        not in carrier_lower
    ]

    # --------------------------------------------------------------
    # Evidence consistency
    # --------------------------------------------------------------

    drug_n = len(drug_values)
    polymer_n = len(polymer_values)
    carrier_n = len(carrier_values)
    condition_n = len(condition_values)

    if polymer_n == 1:

        polymer_status = (
            "SINGLE_ROW_CANDIDATE"
        )

    elif polymer_n > 1:

        polymer_status = (
            "MULTIPLE_ROW_CANDIDATES"
        )

    else:

        polymer_status = (
            "UNRESOLVED"
        )


    if carrier_n == 1:

        carrier_status = (
            "SINGLE_ROW_CANDIDATE"
        )

    elif carrier_n > 1:

        carrier_status = (
            "MULTIPLE_ROW_CANDIDATES"
        )

    else:

        carrier_status = (
            "UNRESOLVED"
        )


    if condition_n == 1:

        condition_status = (
            "EXPLICIT_PH"
        )

    elif condition_n > 1:

        condition_status = (
            "MULTIPLE_EXPLICIT_PH"
        )

    else:

        condition_status = (
            "UNRESOLVED"
        )


    # --------------------------------------------------------------
    # Conservative status
    # --------------------------------------------------------------

    confirmed_fields = sum([
        drug_n == 1,
        polymer_n == 1,
        carrier_n == 1
    ])

    if (
        confirmed_fields == 3
        and condition_n <= 1
    ):

        status = (
            "STRONG_ROW_SUPPORTED_UNIT"
        )

    elif confirmed_fields >= 2:

        status = (
            "GOOD_ROW_SUPPORTED_UNIT"
        )

    elif confirmed_fields == 1:

        status = (
            "PARTIAL_ROW_SUPPORTED_UNIT"
        )

    else:

        status = (
            "NO_ROW_SUPPORTED_METADATA"
        )


    unit_records.append({

        "experimental_unit_proxy":
            unit_id,

        "pmcid":
            group["pmcid"].iloc[0],

        "table_index":
            group["table_index"].iloc[0],

        "source_rows":
            group[
                "row_index"
            ].nunique(),

        "drug":
            ";".join(drug_values)
            if drug_values
            else "NOT_IDENTIFIED",

        "polymer":
            ";".join(polymer_values)
            if polymer_values
            else "NOT_IDENTIFIED",

        "carrier":
            ";".join(carrier_values)
            if carrier_values
            else "NOT_IDENTIFIED",

        "release_condition":
            ";".join(condition_values)
            if condition_values
            else "NOT_IDENTIFIED",

        "drug_n":
            drug_n,

        "polymer_n":
            polymer_n,

        "carrier_n":
            carrier_n,

        "condition_n":
            condition_n,

        "polymer_status":
            polymer_status,

        "carrier_status":
            carrier_status,

        "condition_status":
            condition_status,

        "unit_status_6S":
            status
    })


units6s = pd.DataFrame(
    unit_records
)


# ======================================================================
# 12. MANUAL-REVIEW CLASSIFICATION
# ======================================================================

def review_class(row):

    if (
        row[
            "unit_status_6S"
        ]
        == "STRONG_ROW_SUPPORTED_UNIT"
    ):

        return "HIGH_PRIORITY_MANUAL_CONFIRMATION"

    if (
        row[
            "unit_status_6S"
        ]
        == "GOOD_ROW_SUPPORTED_UNIT"
    ):

        return "MANUAL_CONFIRMATION"

    if (
        row[
            "polymer_status"
        ]
        == "MULTIPLE_ROW_CANDIDATES"
        or
        row[
            "carrier_status"
        ]
        == "MULTIPLE_ROW_CANDIDATES"
    ):

        return "AMBIGUOUS_METADATA"

    return "LOW_EVIDENCE_MANUAL_REVIEW"


units6s[
    "manual_review_class"
] = units6s.apply(
    review_class,
    axis=1
)


# ======================================================================
# 13. STRICT ML CANDIDATE
# ======================================================================

units6s[
    "strict_ml_metadata_candidate"
] = (
    units6s[
        "unit_status_6S"
    ]
    == "STRONG_ROW_SUPPORTED_UNIT"
)


# ======================================================================
# 14. MAP UNIT RESULT BACK TO 60 OBSERVATIONS
# ======================================================================

df6s = df6q.merge(
    units6s[
        [
            "experimental_unit_proxy",
            "unit_status_6S",
            "manual_review_class",
            "strict_ml_metadata_candidate"
        ]
    ],
    left_on="experimental_unit_proxy_6P",
    right_on="experimental_unit_proxy",
    how="left",
    validate="many_to_one"
)

if len(df6s) != 60:

    raise ValueError(
        f"Observation count changed during 6S merge: "
        f"60 -> {len(df6s)}"
    )


# ======================================================================
# 15. FALSE-pH AUDIT
# ======================================================================

old_ph_values = []

if "condition_evidence" in summary.columns:

    for s in summary[
        "condition_evidence"
    ]:

        old_ph_values.extend(
            clean(s).split(";")
        )

old_ph_values = [
    x for x in old_ph_values
    if x
]

false_ph_like = [
    x for x in old_ph_values
    if not re.fullmatch(
        r"pH\s*(?:=|:)?\s*(?:[0-9]|1[0-4])(?:\.[0-9]+)?",
        x,
        flags=re.I
    )
]


# ======================================================================
# 16. SAVE
# ======================================================================

OUT_EVIDENCE = os.path.join(
    VAL,
    "06S_precision_unit_source_evidence.csv"
)

OUT_UNITS = os.path.join(
    VAL,
    "06S_precision_unit_summary.csv"
)

OUT_CANDIDATES = os.path.join(
    VAL,
    "06S_strict_metadata_candidates.csv"
)

OUT_REVIEW = os.path.join(
    VAL,
    "06S_manual_metadata_review.csv"
)

OUT_AUDIT = os.path.join(
    VAL,
    "06S_false_positive_audit.csv"
)

evidence.to_csv(
    OUT_EVIDENCE,
    index=False
)

units6s.to_csv(
    OUT_UNITS,
    index=False
)

units6s[
    units6s[
        "strict_ml_metadata_candidate"
    ]
].to_csv(
    OUT_CANDIDATES,
    index=False
)

units6s[
    ~units6s[
        "strict_ml_metadata_candidate"
    ]
].to_csv(
    OUT_REVIEW,
    index=False
)

pd.DataFrame({
    "old_6R_condition_value": old_ph_values,
    "flagged_false_positive": [
        x in false_ph_like
        for x in old_ph_values
    ]
}).to_csv(
    OUT_AUDIT,
    index=False
)


# ======================================================================
# 17. FINAL REPORT
# ======================================================================

print()
print("=" * 70)
print("CODE 6S COMPLETE")
print("=" * 70)

print(
    "6R evidence rows:",
    len(evidence)
)

print(
    "Experimental units:",
    len(units6s)
)

print()
print("Strict unit status:")
print(
    units6s[
        "unit_status_6S"
    ].value_counts()
)

print()
print("Polymer status:")
print(
    units6s[
        "polymer_status"
    ].value_counts()
)

print()
print("Carrier status:")
print(
    units6s[
        "carrier_status"
    ].value_counts()
)

print()
print("Condition status:")
print(
    units6s[
        "condition_status"
    ].value_counts()
)

print()
print("Manual-review classification:")
print(
    units6s[
        "manual_review_class"
    ].value_counts()
)

print()
print(
    "Strict metadata candidates:",
    int(
        units6s[
            "strict_ml_metadata_candidate"
        ].sum()
    )
)

print()
print(
    "6R suspicious pH values identified:",
    len(false_ph_like)
)

if false_ph_like:

    print()
    print(
        "Examples of filtered values:"
    )

    for x in false_ph_like[:20]:

        print(
            "  ",
            x
        )

print()
print("Strict unit summary:")

print(
    units6s[
        [
            "experimental_unit_proxy",
            "pmcid",
            "table_index",
            "drug",
            "polymer",
            "carrier",
            "release_condition",
            "unit_status_6S"
        ]
    ].to_string(
        index=False
    )
)

print()
print("Observation preservation:")
print(
    "Input:",
    len(df6q)
)

print(
    "Output:",
    len(df6s)
)

print()
print("Saved:")
print(OUT_EVIDENCE)
print(OUT_UNITS)
print(OUT_CANDIDATES)
print(OUT_REVIEW)
print(OUT_AUDIT)

print()
print("IMPORTANT:")
print(
    "Only row-level source evidence is used for strict metadata "
    "candidate classification."
)

print(
    "Table-wide vocabulary is not automatically assigned to a formulation."
)

print(
    "A numeric value is never treated as pH without an explicit pH label."
)

print(
    "No release values were changed."
)

print(
    "No observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were duplicated."
)

print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# CODE 6T — FINAL STRICT METADATA GATE & DATASET STATUS
# ======================================================================

import os
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")
FINAL = os.path.join(ROOT, "data", "final")

os.makedirs(FINAL, exist_ok=True)

print("=" * 70)
print("CODE 6T — FINAL STRICT METADATA GATE & DATASET STATUS")
print("=" * 70)

# ----------------------------------------------------------------------
# Load authoritative release dataset
# ----------------------------------------------------------------------

FILE_6N = os.path.join(
    VAL,
    "06N_analytical_ready_observations.csv"
)

FILE_6S = os.path.join(
    VAL,
    "06S_precision_unit_summary.csv"
)

if not os.path.exists(FILE_6N):
    raise FileNotFoundError(FILE_6N)

if not os.path.exists(FILE_6S):
    raise FileNotFoundError(FILE_6S)

df = pd.read_csv(FILE_6N)
s = pd.read_csv(FILE_6S)

print("6N observations:", len(df))
print("6S unit records:", len(s))

# ----------------------------------------------------------------------
# Basic preservation check
# ----------------------------------------------------------------------

if len(df) != 60:
    raise ValueError(
        f"Expected 60 observations from 6N, found {len(df)}"
    )

if "observation_id" not in df.columns:
    raise ValueError(
        "6N does not contain observation_id"
    )

if df["observation_id"].duplicated().any():
    raise ValueError(
        "Duplicate observation_id detected in 6N"
    )

# ----------------------------------------------------------------------
# Detect release columns safely
# ----------------------------------------------------------------------

time_col = None
release_col = None

for c in [
    "release_time_h",
    "release_time_h_numeric"
]:
    if c in df.columns:
        time_col = c
        break

for c in [
    "release_percent",
    "release_percent_numeric"
]:
    if c in df.columns:
        release_col = c
        break

if time_col is None:
    raise ValueError(
        "Could not identify release-time column."
    )

if release_col is None:
    raise ValueError(
        "Could not identify release-percent column."
    )

# ----------------------------------------------------------------------
# Release validity
# ----------------------------------------------------------------------

df["final_time_h"] = pd.to_numeric(
    df[time_col],
    errors="coerce"
)

df["final_release_percent"] = pd.to_numeric(
    df[release_col],
    errors="coerce"
)

df["release_valid_6T"] = (
    df["final_time_h"].notna()
    &
    df["final_release_percent"].notna()
    &
    (df["final_time_h"] >= 0)
    &
    (df["final_release_percent"] >= 0)
    &
    (df["final_release_percent"] <= 100)
)

# ----------------------------------------------------------------------
# Strict metadata availability
# ----------------------------------------------------------------------

def is_strict_value(x):
    if pd.isna(x):
        return False

    x = str(x).strip()

    if x == "":
        return False

    if x == "NOT_IDENTIFIED":
        return False

    if x.startswith("AMBIGUOUS:"):
        return False

    return True


# Find best available metadata columns
drug_col = next(
    (
        c for c in [
            "final_drug",
            "final_drug_6M",
            "recovered_drug",
            "drug"
        ]
        if c in df.columns
    ),
    None
)

polymer_col = next(
    (
        c for c in [
            "final_polymer",
            "final_polymer_6M",
            "recovered_polymer",
            "polymer"
        ]
        if c in df.columns
    ),
    None
)

carrier_col = next(
    (
        c for c in [
            "final_carrier",
            "final_carrier_6M",
            "recovered_carrier",
            "carrier"
        ]
        if c in df.columns
    ),
    None
)

# ----------------------------------------------------------------------
# Strict metadata flags
# ----------------------------------------------------------------------

if drug_col:
    df["drug_strict_6T"] = df[drug_col].apply(
        is_strict_value
    )
else:
    df["drug_strict_6T"] = False

if polymer_col:
    df["polymer_strict_6T"] = df[polymer_col].apply(
        is_strict_value
    )
else:
    df["polymer_strict_6T"] = False

if carrier_col:
    df["carrier_strict_6T"] = df[carrier_col].apply(
        is_strict_value
    )
else:
    df["carrier_strict_6T"] = False

df["strict_metadata_fields"] = (
    df[
        [
            "drug_strict_6T",
            "polymer_strict_6T",
            "carrier_strict_6T"
        ]
    ].sum(axis=1)
)

# ----------------------------------------------------------------------
# Final conservative status
# ----------------------------------------------------------------------

df["final_6T_status"] = np.select(
    [
        ~df["release_valid_6T"],
        df["strict_metadata_fields"] >= 3,
        df["strict_metadata_fields"] >= 1
    ],
    [
        "RELEASE_INVALID",
        "STRICT_METADATA_READY",
        "METADATA_INCOMPLETE"
    ],
    default="METADATA_INCOMPLETE"
)

# ML eligibility is intentionally VERY strict.
df["ml_candidate_6T"] = (
    df["release_valid_6T"]
    &
    (df["strict_metadata_fields"] >= 3)
)

# ----------------------------------------------------------------------
# Observation type preserved
# ----------------------------------------------------------------------

if "observation_type" in df.columns:
    observation_counts = (
        df["observation_type"]
        .value_counts()
    )
else:
    observation_counts = pd.Series(
        dtype=int
    )

# ----------------------------------------------------------------------
# Save outputs
# ----------------------------------------------------------------------

AUDIT = os.path.join(
    VAL,
    "06T_final_strict_metadata_gate.csv"
)

RELEASE_VALID = os.path.join(
    FINAL,
    "06T_release_validated.csv"
)

ML_CANDIDATES = os.path.join(
    FINAL,
    "06T_strict_ml_candidates.csv"
)

MANUAL = os.path.join(
    FINAL,
    "06T_metadata_manual_review.csv"
)

df.to_csv(
    AUDIT,
    index=False
)

df[
    df["release_valid_6T"]
].to_csv(
    RELEASE_VALID,
    index=False
)

df[
    df["ml_candidate_6T"]
].to_csv(
    ML_CANDIDATES,
    index=False
)

df[
    ~df["ml_candidate_6T"]
].to_csv(
    MANUAL,
    index=False
)

# ----------------------------------------------------------------------
# Final report
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("CODE 6T COMPLETE")
print("=" * 70)

print("Input observations:", len(df))

print(
    "Release-valid:",
    int(df["release_valid_6T"].sum())
)

print(
    "Strict metadata-ready:",
    int(
        (
            df["final_6T_status"]
            == "STRICT_METADATA_READY"
        ).sum()
    )
)

print(
    "ML candidates:",
    int(df["ml_candidate_6T"].sum())
)

print(
    "Metadata incomplete/manual:",
    int(
        (
            df["final_6T_status"]
            == "METADATA_INCOMPLETE"
        ).sum()
    )
)

print(
    "Release invalid:",
    int(
        (
            df["final_6T_status"]
            == "RELEASE_INVALID"
        ).sum()
    )
)

print()
print("Metadata columns detected:")
print("Drug   :", drug_col)
print("Polymer:", polymer_col)
print("Carrier:", carrier_col)

print()
print("Observation types:")
print(observation_counts)

print()
print("Saved:")
print(AUDIT)
print(RELEASE_VALID)
print(ML_CANDIDATES)
print(MANUAL)

print()
print("IMPORTANT:")
print("6F/6N release values remain authoritative.")
print("6S strict metadata evidence is respected.")
print("No ambiguous metadata were promoted.")
print("No observations were generated.")
print("No interpolation was performed.")
print("No observations were duplicated.")
print("No machine learning was performed.")

print("=" * 70)
```

```python
# ======================================================================
# CODE 6U–6X — FINAL CONSERVATIVE VALIDATION PIPELINE
# SINGLE CELL
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")
RAW = os.path.join(ROOT, "data", "raw", "tables")
FINAL = os.path.join(ROOT, "data", "final")

os.makedirs(VAL, exist_ok=True)
os.makedirs(FINAL, exist_ok=True)

print("=" * 70)
print("CODE 6U–6X — FINAL CONSERVATIVE VALIDATION PIPELINE")
print("=" * 70)


# ======================================================================
# HELPERS
# ======================================================================

def clean(x):
    if pd.isna(x):
        return ""
    return re.sub(r"\s+", " ", str(x)).strip()


def strict_value(x):
    if pd.isna(x):
        return False
    s = clean(x)
    return (
        s != ""
        and s != "NOT_IDENTIFIED"
        and not s.startswith("AMBIGUOUS:")
    )


def first_existing(df, names):
    for n in names:
        if n in df.columns:
            return n
    return None


# ======================================================================
# LOAD AUTHORITATIVE DATA
# ======================================================================

F6T = os.path.join(
    VAL,
    "06T_final_strict_metadata_gate.csv"
)

F6S = os.path.join(
    VAL,
    "06S_precision_unit_summary.csv"
)

F3 = os.path.join(
    RAW,
    "03_extracted_table_cells.csv"
)

if not os.path.exists(F6T):
    raise FileNotFoundError(
        "Missing 6T file:\n" + F6T
    )

if not os.path.exists(F6S):
    raise FileNotFoundError(
        "Missing 6S file:\n" + F6S
    )

if not os.path.exists(F3):
    raise FileNotFoundError(
        "Missing Code 3 file:\n" + F3
    )

df = pd.read_csv(F6T)
units = pd.read_csv(F6S)
cells = pd.read_csv(F3)

print("6T observations:", len(df))
print("6S units:", len(units))
print("Code 3 cells:", len(cells))


# ======================================================================
# CODE 6U — TARGETED SAME-TABLE METADATA EVIDENCE
# ======================================================================

print()
print("=" * 70)
print("CODE 6U — TARGETED SAME-TABLE METADATA EVIDENCE")
print("=" * 70)

required = [
    "pmcid",
    "table_index",
    "row_index",
    "column_index",
    "raw_cell"
]

missing = [
    c for c in required
    if c not in cells.columns
]

if missing:
    raise ValueError(
        "Code 3 missing required columns:\n"
        + "\n".join(missing)
    )

for c in [
    "pmcid",
    "table_index",
    "row_index",
    "column_index"
]:
    cells[c] = cells[c].apply(clean)

for c in [
    "pmcid",
    "table_index",
    "row_index"
]:
    if c in df.columns:
        df[c] = df[c].apply(clean)

# Only source tables represented by the 60 validated observations
source_keys = df[
    ["pmcid", "table_index"]
].drop_duplicates()

source_cells = cells.merge(
    source_keys,
    on=["pmcid", "table_index"],
    how="inner"
)

print(
    "Validated source tables:",
    len(source_keys)
)

print(
    "Restricted Code 3 cells:",
    len(source_cells)
)


# ---- Exact-row evidence only -----------------------------------------

DRUGS = [
    "doxorubicin",
    "5-fluorouracil",
    "paclitaxel",
    "curcumin",
    "metformin",
    "ibuprofen"
]

POLYMERS = [
    "PLGA",
    "PLA",
    "PCL",
    "PEG",
    "PVA",
    "PEO",
    "chitosan",
    "alginate",
    "cellulose",
    "gelatin",
    "poloxamer",
    "polyurethane"
]

CARRIERS = [
    "nanoparticle",
    "nanoparticles",
    "nanocarrier",
    "microparticle",
    "microparticles",
    "microsphere",
    "microspheres",
    "liposome",
    "liposomes",
    "hydrogel",
    "hydrogels",
    "nanofiber",
    "nanofibers",
    "film",
    "films",
    "implant",
    "implants",
    "bead",
    "beads",
    "micelle",
    "micelles"
]


def find_terms(text, terms):
    text = clean(text)
    found = []

    for term in terms:
        if re.search(
            r"(?<![A-Za-z])" +
            re.escape(term) +
            r"(?![A-Za-z])",
            text,
            flags=re.I
        ):
            found.append(term)

    return sorted(set(found))


def find_drugs(text):
    text = clean(text)
    found = []

    patterns = {
        "doxorubicin": [
            r"\bdoxorubicin\b",
            r"\bDOX\b"
        ],
        "5-fluorouracil": [
            r"\b5[\s-]?fluorouracil\b",
            r"\b5[\s-]?FU\b"
        ],
        "paclitaxel": [
            r"\bpaclitaxel\b"
        ],
        "curcumin": [
            r"\bcurcumin\b"
        ],
        "metformin": [
            r"\bmetformin\b"
        ],
        "ibuprofen": [
            r"\bibuprofen\b"
        ]
    }

    for name, pats in patterns.items():
        for p in pats:
            if re.search(p, text, flags=re.I):
                found.append(name)
                break

    return sorted(set(found))


def resolve_field(values, prefix=""):
    vals = sorted(
        set(
            clean(v)
            for v in values
            if clean(v)
            and clean(v) != "NOT_IDENTIFIED"
        )
    )

    vals = [
        v for v in vals
        if not v.startswith("AMBIGUOUS:")
    ]

    if len(vals) == 1:
        return vals[0]

    if len(vals) > 1:
        return (
            "AMBIGUOUS:" +
            "|".join(vals)
        )

    return "NOT_IDENTIFIED"


unit_col = first_existing(
    df,
    [
        "experimental_unit_proxy",
        "experimental_unit_proxy_6P",
        "experimental_unit_id_header"
    ]
)

if unit_col is None:
    # Conservative fallback: source table + observation grouping
    df["experimental_unit_proxy_6U"] = (
        df["pmcid"].astype(str)
        + "|T"
        + df["table_index"].astype(str)
    )
    unit_col = "experimental_unit_proxy_6U"


# Create evidence from EXACT SOURCE ROWS only.
evidence_rows = []

for _, obs in df.iterrows():

    pmcid = clean(obs["pmcid"])
    table = clean(obs["table_index"])
    row_index = clean(obs["row_index"])

    sub = source_cells[
        (source_cells["pmcid"] == pmcid) &
        (source_cells["table_index"] == table) &
        (source_cells["row_index"] == row_index)
    ]

    if len(sub) == 0:

        evidence_rows.append({
            "observation_id":
                obs["observation_id"],

            "pmcid": pmcid,
            "table_index": table,
            "row_index": row_index,

            "drug_6U":
                "NOT_IDENTIFIED",

            "polymer_6U":
                "NOT_IDENTIFIED",

            "carrier_6U":
                "NOT_IDENTIFIED",

            "condition_6U":
                "NOT_IDENTIFIED",

            "evidence_status":
                "NO_EXACT_ROW_EVIDENCE"
        })

        continue

    row_text = " | ".join(
        sub["raw_cell"]
        .astype(str)
        .map(clean)
        .tolist()
    )

    header_text = ""

    if "header_text" in sub.columns:
        header_text = " | ".join(
            sub["header_text"]
            .astype(str)
            .map(clean)
            .tolist()
        )

    # Exact row + its own headers only
    text = row_text + " | " + header_text

    drugs = find_drugs(text)
    polymers = find_terms(text, POLYMERS)
    carriers = find_terms(text, CARRIERS)

    # Remove carrier terms from polymer result
    carrier_lower = {
        x.lower() for x in CARRIERS
    }

    polymers = [
        x for x in polymers
        if x.lower() not in carrier_lower
    ]

    # Explicit pH only
    ph_values = []

    for match in re.finditer(
        r"\bpH\s*(?:=|:)?\s*(\d+(?:\.\d+)?)\b",
        text,
        flags=re.I
    ):
        try:
            p = float(match.group(1))
            if 0 <= p <= 14:
                ph_values.append(f"pH {p:g}")
        except Exception:
            pass

    drug = resolve_field(drugs)
    polymer = resolve_field(polymers)
    carrier = resolve_field(carriers)
    condition = resolve_field(ph_values)

    n = sum([
        strict_value(drug),
        strict_value(polymer),
        strict_value(carrier)
    ])

    if n >= 3:
        status = "STRONG_EXACT_ROW_EVIDENCE"
    elif n >= 2:
        status = "GOOD_EXACT_ROW_EVIDENCE"
    elif n == 1:
        status = "PARTIAL_EXACT_ROW_EVIDENCE"
    else:
        status = "NO_STRICT_METADATA"

    evidence_rows.append({
        "observation_id":
            obs["observation_id"],

        "pmcid": pmcid,
        "table_index": table,
        "row_index": row_index,

        "drug_6U": drug,
        "polymer_6U": polymer,
        "carrier_6U": carrier,
        "condition_6U": condition,

        "exact_row_text_6U":
            row_text,

        "header_text_6U":
            header_text,

        "evidence_status":
            status
    })


u = pd.DataFrame(evidence_rows)

if len(u) != 60:
    raise ValueError(
        f"6U changed row count: {len(u)}"
    )

F6U = os.path.join(
    VAL,
    "06U_targeted_metadata_evidence.csv"
)

u.to_csv(
    F6U,
    index=False
)

print()
print("6U observations:", len(u))
print(
    u["evidence_status"].value_counts()
)


# ======================================================================
# CODE 6V — CONSERVATIVE METADATA CONSOLIDATION
# ======================================================================

print()
print("=" * 70)
print("CODE 6V — CONSERVATIVE METADATA CONSOLIDATION")
print("=" * 70)

# 6T remains primary.
# 6U can only add information where 6T is missing.

merged = df.copy()

u2 = u[
    [
        "observation_id",
        "drug_6U",
        "polymer_6U",
        "carrier_6U",
        "condition_6U",
        "evidence_status"
    ]
]

merged = merged.merge(
    u2,
    on="observation_id",
    how="left",
    validate="one_to_one"
)


def consolidate(primary, secondary):

    if strict_value(primary):
        return primary

    if strict_value(secondary):
        return secondary

    if (
        not strict_value(primary)
        and isinstance(secondary, str)
        and secondary.startswith("AMBIGUOUS:")
    ):
        return secondary

    return "NOT_IDENTIFIED"


merged["final_drug_6V"] = [
    consolidate(a, b)
    for a, b in zip(
        merged.get(
            "final_drug",
            pd.Series(
                ["NOT_IDENTIFIED"] * len(merged)
            )
        ),
        merged["drug_6U"]
    )
]

merged["final_polymer_6V"] = [
    consolidate(a, b)
    for a, b in zip(
        merged.get(
            "final_polymer",
            pd.Series(
                ["NOT_IDENTIFIED"] * len(merged)
            )
        ),
        merged["polymer_6U"]
    )
]

merged["final_carrier_6V"] = [
    consolidate(a, b)
    for a, b in zip(
        merged.get(
            "final_carrier",
            pd.Series(
                ["NOT_IDENTIFIED"] * len(merged)
            )
        ),
        merged["carrier_6U"]
    )
]

merged["final_condition_6V"] = [
    consolidate(a, b)
    for a, b in zip(
        merged.get(
            "final_release_condition",
            pd.Series(
                ["NOT_IDENTIFIED"] * len(merged)
            )
        ),
        merged["condition_6U"]
    )
]

merged["strict_metadata_count_6V"] = (
    merged[
        [
            "final_drug_6V",
            "final_polymer_6V",
            "final_carrier_6V"
        ]
    ]
    .applymap(strict_value)
    .sum(axis=1)
)

merged["metadata_status_6V"] = np.select(
    [
        merged["strict_metadata_count_6V"] >= 3,
        merged["strict_metadata_count_6V"] >= 2,
        merged["strict_metadata_count_6V"] == 1
    ],
    [
        "COMPLETE",
        "SUBSTANTIAL",
        "PARTIAL"
    ],
    default="INSUFFICIENT"
)

F6V = os.path.join(
    VAL,
    "06V_conservative_metadata_consolidation.csv"
)

merged.to_csv(
    F6V,
    index=False
)

print(
    merged[
        "metadata_status_6V"
    ].value_counts()
)


# ======================================================================
# CODE 6W — EXPERIMENTAL-UNIT / INDEPENDENCE AUDIT
# ======================================================================

print()
print("=" * 70)
print("CODE 6W — EXPERIMENTAL-UNIT / INDEPENDENCE AUDIT")
print("=" * 70)

if unit_col not in merged.columns:
    merged["experimental_unit_final_6W"] = (
        merged["pmcid"].astype(str)
        + "|T"
        + merged["table_index"].astype(str)
    )
    final_unit_col = "experimental_unit_final_6W"
else:
    final_unit_col = unit_col

merged["time_numeric_6W"] = pd.to_numeric(
    merged["release_time_h"],
    errors="coerce"
)

merged["release_numeric_6W"] = pd.to_numeric(
    merged["release_percent"],
    errors="coerce"
)

merged["release_valid_6W"] = (
    merged["time_numeric_6W"].notna()
    &
    merged["release_numeric_6W"].notna()
    &
    (merged["time_numeric_6W"] >= 0)
    &
    (merged["release_numeric_6W"] >= 0)
    &
    (merged["release_numeric_6W"] <= 100)
)

merged["observations_per_unit_6W"] = (
    merged.groupby(
        final_unit_col
    )["observation_id"]
    .transform("count")
)

merged["repeated_measurement_6W"] = (
    merged["observations_per_unit_6W"] > 1
)

merged["same_article_table_time_n_6W"] = (
    merged.groupby(
        [
            "pmcid",
            "table_index",
            "release_time_h"
        ]
    )["observation_id"]
    .transform("count")
)

merged["same_article_table_time_multiple_6W"] = (
    merged[
        "same_article_table_time_n_6W"
    ] > 1
)

merged["ml_candidate_6W"] = (
    merged["release_valid_6W"]
    &
    (merged["strict_metadata_count_6V"] >= 3)
    &
    (~merged["same_article_table_time_multiple_6W"])
)

F6W = os.path.join(
    VAL,
    "06W_experimental_unit_audit.csv"
)

merged.to_csv(
    F6W,
    index=False
)

print(
    "Observations:",
    len(merged)
)

print(
    "Unique units:",
    merged[final_unit_col].nunique()
)

print(
    "Repeated measurements:",
    int(
        merged[
            "repeated_measurement_6W"
        ].sum()
    )
)

print(
    "Conservative ML candidates:",
    int(
        merged[
            "ml_candidate_6W"
        ].sum()
    )
)


# ======================================================================
# CODE 6X — FINAL DATASET EXPORT & AUDIT
# ======================================================================

print()
print("=" * 70)
print("CODE 6X — FINAL DATASET EXPORT & AUDIT")
print("=" * 70)

# Preservation checks
if len(merged) != 60:
    raise ValueError(
        f"Final observation count changed: {len(merged)}"
    )

if merged["observation_id"].duplicated().any():
    raise ValueError(
        "Duplicate observation IDs detected."
    )

# Release-valid dataset
release_final = merged[
    merged["release_valid_6W"]
].copy()

# Strict ML candidate dataset
ml_final = merged[
    merged["ml_candidate_6W"]
].copy()

# Manual / metadata-limited dataset
manual_final = merged[
    ~merged["ml_candidate_6W"]
].copy()

MASTER = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_FINAL_MASTER.csv"
)

RELEASE = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_RELEASE_VALIDATED.csv"
)

ML = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_ML_CANDIDATES.csv"
)

MANUAL = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_MANUAL_METADATA_REVIEW.csv"
)

AUDIT = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_FINAL_AUDIT.csv"
)

merged.to_csv(
    MASTER,
    index=False
)

release_final.to_csv(
    RELEASE,
    index=False
)

ml_final.to_csv(
    ML,
    index=False
)

manual_final.to_csv(
    MANUAL,
    index=False
)

audit = pd.DataFrame({
    "metric": [
        "Input observations",
        "Final observations",
        "Release-valid observations",
        "Complete metadata",
        "Substantial metadata",
        "Partial metadata",
        "Insufficient metadata",
        "Conservative ML candidates",
        "Manual/review observations",
        "Unique PMCIDs",
        "Unique source tables",
        "Unique experimental units",
        "Duplicate observation IDs"
    ],

    "value": [
        len(df),
        len(merged),

        int(
            merged[
                "release_valid_6W"
            ].sum()
        ),

        int(
            (
                merged[
                    "metadata_status_6V"
                ] == "COMPLETE"
            ).sum()
        ),

        int(
            (
                merged[
                    "metadata_status_6V"
                ] == "SUBSTANTIAL"
            ).sum()
        ),

        int(
            (
                merged[
                    "metadata_status_6V"
                ] == "PARTIAL"
            ).sum()
        ),

        int(
            (
                merged[
                    "metadata_status_6V"
                ] == "INSUFFICIENT"
            ).sum()
        ),

        int(
            merged[
                "ml_candidate_6W"
            ].sum()
        ),

        int(
            (~merged[
                "ml_candidate_6W"
            ]).sum()
        ),

        merged["pmcid"].nunique(),

        merged[
            ["pmcid", "table_index"]
        ].drop_duplicates().shape[0],

        merged[
            final_unit_col
        ].nunique(),

        int(
            merged[
                "observation_id"
            ].duplicated().sum()
        )
    ]
})

audit.to_csv(
    AUDIT,
    index=False
)

print()
print("=" * 70)
print("CODE 6U–6X COMPLETE")
print("=" * 70)

print("Final observations:", len(merged))
print(
    "Release-valid:",
    len(release_final)
)
print(
    "Complete metadata:",
    int(
        (
            merged[
                "metadata_status_6V"
            ] == "COMPLETE"
        ).sum()
    )
)
print(
    "Conservative ML candidates:",
    len(ml_final)
)
print(
    "Manual/review:",
    len(manual_final)
)

print()
print("FINAL FILES")
print(MASTER)
print(RELEASE)
print(ML)
print(MANUAL)
print(AUDIT)

print()
print("IMPORTANT:")
print("Release values remain unchanged.")
print("No interpolation was performed.")
print("No observations were generated.")
print("No observations were duplicated.")
print("Ambiguous metadata were not promoted.")
print("Experimental-unit repetition is retained as repeated measurement.")
print("No machine learning was performed.")

print("=" * 70)
```

```python
# ======================================================================
# STEP 1 — FINAL METADATA RESOLUTION
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")
FINAL = os.path.join(ROOT, "data", "final")

os.makedirs(FINAL, exist_ok=True)

INPUT = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_FINAL_RELEASE_DATASET.csv"
)

if not os.path.exists(INPUT):
    raise FileNotFoundError(INPUT)

df = pd.read_csv(INPUT)

print("=" * 70)
print("STEP 1 — FINAL METADATA RESOLUTION")
print("=" * 70)

if len(df) != 60:
    raise ValueError(
        f"Expected 60 observations, found {len(df)}"
    )

# ----------------------------------------------------------------------
# Conservative metadata already established by the previous pipeline
# ----------------------------------------------------------------------

def valid(x):
    if pd.isna(x):
        return False
    s = str(x).strip()
    return (
        s != ""
        and s != "NOT_IDENTIFIED"
        and not s.startswith("AMBIGUOUS:")
    )

def choose_column(names):
    for c in names:
        if c in df.columns:
            return c
    return None

drug_col = choose_column([
    "final_drug",
    "final_drug_6M",
    "drug_6T",
    "recovered_drug"
])

polymer_col = choose_column([
    "final_polymer",
    "final_polymer_6M",
    "polymer_6T",
    "recovered_polymer"
])

carrier_col = choose_column([
    "final_carrier",
    "final_carrier_6M",
    "carrier_6T",
    "recovered_carrier"
])

condition_col = choose_column([
    "final_release_condition",
    "final_condition_6V",
    "condition_6T",
    "recovered_release_condition"
])

# ----------------------------------------------------------------------
# Preserve existing evidence; do not invent metadata
# ----------------------------------------------------------------------

df["metadata_drug"] = (
    df[drug_col]
    if drug_col
    else "NOT_IDENTIFIED"
)

df["metadata_polymer"] = (
    df[polymer_col]
    if polymer_col
    else "NOT_IDENTIFIED"
)

df["metadata_carrier"] = (
    df[carrier_col]
    if carrier_col
    else "NOT_IDENTIFIED"
)

df["metadata_condition"] = (
    df[condition_col]
    if condition_col
    else "NOT_IDENTIFIED"
)

# ----------------------------------------------------------------------
# Strict metadata count
# ----------------------------------------------------------------------

df["drug_confirmed"] = (
    df["metadata_drug"].apply(valid)
)

df["polymer_confirmed"] = (
    df["metadata_polymer"].apply(valid)
)

df["carrier_confirmed"] = (
    df["metadata_carrier"].apply(valid)
)

df["condition_confirmed"] = (
    df["metadata_condition"].apply(valid)
)

df["confirmed_metadata_count"] = (
    df[
        [
            "drug_confirmed",
            "polymer_confirmed",
            "carrier_confirmed"
        ]
    ]
    .sum(axis=1)
)

# ----------------------------------------------------------------------
# Conservative final classification
# ----------------------------------------------------------------------

df["metadata_resolution_status"] = np.select(
    [
        df["confirmed_metadata_count"] == 3,
        df["confirmed_metadata_count"] == 2,
        df["confirmed_metadata_count"] == 1
    ],
    [
        "FULL_METADATA",
        "PARTIAL_METADATA",
        "LIMITED_METADATA"
    ],
    default="MANUAL_METADATA_REVIEW"
)

# ----------------------------------------------------------------------
# Formulation identifier
# ----------------------------------------------------------------------

if "final_formulation_id" in df.columns:
    df["final_formulation_id"] = (
        df["final_formulation_id"]
        .fillna("UNRESOLVED")
    )
else:
    df["final_formulation_id"] = (
        df["pmcid"].astype(str)
        + "|T"
        + df["table_index"].astype(str)
    )

# ----------------------------------------------------------------------
# Experimental unit
# ----------------------------------------------------------------------

unit_candidates = [
    "experimental_unit_proxy_6P",
    "experimental_unit_proxy",
    "experimental_unit_id_header"
]

unit_col = None

for c in unit_candidates:
    if c in df.columns:
        unit_col = c
        break

if unit_col:
    df["final_experimental_unit"] = (
        df[unit_col]
        .fillna("UNRESOLVED")
    )
else:
    df["final_experimental_unit"] = (
        df["pmcid"].astype(str)
        + "|T"
        + df["table_index"].astype(str)
    )

# ----------------------------------------------------------------------
# Save
# ----------------------------------------------------------------------

OUT = os.path.join(
    FINAL,
    "STEP1_METADATA_RESOLVED.csv"
)

REVIEW = os.path.join(
    FINAL,
    "STEP1_METADATA_MANUAL_REVIEW.csv"
)

AUDIT = os.path.join(
    VAL,
    "STEP1_metadata_resolution_audit.csv"
)

df.to_csv(OUT, index=False)

df[
    df["metadata_resolution_status"]
    == "MANUAL_METADATA_REVIEW"
].to_csv(
    REVIEW,
    index=False
)

audit = pd.DataFrame({
    "metric": [
        "Input observations",
        "Full metadata",
        "Partial metadata",
        "Limited metadata",
        "Manual metadata review",
        "Unique formulations",
        "Unique experimental units"
    ],
    "value": [
        len(df),
        int(
            (
                df["metadata_resolution_status"]
                == "FULL_METADATA"
            ).sum()
        ),
        int(
            (
                df["metadata_resolution_status"]
                == "PARTIAL_METADATA"
            ).sum()
        ),
        int(
            (
                df["metadata_resolution_status"]
                == "LIMITED_METADATA"
            ).sum()
        ),
        int(
            (
                df["metadata_resolution_status"]
                == "MANUAL_METADATA_REVIEW"
            ).sum()
        ),
        df["final_formulation_id"].nunique(),
        df["final_experimental_unit"].nunique()
    ]
})

audit.to_csv(
    AUDIT,
    index=False
)

print()
print(
    df[
        "metadata_resolution_status"
    ].value_counts()
)

print()
print("Output:", OUT)
print("Manual review:", REVIEW)
print("Audit:", AUDIT)

print()
print("STEP 1 COMPLETE")
print("No release values changed.")
print("No observations generated.")
print("No interpolation.")
print("No duplication.")
print("No unsupported metadata promoted.")

# ======================================================================
```

```python
# ======================================================================
# STEP 2 — FINAL ML DATASET PREPARATION & INDEPENDENCE GATE
# ======================================================================

import os
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"
VAL = os.path.join(ROOT, "data", "validation")
FINAL = os.path.join(ROOT, "data", "final")

INPUT = os.path.join(
    FINAL,
    "STEP1_METADATA_RESOLVED.csv"
)

if not os.path.exists(INPUT):
    raise FileNotFoundError(INPUT)

df = pd.read_csv(INPUT)

print("=" * 70)
print("STEP 2 — FINAL ML DATASET PREPARATION")
print("=" * 70)

if len(df) != 60:
    raise ValueError(
        f"Expected 60 observations, found {len(df)}"
    )

# ----------------------------------------------------------------------
# Release validity
# ----------------------------------------------------------------------

df["time_h_ml"] = pd.to_numeric(
    df["release_time_h"],
    errors="coerce"
)

df["release_percent_ml"] = pd.to_numeric(
    df["release_percent"],
    errors="coerce"
)

df["release_valid_ml"] = (
    df["time_h_ml"].notna()
    &
    df["release_percent_ml"].notna()
    &
    (df["time_h_ml"] >= 0)
    &
    (df["release_percent_ml"] >= 0)
    &
    (df["release_percent_ml"] <= 100)
)

# ----------------------------------------------------------------------
# Strict metadata requirement
# ----------------------------------------------------------------------

df["metadata_complete_ml"] = (
    df["metadata_resolution_status"]
    == "FULL_METADATA"
)

# ----------------------------------------------------------------------
# Experimental-unit structure
# ----------------------------------------------------------------------

unit_col = "final_experimental_unit"

df["unit_n"] = (
    df.groupby(
        unit_col
    )["observation_id"]
    .transform("count")
)

df["repeated_measurement"] = (
    df["unit_n"] > 1
)

# ----------------------------------------------------------------------
# Same article/table/time duplication audit
# ----------------------------------------------------------------------

df["article_table_time_n"] = (
    df.groupby(
        [
            "pmcid",
            "table_index",
            "time_h_ml"
        ]
    )["observation_id"]
    .transform("count")
)

df["same_time_multiple"] = (
    df["article_table_time_n"] > 1
)

# ----------------------------------------------------------------------
# IMPORTANT:
# Do NOT pretend repeated time points are independent experiments.
#
# ML candidates require:
#   1. valid release
#   2. complete metadata
#
# Repeated measurements are retained but flagged.
# ----------------------------------------------------------------------

df["ml_ready"] = (
    df["release_valid_ml"]
    &
    df["metadata_complete_ml"]
)

df["ml_exclusion_reason"] = np.select(
    [
        ~df["release_valid_ml"],
        ~df["metadata_complete_ml"],
        df["same_time_multiple"]
    ],
    [
        "INVALID_RELEASE",
        "INCOMPLETE_METADATA",
        "SAME_ARTICLE_TABLE_TIME_GROUP"
    ],
    default="ML_READY"
)

# ----------------------------------------------------------------------
# Do not delete repeated time points from the master data.
# ----------------------------------------------------------------------

ML = df[
    df["ml_ready"]
].copy()

REVIEW = df[
    ~df["ml_ready"]
].copy()

MASTER = os.path.join(
    FINAL,
    "STEP2_FINAL_MASTER_DATASET.csv"
)

ML_OUT = os.path.join(
    FINAL,
    "STEP2_ML_READY_CANDIDATES.csv"
)

REVIEW_OUT = os.path.join(
    FINAL,
    "STEP2_FINAL_REVIEW_DATASET.csv"
)

AUDIT = os.path.join(
    VAL,
    "STEP2_final_ml_readiness_audit.csv"
)

df.to_csv(
    MASTER,
    index=False
)

ML.to_csv(
    ML_OUT,
    index=False
)

REVIEW.to_csv(
    REVIEW_OUT,
    index=False
)

# ----------------------------------------------------------------------
# Audit
# ----------------------------------------------------------------------

audit = pd.DataFrame({
    "metric": [
        "Total observations",
        "Release-valid",
        "Complete metadata",
        "ML-ready candidates",
        "Review-required",
        "Repeated measurements",
        "Same article/table/time groups",
        "Unique experimental units",
        "Unique PMCIDs",
        "Unique source tables"
    ],
    "value": [
        len(df),
        int(df["release_valid_ml"].sum()),
        int(df["metadata_complete_ml"].sum()),
        len(ML),
        len(REVIEW),
        int(df["repeated_measurement"].sum()),
        int(df["same_time_multiple"].sum()),
        df[unit_col].nunique(),
        df["pmcid"].nunique(),
        df[
            ["pmcid", "table_index"]
        ].drop_duplicates().shape[0]
    ]
})

audit.to_csv(
    AUDIT,
    index=False
)

print()
print("FINAL ML STATUS")
print("-" * 50)
print("Total observations:", len(df))
print(
    "Release-valid:",
    int(df["release_valid_ml"].sum())
)
print(
    "Complete metadata:",
    int(df["metadata_complete_ml"].sum())
)
print(
    "ML-ready candidates:",
    len(ML)
)
print(
    "Review-required:",
    len(REVIEW)
)
print(
    "Repeated measurements:",
    int(df["repeated_measurement"].sum())
)

print()
print("Saved:")
print(MASTER)
print(ML_OUT)
print(REVIEW_OUT)
print(AUDIT)

print()
print("STEP 2 COMPLETE")
print("No model was trained.")
print("No observations were fabricated.")
print("No interpolation was performed.")
print("Repeated time points were retained and flagged.")
print("Experimental-unit independence was not assumed.")

print("=" * 70)
```

```python
# ======================================================================
# CODE 6S-R — SOURCE-VERIFIED PRECISION UNIT SUMMARY
# ======================================================================
#
# Purpose:
#   Rebuild the 06S unit summary using the actual six validated PMC
#   source articles.
#
# IMPORTANT:
#   - Release values are NOT changed.
#   - No observations are generated.
#   - PMC12639846 Table 5 comparison rows are NOT treated as independent
#     experimental formulations.
#   - Article/source metadata are only promoted where directly supported.
#   - Release condition is kept conservative.
#   - This stage does NOT train ML.
#
# ======================================================================

import os
import re
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"

FINAL = os.path.join(ROOT, "data", "final")
VAL   = os.path.join(ROOT, "data", "validation")

os.makedirs(FINAL, exist_ok=True)
os.makedirs(VAL, exist_ok=True)

# ----------------------------------------------------------------------
# INPUT
# ----------------------------------------------------------------------

INPUT_CANDIDATES = [
    os.path.join(
        FINAL,
        "AI_Polymer_Drug_Release_FINAL_RELEASE_DATASET.csv"
    ),
    os.path.join(
        FINAL,
        "STEP1_METADATA_RESOLVED.csv"
    ),
    os.path.join(
        VAL,
        "06N_analytical_ready_observations.csv"
    ),
    os.path.join(
        VAL,
        "06F_valid_source_release.csv"
    )
]

INPUT = next(
    (x for x in INPUT_CANDIDATES if os.path.exists(x)),
    None
)

if INPUT is None:
    raise FileNotFoundError(
        "Could not find the validated release dataset."
    )

df = pd.read_csv(INPUT, low_memory=False)

print("=" * 70)
print("CODE 6S-R — SOURCE-VERIFIED PRECISION UNIT SUMMARY")
print("=" * 70)

print("Input observations:", len(df))

if len(df) != 60:
    raise ValueError(
        f"Expected 60 source-validated observations, found {len(df)}"
    )

# ----------------------------------------------------------------------
# NORMALIZE
# ----------------------------------------------------------------------

def clean(x):
    if pd.isna(x):
        return ""
    return str(x).strip()

for c in [
    "pmcid",
    "table_index",
    "row_index",
    "observation_id",
    "release_time_h",
    "release_percent"
]:
    if c not in df.columns:
        raise ValueError(f"Required column missing: {c}")

df["pmcid"] = df["pmcid"].astype(str).str.strip()
df["table_index"] = pd.to_numeric(
    df["table_index"],
    errors="coerce"
)
df["row_index"] = pd.to_numeric(
    df["row_index"],
    errors="coerce"
)

# ----------------------------------------------------------------------
# SIX VALIDATED SOURCES
# ----------------------------------------------------------------------

VALID_PMCIDS = {
    "PMC10380726",
    "PMC12584378",
    "PMC12639846",
    "PMC12804827",
    "PMC12902851",
    "PMC13089678"
}

bad_sources = sorted(
    set(df["pmcid"]) - VALID_PMCIDS
)

if bad_sources:
    raise ValueError(
        "Unexpected PMCID(s) detected:\n"
        + "\n".join(bad_sources)
    )

# ----------------------------------------------------------------------
# SOURCE-VERIFIED FORMULATION METADATA
# ----------------------------------------------------------------------
#
# These are based on direct source-article verification.
# Conditions are deliberately conservative.
# ----------------------------------------------------------------------

SOURCE_META = {

    # --------------------------------------------------------------
    # PMC12804827
    # --------------------------------------------------------------
    "PMC12804827": {
        "drug": "doxorubicin;ICG",
        "polymer": "hyaluronic acid;chitosan;PLGA",
        "carrier": "nanoparticle",
        "formulation": "HA-CS-PLGA-DOX/ICG nanoparticles",
        "source_status": "SOURCE_VERIFIED",
        "condition_note":
            "pH 7.4 and pH 5.5 release series"
    },

    # --------------------------------------------------------------
    # PMC13089678
    # --------------------------------------------------------------
    "PMC13089678": {
        "drug": "5-fluorouracil",
        "polymer": "Ni-MOF",
        "carrier": "nanocarrier",
        "formulation": "FA/Ni-MOF/5-FU nanocarrier",
        "source_status": "SOURCE_VERIFIED",
        "condition_note":
            "pH 5.0 and pH 7.4 release conditions"
    },

    # --------------------------------------------------------------
    # PMC12584378
    # --------------------------------------------------------------
    "PMC12584378": {
        "drug": "betamethasone",
        "polymer": "PVA;HPMC",
        "carrier": "film",
        "formulation": "beeswax-PVA-HPMC oral film",
        "source_status": "SOURCE_VERIFIED",
        "condition_note":
            "simulated saliva / pH 6.8"
    },

    # --------------------------------------------------------------
    # PMC12902851
    # --------------------------------------------------------------
    "PMC12902851": {
        "drug": "neomycin;clotrimazole",
        "polymer": "PVA;chitosan",
        "carrier": "electrospun nanofiber film",
        "formulation": "PVA-CS-NM-CZ electrospun wound dressing",
        "source_status": "SOURCE_VERIFIED",
        "condition_note":
            "PBS, 37 C"
    },

    # --------------------------------------------------------------
    # PMC10380726
    # --------------------------------------------------------------
    "PMC10380726": {
        "drug": "colistin",
        "polymer": "hyaluronic acid",
        "carrier": "conjugate/particle",
        "formulation":
            "cyanocobalamin-modified colistin-hyaluronan conjugate",
        "source_status": "SOURCE_VERIFIED",
        "condition_note":
            "pH 7.4 and pH 5.2"
    },

    # --------------------------------------------------------------
    # PMC12639846
    # --------------------------------------------------------------
    "PMC12639846": {
        "drug": "ketorolac tromethamine",
        "polymer": "sodium alginate;acrylamide",
        "carrier": "hydrogel beads",
        "formulation":
            "SA/Aa ketorolac hydrogel beads",
        "source_status":
            "PRESENT_STUDY_FORMULATION_ONLY",
        "condition_note":
            "release condition not assigned automatically"
    }
}

# ----------------------------------------------------------------------
# INITIALIZE OUTPUT
# ----------------------------------------------------------------------

out = df.copy()

out["source_verified_drug"] = "NOT_IDENTIFIED"
out["source_verified_polymer"] = "NOT_IDENTIFIED"
out["source_verified_carrier"] = "NOT_IDENTIFIED"
out["source_verified_formulation"] = "UNRESOLVED"
out["source_verified_condition"] = "NOT_IDENTIFIED"

out["source_metadata_status"] = "MANUAL_REVIEW"
out["source_evidence_level"] = "ARTICLE_VERIFIED"

# ----------------------------------------------------------------------
# APPLY SOURCE-ARTICLE METADATA
# ----------------------------------------------------------------------

for pmcid, meta in SOURCE_META.items():

    mask = out["pmcid"].eq(pmcid)

    out.loc[
        mask,
        "source_verified_drug"
    ] = meta["drug"]

    out.loc[
        mask,
        "source_verified_polymer"
    ] = meta["polymer"]

    out.loc[
        mask,
        "source_verified_carrier"
    ] = meta["carrier"]

    out.loc[
        mask,
        "source_verified_formulation"
    ] = meta["formulation"]

    out.loc[
        mask,
        "source_verified_condition"
    ] = meta["condition_note"]

    out.loc[
        mask,
        "source_metadata_status"
    ] = meta["source_status"]

# ----------------------------------------------------------------------
# IMPORTANT CORRECTION:
# PMC12639846 TABLE 5
#
# Rows 1–4 are comparison formulations from earlier literature.
# Row 5 is the present-study SA/Aa hydrogel bead formulation.
#
# Your five MAX_RELEASE_EVENT observations correspond to these rows.
#
# They must NOT be treated as five independent experimental units.
# ----------------------------------------------------------------------

mask_126 = out["pmcid"].eq("PMC12639846")

out.loc[
    mask_126,
    "source_metadata_status"
] = "SECONDARY_COMPARISON"

out.loc[
    mask_126,
    "source_evidence_level"
] = "SECONDARY_COMPARISON_TABLE"

out.loc[
    mask_126,
    "source_verified_formulation"
] = "COMPARISON_TABLE_FORMULATION"

# ----------------------------------------------------------------------
# MARK THE PRESENT-STUDY ROW IF IDENTIFIABLE
# ----------------------------------------------------------------------

if "row_index" in out.columns:

    # The source table has five rows.
    # Your validated observations retain row_index 0–4.
    #
    # The final row is the present study's SA/Aa hydrogel beads.
    #

    present_mask = (
        mask_126
        &
        out["row_index"].eq(
            out.loc[mask_126, "row_index"].max()
        )
    )

    out.loc[
        present_mask,
        "source_metadata_status"
    ] = "SOURCE_VERIFIED_PRESENT_STUDY"

    out.loc[
        present_mask,
        "source_evidence_level"
    ] = "EXACT_SOURCE_TABLE"

    out.loc[
        present_mask,
        "source_verified_formulation"
    ] = "SA/Aa hydrogel beads"

    out.loc[
        present_mask,
        "source_verified_drug"
    ] = "ketorolac tromethamine"

    out.loc[
        present_mask,
        "source_verified_polymer"
    ] = "sodium alginate;acrylamide"

    out.loc[
        present_mask,
        "source_verified_carrier"
    ] = "hydrogel beads"

# ----------------------------------------------------------------------
# EXPERIMENTAL UNIT
# ----------------------------------------------------------------------

def make_unit(row):

    pmcid = clean(row["pmcid"])
    table = str(int(row["table_index"])) \
        if pd.notna(row["table_index"]) \
        else "NA"

    formulation = clean(
        row["source_verified_formulation"]
    )

    condition = clean(
        row["source_verified_condition"]
    )

    # Secondary comparison observations get their own
    # non-experimental classification.
    if (
        row["source_metadata_status"]
        == "SECONDARY_COMPARISON"
    ):
        return (
            f"{pmcid}|T{table}|"
            "SECONDARY_COMPARISON"
        )

    return (
        f"{pmcid}|T{table}|"
        f"{formulation}|{condition}"
    )

out["experimental_unit_6S"] = out.apply(
    make_unit,
    axis=1
)

# ----------------------------------------------------------------------
# EXPERIMENTAL TYPE
# ----------------------------------------------------------------------

out["dataset_role_6S"] = np.where(
    out["source_metadata_status"]
    == "SECONDARY_COMPARISON",
    "EXCLUDE_FROM_ML",
    "SOURCE_DATA"
)

# ----------------------------------------------------------------------
# STRICT ML ELIGIBILITY
# ----------------------------------------------------------------------

def usable_metadata(row):

    if row["source_metadata_status"] \
            == "SECONDARY_COMPARISON":
        return False

    fields = [
        row["source_verified_drug"],
        row["source_verified_polymer"],
        row["source_verified_carrier"],
        row["source_verified_formulation"]
    ]

    return all(
        clean(x)
        and clean(x) != "NOT_IDENTIFIED"
        and not clean(x).startswith("AMBIGUOUS")
        for x in fields
    )

out["strict_metadata_ready_6S"] = out.apply(
    usable_metadata,
    axis=1
)

# ----------------------------------------------------------------------
# INDEPENDENCE AUDIT
# ----------------------------------------------------------------------

out["unit_observation_count"] = (
    out.groupby(
        "experimental_unit_6S"
    )["observation_id"]
    .transform("count")
)

out["repeated_measurement_6S"] = (
    out["unit_observation_count"] > 1
)

# ----------------------------------------------------------------------
# UNIT SUMMARY
# ----------------------------------------------------------------------

unit_summary = (
    out.groupby(
        "experimental_unit_6S",
        dropna=False
    )
    .agg(
        pmcid=("pmcid", "first"),
        table_index=("table_index", "first"),
        drug=("source_verified_drug", "first"),
        polymer=("source_verified_polymer", "first"),
        carrier=("source_verified_carrier", "first"),
        formulation=(
            "source_verified_formulation",
            "first"
        ),
        release_condition=(
            "source_verified_condition",
            "first"
        ),
        source_metadata_status=(
            "source_metadata_status",
            "first"
        ),
        source_evidence_level=(
            "source_evidence_level",
            "first"
        ),
        dataset_role=(
            "dataset_role_6S",
            "first"
        ),
        observation_count=(
            "observation_id",
            "count"
        ),
        timepoints=(
            "release_time_h",
            "nunique"
        )
    )
    .reset_index()
)

# ----------------------------------------------------------------------
# UNIT-LEVEL ML FLAG
# ----------------------------------------------------------------------

unit_summary["strict_ml_candidate_unit"] = (
    (unit_summary["dataset_role"]
     != "EXCLUDE_FROM_ML")
    &
    unit_summary["drug"].ne("NOT_IDENTIFIED")
    &
    unit_summary["polymer"].ne("NOT_IDENTIFIED")
    &
    unit_summary["carrier"].ne("NOT_IDENTIFIED")
    &
    unit_summary["formulation"].ne("UNRESOLVED")
)

# ----------------------------------------------------------------------
# SAVE
# ----------------------------------------------------------------------

OBS_FILE = os.path.join(
    VAL,
    "06S_precision_observation_audit.csv"
)

UNIT_FILE = os.path.join(
    VAL,
    "06S_precision_unit_summary.csv"
)

ML_FILE = os.path.join(
    VAL,
    "06S_strict_metadata_candidates.csv"
)

MANUAL_FILE = os.path.join(
    VAL,
    "06S_manual_metadata_review.csv"
)

out.to_csv(
    OBS_FILE,
    index=False
)

unit_summary.to_csv(
    UNIT_FILE,
    index=False
)

out[
    out["strict_metadata_ready_6S"]
].to_csv(
    ML_FILE,
    index=False
)

out[
    ~out["strict_metadata_ready_6S"]
].to_csv(
    MANUAL_FILE,
    index=False
)

# ----------------------------------------------------------------------
# SUMMARY
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("CODE 6S-R COMPLETE")
print("=" * 70)

print(
    "Total observations:",
    len(out)
)

print(
    "\nSource distribution:"
)

print(
    out["pmcid"].value_counts()
)

print(
    "\nMetadata status:"
)

print(
    out["source_metadata_status"].value_counts()
)

print(
    "\nDataset role:"
)

print(
    out["dataset_role_6S"].value_counts()
)

print(
    "\nStrict metadata-ready observations:",
    int(
        out["strict_metadata_ready_6S"].sum()
    )
)

print(
    "Secondary comparison observations:",
    int(
        (
            out["dataset_role_6S"]
            == "EXCLUDE_FROM_ML"
        ).sum()
    )
)

print(
    "Unique experimental units:",
    unit_summary[
        unit_summary["dataset_role"]
        != "EXCLUDE_FROM_ML"
    ]["experimental_unit_6S"].nunique()
)

print(
    "\nRepeated measurements:",
    int(
        out["repeated_measurement_6S"].sum()
    )
)

print("\nSaved:")
print(OBS_FILE)
print(UNIT_FILE)
print(ML_FILE)
print(MANUAL_FILE)

print()
print("IMPORTANT:")
print(
    "Release values remain unchanged."
)
print(
    "PMC12639846 comparison rows are excluded from ML."
)
print(
    "No synthetic observations were generated."
)
print(
    "No interpolation was performed."
)
print(
    "No observations were duplicated."
)
print(
    "No machine learning was performed."
)

print("=" * 70)
```

```python
# ======================================================================
# FINAL ML DATASET PREPARATION — ALL 8 STEPS IN ONE CODE
# ======================================================================
#
# 1. Exclude secondary-comparison observations
# 2. Keep genuine source observations
# 3. Validate metadata
# 4. Encode categorical variables
# 5. Define predictors and target
# 6. Create experimental-unit/source groups
# 7. Perform leakage/independence audit
# 8. Export final ML dataset + audit files
#
# NO MODEL TRAINING
# NO SYNTHETIC DATA
# NO INTERPOLATION
# NO DUPLICATION
# ======================================================================

import os
import numpy as np
import pandas as pd

ROOT = "/content/AI_Polymer_Drug_Release"

VAL = os.path.join(
    ROOT,
    "data",
    "validation"
)

FINAL = os.path.join(
    ROOT,
    "data",
    "final"
)

os.makedirs(VAL, exist_ok=True)
os.makedirs(FINAL, exist_ok=True)

INPUT = os.path.join(
    VAL,
    "06S_precision_observation_audit.csv"
)

if not os.path.exists(INPUT):
    raise FileNotFoundError(
        f"Required 06S-R file not found:\n{INPUT}"
    )

df = pd.read_csv(
    INPUT,
    low_memory=False
)

print("=" * 70)
print("FINAL ML DATASET PREPARATION — 8 STEPS")
print("=" * 70)

print("Input observations:", len(df))

if len(df) != 60:
    raise ValueError(
        f"Expected 60 observations, found {len(df)}"
    )


# ======================================================================
# STEP 1 — EXCLUDE SECONDARY-COMPARISON OBSERVATIONS
# ======================================================================

print("\n" + "=" * 70)
print("STEP 1 — SECONDARY-COMPARISON FILTER")
print("=" * 70)

if "dataset_role_6S" not in df.columns:
    raise ValueError(
        "dataset_role_6S column missing."
    )

df["is_secondary_comparison"] = (
    df["dataset_role_6S"]
    .astype(str)
    .eq("EXCLUDE_FROM_ML")
)

ml = df[
    ~df["is_secondary_comparison"]
].copy()

print(
    "Excluded secondary-comparison:",
    int(df["is_secondary_comparison"].sum())
)

print(
    "Remaining source observations:",
    len(ml)
)

if len(ml) != 56:
    raise ValueError(
        f"Expected 56 genuine source observations, found {len(ml)}"
    )


# ======================================================================
# STEP 2 — RELEASE-VALUE VALIDATION
# ======================================================================

print("\n" + "=" * 70)
print("STEP 2 — RELEASE-VALUE VALIDATION")
print("=" * 70)

required_release = [
    "observation_id",
    "pmcid",
    "table_index",
    "release_time_h",
    "release_percent"
]

missing = [
    c for c in required_release
    if c not in ml.columns
]

if missing:
    raise ValueError(
        "Missing release columns:\n"
        + "\n".join(missing)
    )

ml["release_time_h"] = pd.to_numeric(
    ml["release_time_h"],
    errors="coerce"
)

ml["release_percent"] = pd.to_numeric(
    ml["release_percent"],
    errors="coerce"
)

ml["release_valid"] = (
    ml["release_time_h"].notna()
    &
    ml["release_percent"].notna()
    &
    (ml["release_time_h"] >= 0)
    &
    (ml["release_percent"] >= 0)
    &
    (ml["release_percent"] <= 100)
)

invalid = ml[
    ~ml["release_valid"]
]

print(
    "Release-valid:",
    int(ml["release_valid"].sum())
)

print(
    "Release-invalid:",
    len(invalid)
)

if len(invalid) > 0:
    raise ValueError(
        "Invalid release observations detected."
    )


# ======================================================================
# STEP 3 — STRICT METADATA VALIDATION
# ======================================================================

print("\n" + "=" * 70)
print("STEP 3 — METADATA VALIDATION")
print("=" * 70)

metadata_columns = [
    "source_verified_drug",
    "source_verified_polymer",
    "source_verified_carrier",
    "source_verified_formulation"
]

missing_metadata = [
    c for c in metadata_columns
    if c not in ml.columns
]

if missing_metadata:
    raise ValueError(
        "Required metadata columns missing:\n"
        + "\n".join(missing_metadata)
    )


def metadata_is_valid(value):

    if pd.isna(value):
        return False

    s = str(value).strip()

    if not s:
        return False

    bad = [
        "NOT_IDENTIFIED",
        "UNRESOLVED",
        "AMBIGUOUS",
        "MANUAL_REVIEW"
    ]

    return not any(
        x in s.upper()
        for x in bad
    )


for col in metadata_columns:

    ml[
        col + "_valid"
    ] = ml[col].apply(
        metadata_is_valid
    )

ml["metadata_complete"] = (
    ml[
        [
            c + "_valid"
            for c in metadata_columns
        ]
    ]
    .all(axis=1)
)

print(
    "Complete metadata:",
    int(ml["metadata_complete"].sum())
)

print(
    "Incomplete metadata:",
    int(
        (~ml["metadata_complete"]).sum()
    )
)

# IMPORTANT:
# Only complete metadata observations are eligible for ML.
ml_ready = ml[
    ml["metadata_complete"]
].copy()

print(
    "Strict ML observations:",
    len(ml_ready)
)


# ======================================================================
# STEP 4 — CATEGORICAL FEATURE ENCODING
# ======================================================================

print("\n" + "=" * 70)
print("STEP 4 — CATEGORICAL FEATURE PREPARATION")
print("=" * 70)

# Keep original categorical fields.
# Create stable numeric codes for later ML use.
#
# Codes are identifiers, NOT ordinal biological rankings.

categorical_features = [
    "source_verified_drug",
    "source_verified_polymer",
    "source_verified_carrier",
    "source_verified_condition"
]

for col in categorical_features:

    if col in ml_ready.columns:

        values = (
            ml_ready[col]
            .fillna("NOT_REPORTED")
            .astype(str)
        )

        categories = sorted(
            values.unique()
        )

        mapping = {
            value: i
            for i, value
            in enumerate(categories)
        }

        ml_ready[
            col + "_code"
        ] = values.map(mapping)

        # Save mapping for reproducibility.
        map_file = os.path.join(
            VAL,
            f"ML_encoding_{col}.csv"
        )

        pd.DataFrame({
            "category": categories,
            "code": range(len(categories))
        }).to_csv(
            map_file,
            index=False
        )

        print(
            col,
            "categories:",
            len(categories)
        )


# ======================================================================
# STEP 5 — DEFINE FEATURES AND TARGET
# ======================================================================

print("\n" + "=" * 70)
print("STEP 5 — FEATURE / TARGET DEFINITION")
print("=" * 70)

# Release time is the primary numerical predictor.
#
# Drug/polymer/carrier/condition are categorical formulation descriptors.
#
# Release percent is the target.

feature_columns = [
    "release_time_h",
    "source_verified_drug_code",
    "source_verified_polymer_code",
    "source_verified_carrier_code"
]

condition_available = (
    "source_verified_condition_code"
    in ml_ready.columns
)

if condition_available:
    feature_columns.append(
        "source_verified_condition_code"
    )

target_column = "release_percent"

missing_features = [
    c for c in feature_columns + [target_column]
    if c not in ml_ready.columns
]

if missing_features:
    raise ValueError(
        "Missing ML columns:\n"
        + "\n".join(missing_features)
    )

print(
    "Features:"
)

for f in feature_columns:
    print(" -", f)

print(
    "Target:",
    target_column
)


# ======================================================================
# STEP 6 — EXPERIMENTAL-UNIT / SOURCE GROUPING
# ======================================================================

print("\n" + "=" * 70)
print("STEP 6 — EXPERIMENTAL-UNIT GROUPING")
print("=" * 70)

if "experimental_unit_6S" in ml_ready.columns:

    ml_ready["experimental_unit_id"] = (
        ml_ready["experimental_unit_6S"]
        .fillna("UNRESOLVED_UNIT")
        .astype(str)
    )

else:

    # Conservative fallback.
    ml_ready["experimental_unit_id"] = (
        ml_ready["pmcid"].astype(str)
        + "|T"
        + ml_ready["table_index"].astype(str)
    )

# Source-group identifier.
ml_ready["source_group_id"] = (
    ml_ready["pmcid"].astype(str)
    + "|T"
    + ml_ready["table_index"].astype(str)
)

ml_ready["unit_observation_count"] = (
    ml_ready
    .groupby(
        "experimental_unit_id"
    )["observation_id"]
    .transform("count")
)

ml_ready["repeated_measurement"] = (
    ml_ready["unit_observation_count"] > 1
)

print(
    "Unique experimental units:",
    ml_ready[
        "experimental_unit_id"
    ].nunique()
)

print(
    "Unique source groups:",
    ml_ready[
        "source_group_id"
    ].nunique()
)

print(
    "Repeated measurements:",
    int(
        ml_ready[
            "repeated_measurement"
        ].sum()
    )
)


# ======================================================================
# STEP 7 — LEAKAGE / INDEPENDENCE AUDIT
# ======================================================================

print("\n" + "=" * 70)
print("STEP 7 — LEAKAGE / INDEPENDENCE AUDIT")
print("=" * 70)

# Same-unit observations must remain together during train/test splitting.

ml_ready["article_table_time_n"] = (
    ml_ready
    .groupby(
        [
            "pmcid",
            "table_index",
            "release_time_h"
        ]
    )["observation_id"]
    .transform("count")
)

ml_ready["same_article_table_time"] = (
    ml_ready["article_table_time_n"] > 1
)

# Duplicate observation IDs.
duplicate_ids = (
    ml_ready["observation_id"]
    .duplicated()
)

# Exact value duplicates.
duplicate_values = (
    ml_ready
    .duplicated(
        subset=[
            "pmcid",
            "table_index",
            "release_time_h",
            "release_percent"
        ],
        keep=False
    )
)

print(
    "Duplicate observation IDs:",
    int(duplicate_ids.sum())
)

print(
    "Exact value duplicates:",
    int(duplicate_values.sum())
)

print(
    "Same article/table/time groups:",
    int(
        ml_ready[
            "same_article_table_time"
        ].sum()
    )
)

if duplicate_ids.any():
    raise ValueError(
        "Duplicate observation IDs detected."
    )

# Do NOT delete repeated measurements.
# They are valid observations but must be grouped during validation.


# ======================================================================
# STEP 8 — FINAL ML DATASET EXPORT
# ======================================================================

print("\n" + "=" * 70)
print("STEP 8 — FINAL ML DATASET EXPORT")
print("=" * 70)

# Final clean dataset.
final_columns = [
    "observation_id",
    "pmcid",
    "table_index",
    "row_index",
    "experimental_unit_id",
    "source_group_id",
    "source_verified_drug",
    "source_verified_polymer",
    "source_verified_carrier",
    "source_verified_formulation",
    "source_verified_condition",
    "release_time_h",
    "release_percent",
    "observation_type",
    "repeated_measurement"
]

# Keep only columns that actually exist.
final_columns = [
    c for c in final_columns
    if c in ml_ready.columns
]

final_ml = ml_ready[
    final_columns
].copy()

# Separate numeric ML matrix.
X = ml_ready[
    feature_columns
].copy()

y = ml_ready[
    target_column
].copy()

# Group vector for leakage-safe CV later.
groups = ml_ready[
    "experimental_unit_id"
].copy()

# ----------------------------------------------------------------------
# Save final datasets
# ----------------------------------------------------------------------

ML_DATASET = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_FINAL_ML_DATASET.csv"
)

FEATURES = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_ML_FEATURES.csv"
)

TARGET = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_ML_TARGET.csv"
)

GROUPS = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_ML_GROUPS.csv"
)

EXCLUDED = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_ML_EXCLUDED_OBSERVATIONS.csv"
)

AUDIT = os.path.join(
    VAL,
    "FINAL_ML_DATASET_AUDIT.csv"
)

final_ml.to_csv(
    ML_DATASET,
    index=False
)

X.to_csv(
    FEATURES,
    index=False
)

y.to_csv(
    TARGET,
    index=False
)

pd.DataFrame({
    "observation_id":
        ml_ready["observation_id"],
    "experimental_unit_id":
        groups
}).to_csv(
    GROUPS,
    index=False
)

# Save excluded observations for complete audit trail.
excluded = ml[
    ~ml["metadata_complete"]
].copy()

excluded.to_csv(
    EXCLUDED,
    index=False
)

# ----------------------------------------------------------------------
# FINAL AUDIT
# ----------------------------------------------------------------------

audit = pd.DataFrame({
    "metric": [
        "Original observations",
        "Secondary comparison excluded",
        "Genuine source observations",
        "Release-valid observations",
        "Complete metadata",
        "Final ML observations",
        "Excluded for incomplete metadata",
        "Unique PMCIDs",
        "Unique source tables",
        "Unique experimental units",
        "Repeated measurements",
        "Duplicate observation IDs",
        "Exact value duplicates"
    ],

    "value": [
        len(df),

        int(
            df[
                "is_secondary_comparison"
            ].sum()
        ),

        len(ml),

        int(
            ml["release_valid"].sum()
        ),

        int(
            ml["metadata_complete"].sum()
        ),

        len(final_ml),

        len(excluded),

        ml["pmcid"].nunique(),

        ml[
            [
                "pmcid",
                "table_index"
            ]
        ]
        .drop_duplicates()
        .shape[0],

        ml_ready[
            "experimental_unit_id"
        ].nunique(),

        int(
            ml_ready[
                "repeated_measurement"
            ].sum()
        ),

        int(
            duplicate_ids.sum()
        ),

        int(
            duplicate_values.sum()
        )
    ]
})

audit.to_csv(
    AUDIT,
    index=False
)

# ----------------------------------------------------------------------
# FINAL REPORT
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("FINAL ML DATASET PREPARATION COMPLETE")
print("=" * 70)

print(
    "Original observations:",
    len(df)
)

print(
    "Secondary comparison excluded:",
    int(
        df[
            "is_secondary_comparison"
        ].sum()
    )
)

print(
    "Genuine source observations:",
    len(ml)
)

print(
    "Release-valid:",
    int(
        ml["release_valid"].sum()
    )
)

print(
    "Complete metadata:",
    int(
        ml["metadata_complete"].sum()
    )
)

print(
    "FINAL ML observations:",
    len(final_ml)
)

print(
    "Excluded for incomplete metadata:",
    len(excluded)
)

print(
    "Unique experimental units:",
    ml_ready[
        "experimental_unit_id"
    ].nunique()
)

print(
    "Repeated measurements:",
    int(
        ml_ready[
            "repeated_measurement"
        ].sum()
    )
)

print()
print("FINAL FILES")
print("-" * 70)
print(ML_DATASET)
print(FEATURES)
print(TARGET)
print(GROUPS)
print(EXCLUDED)
print(AUDIT)

print()
print("IMPORTANT")
print("-" * 70)
print(
    "No model was trained."
)
print(
    "Repeated measurements remain retained."
)
print(
    "Experimental units must be kept together during train/test splitting."
)
print(
    "No synthetic observations were generated."
)
print(
    "No interpolation was performed."
)
print(
    "No observations were duplicated."
)
print(
    "No secondary-comparison observations were used for ML."
)

print("=" * 70)
```

```python
# ======================================================================
# FINAL DUPLICATE & INDEPENDENCE AUDIT
# ======================================================================

import os
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"
FINAL = os.path.join(ROOT, "data", "final")
VAL = os.path.join(ROOT, "data", "validation")

INPUT = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_FINAL_ML_DATASET.csv"
)

if not os.path.exists(INPUT):
    raise FileNotFoundError(INPUT)

df = pd.read_csv(INPUT)

print("=" * 70)
print("FINAL DUPLICATE & INDEPENDENCE AUDIT")
print("=" * 70)
print("Input observations:", len(df))

# ----------------------------------------------------------------------
# 1. Observation-ID duplicates
# ----------------------------------------------------------------------

id_dups = df[
    df["observation_id"].duplicated(keep=False)
].copy()

# ----------------------------------------------------------------------
# 2. Exact value duplicates
# ----------------------------------------------------------------------

value_cols = [
    "pmcid",
    "table_index",
    "release_time_h",
    "release_percent"
]

missing = [
    c for c in value_cols
    if c not in df.columns
]

if missing:
    raise ValueError(
        "Missing duplicate-audit columns:\n"
        + "\n".join(missing)
    )

df["exact_value_duplicate"] = (
    df.duplicated(
        subset=value_cols,
        keep=False
    )
)

value_dups = df[
    df["exact_value_duplicate"]
].copy()

# ----------------------------------------------------------------------
# 3. Check whether exact duplicates have different metadata
# ----------------------------------------------------------------------

metadata_cols = [
    "source_verified_drug",
    "source_verified_polymer",
    "source_verified_carrier",
    "source_verified_condition",
    "source_verified_formulation"
]

metadata_cols = [
    c for c in metadata_cols
    if c in df.columns
]

if metadata_cols:

    duplicate_groups = []

    for key, g in value_dups.groupby(
        value_cols,
        dropna=False
    ):

        metadata_variation = {
            c: g[c].nunique(dropna=False)
            for c in metadata_cols
        }

        duplicate_groups.append({
            **dict(
                zip(value_cols, key)
            ),
            "n_rows": len(g),
            "metadata_variation": str(
                metadata_variation
            ),
            "different_metadata": any(
                v > 1
                for v in metadata_variation.values()
            )
        })

    duplicate_group_df = pd.DataFrame(
        duplicate_groups
    )

else:
    duplicate_group_df = pd.DataFrame()

# ----------------------------------------------------------------------
# 4. Same article/table/time audit
# ----------------------------------------------------------------------

group_cols = [
    "pmcid",
    "table_index",
    "release_time_h"
]

df["same_article_table_time_n"] = (
    df.groupby(
        group_cols,
        dropna=False
    )["observation_id"]
    .transform("count")
)

df["same_article_table_time_multiple"] = (
    df["same_article_table_time_n"] > 1
)

same_time = df[
    df["same_article_table_time_multiple"]
].copy()

# ----------------------------------------------------------------------
# 5. Experimental-unit audit
# ----------------------------------------------------------------------

if "experimental_unit_id" not in df.columns:
    raise ValueError(
        "experimental_unit_id is missing."
    )

df["unit_n"] = (
    df.groupby(
        "experimental_unit_id"
    )["observation_id"]
    .transform("count")
)

df["repeated_measurement"] = (
    df["unit_n"] > 1
)

unit_summary = (
    df.groupby(
        "experimental_unit_id",
        dropna=False
    )
    .agg(
        observations=("observation_id", "count"),
        pmcids=("pmcid", "nunique"),
        tables=("table_index", "nunique"),
        timepoints=("release_time_h", "nunique")
    )
    .reset_index()
)

# ----------------------------------------------------------------------
# 6. Conservative duplicate classification
# ----------------------------------------------------------------------

df["duplicate_decision"] = "NOT_DUPLICATE"

if len(value_dups) > 0:

    for key, g in value_dups.groupby(
        value_cols,
        dropna=False
    ):

        idx = g.index

        # Different metadata = potentially legitimate observations.
        if metadata_cols:

            different_metadata = any(
                g[c].nunique(dropna=False) > 1
                for c in metadata_cols
            )

        else:
            different_metadata = False

        if different_metadata:

            df.loc[
                idx,
                "duplicate_decision"
            ] = "POTENTIALLY_LEGITIMATE_DUPLICATE_VALUE"

        else:

            df.loc[
                idx,
                "duplicate_decision"
            ] = "REVIEW_TRUE_DUPLICATE"

# ----------------------------------------------------------------------
# 7. NEVER automatically delete duplicates
# ----------------------------------------------------------------------

true_duplicate_review = df[
    df["duplicate_decision"]
    == "REVIEW_TRUE_DUPLICATE"
].copy()

legitimate_value_duplicates = df[
    df["duplicate_decision"]
    == "POTENTIALLY_LEGITIMATE_DUPLICATE_VALUE"
].copy()

# ----------------------------------------------------------------------
# 8. ML eligibility after audit
# ----------------------------------------------------------------------

# Conservative rule:
# No observation is removed automatically.
#
# If true duplicate review exists, ML should wait for manual confirmation.

if len(true_duplicate_review) == 0:

    df["final_ml_audit_status"] = (
        "READY_FOR_GROUPED_MODEL_VALIDATION"
    )

else:

    df["final_ml_audit_status"] = (
        "DUPLICATE_REVIEW_REQUIRED"
    )

# ----------------------------------------------------------------------
# 9. Save audit files
# ----------------------------------------------------------------------

AUDIT = os.path.join(
    VAL,
    "FINAL_DUPLICATE_INDEPENDENCE_AUDIT.csv"
)

DUPLICATES = os.path.join(
    VAL,
    "FINAL_EXACT_VALUE_DUPLICATES.csv"
)

REVIEW = os.path.join(
    VAL,
    "FINAL_TRUE_DUPLICATE_REVIEW.csv"
)

UNITS = os.path.join(
    VAL,
    "FINAL_EXPERIMENTAL_UNIT_SUMMARY.csv"
)

df.to_csv(
    AUDIT,
    index=False
)

value_dups.to_csv(
    DUPLICATES,
    index=False
)

true_duplicate_review.to_csv(
    REVIEW,
    index=False
)

unit_summary.to_csv(
    UNITS,
    index=False
)

# ----------------------------------------------------------------------
# 10. FINAL REPORT
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("FINAL AUDIT RESULTS")
print("=" * 70)

print(
    "Observations:",
    len(df)
)

print(
    "Duplicate observation IDs:",
    len(id_dups)
)

print(
    "Exact value-duplicate observations:",
    len(value_dups)
)

print(
    "Potentially legitimate value duplicates:",
    len(legitimate_value_duplicates)
)

print(
    "True-duplicate review observations:",
    len(true_duplicate_review)
)

print(
    "Same article/table/time observations:",
    int(
        df[
            "same_article_table_time_multiple"
        ].sum()
    )
)

print(
    "Unique experimental units:",
    df[
        "experimental_unit_id"
    ].nunique()
)

print(
    "Repeated measurements:",
    int(
        df[
            "repeated_measurement"
        ].sum()
    )
)

print()
print(
    "FINAL STATUS:",
    df[
        "final_ml_audit_status"
    ].iloc[0]
)

print()
print("Saved:")
print(AUDIT)
print(DUPLICATES)
print(REVIEW)
print(UNITS)

print()
print("IMPORTANT:")
print("No observations were automatically deleted.")
print("Repeated time points remain repeated measurements.")
print("Experimental units must remain grouped during validation.")
print("Exact value duplicates require metadata/source interpretation.")
print("No machine learning was performed.")

print("=" * 70)
```

```python
# ======================================================================
# FINAL DUPLICATE RESOLUTION — SOURCE-LEVEL CHECK
# ======================================================================

import os
import pandas as pd
import numpy as np

ROOT = "/content/AI_Polymer_Drug_Release"
FINAL = os.path.join(ROOT, "data", "final")
VAL = os.path.join(ROOT, "data", "validation")

ML_FILE = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_FINAL_ML_DATASET.csv"
)

CODE3 = os.path.join(
    ROOT,
    "data",
    "raw",
    "tables",
    "03_extracted_table_cells.csv"
)

if not os.path.exists(ML_FILE):
    raise FileNotFoundError(ML_FILE)

if not os.path.exists(CODE3):
    raise FileNotFoundError(CODE3)

df = pd.read_csv(ML_FILE, low_memory=False)
cells = pd.read_csv(CODE3, low_memory=False)

print("=" * 70)
print("FINAL DUPLICATE RESOLUTION — SOURCE-LEVEL CHECK")
print("=" * 70)

print("ML observations:", len(df))
print("Code 3 cells:", len(cells))

# ----------------------------------------------------------------------
# EXACT DUPLICATES
# ----------------------------------------------------------------------

dup_cols = [
    "pmcid",
    "table_index",
    "release_time_h",
    "release_percent"
]

dups = df[
    df.duplicated(
        subset=dup_cols,
        keep=False
    )
].copy()

print(
    "Exact duplicate observations:",
    len(dups)
)

# ----------------------------------------------------------------------
# SOURCE TABLE RECONSTRUCTION
# ----------------------------------------------------------------------

cells["pmcid"] = cells["pmcid"].astype(str)
cells["table_index"] = pd.to_numeric(
    cells["table_index"],
    errors="coerce"
)
cells["row_index"] = pd.to_numeric(
    cells["row_index"],
    errors="coerce"
)

# Keep only tables represented by duplicate observations.
keys = dups[
    ["pmcid", "table_index"]
].drop_duplicates()

source = cells.merge(
    keys,
    on=["pmcid", "table_index"],
    how="inner"
)

print(
    "Relevant Code 3 cells:",
    len(source)
)

# ----------------------------------------------------------------------
# BUILD SOURCE ROW TEXT
# ----------------------------------------------------------------------

row_text = (
    source.groupby(
        [
            "pmcid",
            "table_index",
            "row_index"
        ],
        dropna=False
    )
    .agg(
        row_text=("row_text", "first"),
        full_table_text=("full_table_text", "first"),
        header_text=("header_text", "first")
    )
    .reset_index()
)

# ----------------------------------------------------------------------
# MERGE DUPLICATE OBSERVATIONS WITH SOURCE ROWS
# ----------------------------------------------------------------------

audit = dups.merge(
    row_text,
    on=[
        "pmcid",
        "table_index"
    ],
    how="left"
)

# ----------------------------------------------------------------------
# SEARCH SOURCE TABLE FOR FORMULATION TERMS
# ----------------------------------------------------------------------

FORMULATION_TERMS = [
    "polymer",
    "chitosan",
    "alginate",
    "cellulose",
    "PVA",
    "HPMC",
    "PLGA",
    "hyaluronic",
    "nanoparticle",
    "nanoparticles",
    "microparticle",
    "microparticles",
    "hydrogel",
    "bead",
    "film",
    "MOF",
    "drug",
    "DOX",
    "ICG",
    "5-FU",
    "fluorouracil",
    "betamethasone",
    "neomycin",
    "clotrimazole",
    "colistin"
]

def source_terms(row):

    text = " ".join([
        str(row.get("row_text", "")),
        str(row.get("full_table_text", "")),
        str(row.get("header_text", ""))
    ]).lower()

    found = []

    for term in FORMULATION_TERMS:

        if term.lower() in text:
            found.append(term)

    return ";".join(
        sorted(set(found))
    )

audit["source_formulation_terms"] = audit.apply(
    source_terms,
    axis=1
)

# ----------------------------------------------------------------------
# DETERMINE WHETHER DUPLICATE VALUES HAVE DIFFERENT FORMULATION CONTEXT
# ----------------------------------------------------------------------

metadata_cols = [
    "source_verified_drug",
    "source_verified_polymer",
    "source_verified_carrier",
    "source_verified_formulation",
    "source_verified_condition"
]

metadata_cols = [
    c for c in metadata_cols
    if c in audit.columns
]

audit["metadata_context"] = (
    audit[metadata_cols]
    .astype(str)
    .agg(
        " | ".join,
        axis=1
    )
)

context_counts = (
    audit.groupby(
        dup_cols,
        dropna=False
    )["metadata_context"]
    .nunique()
    .reset_index(
        name="metadata_context_count"
    )
)

audit = audit.merge(
    context_counts,
    on=dup_cols,
    how="left"
)

# ----------------------------------------------------------------------
# CLASSIFICATION
# ----------------------------------------------------------------------

audit["duplicate_resolution"] = np.where(
    audit["metadata_context_count"] > 1,
    "LEGITIMATE_EQUAL_VALUE_DIFFERENT_CONTEXT",
    "TRUE_DUPLICATE_EXTRACTION"
)

# ----------------------------------------------------------------------
# IMPORTANT:
# Do NOT delete until classification is complete.
# ----------------------------------------------------------------------

true_dups = audit[
    audit["duplicate_resolution"]
    == "TRUE_DUPLICATE_EXTRACTION"
].copy()

legit_equal = audit[
    audit["duplicate_resolution"]
    == "LEGITIMATE_EQUAL_VALUE_DIFFERENT_CONTEXT"
].copy()

# ----------------------------------------------------------------------
# SAVE
# ----------------------------------------------------------------------

AUDIT = os.path.join(
    VAL,
    "FINAL_DUPLICATE_SOURCE_RESOLUTION.csv"
)

TRUE_DUP = os.path.join(
    VAL,
    "FINAL_TRUE_DUPLICATES_CONFIRMED.csv"
)

LEGIT = os.path.join(
    VAL,
    "FINAL_EQUAL_VALUES_DIFFERENT_CONTEXT.csv"
)

audit.to_csv(
    AUDIT,
    index=False
)

true_dups.to_csv(
    TRUE_DUP,
    index=False
)

legit_equal.to_csv(
    LEGIT,
    index=False
)

# ----------------------------------------------------------------------
# REPORT
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("DUPLICATE RESOLUTION COMPLETE")
print("=" * 70)

print(
    "Exact duplicate observations:",
    len(dups)
)

print(
    "True duplicate extractions:",
    len(true_dups)
)

print(
    "Equal values with different context:",
    len(legit_equal)
)

print()

if len(true_dups) > 0:
    print(
        "STATUS: TRUE_DUPLICATE_REVIEW_REQUIRED"
    )
else:
    print(
        "STATUS: NO_CONFIRMED_TRUE_DUPLICATES"
    )

print()
print("Saved:")
print(AUDIT)
print(TRUE_DUP)
print(LEGIT)

print()
print("No observations were deleted.")
print("No release values were modified.")
print("No interpolation.")
print("No synthetic observations.")
print("No machine learning.")

print("=" * 70)
```

```python
# ======================================================================
# FINAL PROJECT PIPELINE
# DUPLICATE RESOLUTION → FINAL DATASET → GROUPED CV → ML → RESULTS
# ======================================================================
#
# IMPORTANT:
# - Corrects the previous source-merge duplication bug.
# - 6 exact duplicate observations are handled at observation level.
# - We NEVER remove 44 rows.
# - Repeated time points remain repeated measurements.
# - Experimental units are kept together during validation.
# - Leave-One-Experimental-Unit-Out CV is used.
# - No synthetic data.
# - No interpolation.
# ======================================================================

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

from sklearn.model_selection import LeaveOneGroupOut
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ======================================================================
# PATHS
# ======================================================================

ROOT = "/content/AI_Polymer_Drug_Release"

FINAL = os.path.join(
    ROOT,
    "data",
    "final"
)

VAL = os.path.join(
    ROOT,
    "data",
    "validation"
)

RESULTS = os.path.join(
    ROOT,
    "data",
    "results"
)

os.makedirs(FINAL, exist_ok=True)
os.makedirs(VAL, exist_ok=True)
os.makedirs(RESULTS, exist_ok=True)

INPUT = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_FINAL_ML_DATASET.csv"
)

if not os.path.exists(INPUT):
    raise FileNotFoundError(
        f"Final ML dataset not found:\n{INPUT}"
    )

df = pd.read_csv(
    INPUT,
    low_memory=False
)

print("=" * 75)
print("FINAL PROJECT PIPELINE")
print("=" * 75)

print("Initial ML observations:", len(df))


# ======================================================================
# PART 1 — CORRECT DUPLICATE RESOLUTION
# ======================================================================

print("\n" + "=" * 75)
print("PART 1 — CORRECT DUPLICATE RESOLUTION")
print("=" * 75)

duplicate_key = [
    "pmcid",
    "table_index",
    "release_time_h",
    "release_percent"
]

missing = [
    c for c in duplicate_key
    if c not in df.columns
]

if missing:
    raise ValueError(
        "Missing duplicate-key columns:\n"
        + "\n".join(missing)
    )

df["exact_duplicate_group"] = (
    df.groupby(
        duplicate_key,
        dropna=False
    )["observation_id"]
    .transform("count")
)

df["exact_duplicate"] = (
    df["exact_duplicate_group"] > 1
)

duplicate_rows = df[
    df["exact_duplicate"]
].copy()

print(
    "Exact duplicate rows:",
    len(duplicate_rows)
)

print(
    "Duplicate groups:",
    duplicate_rows[
        duplicate_key
    ].drop_duplicates().shape[0]
)

# ----------------------------------------------------------------------
# IMPORTANT:
#
# If there are 6 duplicate rows, we remove only the extra copy within
# each exact duplicate group.
#
# Example:
#   A
#   A
#
# becomes:
#   A
#
# NOT:
#   delete all observations belonging to the source.
# ----------------------------------------------------------------------

df["duplicate_rank"] = (
    df.groupby(
        duplicate_key,
        dropna=False
    ).cumcount()
)

df["duplicate_keep"] = (
    df["duplicate_rank"] == 0
)

removed_duplicates = df[
    ~df["duplicate_keep"]
].copy()

clean = df[
    df["duplicate_keep"]
].copy()

print(
    "Rows removed as exact duplicate copies:",
    len(removed_duplicates)
)

print(
    "Observations after duplicate resolution:",
    len(clean)
)

# Save duplicate audit
removed_duplicates.to_csv(
    os.path.join(
        VAL,
        "FINAL_DUPLICATE_ROWS_REMOVED.csv"
    ),
    index=False
)

duplicate_audit = df[
    df["exact_duplicate"]
].copy()

duplicate_audit.to_csv(
    os.path.join(
        VAL,
        "FINAL_DUPLICATE_RESOLUTION_AUDIT.csv"
    ),
    index=False
)


# ======================================================================
# PART 2 — FINAL RELEASE VALIDATION
# ======================================================================

print("\n" + "=" * 75)
print("PART 2 — FINAL RELEASE VALIDATION")
print("=" * 75)

clean["release_time_h"] = pd.to_numeric(
    clean["release_time_h"],
    errors="coerce"
)

clean["release_percent"] = pd.to_numeric(
    clean["release_percent"],
    errors="coerce"
)

clean["release_valid"] = (
    clean["release_time_h"].notna()
    &
    clean["release_percent"].notna()
    &
    (clean["release_time_h"] >= 0)
    &
    (clean["release_percent"] >= 0)
    &
    (clean["release_percent"] <= 100)
)

invalid = clean[
    ~clean["release_valid"]
].copy()

print(
    "Release-valid:",
    int(clean["release_valid"].sum())
)

print(
    "Release-invalid:",
    len(invalid)
)

if len(invalid) > 0:
    raise ValueError(
        "Invalid release observations remain."
    )


# ======================================================================
# PART 3 — REQUIRED METADATA
# ======================================================================

print("\n" + "=" * 75)
print("PART 3 — METADATA CHECK")
print("=" * 75)

required_metadata = [
    "source_verified_drug",
    "source_verified_polymer",
    "source_verified_carrier",
    "source_verified_condition",
    "experimental_unit_id"
]

missing = [
    c
    for c in required_metadata
    if c not in clean.columns
]

if missing:
    raise ValueError(
        "Required metadata missing:\n"
        + "\n".join(missing)
    )

def valid_metadata(x):

    if pd.isna(x):
        return False

    s = str(x).strip()

    if not s:
        return False

    bad = [
        "NOT_IDENTIFIED",
        "UNRESOLVED",
        "AMBIGUOUS",
        "MANUAL_REVIEW"
    ]

    return not any(
        b in s.upper()
        for b in bad
    )

for c in required_metadata[:-1]:

    clean[c + "_valid"] = (
        clean[c]
        .apply(valid_metadata)
    )

clean["metadata_complete"] = (
    clean[
        [
            c + "_valid"
            for c in required_metadata[:-1]
        ]
    ]
    .all(axis=1)
)

print(
    "Complete metadata:",
    int(
        clean["metadata_complete"].sum()
    )
)

if not clean["metadata_complete"].all():
    raise ValueError(
        "Incomplete metadata detected. "
        "Do not proceed with ML."
    )


# ======================================================================
# PART 4 — EXPERIMENTAL-UNIT AUDIT
# ======================================================================

print("\n" + "=" * 75)
print("PART 4 — EXPERIMENTAL-UNIT AUDIT")
print("=" * 75)

clean["experimental_unit_id"] = (
    clean["experimental_unit_id"]
    .astype(str)
)

clean["unit_n"] = (
    clean.groupby(
        "experimental_unit_id"
    )["observation_id"]
    .transform("count")
)

clean["repeated_measurement"] = (
    clean["unit_n"] > 1
)

print(
    "Unique experimental units:",
    clean[
        "experimental_unit_id"
    ].nunique()
)

print(
    "Repeated measurements:",
    int(
        clean[
            "repeated_measurement"
        ].sum()
    )
)

unit_summary = (
    clean.groupby(
        "experimental_unit_id"
    )
    .agg(
        observations=(
            "observation_id",
            "count"
        ),
        unique_times=(
            "release_time_h",
            "nunique"
        ),
        unique_articles=(
            "pmcid",
            "nunique"
        ),
        unique_tables=(
            "table_index",
            "nunique"
        ),
        mean_release=(
            "release_percent",
            "mean"
        )
    )
    .reset_index()
)

unit_summary.to_csv(
    os.path.join(
        VAL,
        "FINAL_EXPERIMENTAL_UNIT_AUDIT.csv"
    ),
    index=False
)


# ======================================================================
# PART 5 — DEFINE ML VARIABLES
# ======================================================================

print("\n" + "=" * 75)
print("PART 5 — FEATURE / TARGET DEFINITION")
print("=" * 75)

numeric_features = [
    "release_time_h"
]

categorical_features = [
    "source_verified_drug",
    "source_verified_polymer",
    "source_verified_carrier",
    "source_verified_condition"
]

target = "release_percent"

print("Numeric feature:")
print("  release_time_h")

print("Categorical features:")
for c in categorical_features:
    print(" ", c)

print("Target:")
print(" ", target)


# ======================================================================
# PART 6 — GROUPED LEAVE-ONE-UNIT-OUT CROSS VALIDATION
# ======================================================================

print("\n" + "=" * 75)
print("PART 6 — GROUPED CROSS-VALIDATION")
print("=" * 75)

X = clean[
    numeric_features +
    categorical_features
].copy()

y = clean[
    target
].copy()

groups = clean[
    "experimental_unit_id"
].copy()

n_groups = groups.nunique()

print(
    "Independent experimental-unit groups:",
    n_groups
)

if n_groups < 3:
    raise ValueError(
        "Too few independent groups for meaningful grouped CV."
    )

logo = LeaveOneGroupOut()


# ======================================================================
# PREPROCESSOR
# ======================================================================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)


# ======================================================================
# MODELS
# ======================================================================

models = {

    "Ridge": Ridge(
        alpha=1.0
    ),

    "RandomForest": RandomForestRegressor(
        n_estimators=300,
        max_depth=5,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ),

    "GradientBoosting": GradientBoostingRegressor(
        n_estimators=150,
        learning_rate=0.03,
        max_depth=2,
        min_samples_leaf=2,
        random_state=42
    )
}


# ======================================================================
# GROUPED CV
# ======================================================================

all_results = []
prediction_rows = []

for model_name, model in models.items():

    print(
        "\nRunning:",
        model_name
    )

    predictions = np.full(
        len(clean),
        np.nan
    )

    fold = 0

    for train_idx, test_idx in logo.split(
        X,
        y,
        groups=groups
    ):

        fold += 1

        X_train = X.iloc[
            train_idx
        ]

        X_test = X.iloc[
            test_idx
        ]

        y_train = y.iloc[
            train_idx
        ]

        y_test = y.iloc[
            test_idx
        ]

        train_groups = groups.iloc[
            train_idx
        ].unique()

        test_groups = groups.iloc[
            test_idx
        ].unique()

        # --------------------------------------------------------------
        # Leakage check
        # --------------------------------------------------------------

        overlap = set(
            train_groups
        ).intersection(
            set(test_groups)
        )

        if overlap:
            raise RuntimeError(
                "GROUP LEAKAGE DETECTED: "
                + str(overlap)
            )

        pipe = Pipeline(
            steps=[
                (
                    "preprocessor",
                    preprocessor
                ),
                (
                    "model",
                    model
                )
            ]
        )

        pipe.fit(
            X_train,
            y_train
        )

        pred = pipe.predict(
            X_test
        )

        predictions[
            test_idx
        ] = pred

        prediction_rows.append(
            pd.DataFrame({
                "model": model_name,
                "observation_id":
                    clean.iloc[
                        test_idx
                    ]["observation_id"].values,
                "experimental_unit_id":
                    clean.iloc[
                        test_idx
                    ]["experimental_unit_id"].values,
                "pmcid":
                    clean.iloc[
                        test_idx
                    ]["pmcid"].values,
                "time_h":
                    clean.iloc[
                        test_idx
                    ]["release_time_h"].values,
                "actual_release_percent":
                    y_test.values,
                "predicted_release_percent":
                    pred,
                "fold": fold
            })
        )

    # ------------------------------------------------------------------
    # GLOBAL OUT-OF-FOLD METRICS
    # ------------------------------------------------------------------

    valid_pred = ~np.isnan(
        predictions
    )

    actual = y.values[
        valid_pred
    ]

    predicted = predictions[
        valid_pred
    ]

    mae = mean_absolute_error(
        actual,
        predicted
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )

    r2 = r2_score(
        actual,
        predicted
    )

    all_results.append({
        "model": model_name,
        "n_observations": len(actual),
        "n_groups": n_groups,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

    print(
        f"{model_name}: "
        f"R2={r2:.4f}, "
        f"RMSE={rmse:.4f}, "
        f"MAE={mae:.4f}"
    )


# ======================================================================
# PART 7 — RESULTS + PREDICTIONS
# ======================================================================

print("\n" + "=" * 75)
print("PART 7 — MODEL COMPARISON")
print("=" * 75)

results = pd.DataFrame(
    all_results
)

results = results.sort_values(
    "RMSE"
).reset_index(
    drop=True
)

print(
    results.to_string(
        index=False
    )
)

predictions = pd.concat(
    prediction_rows,
    ignore_index=True
)

results_file = os.path.join(
    RESULTS,
    "FINAL_GROUPED_CV_MODEL_COMPARISON.csv"
)

predictions_file = os.path.join(
    RESULTS,
    "FINAL_GROUPED_CV_PREDICTIONS.csv"
)

results.to_csv(
    results_file,
    index=False
)

predictions.to_csv(
    predictions_file,
    index=False
)


# ======================================================================
# PART 8 — FINAL MODEL FIT + FEATURE IMPORTANCE
# ======================================================================

print("\n" + "=" * 75)
print("PART 8 — FINAL MODEL FIT / FEATURE IMPORTANCE")
print("=" * 75)

# Fit each model on the complete cleaned dataset.
#
# IMPORTANT:
# These fitted models are for final interpretation.
# The grouped CV metrics above are the estimates of generalization.
# ----------------------------------------------------------------------

feature_importance_records = []

for model_name, model in models.items():

    pipe = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "model",
                model
            )
        ]
    )

    pipe.fit(
        X,
        y
    )

    # --------------------------------------------------------------
    # Extract feature names
    # --------------------------------------------------------------

    prep = pipe.named_steps[
        "preprocessor"
    ]

    estimator = pipe.named_steps[
        "model"
    ]

    try:

        feature_names = (
            prep
            .get_feature_names_out()
        )

    except Exception:

        feature_names = [
            f"feature_{i}"
            for i in range(
                len(
                    estimator
                    .feature_importances_
                    if hasattr(
                        estimator,
                        "feature_importances_"
                    )
                    else estimator.coef_
                )
            )
        ]

    # --------------------------------------------------------------
    # Importance
    # --------------------------------------------------------------

    if hasattr(
        estimator,
        "feature_importances_"
    ):

        importance = (
            estimator
            .feature_importances_
        )

    elif hasattr(
        estimator,
        "coef_"
    ):

        importance = np.abs(
            np.asarray(
                estimator.coef_
            ).ravel()
        )

    else:

        importance = np.zeros(
            len(feature_names)
        )

    fi = pd.DataFrame({
        "model": model_name,
        "feature": feature_names,
        "importance": importance
    })

    fi = fi.sort_values(
        "importance",
        ascending=False
    )

    feature_importance_records.append(
        fi
    )

feature_importance = pd.concat(
    feature_importance_records,
    ignore_index=True
)

feature_file = os.path.join(
    RESULTS,
    "FINAL_MODEL_FEATURE_IMPORTANCE.csv"
)

feature_importance.to_csv(
    feature_file,
    index=False
)

print(
    "\nTop features:"
)

print(
    feature_importance
    .sort_values(
        ["model", "importance"],
        ascending=[True, False]
    )
    .groupby("model")
    .head(10)
    .to_string(index=False)
)


# ======================================================================
# FINAL MASTER DATASET
# ======================================================================

final_master = clean.copy()

final_master_file = os.path.join(
    FINAL,
    "AI_Polymer_Drug_Release_FINAL_CLEAN_ML_DATASET.csv"
)

final_master.to_csv(
    final_master_file,
    index=False
)


# ======================================================================
# FINAL PROJECT AUDIT
# ======================================================================

audit = pd.DataFrame({
    "metric": [
        "Initial ML observations",
        "Exact duplicate rows",
        "Duplicate rows removed",
        "Final clean observations",
        "Release-valid observations",
        "Complete metadata",
        "Unique experimental units",
        "Repeated measurements",
        "Unique PMCIDs",
        "Unique source tables",
        "Models evaluated",
        "Validation method"
    ],

    "value": [
        len(df),
        int(
            df["exact_duplicate"].sum()
        ),
        len(removed_duplicates),
        len(clean),
        int(
            clean["release_valid"].sum()
        ),
        int(
            clean["metadata_complete"].sum()
        ),
        clean[
            "experimental_unit_id"
        ].nunique(),
        int(
            clean[
                "repeated_measurement"
            ].sum()
        ),
        clean["pmcid"].nunique(),
        clean[
            [
                "pmcid",
                "table_index"
            ]
        ]
        .drop_duplicates()
        .shape[0],
        len(models),
        "Leave-One-Experimental-Unit-Out"
    ]
})

audit_file = os.path.join(
    RESULTS,
    "FINAL_PROJECT_AUDIT.csv"
)

audit.to_csv(
    audit_file,
    index=False
)


# ======================================================================
# FINAL REPORT
# ======================================================================

print("\n" + "=" * 75)
print("FINAL PROJECT PIPELINE COMPLETE")
print("=" * 75)

print(
    "Initial ML observations:",
    len(df)
)

print(
    "Exact duplicate rows:",
    int(
        df["exact_duplicate"].sum()
    )
)

print(
    "Duplicate copies removed:",
    len(removed_duplicates)
)

print(
    "Final clean observations:",
    len(clean)
)

print(
    "Unique experimental units:",
    clean[
        "experimental_unit_id"
    ].nunique()
)

print(
    "Repeated measurements:",
    int(
        clean[
            "repeated_measurement"
        ].sum()
    )
)

print()
print("MODEL RESULTS")
print(
    results.to_string(
        index=False
    )
)

print()
print("FILES CREATED")
print("-" * 75)

print(final_master_file)
print(results_file)
print(predictions_file)
print(feature_file)
print(audit_file)

print()
print("=" * 75)
print("IMPORTANT SCIENTIFIC INTERPRETATION")
print("=" * 75)

print(
    "Grouped CV was used to prevent repeated measurements from "
    "the same experimental unit entering both train and test."
)

print(
    "The six independent experimental units are the limiting factor "
    "for generalization."
)

print(
    "Model performance must therefore be interpreted as exploratory, "
    "not as strong evidence of population-level predictive validity."
)

print(
    "No ANN was used because the independent-unit sample size is too small "
    "to justify a complex neural network."
)

print(
    "No synthetic observations were generated."
)

print(
    "No interpolation was performed."
)

print(
    "No observations were intentionally duplicated."
)

print(
    "The model was not allowed to use observation identity as a predictor."
)

print("=" * 75)
```
