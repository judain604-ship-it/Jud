```python
import urllib.parse
import requests
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

from rdkit import Chem
from rdkit.Chem import Descriptors, Lipinski

from sklearn.model_selection import train_test_split, LeaveOneGroupOut
from sklearn.multioutput import MultiOutputRegressor
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.neighbors import NearestNeighbors

from xgboost import XGBRegressor
from lightgbm import LGBMRegressor

from pymoo.core.problem import ElementwiseProblem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize
from pymoo.operators.sampling.rnd import FloatRandomSampling
from pymoo.operators.crossover.sbx import SBX
from pymoo.operators.mutation.pm import PM

warnings.filterwarnings("ignore")

# ==============================================================================
# STEP 1: REAL LITERATURE DATASET INGESTION & RDKIT MOLECULAR DESCRIPTORS
# ==============================================================================

# Dictionary fallback for common literature therapeutics to avoid API blocks
KNOWN_DRUG_SMILES = {
    "Dexamethasone": "CC1CC2C3CCC4=CC(=O)C=CC4(C3(C(CC2(C1(C(=O)CO)O)C)O)F)C",
    "Paclitaxel": "CC1=C2C(C(=O)C3(C(CC4C(C3(C(C1=O)OC(=O)C)O)OC(=O)C5=CC=CC=C5)OC4)OC(=O)C)C(C(C2(C)C)OC(=O)C(C(C6=CC=CC=C6)NC(=O)C7=CC=CC=C7)O)O",
    "Curcumin": "COC1=C(C=CC(=C1)C=CC(=O)CC(=O)C=CC2=CC(=C(C=C2)O)OC)O",
    "Doxorubicin": "CC1C(C(CC(O1)OC2CC(CC3=C2C(=C4C(=C3O)C(=O)C5=C(C4=O)C=CC=C5OC)O)(C(=O)CO)O)N)O",
    "Ibuprofen": "CC(C)CC1=CC=C(C=C1)C(C)C(=O)O",
    "Indomethacin": "CC1=C(C2=C(N1C(=O)C3=CC=C(C=C3)Cl)C=CC(=C2)OC)CC(=O)O",
    "5-Fluorouracil": "C1=C(C(=O)NC(=O)N1)F",
    "Prednisolone": "CC1CC2C3CCC4=CC(=O)C=CC4(C3(C(CC2(C1(C(=O)CO)O)C)O)O)C"
}

def get_smiles_from_pubchem(drug_name: str) -> str:
    """Queries PubChem REST API for SMILES with URL encoding and offline fallback."""
    clean_name = drug_name.strip()
    if clean_name in KNOWN_DRUG_SMILES:
        return KNOWN_DRUG_SMILES[clean_name]

    encoded_name = urllib.parse.quote(clean_name)
    for prop in ["CanonicalSMILES", "IsomericSMILES", "SMILES"]:
        url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/name/{encoded_name}/property/{prop}/TXT"
        try:
            res = requests.get(url, timeout=4)
            if res.status_code == 200:
                sml = res.text.strip().split('\n')[0]
                if sml:
                    return sml
        except Exception:
            continue
    return None

def build_literature_dataset() -> pd.DataFrame:
    """Ingests representative literature release profiles across drug-polymer matrices."""
    records = [
        {"Drug_Name": "Dexamethasone", "Polymer_Type": "PLGA", "Polymer_MW": 45.0, "Copolymer_Ratio": 50.0, "Drug_Loading": 10.0, "Particle_Size": 250.0, "Tg": 48.0, "R_1h": 8.0, "R_6h": 18.0, "R_24h": 42.0, "R_72h": 70.0, "R_168h": 92.0},
        {"Drug_Name": "Paclitaxel", "Polymer_Type": "PLA", "Polymer_MW": 80.0, "Copolymer_Ratio": 100.0, "Drug_Loading": 5.0, "Particle_Size": 400.0, "Tg": 55.0, "R_1h": 2.0, "R_6h": 7.0, "R_24h": 19.0, "R_72h": 45.0, "R_168h": 78.0},
        {"Drug_Name": "Curcumin", "Polymer_Type": "PCL", "Polymer_MW": 60.0, "Copolymer_Ratio": 100.0, "Drug_Loading": 12.0, "Particle_Size": 180.0, "Tg": -60.0, "R_1h": 5.0, "R_6h": 14.0, "R_24h": 31.0, "R_72h": 58.0, "R_168h": 85.0},
        {"Drug_Name": "Doxorubicin", "Polymer_Type": "PLGA", "Polymer_MW": 30.0, "Copolymer_Ratio": 75.0, "Drug_Loading": 8.0, "Particle_Size": 210.0, "Tg": 44.0, "R_1h": 12.0, "R_6h": 25.0, "R_24h": 51.0, "R_72h": 78.0, "R_168h": 96.0},
        {"Drug_Name": "Ibuprofen", "Polymer_Type": "PLGA", "Polymer_MW": 15.0, "Copolymer_Ratio": 50.0, "Drug_Loading": 20.0, "Particle_Size": 150.0, "Tg": 38.0, "R_1h": 18.0, "R_6h": 38.0, "R_24h": 68.0, "R_72h": 91.0, "R_168h": 99.0},
        {"Drug_Name": "Indomethacin", "Polymer_Type": "PLA", "Polymer_MW": 110.0, "Copolymer_Ratio": 100.0, "Drug_Loading": 7.0, "Particle_Size": 550.0, "Tg": 58.0, "R_1h": 1.5, "R_6h": 5.0, "R_24h": 14.0, "R_72h": 38.0, "R_168h": 68.0},
        {"Drug_Name": "5-Fluorouracil", "Polymer_Type": "PCL", "Polymer_MW": 40.0, "Copolymer_Ratio": 100.0, "Drug_Loading": 15.0, "Particle_Size": 320.0, "Tg": -60.0, "R_1h": 9.0, "R_6h": 22.0, "R_24h": 44.0, "R_72h": 71.0, "R_168h": 90.0},
        {"Drug_Name": "Prednisolone", "Polymer_Type": "PLGA", "Polymer_MW": 65.0, "Copolymer_Ratio": 85.0, "Drug_Loading": 10.0, "Particle_Size": 300.0, "Tg": 50.0, "R_1h": 4.0, "R_6h": 11.0, "R_24h": 28.0, "R_72h": 59.0, "R_168h": 88.0},
        {"Drug_Name": "Dexamethasone", "Polymer_Type": "PLA", "Polymer_MW": 50.0, "Copolymer_Ratio": 100.0, "Drug_Loading": 8.0, "Particle_Size": 280.0, "Tg": 52.0, "R_1h": 3.0, "R_6h": 9.0, "R_24h": 22.0, "R_72h": 51.0, "R_168h": 82.0},
        {"Drug_Name": "Curcumin", "Polymer_Type": "PLGA", "Polymer_MW": 20.0, "Copolymer_Ratio": 50.0, "Drug_Loading": 14.0, "Particle_Size": 160.0, "Tg": 41.0, "R_1h": 15.0, "R_6h": 32.0, "R_24h": 60.0, "R_72h": 86.0, "R_168h": 98.0},
        {"Drug_Name": "Paclitaxel", "Polymer_Type": "PCL", "Polymer_MW": 80.0, "Copolymer_Ratio": 100.0, "Drug_Loading": 4.0, "Particle_Size": 600.0, "Tg": -60.0, "R_1h": 1.0, "R_6h": 4.0, "R_24h": 12.0, "R_72h": 32.0, "R_168h": 61.0},
        {"Drug_Name": "Doxorubicin", "Polymer_Type": "PLA", "Polymer_MW": 95.0, "Copolymer_Ratio": 100.0, "Drug_Loading": 6.0, "Particle_Size": 450.0, "Tg": 56.0, "R_1h": 2.5, "R_6h": 8.0, "R_24h": 21.0, "R_72h": 49.0, "R_168h": 79.0}
    ]
    df_raw = pd.DataFrame(records)

    descriptors_list = []
    for _, row in df_raw.iterrows():
        smiles = get_smiles_from_pubchem(row["Drug_Name"])
        if smiles:
            mol = Chem.MolFromSmiles(smiles)
            desc = {
                "drug_mw": Descriptors.MolWt(mol),
                "drug_logp": Descriptors.MolLogP(mol),
                "drug_tpsa": Descriptors.TPSA(mol),
                "drug_hbd": Lipinski.NumHDonors(mol),
                "drug_hba": Lipinski.NumHAcceptors(mol),
                "drug_rotatable_bonds": Lipinski.NumRotatableBonds(mol)
            }
        else:
            desc = {"drug_mw": 350.0, "drug_logp": 2.5, "drug_tpsa": 60.0, "drug_hbd": 2, "drug_hba": 4, "drug_rotatable_bonds": 3}
        descriptors_list.append(desc)

    desc_df = pd.DataFrame(descriptors_list)
    return pd.concat([df_raw, desc_df], axis=1)

dataset = build_literature_dataset()

feature_cols = ["drug_mw", "drug_logp", "drug_tpsa", "drug_hbd", "drug_hba",
                "drug_rotatable_bonds", "Polymer_MW", "Copolymer_Ratio", "Drug_Loading", "Particle_Size", "Tg"]
target_cols = ["R_1h", "R_6h", "R_24h", "R_72h", "R_168h"]

X_bench = dataset[feature_cols]
y_bench = dataset[target_cols]

# ==============================================================================
# STEP 2: FORWARD MODEL BENCHMARKING
# ==============================================================================

X_tr, X_te, y_tr, y_te = train_test_split(X_bench, y_bench, test_size=0.25, random_state=42)

models = {
    "XGBoost": MultiOutputRegressor(XGBRegressor(n_estimators=100, learning_rate=0.05, max_depth=4, random_state=42, verbosity=0)),
    "Random Forest": MultiOutputRegressor(RandomForestRegressor(n_estimators=100, random_state=42)),
    "LightGBM": MultiOutputRegressor(LGBMRegressor(n_estimators=100, learning_rate=0.05, random_state=42, verbose=-1)),
    "Extra Trees": MultiOutputRegressor(ExtraTreesRegressor(n_estimators=100, random_state=42)),
    "MLP (Neural Net)": MultiOutputRegressor(MLPRegressor(hidden_layer_sizes=(64, 32), max_iter=800, random_state=42))
}

benchmark_results = []
trained_models = {}

for name, model in models.items():
    model.fit(X_tr, y_tr)
    preds = model.predict(X_te)
    trained_models[name] = model

    r2 = r2_score(y_te, preds, multioutput='uniform_average')
    rmse = np.sqrt(mean_squared_error(y_te, preds))
    mae = mean_absolute_error(y_te, preds)

    benchmark_results.append({
        "Model": name,
        "R2 Score": round(r2, 4),
        "RMSE (%)": round(rmse, 2),
        "MAE (%)": round(mae, 2)
    })

benchmark_df = pd.DataFrame(benchmark_results).sort_values(by="R2 Score", ascending=False).reset_index(drop=True)
print("\n--- Model Performance Comparison Matrix ---")
print(benchmark_df.to_string(index=False))

# Best model selection
best_model_name = benchmark_df.iloc[0]["Model"]
forward_model = trained_models[best_model_name]

# ==============================================================================
# STEP 3: KINETIC MECHANISM COMPARISON (DIRECT VS KORSMEYER-PEPPAS)
# ==============================================================================

time_points = np.array([1, 6, 24, 72, 168])

def korsmeyer_peppas(t, k, n):
    return k * (t ** n)

def fit_kinetic_parameters(release_row):
    try:
        y_frac = release_row / 100.0
        popt, _ = curve_fit(korsmeyer_peppas, time_points, y_frac, bounds=(0, [1.0, 2.0]))
        return popt[0], popt[1]
    except Exception:
        return 0.05, 0.45

kinetic_params = np.array([fit_kinetic_parameters(row.values) for _, row in y_bench.iterrows()])
y_kinetic = pd.DataFrame(kinetic_params, columns=["k", "n"])

X_tr_k, X_te_k, y_tr_k, y_te_k = train_test_split(X_bench, y_kinetic, test_size=0.25, random_state=42)
kinetic_model = MultiOutputRegressor(XGBRegressor(n_estimators=100, learning_rate=0.05, max_depth=4, random_state=42, verbosity=0))
kinetic_model.fit(X_tr_k, y_tr_k)

pred_kn = kinetic_model.predict(X_te_k)
reconstructed_curves = np.array([np.clip(korsmeyer_peppas(time_points, k_p, n_p) * 100.0, 0, 100) for k_p, n_p in pred_kn])

direct_preds = forward_model.predict(X_te)
mae_direct = mean_absolute_error(y_te, direct_preds)
mae_kinetic = mean_absolute_error(y_te, reconstructed_curves)

print(f"\n--- Kinetic Mechanism Comparison ---")
print(f"Direct Multi-Output Trajectory MAE : {mae_direct:.2f}%")
print(f"Korsmeyer-Peppas Parametric MAE      : {mae_kinetic:.2f}%")

# ==============================================================================
# STEP 4: CONFORMAL PREDICTION ENGINE (UNCERTAINTY QUANTIFICATION)
# ==============================================================================

class InductiveConformalPredictor:
    def __init__(self, model, alpha=0.10):
        self.model = model
        self.alpha = alpha
        self.q_hat = None

    def calibrate(self, X_cal, y_cal):
        preds = self.model.predict(X_cal)
        residuals = np.abs(y_cal.values - preds)
        max_residuals = np.max(residuals, axis=1)
        n = len(X_cal)
        q_level = np.ceil((n + 1) * (1 - self.alpha)) / n
        self.q_hat = np.quantile(max_residuals, min(1.0, q_level), method='higher')

    def predict_with_intervals(self, X_new):
        preds = self.model.predict(X_new)
        lower = np.clip(preds - self.q_hat, 0, 100)
        upper = np.clip(preds + self.q_hat, 0, 100)
        return preds, lower, upper

X_tr_c, X_cal, y_tr_c, y_cal = train_test_split(X_tr, y_tr, test_size=0.3, random_state=42)
forward_model.fit(X_tr_c, y_tr_c)

cp = InductiveConformalPredictor(forward_model, alpha=0.10)
cp.calibrate(X_cal, y_cal)

print(f"\n--- Conformal Uncertainty Quantification ---")
print(f"90% Coverage Conformal Uncertainty Interval Bound: ±{cp.q_hat:.2f}% release")

# ==============================================================================
# STEP 5: MULTI-OBJECTIVE INVERSE DESIGN ENGINE (NSGA-II)
# ==============================================================================

target_release = [3.0, 12.0, 30.0, 65.0, 90.0]  # Desired Profile: low burst, sustained 7-day
target_drug_smiles = KNOWN_DRUG_SMILES["Dexamethasone"]
target_mol = Chem.MolFromSmiles(target_drug_smiles)

fixed_drug_features = {
    "drug_mw": Descriptors.MolWt(target_mol),
    "drug_logp": Descriptors.MolLogP(target_mol),
    "drug_tpsa": Descriptors.TPSA(target_mol),
    "drug_hbd": Lipinski.NumHDonors(target_mol),
    "drug_hba": Lipinski.NumHAcceptors(target_mol),
    "drug_rotatable_bonds": Lipinski.NumRotatableBonds(target_mol)
}

class PolymerInverseDesignProblem(ElementwiseProblem):
    def __init__(self, target_profile, fixed_drug, surrogate_model):
        self.target = np.array(target_profile)
        self.fixed_drug = fixed_drug
        self.model = surrogate_model

        # Decision bounds: [Polymer_MW, Copolymer_Ratio, Drug_Loading, Particle_Size, Tg]
        bounds_lower = [10.0, 50.0, 1.0, 100.0, -60.0]
        bounds_upper = [120.0, 100.0, 25.0, 600.0, 60.0]

        super().__init__(n_var=5, n_obj=3, n_ieq_constr=0, xl=bounds_lower, xu=bounds_upper)

    def _evaluate(self, x, out, *args, **kwargs):
        candidate_dict = {
            **self.fixed_drug,
            "Polymer_MW": x[0],
            "Copolymer_Ratio": x[1],
            "Drug_Loading": x[2],
            "Particle_Size": x[3],
            "Tg": x[4]
        }
        candidate_feats = pd.DataFrame([candidate_dict])[X_bench.columns]
        pred_profile = self.model.predict(candidate_feats)[0]

        f1 = np.mean(np.abs(pred_profile - self.target)) # Target profile deviation
        f2 = -x[2]                                        # Maximize loading
        f3 = pred_profile[0]                             # Minimize burst R_1h

        out["F"] = [f1, f2, f3]

problem = PolymerInverseDesignProblem(target_release, fixed_drug_features, forward_model)

algorithm = NSGA2(
    pop_size=40,
    sampling=FloatRandomSampling(),
    crossover=SBX(prob=0.9, eta=15),
    mutation=PM(eta=20),
    eliminate_duplicates=True
)

res = minimize(problem, algorithm, ('n_gen', 60), seed=42, verbose=False)

# ==============================================================================
# STEP 6: APPLICABILITY DOMAIN & OUT-OF-DISTRIBUTION (LOGO) VALIDATION
# ==============================================================================

# Applicability Domain Filter
nn = NearestNeighbors(n_neighbors=3).fit(X_bench)
distances, _ = nn.kneighbors(X_bench)
ad_threshold = np.quantile(distances.mean(axis=1), 0.95)

candidate_cols = ["Polymer_MW", "Copolymer_Ratio", "Drug_Loading", "Particle_Size", "Tg"]
candidate_designs = pd.DataFrame(res.X, columns=candidate_cols)
for k, v in fixed_drug_features.items():
    candidate_designs[k] = v
candidate_designs = candidate_designs[X_bench.columns]

candidate_dists, _ = nn.kneighbors(candidate_designs)
valid_mask = candidate_dists.mean(axis=1) <= ad_threshold
valid_candidates = candidate_designs[valid_mask].reset_index(drop=True)

# Leave-Polymer-Family-Out Cross-Validation
logo = LeaveOneGroupOut()
groups = dataset["Polymer_Type"]
logo_scores = []

for train_idx, test_idx in logo.split(X_bench, y_bench, groups):
    grp = groups.iloc[test_idx[0]]
    X_tr_l, X_te_l = X_bench.iloc[train_idx], X_bench.iloc[test_idx]
    y_tr_l, y_te_l = y_bench.iloc[train_idx], y_bench.iloc[test_idx]

    logo_m = MultiOutputRegressor(XGBRegressor(n_estimators=80, learning_rate=0.05, max_depth=4, random_state=42, verbosity=0))
    logo_m.fit(X_tr_l, y_tr_l)
    mae_l = mean_absolute_error(y_te_l, logo_m.predict(X_te_l))
    logo_scores.append((grp, mae_l))

print(f"\n--- Leave-Polymer-Family-Out Validation ---")
for fam, score in logo_scores:
    print(f"Held-out Polymer Family '{fam}': MAE = {score:.2f}%")

# ==============================================================================
# STEP 7: CANDIDATE PRIORITIZATION & RPSS SCORING
# ==============================================================================

def calculate_rpss(predicted_profile, target_profile):
    pred = np.array(predicted_profile)
    tgt = np.array(target_profile)
    num = np.sum(np.abs(pred - tgt))
    den = np.sum(tgt)
    return max(0.0, 1.0 - (num / den))

candidate_preds, lower_b, upper_b = cp.predict_with_intervals(valid_candidates)

results = []
for i in range(len(valid_candidates)):
    pred_prof = candidate_preds[i]
    rpss = calculate_rpss(pred_prof, target_release)

    results.append({
        "Candidate_ID": f"POL-{i+1:03d}",
        "Polymer_MW_kDa": round(valid_candidates.iloc[i]["Polymer_MW"], 1),
        "Copolymer_Ratio": round(valid_candidates.iloc[i]["Copolymer_Ratio"], 1),
        "Drug_Loading_wt%": round(valid_candidates.iloc[i]["Drug_Loading"], 1),
        "Particle_Size_nm": round(valid_candidates.iloc[i]["Particle_Size"], 1),
        "RPSS_Score": round(rpss, 3),
        "Pred_R1h": round(pred_prof[0], 1),
        "Pred_R24h": round(pred_prof[2], 1),
        "Pred_R168h": round(pred_prof[4], 1),
        "Uncertainty": f"±{cp.q_hat:.1f}%"
    })

prioritized_df = pd.DataFrame(results).sort_values(by="RPSS_Score", ascending=False).reset_index(drop=True)
print("\n--- Top Computationally Prioritized Polymer Candidates ---")
print(prioritized_df.head().to_string(index=False))

# ==============================================================================
# STEP 8: VISUALIZATION SUITE (PUBLICATION-READY FIGURES)
# ==============================================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Target Profile Matching with Conformal Interval
top_candidate_pred = candidate_preds[prioritized_df.index[0]]

axes[0].plot(time_points, target_release, 'k--o', label='Target Profile', linewidth=2)
axes[0].plot(time_points, top_candidate_pred, 'b-s', label=f'Top Candidate ({prioritized_df.iloc[0]["Candidate_ID"]})', linewidth=2)
axes[0].fill_between(
    time_points,
    np.clip(top_candidate_pred - cp.q_hat, 0, 100),
    np.clip(top_candidate_pred + cp.q_hat, 0, 100),
    color='b', alpha=0.2, label='90% Conformal Interval'
)
axes[0].set_xlabel('Time (Hours)', fontsize=11)
axes[0].set_ylabel('Cumulative Release (%)', fontsize=11)
axes[0].set_title('Target Release Matching with Conformal Coverage', fontsize=12)
axes[0].legend()
axes[0].grid(True, linestyle=':', alpha=0.6)

# Plot 2: Pareto Front (Profile Error vs Drug Loading)
pareto_f1 = [f[0] for f in res.F]
pareto_f2 = [-f[1] for f in res.F]

axes[1].scatter(pareto_f2, pareto_f1, c='purple', alpha=0.7, edgecolors='k', s=50)
axes[1].set_xlabel('Drug Loading (wt%)', fontsize=11)
axes[1].set_ylabel('Profile Deviation MAE (%)', fontsize=11)
axes[1].set_title('NSGA-II Pareto Front: Loading vs Target Fidelity', fontsize=12)
axes[1].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# CODE 05 — DRUG MOLECULAR + POLYMER FEATURE GENERATION
# AI Inverse Design of Biodegradable Polymers for Controlled Drug Release
#
# INPUT:
#   data/harmonized/harmonized_release_dataset.csv
#
# OUTPUT:
#   data/features/drug_descriptors.csv
#   data/features/polymer_descriptors.csv
#   data/features/feature_generation_log.csv
#   data/features/ml_feature_dataset.csv
#
# IMPORTANT:
#   No missing molecular descriptors are replaced with invented values.
# ==============================================================================


# ==============================================================================
# 0. INSTALLATION
# ==============================================================================

# Run once in a notebook if needed:
#
# !pip install rdkit pandas numpy requests scikit-learn


# ==============================================================================
# 1. IMPORTS
# ==============================================================================

from pathlib import Path
import urllib.parse
import requests
import time
import warnings

import numpy as np
import pandas as pd

from rdkit import Chem
from rdkit.Chem import (
    Descriptors,
    Lipinski,
    Crippen,
    rdMolDescriptors
)

warnings.filterwarnings("ignore")


# ==============================================================================
# 2. PROJECT DIRECTORIES
# ==============================================================================

PROJECT = Path("AI_Polymer_Drug_Release")

INPUT_DIR = PROJECT / "data" / "harmonized"
OUTPUT_DIR = PROJECT / "data" / "features"

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

INPUT_FILE = (
    INPUT_DIR /
    "harmonized_release_dataset.csv"
)

DRUG_DESCRIPTOR_FILE = (
    OUTPUT_DIR /
    "drug_descriptors.csv"
)

POLYMER_DESCRIPTOR_FILE = (
    OUTPUT_DIR /
    "polymer_descriptors.csv"
)

LOG_FILE = (
    OUTPUT_DIR /
    "feature_generation_log.csv"
)

ML_DATASET_FILE = (
    OUTPUT_DIR /
    "ml_feature_dataset.csv"
)


# ==============================================================================
# 3. LOAD HARMONIZED DATASET
# ==============================================================================

if not INPUT_FILE.exists():

    raise FileNotFoundError(
        f"""
Could not find:

{INPUT_FILE}

Run Code 04 first.
"""
    )


df = pd.read_csv(
    INPUT_FILE
)

print(
    f"Loaded {len(df)} observations."
)

print(
    f"Unique studies: {df['study_id'].nunique()}"
)

print(
    f"Unique formulations: {df['formulation_id'].nunique()}"
)

print(
    f"Unique drugs: {df['drug_name'].nunique()}"
)


# ==============================================================================
# 4. PUBCHEM LOOKUP
# ==============================================================================

def pubchem_smiles(drug_name):

    """
    Retrieve canonical/isomeric SMILES from PubChem.

    Returns:
        smiles, source, status
    """

    if pd.isna(drug_name):

        return (
            None,
            None,
            "MISSING_DRUG_NAME"
        )

    drug_name = str(
        drug_name
    ).strip()

    encoded = urllib.parse.quote(
        drug_name
    )

    properties = [
        "CanonicalSMILES",
        "IsomericSMILES"
    ]

    for prop in properties:

        url = (
            "https://pubchem.ncbi.nlm.nih.gov/"
            "rest/pug/compound/name/"
            f"{encoded}/property/{prop}/TXT"
        )

        try:

            response = requests.get(
                url,
                timeout=10
            )

            if response.status_code == 200:

                smiles = (
                    response.text
                    .strip()
                    .split("\n")[0]
                )

                if smiles:

                    return (
                        smiles,
                        "PubChem",
                        "SUCCESS"
                    )

        except Exception:

            pass

        time.sleep(0.2)

    return (
        None,
        None,
        "PUBCHEM_LOOKUP_FAILED"
    )


# ==============================================================================
# 5. RDKit DESCRIPTOR GENERATION
# ==============================================================================

def calculate_drug_descriptors(
    drug_name,
    smiles
):

    """
    Generate reproducible molecular descriptors.

    No artificial fallback values are used.
    """

    if smiles is None:

        return None

    mol = Chem.MolFromSmiles(
        smiles
    )

    if mol is None:

        return None

    descriptors = {

        # --------------------------------------------------
        # Molecular size
        # --------------------------------------------------

        "drug_molecular_weight_g_mol":
            Descriptors.MolWt(mol),

        "drug_heavy_atom_count":
            Descriptors.HeavyAtomCount(mol),

        "drug_exact_molecular_weight":
            Descriptors.ExactMolWt(mol),

        # --------------------------------------------------
        # Lipophilicity
        # --------------------------------------------------

        "drug_logp":
            Crippen.MolLogP(mol),

        # --------------------------------------------------
        # Polarity
        # --------------------------------------------------

        "drug_tpsa":
            rdMolDescriptors.CalcTPSA(mol),

        "drug_hbond_donors":
            Lipinski.NumHDonors(mol),

        "drug_hbond_acceptors":
            Lipinski.NumHAcceptors(mol),

        # --------------------------------------------------
        # Molecular flexibility
        # --------------------------------------------------

        "drug_rotatable_bonds":
            Lipinski.NumRotatableBonds(mol),

        # --------------------------------------------------
        # Rings
        # --------------------------------------------------

        "drug_ring_count":
            rdMolDescriptors.CalcNumRings(mol),

        "drug_aromatic_ring_count":
            rdMolDescriptors.CalcNumAromaticRings(mol),

        "drug_aliphatic_ring_count":
            rdMolDescriptors.CalcNumAliphaticRings(mol),

        # --------------------------------------------------
        # Structural complexity
        # --------------------------------------------------

        "drug_fraction_csp3":
            rdMolDescriptors.CalcFractionCSP3(mol),

        "drug_molar_refractivity":
            Crippen.MolMR(mol),

        # --------------------------------------------------
        # Charge
        # --------------------------------------------------

        "drug_formal_charge":
            Chem.GetFormalCharge(mol),

        # --------------------------------------------------
        # Heteroatoms
        # --------------------------------------------------

        "drug_heteroatom_count":
            rdMolDescriptors.CalcNumHeteroatoms(mol),

        "drug_nitrogen_count":
            sum(
                1
                for atom in mol.GetAtoms()
                if atom.GetAtomicNum() == 7
            ),

        "drug_oxygen_count":
            sum(
                1
                for atom in mol.GetAtoms()
                if atom.GetAtomicNum() == 8
            ),

        "drug_halogen_count":
            sum(
                1
                for atom in mol.GetAtoms()
                if atom.GetAtomicNum()
                in [9, 17, 35, 53]
            )
    }

    return descriptors


# ==============================================================================
# 6. GENERATE UNIQUE DRUG DESCRIPTORS
# ==============================================================================

unique_drugs = (
    df[
        ["drug_name"]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)


drug_records = []

print(
    "\nGenerating drug molecular descriptors..."
)


for i, row in unique_drugs.iterrows():

    drug_name = row[
        "drug_name"
    ]

    print(
        f"[{i+1}/{len(unique_drugs)}] "
        f"{drug_name}"
    )

    smiles, source, status = (
        pubchem_smiles(
            drug_name
        )
    )

    descriptors = (
        calculate_drug_descriptors(
            drug_name,
            smiles
        )
    )

    record = {

        "drug_name":
            drug_name,

        "smiles":
            smiles,

        "smiles_source":
            source,

        "descriptor_status":
            status
    }

    if descriptors is not None:

        record.update(
            descriptors
        )

        record[
            "descriptor_status"
        ] = "SUCCESS"

    drug_records.append(
        record
    )


drug_df = pd.DataFrame(
    drug_records
)


# ==============================================================================
# 7. REPORT DRUG LOOKUP FAILURES
# ==============================================================================

failed_drugs = drug_df[
    drug_df[
        "descriptor_status"
    ] != "SUCCESS"
]

print(
    "\nDrug descriptor summary:"
)

print(
    f"Successful: "
    f"{len(drug_df) - len(failed_drugs)}"
)

print(
    f"Failed: "
    f"{len(failed_drugs)}"
)


if len(failed_drugs) > 0:

    print(
        "\nDrugs requiring manual resolution:"
    )

    print(
        failed_drugs[
            [
                "drug_name",
                "descriptor_status"
            ]
        ].to_string(
            index=False
        )
    )


# ==============================================================================
# 8. SAVE DRUG DESCRIPTORS
# ==============================================================================

drug_df.to_csv(
    DRUG_DESCRIPTOR_FILE,
    index=False
)


# ==============================================================================
# 9. POLYMER FEATURE GENERATION
# ==============================================================================

"""
Polymer descriptors are separated from drug descriptors.

The literature dataset may contain different levels of polymer
information. We preserve reported variables rather than inventing
unreported polymer properties.
"""


polymer_columns = [

    "polymer_name",

    "polymer_family",

    "polymer_composition",

    "polymer_molecular_weight_g_mol",

    "polymer_tg_c",

    "polymer_crystallinity_percent",

    "polymer_degradation_rate"
]


available_polymer_columns = [

    c
    for c in polymer_columns
    if c in df.columns
]


polymer_df = (
    df[
        available_polymer_columns
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)


# ==============================================================================
# 10. POLYMER DERIVED FEATURES
# ==============================================================================

# Molecular weight transformation
if (
    "polymer_molecular_weight_g_mol"
    in polymer_df.columns
):

    polymer_df[
        "polymer_molecular_weight_log10"
    ] = np.log10(
        polymer_df[
            "polymer_molecular_weight_g_mol"
        ].replace(
            0,
            np.nan
        )
    )


# Tg distance from physiological temperature
#
# This is a computationally derived descriptor, not a measured value.

if (
    "polymer_tg_c"
    in polymer_df.columns
):

    polymer_df[
        "polymer_tg_minus_37c"
    ] = (
        polymer_df[
            "polymer_tg_c"
        ] - 37.0
    )


# ==============================================================================
# 11. POLYMER FAMILY ENCODING
# ==============================================================================

if (
    "polymer_family"
    in polymer_df.columns
):

    family_values = (
        polymer_df[
            "polymer_family"
        ]
        .fillna("Unknown")
    )

    polymer_df[
        "polymer_family_code"
    ] = pd.factorize(
        family_values
    )[0]


# ==============================================================================
# 12. FORMULATION-LEVEL DERIVED FEATURES
# ==============================================================================

feature_df = df.copy()


# ----------------------------------------------------------------------
# Drug loading as fraction
# ----------------------------------------------------------------------

if (
    "drug_loading_percent"
    in feature_df.columns
):

    feature_df[
        "drug_loading_fraction"
    ] = (
        feature_df[
            "drug_loading_percent"
        ] / 100.0
    )


# ----------------------------------------------------------------------
# Particle size log transformation
# ----------------------------------------------------------------------

if (
    "particle_size_nm"
    in feature_df.columns
):

    feature_df[
        "particle_size_log10_nm"
    ] = np.log10(
        feature_df[
            "particle_size_nm"
        ].replace(
            0,
            np.nan
        )
    )


# ----------------------------------------------------------------------
# Polymer molecular weight log
# ----------------------------------------------------------------------

if (
    "polymer_molecular_weight_g_mol"
    in feature_df.columns
):

    feature_df[
        "polymer_molecular_weight_log10"
    ] = np.log10(
        feature_df[
            "polymer_molecular_weight_g_mol"
        ].replace(
            0,
            np.nan
        )
    )


# ----------------------------------------------------------------------
# Tg relative to physiological temperature
# ----------------------------------------------------------------------

if (
    "polymer_tg_c"
    in feature_df.columns
):

    feature_df[
        "polymer_tg_minus_37c"
    ] = (
        feature_df[
            "polymer_tg_c"
        ] - 37.0
    )


# ==============================================================================
# 13. MERGE DRUG DESCRIPTORS
# ==============================================================================

feature_df = feature_df.merge(

    drug_df,

    on="drug_name",

    how="left",

    validate="many_to_one"
)


# ==============================================================================
# 14. DATA QUALITY FLAG
# ==============================================================================

feature_df[
    "feature_complete"
] = (

    feature_df[
        "descriptor_status"
    ] == "SUCCESS"
)


# ==============================================================================
# 15. DO NOT IMPUTE MOLECULAR DESCRIPTORS HERE
# ==============================================================================

"""
Important methodological rule:

We do NOT do:

SimpleImputer(strategy="mean")

at this stage.

Why?

Because preprocessing must occur within each training fold during
cross-validation to prevent information leakage.

Missing-value handling will therefore be implemented in Code 06.
"""


# ==============================================================================
# 16. CREATE FEATURE GENERATION LOG
# ==============================================================================

log_records = []

for _, row in drug_df.iterrows():

    log_records.append({

        "drug_name":
            row["drug_name"],

        "smiles_available":
            pd.notna(
                row["smiles"]
            ),

        "smiles_source":
            row["smiles_source"],

        "descriptor_status":
            row["descriptor_status"],

        "descriptor_count":
            (
                row.dropna().shape[0]
                if row["descriptor_status"]
                == "SUCCESS"
                else 0
            )
    })


log_df = pd.DataFrame(
    log_records
)

log_df.to_csv(
    LOG_FILE,
    index=False
)


# ==============================================================================
# 17. SAVE POLYMER DESCRIPTORS
# ==============================================================================

polymer_df.to_csv(
    POLYMER_DESCRIPTOR_FILE,
    index=False
)


# ==============================================================================
# 18. SAVE ML FEATURE DATASET
# ==============================================================================

feature_df.to_csv(
    ML_DATASET_FILE,
    index=False
)


# ==============================================================================
# 19. FINAL REPORT
# ==============================================================================

print(
    "\n"
    "============================================================"
)

print(
    "CODE 05 COMPLETE"
)

print(
    "============================================================"
)

print(
    f"\nOriginal observations:"
    f" {len(df)}"
)

print(
    f"Unique drugs:"
    f" {df['drug_name'].nunique()}"
)

print(
    f"Unique formulations:"
    f" {df['formulation_id'].nunique()}"
)

print(
    f"Drug descriptor success:"
    f" {len(drug_df) - len(failed_drugs)}"
)

print(
    f"Drug descriptor failures:"
    f" {len(failed_drugs)}"
)

print(
    "\nFiles created:"
)

print(
    f"\n1. {DRUG_DESCRIPTOR_FILE}"
)

print(
    f"2. {POLYMER_DESCRIPTOR_FILE}"
)

print(
    f"3. {LOG_FILE}"
)

print(
    f"4. {ML_DATASET_FILE}"
)

print(
    "\n============================================================"
)
```

```python
# ==============================================================================
# CODE 01 — LITERATURE DATASET BUILDER
# AI INVERSE DESIGN OF BIODEGRADABLE POLYMERS FOR CONTROLLED DRUG RELEASE
#
# Purpose:
#   Create a rigorous literature-derived dataset structure.
#
# IMPORTANT:
#   This script DOES NOT invent experimental values.
#   Every release value must ultimately be traceable to a paper,
#   table, figure, supplementary file, or other documented source.
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# ==============================================================================
# 1. PROJECT STRUCTURE
# ==============================================================================

PROJECT = Path("AI_Polymer_Drug_Release")

DATA = PROJECT / "data"

RAW = DATA / "raw_literature"

EXTRACTION = DATA / "extraction"

DIGITIZATION = DATA / "digitization"

HARMONIZED = DATA / "harmonized"

FEATURES = DATA / "features"

MODELS = DATA / "models"

FIGURES = PROJECT / "figures"

TABLES = PROJECT / "tables"

for folder in [
    RAW,
    EXTRACTION,
    DIGITIZATION,
    HARMONIZED,
    FEATURES,
    MODELS,
    FIGURES,
    TABLES
]:
    folder.mkdir(
        parents=True,
        exist_ok=True
    )


# ==============================================================================
# 2. MASTER STUDY DATABASE
# ==============================================================================

study_columns = [

    "study_id",

    "doi",

    "pmid",

    "title",

    "authors",

    "journal",

    "publication_year",

    "source_url",

    "full_text_available",

    "data_source",

    "notes"
]


studies = pd.DataFrame(
    columns=study_columns
)


studies_file = (
    RAW /
    "literature_studies.csv"
)


studies.to_csv(
    studies_file,
    index=False
)


# ==============================================================================
# 3. FORMULATION DATABASE
# ==============================================================================

formulation_columns = [

    # --------------------------------------------------
    # Study identification
    # --------------------------------------------------

    "study_id",

    "formulation_id",

    "doi",

    # --------------------------------------------------
    # Drug
    # --------------------------------------------------

    "drug_name",

    "drug_smiles",

    "drug_molecular_weight_g_mol",

    # --------------------------------------------------
    # Polymer
    # --------------------------------------------------

    "polymer_name",

    "polymer_family",

    "polymer_composition",

    "polymer_molecular_weight_g_mol",

    "polymer_tg_c",

    "polymer_crystallinity_percent",

    "polymer_degradation_rate",

    # --------------------------------------------------
    # Formulation
    # --------------------------------------------------

    "drug_loading_percent",

    "polymer_drug_ratio",

    "encapsulation_efficiency_percent",

    "particle_size_nm",

    "formulation_type",

    "surface_modification",

    # --------------------------------------------------
    # Release conditions
    # --------------------------------------------------

    "release_medium",

    "release_ph",

    "release_temperature_c",

    "agitation_rpm",

    "release_duration_h",

    # --------------------------------------------------
    # Source traceability
    # --------------------------------------------------

    "source_table",

    "source_figure",

    "source_page",

    "source_section",

    "data_extraction_method",

    "extraction_notes",

    "data_quality_flag"
]


formulations = pd.DataFrame(
    columns=formulation_columns
)


formulations_file = (
    EXTRACTION /
    "literature_formulations.csv"
)


formulations.to_csv(
    formulations_file,
    index=False
)


# ==============================================================================
# 4. RELEASE-CURVE DATABASE
# ==============================================================================

release_columns = [

    "study_id",

    "formulation_id",

    "doi",

    "drug_name",

    "polymer_name",

    "polymer_family",

    "time_h",

    "release_percent",

    "release_sd",

    "release_sem",

    "n_replicates",

    "source_figure",

    "source_table",

    "source_page",

    "data_extraction_method",

    "digitization_software",

    "digitization_notes",

    "data_quality_flag"
]


release_data = pd.DataFrame(
    columns=release_columns
)


release_file = (
    DIGITIZATION /
    "release_curves_raw.csv"
)


release_data.to_csv(
    release_file,
    index=False
)


# ==============================================================================
# 5. DATA DICTIONARY
# ==============================================================================

data_dictionary = {

    "study_id":
        "Unique identifier for each literature study.",

    "formulation_id":
        "Unique identifier for each polymer-drug formulation.",

    "doi":
        "Digital Object Identifier of the source publication.",

    "drug_name":
        "Name of the therapeutic compound.",

    "drug_smiles":
        "Canonical or isomeric SMILES obtained from a documented chemical source.",

    "drug_molecular_weight_g_mol":
        "Drug molecular weight in g/mol.",

    "polymer_name":
        "Reported polymer name.",

    "polymer_family":
        "Polymer family such as PLGA, PLA, PCL, etc.",

    "polymer_composition":
        "Copolymer composition or formulation composition as reported.",

    "polymer_molecular_weight_g_mol":
        "Polymer molecular weight as reported.",

    "polymer_tg_c":
        "Glass-transition temperature in °C.",

    "polymer_crystallinity_percent":
        "Crystallinity percentage if reported.",

    "polymer_degradation_rate":
        "Reported degradation rate or corresponding quantitative descriptor.",

    "drug_loading_percent":
        "Drug loading percentage.",

    "polymer_drug_ratio":
        "Polymer-to-drug ratio.",

    "encapsulation_efficiency_percent":
        "Encapsulation efficiency percentage.",

    "particle_size_nm":
        "Particle size in nanometers.",

    "formulation_type":
        "Microsphere, nanoparticle, film, implant, depot, etc.",

    "surface_modification":
        "Surface modification if reported.",

    "release_medium":
        "Experimental release medium.",

    "release_ph":
        "Release-test pH.",

    "release_temperature_c":
        "Release-test temperature.",

    "agitation_rpm":
        "Agitation speed if reported.",

    "release_duration_h":
        "Total release duration.",

    "time_h":
        "Release measurement time in hours.",

    "release_percent":
        "Cumulative drug release percentage.",

    "release_sd":
        "Reported standard deviation.",

    "release_sem":
        "Reported standard error of mean.",

    "n_replicates":
        "Number of experimental replicates.",

    "source_figure":
        "Figure from which data were extracted.",

    "source_table":
        "Table from which data were extracted.",

    "source_page":
        "Page containing the relevant information.",

    "data_extraction_method":
        "Direct table extraction, manually reported value, graph digitization, etc.",

    "digitization_software":
        "Software used for graph digitization.",

    "data_quality_flag":
        "PASS, REVIEW, or EXCLUDE."
}


dictionary_df = pd.DataFrame(
    [
        {
            "variable": key,
            "definition": value
        }
        for key, value in data_dictionary.items()
    ]
)


dictionary_file = (
    RAW /
    "data_dictionary.csv"
)


dictionary_df.to_csv(
    dictionary_file,
    index=False
)


# ==============================================================================
# 6. SEARCH STRATEGY FILE
# ==============================================================================

search_strategy = [

    {
        "search_id": "S001",
        "query":
        '"biodegradable polymer" AND "controlled drug release" AND PLGA',
        "database":
        "PubMed/Scopus/Web of Science",
        "status":
        "planned"
    },

    {
        "search_id": "S002",
        "query":
        '"PLGA" AND "drug release" AND nanoparticle',
        "database":
        "PubMed/Scopus/Web of Science",
        "status":
        "planned"
    },

    {
        "search_id": "S003",
        "query":
        '"PLA" AND "controlled release" AND drug',
        "database":
        "PubMed/Scopus/Web of Science",
        "status":
        "planned"
    },

    {
        "search_id": "S004",
        "query":
        '"PCL" AND "controlled drug release"',
        "database":
        "PubMed/Scopus/Web of Science",
        "status":
        "planned"
    },

    {
        "search_id": "S005",
        "query":
        '"biodegradable polymer" AND "drug release profile"',
        "database":
        "PubMed/Scopus/Web of Science",
        "status":
        "planned"
    }
]


search_df = pd.DataFrame(
    search_strategy
)


search_file = (
    RAW /
    "literature_search_strategy.csv"
)


search_df.to_csv(
    search_file,
    index=False
)


# ==============================================================================
# 7. DATA EXTRACTION TEMPLATE
# ==============================================================================

extraction_template = pd.DataFrame(
    columns=[
        "study_id",
        "doi",
        "formulation_id",
        "drug_name",
        "polymer_name",
        "polymer_family",
        "drug_loading_percent",
        "particle_size_nm",
        "polymer_molecular_weight_g_mol",
        "polymer_tg_c",
        "release_medium",
        "release_ph",
        "release_temperature_c",
        "time_h",
        "release_percent",
        "source_figure",
        "source_table",
        "data_extraction_method",
        "extraction_notes",
        "data_quality_flag"
    ]
)


template_file = (
    EXTRACTION /
    "manual_extraction_template.csv"
)


extraction_template.to_csv(
    template_file,
    index=False
)


# ==============================================================================
# 8. README
# ==============================================================================

readme = """
AI-ASSISTED INVERSE DESIGN OF BIODEGRADABLE POLYMERS
FOR CONTROLLED DRUG RELEASE

DATA PIPELINE

01. Literature identification
02. Study-level source registration
03. Formulation-level extraction
04. Release-curve extraction/digitization
05. Quality control
06. Molecular/polymer descriptor generation
07. ML-ready dataset construction
08. Nested/grouped model validation
09. Explainability
10. Uncertainty quantification
11. Multi-objective inverse design

IMPORTANT:

No experimental release values should be invented.

Every release observation must be traceable to:
- DOI
- study
- formulation
- figure/table/page
- extraction method

Graph digitization should preserve the original reported curve.

Missing variables should remain missing until the modeling pipeline
handles them using training-fold-only preprocessing.

Do not randomly duplicate observations.

Do not treat multiple time points from the same formulation as
independent formulations during train/test splitting.
"""


readme_file = (
    PROJECT /
    "README_DATA_PIPELINE.txt"
)


readme_file.write_text(
    readme,
    encoding="utf-8"
)


# ==============================================================================
# 9. FINAL OUTPUT
# ==============================================================================

print(
    "\n============================================================"
)

print(
    "CODE 01 — LITERATURE DATASET FRAMEWORK CREATED"
)

print(
    "============================================================"
)

print(
    "\nProject directory:"
)

print(
    PROJECT.resolve()
)

print(
    "\nCreated files:"
)

print(
    f"1. {studies_file}"
)

print(
    f"2. {formulations_file}"
)

print(
    f"3. {release_file}"
)

print(
    f"4. {dictionary_file}"
)

print(
    f"5. {search_file}"
)

print(
    f"6. {template_file}"
)

print(
    f"7. {readme_file}"
)

print(
    "\nNext:"
)

print(
    "Populate the literature source and formulation files "
    "with REAL published data."
)

print(
    "\n============================================================"
)
```

```python
# ==============================================================================
# CODE 02 — REAL LITERATURE DISCOVERY
# AI INVERSE DESIGN OF BIODEGRADABLE POLYMERS
# FOR CONTROLLED DRUG RELEASE
#
# Purpose:
#   Search PubMed for real literature.
#   Retrieve PMID, DOI, title, journal, year, authors and abstract.
#   Remove duplicate publications.
#
# IMPORTANT:
#   This code does NOT invent experimental data.
#   It only identifies real publications for subsequent extraction.
# ==============================================================================

import requests
import xml.etree.ElementTree as ET
import pandas as pd
import numpy as np
import re
import time
from pathlib import Path


# ==============================================================================
# 1. PROJECT PATHS
# ==============================================================================

PROJECT = Path("AI_Polymer_Drug_Release")

RAW = PROJECT / "data" / "raw_literature"

RAW.mkdir(
    parents=True,
    exist_ok=True
)


# ==============================================================================
# 2. SEARCH TERMS
# ==============================================================================

SEARCH_QUERIES = {

    "Q1":
        '("biodegradable polymers"[Title/Abstract]) '
        'AND ("controlled drug release"[Title/Abstract])',

    "Q2":
        '(PLGA[Title/Abstract]) '
        'AND ("drug release"[Title/Abstract])',

    "Q3":
        '(PLA[Title/Abstract]) '
        'AND ("controlled release"[Title/Abstract])',

    "Q4":
        '(PCL[Title/Abstract]) '
        'AND ("controlled drug release"[Title/Abstract])',

    "Q5":
        '("poly(lactic-co-glycolic acid)"[Title/Abstract]) '
        'AND ("drug release"[Title/Abstract])',

    "Q6":
        '("polycaprolactone"[Title/Abstract]) '
        'AND ("drug delivery"[Title/Abstract])',

    "Q7":
        '("biodegradable polymer"[Title/Abstract]) '
        'AND ("release kinetics"[Title/Abstract])',

    "Q8":
        '("polymeric nanoparticles"[Title/Abstract]) '
        'AND ("drug release"[Title/Abstract])',

    "Q9":
        '("polymeric microspheres"[Title/Abstract]) '
        'AND ("drug release"[Title/Abstract])',

    "Q10":
        '("polymeric microparticles"[Title/Abstract]) '
        'AND ("controlled release"[Title/Abstract])'
}


# ==============================================================================
# 3. PUBMED SEARCH FUNCTION
# ==============================================================================

def pubmed_search(
    query,
    retmax=200
):

    url = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/esearch.fcgi"

    params = {

        "db": "pubmed",

        "term": query,

        "retmode": "json",

        "retmax": retmax,

        "sort": "relevance"
    }

    response = requests.get(
        url,
        params=params,
        timeout=30
    )

    response.raise_for_status()

    data = response.json()

    return data[
        "esearchresult"
    ][
        "idlist"
    ]


# ==============================================================================
# 4. FETCH PUBMED RECORDS
# ==============================================================================

def fetch_pubmed_records(
    pmids
):

    if not pmids:
        return []

    url = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/efetch.fcgi"

    params = {

        "db": "pubmed",

        "id": ",".join(pmids),

        "retmode": "xml"
    }

    response = requests.get(
        url,
        params=params,
        timeout=60
    )

    response.raise_for_status()

    root = ET.fromstring(
        response.text
    )

    records = []

    for article in root.findall(
        ".//PubmedArticle"
    ):

        pmid_element = article.find(
            ".//PMID"
        )

        pmid = (
            pmid_element.text
            if pmid_element is not None
            else None
        )

        # ----------------------------------------------------------
        # Title
        # ----------------------------------------------------------

        title_element = article.find(
            ".//ArticleTitle"
        )

        title = (
            "".join(
                title_element.itertext()
            )
            if title_element is not None
            else None
        )

        # ----------------------------------------------------------
        # Abstract
        # ----------------------------------------------------------

        abstract_parts = []

        for abstract_text in article.findall(
            ".//Abstract/AbstractText"
        ):

            label = (
                abstract_text.attrib.get(
                    "Label",
                    ""
                )
            )

            text = "".join(
                abstract_text.itertext()
            )

            if label:

                abstract_parts.append(
                    f"{label}: {text}"
                )

            else:

                abstract_parts.append(
                    text
                )

        abstract = " ".join(
            abstract_parts
        )

        # ----------------------------------------------------------
        # Journal
        # ----------------------------------------------------------

        journal_element = article.find(
            ".//Journal/Title"
        )

        journal = (
            journal_element.text
            if journal_element is not None
            else None
        )

        # ----------------------------------------------------------
        # Publication year
        # ----------------------------------------------------------

        year = None

        for path in [
            ".//PubDate/Year",
            ".//PubDate/MedlineDate"
        ]:

            element = article.find(
                path
            )

            if element is not None:

                match = re.search(
                    r"(19|20)\d{2}",
                    element.text or ""
                )

                if match:

                    year = int(
                        match.group()
                    )

                    break

        # ----------------------------------------------------------
        # DOI
        # ----------------------------------------------------------

        doi = None

        for aid in article.findall(
            ".//ArticleId"
        ):

            if aid.attrib.get(
                "IdType"
            ) == "doi":

                doi = aid.text

                break

        # ----------------------------------------------------------
        # Authors
        # ----------------------------------------------------------

        authors = []

        for author in article.findall(
            ".//Author"
        ):

            lastname = author.find(
                "LastName"
            )

            initials = author.find(
                "Initials"
            )

            if lastname is not None:

                name = lastname.text

                if initials is not None:

                    name += " " + initials.text

                authors.append(
                    name
                )

        authors_text = "; ".join(
            authors
        )

        # ----------------------------------------------------------
        # Publication types
        # ----------------------------------------------------------

        publication_types = []

        for pt in article.findall(
            ".//PublicationType"
        ):

            if pt.text:

                publication_types.append(
                    pt.text
                )

        # ----------------------------------------------------------
        # MeSH terms
        # ----------------------------------------------------------

        mesh_terms = []

        for mesh in article.findall(
            ".//MeshHeading/DescriptorName"
        ):

            if mesh.text:

                mesh_terms.append(
                    mesh.text
                )

        # ----------------------------------------------------------
        # PMID URL
        # ----------------------------------------------------------

        pubmed_url = (
            f"https://pubmed.ncbi.nlm.nih.gov/"
            f"{pmid}/"
            if pmid
            else None
        )

        records.append({

            "pmid":
                pmid,

            "doi":
                doi,

            "title":
                title,

            "authors":
                authors_text,

            "journal":
                journal,

            "publication_year":
                year,

            "abstract":
                abstract,

            "publication_types":
                "; ".join(
                    publication_types
                ),

            "mesh_terms":
                "; ".join(
                    mesh_terms
                ),

            "pubmed_url":
                pubmed_url
        })

    return records


# ==============================================================================
# 5. RUN ALL SEARCHES
# ==============================================================================

all_records = []

search_log = []

print(
    "\nStarting PubMed literature discovery...\n"
)


for search_id, query in SEARCH_QUERIES.items():

    print(
        f"Running {search_id}: {query}"
    )

    try:

        pmids = pubmed_search(
            query,
            retmax=200
        )

        print(
            f"  Found {len(pmids)} records"
        )

        records = fetch_pubmed_records(
            pmids
        )

        for record in records:

            record[
                "search_id"
            ] = search_id

            record[
                "search_query"
            ] = query

            all_records.append(
                record
            )

        search_log.append({

            "search_id":
                search_id,

            "query":
                query,

            "records_found":
                len(records),

            "status":
                "SUCCESS"
        })

    except Exception as e:

        print(
            f"  ERROR: {e}"
        )

        search_log.append({

            "search_id":
                search_id,

            "query":
                query,

            "records_found":
                0,

            "status":
                f"ERROR: {e}"
        })

    # Respect NCBI request limits
    time.sleep(
        0.5
    )


# ==============================================================================
# 6. CREATE DATAFRAME
# ==============================================================================

literature_df = pd.DataFrame(
    all_records
)


if literature_df.empty:

    raise RuntimeError(
        """
No PubMed records were retrieved.

Check your internet connection or PubMed availability.
"""
    )


print(
    "\nTotal retrieved records:"
)

print(
    len(literature_df)
)


# ==============================================================================
# 7. REMOVE DUPLICATES BY PMID
# ==============================================================================

literature_df = (
    literature_df
    .drop_duplicates(
        subset=["pmid"]
    )
    .reset_index(
        drop=True
    )
)


# ==============================================================================
# 8. REMOVE DUPLICATES BY DOI
# ==============================================================================

literature_df[
    "doi_clean"
] = (
    literature_df[
        "doi"
    ]
    .fillna("")
    .str.lower()
    .str.strip()
)


has_doi = (
    literature_df[
        "doi_clean"
    ] != ""
)


literature_df_doi = (
    literature_df[
        has_doi
    ]
    .drop_duplicates(
        subset=["doi_clean"]
    )
)


literature_df_no_doi = (
    literature_df[
        ~has_doi
    ]
)


literature_df = pd.concat(
    [
        literature_df_doi,
        literature_df_no_doi
    ],
    ignore_index=True
)


# ==============================================================================
# 9. BASIC RELEVANCE SCREENING
# ==============================================================================

def relevance_score(row):

    text = (
        str(row["title"])
        + " "
        + str(row["abstract"])
    ).lower()

    score = 0

    # Biodegradable polymer
    polymer_terms = [

        "plga",
        "pla",
        "pcl",
        "polylactic",
        "poly(lactic",
        "polycaprolactone",
        "biodegradable polymer",
        "polymeric nanoparticle",
        "polymeric microsphere",
        "polymeric microparticle"
    ]

    # Drug release
    release_terms = [

        "drug release",
        "controlled release",
        "sustained release",
        "release kinetics",
        "drug delivery",
        "cumulative release"
    ]

    for term in polymer_terms:

        if term in text:

            score += 1

    for term in release_terms:

        if term in text:

            score += 1

    return score


literature_df[
    "relevance_score"
] = literature_df.apply(
    relevance_score,
    axis=1
)


# ==============================================================================
# 10. SCREENING DECISION
# ==============================================================================

literature_df[
    "screening_decision"
] = np.where(

    literature_df[
        "relevance_score"
    ] >= 2,

    "POTENTIALLY_RELEVANT",

    "REVIEW"
)


# ==============================================================================
# 11. ASSIGN STUDY IDs
# ==============================================================================

literature_df[
    "study_id"
] = [

    f"STUDY-{i:05d}"

    for i in range(
        1,
        len(literature_df) + 1
    )
]


# ==============================================================================
# 12. SAVE COMPLETE SEARCH DATABASE
# ==============================================================================

output_file = (
    RAW /
    "pubmed_literature_database.csv"
)


literature_df.to_csv(
    output_file,
    index=False
)


# ==============================================================================
# 13. SAVE SEARCH LOG
# ==============================================================================

search_log_df = pd.DataFrame(
    search_log
)


search_log_file = (
    RAW /
    "pubmed_search_log.csv"
)


search_log_df.to_csv(
    search_log_file,
    index=False
)


# ==============================================================================
# 14. CREATE SCREENING FILE
# ==============================================================================

screening_columns = [

    "study_id",

    "pmid",

    "doi",

    "title",

    "authors",

    "journal",

    "publication_year",

    "abstract",

    "pubmed_url",

    "relevance_score",

    "screening_decision",

    # Manual fields
    "full_text_checked",

    "biodegradable_polymer",

    "controlled_drug_release",

    "quantitative_release_data",

    "formulation_data_available",

    "include_final_dataset",

    "exclusion_reason",

    "reviewer_notes"
]


for column in screening_columns:

    if column not in literature_df.columns:

        literature_df[
            column
        ] = ""


screening_df = literature_df[
    screening_columns
].copy()


screening_file = (
    RAW /
    "literature_screening.csv"
)


screening_df.to_csv(
    screening_file,
    index=False
)


# ==============================================================================
# 15. PRINT RESULTS
# ==============================================================================

print(
    "\n============================================================"
)

print(
    "CODE 02 COMPLETE"
)

print(
    "============================================================"
)

print(
    f"\nUnique publications:"
    f" {len(literature_df)}"
)

print(
    "\nPotentially relevant:"
)

print(
    (
        literature_df[
            "screening_decision"
        ]
        == "POTENTIALLY_RELEVANT"
    ).sum()
)

print(
    "\nFiles created:"
)

print(
    f"\n1. {output_file}"
)

print(
    f"2. {search_log_file}"
)

print(
    f"3. {screening_file}"
)

print(
    "\nNext step:"
)

print(
    "Review the screening file and identify papers that contain"
)

print(
    "quantitative polymer/drug-release formulation data."
)

print(
    "\n============================================================"
)
```

```python
# ==============================================================================
# CODE 03 — AUTOMATED LITERATURE SCREENING & PRIORITIZATION
#
# AI INVERSE DESIGN OF BIODEGRADABLE POLYMERS
# FOR CONTROLLED DRUG RELEASE
#
# Input:
#   pubmed_literature_database.csv
#
# Output:
#   01_screening_all.csv
#   02_high_priority_primary_studies.csv
#   03_excluded_reviews.csv
#   04_screening_statistics.csv
#
# IMPORTANT:
#   This is a PRIORITIZATION system.
#   Final inclusion must be verified from the actual paper/full text.
# ==============================================================================


# ==============================================================================
# 1. IMPORTS
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re


# ==============================================================================
# 2. PATHS
# ==============================================================================

PROJECT = Path(
    "AI_Polymer_Drug_Release"
)

RAW = (
    PROJECT /
    "data" /
    "raw_literature"
)

SCREENING = (
    PROJECT /
    "data" /
    "screening"
)

SCREENING.mkdir(
    parents=True,
    exist_ok=True
)


INPUT_FILE = (
    RAW /
    "pubmed_literature_database.csv"
)


if not INPUT_FILE.exists():

    raise FileNotFoundError(
        f"""
Could not find:

{INPUT_FILE}

Run Code 02 first.
"""
    )


# ==============================================================================
# 3. LOAD DATABASE
# ==============================================================================

df = pd.read_csv(
    INPUT_FILE
)

print(
    "Loaded publications:",
    len(df)
)


# ==============================================================================
# 4. TEXT NORMALIZATION
# ==============================================================================

def normalize_text(value):

    if pd.isna(value):

        return ""

    value = str(value)

    value = value.lower()

    value = re.sub(
        r"\s+",
        " ",
        value
    )

    return value.strip()


df[
    "search_text"
] = (

    df["title"]
    .fillna("")
    .astype(str)

    + " "

    + df["abstract"]
    .fillna("")
    .astype(str)

    + " "

    + df["publication_types"]
    .fillna("")
    .astype(str)

).apply(
    normalize_text
)


# ==============================================================================
# 5. DEFINE RELEVANCE TERMS
# ==============================================================================

POLYMER_TERMS = [

    "plga",

    "poly(lactic-co-glycolic acid)",

    "polylactic acid",

    "poly(lactic acid)",

    "pla",

    "polycaprolactone",

    "pcl",

    "biodegradable polymer",

    "biodegradable polymers",

    "biodegradable polymeric",

    "polymeric nanoparticle",

    "polymeric nanoparticles",

    "polymeric microsphere",

    "polymeric microspheres",

    "polymeric microparticle",

    "polymeric microparticles",

    "polymer matrix",

    "polymer depot",

    "polymer implant",

    "polymer film",

    "polymer-based drug delivery",

    "polymer based drug delivery"
]


RELEASE_TERMS = [

    "drug release",

    "controlled release",

    "sustained release",

    "prolonged release",

    "extended release",

    "cumulative release",

    "release kinetics",

    "release profile",

    "release profiles",

    "in vitro release",

    "drug-releasing",

    "drug releasing"
]


QUANTITATIVE_TERMS = [

    "percent release",

    "% release",

    "cumulative release",

    "release percentage",

    "release kinetics",

    "release profile",

    "release curve",

    "release data",

    "hours",

    "days",

    "h)",

    "24 h",

    "48 h",

    "72 h",

    "168 h"
]


EXPERIMENTAL_TERMS = [

    "prepared",

    "synthesized",

    "fabricated",

    "formulated",

    "encapsulated",

    "characterized",

    "experimental",

    "in vitro",

    "incubated",

    "measured",

    "investigated"
]


# ==============================================================================
# 6. REVIEW / NON-PRIMARY STUDY TERMS
# ==============================================================================

REVIEW_TERMS = [

    "review",

    "systematic review",

    "scoping review",

    "narrative review",

    "comprehensive overview",

    "perspectives and challenges",

    "update on",

    "state of the art",

    "recent advances",

    "current perspectives",

    "overview",

    "bibliometric",

    "meta-analysis"
]


# ==============================================================================
# 7. SCORE FUNCTION
# ==============================================================================

def count_terms(
    text,
    terms
):

    return sum(
        term in text
        for term in terms
    )


def calculate_screening_score(row):

    text = row[
        "search_text"
    ]

    polymer_score = count_terms(
        text,
        POLYMER_TERMS
    )

    release_score = count_terms(
        text,
        RELEASE_TERMS
    )

    quantitative_score = count_terms(
        text,
        QUANTITATIVE_TERMS
    )

    experimental_score = count_terms(
        text,
        EXPERIMENTAL_TERMS
    )

    review_score = count_terms(
        text,
        REVIEW_TERMS
    )

    # ----------------------------------------------------------
    # Primary relevance score
    # ----------------------------------------------------------

    score = 0

    score += (
        polymer_score * 2
    )

    score += (
        release_score * 3
    )

    score += (
        quantitative_score * 2
    )

    score += (
        experimental_score
    )

    # Penalize likely reviews
    score -= (
        review_score * 4
    )

    # ----------------------------------------------------------
    # Explicit publication type penalty
    # ----------------------------------------------------------

    publication_types = normalize_text(
        row[
            "publication_types"
        ]
    )

    if "review" in publication_types:

        score -= 10

    if "meta-analysis" in publication_types:

        score -= 10

    return score


df[
    "screening_score"
] = df.apply(
    calculate_screening_score,
    axis=1
)


# ==============================================================================
# 8. INDIVIDUAL COMPONENT SCORES
# ==============================================================================

df[
    "polymer_score"
] = df[
    "search_text"
].apply(
    lambda x:
    count_terms(
        x,
        POLYMER_TERMS
    )
)


df[
    "release_score"
] = df[
    "search_text"
].apply(
    lambda x:
    count_terms(
        x,
        RELEASE_TERMS
    )
)


df[
    "quantitative_score"
] = df[
    "search_text"
].apply(
    lambda x:
    count_terms(
        x,
        QUANTITATIVE_TERMS
    )
)


df[
    "experimental_score"
] = df[
    "search_text"
].apply(
    lambda x:
    count_terms(
        x,
        EXPERIMENTAL_TERMS
    )
)


df[
    "review_score"
] = df[
    "search_text"
].apply(
    lambda x:
    count_terms(
        x,
        REVIEW_TERMS
    )
)


# ==============================================================================
# 9. STUDY TYPE CLASSIFICATION
# ==============================================================================

def classify_study(row):

    text = row[
        "search_text"
    ]

    publication_type = normalize_text(
        row[
            "publication_types"
        ]
    )

    if (
        "review" in publication_type
        or "review" in text
        or "systematic review" in text
        or "meta-analysis" in text
    ):

        return "REVIEW_OR_SYNTHESIS"

    experimental_terms_found = (
        count_terms(
            text,
            EXPERIMENTAL_TERMS
        )
    )

    quantitative_terms_found = (
        count_terms(
            text,
            QUANTITATIVE_TERMS
        )
    )

    if (
        experimental_terms_found >= 2
        and quantitative_terms_found >= 2
    ):

        return "LIKELY_PRIMARY_QUANTITATIVE"

    if (
        experimental_terms_found >= 1
    ):

        return "POSSIBLE_PRIMARY_STUDY"

    return "UNCERTAIN"


df[
    "study_type"
] = df.apply(
    classify_study,
    axis=1
)


# ==============================================================================
# 10. FINAL AUTOMATED SCREENING CATEGORY
# ==============================================================================

def screening_category(row):

    score = row[
        "screening_score"
    ]

    study_type = row[
        "study_type"
    ]

    if (
        study_type
        == "LIKELY_PRIMARY_QUANTITATIVE"
        and score >= 10
    ):

        return "HIGH_PRIORITY"

    if (
        study_type
        == "POSSIBLE_PRIMARY_STUDY"
        and score >= 8
    ):

        return "MEDIUM_PRIORITY"

    if (
        study_type
        == "REVIEW_OR_SYNTHESIS"
    ):

        return "BACKGROUND_ONLY"

    return "LOW_PRIORITY"


df[
    "automated_category"
] = df.apply(
    screening_category,
    axis=1
)


# ==============================================================================
# 11. EXCLUDE OBVIOUSLY IRRELEVANT TOPICS
# ==============================================================================

IRRELEVANT_TERMS = [

    "bone tissue engineering",
    "wound healing scaffold",
    "gene delivery",
    "mrna delivery",
    "sirna delivery",
    "dna delivery",

    # These are only excluded if no conventional drug-release
    # signal is also present.
    "vaccine delivery",
]


def potential_irrelevance(row):

    text = row[
        "search_text"
    ]

    found = [
        term
        for term in IRRELEVANT_TERMS
        if term in text
    ]

    if not found:

        return ""

    # Do not automatically exclude if explicit drug-release
    # terminology is present.
    release_hits = count_terms(
        text,
        RELEASE_TERMS
    )

    if release_hits >= 2:

        return ""

    return "; ".join(
        found
    )


df[
    "potential_irrelevance"
] = df.apply(
    potential_irrelevance,
    axis=1
)


# ==============================================================================
# 12. MANUAL SCREENING FIELDS
# ==============================================================================

manual_fields = [

    "full_text_available",

    "full_text_checked",

    "primary_experimental_study",

    "biodegradable_polymer_confirmed",

    "drug_release_experiment_confirmed",

    "quantitative_release_curve_available",

    "formulation_variables_available",

    "polymer_identity_available",

    "drug_identity_available",

    "include_final_dataset",

    "exclusion_reason",

    "reviewer_notes"
]


for column in manual_fields:

    df[
        column
    ] = ""


# ==============================================================================
# 13. SORT BY PRIORITY
# ==============================================================================

df = df.sort_values(

    by=[
        "automated_category",
        "screening_score",
        "quantitative_score",
        "publication_year"
    ],

    ascending=[
        True,
        False,
        False,
        False
    ]
).reset_index(
    drop=True
)


# ==============================================================================
# 14. SAVE COMPLETE SCREENING DATABASE
# ==============================================================================

all_screening_file = (
    SCREENING /
    "01_screening_all.csv"
)


df.to_csv(
    all_screening_file,
    index=False
)


# ==============================================================================
# 15. HIGH PRIORITY STUDIES
# ==============================================================================

high_priority = df[
    df[
        "automated_category"
    ]
    == "HIGH_PRIORITY"
].copy()


high_priority_file = (
    SCREENING /
    "02_high_priority_primary_studies.csv"
)


high_priority.to_csv(
    high_priority_file,
    index=False
)


# ==============================================================================
# 16. MEDIUM PRIORITY STUDIES
# ==============================================================================

medium_priority = df[
    df[
        "automated_category"
    ]
    == "MEDIUM_PRIORITY"
].copy()


medium_priority_file = (
    SCREENING /
    "03_medium_priority_studies.csv"
)


medium_priority.to_csv(
    medium_priority_file,
    index=False
)


# ==============================================================================
# 17. BACKGROUND / REVIEW PAPERS
# ==============================================================================

background = df[
    df[
        "automated_category"
    ]
    == "BACKGROUND_ONLY"
].copy()


background_file = (
    SCREENING /
    "04_background_reviews.csv"
)


background.to_csv(
    background_file,
    index=False
)


# ==============================================================================
# 18. STATISTICS
# ==============================================================================

statistics = pd.DataFrame({

    "category": [

        "Total publications",

        "High priority",

        "Medium priority",

        "Background/reviews",

        "Low priority",

        "Likely primary quantitative",

        "Possible primary study"
    ],

    "count": [

        len(df),

        (
            df[
                "automated_category"
            ]
            == "HIGH_PRIORITY"
        ).sum(),

        (
            df[
                "automated_category"
            ]
            == "MEDIUM_PRIORITY"
        ).sum(),

        (
            df[
                "automated_category"
            ]
            == "BACKGROUND_ONLY"
        ).sum(),

        (
            df[
                "automated_category"
            == "LOW_PRIORITY"
        ).sum(),

        (
            df[
                "study_type"
            ]
            == "LIKELY_PRIMARY_QUANTITATIVE"
        ).sum(),

        (
            df[
                "study_type"
            ]
            == "POSSIBLE_PRIMARY_STUDY"
        ).sum()
    ]
})


statistics_file = (
    SCREENING /
    "05_screening_statistics.csv"
)


statistics.to_csv(
    statistics_file,
    index=False
)


# ==============================================================================
# 19. PRINT SUMMARY
# ==============================================================================

print(
    "\n============================================================"
)

print(
    "CODE 03 COMPLETE"
)

print(
    "============================================================"
)

print(
    f"\nTotal publications: {len(df)}"
)

print(
    f"High priority: "
    f"{len(high_priority)}"
)

print(
    f"Medium priority: "
    f"{len(medium_priority)}"
)

print(
    f"Background/reviews: "
    f"{len(background)}"
)

print(
    "\nStudy-type classification:"
)

print(
    df[
        "study_type"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nFiles created:"
)

print(
    f"\n{all_screening_file}"
)

print(
    high_priority_file
)

print(
    medium_priority_file
)

print(
    background_file
)

print(
    statistics_file
)

print(
    "\n============================================================"
)
```

```python
# ==============================================================================
# CODE 03 — RECOVERY / COMPLETE SCREENING VERSION
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

# ------------------------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------------------------

PROJECT = Path("AI_Polymer_Drug_Release")

RAW = PROJECT / "data" / "raw_literature"

SCREENING = PROJECT / "data" / "screening"

SCREENING.mkdir(
    parents=True,
    exist_ok=True
)

INPUT_FILE = RAW / "pubmed_literature_database.csv"

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"\nCould not find:\n{INPUT_FILE}\n\n"
        "Run Code 02 first."
    )

# ------------------------------------------------------------------------------
# 2. LOAD THE 1,375-PAPER DATABASE
# ------------------------------------------------------------------------------

df = pd.read_csv(INPUT_FILE)

print("Loaded publications:", len(df))

# ------------------------------------------------------------------------------
# 3. NORMALIZE TEXT
# ------------------------------------------------------------------------------

def normalize_text(value):

    if pd.isna(value):
        return ""

    value = str(value).lower()

    value = re.sub(r"\s+", " ", value)

    return value.strip()


df["search_text"] = (
    df["title"].fillna("").astype(str)
    + " "
    + df["abstract"].fillna("").astype(str)
    + " "
    + df["publication_types"].fillna("").astype(str)
).apply(normalize_text)

# ------------------------------------------------------------------------------
# 4. KEYWORD GROUPS
# ------------------------------------------------------------------------------

POLYMER_TERMS = [
    "plga",
    "poly(lactic-co-glycolic acid)",
    "polylactic acid",
    "poly(lactic acid)",
    "pla",
    "polycaprolactone",
    "pcl",
    "biodegradable polymer",
    "biodegradable polymers",
    "biodegradable polymeric",
    "polymeric nanoparticle",
    "polymeric nanoparticles",
    "polymeric microsphere",
    "polymeric microspheres",
    "polymeric microparticle",
    "polymeric microparticles",
    "polymer matrix",
    "polymer depot",
    "polymer implant",
    "polymer film",
    "polymer-based drug delivery",
    "polymer based drug delivery"
]

RELEASE_TERMS = [
    "drug release",
    "controlled release",
    "sustained release",
    "prolonged release",
    "extended release",
    "cumulative release",
    "release kinetics",
    "release profile",
    "release profiles",
    "in vitro release",
    "drug-releasing",
    "drug releasing"
]

QUANTITATIVE_TERMS = [
    "percent release",
    "% release",
    "cumulative release",
    "release percentage",
    "release kinetics",
    "release profile",
    "release curve",
    "release data",
    "24 h",
    "48 h",
    "72 h",
    "168 h",
    "hours",
    "days"
]

EXPERIMENTAL_TERMS = [
    "prepared",
    "synthesized",
    "fabricated",
    "formulated",
    "encapsulated",
    "characterized",
    "experimental",
    "in vitro",
    "incubated",
    "measured",
    "investigated"
]

REVIEW_TERMS = [
    "review",
    "systematic review",
    "scoping review",
    "narrative review",
    "comprehensive overview",
    "perspectives and challenges",
    "update on",
    "state of the art",
    "recent advances",
    "current perspectives",
    "overview",
    "bibliometric",
    "meta-analysis"
]

# ------------------------------------------------------------------------------
# 5. TERM COUNTER
# ------------------------------------------------------------------------------

def count_terms(text, terms):

    return sum(
        term in text
        for term in terms
    )

# ------------------------------------------------------------------------------
# 6. COMPONENT SCORES
# ------------------------------------------------------------------------------

df["polymer_score"] = df["search_text"].apply(
    lambda x: count_terms(x, POLYMER_TERMS)
)

df["release_score"] = df["search_text"].apply(
    lambda x: count_terms(x, RELEASE_TERMS)
)

df["quantitative_score"] = df["search_text"].apply(
    lambda x: count_terms(x, QUANTITATIVE_TERMS)
)

df["experimental_score"] = df["search_text"].apply(
    lambda x: count_terms(x, EXPERIMENTAL_TERMS)
)

df["review_score"] = df["search_text"].apply(
    lambda x: count_terms(x, REVIEW_TERMS)
)

# ------------------------------------------------------------------------------
# 7. SCREENING SCORE
# ------------------------------------------------------------------------------

def calculate_screening_score(row):

    score = 0

    score += row["polymer_score"] * 2

    score += row["release_score"] * 3

    score += row["quantitative_score"] * 2

    score += row["experimental_score"]

    score -= row["review_score"] * 4

    publication_types = normalize_text(
        row["publication_types"]
    )

    if "review" in publication_types:
        score -= 10

    if "meta-analysis" in publication_types:
        score -= 10

    return score


df["screening_score"] = df.apply(
    calculate_screening_score,
    axis=1
)

# ------------------------------------------------------------------------------
# 8. STUDY TYPE
# ------------------------------------------------------------------------------

def classify_study(row):

    text = row["search_text"]

    publication_type = normalize_text(
        row["publication_types"]
    )

    # Explicit review
    if (
        "review" in publication_type
        or "review" in text
        or "systematic review" in text
        or "meta-analysis" in text
    ):
        return "REVIEW_OR_SYNTHESIS"

    experimental_hits = count_terms(
        text,
        EXPERIMENTAL_TERMS
    )

    quantitative_hits = count_terms(
        text,
        QUANTITATIVE_TERMS
    )

    if (
        experimental_hits >= 2
        and quantitative_hits >= 2
    ):
        return "LIKELY_PRIMARY_QUANTITATIVE"

    if experimental_hits >= 1:
        return "POSSIBLE_PRIMARY_STUDY"

    return "UNCERTAIN"


df["study_type"] = df.apply(
    classify_study,
    axis=1
)

# ------------------------------------------------------------------------------
# 9. AUTOMATED CATEGORY
# ------------------------------------------------------------------------------

def screening_category(row):

    score = row["screening_score"]

    study_type = row["study_type"]

    if (
        study_type == "LIKELY_PRIMARY_QUANTITATIVE"
        and score >= 10
    ):
        return "HIGH_PRIORITY"

    if (
        study_type == "POSSIBLE_PRIMARY_STUDY"
        and score >= 8
    ):
        return "MEDIUM_PRIORITY"

    if study_type == "REVIEW_OR_SYNTHESIS":
        return "BACKGROUND_ONLY"

    return "LOW_PRIORITY"


df["automated_category"] = df.apply(
    screening_category,
    axis=1
)

# ------------------------------------------------------------------------------
# 10. MANUAL SCREENING FIELDS
# ------------------------------------------------------------------------------

manual_fields = [
    "full_text_available",
    "full_text_checked",
    "primary_experimental_study",
    "biodegradable_polymer_confirmed",
    "drug_release_experiment_confirmed",
    "quantitative_release_curve_available",
    "formulation_variables_available",
    "polymer_identity_available",
    "drug_identity_available",
    "include_final_dataset",
    "exclusion_reason",
    "reviewer_notes"
]

for column in manual_fields:
    df[column] = ""

# ------------------------------------------------------------------------------
# 11. SORT
# ------------------------------------------------------------------------------

df = df.sort_values(
    by=[
        "screening_score",
        "quantitative_score",
        "release_score",
        "publication_year"
    ],
    ascending=[
        False,
        False,
        False,
        False
    ]
).reset_index(drop=True)

# ------------------------------------------------------------------------------
# 12. SAVE ALL SCREENING
# ------------------------------------------------------------------------------

all_screening_file = (
    SCREENING /
    "01_screening_all.csv"
)

df.to_csv(
    all_screening_file,
    index=False
)

# ------------------------------------------------------------------------------
# 13. HIGH PRIORITY
# ------------------------------------------------------------------------------

high_priority = df[
    df["automated_category"] == "HIGH_PRIORITY"
].copy()

high_priority_file = (
    SCREENING /
    "02_high_priority_primary_studies.csv"
)

high_priority.to_csv(
    high_priority_file,
    index=False
)

# ------------------------------------------------------------------------------
# 14. MEDIUM PRIORITY
# ------------------------------------------------------------------------------

medium_priority = df[
    df["automated_category"] == "MEDIUM_PRIORITY"
].copy()

medium_priority_file = (
    SCREENING /
    "03_medium_priority_studies.csv"
)

medium_priority.to_csv(
    medium_priority_file,
    index=False
)

# ------------------------------------------------------------------------------
# 15. REVIEWS / BACKGROUND
# ------------------------------------------------------------------------------

background = df[
    df["automated_category"] == "BACKGROUND_ONLY"
].copy()

background_file = (
    SCREENING /
    "04_background_reviews.csv"
)

background.to_csv(
    background_file,
    index=False
)

# ------------------------------------------------------------------------------
# 16. STATISTICS — CORRECTED
# ------------------------------------------------------------------------------

category_counts = (
    df["automated_category"]
    .value_counts()
)

study_type_counts = (
    df["study_type"]
    .value_counts()
)

statistics = pd.DataFrame({
    "category": [
        "Total publications",
        "High priority",
        "Medium priority",
        "Background/reviews",
        "Low priority",
        "Likely primary quantitative",
        "Possible primary study"
    ],

    "count": [
        len(df),

        category_counts.get(
            "HIGH_PRIORITY",
            0
        ),

        category_counts.get(
            "MEDIUM_PRIORITY",
            0
        ),

        category_counts.get(
            "BACKGROUND_ONLY",
            0
        ),

        category_counts.get(
            "LOW_PRIORITY",
            0
        ),

        study_type_counts.get(
            "LIKELY_PRIMARY_QUANTITATIVE",
            0
        ),

        study_type_counts.get(
            "POSSIBLE_PRIMARY_STUDY",
            0
        )
    ]
})

statistics_file = (
    SCREENING /
    "05_screening_statistics.csv"
)

statistics.to_csv(
    statistics_file,
    index=False
)

# ------------------------------------------------------------------------------
# 17. FINAL OUTPUT
# ------------------------------------------------------------------------------

print(
    "\n============================================================"
)

print(
    "CODE 03 COMPLETED SUCCESSFULLY"
)

print(
    "============================================================"
)

print(
    f"\nTotal publications: {len(df)}"
)

print(
    f"High priority: "
    f"{category_counts.get('HIGH_PRIORITY', 0)}"
)

print(
    f"Medium priority: "
    f"{category_counts.get('MEDIUM_PRIORITY', 0)}"
)

print(
    f"Background/reviews: "
    f"{category_counts.get('BACKGROUND_ONLY', 0)}"
)

print(
    f"Low priority: "
    f"{category_counts.get('LOW_PRIORITY', 0)}"
)

print(
    "\nStudy-type classification:"
)

print(
    df["study_type"]
    .value_counts()
    .to_string()
)

print(
    "\nFiles created:"
)

print(
    all_screening_file
)

print(
    high_priority_file
)

print(
    medium_priority_file
)

print(
    background_file
)

print(
    statistics_file
)

print(
    "\n============================================================"
)
```

```python
# ==============================================================================
# CODE 04A — HIGH-PRIORITY PAPER SCOPE FILTER
# ==============================================================================

import pandas as pd
import re
from pathlib import Path

INPUT = Path(
    "AI_Polymer_Drug_Release/data/screening/"
    "02_high_priority_primary_studies.csv"
)

OUTPUT_DIR = Path(
    "AI_Polymer_Drug_Release/data/screening"
)

df = pd.read_csv(INPUT)

# ------------------------------------------------------------------------------
# Normalize text
# ------------------------------------------------------------------------------

def clean(x):
    if pd.isna(x):
        return ""
    return str(x).lower()

df["text"] = (
    df["title"].apply(clean)
    + " "
    + df["abstract"].apply(clean)
)

# ------------------------------------------------------------------------------
# Target polymer classes
# ------------------------------------------------------------------------------

POLYMERS = [
    "plga",
    "poly(lactic-co-glycolic acid)",
    "poly(lactic acid)",
    "polylactic acid",
    "pla",
    "polycaprolactone",
    "pcl",
    "poly(propylene fumarate)",
    "polymeric"
]

# ------------------------------------------------------------------------------
# Drug / therapeutic cargo
# ------------------------------------------------------------------------------

SMALL_MOLECULE_TERMS = [
    "dexamethasone",
    "paclitaxel",
    "curcumin",
    "ibuprofen",
    "indomethacin",
    "clindamycin",
    "donepezil",
    "gefitinib",
    "amoxicillin",
    "olanzapine",
    "cloxacillin",
    "ibrutinib",
    "sirolimus",
    "tacrolimus",
    "5-fluorouracil",
    "furosemide",
    "drug-loaded",
    "drug release",
    "small molecule",
    "therapeutic agent"
]

# ------------------------------------------------------------------------------
# Exclusion categories
# ------------------------------------------------------------------------------

GENETIC_TERMS = [
    "sirna",
    "mrna",
    "mirna",
    "oligonucleotide",
    "dna delivery",
    "gene delivery",
    "gene therapy",
    "nucleic acid"
]

PROTEIN_TERMS = [
    "protein drug",
    "protein delivery",
    "protein encapsulation",
    "peptide delivery",
    "peptide drug"
]

TISSUE_ENGINEERING_TERMS = [
    "tissue engineering",
    "bone tissue engineering",
    "scaffold",
    "regeneration scaffold",
    "bioprinted scaffold"
]

# ------------------------------------------------------------------------------
# Classification
# ------------------------------------------------------------------------------

def classify(row):

    text = row["text"]

    polymer_hits = [
        term for term in POLYMERS
        if term in text
    ]

    drug_hits = [
        term for term in SMALL_MOLECULE_TERMS
        if term in text
    ]

    genetic_hits = [
        term for term in GENETIC_TERMS
        if term in text
    ]

    protein_hits = [
        term for term in PROTEIN_TERMS
        if term in text
    ]

    tissue_hits = [
        term for term in TISSUE_ENGINEERING_TERMS
        if term in text
    ]

    # ----------------------------------------------------------
    # Strong exclusion
    # ----------------------------------------------------------

    if genetic_hits and not drug_hits:
        return "EXCLUDE_GENETIC_CARGO"

    if tissue_hits and not drug_hits:
        return "EXCLUDE_TISSUE_ENGINEERING"

    if protein_hits and not drug_hits:
        return "SECONDARY_PROTEIN_OR_PEPTIDE"

    # ----------------------------------------------------------
    # Core candidate
    # ----------------------------------------------------------

    if (
        len(polymer_hits) >= 1
        and len(drug_hits) >= 1
    ):
        return "CORE_CANDIDATE"

    # ----------------------------------------------------------
    # Secondary
    # ----------------------------------------------------------

    if len(polymer_hits) >= 1:
        return "SECONDARY_REQUIRES_FULL_TEXT"

    return "REQUIRES_MANUAL_REVIEW"


df["scope_category"] = df.apply(
    classify,
    axis=1
)

# ------------------------------------------------------------------------------
# Evidence indicators
# ------------------------------------------------------------------------------

RELEASE_TERMS = [
    "release kinetics",
    "release profile",
    "cumulative release",
    "controlled release",
    "sustained release",
    "in vitro release",
    "drug release"
]

FORMULATION_TERMS = [
    "drug loading",
    "encapsulation efficiency",
    "particle size",
    "microsphere",
    "nanoparticle",
    "polymer molecular weight",
    "polymer ratio",
    "copolymer ratio",
    "polymer composition"
]

df["release_evidence"] = df["text"].apply(
    lambda x: sum(
        term in x
        for term in RELEASE_TERMS
    )
)

df["formulation_evidence"] = df["text"].apply(
    lambda x: sum(
        term in x
        for term in FORMULATION_TERMS
    )
)

# ------------------------------------------------------------------------------
# Priority score
# ------------------------------------------------------------------------------

df["data_priority_score"] = (
    df["release_evidence"] * 3
    + df["formulation_evidence"] * 2
    + df["screening_score"]
)

# ------------------------------------------------------------------------------
# Sort
# ------------------------------------------------------------------------------

df = df.sort_values(
    "data_priority_score",
    ascending=False
).reset_index(drop=True)

# ------------------------------------------------------------------------------
# Save complete scope classification
# ------------------------------------------------------------------------------

all_file = (
    OUTPUT_DIR /
    "04A_high_priority_scope_classification.csv"
)

df.to_csv(
    all_file,
    index=False
)

# ------------------------------------------------------------------------------
# Save core candidates
# ------------------------------------------------------------------------------

core = df[
    df["scope_category"] == "CORE_CANDIDATE"
].copy()

core_file = (
    OUTPUT_DIR /
    "04A_core_candidates.csv"
)

core.to_csv(
    core_file,
    index=False
)

# ------------------------------------------------------------------------------
# Save secondary candidates
# ------------------------------------------------------------------------------

secondary = df[
    df["scope_category"].isin([
        "SECONDARY_REQUIRES_FULL_TEXT",
        "SECONDARY_PROTEIN_OR_PEPTIDE",
        "REQUIRES_MANUAL_REVIEW"
    ])
].copy()

secondary_file = (
    OUTPUT_DIR /
    "04A_secondary_candidates.csv"
)

secondary.to_csv(
    secondary_file,
    index=False
)

# ------------------------------------------------------------------------------
# Output
# ------------------------------------------------------------------------------

print("=" * 70)
print("CODE 04A COMPLETE")
print("=" * 70)

print("\nTotal high-priority papers:", len(df))

print("\nScope classification:")
print(
    df["scope_category"]
    .value_counts()
    .to_string()
)

print("\nCore candidates:", len(core))

print("\nTop 30 core candidates:")

print(
    core[
        [
            "study_id",
            "pmid",
            "doi",
            "title",
            "journal",
            "publication_year",
            "release_evidence",
            "formulation_evidence",
            "data_priority_score"
        ]
    ]
    .head(30)
    .to_string(index=False)
)

print("\nFiles created:")
print(all_file)
print(core_file)
print(secondary_file)
```

```python
# ==============================================================================
# CODE 04B — FULL-TEXT AVAILABILITY + QUANTITATIVE DATA SCREENING
# ==============================================================================
#
# Purpose:
#   1. Load the 89 core candidates
#   2. Query Europe PMC using PMID/DOI
#   3. Determine full-text availability
#   4. Download openly accessible full text when available
#   5. Search full text for quantitative release/formulation evidence
#   6. Create a ranked full-text extraction queue
#
# IMPORTANT:
#   Missing numerical values are NOT fabricated.
# ==============================================================================

import os
import re
import time
import requests
import pandas as pd
import numpy as np

from pathlib import Path
from xml.etree import ElementTree as ET

# ------------------------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------------------------

PROJECT = Path("AI_Polymer_Drug_Release")

INPUT_FILE = (
    PROJECT /
    "data" /
    "screening" /
    "04A_core_candidates.csv"
)

OUTPUT_DIR = (
    PROJECT /
    "data" /
    "fulltext"
)

TEXT_DIR = OUTPUT_DIR / "texts"

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

TEXT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

if not INPUT_FILE.exists():

    raise FileNotFoundError(
        f"\nCould not find:\n{INPUT_FILE}\n\n"
        "Run Code 04A first."
    )

# ------------------------------------------------------------------------------
# 2. LOAD CORE CANDIDATES
# ------------------------------------------------------------------------------

df = pd.read_csv(INPUT_FILE)

print("=" * 75)
print("CODE 04B — FULL-TEXT AVAILABILITY SCREENING")
print("=" * 75)

print(
    "\nCore candidates loaded:",
    len(df)
)

# ------------------------------------------------------------------------------
# 3. EUROPE PMC API
# ------------------------------------------------------------------------------

SEARCH_URL = (
    "https://www.ebi.ac.uk/europepmc/webservices/rest/search"
)

FULLTEXT_URL = (
    "https://www.ebi.ac.uk/europepmc/webservices/rest"
)

session = requests.Session()

session.headers.update({
    "User-Agent":
    "AI-Polymer-Drug-Release-Research/1.0"
})

# ------------------------------------------------------------------------------
# 4. SAFE TEXT FUNCTION
# ------------------------------------------------------------------------------

def clean_text(x):

    if pd.isna(x):
        return ""

    return str(x)

# ------------------------------------------------------------------------------
# 5. EUROPE PMC SEARCH
# ------------------------------------------------------------------------------

def search_europe_pmc(pmid=None, doi=None):

    queries = []

    if pd.notna(pmid) and str(pmid).strip():

        queries.append(
            f"EXT_ID:{str(pmid).strip()}"
        )

    if pd.notna(doi) and str(doi).strip():

        doi_clean = str(doi).strip()

        queries.append(
            f'DOI:"{doi_clean}"'
        )

    for query in queries:

        params = {
            "query": query,
            "format": "json",
            "pageSize": 5
        }

        try:

            response = session.get(
                SEARCH_URL,
                params=params,
                timeout=20
            )

            if response.status_code != 200:
                continue

            data = response.json()

            results = data.get(
                "resultList",
                {}
            ).get(
                "result",
                []
            )

            if results:
                return results[0]

        except Exception as e:

            continue

    return None

# ------------------------------------------------------------------------------
# 6. FULL-TEXT RETRIEVAL
# ------------------------------------------------------------------------------

def get_fulltext_xml(pmcid):

    if not pmcid:
        return None

    url = (
        f"{FULLTEXT_URL}/"
        f"{pmcid}/fullTextXML"
    )

    try:

        response = session.get(
            url,
            timeout=30
        )

        if response.status_code == 200:

            text = response.text

            if "<article" in text:

                return text

    except Exception:

        pass

    return None

# ------------------------------------------------------------------------------
# 7. XML → PLAIN TEXT
# ------------------------------------------------------------------------------

def xml_to_text(xml_text):

    try:

        root = ET.fromstring(
            xml_text
        )

        pieces = []

        for element in root.iter():

            if element.text:

                text = element.text.strip()

                if text:
                    pieces.append(text)

        text = " ".join(pieces)

        text = re.sub(
            r"\s+",
            " ",
            text
        )

        return text.strip()

    except Exception:

        # fallback
        text = re.sub(
            r"<[^>]+>",
            " ",
            xml_text
        )

        text = re.sub(
            r"\s+",
            " ",
            text
        )

        return text.strip()

# ------------------------------------------------------------------------------
# 8. QUANTITATIVE SEARCH TERMS
# ------------------------------------------------------------------------------

RELEASE_TERMS = [

    "cumulative release",
    "drug release",
    "release kinetics",
    "release profile",
    "release profiles",
    "controlled release",
    "sustained release",
    "in vitro release",
    "percent release",
    "% release",
    "released",
    "release rate"
]

FORMULATION_TERMS = [

    "drug loading",
    "drug load",
    "encapsulation efficiency",
    "particle size",
    "particle diameter",
    "mean diameter",
    "polymer molecular weight",
    "molecular weight",
    "polymer ratio",
    "polymer composition",
    "polymer concentration",
    "drug concentration",
    "drug/polymer ratio",
    "drug polymer ratio"
]

TIME_TERMS = [

    "1 h",
    "2 h",
    "4 h",
    "6 h",
    "8 h",
    "12 h",
    "24 h",
    "48 h",
    "72 h",
    "96 h",
    "120 h",
    "168 h",
    "7 days",
    "14 days",
    "21 days",
    "28 days",
    "days"
]

NUMERIC_PATTERN = re.compile(
    r"""
    (?:
        \d+(?:\.\d+)?
        |
        \.\d+
    )
    \s*
    (?:
        %
        |
        mg
        |
        µg
        |
        ug
        |
        nm
        |
        µm
        |
        kDa
        |
        Da
        |
        h
        |
        hr
        |
        hours?
        |
        days?
    )
    """,
    re.IGNORECASE |
    re.VERBOSE
)

# ------------------------------------------------------------------------------
# 9. COUNT EVIDENCE
# ------------------------------------------------------------------------------

def count_terms(text, terms):

    text = text.lower()

    return sum(
        term.lower() in text
        for term in terms
    )

# ------------------------------------------------------------------------------
# 10. EXTRACT NUMERICAL SNIPPETS
# ------------------------------------------------------------------------------

def numerical_snippets(text):

    snippets = []

    matches = list(
        NUMERIC_PATTERN.finditer(
            text
        )
    )

    for match in matches[:100]:

        start = max(
            0,
            match.start() - 100
        )

        end = min(
            len(text),
            match.end() + 100
        )

        snippets.append(
            text[start:end]
        )

    return snippets

# ------------------------------------------------------------------------------
# 11. PROCESS PAPERS
# ------------------------------------------------------------------------------

records = []

for i, row in df.iterrows():

    study_id = clean_text(
        row.get("study_id")
    )

    pmid = clean_text(
        row.get("pmid")
    )

    doi = clean_text(
        row.get("doi")
    )

    title = clean_text(
        row.get("title")
    )

    print(
        f"[{i+1:03d}/{len(df)}] "
        f"{study_id} | PMID={pmid}"
    )

    result = search_europe_pmc(
        pmid=pmid,
        doi=doi
    )

    pmc_id = None
    fulltext_available = False
    fulltext_text = ""

    europepmc_id = None

    if result:

        europepmc_id = result.get(
            "id"
        )

        pmc_id = result.get(
            "pmcid"
        )

        # ----------------------------------------------------------
        # Retrieve public full text
        # ----------------------------------------------------------

        if pmc_id:

            xml = get_fulltext_xml(
                pmc_id
            )

            if xml:

                fulltext_available = True

                fulltext_text = xml_to_text(
                    xml
                )

                # Save raw XML
                xml_file = (
                    TEXT_DIR /
                    f"{study_id}.xml"
                )

                with open(
                    xml_file,
                    "w",
                    encoding="utf-8"
                ) as f:

                    f.write(xml)

                # Save extracted text
                txt_file = (
                    TEXT_DIR /
                    f"{study_id}.txt"
                )

                with open(
                    txt_file,
                    "w",
                    encoding="utf-8"
                ) as f:

                    f.write(fulltext_text)

    # --------------------------------------------------------------
    # Evidence analysis
    # --------------------------------------------------------------

    release_hits = count_terms(
        fulltext_text,
        RELEASE_TERMS
    )

    formulation_hits = count_terms(
        fulltext_text,
        FORMULATION_TERMS
    )

    time_hits = count_terms(
        fulltext_text,
        TIME_TERMS
    )

    numerical_matches = NUMERIC_PATTERN.findall(
        fulltext_text
    )

    numerical_count = len(
        numerical_matches
    )

    snippets = numerical_snippets(
        fulltext_text
    )

    # --------------------------------------------------------------
    # Quantitative-data score
    # --------------------------------------------------------------

    quantitative_score = (
        release_hits * 3
        + formulation_hits * 2
        + min(time_hits, 10)
        + min(numerical_count / 10, 10)
    )

    # --------------------------------------------------------------
    # Availability category
    # --------------------------------------------------------------

    if not result:

        availability = (
            "NOT_FOUND_IN_EUROPE_PMC"
        )

    elif fulltext_available:

        if (
            release_hits >= 2
            and formulation_hits >= 2
            and numerical_count >= 10
        ):

            availability = (
                "FULLTEXT_HIGH_QUANTITATIVE"
            )

        elif (
            release_hits >= 1
            and numerical_count >= 5
        ):

            availability = (
                "FULLTEXT_POSSIBLE_QUANTITATIVE"
            )

        else:

            availability = (
                "FULLTEXT_LOW_QUANTITATIVE"
            )

    else:

        availability = (
            "METADATA_ONLY_OR_PAYWALLED"
        )

    records.append({

        "study_id": study_id,

        "pmid": pmid,

        "doi": doi,

        "title": title,

        "europepmc_id":
            europepmc_id,

        "pmcid":
            pmc_id,

        "fulltext_available":
            fulltext_available,

        "availability_category":
            availability,

        "release_term_hits":
            release_hits,

        "formulation_term_hits":
            formulation_hits,

        "time_term_hits":
            time_hits,

        "numeric_measurement_count":
            numerical_count,

        "quantitative_score":
            round(
                quantitative_score,
                2
            ),

        "text_file":
            (
                f"{study_id}.txt"
                if fulltext_available
                else ""
            )

    })

    # Be polite to public API
    time.sleep(0.15)

# ------------------------------------------------------------------------------
# 12. CREATE RESULTS TABLE
# ------------------------------------------------------------------------------

availability_df = pd.DataFrame(
    records
)

# ------------------------------------------------------------------------------
# 13. MERGE WITH ORIGINAL DATA
# ------------------------------------------------------------------------------

merged = df.merge(
    availability_df[
        [
            "study_id",
            "europepmc_id",
            "pmcid",
            "fulltext_available",
            "availability_category",
            "release_term_hits",
            "formulation_term_hits",
            "time_term_hits",
            "numeric_measurement_count",
            "quantitative_score",
            "text_file"
        ]
    ],
    on="study_id",
    how="left"
)

# ------------------------------------------------------------------------------
# 14. RANK
# ------------------------------------------------------------------------------

merged = merged.sort_values(
    by=[
        "fulltext_available",
        "quantitative_score",
        "release_term_hits",
        "formulation_term_hits"
    ],
    ascending=[
        False,
        False,
        False,
        False
    ]
).reset_index(drop=True)

# ------------------------------------------------------------------------------
# 15. SAVE ALL
# ------------------------------------------------------------------------------

all_output = (
    OUTPUT_DIR /
    "04B_fulltext_screening_all.csv"
)

merged.to_csv(
    all_output,
    index=False
)

# ------------------------------------------------------------------------------
# 16. HIGH-QUANTITATIVE QUEUE
# ------------------------------------------------------------------------------

high_quant = merged[
    merged["availability_category"]
    == "FULLTEXT_HIGH_QUANTITATIVE"
].copy()

high_output = (
    OUTPUT_DIR /
    "04B_high_quantitative_extraction_queue.csv"
)

high_quant.to_csv(
    high_output,
    index=False
)

# ------------------------------------------------------------------------------
# 17. POSSIBLE QUANTITATIVE QUEUE
# ------------------------------------------------------------------------------

possible_quant = merged[
    merged["availability_category"]
    == "FULLTEXT_POSSIBLE_QUANTITATIVE"
].copy()

possible_output = (
    OUTPUT_DIR /
    "04B_possible_quantitative_extraction_queue.csv"
)

possible_quant.to_csv(
    possible_output,
    index=False
)

# ------------------------------------------------------------------------------
# 18. SUMMARY
# ------------------------------------------------------------------------------

print("\n")
print("=" * 75)
print("CODE 04B COMPLETE")
print("=" * 75)

print(
    "\nTotal papers processed:",
    len(merged)
)

print(
    "\nAvailability classification:"
)

print(
    merged[
        "availability_category"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nPublic full texts obtained:",
    int(
        merged[
            "fulltext_available"
        ].sum()
    )
)

print(
    "\nHigh quantitative candidates:",
    len(high_quant)
)

print(
    "Possible quantitative candidates:",
    len(possible_quant)
)

print(
    "\nTop quantitative papers:"
)

display_cols = [
    "study_id",
    "pmid",
    "title",
    "pmcid",
    "fulltext_available",
    "release_term_hits",
    "formulation_term_hits",
    "numeric_measurement_count",
    "quantitative_score"
]

print(
    merged[
        display_cols
    ]
    .head(30)
    .to_string(index=False)
)

print(
    "\nFiles created:"
)

print(
    all_output
)

print(
    high_output
)

print(
    possible_output
)

print(
    "\nFull-text files:",
    TEXT_DIR
)

print(
    "=" * 75
)
```

```python
# ==============================================================================
# CODE 04B — FIX FINAL MERGE / SORT / SAVE
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT = Path("AI_Polymer_Drug_Release")

INPUT_FILE = (
    PROJECT /
    "data" /
    "screening" /
    "04A_core_candidates.csv"
)

OUTPUT_DIR = (
    PROJECT /
    "data" /
    "fulltext"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------------------------
# 1. RELOAD ORIGINAL 89 CORE CANDIDATES
# ------------------------------------------------------------------------------

df = pd.read_csv(INPUT_FILE)

print("Core candidates:", len(df))

# ------------------------------------------------------------------------------
# 2. CHECK WHETHER AVAILABILITY RESULTS EXIST
# ------------------------------------------------------------------------------

if "availability_df" not in globals():

    raise RuntimeError(
        "availability_df is not present in memory. "
        "Run Code 04B again from the beginning."
    )

print(
    "Full-text screening records:",
    len(availability_df)
)

# ------------------------------------------------------------------------------
# 3. REMOVE POSSIBLE DUPLICATE COLUMNS BEFORE MERGING
# ------------------------------------------------------------------------------

# Only take the fields actually generated by the full-text screening.

availability_cols = [
    "study_id",
    "europepmc_id",
    "pmcid",
    "fulltext_available",
    "availability_category",
    "release_term_hits",
    "formulation_term_hits",
    "time_term_hits",
    "numeric_measurement_count",
    "quantitative_score",
    "text_file"
]

# Keep only columns that actually exist
availability_cols = [
    c for c in availability_cols
    if c in availability_df.columns
]

availability_clean = (
    availability_df[
        availability_cols
    ]
    .drop_duplicates(
        subset="study_id"
    )
    .copy()
)

# ------------------------------------------------------------------------------
# 4. MERGE
# ------------------------------------------------------------------------------

merged = df.merge(
    availability_clean,
    on="study_id",
    how="left",
    suffixes=("", "_fulltext")
)

# ------------------------------------------------------------------------------
# 5. FIX QUANTITATIVE SCORE
# ------------------------------------------------------------------------------

# If a collision occurred, prefer the full-text quantitative score.

if (
    "quantitative_score_fulltext"
    in merged.columns
):

    merged["quantitative_score"] = (
        merged["quantitative_score_fulltext"]
    )

    merged.drop(
        columns=[
            "quantitative_score_fulltext"
        ],
        inplace=True
    )

elif "quantitative_score" not in merged.columns:

    merged["quantitative_score"] = 0.0

# ------------------------------------------------------------------------------
# 6. CONVERT NUMERIC FIELDS
# ------------------------------------------------------------------------------

numeric_columns = [
    "quantitative_score",
    "release_term_hits",
    "formulation_term_hits",
    "time_term_hits",
    "numeric_measurement_count"
]

for col in numeric_columns:

    if col in merged.columns:

        merged[col] = pd.to_numeric(
            merged[col],
            errors="coerce"
        ).fillna(0)

# ------------------------------------------------------------------------------
# 7. FULL-TEXT BOOLEAN
# ------------------------------------------------------------------------------

if "fulltext_available" in merged.columns:

    merged["fulltext_available"] = (
        merged["fulltext_available"]
        .fillna(False)
        .astype(bool)
    )

# ------------------------------------------------------------------------------
# 8. SORT — CORRECTED
# ------------------------------------------------------------------------------

sort_columns = [
    c for c in [
        "fulltext_available",
        "quantitative_score",
        "release_term_hits",
        "formulation_term_hits",
        "numeric_measurement_count"
    ]
    if c in merged.columns
]

merged = merged.sort_values(
    by=sort_columns,
    ascending=[
        False,
        False,
        False,
        False,
        False
    ][:len(sort_columns)]
).reset_index(
    drop=True
)

# ------------------------------------------------------------------------------
# 9. SAVE COMPLETE RESULTS
# ------------------------------------------------------------------------------

all_output = (
    OUTPUT_DIR /
    "04B_fulltext_screening_all.csv"
)

merged.to_csv(
    all_output,
    index=False
)

# ------------------------------------------------------------------------------
# 10. HIGH-QUANTITATIVE EXTRACTION QUEUE
# ------------------------------------------------------------------------------

high_quant = merged[
    merged["availability_category"]
    == "FULLTEXT_HIGH_QUANTITATIVE"
].copy()

high_output = (
    OUTPUT_DIR /
    "04B_high_quantitative_extraction_queue.csv"
)

high_quant.to_csv(
    high_output,
    index=False
)

# ------------------------------------------------------------------------------
# 11. POSSIBLE-QUANTITATIVE QUEUE
# ------------------------------------------------------------------------------

possible_quant = merged[
    merged["availability_category"]
    == "FULLTEXT_POSSIBLE_QUANTITATIVE"
].copy()

possible_output = (
    OUTPUT_DIR /
    "04B_possible_quantitative_extraction_queue.csv"
)

possible_quant.to_csv(
    possible_output,
    index=False
)

# ------------------------------------------------------------------------------
# 12. LOW-QUANTITATIVE FULL TEXT
# ------------------------------------------------------------------------------

low_quant = merged[
    merged["availability_category"]
    == "FULLTEXT_LOW_QUANTITATIVE"
].copy()

low_output = (
    OUTPUT_DIR /
    "04B_fulltext_low_quantitative.csv"
)

low_quant.to_csv(
    low_output,
    index=False
)

# ------------------------------------------------------------------------------
# 13. SUMMARY
# ------------------------------------------------------------------------------

print()
print("=" * 75)
print("CODE 04B FINISHED SUCCESSFULLY")
print("=" * 75)

print(
    "\nTotal papers processed:",
    len(merged)
)

print(
    "\nAvailability classification:"
)

print(
    merged[
        "availability_category"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nPublic full texts obtained:",
    int(
        merged[
            "fulltext_available"
        ].sum()
    )
)

print(
    "\nHigh quantitative candidates:",
    len(high_quant)
)

print(
    "Possible quantitative candidates:",
    len(possible_quant)
)

print(
    "Low quantitative full texts:",
    len(low_quant)
)

# ------------------------------------------------------------------------------
# 14. TOP PAPERS
# ------------------------------------------------------------------------------

display_cols = [
    "study_id",
    "pmid",
    "doi",
    "title",
    "pmcid",
    "fulltext_available",
    "release_term_hits",
    "formulation_term_hits",
    "time_term_hits",
    "numeric_measurement_count",
    "quantitative_score"
]

display_cols = [
    c for c in display_cols
    if c in merged.columns
]

print(
    "\nTOP 30 FULL-TEXT QUANTITATIVE CANDIDATES:"
)

print(
    merged[
        display_cols
    ]
    .head(30)
    .to_string(index=False)
)

# ------------------------------------------------------------------------------
# 15. FILES
# ------------------------------------------------------------------------------

print(
    "\nFiles created:"
)

print(
    all_output
)

print(
    high_output
)

print(
    possible_output
)

print(
    low_output
)

print(
    "\nFull-text directory:"
)

print(
    PROJECT /
    "data" /
    "fulltext" /
    "texts"
)

print(
    "=" * 75
)
```

```python
# ==============================================================================
# CODE 05 — STRUCTURED QUANTITATIVE EXTRACTION FROM FULL-TEXT PAPERS
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re
import json
import warnings

warnings.filterwarnings("ignore")

# ------------------------------------------------------------------------------
# PATHS
# ------------------------------------------------------------------------------

BASE_DIR = Path("AI_Polymer_Drug_Release")

SCREEN_FILE = (
    BASE_DIR /
    "data/fulltext/04B_high_quantitative_extraction_queue.csv"
)

TEXT_DIR = (
    BASE_DIR /
    "data/fulltext/texts"
)

OUTPUT_DIR = (
    BASE_DIR /
    "data/extraction"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 78)
print("CODE 05 — STRUCTURED QUANTITATIVE EXTRACTION")
print("=" * 78)

# ------------------------------------------------------------------------------
# LOAD SCREENING QUEUE
# ------------------------------------------------------------------------------

queue = pd.read_csv(SCREEN_FILE)

print(f"\nHigh-quantitative papers loaded: {len(queue)}")

print("\nColumns:")
print(queue.columns.tolist())

# ------------------------------------------------------------------------------
# HELPER FUNCTIONS
# ------------------------------------------------------------------------------

def safe_text(value):
    if pd.isna(value):
        return ""
    return str(value)


def normalize_whitespace(text):
    text = safe_text(text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()


def extract_first_number(pattern, text, flags=re.I):
    """
    Return first numerical value matching regex.
    """
    match = re.search(pattern, text, flags)
    if match:
        try:
            return float(match.group(1))
        except Exception:
            return np.nan
    return np.nan


def extract_numbers(text):
    """
    Extract numerical values from text.
    """
    vals = re.findall(
        r"(?<![A-Za-z])[-+]?\d+(?:\.\d+)?",
        text
    )

    output = []

    for v in vals:
        try:
            output.append(float(v))
        except Exception:
            pass

    return output


# ------------------------------------------------------------------------------
# TEXT FILE LOCATOR
# ------------------------------------------------------------------------------

def locate_text_file(study_id, pmcid=None):

    possible_names = []

    if pmcid and not pd.isna(pmcid):
        possible_names.extend([
            f"{pmcid}.txt",
            f"{pmcid}.xml",
            f"{pmcid}.html",
        ])

    possible_names.extend([
        f"{study_id}.txt",
        f"{study_id}.xml",
        f"{study_id}.html",
    ])

    for name in possible_names:

        candidate = TEXT_DIR / name

        if candidate.exists():
            return candidate

    # recursive search
    for candidate in TEXT_DIR.rglob("*"):

        if not candidate.is_file():
            continue

        if study_id in candidate.name:
            return candidate

        if pmcid and not pd.isna(pmcid):
            if str(pmcid) in candidate.name:
                return candidate

    return None


# ------------------------------------------------------------------------------
# KEYWORD WINDOWS
# ------------------------------------------------------------------------------

def keyword_windows(text, keywords, window=700):

    text = normalize_whitespace(text)

    windows = []

    lower_text = text.lower()

    for keyword in keywords:

        keyword = keyword.lower()

        start = 0

        while True:

            idx = lower_text.find(keyword, start)

            if idx == -1:
                break

            left = max(0, idx - window)
            right = min(len(text), idx + len(keyword) + window)

            windows.append(text[left:right])

            start = idx + len(keyword)

    return windows


# ------------------------------------------------------------------------------
# POLYMER EXTRACTION
# ------------------------------------------------------------------------------

POLYMER_PATTERNS = {

    "PLGA": r"\bPLGA\b|poly\s*\(\s*lactic[- ]co[- ]glycolic acid",

    "PLA": r"\bPLA\b|poly\s*\(\s*lactic acid",

    "PCL": r"\bPCL\b|poly\s*\(\s*caprolactone",

    "PEG": r"\bPEG\b|polyethylene glycol",

    "Chitosan": r"\bchitosan\b",

    "Cellulose": r"\bcellulose\b",

    "PVP": r"\bPVP\b|polyvinylpyrrolidone",

    "PVA": r"\bPVA\b|polyvinyl alcohol",

    "Polypeptide": r"\bpolypeptide\b",

    "Polyester": r"\bpolyester\b"
}


def detect_polymers(text):

    detected = []

    for polymer, pattern in POLYMER_PATTERNS.items():

        if re.search(pattern, text, flags=re.I):
            detected.append(polymer)

    return ";".join(sorted(set(detected)))


# ------------------------------------------------------------------------------
# DRUG EXTRACTION
# ------------------------------------------------------------------------------

KNOWN_DRUGS = [
    "dexamethasone",
    "dexamethasone phosphate",
    "paclitaxel",
    "sirolimus",
    "tacrolimus",
    "deferoxamine",
    "clindamycin",
    "clindamycin phosphate",
    "glipizide",
    "donepezil",
    "ibrutinib",
    "modafinil",
    "artemisinin",
    "simvastatin",
    "indomethacin",
    "ciprofloxacin",
    "amoxicillin",
    "azithromycin",
    "gefitinib",
    "salinomycin",
    "furosemide",
    "brimonidine",
    "brimonidine tartrate",
    "olanzapine",
]


def detect_drugs(text):

    detected = []

    lower = text.lower()

    for drug in KNOWN_DRUGS:

        if drug.lower() in lower:
            detected.append(drug)

    return ";".join(sorted(set(detected)))


# ------------------------------------------------------------------------------
# RELEASE-TIME EXTRACTION
# ------------------------------------------------------------------------------

def extract_release_time_values(text):

    """
    Search for common time + release expressions.
    This does NOT assume that every number is a release value.
    """

    patterns = [

        r"(\d+(?:\.\d+)?)\s*(?:h|hr|hrs|hours?)"
        r".{0,100}?"
        r"(\d+(?:\.\d+)?)\s*%",

        r"(\d+(?:\.\d+)?)\s*%"
        r".{0,100}?"
        r"(\d+(?:\.\d+)?)\s*(?:h|hr|hrs|hours?)",

    ]

    matches = []

    for pattern in patterns:

        for match in re.finditer(pattern, text, flags=re.I):

            values = []

            for group in match.groups():

                try:
                    values.append(float(group))
                except Exception:
                    pass

            if len(values) == 2:

                matches.append({
                    "time_candidate": values[0],
                    "release_candidate": values[1],
                    "context": match.group(0)
                })

    return matches


# ------------------------------------------------------------------------------
# RELEASE DATA TABLE DETECTION
# ------------------------------------------------------------------------------

def extract_release_rows(text):

    """
    Detect lines containing time/release information.

    This is deliberately conservative.
    It creates candidate observations that will later be manually/algorithmically
    validated rather than pretending every number is a true release datum.
    """

    lines = re.split(r"[\n\r]+", text)

    rows = []

    for line_no, line in enumerate(lines):

        line_clean = normalize_whitespace(line)

        if len(line_clean) < 5:
            continue

        lower = line_clean.lower()

        release_context = any(
            term in lower
            for term in [
                "release",
                "cumulative release",
                "drug release",
                "released",
                "release profile"
            ]
        )

        if not release_context:
            continue

        numbers = extract_numbers(line_clean)

        if len(numbers) >= 2:

            rows.append({
                "line_no": line_no,
                "text": line_clean,
                "numbers": numbers
            })

    return rows


# ------------------------------------------------------------------------------
# MAIN EXTRACTION
# ------------------------------------------------------------------------------

paper_records = []
release_records = []

for idx, row in queue.iterrows():

    study_id = safe_text(row.get("study_id"))
    pmcid = row.get("pmcid", np.nan)

    print(
        f"[{idx+1:02d}/{len(queue)}] "
        f"{study_id} | PMCID={pmcid}"
    )

    text_file = locate_text_file(
        study_id,
        pmcid
    )

    if text_file is None:

        print("   WARNING: text file not found")

        paper_records.append({
            "study_id": study_id,
            "pmid": row.get("pmid"),
            "pmcid": pmcid,
            "doi": row.get("doi"),
            "title": row.get("title"),
            "text_file": None,
            "text_available": False,
            "detected_polymers": "",
            "detected_drugs": "",
            "release_candidate_count": 0,
            "extraction_status": "TEXT_NOT_FOUND"
        })

        continue

    try:

        text = text_file.read_text(
            encoding="utf-8",
            errors="ignore"
        )

    except Exception:

        print("   WARNING: could not read text")

        paper_records.append({
            "study_id": study_id,
            "pmid": row.get("pmid"),
            "pmcid": pmcid,
            "doi": row.get("doi"),
            "title": row.get("title"),
            "text_file": str(text_file),
            "text_available": False,
            "detected_polymers": "",
            "detected_drugs": "",
            "release_candidate_count": 0,
            "extraction_status": "READ_ERROR"
        })

        continue

    text = normalize_whitespace(text)

    polymers = detect_polymers(text)

    drugs = detect_drugs(text)

    release_rows = extract_release_rows(text)

    release_matches = extract_release_time_values(text)

    # ----------------------------------------------------------
    # Save candidate release observations
    # ----------------------------------------------------------

    for r in release_rows:

        release_records.append({
            "study_id": study_id,
            "pmid": row.get("pmid"),
            "pmcid": pmcid,
            "doi": row.get("doi"),
            "title": row.get("title"),
            "source_file": str(text_file),
            "line_no": r["line_no"],
            "candidate_numbers": json.dumps(r["numbers"]),
            "source_text": r["text"]
        })

    # ----------------------------------------------------------
    # Paper-level record
    # ----------------------------------------------------------

    paper_records.append({

        "study_id": study_id,

        "pmid": row.get("pmid"),

        "pmcid": pmcid,

        "doi": row.get("doi"),

        "title": row.get("title"),

        "journal": row.get("journal"),

        "publication_year": row.get("publication_year"),

        "text_file": str(text_file),

        "text_available": True,

        "text_length": len(text),

        "detected_polymers": polymers,

        "detected_drugs": drugs,

        "release_candidate_count": len(release_rows),

        "time_release_match_count": len(release_matches),

        "extraction_status": "CANDIDATES_EXTRACTED"

    })


# ------------------------------------------------------------------------------
# DATAFRAMES
# ------------------------------------------------------------------------------

paper_df = pd.DataFrame(paper_records)

release_df = pd.DataFrame(release_records)


# ------------------------------------------------------------------------------
# SAVE RAW EXTRACTION OUTPUTS
# ------------------------------------------------------------------------------

paper_output = (
    OUTPUT_DIR /
    "05_paper_level_extraction.csv"
)

release_output = (
    OUTPUT_DIR /
    "05_release_candidate_observations.csv"
)

paper_df.to_csv(
    paper_output,
    index=False
)

release_df.to_csv(
    release_output,
    index=False
)


# ------------------------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("CODE 05 COMPLETE")
print("=" * 78)

print(
    f"\nPapers processed: "
    f"{len(paper_df)}"
)

print(
    f"Readable full texts: "
    f"{paper_df['text_available'].sum()}"
)

print(
    f"Total candidate release rows: "
    f"{len(release_df)}"
)

print(
    f"\nPaper-level output:\n"
    f"{paper_output}"
)

print(
    f"\nRelease candidate output:\n"
    f"{release_output}"
)

print("\nDetected polymers:")
print(
    paper_df["detected_polymers"]
    .replace("", np.nan)
    .dropna()
    .value_counts()
    .head(20)
)

print("\nDetected drugs:")
print(
    paper_df["detected_drugs"]
    .replace("", np.nan)
    .dropna()
    .value_counts()
    .head(20)
)

print("\nTop papers by candidate release observations:")

display(
    paper_df[
        [
            "study_id",
            "title",
            "detected_polymers",
            "detected_drugs",
            "release_candidate_count",
            "time_release_match_count"
        ]
    ]
    .sort_values(
        "release_candidate_count",
        ascending=False
    )
    .head(20)
)

print("\n" + "=" * 78)
```

```python
# ==============================================================================
# CODE 06 — QUANTITATIVE RELEASE OBSERVATION EXTRACTION
# ==============================================================================

from pathlib import Path
import pandas as pd
import re
import numpy as np

BASE = Path("AI_Polymer_Drug_Release")

INPUT_FILE = BASE / "data/extraction/05_release_candidate_observations.csv"
FULLTEXT_DIR = BASE / "data/fulltext/texts"

OUTPUT_DIR = BASE / "data/extraction"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = OUTPUT_DIR / "06_quantitative_release_observations.csv"
VALIDATION_FILE = OUTPUT_DIR / "06_extraction_validation.csv"

print("=" * 78)
print("CODE 06 — QUANTITATIVE RELEASE OBSERVATION EXTRACTION")
print("=" * 78)

# ------------------------------------------------------------------------------
# 1. LOAD INPUT
# ------------------------------------------------------------------------------

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"""
Could not find:
{INPUT_FILE}

Run Code 05 first.
"""
    )

df = pd.read_csv(INPUT_FILE)

print(f"Input rows: {len(df)}")
print(f"Input columns: {len(df.columns)}")

# ------------------------------------------------------------------------------
# 2. HELPER FUNCTIONS
# ------------------------------------------------------------------------------

def clean_text(x):
    if pd.isna(x):
        return ""
    return str(x).replace("\x00", " ").strip()


def first_match(pattern, text, flags=re.I):
    m = re.search(pattern, text, flags)
    return m.group(1).strip() if m else ""


def detect_time_values(text):
    """
    Detect common release-time expressions.
    Returns unique numeric hour values.
    """

    values = []

    patterns = [
        r'(\d+(?:\.\d+)?)\s*(?:hours?|hrs?|h)\b',
        r'(\d+(?:\.\d+)?)\s*(?:days?|d)\b',
        r'(\d+(?:\.\d+)?)\s*(?:minutes?|mins?|min)\b'
    ]

    for pattern in patterns:
        for m in re.finditer(pattern, text, re.I):
            value = float(m.group(1))

            if "day" in pattern:
                value *= 24
            elif "minute" in pattern or "min" in pattern:
                value /= 60

            values.append(value)

    return sorted(set(round(v, 6) for v in values))


def detect_release_percentages(text):
    """
    Detect values explicitly expressed as percentage release.
    """

    values = []

    patterns = [
        r'(\d+(?:\.\d+)?)\s*%\s*(?:drug\s*)?(?:release|released)',
        r'(?:release|released)[^\n]{0,50}?(\d+(?:\.\d+)?)\s*%',
        r'(\d+(?:\.\d+)?)\s*%\s*cumulative\s+release'
    ]

    for pattern in patterns:
        for m in re.finditer(pattern, text, re.I):
            value = float(m.group(1))

            if 0 <= value <= 100:
                values.append(value)

    return sorted(set(round(v, 4) for v in values))


def detect_release_amounts(text):
    """
    Detect common amount expressions.
    """

    values = []

    patterns = [
        r'(\d+(?:\.\d+)?)\s*(?:µg|μg|ug|mg|g)\s*(?:of\s+)?(?:drug\s*)?(?:was\s*)?(?:released|release)',
        r'(?:released|release)[^\n]{0,40}?(\d+(?:\.\d+)?)\s*(µg|μg|ug|mg|g)'
    ]

    for pattern in patterns:
        for m in re.finditer(pattern, text, re.I):
            try:
                if len(m.groups()) >= 2:
                    value = float(m.group(1))
                    unit = m.group(2)
                    values.append((value, unit))
                else:
                    value = float(m.group(1))
                    values.append((value, "unknown"))
            except Exception:
                pass

    return values


def extract_contexts(text, pattern, window=300):
    """
    Return short text contexts around matches.
    """

    contexts = []

    for m in re.finditer(pattern, text, re.I):
        start = max(0, m.start() - window)
        end = min(len(text), m.end() + window)
        contexts.append(text[start:end].replace("\n", " "))

    return contexts


# ------------------------------------------------------------------------------
# 3. DETECT TEXT COLUMN
# ------------------------------------------------------------------------------

text_candidates = [
    "text",
    "full_text",
    "article_text",
    "content",
    "body"
]

text_column = None

for col in text_candidates:
    if col in df.columns:
        text_column = col
        break

if text_column is None:
    raise KeyError(
        "Could not find a full-text column. "
        f"Available columns are: {list(df.columns)}"
    )

print(f"Using text column: {text_column}")

# ------------------------------------------------------------------------------
# 4. EXTRACT OBSERVATIONS
# ------------------------------------------------------------------------------

observations = []
validation = []

for idx, row in df.iterrows():

    study_id = row.get("study_id", "")
    pmid = row.get("pmid", "")
    doi = row.get("doi", "")
    title = row.get("title", "")

    text = clean_text(row.get(text_column, ""))

    # If text is missing, try the stored text file
    if not text and study_id:

        possible_files = list(FULLTEXT_DIR.glob(f"*{study_id}*"))

        if possible_files:
            try:
                text = possible_files[0].read_text(
                    encoding="utf-8",
                    errors="ignore"
                )
            except Exception:
                pass

    # Existing parser outputs, when available
    polymers = clean_text(row.get("detected_polymers", ""))
    drugs = clean_text(row.get("detected_drugs", ""))

    # Detect quantitative information
    times = detect_time_values(text)
    release_pct = detect_release_percentages(text)
    release_amounts = detect_release_amounts(text)

    release_contexts = extract_contexts(
        text,
        r'(?:drug\s+release|cumulative\s+release|release\s+kinetics|released)',
        window=250
    )

    # --------------------------------------------------------------------------
    # Create candidate rows
    # --------------------------------------------------------------------------

    n_candidates = max(
        len(times),
        len(release_pct),
        len(release_amounts),
        1
    )

    # We do not claim that values correspond to each other automatically.
    # Instead, mark them as candidate observations.

    for j in range(n_candidates):

        time_h = times[j] if j < len(times) else np.nan
        release_value = (
            release_pct[j]
            if j < len(release_pct)
            else np.nan
        )

        amount_value = (
            release_amounts[j][0]
            if j < len(release_amounts)
            else np.nan
        )

        amount_unit = (
            release_amounts[j][1]
            if j < len(release_amounts)
            else ""
        )

        context = (
            release_contexts[j]
            if j < len(release_contexts)
            else ""
        )

        # Confidence is deliberately conservative
        if not pd.isna(time_h) and not pd.isna(release_value):
            confidence = "HIGH"
        elif not pd.isna(release_value):
            confidence = "MEDIUM"
        elif not pd.isna(time_h):
            confidence = "LOW"
        else:
            confidence = "VERY_LOW"

        observations.append({
            "observation_id": f"{study_id}-OBS-{j+1:03d}",
            "study_id": study_id,
            "pmid": pmid,
            "doi": doi,
            "title": title,

            "detected_polymers": polymers,
            "detected_drugs": drugs,

            "time_h_candidate": time_h,

            "cumulative_release_pct_candidate": release_value,

            "release_amount_candidate": amount_value,
            "release_amount_unit": amount_unit,

            "release_context": context,

            "confidence": confidence,

            "automated_mapping_status":
                "CANDIDATE_NOT_VALIDATED"
        })

    validation.append({
        "study_id": study_id,
        "pmid": pmid,
        "title": title,
        "text_length": len(text),
        "time_candidates": len(times),
        "release_percentage_candidates": len(release_pct),
        "release_amount_candidates": len(release_amounts),
        "release_contexts": len(release_contexts),
        "candidate_observation_rows": n_candidates
    })

# ------------------------------------------------------------------------------
# 5. SAVE
# ------------------------------------------------------------------------------

obs_df = pd.DataFrame(observations)
validation_df = pd.DataFrame(validation)

obs_df.to_csv(
    OUTPUT_FILE,
    index=False
)

validation_df.to_csv(
    VALIDATION_FILE,
    index=False
)

# ------------------------------------------------------------------------------
# 6. SUMMARY
# ------------------------------------------------------------------------------

print()
print("=" * 78)
print("CODE 06 COMPLETE")
print("=" * 78)

print(f"Source papers processed: {len(df)}")
print(f"Candidate observation rows: {len(obs_df)}")

print()
print("Confidence distribution:")
print(obs_df["confidence"].value_counts(dropna=False))

print()
print("Papers with candidate release percentages:")
print(
    validation_df[
        validation_df["release_percentage_candidates"] > 0
    ].shape[0]
)

print()
print("Papers with candidate time values:")
print(
    validation_df[
        validation_df["time_candidates"] > 0
    ].shape[0]
)

print()
print("Files created:")
print(OUTPUT_FILE)
print(VALIDATION_FILE)

print("=" * 78)
```

```python
# ==============================================================================
# CODE 06 — QUANTITATIVE RELEASE OBSERVATION EXTRACTION
# ==============================================================================

from pathlib import Path
import pandas as pd
import re
import numpy as np

BASE = Path("AI_Polymer_Drug_Release")

INPUT_FILE = BASE / "data/extraction/05_release_candidate_observations.csv"
OUTPUT_DIR = BASE / "data/extraction"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = OUTPUT_DIR / "06_quantitative_release_observations.csv"
VALIDATION_FILE = OUTPUT_DIR / "06_extraction_validation.csv"

print("=" * 78)
print("CODE 06 — QUANTITATIVE RELEASE OBSERVATION EXTRACTION")
print("=" * 78)

# ------------------------------------------------------------------------------
# 1. LOAD INPUT
# ------------------------------------------------------------------------------

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"Could not find:\n{INPUT_FILE}\n\nRun Code 05 first."
    )

df = pd.read_csv(INPUT_FILE)

print(f"Input rows: {len(df)}")
print(f"Input columns: {len(df.columns)}")
print(f"Available columns: {list(df.columns)}")

# ------------------------------------------------------------------------------
# 2. USE THE ACTUAL CODE-05 TEXT COLUMN
# ------------------------------------------------------------------------------

TEXT_COLUMN = "source_text"

if TEXT_COLUMN not in df.columns:
    raise KeyError(
        f"Expected column '{TEXT_COLUMN}' was not found.\n"
        f"Available columns: {list(df.columns)}"
    )

print(f"Using full-text column: {TEXT_COLUMN}")

# ------------------------------------------------------------------------------
# 3. HELPER FUNCTIONS
# ------------------------------------------------------------------------------

def clean_text(x):
    if pd.isna(x):
        return ""
    return str(x).replace("\x00", " ").strip()


def detect_time_values(text):

    values = []

    patterns = [
        (r'(\d+(?:\.\d+)?)\s*(?:hours?|hrs?|h)\b', 1),
        (r'(\d+(?:\.\d+)?)\s*(?:days?|d)\b', 24),
        (r'(\d+(?:\.\d+)?)\s*(?:minutes?|mins?|min)\b', 1/60)
    ]

    for pattern, multiplier in patterns:

        for m in re.finditer(pattern, text, re.I):

            try:
                value = float(m.group(1)) * multiplier

                if 0 <= value <= 10000:
                    values.append(value)

            except Exception:
                pass

    return sorted(set(round(v, 6) for v in values))


def detect_release_percentages(text):

    values = []

    patterns = [
        r'(\d+(?:\.\d+)?)\s*%\s*(?:cumulative\s+)?(?:drug\s*)?(?:release|released)',
        r'(?:cumulative\s+)?(?:drug\s+)?(?:release|released)[^\n]{0,80}?(\d+(?:\.\d+)?)\s*%',
        r'(\d+(?:\.\d+)?)\s*%\s*cumulative\s+release'
    ]

    for pattern in patterns:

        for m in re.finditer(pattern, text, re.I):

            try:
                value = float(m.group(1))

                if 0 <= value <= 100:
                    values.append(value)

            except Exception:
                pass

    return sorted(set(round(v, 4) for v in values))


def detect_release_amounts(text):

    values = []

    patterns = [
        r'(\d+(?:\.\d+)?)\s*(µg|μg|ug|mg|g)\s*(?:of\s+)?(?:drug\s*)?(?:was\s*)?(?:released|release)',
        r'(?:released|release)[^\n]{0,60}?(\d+(?:\.\d+)?)\s*(µg|μg|ug|mg|g)'
    ]

    for pattern in patterns:

        for m in re.finditer(pattern, text, re.I):

            try:
                values.append(
                    (
                        float(m.group(1)),
                        m.group(2)
                    )
                )
            except Exception:
                pass

    return values


def extract_release_contexts(text):

    pattern = (
        r'(?:drug\s+release|'
        r'cumulative\s+release|'
        r'release\s+kinetics|'
        r'released)'
    )

    contexts = []

    for m in re.finditer(pattern, text, re.I):

        start = max(0, m.start() - 300)
        end = min(len(text), m.end() + 500)

        context = text[start:end]

        context = re.sub(r'\s+', ' ', context)

        contexts.append(context.strip())

    return contexts


# ------------------------------------------------------------------------------
# 4. EXTRACT CANDIDATE OBSERVATIONS
# ------------------------------------------------------------------------------

observations = []
validation = []

for idx, row in df.iterrows():

    study_id = clean_text(row.get("study_id"))
    pmid = clean_text(row.get("pmid"))
    pmcid = clean_text(row.get("pmcid"))
    doi = clean_text(row.get("doi"))
    title = clean_text(row.get("title"))

    text = clean_text(row.get(TEXT_COLUMN))

    polymers = clean_text(row.get("detected_polymers"))
    drugs = clean_text(row.get("detected_drugs"))

    # --------------------------------------------------------------------------
    # Detect numerical information
    # --------------------------------------------------------------------------

    times = detect_time_values(text)

    release_percentages = detect_release_percentages(text)

    release_amounts = detect_release_amounts(text)

    contexts = extract_release_contexts(text)

    # --------------------------------------------------------------------------
    # Number of candidate rows
    # --------------------------------------------------------------------------

    n_candidates = max(
        len(times),
        len(release_percentages),
        len(release_amounts),
        1
    )

    # --------------------------------------------------------------------------
    # Candidate observations
    # --------------------------------------------------------------------------

    for j in range(n_candidates):

        time_value = (
            times[j]
            if j < len(times)
            else np.nan
        )

        release_value = (
            release_percentages[j]
            if j < len(release_percentages)
            else np.nan
        )

        amount_value = (
            release_amounts[j][0]
            if j < len(release_amounts)
            else np.nan
        )

        amount_unit = (
            release_amounts[j][1]
            if j < len(release_amounts)
            else ""
        )

        context = (
            contexts[j]
            if j < len(contexts)
            else ""
        )

        # ----------------------------------------------------------------------
        # Conservative confidence
        # ----------------------------------------------------------------------

        if (
            not pd.isna(time_value)
            and not pd.isna(release_value)
        ):
            confidence = "HIGH_CANDIDATE"

        elif not pd.isna(release_value):
            confidence = "MEDIUM_CANDIDATE"

        elif not pd.isna(time_value):
            confidence = "LOW_CANDIDATE"

        else:
            confidence = "VERY_LOW"

        observations.append({

            "observation_id":
                f"{study_id}-OBS-{j+1:03d}",

            "study_id":
                study_id,

            "pmid":
                pmid,

            "pmcid":
                pmcid,

            "doi":
                doi,

            "title":
                title,

            "detected_polymers":
                polymers,

            "detected_drugs":
                drugs,

            "time_h_candidate":
                time_value,

            "cumulative_release_pct_candidate":
                release_value,

            "release_amount_candidate":
                amount_value,

            "release_amount_unit":
                amount_unit,

            "source_context":
                context,

            "confidence":
                confidence,

            "mapping_status":
                "CANDIDATE_NOT_VALIDATED"
        })

    # --------------------------------------------------------------------------
    # Paper-level validation record
    # --------------------------------------------------------------------------

    validation.append({

        "study_id":
            study_id,

        "pmid":
            pmid,

        "title":
            title,

        "text_length":
            len(text),

        "time_candidates":
            len(times),

        "release_percentage_candidates":
            len(release_percentages),

        "release_amount_candidates":
            len(release_amounts),

        "release_contexts":
            len(contexts),

        "candidate_observation_rows":
            n_candidates
    })


# ------------------------------------------------------------------------------
# 5. CREATE DATAFRAMES
# ------------------------------------------------------------------------------

obs_df = pd.DataFrame(observations)

validation_df = pd.DataFrame(validation)


# ------------------------------------------------------------------------------
# 6. REMOVE EXACT DUPLICATES
# ------------------------------------------------------------------------------

if len(obs_df) > 0:

    obs_df = obs_df.drop_duplicates(
        subset=[
            "study_id",
            "time_h_candidate",
            "cumulative_release_pct_candidate",
            "release_amount_candidate",
            "release_amount_unit"
        ]
    ).reset_index(drop=True)


# ------------------------------------------------------------------------------
# 7. SAVE
# ------------------------------------------------------------------------------

obs_df.to_csv(
    OUTPUT_FILE,
    index=False
)

validation_df.to_csv(
    VALIDATION_FILE,
    index=False
)


# ------------------------------------------------------------------------------
# 8. SUMMARY
# ------------------------------------------------------------------------------

print()
print("=" * 78)
print("CODE 06 COMPLETE")
print("=" * 78)

print(f"Source papers processed: {len(df)}")
print(f"Candidate observation rows: {len(obs_df)}")

print()
print("Confidence distribution:")

if len(obs_df) > 0:
    print(
        obs_df["confidence"]
        .value_counts(dropna=False)
    )
else:
    print("No observations generated.")

print()
print(
    "Papers with candidate release percentages:",
    (
        validation_df[
            validation_df["release_percentage_candidates"] > 0
        ].shape[0]
    )
)

print(
    "Papers with candidate time values:",
    (
        validation_df[
            validation_df["time_candidates"] > 0
        ].shape[0]
    )
)

print()
print("Files created:")
print(OUTPUT_FILE)
print(VALIDATION_FILE)

print("=" * 78)
```

```python
# ==============================================================================
# CODE 07 — QUANTITATIVE RELEASE VALIDATION AND TIME–RELEASE PAIRING
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

BASE = Path("AI_Polymer_Drug_Release")
INPUT_FILE = BASE / "data/extraction/06_quantitative_release_observations.csv"

OUTPUT_DIR = BASE / "data/extraction"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

VALIDATED_FILE = OUTPUT_DIR / "07_validated_release_observations.csv"
REJECTED_FILE = OUTPUT_DIR / "07_rejected_release_candidates.csv"
SUMMARY_FILE = OUTPUT_DIR / "07_validation_summary.csv"

print("=" * 78)
print("CODE 07 — QUANTITATIVE RELEASE VALIDATION")
print("=" * 78)

# ------------------------------------------------------------------------------
# 1. LOAD CODE 06
# ------------------------------------------------------------------------------

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"Could not find:\n{INPUT_FILE}\n\nRun Code 06 first."
    )

df = pd.read_csv(INPUT_FILE)

print(f"Candidate observations loaded: {len(df)}")

required_columns = [
    "study_id",
    "pmid",
    "title",
    "time_h_candidate",
    "cumulative_release_pct_candidate",
    "source_context",
    "confidence"
]

missing = [
    c for c in required_columns
    if c not in df.columns
]

if missing:
    raise KeyError(
        f"Missing required columns: {missing}\n"
        f"Available columns: {list(df.columns)}"
    )

# ------------------------------------------------------------------------------
# 2. CLEAN NUMERIC FIELDS
# ------------------------------------------------------------------------------

df["time_h_candidate"] = pd.to_numeric(
    df["time_h_candidate"],
    errors="coerce"
)

df["cumulative_release_pct_candidate"] = pd.to_numeric(
    df["cumulative_release_pct_candidate"],
    errors="coerce"
)

# ------------------------------------------------------------------------------
# 3. VALIDATION FUNCTIONS
# ------------------------------------------------------------------------------

def context_has_release_language(text):
    text = str(text).lower()

    terms = [
        "drug release",
        "cumulative release",
        "release profile",
        "release kinetics",
        "released",
        "release rate"
    ]

    return any(term in text for term in terms)


def context_has_release_percentage(text):
    text = str(text).lower()

    patterns = [
        r'\d+(?:\.\d+)?\s*%',
        r'percentage',
        r'percent'
    ]

    return any(
        re.search(pattern, text)
        for pattern in patterns
    )


def likely_non_release_percentage(text):
    """
    Identify contexts where a percentage is more likely to represent
    something other than cumulative drug release.
    """

    text = str(text).lower()

    exclusion_terms = [
        "encapsulation efficiency",
        "encapsulation yield",
        "entrapment efficiency",
        "drug loading",
        "loading efficiency",
        "yield",
        "cell viability",
        "viability",
        "degradation",
        "swelling",
        "water uptake",
        "porosity",
        "weight loss",
        "recovery"
    ]

    return any(term in text for term in exclusion_terms)


def assign_validation(row):

    time_value = row["time_h_candidate"]
    release_value = row["cumulative_release_pct_candidate"]

    context = str(row["source_context"])

    # --------------------------------------------------------------------------
    # No release percentage
    # --------------------------------------------------------------------------

    if pd.isna(release_value):

        return (
            "REJECTED",
            "No cumulative release percentage candidate"
        )

    # --------------------------------------------------------------------------
    # Impossible release percentage
    # --------------------------------------------------------------------------

    if release_value < 0 or release_value > 100:

        return (
            "REJECTED",
            "Release percentage outside 0–100%"
        )

    # --------------------------------------------------------------------------
    # No time
    # --------------------------------------------------------------------------

    if pd.isna(time_value):

        return (
            "PERCENTAGE_ONLY",
            "Release percentage detected but no associated time candidate"
        )

    # --------------------------------------------------------------------------
    # Invalid time
    # --------------------------------------------------------------------------

    if time_value < 0:

        return (
            "REJECTED",
            "Negative release time"
        )

    # --------------------------------------------------------------------------
    # Check release terminology
    # --------------------------------------------------------------------------

    if not context_has_release_language(context):

        return (
            "REJECTED",
            "Context lacks explicit drug-release terminology"
        )

    # --------------------------------------------------------------------------
    # Check likely non-release percentages
    # --------------------------------------------------------------------------

    if likely_non_release_percentage(context):

        return (
            "REVIEW",
            "Percentage may represent loading, efficiency, viability, yield, or another non-release variable"
        )

    # --------------------------------------------------------------------------
    # High-confidence candidate
    # --------------------------------------------------------------------------

    return (
        "VALIDATED_CANDIDATE",
        "Time and cumulative release percentage detected in release context"
    )


# ------------------------------------------------------------------------------
# 4. VALIDATE
# ------------------------------------------------------------------------------

results = []

for _, row in df.iterrows():

    status, reason = assign_validation(row)

    results.append({
        "observation_id":
            row.get("observation_id", ""),

        "study_id":
            row.get("study_id", ""),

        "pmid":
            row.get("pmid", ""),

        "pmcid":
            row.get("pmcid", ""),

        "doi":
            row.get("doi", ""),

        "title":
            row.get("title", ""),

        "detected_polymers":
            row.get("detected_polymers", ""),

        "detected_drugs":
            row.get("detected_drugs", ""),

        "time_h_candidate":
            row.get("time_h_candidate", np.nan),

        "cumulative_release_pct_candidate":
            row.get(
                "cumulative_release_pct_candidate",
                np.nan
            ),

        "release_amount_candidate":
            row.get(
                "release_amount_candidate",
                np.nan
            ),

        "release_amount_unit":
            row.get(
                "release_amount_unit",
                ""
            ),

        "source_context":
            row.get("source_context", ""),

        "original_confidence":
            row.get("confidence", ""),

        "validation_status":
            status,

        "validation_reason":
            reason
    })

validated_all = pd.DataFrame(results)

# ------------------------------------------------------------------------------
# 5. SPLIT VALIDATED / REVIEW / REJECTED
# ------------------------------------------------------------------------------

validated = validated_all[
    validated_all["validation_status"] ==
    "VALIDATED_CANDIDATE"
].copy()

review = validated_all[
    validated_all["validation_status"] ==
    "REVIEW"
].copy()

percentage_only = validated_all[
    validated_all["validation_status"] ==
    "PERCENTAGE_ONLY"
].copy()

rejected = validated_all[
    validated_all["validation_status"] ==
    "REJECTED"
].copy()

# ------------------------------------------------------------------------------
# 6. ADD FINAL PROVISIONAL VARIABLES
# ------------------------------------------------------------------------------

if len(validated) > 0:

    validated["validated_time_h"] = (
        validated["time_h_candidate"]
    )

    validated["validated_release_pct"] = (
        validated["cumulative_release_pct_candidate"]
    )

    validated["validation_confidence"] = "PROVISIONAL"

    validated["mapping_status"] = (
        "TIME_RELEASE_PAIR_PROVISIONAL"
    )

# ------------------------------------------------------------------------------
# 7. SAVE
# ------------------------------------------------------------------------------

validated.to_csv(
    VALIDATED_FILE,
    index=False
)

rejected_and_review = pd.concat(
    [rejected, review, percentage_only],
    ignore_index=True
)

REJECTED_FILE = (
    OUTPUT_DIR /
    "07_rejected_release_candidates.csv"
)

rejected_and_review.to_csv(
    REJECTED_FILE,
    index=False
)

# ------------------------------------------------------------------------------
# 8. PAPER-LEVEL SUMMARY
# ------------------------------------------------------------------------------

summary = (
    validated_all
    .groupby("study_id")
    .agg(
        candidate_rows=("study_id", "size"),
        validated_pairs=(
            "validation_status",
            lambda x: (x == "VALIDATED_CANDIDATE").sum()
        ),
        review_rows=(
            "validation_status",
            lambda x: (x == "REVIEW").sum()
        ),
        percentage_only=(
            "validation_status",
            lambda x: (x == "PERCENTAGE_ONLY").sum()
        ),
        rejected_rows=(
            "validation_status",
            lambda x: (x == "REJECTED").sum()
        )
    )
    .reset_index()
)

summary.to_csv(
    SUMMARY_FILE,
    index=False
)

# ------------------------------------------------------------------------------
# 9. REPORT
# ------------------------------------------------------------------------------

print()
print("=" * 78)
print("CODE 07 COMPLETE")
print("=" * 78)

print(f"Input candidate rows: {len(df)}")
print(f"Provisional validated pairs: {len(validated)}")
print(f"Manual-review candidates: {len(review)}")
print(f"Percentage-only candidates: {len(percentage_only)}")
print(f"Rejected candidates: {len(rejected)}")

print()
print("Validation status:")
print(
    validated_all["validation_status"]
    .value_counts(dropna=False)
)

print()
print("Papers containing provisional validated pairs:")

if len(validated) > 0:
    print(
        validated["study_id"]
        .nunique()
    )
else:
    print(0)

print()
print("Files created:")
print(VALIDATED_FILE)
print(REJECTED_FILE)
print(SUMMARY_FILE)

print("=" * 78)
```

```python
# ==============================================================================
# CODE 08 — QUANTITATIVE DATASET HARMONIZATION & FINAL VALIDATION
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

BASE = Path("AI_Polymer_Drug_Release")

INPUT_FILE = BASE / "data/extraction/07_validated_release_observations.csv"
OUT_DIR = BASE / "data/extraction"
OUT_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 78)
print("CODE 08 — QUANTITATIVE DATASET HARMONIZATION & FINAL VALIDATION")
print("=" * 78)

# ------------------------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------------------------

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"\nCould not find:\n{INPUT_FILE}\n\n"
        "Run CODE 07 first."
    )

df = pd.read_csv(INPUT_FILE)

print(f"\nInput rows: {len(df)}")
print(f"Input columns: {len(df.columns)}")
print("\nAvailable columns:")
print(list(df.columns))

# ------------------------------------------------------------------------------
# 2. BASIC CLEANING
# ------------------------------------------------------------------------------

df = df.copy()

# Normalize column names
df.columns = [
    str(c).strip().lower().replace(" ", "_")
    for c in df.columns
]

# Replace empty strings
df = df.replace(r"^\s*$", np.nan, regex=True)

# ------------------------------------------------------------------------------
# 3. HELPER FUNCTIONS
# ------------------------------------------------------------------------------

def first_existing(frame, candidates):
    for c in candidates:
        if c in frame.columns:
            return c
    return None


def clean_text(x):
    if pd.isna(x):
        return ""
    return str(x).strip()


def normalize_polymer(x):
    x = clean_text(x)
    if not x:
        return np.nan

    x = x.lower()

    replacements = {
        "poly(lactic-co-glycolic acid)": "PLGA",
        "poly(lactic-co-glycolic acid) (plga)": "PLGA",
        "plg": "PLGA",
        "poly(lactic acid)": "PLA",
        "polycaprolactone": "PCL",
        "polyethylene glycol": "PEG",
        "polyvinyl alcohol": "PVA",
        "polyvinylpyrrolidone": "PVP",
        "chitosan": "Chitosan",
        "cellulose": "Cellulose",
    }

    for old, new in replacements.items():
        if old in x:
            x = x.replace(old, new.lower())

    return x.strip()


def normalize_drug(x):
    x = clean_text(x)
    if not x:
        return np.nan

    x = x.lower().strip()

    aliases = {
        "dexamethasone phosphate": "dexamethasone",
        "dexamethasone sodium phosphate": "dexamethasone",
        "paclitaxel": "paclitaxel",
        "ciprofloxacin": "ciprofloxacin",
        "simvastatin": "simvastatin",
        "glipizide": "glipizide",
        "deferoxamine": "deferoxamine",
        "donepezil": "donepezil",
        "artemisinin": "artemisinin",
        "indomethacin": "indomethacin",
        "salinomycin": "salinomycin",
        "clindamycin phosphate": "clindamycin",
        "clindamycin": "clindamycin",
        "sirolimus": "sirolimus",
        "tacrolimus": "tacrolimus",
        "modafinil": "modafinil",
        "brimonidine tartrate": "brimonidine",
        "brimonidine": "brimonidine",
        "azithromycin": "azithromycin",
    }

    for old, new in aliases.items():
        if old in x:
            return new

    return x


def numeric_value(x):
    if pd.isna(x):
        return np.nan

    if isinstance(x, (int, float, np.integer, np.floating)):
        return float(x)

    m = re.search(r"-?\d+(?:\.\d+)?", str(x))

    if m:
        try:
            return float(m.group())
        except:
            return np.nan

    return np.nan


# ------------------------------------------------------------------------------
# 4. IDENTIFY IMPORTANT COLUMNS
# ------------------------------------------------------------------------------

study_col = first_existing(
    df,
    ["study_id", "study", "studyid"]
)

pmid_col = first_existing(
    df,
    ["pmid"]
)

pmcid_col = first_existing(
    df,
    ["pmcid"]
)

doi_col = first_existing(
    df,
    ["doi"]
)

title_col = first_existing(
    df,
    ["title"]
)

polymer_col = first_existing(
    df,
    [
        "polymer",
        "detected_polymer",
        "detected_polymers",
        "polymer_identity",
        "polymer_composition"
    ]
)

drug_col = first_existing(
    df,
    [
        "drug",
        "detected_drug",
        "detected_drugs",
        "drug_identity"
    ]
)

time_col = first_existing(
    df,
    [
        "release_time",
        "time",
        "time_value",
        "release_time_value",
        "time_numeric"
    ]
)

release_col = first_existing(
    df,
    [
        "release_percent",
        "release_percentage",
        "cumulative_release",
        "cumulative_release_percent",
        "release_value",
        "percentage"
    ]
)

validation_col = first_existing(
    df,
    [
        "validation_status",
        "status"
    ]
)

print("\nDetected important columns:")
print("Study:", study_col)
print("PMID:", pmid_col)
print("PMCID:", pmcid_col)
print("DOI:", doi_col)
print("Title:", title_col)
print("Polymer:", polymer_col)
print("Drug:", drug_col)
print("Time:", time_col)
print("Release:", release_col)
print("Validation:", validation_col)

# ------------------------------------------------------------------------------
# 5. CREATE STANDARDIZED COLUMNS
# ------------------------------------------------------------------------------

if study_col:
    df["study_id_std"] = df[study_col].astype(str).str.strip()
else:
    df["study_id_std"] = np.nan

if pmid_col:
    df["pmid_std"] = df[pmid_col]
else:
    df["pmid_std"] = np.nan

if pmcid_col:
    df["pmcid_std"] = df[pmcid_col]
else:
    df["pmcid_std"] = np.nan

if doi_col:
    df["doi_std"] = df[doi_col].astype(str).str.strip()
else:
    df["doi_std"] = np.nan

if title_col:
    df["title_std"] = df[title_col].astype(str).str.strip()
else:
    df["title_std"] = np.nan

if polymer_col:
    df["polymer_std"] = df[polymer_col].apply(normalize_polymer)
else:
    df["polymer_std"] = np.nan

if drug_col:
    df["drug_std"] = df[drug_col].apply(normalize_drug)
else:
    df["drug_std"] = np.nan

if time_col:
    df["release_time_std"] = df[time_col].apply(numeric_value)
else:
    df["release_time_std"] = np.nan

if release_col:
    df["release_percent_std"] = df[release_col].apply(numeric_value)
else:
    df["release_percent_std"] = np.nan

# ------------------------------------------------------------------------------
# 6. VALIDATE NUMERIC VALUES
# ------------------------------------------------------------------------------

df["time_valid"] = (
    df["release_time_std"].notna() &
    (df["release_time_std"] >= 0)
)

df["release_percent_valid"] = (
    df["release_percent_std"].notna() &
    (df["release_percent_std"] >= 0) &
    (df["release_percent_std"] <= 100)
)

# ------------------------------------------------------------------------------
# 7. CHECK FOR IMPOSSIBLE RELEASE VALUES
# ------------------------------------------------------------------------------

invalid_release = df[
    df["release_percent_std"].notna() &
    ~df["release_percent_valid"]
].copy()

print("\nInvalid release percentages:", len(invalid_release))

# ------------------------------------------------------------------------------
# 8. CHECK FOR DUPLICATES
# ------------------------------------------------------------------------------

dedup_columns = [
    c for c in [
        "study_id_std",
        "drug_std",
        "polymer_std",
        "release_time_std",
        "release_percent_std"
    ]
    if c in df.columns
]

if dedup_columns:
    duplicate_mask = df.duplicated(
        subset=dedup_columns,
        keep=False
    )
else:
    duplicate_mask = pd.Series(False, index=df.index)

duplicate_rows = df[duplicate_mask].copy()

print("Potential duplicate rows:", len(duplicate_rows))

# ------------------------------------------------------------------------------
# 9. CREATE FINAL ELIGIBILITY FLAG
# ------------------------------------------------------------------------------

df["final_quantitative_candidate"] = (
    df["time_valid"] &
    df["release_percent_valid"]
)

# Keep existing validation information when available
if validation_col:
    df["original_validation_status"] = df[validation_col].astype(str)
else:
    df["original_validation_status"] = np.nan

# ------------------------------------------------------------------------------
# 10. FLAG MANUAL REVIEW
# ------------------------------------------------------------------------------

df["manual_review_required"] = False

# Missing key quantitative information
df.loc[
    ~df["time_valid"] |
    ~df["release_percent_valid"],
    "manual_review_required"
] = True

# Duplicate observations
if len(duplicate_rows) > 0:
    df.loc[duplicate_mask, "manual_review_required"] = True

# ------------------------------------------------------------------------------
# 11. FINAL CLEAN DATASET
# ------------------------------------------------------------------------------

final_cols = [
    "study_id_std",
    "pmid_std",
    "pmcid_std",
    "doi_std",
    "title_std",
    "polymer_std",
    "drug_std",
    "release_time_std",
    "release_percent_std",
    "time_valid",
    "release_percent_valid",
    "final_quantitative_candidate",
    "manual_review_required",
    "original_validation_status"
]

final_cols = [c for c in final_cols if c in df.columns]

final_df = df[final_cols].copy()

# Remove exact duplicate rows from final candidate dataset
final_candidates = final_df[
    final_df["final_quantitative_candidate"]
].copy()

final_candidates = final_candidates.drop_duplicates(
    subset=[
        c for c in [
            "study_id_std",
            "drug_std",
            "polymer_std",
            "release_time_std",
            "release_percent_std"
        ]
        if c in final_candidates.columns
    ]
)

# ------------------------------------------------------------------------------
# 12. PAPER-LEVEL SUMMARY
# ------------------------------------------------------------------------------

paper_summary = (
    final_df
    .groupby("study_id_std", dropna=False)
    .agg(
        candidate_rows=("study_id_std", "size"),
        valid_quantitative_rows=(
            "final_quantitative_candidate",
            "sum"
        ),
        manual_review_rows=(
            "manual_review_required",
            "sum"
        ),
        drugs=("drug_std", lambda x: "; ".join(
            sorted(set(
                str(v) for v in x.dropna()
                if str(v).strip()
            ))
        )),
        polymers=("polymer_std", lambda x: "; ".join(
            sorted(set(
                str(v) for v in x.dropna()
                if str(v).strip()
            ))
        ))
    )
    .reset_index()
)

# ------------------------------------------------------------------------------
# 13. SUMMARY STATISTICS
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("CODE 08 COMPLETE")
print("=" * 78)

print("\nInput candidate rows:", len(df))
print(
    "Rows with valid time + release percentage:",
    int(df["final_quantitative_candidate"].sum())
)
print(
    "Rows requiring manual review:",
    int(df["manual_review_required"].sum())
)
print(
    "Invalid release values:",
    len(invalid_release)
)
print(
    "Potential duplicate rows:",
    len(duplicate_rows)
)
print(
    "Unique papers:",
    df["study_id_std"].nunique()
)

print("\nFinal candidate rows after duplicate removal:",
      len(final_candidates))

print("\nRelease percentage summary:")

if len(final_candidates) > 0:
    print(
        final_candidates["release_percent_std"]
        .describe()
    )
else:
    print("No fully validated quantitative rows available.")

# ------------------------------------------------------------------------------
# 14. OUTPUT FILES
# ------------------------------------------------------------------------------

all_output = (
    OUT_DIR /
    "08_harmonized_quantitative_observations_all.csv"
)

candidate_output = (
    OUT_DIR /
    "08_final_quantitative_release_dataset.csv"
)

review_output = (
    OUT_DIR /
    "08_manual_review_candidates.csv"
)

paper_output = (
    OUT_DIR /
    "08_paper_level_quantitative_summary.csv"
)

invalid_output = (
    OUT_DIR /
    "08_invalid_release_values.csv"
)

final_df.to_csv(all_output, index=False)

final_candidates.to_csv(
    candidate_output,
    index=False
)

df[
    df["manual_review_required"]
].to_csv(
    review_output,
    index=False
)

paper_summary.to_csv(
    paper_output,
    index=False
)

invalid_release.to_csv(
    invalid_output,
    index=False
)

print("\nFiles created:")
print(all_output)
print(candidate_output)
print(review_output)
print(paper_output)
print(invalid_output)

print("\n" + "=" * 78)
```

```python
# ==============================================================================
# CODE 08 FIXED — QUANTITATIVE DATASET HARMONIZATION
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

BASE = Path("AI_Polymer_Drug_Release")
INPUT_FILE = BASE / "data/extraction/07_validated_release_observations.csv"
OUT_DIR = BASE / "data/extraction"

print("=" * 78)
print("CODE 08 FIXED — QUANTITATIVE DATASET HARMONIZATION")
print("=" * 78)

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"Could not find:\n{INPUT_FILE}\n\nRun CODE 07 first."
    )

df = pd.read_csv(INPUT_FILE)

print("\nInput rows:", len(df))
print("Input columns:", len(df.columns))
print("\nColumns:")
print(list(df.columns))

# ------------------------------------------------------------------------------
# 1. BASIC CLEANING
# ------------------------------------------------------------------------------

df = df.copy()

df.columns = [
    str(c).strip().lower()
    for c in df.columns
]

df = df.replace(r"^\s*$", np.nan, regex=True)

# ------------------------------------------------------------------------------
# 2. NUMERIC CLEANER
# ------------------------------------------------------------------------------

def numeric_value(x):

    if pd.isna(x):
        return np.nan

    if isinstance(x, (int, float, np.integer, np.floating)):
        return float(x)

    text = str(x).strip()

    # Remove percentage sign
    text = text.replace("%", "")

    match = re.search(
        r"-?\d+(?:\.\d+)?",
        text
    )

    if match:
        try:
            return float(match.group())
        except:
            return np.nan

    return np.nan


# ------------------------------------------------------------------------------
# 3. NORMALIZE POLYMERS
# ------------------------------------------------------------------------------

def normalize_polymer(x):

    if pd.isna(x):
        return np.nan

    text = str(x).strip()

    if not text:
        return np.nan

    return text


# ------------------------------------------------------------------------------
# 4. NORMALIZE DRUGS
# ------------------------------------------------------------------------------

def normalize_drug(x):

    if pd.isna(x):
        return np.nan

    text = str(x).strip()

    if not text:
        return np.nan

    return text


# ------------------------------------------------------------------------------
# 5. MAP ACTUAL CODE 07 COLUMNS
# ------------------------------------------------------------------------------

# These are the ACTUAL columns produced by Code 07.

df["study_id_std"] = df["study_id"]

df["pmid_std"] = df["pmid"]

df["pmcid_std"] = df["pmcid"]

df["doi_std"] = df["doi"]

df["title_std"] = df["title"]

df["polymer_std"] = (
    df["detected_polymers"]
    .apply(normalize_polymer)
)

df["drug_std"] = (
    df["detected_drugs"]
    .apply(normalize_drug)
)

# CRITICAL FIX:
# Use validated_time_h from Code 07
df["release_time_std"] = (
    df["validated_time_h"]
    .apply(numeric_value)
)

# CRITICAL FIX:
# Use validated_release_pct from Code 07
df["release_percent_std"] = (
    df["validated_release_pct"]
    .apply(numeric_value)
)

# ------------------------------------------------------------------------------
# 6. CHECK VALIDATION STATUS
# ------------------------------------------------------------------------------

print("\nValidation status:")
print(
    df["validation_status"]
    .value_counts(dropna=False)
)

# ------------------------------------------------------------------------------
# 7. NUMERIC VALIDITY
# ------------------------------------------------------------------------------

df["time_valid"] = (
    df["release_time_std"].notna()
    &
    (df["release_time_std"] >= 0)
)

df["release_percent_valid"] = (
    df["release_percent_std"].notna()
    &
    (df["release_percent_std"] >= 0)
    &
    (df["release_percent_std"] <= 100)
)

# ------------------------------------------------------------------------------
# 8. FINAL QUANTITATIVE ELIGIBILITY
# ------------------------------------------------------------------------------

df["final_quantitative_candidate"] = (
    df["time_valid"]
    &
    df["release_percent_valid"]
)

# ------------------------------------------------------------------------------
# 9. CHECK RELEASE VALUES
# ------------------------------------------------------------------------------

invalid_release = df[
    df["release_percent_std"].notna()
    &
    ~df["release_percent_valid"]
].copy()

print(
    "\nInvalid release percentages:",
    len(invalid_release)
)

# ------------------------------------------------------------------------------
# 10. DUPLICATE CHECK
# ------------------------------------------------------------------------------

# IMPORTANT:
# Include observation_id so genuinely separate observations
# are NOT falsely treated as duplicates.

duplicate_columns = [
    "study_id_std",
    "drug_std",
    "polymer_std",
    "release_time_std",
    "release_percent_std"
]

duplicate_mask = df.duplicated(
    subset=duplicate_columns,
    keep=False
)

duplicate_rows = df[
    duplicate_mask
].copy()

print(
    "Potential duplicate rows:",
    len(duplicate_rows)
)

# ------------------------------------------------------------------------------
# 11. MANUAL REVIEW FLAG
# ------------------------------------------------------------------------------

df["manual_review_required"] = False

df.loc[
    ~df["final_quantitative_candidate"],
    "manual_review_required"
] = True

# Preserve Code 07 mapping information
df["mapping_status_std"] = df["mapping_status"]

df["validation_confidence_std"] = (
    df["validation_confidence"]
)

# ------------------------------------------------------------------------------
# 12. CREATE HARMONIZED DATASET
# ------------------------------------------------------------------------------

harmonized_columns = [
    "observation_id",
    "study_id_std",
    "pmid_std",
    "pmcid_std",
    "doi_std",
    "title_std",
    "polymer_std",
    "drug_std",
    "release_time_std",
    "release_percent_std",
    "release_amount_candidate",
    "release_amount_unit",
    "source_context",
    "original_confidence",
    "validation_status",
    "validation_reason",
    "validation_confidence_std",
    "mapping_status_std",
    "time_valid",
    "release_percent_valid",
    "final_quantitative_candidate",
    "manual_review_required"
]

harmonized_columns = [
    c for c in harmonized_columns
    if c in df.columns
]

harmonized = df[
    harmonized_columns
].copy()

# ------------------------------------------------------------------------------
# 13. FINAL QUANTITATIVE DATASET
# ------------------------------------------------------------------------------

final_candidates = harmonized[
    harmonized["final_quantitative_candidate"]
].copy()

# Remove exact duplicate observations only
final_candidates = final_candidates.drop_duplicates(
    subset=[
        "study_id_std",
        "drug_std",
        "polymer_std",
        "release_time_std",
        "release_percent_std"
    ]
)

# ------------------------------------------------------------------------------
# 14. PAPER-LEVEL SUMMARY
# ------------------------------------------------------------------------------

paper_summary = (
    harmonized
    .groupby(
        "study_id_std",
        dropna=False
    )
    .agg(
        candidate_rows=(
            "observation_id",
            "count"
        ),

        valid_quantitative_rows=(
            "final_quantitative_candidate",
            "sum"
        ),

        manual_review_rows=(
            "manual_review_required",
            "sum"
        ),

        drugs=(
            "drug_std",
            lambda x:
            "; ".join(
                sorted(
                    set(
                        str(v)
                        for v in x.dropna()
                        if str(v).strip()
                    )
                )
            )
        ),

        polymers=(
            "polymer_std",
            lambda x:
            "; ".join(
                sorted(
                    set(
                        str(v)
                        for v in x.dropna()
                        if str(v).strip()
                    )
                )
            )
        )
    )
    .reset_index()
)

# ------------------------------------------------------------------------------
# 15. SUMMARY
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("CODE 08 FIXED COMPLETE")
print("=" * 78)

print("\nInput candidate rows:", len(df))

print(
    "Rows with valid time + release percentage:",
    int(df["final_quantitative_candidate"].sum())
)

print(
    "Rows requiring manual review:",
    int(df["manual_review_required"].sum())
)

print(
    "Invalid release values:",
    len(invalid_release)
)

print(
    "Unique papers:",
    df["study_id_std"].nunique()
)

print(
    "Final quantitative rows after exact duplicate removal:",
    len(final_candidates)
)

if len(final_candidates) > 0:

    print("\nRelease percentage statistics:")
    print(
        final_candidates[
            "release_percent_std"
        ].describe()
    )

    print("\nRelease time statistics:")
    print(
        final_candidates[
            "release_time_std"
        ].describe()
    )

else:

    print(
        "\nWARNING: No quantitative observations survived."
    )

# ------------------------------------------------------------------------------
# 16. PAPER-LEVEL VALIDATION
# ------------------------------------------------------------------------------

print("\nPapers with valid quantitative observations:")

valid_papers = paper_summary[
    paper_summary["valid_quantitative_rows"] > 0
]

print(
    valid_papers[
        [
            "study_id_std",
            "valid_quantitative_rows",
            "drugs",
            "polymers"
        ]
    ].to_string(index=False)
)

# ------------------------------------------------------------------------------
# 17. SAVE FILES
# ------------------------------------------------------------------------------

all_output = (
    OUT_DIR /
    "08_harmonized_quantitative_observations_all.csv"
)

candidate_output = (
    OUT_DIR /
    "08_final_quantitative_release_dataset.csv"
)

review_output = (
    OUT_DIR /
    "08_manual_review_candidates.csv"
)

paper_output = (
    OUT_DIR /
    "08_paper_level_quantitative_summary.csv"
)

invalid_output = (
    OUT_DIR /
    "08_invalid_release_values.csv"
)

harmonized.to_csv(
    all_output,
    index=False
)

final_candidates.to_csv(
    candidate_output,
    index=False
)

harmonized[
    harmonized["manual_review_required"]
].to_csv(
    review_output,
    index=False
)

paper_summary.to_csv(
    paper_output,
    index=False
)

invalid_release.to_csv(
    invalid_output,
    index=False
)

print("\nFiles created:")

print(all_output)
print(candidate_output)
print(review_output)
print(paper_output)
print(invalid_output)

print("\n" + "=" * 78)
```

```python
# ==============================================================================
# CODE 09 — DATASET QUALITY ASSESSMENT & EXPLORATORY ANALYSIS
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

BASE_DIR = Path("AI_Polymer_Drug_Release")

INPUT_FILE = (
    BASE_DIR
    / "data"
    / "extraction"
    / "08_final_quantitative_release_dataset.csv"
)

OUTPUT_DIR = BASE_DIR / "data" / "analysis"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 78)
print("CODE 09 — DATASET QUALITY & EXPLORATORY ANALYSIS")
print("=" * 78)

# ------------------------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------------------------

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"\nCould not find:\n{INPUT_FILE}\n\n"
        "Run CODE 08 FIXED first."
    )

df = pd.read_csv(INPUT_FILE)

print(f"\nInput file: {INPUT_FILE}")
print(f"Rows: {len(df)}")
print(f"Columns: {len(df.columns)}")

print("\nAvailable columns:")
print(df.columns.tolist())

# ------------------------------------------------------------------------------
# 2. IDENTIFY IMPORTANT COLUMNS
# ------------------------------------------------------------------------------

def find_column(df, candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None


study_col = find_column(
    df,
    ["study_id_std", "study_id", "study_id_x"]
)

pmid_col = find_column(
    df,
    ["pmid_std", "pmid"]
)

polymer_col = find_column(
    df,
    [
        "polymer_std",
        "polymer",
        "detected_polymers",
        "polymer_composition"
    ]
)

drug_col = find_column(
    df,
    [
        "drug_std",
        "drug",
        "detected_drugs"
    ]
)

time_col = find_column(
    df,
    [
        "release_time_std",
        "release_time_h",
        "validated_time_h",
        "time_h"
    ]
)

release_col = find_column(
    df,
    [
        "release_percent_std",
        "validated_release_pct",
        "cumulative_release_pct_candidate",
        "release_percent"
    ]
)

print("\nDetected columns:")
print("Study:", study_col)
print("PMID:", pmid_col)
print("Polymer:", polymer_col)
print("Drug:", drug_col)
print("Time:", time_col)
print("Release:", release_col)

required = {
    "study": study_col,
    "time": time_col,
    "release": release_col
}

missing = [k for k, v in required.items() if v is None]

if missing:
    raise KeyError(
        f"Could not identify required columns: {missing}\n"
        f"Available columns: {df.columns.tolist()}"
    )

# ------------------------------------------------------------------------------
# 3. NUMERIC CONVERSION
# ------------------------------------------------------------------------------

df[time_col] = pd.to_numeric(
    df[time_col],
    errors="coerce"
)

df[release_col] = pd.to_numeric(
    df[release_col],
    errors="coerce"
)

# ------------------------------------------------------------------------------
# 4. BASIC DATA QUALITY
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("DATA QUALITY")
print("=" * 78)

print("\nMissing values:")
missing_summary = df.isna().sum()
print(missing_summary[missing_summary > 0])

print("\nComplete quantitative rows:")
complete_mask = (
    df[time_col].notna()
    & df[release_col].notna()
)

print(
    f"{complete_mask.sum()} / {len(df)} "
    f"({100 * complete_mask.mean():.1f}%)"
)

# ------------------------------------------------------------------------------
# 5. RANGE CHECKS
# ------------------------------------------------------------------------------

invalid_time = (
    df[time_col].notna()
    & (df[time_col] < 0)
)

invalid_release = (
    df[release_col].notna()
    & (
        (df[release_col] < 0)
        | (df[release_col] > 100)
    )
)

print("\nRange validation:")
print("Negative release times:", int(invalid_time.sum()))
print("Invalid release percentages:", int(invalid_release.sum()))

# ------------------------------------------------------------------------------
# 6. UNIQUE PAPERS
# ------------------------------------------------------------------------------

unique_studies = df[study_col].nunique()

print("\nUnique studies:", unique_studies)
print("Total observations:", len(df))

if pmid_col:
    print("Unique PMIDs:", df[pmid_col].nunique())

# ------------------------------------------------------------------------------
# 7. OBSERVATIONS PER PAPER
# ------------------------------------------------------------------------------

paper_counts = (
    df.groupby(study_col)
    .size()
    .reset_index(name="observation_count")
    .sort_values(
        "observation_count",
        ascending=False
    )
)

print("\nObservations per paper:")
print(paper_counts.to_string(index=False))

# ------------------------------------------------------------------------------
# 8. RELEASE STATISTICS
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("RELEASE STATISTICS")
print("=" * 78)

print(
    df[release_col]
    .describe()
    .to_string()
)

print("\nRelease-time statistics:")
print(
    df[time_col]
    .describe()
    .to_string()
)

# ------------------------------------------------------------------------------
# 9. UNIQUE DRUGS / POLYMERS
# ------------------------------------------------------------------------------

if drug_col:
    print("\n" + "=" * 78)
    print("DRUG INFORMATION")
    print("=" * 78)

    drug_counts = (
        df[drug_col]
        .fillna("UNKNOWN")
        .value_counts()
    )

    print(drug_counts.to_string())

if polymer_col:
    print("\n" + "=" * 78)
    print("POLYMER INFORMATION")
    print("=" * 78)

    polymer_counts = (
        df[polymer_col]
        .fillna("UNKNOWN")
        .value_counts()
    )

    print(polymer_counts.to_string())

# ------------------------------------------------------------------------------
# 10. TIME-RELEASE CORRELATION
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("TIME–RELEASE RELATIONSHIP")
print("=" * 78)

valid_pair = df[
    df[time_col].notna()
    & df[release_col].notna()
].copy()

if len(valid_pair) >= 3:

    pearson_r = valid_pair[
        [time_col, release_col]
    ].corr(
        method="pearson"
    ).iloc[0, 1]

    spearman_r = valid_pair[
        [time_col, release_col]
    ].corr(
        method="spearman"
    ).iloc[0, 1]

    print("Valid pairs:", len(valid_pair))
    print("Pearson correlation:", round(pearson_r, 4))
    print("Spearman correlation:", round(spearman_r, 4))

else:
    pearson_r = np.nan
    spearman_r = np.nan
    print("Insufficient observations for correlation.")

# ------------------------------------------------------------------------------
# 11. SORTED RELEASE OBSERVATIONS
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("QUANTITATIVE OBSERVATIONS")
print("=" * 78)

display_cols = [
    c for c in [
        study_col,
        pmid_col,
        polymer_col,
        drug_col,
        time_col,
        release_col
    ]
    if c is not None
]

print(
    valid_pair[
        display_cols
    ]
    .sort_values(
        [study_col, time_col]
    )
    .to_string(index=False)
)

# ------------------------------------------------------------------------------
# 12. PAPER-LEVEL SUMMARY
# ------------------------------------------------------------------------------

paper_summary = (
    valid_pair
    .groupby(study_col)
    .agg(
        observations=(release_col, "count"),
        min_time_h=(time_col, "min"),
        max_time_h=(time_col, "max"),
        min_release_pct=(release_col, "min"),
        max_release_pct=(release_col, "max"),
        mean_release_pct=(release_col, "mean")
    )
    .reset_index()
)

paper_summary["release_range_pct"] = (
    paper_summary["max_release_pct"]
    - paper_summary["min_release_pct"]
)

# ------------------------------------------------------------------------------
# 13. SAVE QUALITY REPORTS
# ------------------------------------------------------------------------------

QUALITY_FILE = (
    OUTPUT_DIR
    / "09_dataset_quality_report.csv"
)

PAPER_SUMMARY_FILE = (
    OUTPUT_DIR
    / "09_paper_level_release_summary.csv"
)

OBSERVATION_FILE = (
    OUTPUT_DIR
    / "09_valid_quantitative_observations.csv"
)

quality_report = pd.DataFrame({
    "metric": [
        "total_observations",
        "unique_studies",
        "complete_time_release_pairs",
        "invalid_time_values",
        "invalid_release_values",
        "minimum_release_pct",
        "maximum_release_pct",
        "mean_release_pct",
        "median_release_pct",
        "minimum_time_h",
        "maximum_time_h",
        "mean_time_h",
        "median_time_h",
        "pearson_time_release",
        "spearman_time_release"
    ],
    "value": [
        len(df),
        unique_studies,
        len(valid_pair),
        int(invalid_time.sum()),
        int(invalid_release.sum()),
        valid_pair[release_col].min(),
        valid_pair[release_col].max(),
        valid_pair[release_col].mean(),
        valid_pair[release_col].median(),
        valid_pair[time_col].min(),
        valid_pair[time_col].max(),
        valid_pair[time_col].mean(),
        valid_pair[time_col].median(),
        pearson_r,
        spearman_r
    ]
})

quality_report.to_csv(
    QUALITY_FILE,
    index=False
)

paper_summary.to_csv(
    PAPER_SUMMARY_FILE,
    index=False
)

valid_pair.to_csv(
    OBSERVATION_FILE,
    index=False
)

# ------------------------------------------------------------------------------
# 14. FINAL OUTPUT
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("CODE 09 COMPLETE")
print("=" * 78)

print("\nDataset:")
print("  Total observations:", len(df))
print("  Valid quantitative pairs:", len(valid_pair))
print("  Unique studies:", unique_studies)

print("\nQuality:")
print("  Invalid time values:", int(invalid_time.sum()))
print("  Invalid release values:", int(invalid_release.sum()))

print("\nFiles created:")
print(QUALITY_FILE)
print(PAPER_SUMMARY_FILE)
print(OBSERVATION_FILE)

print("=" * 78)
```

```python
# ==============================================================================
# CODE 10 — METADATA RECOVERY & QUANTITATIVE OBSERVATION MAPPING
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

BASE_DIR = Path("AI_Polymer_Drug_Release")

QUANT_FILE = (
    BASE_DIR
    / "data"
    / "extraction"
    / "08_final_quantitative_release_dataset.csv"
)

PAPER_FILE = (
    BASE_DIR
    / "extraction"
    / "05_paper_level_extraction.csv"
)

CANDIDATE_FILE = (
    BASE_DIR
    / "data"
    / "extraction"
    / "05_release_candidate_observations.csv"
)

FULLTEXT_DIR = (
    BASE_DIR
    / "data"
    / "fulltext"
    / "texts"
)

OUTPUT_DIR = (
    BASE_DIR
    / "data"
    / "extraction"
    / "10_metadata_recovery"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 78)
print("CODE 10 — METADATA RECOVERY & QUANTITATIVE OBSERVATION MAPPING")
print("=" * 78)

# ------------------------------------------------------------------------------
# 1. LOAD QUANTITATIVE DATA
# ------------------------------------------------------------------------------

if not QUANT_FILE.exists():
    raise FileNotFoundError(
        f"\nCould not find:\n{QUANT_FILE}\n\n"
        "Run CODE 08 FIXED first."
    )

df = pd.read_csv(QUANT_FILE)

print("\nQuantitative observations loaded:", len(df))

# ------------------------------------------------------------------------------
# 2. LOAD EARLIER EXTRACTION FILES
# ------------------------------------------------------------------------------

paper_df = None
candidate_df = None

if PAPER_FILE.exists():
    paper_df = pd.read_csv(PAPER_FILE)
    print("Paper-level extraction loaded:", len(paper_df))
else:
    print("Paper-level extraction file not found.")

if CANDIDATE_FILE.exists():
    candidate_df = pd.read_csv(CANDIDATE_FILE)
    print("Candidate extraction loaded:", len(candidate_df))
else:
    print("Candidate extraction file not found.")

# ------------------------------------------------------------------------------
# 3. IDENTIFY STUDY / TEXT COLUMNS
# ------------------------------------------------------------------------------

def find_column(frame, candidates):
    if frame is None:
        return None

    for col in candidates:
        if col in frame.columns:
            return col

    return None


study_col = find_column(
    df,
    ["study_id_std", "study_id"]
)

time_col = find_column(
    df,
    ["release_time_std", "validated_time_h"]
)

release_col = find_column(
    df,
    ["release_percent_std", "validated_release_pct"]
)

if study_col is None:
    raise KeyError(
        f"Could not identify study column.\n"
        f"Available columns: {df.columns.tolist()}"
    )

if time_col is None or release_col is None:
    raise KeyError(
        "Could not identify quantitative columns."
    )

# ------------------------------------------------------------------------------
# 4. POLYMER / DRUG DICTIONARIES
# ------------------------------------------------------------------------------

POLYMERS = {
    "PLGA": [
        "poly(lactic-co-glycolic acid)",
        "poly(lactic-co-glycolic)",
        "poly(lactide-co-glycolide)",
        "poly(lactide-co-glycolide)",
        "plga"
    ],

    "PLA": [
        "poly(lactic acid)",
        "polylactic acid",
        "poly-lactic acid",
        "pla"
    ],

    "PCL": [
        "polycaprolactone",
        "pcl"
    ],

    "PEG": [
        "polyethylene glycol",
        "poly(ethylene glycol)",
        "peg"
    ],

    "PVA": [
        "polyvinyl alcohol",
        "poly(vinyl alcohol)",
        "pva"
    ],

    "PVP": [
        "polyvinylpyrrolidone",
        "poly(vinylpyrrolidone)",
        "pvp"
    ],

    "Chitosan": [
        "chitosan"
    ],

    "Cellulose": [
        "cellulose",
        "cellulose nanocrystal",
        "cellulose nanocrystals"
    ],

    "Polyester": [
        "polyester"
    ],

    "Polypeptide": [
        "polypeptide"
    ]
}

DRUGS = [
    "dexamethasone",
    "dexamethasone phosphate",
    "paclitaxel",
    "sirolimus",
    "tacrolimus",
    "clindamycin",
    "clindamycin phosphate",
    "glipizide",
    "deferoxamine",
    "donepezil",
    "simvastatin",
    "ciprofloxacin",
    "indomethacin",
    "salinomycin",
    "artemisinin",
    "brimonidine",
    "brimonidine tartrate",
    "modafinil",
    "azithromycin",
    "ibrutinib",
    "furosemide"
]

# ------------------------------------------------------------------------------
# 5. NORMALIZATION
# ------------------------------------------------------------------------------

def normalize_text(value):
    if pd.isna(value):
        return ""

    text = str(value).lower()

    text = text.replace("–", "-")
    text = text.replace("—", "-")

    return re.sub(
        r"\s+",
        " ",
        text
    ).strip()


def detect_polymers(text):
    text = normalize_text(text)

    found = []

    for polymer, aliases in POLYMERS.items():
        for alias in aliases:
            if alias.lower() in text:
                found.append(polymer)
                break

    return sorted(set(found))


def detect_drugs(text):
    text = normalize_text(text)

    found = []

    for drug in DRUGS:
        if drug.lower() in text:
            found.append(drug)

    return sorted(set(found))

# ------------------------------------------------------------------------------
# 6. BUILD PAPER-LEVEL TEXT INDEX
# ------------------------------------------------------------------------------

print("\nBuilding paper-level metadata index...")

paper_records = {}

# From CODE 05 paper-level extraction
if paper_df is not None:

    paper_study_col = find_column(
        paper_df,
        ["study_id", "study_id_std"]
    )

    text_candidates = [
        "text",
        "source_text",
        "full_text",
        "reviewer_notes"
    ]

    for _, row in paper_df.iterrows():

        if paper_study_col is None:
            continue

        study = str(
            row[paper_study_col]
        )

        text_parts = []

        for col in text_candidates:
            if col in paper_df.columns:
                value = row[col]

                if pd.notna(value):
                    text_parts.append(
                        str(value)
                    )

        paper_records.setdefault(
            study,
            []
        ).extend(text_parts)

# ------------------------------------------------------------------------------
# 7. ADD FULL-TEXT FILES
# ------------------------------------------------------------------------------

if FULLTEXT_DIR.exists():

    for txt_file in FULLTEXT_DIR.glob("*.txt"):

        try:
            text = txt_file.read_text(
                encoding="utf-8",
                errors="ignore"
            )
        except Exception:
            continue

        # Try to identify study ID from filename
        match = re.search(
            r"(STUDY-\d+)",
            txt_file.name,
            flags=re.I
        )

        if match:

            study = match.group(1).upper()

            paper_records.setdefault(
                study,
                []
            ).append(text)

print(
    "Studies with metadata text:",
    len(paper_records)
)

# ------------------------------------------------------------------------------
# 8. ADD CANDIDATE SOURCE TEXT
# ------------------------------------------------------------------------------

candidate_records = {}

if candidate_df is not None:

    cand_study_col = find_column(
        candidate_df,
        ["study_id", "study_id_std"]
    )

    cand_text_col = find_column(
        candidate_df,
        [
            "source_text",
            "text",
            "source_context"
        ]
    )

    if cand_study_col and cand_text_col:

        for _, row in candidate_df.iterrows():

            study = str(
                row[cand_study_col]
            )

            text = row[cand_text_col]

            if pd.notna(text):

                candidate_records.setdefault(
                    study,
                    []
                ).append(
                    str(text)
                )

print(
    "Studies with candidate source text:",
    len(candidate_records)
)

# ------------------------------------------------------------------------------
# 9. RECOVER METADATA
# ------------------------------------------------------------------------------

metadata_rows = []

for study in df[study_col].dropna().unique():

    study = str(study)

    text_parts = []

    text_parts.extend(
        paper_records.get(study, [])
    )

    text_parts.extend(
        candidate_records.get(study, [])
    )

    combined_text = "\n".join(
        text_parts
    )

    polymers = detect_polymers(
        combined_text
    )

    drugs = detect_drugs(
        combined_text
    )

    metadata_rows.append({
        "study_id": study,
        "recovered_polymer_count": len(polymers),
        "recovered_polymers": ";".join(polymers),
        "recovered_drug_count": len(drugs),
        "recovered_drugs": ";".join(drugs),
        "metadata_text_available": bool(
            combined_text.strip()
        ),
        "metadata_text_length": len(
            combined_text
        )
    })

metadata_df = pd.DataFrame(
    metadata_rows
)

print("\nRecovered metadata:")
print(
    metadata_df[
        [
            "study_id",
            "recovered_polymers",
            "recovered_drugs"
        ]
    ].to_string(index=False)
)

# ------------------------------------------------------------------------------
# 10. MERGE METADATA WITH QUANTITATIVE OBSERVATIONS
# ------------------------------------------------------------------------------

out = df.merge(
    metadata_df,
    left_on=study_col,
    right_on="study_id",
    how="left"
)

# ------------------------------------------------------------------------------
# 11. METADATA CONFIDENCE
# ------------------------------------------------------------------------------

def classify_metadata(row):

    polymer_count = row[
        "recovered_polymer_count"
    ]

    drug_count = row[
        "recovered_drug_count"
    ]

    if polymer_count > 0 and drug_count > 0:
        return "HIGH"

    if polymer_count > 0 or drug_count > 0:
        return "PARTIAL"

    return "MISSING"


out["metadata_recovery_status"] = out.apply(
    classify_metadata,
    axis=1
)

# ------------------------------------------------------------------------------
# 12. OBSERVATION-LEVEL IDs
# ------------------------------------------------------------------------------

out["mapped_observation_id"] = [
    f"QOBS-{i:04d}"
    for i in range(1, len(out) + 1)
]

# ------------------------------------------------------------------------------
# 13. STUDY-LEVEL INDEPENDENCE FLAG
# ------------------------------------------------------------------------------

study_observation_counts = (
    out.groupby(study_col)
    .size()
    .to_dict()
)

out["observations_in_same_study"] = (
    out[study_col]
    .map(study_observation_counts)
)

out["repeated_measurement_study"] = (
    out["observations_in_same_study"] > 1
)

# ------------------------------------------------------------------------------
# 14. SAVE STUDY-LEVEL METADATA
# ------------------------------------------------------------------------------

METADATA_FILE = (
    OUTPUT_DIR
    / "10_recovered_study_metadata.csv"
)

metadata_df.to_csv(
    METADATA_FILE,
    index=False
)

# ------------------------------------------------------------------------------
# 15. SAVE MAPPED OBSERVATIONS
# ------------------------------------------------------------------------------

MAPPED_FILE = (
    OUTPUT_DIR
    / "10_mapped_quantitative_observations.csv"
)

out.to_csv(
    MAPPED_FILE,
    index=False
)

# ------------------------------------------------------------------------------
# 16. SAVE MISSING METADATA
# ------------------------------------------------------------------------------

MISSING_FILE = (
    OUTPUT_DIR
    / "10_metadata_missing_for_manual_review.csv"
)

missing_metadata = out[
    out["metadata_recovery_status"] == "MISSING"
].copy()

missing_metadata.to_csv(
    MISSING_FILE,
    index=False
)

# ------------------------------------------------------------------------------
# 17. SUMMARY
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("CODE 10 COMPLETE")
print("=" * 78)

print(
    "\nQuantitative observations:",
    len(out)
)

print(
    "Unique studies:",
    out[study_col].nunique()
)

print(
    "\nMetadata recovery status:"
)

print(
    out[
        "metadata_recovery_status"
    ].value_counts()
)

print(
    "\nStudies with recovered polymers:",
    (
        metadata_df[
            "recovered_polymer_count"
        ] > 0
    ).sum()
)

print(
    "Studies with recovered drugs:",
    (
        metadata_df[
            "recovered_drug_count"
        ] > 0
    ).sum()
)

print(
    "\nRepeated-measurement observations:",
    int(
        out[
            "repeated_measurement_study"
        ].sum()
    )
)

print("\nFiles created:")
print(METADATA_FILE)
print(MAPPED_FILE)
print(MISSING_FILE)

print("=" * 78)
```

```python
# ==============================================================================
# CODE 11 — EXPERIMENTAL FORMULATION VERIFICATION
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

BASE_DIR = Path("AI_Polymer_Drug_Release")

INPUT_FILE = (
    BASE_DIR
    / "data"
    / "extraction"
    / "10_metadata_recovery"
    / "10_mapped_quantitative_observations.csv"
)

CANDIDATE_FILE = (
    BASE_DIR
    / "data"
    / "extraction"
    / "05_release_candidate_observations.csv"
)

OUTPUT_DIR = (
    BASE_DIR
    / "data"
    / "extraction"
    / "11_formulation_verification"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 78)
print("CODE 11 — EXPERIMENTAL FORMULATION VERIFICATION")
print("=" * 78)

# ------------------------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------------------------

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"\nCould not find:\n{INPUT_FILE}\n\n"
        "Run CODE 10 first."
    )

df = pd.read_csv(INPUT_FILE)

print("\nInput observations:", len(df))
print("Input columns:", len(df.columns))

candidate_df = None

if CANDIDATE_FILE.exists():
    candidate_df = pd.read_csv(CANDIDATE_FILE)
    print("Candidate source rows:", len(candidate_df))

# ------------------------------------------------------------------------------
# 2. COLUMN DETECTION
# ------------------------------------------------------------------------------

def find_column(frame, candidates):
    for c in candidates:
        if c in frame.columns:
            return c
    return None


study_col = find_column(
    df,
    ["study_id_std", "study_id"]
)

time_col = find_column(
    df,
    ["release_time_std", "validated_time_h"]
)

release_col = find_column(
    df,
    ["release_percent_std", "validated_release_pct"]
)

if study_col is None:
    raise KeyError("Study column not found.")

# ------------------------------------------------------------------------------
# 3. TERMINOLOGY
# ------------------------------------------------------------------------------

POLYMER_ALIASES = {
    "PLGA": [
        "plga",
        "poly(lactic-co-glycolic acid)",
        "poly(lactic-co-glycolic)",
        "poly(lactide-co-glycolide)"
    ],
    "PLA": [
        "pla",
        "poly(lactic acid)",
        "polylactic acid"
    ],
    "PCL": [
        "pcl",
        "polycaprolactone"
    ],
    "PEG": [
        "peg",
        "polyethylene glycol",
        "poly(ethylene glycol)"
    ],
    "PVA": [
        "pva",
        "polyvinyl alcohol",
        "poly(vinyl alcohol)"
    ],
    "PVP": [
        "pvp",
        "polyvinylpyrrolidone",
        "poly(vinylpyrrolidone)"
    ],
    "Chitosan": [
        "chitosan"
    ],
    "Cellulose": [
        "cellulose",
        "cellulose nanocrystal",
        "cellulose nanocrystals"
    ]
}

DRUGS = [
    "dexamethasone phosphate",
    "dexamethasone",
    "paclitaxel",
    "sirolimus",
    "tacrolimus",
    "clindamycin phosphate",
    "clindamycin",
    "glipizide",
    "deferoxamine",
    "donepezil",
    "simvastatin",
    "ciprofloxacin",
    "indomethacin",
    "salinomycin",
    "artemisinin",
    "brimonidine tartrate",
    "brimonidine",
    "modafinil",
    "azithromycin",
    "ibrutinib",
    "furosemide"
]

EXPERIMENTAL_TERMS = [
    "formulation",
    "prepared",
    "preparation",
    "fabricated",
    "fabrication",
    "loaded",
    "drug-loaded",
    "drug loaded",
    "encapsulated",
    "encapsulation",
    "microsphere",
    "microspheres",
    "nanoparticle",
    "nanoparticles",
    "nanofiber",
    "nanofibers",
    "implant",
    "implants",
    "microneedle",
    "microneedles",
    "film",
    "films",
    "scaffold",
    "release study",
    "release studies",
    "in vitro release",
    "drug release"
]

NON_EXPERIMENTAL_TERMS = [
    "review",
    "reviews",
    "previous studies",
    "reported previously",
    "literature",
    "reference",
    "references",
    "clinical guideline",
    "background",
    "however, studies",
    "other studies"
]

# ------------------------------------------------------------------------------
# 4. NORMALIZATION
# ------------------------------------------------------------------------------

def norm(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()

    text = text.replace("–", "-")
    text = text.replace("—", "-")

    return re.sub(r"\s+", " ", text).strip()


def polymer_hits(text):
    text = norm(text)

    found = []

    for polymer, aliases in POLYMER_ALIASES.items():
        for alias in aliases:
            if alias in text:
                found.append(polymer)
                break

    return sorted(set(found))


def drug_hits(text):
    text = norm(text)

    found = []

    for drug in DRUGS:
        if drug in text:
            found.append(drug)

    return sorted(set(found))


# ------------------------------------------------------------------------------
# 5. BUILD SOURCE TEXT BY STUDY
# ------------------------------------------------------------------------------

source_by_study = {}

if candidate_df is not None:

    cand_study = find_column(
        candidate_df,
        ["study_id", "study_id_std"]
    )

    cand_text = find_column(
        candidate_df,
        [
            "source_text",
            "source_context",
            "text"
        ]
    )

    if cand_study and cand_text:

        for _, row in candidate_df.iterrows():

            study = str(row[cand_study])

            text = row[cand_text]

            if pd.notna(text):

                source_by_study.setdefault(
                    study,
                    []
                ).append(str(text))


# ------------------------------------------------------------------------------
# 6. FIND EXPERIMENTAL CONTEXT
# ------------------------------------------------------------------------------

def experimental_context(text):

    text = str(text)

    # Split into sentences.
    sentences = re.split(
        r"(?<=[.!?])\s+",
        text
    )

    selected = []

    for sentence in sentences:

        s = norm(sentence)

        has_experimental = any(
            term in s
            for term in EXPERIMENTAL_TERMS
        )

        if has_experimental:
            selected.append(sentence)

    return " ".join(selected)


# ------------------------------------------------------------------------------
# 7. SCORE FORMULATION EVIDENCE
# ------------------------------------------------------------------------------

def evidence_score(text, polymer_list, drug_list):

    t = norm(text)

    score = 0
    evidence = []

    experimental_hits = [
        term
        for term in EXPERIMENTAL_TERMS
        if term in t
    ]

    if experimental_hits:
        score += min(
            4,
            len(set(experimental_hits))
        )

        evidence.append(
            "experimental_terms"
        )

    if polymer_list:
        score += 2
        evidence.append(
            "polymer_detected"
        )

    if drug_list:
        score += 2
        evidence.append(
            "drug_detected"
        )

    # Strong formulations usually contain words such as
    # loaded, encapsulated, prepared, fabricated, etc.
    formulation_words = [
        "loaded",
        "encapsulated",
        "prepared",
        "fabricated",
        "formulated",
        "incorporated"
    ]

    if any(
        word in t
        for word in formulation_words
    ):
        score += 2
        evidence.append(
            "formulation_action"
        )

    return score, ";".join(evidence)


# ------------------------------------------------------------------------------
# 8. PAPER-LEVEL VERIFICATION
# ------------------------------------------------------------------------------

verification_rows = []

for study in df[study_col].dropna().unique():

    study = str(study)

    source_text = "\n".join(
        source_by_study.get(
            study,
            []
        )
    )

    if not source_text:

        verification_rows.append({
            "study_id": study,
            "experimental_context": "",
            "confirmed_polymers": "",
            "confirmed_drugs": "",
            "evidence_score": 0,
            "verification_status": "UNCERTAIN",
            "verification_reason":
                "No candidate source text available"
        })

        continue

    exp_text = experimental_context(
        source_text
    )

    polymers = polymer_hits(
        exp_text
    )

    drugs = drug_hits(
        exp_text
    )

    score, evidence = evidence_score(
        exp_text,
        polymers,
        drugs
    )

    # Remove obvious non-experimental references
    # when they dominate the extracted context.
    non_exp_hits = sum(
        term in norm(exp_text)
        for term in NON_EXPERIMENTAL_TERMS
    )

    if score >= 7 and polymers and drugs:

        status = "CONFIRMED_EXPERIMENTAL"

    elif score >= 4 and (polymers or drugs):

        status = "LIKELY_EXPERIMENTAL"

    elif polymers or drugs:

        status = "MENTION_ONLY"

    else:

        status = "UNCERTAIN"

    verification_rows.append({
        "study_id": study,
        "experimental_context":
            exp_text[:5000],
        "confirmed_polymers":
            ";".join(polymers),
        "confirmed_drugs":
            ";".join(drugs),
        "evidence_score":
            score,
        "non_experimental_term_count":
            non_exp_hits,
        "evidence_flags":
            evidence,
        "verification_status":
            status,
        "verification_reason":
            "Experimental context evaluated"
    })


verification_df = pd.DataFrame(
    verification_rows
)

# ------------------------------------------------------------------------------
# 9. MERGE BACK TO OBSERVATIONS
# ------------------------------------------------------------------------------

out = df.merge(
    verification_df,
    left_on=study_col,
    right_on="study_id",
    how="left"
)

# ------------------------------------------------------------------------------
# 10. METADATA QUALITY FLAGS
# ------------------------------------------------------------------------------

out["usable_formulation_metadata"] = (
    out["verification_status"].isin(
        [
            "CONFIRMED_EXPERIMENTAL",
            "LIKELY_EXPERIMENTAL"
        ]
    )
)

out["manual_formulation_review"] = (
    ~out["usable_formulation_metadata"]
)

# ------------------------------------------------------------------------------
# 11. SAVE
# ------------------------------------------------------------------------------

VERIFICATION_FILE = (
    OUTPUT_DIR
    / "11_formulation_verification_by_study.csv"
)

MAPPED_FILE = (
    OUTPUT_DIR
    / "11_quantitative_observations_with_verified_metadata.csv"
)

MANUAL_FILE = (
    OUTPUT_DIR
    / "11_manual_formulation_review.csv"
)

verification_df.to_csv(
    VERIFICATION_FILE,
    index=False
)

out.to_csv(
    MAPPED_FILE,
    index=False
)

out[
    out["manual_formulation_review"]
].to_csv(
    MANUAL_FILE,
    index=False
)

# ------------------------------------------------------------------------------
# 12. SUMMARY
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("FORMULATION VERIFICATION SUMMARY")
print("=" * 78)

print(
    "\nVerification status:"
)

print(
    verification_df[
        "verification_status"
    ].value_counts()
)

print(
    "\nStudies with confirmed polymers:"
)

print(
    verification_df[
        verification_df[
            "confirmed_polymers"
        ].astype(str).str.len() > 0
    ][
        [
            "study_id",
            "confirmed_polymers",
            "confirmed_drugs",
            "evidence_score",
            "verification_status"
        ]
    ].to_string(index=False)
)

print(
    "\nQuantitative observations with usable formulation metadata:",
    int(
        out[
            "usable_formulation_metadata"
        ].sum()
    )
)

print(
    "Observations requiring manual formulation review:",
    int(
        out[
            "manual_formulation_review"
        ].sum()
    )
)

print("\nFiles created:")
print(VERIFICATION_FILE)
print(MAPPED_FILE)
print(MANUAL_FILE)

print("=" * 78)
print("CODE 11 COMPLETE")
print("=" * 78)
```

```python
# ==============================================================================
# CODE 12 — FINAL DATASET ASSEMBLY & ANALYTICAL READINESS
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 78)
print("CODE 12 — FINAL DATASET ASSEMBLY & ANALYTICAL READINESS")
print("=" * 78)

BASE = Path("AI_Polymer_Drug_Release")

INPUT_FILE = (
    BASE / "data/extraction/11_formulation_verification/"
    "11_quantitative_observations_with_verified_metadata.csv"
)

OUTPUT_DIR = BASE / "data/final"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"\nCould not find:\n{INPUT_FILE}\n\n"
        "Run CODE 11 first."
    )

df = pd.read_csv(INPUT_FILE)

print(f"\nInput file: {INPUT_FILE}")
print(f"Input rows: {len(df)}")
print(f"Input columns: {len(df.columns)}")
print("\nAvailable columns:")
print(df.columns.tolist())

# ------------------------------------------------------------------------------
# Helper: find columns robustly
# ------------------------------------------------------------------------------

def find_col(possible_names):
    for c in possible_names:
        if c in df.columns:
            return c
    return None


study_col = find_col([
    "study_id",
    "study_id_std"
])

pmid_col = find_col([
    "pmid",
    "pmid_std"
])

doi_col = find_col([
    "doi",
    "doi_std"
])

title_col = find_col([
    "title",
    "title_std"
])

polymer_col = find_col([
    "confirmed_polymers",
    "polymer_verified",
    "recovered_polymers",
    "polymer_std"
])

drug_col = find_col([
    "confirmed_drugs",
    "drug_verified",
    "recovered_drugs",
    "drug_std"
])

time_col = find_col([
    "release_time_std",
    "validated_time_h",
    "time_h",
    "release_time"
])

release_col = find_col([
    "release_percent_std",
    "validated_release_pct",
    "cumulative_release_pct",
    "release_percent"
])

verification_col = find_col([
    "verification_status"
])

confidence_col = find_col([
    "verification_confidence",
    "validation_confidence_std",
    "validation_confidence"
])

print("\nDetected columns:")
print("Study:", study_col)
print("PMID:", pmid_col)
print("DOI:", doi_col)
print("Title:", title_col)
print("Polymer:", polymer_col)
print("Drug:", drug_col)
print("Time:", time_col)
print("Release:", release_col)
print("Verification:", verification_col)

required = {
    "study": study_col,
    "time": time_col,
    "release": release_col,
    "polymer": polymer_col,
    "drug": drug_col
}

missing = [k for k, v in required.items() if v is None]

if missing:
    raise KeyError(
        "Could not identify required columns: "
        + ", ".join(missing)
    )

# ------------------------------------------------------------------------------
# Create clean analysis dataframe
# ------------------------------------------------------------------------------

analysis = pd.DataFrame()

analysis["study_id"] = df[study_col].astype(str)

if pmid_col:
    analysis["pmid"] = df[pmid_col]

if doi_col:
    analysis["doi"] = df[doi_col]

if title_col:
    analysis["title"] = df[title_col]

analysis["polymer"] = (
    df[polymer_col]
    .replace(["UNKNOWN", "nan", "None", ""], np.nan)
)

analysis["drug"] = (
    df[drug_col]
    .replace(["UNKNOWN", "nan", "None", ""], np.nan)
)

analysis["release_time_h"] = pd.to_numeric(
    df[time_col],
    errors="coerce"
)

analysis["release_percent"] = pd.to_numeric(
    df[release_col],
    errors="coerce"
)

if verification_col:
    analysis["verification_status"] = df[verification_col]

if confidence_col:
    analysis["verification_confidence"] = df[confidence_col]

# ------------------------------------------------------------------------------
# Basic validity checks
# ------------------------------------------------------------------------------

analysis["valid_time"] = (
    analysis["release_time_h"].notna()
    & (analysis["release_time_h"] >= 0)
)

analysis["valid_release"] = (
    analysis["release_percent"].notna()
    & (analysis["release_percent"] >= 0)
    & (analysis["release_percent"] <= 100)
)

analysis["analytically_valid"] = (
    analysis["valid_time"]
    & analysis["valid_release"]
    & analysis["polymer"].notna()
    & analysis["drug"].notna()
)

# ------------------------------------------------------------------------------
# Repeated-measurement identification
# ------------------------------------------------------------------------------

study_counts = (
    analysis.groupby("study_id")
    .size()
    .rename("study_observation_count")
)

analysis = analysis.merge(
    study_counts,
    left_on="study_id",
    right_index=True,
    how="left"
)

analysis["repeated_measurement"] = (
    analysis["study_observation_count"] > 1
)

# ------------------------------------------------------------------------------
# Sort observations within each study
# ------------------------------------------------------------------------------

analysis = analysis.sort_values(
    ["study_id", "release_time_h"],
    ascending=[True, True]
).reset_index(drop=True)

# ------------------------------------------------------------------------------
# Observation sequence
# ------------------------------------------------------------------------------

analysis["observation_number"] = (
    analysis.groupby("study_id")
    .cumcount() + 1
)

# ------------------------------------------------------------------------------
# Within-study release change
# ------------------------------------------------------------------------------

analysis["previous_release_percent"] = (
    analysis.groupby("study_id")["release_percent"]
    .shift(1)
)

analysis["release_change_percent"] = (
    analysis["release_percent"]
    - analysis["previous_release_percent"]
)

# ------------------------------------------------------------------------------
# Flag potentially problematic curves
# ------------------------------------------------------------------------------

analysis["non_monotonic_release"] = False

for study, idx in analysis.groupby("study_id").groups.items():

    values = analysis.loc[idx, "release_percent"]

    if len(values) > 1:

        diffs = values.diff().dropna()

        if (diffs < -1e-9).any():
            analysis.loc[idx, "non_monotonic_release"] = True

# ------------------------------------------------------------------------------
# Final analytical inclusion
# ------------------------------------------------------------------------------

analysis["final_analysis_inclusion"] = (
    analysis["analytically_valid"]
    & analysis["polymer"].notna()
    & analysis["drug"].notna()
)

# ------------------------------------------------------------------------------
# Summary statistics
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("FINAL DATASET QUALITY")
print("=" * 78)

print("\nTotal observations:", len(analysis))

print(
    "Analytically valid observations:",
    int(analysis["analytically_valid"].sum())
)

print(
    "Final analysis observations:",
    int(analysis["final_analysis_inclusion"].sum())
)

print(
    "Unique studies:",
    analysis["study_id"].nunique()
)

print(
    "Repeated-measurement observations:",
    int(analysis["repeated_measurement"].sum())
)

print(
    "Studies with repeated measurements:",
    int(
        (study_counts > 1).sum()
    )
)

print("\nMissing polymer:")
print(analysis["polymer"].isna().sum())

print("\nMissing drug:")
print(analysis["drug"].isna().sum())

print("\nInvalid release values:")
print((~analysis["valid_release"]).sum())

print("\nInvalid time values:")
print((~analysis["valid_time"]).sum())

# ------------------------------------------------------------------------------
# Release statistics
# ------------------------------------------------------------------------------

valid_release = analysis.loc[
    analysis["final_analysis_inclusion"],
    "release_percent"
]

valid_time = analysis.loc[
    analysis["final_analysis_inclusion"],
    "release_time_h"
]

print("\n" + "=" * 78)
print("RELEASE STATISTICS")
print("=" * 78)

if len(valid_release) > 0:
    print(valid_release.describe())

print("\nRelease-time statistics:")

if len(valid_time) > 0:
    print(valid_time.describe())

# ------------------------------------------------------------------------------
# Study-level summary
# ------------------------------------------------------------------------------

study_summary = (
    analysis.groupby("study_id")
    .agg(
        observation_count=("release_percent", "size"),
        polymer=("polymer", "first"),
        drug=("drug", "first"),
        minimum_time_h=("release_time_h", "min"),
        maximum_time_h=("release_time_h", "max"),
        minimum_release_pct=("release_percent", "min"),
        maximum_release_pct=("release_percent", "max"),
        mean_release_pct=("release_percent", "mean"),
        repeated_measurements=("repeated_measurement", "max"),
        non_monotonic_release=("non_monotonic_release", "max"),
    )
    .reset_index()
)

# ------------------------------------------------------------------------------
# Polymer-level summary
# ------------------------------------------------------------------------------

polymer_summary = (
    analysis[
        analysis["final_analysis_inclusion"]
    ]
    .groupby("polymer")
    .agg(
        observation_count=("release_percent", "size"),
        study_count=("study_id", "nunique"),
        mean_release_pct=("release_percent", "mean"),
        median_release_pct=("release_percent", "median"),
        min_release_pct=("release_percent", "min"),
        max_release_pct=("release_percent", "max"),
    )
    .reset_index()
    .sort_values("study_count", ascending=False)
)

# ------------------------------------------------------------------------------
# Drug-level summary
# ------------------------------------------------------------------------------

drug_summary = (
    analysis[
        analysis["final_analysis_inclusion"]
    ]
    .groupby("drug")
    .agg(
        observation_count=("release_percent", "size"),
        study_count=("study_id", "nunique"),
        mean_release_pct=("release_percent", "mean"),
        median_release_pct=("release_percent", "median"),
        min_release_pct=("release_percent", "min"),
        max_release_pct=("release_percent", "max"),
    )
    .reset_index()
    .sort_values("study_count", ascending=False)
)

# ------------------------------------------------------------------------------
# Save outputs
# ------------------------------------------------------------------------------

final_file = OUTPUT_DIR / "12_final_analysis_dataset.csv"
study_file = OUTPUT_DIR / "12_study_level_summary.csv"
polymer_file = OUTPUT_DIR / "12_polymer_level_summary.csv"
drug_file = OUTPUT_DIR / "12_drug_level_summary.csv"

analysis.to_csv(
    final_file,
    index=False
)

study_summary.to_csv(
    study_file,
    index=False
)

polymer_summary.to_csv(
    polymer_file,
    index=False
)

drug_summary.to_csv(
    drug_file,
    index=False
)

# ------------------------------------------------------------------------------
# Final display
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("FINAL ANALYSIS DATASET")
print("=" * 78)

print(
    analysis[
        [
            "study_id",
            "polymer",
            "drug",
            "release_time_h",
            "release_percent",
            "study_observation_count",
            "repeated_measurement",
            "final_analysis_inclusion"
        ]
    ].to_string(index=False)
)

print("\n" + "=" * 78)
print("FILES CREATED")
print("=" * 78)

print(final_file)
print(study_file)
print(polymer_file)
print(drug_file)

print("\n" + "=" * 78)
print("CODE 12 COMPLETE")
print("=" * 78)
```

```python
# ==============================================================================
# CODE 13 — RELEASE-CURVE STRUCTURE & STUDY-LEVEL KINETIC READINESS
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 78)
print("CODE 13 — RELEASE-CURVE STRUCTURE & STUDY-LEVEL KINETIC READINESS")
print("=" * 78)

BASE = Path("AI_Polymer_Drug_Release")

INPUT_FILE = (
    BASE / "data/final/12_final_analysis_dataset.csv"
)

OUTPUT_DIR = BASE / "data/analysis/13_release_curve"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"\nCould not find:\n{INPUT_FILE}\n\n"
        "Run CODE 12 first."
    )

df = pd.read_csv(INPUT_FILE)

# ------------------------------------------------------------------------------
# Keep only final analytical observations
# ------------------------------------------------------------------------------

if "final_analysis_inclusion" in df.columns:
    df = df[
        df["final_analysis_inclusion"] == True
    ].copy()

print("\nAnalytical observations:", len(df))

# ------------------------------------------------------------------------------
# Required columns
# ------------------------------------------------------------------------------

required = [
    "study_id",
    "release_time_h",
    "release_percent",
    "polymer",
    "drug"
]

missing = [
    c for c in required
    if c not in df.columns
]

if missing:
    raise KeyError(
        f"Missing required columns: {missing}"
    )

# ------------------------------------------------------------------------------
# Numeric conversion
# ------------------------------------------------------------------------------

df["release_time_h"] = pd.to_numeric(
    df["release_time_h"],
    errors="coerce"
)

df["release_percent"] = pd.to_numeric(
    df["release_percent"],
    errors="coerce"
)

df = df.dropna(
    subset=[
        "study_id",
        "release_time_h",
        "release_percent"
    ]
).copy()

# ------------------------------------------------------------------------------
# Sort
# ------------------------------------------------------------------------------

df = df.sort_values(
    ["study_id", "release_time_h"]
).reset_index(drop=True)

# ------------------------------------------------------------------------------
# Study-level curve characterization
# ------------------------------------------------------------------------------

records = []

for study_id, g in df.groupby("study_id"):

    g = g.sort_values("release_time_h").copy()

    times = g["release_time_h"].to_numpy()
    releases = g["release_percent"].to_numpy()

    n = len(g)

    first_time = float(times[0])
    last_time = float(times[-1])

    first_release = float(releases[0])
    last_release = float(releases[-1])

    time_span = last_time - first_time
    release_change = last_release - first_release

    # --------------------------------------------------------------------------
    # Monotonicity
    # --------------------------------------------------------------------------

    if n > 1:
        diffs = np.diff(releases)

        decreasing_steps = int(
            np.sum(diffs < -1e-9)
        )

        increasing_steps = int(
            np.sum(diffs > 1e-9)
        )

        monotonic_non_decreasing = (
            decreasing_steps == 0
        )

    else:
        decreasing_steps = 0
        increasing_steps = 0
        monotonic_non_decreasing = np.nan

    # --------------------------------------------------------------------------
    # Approximate average release rate
    # --------------------------------------------------------------------------

    if time_span > 0:
        average_rate = (
            release_change / time_span
        )
    else:
        average_rate = np.nan

    # --------------------------------------------------------------------------
    # Curve classification
    # --------------------------------------------------------------------------

    if n == 1:
        curve_type = "SINGLE_POINT"

    elif n == 2:
        curve_type = "TWO_POINT"

    else:
        curve_type = "MULTI_POINT_CURVE"

    # --------------------------------------------------------------------------
    # Kinetic readiness
    # --------------------------------------------------------------------------

    if n >= 4 and time_span > 0:
        kinetic_readiness = "HIGH"

    elif n == 3 and time_span > 0:
        kinetic_readiness = "MODERATE"

    elif n == 2 and time_span > 0:
        kinetic_readiness = "LOW"

    else:
        kinetic_readiness = "NOT_SUITABLE"

    # --------------------------------------------------------------------------
    # Potential burst release
    # --------------------------------------------------------------------------

    burst_release_flag = False

    if n >= 3:

        early_mask = times <= 1.0

        if early_mask.sum() >= 1:

            early_max = releases[early_mask].max()

            if early_max >= 30:
                burst_release_flag = True

    # --------------------------------------------------------------------------
    # Study metadata
    # --------------------------------------------------------------------------

    polymer = g["polymer"].iloc[0]
    drug = g["drug"].iloc[0]

    records.append({
        "study_id": study_id,
        "polymer": polymer,
        "drug": drug,
        "number_of_timepoints": n,
        "first_time_h": first_time,
        "last_time_h": last_time,
        "time_span_h": time_span,
        "first_release_pct": first_release,
        "last_release_pct": last_release,
        "release_change_pct": release_change,
        "average_release_rate_pct_per_h": average_rate,
        "increasing_steps": increasing_steps,
        "decreasing_steps": decreasing_steps,
        "monotonic_non_decreasing": monotonic_non_decreasing,
        "curve_type": curve_type,
        "kinetic_readiness": kinetic_readiness,
        "burst_release_flag": burst_release_flag
    })

study_curves = pd.DataFrame(records)

# ------------------------------------------------------------------------------
# Print study-level results
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("STUDY-LEVEL RELEASE CURVE STRUCTURE")
print("=" * 78)

print(
    study_curves[
        [
            "study_id",
            "number_of_timepoints",
            "first_time_h",
            "last_time_h",
            "time_span_h",
            "first_release_pct",
            "last_release_pct",
            "release_change_pct",
            "curve_type",
            "kinetic_readiness",
            "burst_release_flag"
        ]
    ].to_string(index=False)
)

# ------------------------------------------------------------------------------
# Distribution of curve types
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("CURVE TYPE DISTRIBUTION")
print("=" * 78)

print(
    study_curves["curve_type"]
    .value_counts()
)

print("\nKinetic readiness:")
print(
    study_curves["kinetic_readiness"]
    .value_counts()
)

print("\nBurst-release flags:")
print(
    study_curves["burst_release_flag"]
    .value_counts()
)

# ------------------------------------------------------------------------------
# Monotonicity
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("MONOTONICITY")
print("=" * 78)

print(
    study_curves[
        [
            "study_id",
            "decreasing_steps",
            "monotonic_non_decreasing"
        ]
    ].to_string(index=False)
)

# ------------------------------------------------------------------------------
# Study contribution
# ------------------------------------------------------------------------------

study_counts = (
    df.groupby("study_id")
    .size()
    .reset_index(
        name="observation_count"
    )
    .sort_values(
        "observation_count",
        ascending=False
    )
)

print("\n" + "=" * 78)
print("OBSERVATION CONTRIBUTION BY STUDY")
print("=" * 78)

print(
    study_counts.to_string(index=False)
)

# ------------------------------------------------------------------------------
# Concentration of observations
# ------------------------------------------------------------------------------

total_obs = len(df)

top3_obs = (
    study_counts
    .head(3)["observation_count"]
    .sum()
)

top3_fraction = (
    top3_obs / total_obs
    if total_obs > 0
    else np.nan
)

print(
    f"\nTop 3 studies contribute "
    f"{top3_obs}/{total_obs} observations "
    f"({top3_fraction * 100:.1f}%)."
)

# ------------------------------------------------------------------------------
# Timepoint-level table
# ------------------------------------------------------------------------------

curve_points = df[
    [
        "study_id",
        "polymer",
        "drug",
        "release_time_h",
        "release_percent"
    ]
].copy()

curve_points["time_log10"] = np.log10(
    curve_points["release_time_h"]
    .replace(0, np.nan)
)

# ------------------------------------------------------------------------------
# Save outputs
# ------------------------------------------------------------------------------

study_file = (
    OUTPUT_DIR /
    "13_study_level_release_curve_summary.csv"
)

points_file = (
    OUTPUT_DIR /
    "13_release_curve_points.csv"
)

contribution_file = (
    OUTPUT_DIR /
    "13_study_observation_contribution.csv"
)

study_curves.to_csv(
    study_file,
    index=False
)

curve_points.to_csv(
    points_file,
    index=False
)

study_counts.to_csv(
    contribution_file,
    index=False
)

print("\n" + "=" * 78)
print("FILES CREATED")
print("=" * 78)

print(study_file)
print(points_file)
print(contribution_file)

print("\n" + "=" * 78)
print("CODE 13 COMPLETE")
print("=" * 78)
```

```python
# ==============================================================================
# CODE 14 — RELEASE KINETIC MODELING & MECHANISM ANALYSIS
# ==============================================================================

import os
import warnings
import numpy as np
import pandas as pd

from scipy.optimize import curve_fit
from scipy.stats import linregress

warnings.filterwarnings("ignore")

BASE = "AI_Polymer_Drug_Release"

INPUT = os.path.join(
    BASE,
    "data",
    "final",
    "12_final_analysis_dataset.csv"
)

OUTDIR = os.path.join(
    BASE,
    "data",
    "analysis",
    "14_release_kinetics"
)

os.makedirs(OUTDIR, exist_ok=True)

print("=" * 78)
print("CODE 14 — RELEASE KINETIC MODELING & MECHANISM ANALYSIS")
print("=" * 78)

# ------------------------------------------------------------------------------
# 1. LOAD FINAL DATASET
# ------------------------------------------------------------------------------

df = pd.read_csv(INPUT)

print(f"\nInput rows: {len(df)}")
print("Columns:")
print(list(df.columns))

# ------------------------------------------------------------------------------
# 2. DETECT REQUIRED COLUMNS
# ------------------------------------------------------------------------------

def find_col(df, candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None

study_col = find_col(df, [
    "study_id",
    "study_id_std"
])

time_col = find_col(df, [
    "release_time_h",
    "release_time_std",
    "time_h"
])

release_col = find_col(df, [
    "release_percent",
    "release_percent_std",
    "cumulative_release_pct"
])

if study_col is None or time_col is None or release_col is None:
    raise KeyError(
        "Required columns not found.\n"
        f"Study={study_col}, Time={time_col}, Release={release_col}\n"
        f"Available columns: {list(df.columns)}"
    )

print("\nDetected columns:")
print("Study:", study_col)
print("Time:", time_col)
print("Release:", release_col)

# ------------------------------------------------------------------------------
# 3. CLEAN DATA
# ------------------------------------------------------------------------------

work = df[[study_col, time_col, release_col]].copy()

work.columns = ["study_id", "time_h", "release_percent"]

work["time_h"] = pd.to_numeric(work["time_h"], errors="coerce")
work["release_percent"] = pd.to_numeric(
    work["release_percent"],
    errors="coerce"
)

work = work.dropna(
    subset=["study_id", "time_h", "release_percent"]
)

work = work[
    (work["time_h"] >= 0) &
    (work["release_percent"] >= 0) &
    (work["release_percent"] <= 100)
].copy()

work = work.sort_values(
    ["study_id", "time_h"]
).reset_index(drop=True)

# ------------------------------------------------------------------------------
# 4. IDENTIFY MULTI-POINT STUDIES
# ------------------------------------------------------------------------------

study_counts = (
    work.groupby("study_id")
    .size()
    .reset_index(name="number_of_timepoints")
)

kinetic_studies = study_counts[
    study_counts["number_of_timepoints"] >= 3
].copy()

single_point_studies = study_counts[
    study_counts["number_of_timepoints"] < 3
].copy()

print("\n" + "=" * 78)
print("KINETIC ELIGIBILITY")
print("=" * 78)

print("\nStudy-level timepoint counts:")
print(study_counts.to_string(index=False))

print(
    f"\nKinetic studies (>=3 points): "
    f"{len(kinetic_studies)}"
)

print(
    f"Single/insufficient-point studies excluded from kinetic fitting: "
    f"{len(single_point_studies)}"
)

# ------------------------------------------------------------------------------
# 5. MODEL DEFINITIONS
# ------------------------------------------------------------------------------

# Zero-order:
# Q = Q0 + k0*t

def zero_order(t, q0, k0):
    return q0 + k0 * t


# First-order cumulative release:
# Q = Qinf * (1 - exp(-k*t))
#
# This is fitted directly to cumulative release.

def first_order(t, qinf, k1):
    return qinf * (1.0 - np.exp(-k1 * t))


# Higuchi:
# Q = kh * sqrt(t)
#
# For cumulative percentage release, this is an empirical transformation.

def higuchi(t, kh):
    return kh * np.sqrt(t)


# Korsmeyer-Peppas:
# M_t/M_inf = k*t^n
#
# We use fraction released and exclude points at exactly zero release.

def korsmeyer_peppas(t, k, n):
    return k * np.power(t, n)


# ------------------------------------------------------------------------------
# 6. FITTING FUNCTION
# ------------------------------------------------------------------------------

def calculate_metrics(y, yhat, n_parameters):
    y = np.asarray(y, dtype=float)
    yhat = np.asarray(yhat, dtype=float)

    residuals = y - yhat

    sse = np.sum(residuals ** 2)
    mse = np.mean(residuals ** 2)
    rmse = np.sqrt(mse)

    ss_tot = np.sum((y - np.mean(y)) ** 2)

    if ss_tot > 0:
        r2 = 1 - (sse / ss_tot)
    else:
        r2 = np.nan

    n = len(y)

    if n > n_parameters and sse > 0:
        aic = (
            n * np.log(sse / n)
            + 2 * n_parameters
        )
    else:
        aic = np.nan

    return r2, rmse, sse, aic


def fit_model(
    study_id,
    model_name,
    x,
    y,
    function,
    p0,
    bounds
):
    result = {
        "study_id": study_id,
        "model": model_name,
        "n_points": len(x),
        "status": "FAILED",
        "r2": np.nan,
        "rmse": np.nan,
        "sse": np.nan,
        "aic": np.nan,
        "parameter_1": np.nan,
        "parameter_2": np.nan,
        "parameter_1_name": "",
        "parameter_2_name": "",
        "prediction_equation": "",
        "warning": ""
    }

    try:

        if len(x) < 3:
            result["warning"] = "Insufficient timepoints"
            return result

        popt, pcov = curve_fit(
            function,
            x,
            y,
            p0=p0,
            bounds=bounds,
            maxfev=50000
        )

        yhat = function(x, *popt)

        r2, rmse, sse, aic = calculate_metrics(
            y,
            yhat,
            len(popt)
        )

        result["status"] = "SUCCESS"
        result["r2"] = r2
        result["rmse"] = rmse
        result["sse"] = sse
        result["aic"] = aic

        result["parameter_1"] = popt[0]

        if len(popt) > 1:
            result["parameter_2"] = popt[1]

        if model_name == "Zero-order":
            result["parameter_1_name"] = "Q0_percent"
            result["parameter_2_name"] = "k0_percent_per_h"
            result["prediction_equation"] = (
                f"Q = {popt[0]:.6g} + {popt[1]:.6g}t"
            )

        elif model_name == "First-order":
            result["parameter_1_name"] = "Qinf_percent"
            result["parameter_2_name"] = "k1_per_h"
            result["prediction_equation"] = (
                f"Q = {popt[0]:.6g}(1-exp(-{popt[1]:.6g}t))"
            )

        elif model_name == "Higuchi":
            result["parameter_1_name"] = "Kh_percent_per_sqrt_h"
            result["prediction_equation"] = (
                f"Q = {popt[0]:.6g}sqrt(t)"
            )

        elif model_name == "Korsmeyer-Peppas":
            result["parameter_1_name"] = "k"
            result["parameter_2_name"] = "n"
            result["prediction_equation"] = (
                f"Mt/Minf = {popt[0]:.6g} t^{popt[1]:.6g}"
            )

        return result

    except Exception as e:

        result["warning"] = str(e)
        return result


# ------------------------------------------------------------------------------
# 7. FIT MODELS STUDY-BY-STUDY
# ------------------------------------------------------------------------------

results = []
curve_points = []

for study_id in kinetic_studies["study_id"]:

    s = work[
        work["study_id"] == study_id
    ].copy()

    s = s.sort_values("time_h")

    x = s["time_h"].to_numpy(dtype=float)
    y = s["release_percent"].to_numpy(dtype=float)

    # Store points used in modelling
    for i, (tx, yy) in enumerate(zip(x, y), start=1):
        curve_points.append({
            "study_id": study_id,
            "timepoint_number": i,
            "time_h": tx,
            "release_percent": yy
        })

    max_release = float(np.max(y))

    # --------------------------------------------------------------------------
    # ZERO ORDER
    # --------------------------------------------------------------------------

    results.append(
        fit_model(
            study_id,
            "Zero-order",
            x,
            y,
            zero_order,
            p0=[0, max_release / max(np.max(x), 1)],
            bounds=([-np.inf, -np.inf], [np.inf, np.inf])
        )
    )

    # --------------------------------------------------------------------------
    # FIRST ORDER
    # --------------------------------------------------------------------------

    results.append(
        fit_model(
            study_id,
            "First-order",
            x,
            y,
            first_order,
            p0=[min(100, max_release * 1.1), 0.1],
            bounds=([0, 0], [200, np.inf])
        )
    )

    # --------------------------------------------------------------------------
    # HIGUCHI
    # --------------------------------------------------------------------------

    results.append(
        fit_model(
            study_id,
            "Higuchi",
            x,
            y,
            higuchi,
            p0=[max_release / np.sqrt(max(np.max(x), 1))],
            bounds=([0], [np.inf])
        )
    )

    # --------------------------------------------------------------------------
    # KORSMEYER-PEPPAS
    # --------------------------------------------------------------------------
    #
    # Exclude t=0 because log(0) is undefined.
    # Also use Mt/Minf <= 0.6 as the classical early-release region
    # when enough points are available.
    #
    # This is intentionally conservative.

    kp = s[
        (s["time_h"] > 0) &
        (s["release_percent"] > 0)
    ].copy()

    if max_release > 0:
        kp["fraction_release"] = (
            kp["release_percent"] / max_release
        )

    # Classical early-time region
    kp_early = kp[
        kp["fraction_release"] <= 0.60
    ].copy()

    if len(kp_early) >= 3:

        xkp = kp_early["time_h"].to_numpy(dtype=float)
        ykp = kp_early["fraction_release"].to_numpy(dtype=float)

        kp_result = fit_model(
            study_id,
            "Korsmeyer-Peppas",
            xkp,
            ykp,
            korsmeyer_peppas,
            p0=[0.5, 0.5],
            bounds=([0, 0], [np.inf, 2])
        )

        kp_result["n_points_used"] = len(xkp)
        kp_result["kp_region"] = "Mt/Minf <= 0.60"

    else:

        kp_result = {
            "study_id": study_id,
            "model": "Korsmeyer-Peppas",
            "n_points": len(x),
            "n_points_used": len(kp_early),
            "status": "NOT_FITTED",
            "r2": np.nan,
            "rmse": np.nan,
            "sse": np.nan,
            "aic": np.nan,
            "parameter_1": np.nan,
            "parameter_2": np.nan,
            "parameter_1_name": "k",
            "parameter_2_name": "n",
            "prediction_equation": "",
            "warning": (
                "Fewer than 3 usable early-time points "
                "with Mt/Minf <= 0.60"
            ),
            "kp_region": "Mt/Minf <= 0.60"
        }

    results.append(kp_result)


kinetic_results = pd.DataFrame(results)

# ------------------------------------------------------------------------------
# 8. ADD KORSMEYER-PEPPAS INTERPRETATION
# ------------------------------------------------------------------------------

def interpret_n(n):

    if pd.isna(n):
        return "Not determined"

    if n < 0.45:
        return "Fickian diffusion"

    elif n < 0.89:
        return "Anomalous/non-Fickian transport"

    elif abs(n - 0.89) <= 0.05:
        return "Case-II / relaxation-dominated boundary"

    else:
        return "Super Case-II transport"

kinetic_results["mechanism_interpretation"] = (
    kinetic_results.apply(
        lambda r: (
            interpret_n(r["parameter_2"])
            if r["model"] == "Korsmeyer-Peppas"
            else ""
        ),
        axis=1
    )
)

# ------------------------------------------------------------------------------
# 9. MODEL RANKING
# ------------------------------------------------------------------------------

successful = kinetic_results[
    kinetic_results["status"] == "SUCCESS"
].copy()

successful["rank_r2"] = (
    successful.groupby("study_id")["r2"]
    .rank(
        ascending=False,
        method="min"
    )
)

successful["rank_rmse"] = (
    successful.groupby("study_id")["rmse"]
    .rank(
        ascending=True,
        method="min"
    )
)

successful["rank_aic"] = (
    successful.groupby("study_id")["aic"]
    .rank(
        ascending=True,
        method="min"
    )
)

# Composite rank:
# Lower is better.
successful["composite_rank_score"] = (
    successful["rank_r2"]
    + successful["rank_rmse"]
    + successful["rank_aic"].fillna(successful["rank_r2"])
)

successful["model_rank"] = (
    successful.groupby("study_id")["composite_rank_score"]
    .rank(
        ascending=True,
        method="min"
    )
)

# ------------------------------------------------------------------------------
# 10. BEST MODEL PER STUDY
# ------------------------------------------------------------------------------

best_models = (
    successful[
        successful["model_rank"] == 1
    ]
    .sort_values("study_id")
    .copy()
)

# ------------------------------------------------------------------------------
# 11. STUDY-LEVEL KINETIC SUMMARY
# ------------------------------------------------------------------------------

curve_summary = []

for study_id in kinetic_studies["study_id"]:

    s = work[
        work["study_id"] == study_id
    ].sort_values("time_h")

    first_release = float(s["release_percent"].iloc[0])
    last_release = float(s["release_percent"].iloc[-1])

    summary = {
        "study_id": study_id,
        "n_timepoints": len(s),
        "first_time_h": float(s["time_h"].iloc[0]),
        "last_time_h": float(s["time_h"].iloc[-1]),
        "first_release_percent": first_release,
        "last_release_percent": last_release,
        "release_change_percent": last_release - first_release
    }

    best = best_models[
        best_models["study_id"] == study_id
    ]

    if len(best) > 0:

        b = best.iloc[0]

        summary["best_model"] = b["model"]
        summary["best_model_r2"] = b["r2"]
        summary["best_model_rmse"] = b["rmse"]
        summary["best_model_aic"] = b["aic"]

    else:

        summary["best_model"] = ""
        summary["best_model_r2"] = np.nan
        summary["best_model_rmse"] = np.nan
        summary["best_model_aic"] = np.nan

    kp_rows = kinetic_results[
        (kinetic_results["study_id"] == study_id) &
        (kinetic_results["model"] == "Korsmeyer-Peppas") &
        (kinetic_results["status"] == "SUCCESS")
    ]

    if len(kp_rows) > 0:

        kp_row = kp_rows.iloc[0]

        summary["kp_k"] = kp_row["parameter_1"]
        summary["kp_n"] = kp_row["parameter_2"]
        summary["kp_r2"] = kp_row["r2"]
        summary["kp_mechanism"] = kp_row[
            "mechanism_interpretation"
        ]

    else:

        summary["kp_k"] = np.nan
        summary["kp_n"] = np.nan
        summary["kp_r2"] = np.nan
        summary["kp_mechanism"] = "Not determined"

    curve_summary.append(summary)

curve_summary = pd.DataFrame(curve_summary)

# ------------------------------------------------------------------------------
# 12. MODEL VALIDITY WARNINGS
# ------------------------------------------------------------------------------

warnings_rows = []

for study_id in kinetic_studies["study_id"]:

    s = work[
        work["study_id"] == study_id
    ].sort_values("time_h")

    n_points = len(s)

    if n_points < 5:
        warnings_rows.append({
            "study_id": study_id,
            "warning_type": "LOW_TIMEPOINT_COUNT",
            "severity": "HIGH",
            "message": (
                f"Only {n_points} timepoints available; "
                "kinetic model discrimination is limited."
            )
        })

    if len(s) == 4:
        warnings_rows.append({
            "study_id": study_id,
            "warning_type": "FOUR_POINT_CURVE",
            "severity": "HIGH",
            "message": (
                "Only four observations are available; "
                "model ranking should be interpreted cautiously."
            )
        })

    if len(s) == 5:
        warnings_rows.append({
            "study_id": study_id,
            "warning_type": "FIVE_POINT_CURVE",
            "severity": "MODERATE",
            "message": (
                "Five observations permit exploratory kinetic fitting "
                "but provide limited mechanistic discrimination."
            )
        })

warnings_df = pd.DataFrame(warnings_rows)

# ------------------------------------------------------------------------------
# 13. SAVE OUTPUTS
# ------------------------------------------------------------------------------

kinetic_results.to_csv(
    os.path.join(
        OUTDIR,
        "14_kinetic_model_results.csv"
    ),
    index=False
)

best_models.to_csv(
    os.path.join(
        OUTDIR,
        "14_best_model_by_study.csv"
    ),
    index=False
)

curve_summary.to_csv(
    os.path.join(
        OUTDIR,
        "14_study_level_kinetic_summary.csv"
    ),
    index=False
)

pd.DataFrame(curve_points).to_csv(
    os.path.join(
        OUTDIR,
        "14_kinetic_curve_points.csv"
    ),
    index=False
)

warnings_df.to_csv(
    os.path.join(
        OUTDIR,
        "14_kinetic_model_warnings.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# 14. CONSOLE REPORT
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("KINETIC MODELING RESULTS")
print("=" * 78)

print("\nModels fitted:")
print(
    kinetic_results[
        [
            "study_id",
            "model",
            "n_points",
            "status",
            "r2",
            "rmse",
            "aic"
        ]
    ].to_string(index=False)
)

print("\n" + "=" * 78)
print("BEST MODEL BY STUDY")
print("=" * 78)

if len(best_models) > 0:

    print(
        best_models[
            [
                "study_id",
                "model",
                "r2",
                "rmse",
                "aic"
            ]
        ].to_string(index=False)
    )

else:

    print("No model could be selected.")

print("\n" + "=" * 78)
print("KORSMEYER–PEPPAS RESULTS")
print("=" * 78)

kp_display = kinetic_results[
    kinetic_results["model"] == "Korsmeyer-Peppas"
][
    [
        "study_id",
        "status",
        "n_points",
        "parameter_1",
        "parameter_2",
        "r2",
        "mechanism_interpretation",
        "warning"
    ]
]

print(kp_display.to_string(index=False))

print("\n" + "=" * 78)
print("STUDY-LEVEL SUMMARY")
print("=" * 78)

print(curve_summary.to_string(index=False))

print("\n" + "=" * 78)
print("CODE 14 COMPLETE")
print("=" * 78)

print(f"""
Kinetic studies analyzed: {len(kinetic_studies)}
Studies excluded from kinetic fitting: {len(single_point_studies)}

Output directory:
{OUTDIR}

Files created:
- 14_kinetic_model_results.csv
- 14_best_model_by_study.csv
- 14_study_level_kinetic_summary.csv
- 14_kinetic_curve_points.csv
- 14_kinetic_model_warnings.csv
""")
```

```python
# ==============================================================================
# CODE 15 — STUDY-LEVEL INDEPENDENCE, HETEROGENEITY & ANALYTICAL ELIGIBILITY
# ==============================================================================
#
# Purpose:
#   1. Assess independence of quantitative observations.
#   2. Quantify contribution of repeated-measurement studies.
#   3. Create study-level analytical summaries.
#   4. Identify which analyses are defensible.
#   5. Flag pseudoreplication risks.
#   6. Assess comparability of polymer/drug/formulation metadata.
#
# Input:
#   AI_Polymer_Drug_Release/data/final/12_final_analysis_dataset.csv
#
# Outputs:
#   AI_Polymer_Drug_Release/data/analysis/15_independence/
#
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

# ------------------------------------------------------------------------------
# PATHS
# ------------------------------------------------------------------------------

BASE = Path("AI_Polymer_Drug_Release")

INPUT = BASE / "data/final/12_final_analysis_dataset.csv"
OUTDIR = BASE / "data/analysis/15_independence"
OUTDIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------------------------

df = pd.read_csv(INPUT)

print("=" * 78)
print("CODE 15 — STUDY-LEVEL INDEPENDENCE & HETEROGENEITY")
print("=" * 78)

print("\nInput rows:", len(df))
print("Input columns:", len(df.columns))
print("Columns:")
print(list(df.columns))

# ------------------------------------------------------------------------------
# COLUMN DETECTION
# ------------------------------------------------------------------------------

def find_col(df, candidates, required=True):
    for c in candidates:
        if c in df.columns:
            return c
    if required:
        raise KeyError(
            f"Required column not found. Tried: {candidates}\n"
            f"Available columns: {list(df.columns)}"
        )
    return None

study_col = find_col(df, ["study_id", "study_id_std"])
polymer_col = find_col(
    df,
    ["polymer", "polymer_std", "confirmed_polymers"],
    required=False
)
drug_col = find_col(
    df,
    ["drug", "drug_std", "confirmed_drugs"],
    required=False
)
time_col = find_col(
    df,
    ["release_time_h", "release_time_std"],
    required=False
)
release_col = find_col(
    df,
    ["release_percent", "release_percent_std"],
    required=False
)

print("\nDetected columns:")
print("Study:", study_col)
print("Polymer:", polymer_col)
print("Drug:", drug_col)
print("Time:", time_col)
print("Release:", release_col)

# ------------------------------------------------------------------------------
# STANDARDIZE
# ------------------------------------------------------------------------------

work = df.copy()

work["study_id_analysis"] = work[study_col].astype(str).str.strip()

if polymer_col:
    work["polymer_analysis"] = (
        work[polymer_col]
        .replace(["", "nan", "NaN", "UNKNOWN", "Unknown"], np.nan)
    )
else:
    work["polymer_analysis"] = np.nan

if drug_col:
    work["drug_analysis"] = (
        work[drug_col]
        .replace(["", "nan", "NaN", "UNKNOWN", "Unknown"], np.nan)
    )
else:
    work["drug_analysis"] = np.nan

if time_col:
    work["release_time_analysis"] = pd.to_numeric(
        work[time_col], errors="coerce"
    )
else:
    work["release_time_analysis"] = np.nan

if release_col:
    work["release_percent_analysis"] = pd.to_numeric(
        work[release_col], errors="coerce"
    )
else:
    work["release_percent_analysis"] = np.nan

# Keep only final-analysis rows where such a flag exists
if "final_analysis_inclusion" in work.columns:
    inclusion = work["final_analysis_inclusion"].astype(str).str.lower()
    work = work[
        inclusion.isin(["true", "1", "yes"])
    ].copy()

print("\nFinal analytical observations:", len(work))

# ------------------------------------------------------------------------------
# BASIC VALIDATION
# ------------------------------------------------------------------------------

work["valid_time"] = (
    work["release_time_analysis"].notna()
    & (work["release_time_analysis"] >= 0)
)

work["valid_release"] = (
    work["release_percent_analysis"].notna()
    & work["release_percent_analysis"].between(0, 100)
)

work["analytically_valid"] = (
    work["valid_time"] & work["valid_release"]
)

valid = work[work["analytically_valid"]].copy()

print("\nValid observations:", len(valid))
print("Invalid observations:", len(work) - len(valid))

# ------------------------------------------------------------------------------
# STUDY-LEVEL OBSERVATION COUNTS
# ------------------------------------------------------------------------------

study_counts = (
    valid.groupby("study_id_analysis")
    .size()
    .reset_index(name="observation_count")
    .sort_values(
        ["observation_count", "study_id_analysis"],
        ascending=[False, True]
    )
)

study_counts["repeated_measurement"] = (
    study_counts["observation_count"] > 1
)

study_counts["proportion_of_all_observations"] = (
    study_counts["observation_count"] / len(valid)
)

print("\n" + "=" * 78)
print("OBSERVATION CONTRIBUTION BY STUDY")
print("=" * 78)

print(study_counts.to_string(index=False))

# ------------------------------------------------------------------------------
# STUDY-LEVEL SUMMARY
# ------------------------------------------------------------------------------

summary_rows = []

for study_id, g in valid.groupby("study_id_analysis"):

    g = g.sort_values("release_time_analysis")

    row = {
        "study_id": study_id,
        "observation_count": len(g),
        "repeated_measurement": len(g) > 1,
        "first_time_h": g["release_time_analysis"].min(),
        "last_time_h": g["release_time_analysis"].max(),
        "time_span_h": (
            g["release_time_analysis"].max()
            - g["release_time_analysis"].min()
        ),
        "first_release_percent": g["release_percent_analysis"].iloc[0],
        "last_release_percent": g["release_percent_analysis"].iloc[-1],
        "mean_release_percent": g["release_percent_analysis"].mean(),
        "median_release_percent": g["release_percent_analysis"].median(),
        "min_release_percent": g["release_percent_analysis"].min(),
        "max_release_percent": g["release_percent_analysis"].max(),
    }

    if polymer_col:
        polymers = g["polymer_analysis"].dropna().astype(str).unique()
        row["polymer"] = ";".join(sorted(polymers)) if len(polymers) else np.nan
    else:
        row["polymer"] = np.nan

    if drug_col:
        drugs = g["drug_analysis"].dropna().astype(str).unique()
        row["drug"] = ";".join(sorted(drugs)) if len(drugs) else np.nan
    else:
        row["drug"] = np.nan

    # Curve eligibility
    row["kinetic_eligible"] = len(g) >= 3

    # Monotonicity
    if len(g) >= 2:
        diffs = np.diff(g["release_percent_analysis"].values)
        row["decreasing_steps"] = int(np.sum(diffs < 0))
        row["monotonic_non_decreasing"] = bool(np.all(diffs >= 0))
    else:
        row["decreasing_steps"] = np.nan
        row["monotonic_non_decreasing"] = np.nan

    summary_rows.append(row)

study_summary = pd.DataFrame(summary_rows)

# ------------------------------------------------------------------------------
# STUDY-LEVEL HETEROGENEITY
# ------------------------------------------------------------------------------

study_summary["release_change_percent"] = (
    study_summary["last_release_percent"]
    - study_summary["first_release_percent"]
)

study_summary["release_range_percent"] = (
    study_summary["max_release_percent"]
    - study_summary["min_release_percent"]
)

# ------------------------------------------------------------------------------
# DOMINANCE / PSEUDOREPLICATION
# ------------------------------------------------------------------------------

top3 = (
    study_counts.head(3)["observation_count"].sum()
)

top5 = (
    study_counts.head(5)["observation_count"].sum()
)

total_obs = len(valid)
n_studies = study_counts["study_id_analysis"].nunique()

dominance = pd.DataFrame({
    "metric": [
        "total_observations",
        "unique_studies",
        "top_1_observation_count",
        "top_3_observation_count",
        "top_5_observation_count",
        "top_3_observation_fraction",
        "top_5_observation_fraction",
        "repeated_measurement_studies",
        "single_observation_studies",
    ],
    "value": [
        total_obs,
        n_studies,
        study_counts.iloc[0]["observation_count"]
            if len(study_counts) else 0,
        top3,
        top5,
        top3 / total_obs if total_obs else np.nan,
        top5 / total_obs if total_obs else np.nan,
        int(study_counts["repeated_measurement"].sum()),
        int((~study_counts["repeated_measurement"]).sum()),
    ]
})

# ------------------------------------------------------------------------------
# EFFECTIVE INDEPENDENT SAMPLE SIZE
# ------------------------------------------------------------------------------

# Number of independent units is the number of studies, not observations.
effective_n = n_studies

# Naive observation-to-study ratio
inflation_factor = (
    total_obs / effective_n
    if effective_n > 0 else np.nan
)

effective_sample = pd.DataFrame({
    "metric": [
        "independent_study_units",
        "observation_units",
        "naive_observation_to_study_ratio",
        "pseudoreplication_risk",
    ],
    "value": [
        effective_n,
        total_obs,
        inflation_factor,
        (
            "HIGH"
            if inflation_factor > 1.5
            else "MODERATE"
            if inflation_factor > 1
            else "LOW"
        )
    ]
})

# ------------------------------------------------------------------------------
# POLYMER COMPARABILITY
# ------------------------------------------------------------------------------

if polymer_col:

    polymer_study_counts = (
        study_summary
        .dropna(subset=["polymer"])
        .groupby("polymer")
        .agg(
            study_count=("study_id", "nunique"),
            observation_count=("observation_count", "sum")
        )
        .reset_index()
        .sort_values("study_count", ascending=False)
    )

else:
    polymer_study_counts = pd.DataFrame(
        columns=["polymer", "study_count", "observation_count"]
    )

# ------------------------------------------------------------------------------
# DRUG COMPARABILITY
# ------------------------------------------------------------------------------

if drug_col:

    drug_study_counts = (
        study_summary
        .dropna(subset=["drug"])
        .groupby("drug")
        .agg(
            study_count=("study_id", "nunique"),
            observation_count=("observation_count", "sum")
        )
        .reset_index()
        .sort_values("study_count", ascending=False)
    )

else:
    drug_study_counts = pd.DataFrame(
        columns=["drug", "study_count", "observation_count"]
    )

# ------------------------------------------------------------------------------
# ANALYTICAL ELIGIBILITY
# ------------------------------------------------------------------------------

eligibility = []

for _, row in study_summary.iterrows():

    reasons = []

    if row["observation_count"] < 3:
        reasons.append("INSUFFICIENT_TIMEPOINTS_FOR_KINETIC_MODELING")

    if pd.isna(row["polymer"]):
        reasons.append("MISSING_POLYMER_METADATA")

    if pd.isna(row["drug"]):
        reasons.append("MISSING_DRUG_METADATA")

    if row["observation_count"] == 1:
        reasons.append("SINGLE_OBSERVATION_STUDY")

    eligibility.append({
        "study_id": row["study_id"],
        "observation_count": row["observation_count"],
        "kinetic_eligible": row["kinetic_eligible"],
        "polymer_available": not pd.isna(row["polymer"]),
        "drug_available": not pd.isna(row["drug"]),
        "analytical_notes": (
            "; ".join(reasons)
            if reasons else
            "SUITABLE_FOR_STUDY_LEVEL_ANALYSIS"
        )
    })

eligibility_df = pd.DataFrame(eligibility)

# ------------------------------------------------------------------------------
# ANALYTICAL RECOMMENDATION
# ------------------------------------------------------------------------------

recommendations = []

recommendations.append({
    "analysis": "Observation-level correlation",
    "status": "CAUTION",
    "reason": (
        "Repeated observations from the same study violate simple "
        "independence assumptions."
    )
})

recommendations.append({
    "analysis": "Study-level descriptive analysis",
    "status": "RECOMMENDED",
    "reason": (
        "Study is the independent unit and is appropriate for "
        "descriptive synthesis."
    )
})

recommendations.append({
    "analysis": "Kinetic modeling",
    "status": "LIMITED",
    "reason": (
        f"Only {int(study_summary['kinetic_eligible'].sum())} studies "
        "contain at least three quantitative timepoints."
    )
})

recommendations.append({
    "analysis": "Polymer-group statistical comparison",
    "status": "LIKELY_LIMITED",
    "reason": (
        "Polymer formulations and drugs differ across studies, "
        "creating substantial between-study heterogeneity."
    )
})

recommendations.append({
    "analysis": "Drug-group comparison",
    "status": "LIKELY_LIMITED",
    "reason": (
        "Drug identity, formulation, and experimental conditions "
        "are not standardized across studies."
    )
})

recommendations.append({
    "analysis": "Machine learning",
    "status": "NOT_RECOMMENDED",
    "reason": (
        "The current dataset contains only "
        f"{n_studies} independent studies and "
        f"{total_obs} observations; this is too small and heterogeneous "
        "for reliable predictive ML."
    )
})

recommendations_df = pd.DataFrame(recommendations)

# ------------------------------------------------------------------------------
# SAVE OUTPUTS
# ------------------------------------------------------------------------------

study_counts.to_csv(
    OUTDIR / "15_study_observation_contribution.csv",
    index=False
)

study_summary.to_csv(
    OUTDIR / "15_study_level_summary.csv",
    index=False
)

dominance.to_csv(
    OUTDIR / "15_observation_dominance_report.csv",
    index=False
)

effective_sample.to_csv(
    OUTDIR / "15_effective_sample_size_report.csv",
    index=False
)

eligibility_df.to_csv(
    OUTDIR / "15_study_analytical_eligibility.csv",
    index=False
)

polymer_study_counts.to_csv(
    OUTDIR / "15_polymer_comparability.csv",
    index=False
)

drug_study_counts.to_csv(
    OUTDIR / "15_drug_comparability.csv",
    index=False
)

recommendations_df.to_csv(
    OUTDIR / "15_analysis_recommendations.csv",
    index=False
)

# ------------------------------------------------------------------------------
# CONSOLE SUMMARY
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("INDEPENDENCE SUMMARY")
print("=" * 78)

print("Independent studies:", n_studies)
print("Observation units:", total_obs)
print(
    "Repeated-measurement studies:",
    int(study_counts["repeated_measurement"].sum())
)
print(
    "Single-observation studies:",
    int((~study_counts["repeated_measurement"]).sum())
)

print(
    f"Top 3 studies contribute {top3}/{total_obs} "
    f"observations ({100*top3/total_obs:.1f}%)."
)

print(
    f"Observation-to-study ratio: {inflation_factor:.2f}"
)

print("\n" + "=" * 78)
print("ANALYTICAL RECOMMENDATIONS")
print("=" * 78)

print(recommendations_df.to_string(index=False))

print("\n" + "=" * 78)
print("CODE 15 COMPLETE")
print("=" * 78)

print("\nFiles created:")
for f in sorted(OUTDIR.glob("15_*.csv")):
    print("-", f)

print("=" * 78)
```

```python
# ==============================================================================
# CODE 16 — POLYMER–DRUG COMPARABILITY & EVIDENCE MATRIX
# ==============================================================================
#
# Purpose:
#   1. Normalize polymer and drug labels.
#   2. Identify polymer families and drug families.
#   3. Map polymer–drug combinations by study.
#   4. Separate single-point from multi-point evidence.
#   5. Assess cross-study comparability.
#   6. Identify defensible and non-defensible comparisons.
#   7. Generate a study-level evidence matrix for Results synthesis.
#
# Input:
#   AI_Polymer_Drug_Release/data/final/12_final_analysis_dataset.csv
#
# Output:
#   AI_Polymer_Drug_Release/data/analysis/16_comparability/
#
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

# ------------------------------------------------------------------------------
# PATHS
# ------------------------------------------------------------------------------

BASE = Path("AI_Polymer_Drug_Release")

INPUT = BASE / "data/final/12_final_analysis_dataset.csv"
OUTDIR = BASE / "data/analysis/16_comparability"
OUTDIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------------------------
# LOAD
# ------------------------------------------------------------------------------

df = pd.read_csv(INPUT)

print("=" * 78)
print("CODE 16 — POLYMER–DRUG COMPARABILITY & EVIDENCE MATRIX")
print("=" * 78)

print("\nInput rows:", len(df))
print("Input columns:", len(df.columns))

# ------------------------------------------------------------------------------
# COLUMN DETECTION
# ------------------------------------------------------------------------------

def find_col(df, candidates, required=True):
    for c in candidates:
        if c in df.columns:
            return c
    if required:
        raise KeyError(
            f"Could not find required column. Tried: {candidates}\n"
            f"Available columns: {list(df.columns)}"
        )
    return None

study_col = find_col(df, ["study_id", "study_id_std"])
polymer_col = find_col(
    df,
    ["polymer", "polymer_std", "confirmed_polymers"],
    required=False
)
drug_col = find_col(
    df,
    ["drug", "drug_std", "confirmed_drugs"],
    required=False
)
time_col = find_col(
    df,
    ["release_time_h", "release_time_std"],
    required=False
)
release_col = find_col(
    df,
    ["release_percent", "release_percent_std"],
    required=False
)

print("\nDetected columns:")
print("Study:", study_col)
print("Polymer:", polymer_col)
print("Drug:", drug_col)
print("Time:", time_col)
print("Release:", release_col)

# ------------------------------------------------------------------------------
# FINAL ANALYSIS FILTER
# ------------------------------------------------------------------------------

work = df.copy()

if "final_analysis_inclusion" in work.columns:

    flag = (
        work["final_analysis_inclusion"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    work = work[
        flag.isin(["true", "1", "yes"])
    ].copy()

print("\nFinal analysis rows:", len(work))

# ------------------------------------------------------------------------------
# STANDARDIZED FIELDS
# ------------------------------------------------------------------------------

work["study_id"] = work[study_col].astype(str).str.strip()

if polymer_col:
    work["polymer_raw"] = work[polymer_col].astype(str)
else:
    work["polymer_raw"] = "UNKNOWN"

if drug_col:
    work["drug_raw"] = work[drug_col].astype(str)
else:
    work["drug_raw"] = "UNKNOWN"

if time_col:
    work["time_h"] = pd.to_numeric(
        work[time_col], errors="coerce"
    )
else:
    work["time_h"] = np.nan

if release_col:
    work["release_percent"] = pd.to_numeric(
        work[release_col], errors="coerce"
    )
else:
    work["release_percent"] = np.nan

# ------------------------------------------------------------------------------
# POLYMER NORMALIZATION
# ------------------------------------------------------------------------------

POLYMER_MAP = {
    "poly(lactic-co-glycolic acid)": "PLGA",
    "poly(lactic-co-glycolic acid) (plga)": "PLGA",
    "polylactic-co-glycolic acid": "PLGA",
    "plga": "PLGA",

    "polycaprolactone": "PCL",
    "poly(ε-caprolactone)": "PCL",
    "pcl": "PCL",

    "polyethylene glycol": "PEG",
    "poly(ethylene glycol)": "PEG",
    "peg": "PEG",

    "poly(lactic acid)": "PLA",
    "polylactic acid": "PLA",
    "pla": "PLA",

    "polyvinyl alcohol": "PVA",
    "poly(vinyl alcohol)": "PVA",
    "pva": "PVA",

    "polyvinylpyrrolidone": "PVP",
    "poly(vinylpyrrolidone)": "PVP",
    "pvp": "PVP",

    "chitosan": "Chitosan",
    "cellulose": "Cellulose",
    "polyester": "Polyester",
    "polypeptide": "Polypeptide",
}

def normalize_polymer_token(x):

    if pd.isna(x):
        return None

    x = str(x).strip()

    if x.lower() in ["nan", "none", "", "unknown"]:
        return None

    key = x.lower()

    return POLYMER_MAP.get(key, x)

def split_polymer_string(x):

    if pd.isna(x):
        return []

    x = str(x)

    if x.lower() in ["nan", "none", "", "unknown"]:
        return []

    # Current pipeline uses semicolon-delimited polymer names
    tokens = re.split(r"[;|,]+", x)

    result = []

    for token in tokens:

        token = token.strip()

        if not token:
            continue

        normalized = normalize_polymer_token(token)

        if normalized and normalized not in result:
            result.append(normalized)

    return result

work["polymer_list"] = work["polymer_raw"].apply(split_polymer_string)

work["polymer_normalized"] = work["polymer_list"].apply(
    lambda x: ";".join(sorted(x)) if x else np.nan
)

work["polymer_count"] = work["polymer_list"].apply(len)

# ------------------------------------------------------------------------------
# DRUG NORMALIZATION
# ------------------------------------------------------------------------------

DRUG_MAP = {
    "dexamethasone": "Dexamethasone",
    "dexamethasone phosphate": "Dexamethasone",
    "paclitaxel": "Paclitaxel",
    "sirolimus": "Sirolimus",
    "tacrolimus": "Tacrolimus",
    "deferoxamine": "Deferoxamine",
    "clindamycin": "Clindamycin",
    "clindamycin phosphate": "Clindamycin",
    "glipizide": "Glipizide",
    "donepezil": "Donepezil",
    "simvastatin": "Simvastatin",
    "indomethacin": "Indomethacin",
    "ciprofloxacin": "Ciprofloxacin",
    "salinomycin": "Salinomycin",
    "artemisinin": "Artemisinin",
    "brimonidine": "Brimonidine",
    "brimonidine tartrate": "Brimonidine",
    "modafinil": "Modafinil",
    "azithromycin": "Azithromycin",
}

def split_drug_string(x):

    if pd.isna(x):
        return []

    x = str(x)

    if x.lower() in ["nan", "none", "", "unknown"]:
        return []

    tokens = re.split(r"[;|,]+", x)

    result = []

    for token in tokens:

        token = token.strip()

        if not token:
            continue

        key = token.lower()

        normalized = DRUG_MAP.get(key, token)

        if normalized not in result:
            result.append(normalized)

    return result

work["drug_list"] = work["drug_raw"].apply(split_drug_string)

work["drug_normalized"] = work["drug_list"].apply(
    lambda x: ";".join(sorted(x)) if x else np.nan
)

work["drug_count"] = work["drug_list"].apply(len)

# ------------------------------------------------------------------------------
# STUDY-LEVEL OBSERVATION COUNTS
# ------------------------------------------------------------------------------

study_counts = (
    work.groupby("study_id")
    .agg(
        observation_count=("study_id", "size"),
        polymer_count=("polymer_count", "max"),
        drug_count=("drug_count", "max"),
        first_time_h=("time_h", "min"),
        last_time_h=("time_h", "max"),
        min_release_percent=("release_percent", "min"),
        max_release_percent=("release_percent", "max"),
    )
    .reset_index()
)

study_counts["multi_point"] = (
    study_counts["observation_count"] >= 2
)

study_counts["kinetic_eligible"] = (
    study_counts["observation_count"] >= 3
)

# ------------------------------------------------------------------------------
# STUDY-LEVEL POLYMER / DRUG MAPPING
# ------------------------------------------------------------------------------

study_metadata = (
    work.groupby("study_id")
    .agg(
        polymer=("polymer_normalized",
                 lambda x: ";".join(
                     sorted(set(
                         v for v in x.dropna()
                         if str(v).strip()
                     ))
                 ) if x.notna().any() else np.nan),

        drug=("drug_normalized",
              lambda x: ";".join(
                  sorted(set(
                      v for v in x.dropna()
                      if str(v).strip()
                  ))
              ) if x.notna().any() else np.nan),
    )
    .reset_index()
)

study_level = study_counts.merge(
    study_metadata,
    on="study_id",
    how="left"
)

# ------------------------------------------------------------------------------
# COMPARABILITY CLASSIFICATION
# ------------------------------------------------------------------------------

def classify_comparability(row):

    reasons = []

    if row["observation_count"] == 1:
        reasons.append("SINGLE_TIMEPOINT")

    if row["observation_count"] >= 3:
        curve = True
    else:
        curve = False

    if pd.isna(row["polymer"]):
        reasons.append("MISSING_POLYMER")

    if pd.isna(row["drug"]):
        reasons.append("MISSING_DRUG")

    if curve and not reasons:
        return "HIGHER_WITHIN_STUDY_COMPARABILITY"

    if not curve and not reasons:
        return "DESCRIPTIVE_SINGLE_TIMEPOINT"

    if "MISSING_DRUG" in reasons or "MISSING_POLYMER" in reasons:
        return "LIMITED_METADATA_COMPARABILITY"

    return "LIMITED_COMPARABILITY"

study_level["comparability_class"] = study_level.apply(
    classify_comparability,
    axis=1
)

# ------------------------------------------------------------------------------
# POLYMER FAMILY FREQUENCY
# ------------------------------------------------------------------------------

polymer_rows = []

for _, row in study_level.iterrows():

    if pd.isna(row["polymer"]):
        continue

    for polymer in str(row["polymer"]).split(";"):

        polymer = polymer.strip()

        if polymer:

            polymer_rows.append({
                "study_id": row["study_id"],
                "polymer": polymer,
                "observation_count": row["observation_count"],
                "kinetic_eligible": row["kinetic_eligible"]
            })

polymer_study = pd.DataFrame(polymer_rows)

if len(polymer_study):

    polymer_summary = (
        polymer_study.groupby("polymer")
        .agg(
            study_count=("study_id", "nunique"),
            observation_count=("observation_count", "sum"),
            kinetic_study_count=("kinetic_eligible", "sum")
        )
        .reset_index()
        .sort_values(
            ["study_count", "observation_count"],
            ascending=False
        )
    )

else:

    polymer_summary = pd.DataFrame(
        columns=[
            "polymer",
            "study_count",
            "observation_count",
            "kinetic_study_count"
        ]
    )

# ------------------------------------------------------------------------------
# DRUG FREQUENCY
# ------------------------------------------------------------------------------

drug_rows = []

for _, row in study_level.iterrows():

    if pd.isna(row["drug"]):
        continue

    for drug in str(row["drug"]).split(";"):

        drug = drug.strip()

        if drug:

            drug_rows.append({
                "study_id": row["study_id"],
                "drug": drug,
                "observation_count": row["observation_count"],
                "kinetic_eligible": row["kinetic_eligible"]
            })

drug_study = pd.DataFrame(drug_rows)

if len(drug_study):

    drug_summary = (
        drug_study.groupby("drug")
        .agg(
            study_count=("study_id", "nunique"),
            observation_count=("observation_count", "sum"),
            kinetic_study_count=("kinetic_eligible", "sum")
        )
        .reset_index()
        .sort_values(
            ["study_count", "observation_count"],
            ascending=False
        )
    )

else:

    drug_summary = pd.DataFrame(
        columns=[
            "drug",
            "study_count",
            "observation_count",
            "kinetic_study_count"
        ]
    )

# ------------------------------------------------------------------------------
# POLYMER–DRUG PAIRS
# ------------------------------------------------------------------------------

pair_rows = []

for _, row in study_level.iterrows():

    polymers = (
        []
        if pd.isna(row["polymer"])
        else [x.strip() for x in str(row["polymer"]).split(";") if x.strip()]
    )

    drugs = (
        []
        if pd.isna(row["drug"])
        else [x.strip() for x in str(row["drug"]).split(";") if x.strip()]
    )

    if polymers and drugs:

        for polymer in polymers:

            for drug in drugs:

                pair_rows.append({
                    "study_id": row["study_id"],
                    "polymer": polymer,
                    "drug": drug,
                    "observation_count": row["observation_count"],
                    "kinetic_eligible": row["kinetic_eligible"]
                })

pair_df = pd.DataFrame(pair_rows)

if len(pair_df):

    pair_summary = (
        pair_df.groupby(["polymer", "drug"])
        .agg(
            study_count=("study_id", "nunique"),
            observation_count=("observation_count", "sum"),
            kinetic_study_count=("kinetic_eligible", "sum")
        )
        .reset_index()
        .sort_values(
            ["study_count", "observation_count"],
            ascending=False
        )
    )

else:

    pair_summary = pd.DataFrame(
        columns=[
            "polymer",
            "drug",
            "study_count",
            "observation_count",
            "kinetic_study_count"
        ]
    )

# ------------------------------------------------------------------------------
# COMPARABLE PAIR IDENTIFICATION
# ------------------------------------------------------------------------------

if len(pair_summary):

    pair_summary["direct_cross_study_comparison_possible"] = (
        pair_summary["study_count"] >= 2
    )

    pair_summary["comparison_strength"] = np.select(
        [
            pair_summary["study_count"] >= 3,
            pair_summary["study_count"] == 2,
            pair_summary["study_count"] == 1,
        ],
        [
            "STRONGER_REPLICATION",
            "LIMITED_REPLICATION",
            "SINGLE_STUDY_ONLY"
        ],
        default="UNKNOWN"
    )

else:

    pair_summary["direct_cross_study_comparison_possible"] = []
    pair_summary["comparison_strength"] = []

# ------------------------------------------------------------------------------
# TIMEPOINT COMPARABILITY
# ------------------------------------------------------------------------------

# Clinically common windows, used only for descriptive mapping.
TIME_WINDOWS = {
    "early_0_to_1h": (0, 1),
    "early_1_to_6h": (1, 6),
    "intermediate_6_to_24h": (6, 24),
    "late_24_to_72h": (24, 72),
    "very_late_gt_72h": (72, np.inf)
}

time_rows = []

for _, row in work.iterrows():

    t = row["time_h"]

    if pd.isna(t):
        window = "UNKNOWN"
    else:

        window = "UNKNOWN"

        for name, (low, high) in TIME_WINDOWS.items():

            if low <= t < high:
                window = name
                break

        # Include exactly 72 h in late window
        if t == 72:
            window = "late_24_to_72h"

    time_rows.append({
        "study_id": row["study_id"],
        "time_h": t,
        "release_percent": row["release_percent"],
        "time_window": window
    })

time_df = pd.DataFrame(time_rows)

time_summary = (
    time_df.groupby("time_window", dropna=False)
    .agg(
        observation_count=("study_id", "size"),
        study_count=("study_id", "nunique"),
        mean_release_percent=("release_percent", "mean"),
        median_release_percent=("release_percent", "median"),
        min_release_percent=("release_percent", "min"),
        max_release_percent=("release_percent", "max")
    )
    .reset_index()
)

# ------------------------------------------------------------------------------
# EVIDENCE MATRIX
# ------------------------------------------------------------------------------

evidence_matrix = study_level.copy()

evidence_matrix["evidence_level"] = np.select(
    [
        evidence_matrix["kinetic_eligible"],
        evidence_matrix["multi_point"],
        evidence_matrix["observation_count"] == 1
    ],
    [
        "MULTI_TIMEPOINT_KINETIC_EVIDENCE",
        "MULTI_TIMEPOINT_NON_KINETIC_EVIDENCE",
        "SINGLE_TIMEPOINT_EVIDENCE"
    ],
    default="OTHER"
)

evidence_matrix["recommended_use"] = np.select(
    [
        evidence_matrix["kinetic_eligible"],
        evidence_matrix["observation_count"] == 1
    ],
    [
        "KINETIC_AND_DESCRIPTIVE",
        "DESCRIPTIVE_ONLY"
    ],
    default="DESCRIPTIVE_WITH_CAUTION"
)

# ------------------------------------------------------------------------------
# CROSS-STUDY COMPARISON FLAGS
# ------------------------------------------------------------------------------

cross_study = []

for polymer, g in polymer_study.groupby("polymer"):

    n = g["study_id"].nunique()

    cross_study.append({
        "entity_type": "POLYMER",
        "entity": polymer,
        "study_count": n,
        "observation_count": g["observation_count"].sum(),
        "cross_study_comparison": (
            "POTENTIALLY_POSSIBLE"
            if n >= 2
            else "NOT_POSSIBLE_SINGLE_STUDY"
        )
    })

for drug, g in drug_study.groupby("drug"):

    n = g["study_id"].nunique()

    cross_study.append({
        "entity_type": "DRUG",
        "entity": drug,
        "study_count": n,
        "observation_count": g["observation_count"].sum(),
        "cross_study_comparison": (
            "POTENTIALLY_POSSIBLE"
            if n >= 2
            else "NOT_POSSIBLE_SINGLE_STUDY"
        )
    })

cross_study_df = pd.DataFrame(cross_study)

# ------------------------------------------------------------------------------
# ANALYTICAL RECOMMENDATIONS
# ------------------------------------------------------------------------------

recommendations = pd.DataFrame([
    {
        "analysis": "Study-level descriptive synthesis",
        "status": "RECOMMENDED",
        "reason":
            "Each study represents an independent evidence unit."
    },
    {
        "analysis": "Within-study release-curve analysis",
        "status": "RECOMMENDED",
        "reason":
            "Repeated timepoints are naturally handled within individual studies."
    },
    {
        "analysis": "Kinetic modeling",
        "status": "RESTRICTED",
        "reason":
            "Only studies with at least three usable timepoints should be modeled."
    },
    {
        "analysis": "Pooled observation-level regression",
        "status": "NOT_RECOMMENDED",
        "reason":
            "Repeated observations create pseudoreplication and violate simple independence."
    },
    {
        "analysis": "Polymer-group comparison",
        "status": "CAUTION",
        "reason":
            "Polymer identity is confounded by drug identity and formulation differences."
    },
    {
        "analysis": "Drug-group comparison",
        "status": "CAUTION",
        "reason":
            "Different drugs have different physicochemical and release characteristics."
    },
    {
        "analysis": "Formal meta-analysis",
        "status": "NOT_RECOMMENDED",
        "reason":
            "The extracted outcomes are heterogeneous release percentages at different times rather than a common effect measure."
    },
    {
        "analysis": "Machine learning",
        "status": "NOT_RECOMMENDED",
        "reason":
            "The number of independent studies is too small and formulations are highly heterogeneous."
    },
    {
        "analysis": "Narrative quantitative synthesis",
        "status": "RECOMMENDED",
        "reason":
            "The current evidence supports structured quantitative comparison without unjustified pooling."
    }
])

# ------------------------------------------------------------------------------
# SAVE
# ------------------------------------------------------------------------------

work[
    [
        "study_id",
        "polymer_raw",
        "polymer_normalized",
        "drug_raw",
        "drug_normalized",
        "time_h",
        "release_percent"
    ]
].to_csv(
    OUTDIR / "16_normalized_observations.csv",
    index=False
)

study_level.to_csv(
    OUTDIR / "16_study_level_evidence_matrix.csv",
    index=False
)

polymer_summary.to_csv(
    OUTDIR / "16_polymer_summary.csv",
    index=False
)

drug_summary.to_csv(
    OUTDIR / "16_drug_summary.csv",
    index=False
)

pair_summary.to_csv(
    OUTDIR / "16_polymer_drug_pairs.csv",
    index=False
)

time_summary.to_csv(
    OUTDIR / "16_time_window_summary.csv",
    index=False
)

cross_study_df.to_csv(
    OUTDIR / "16_cross_study_comparison_flags.csv",
    index=False
)

recommendations.to_csv(
    OUTDIR / "16_analytical_recommendations.csv",
    index=False
)

# ------------------------------------------------------------------------------
# CONSOLE OUTPUT
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("STUDY-LEVEL EVIDENCE MATRIX")
print("=" * 78)

print(
    evidence_matrix[
        [
            "study_id",
            "polymer",
            "drug",
            "observation_count",
            "kinetic_eligible",
            "evidence_level",
            "comparability_class",
            "recommended_use"
        ]
    ].to_string(index=False)
)

print("\n" + "=" * 78)
print("POLYMER SUMMARY")
print("=" * 78)

print(polymer_summary.to_string(index=False))

print("\n" + "=" * 78)
print("DRUG SUMMARY")
print("=" * 78)

print(drug_summary.to_string(index=False))

print("\n" + "=" * 78)
print("POLYMER–DRUG PAIR SUMMARY")
print("=" * 78)

print(pair_summary.to_string(index=False))

print("\n" + "=" * 78)
print("TIME-WINDOW SUMMARY")
print("=" * 78)

print(time_summary.to_string(index=False))

print("\n" + "=" * 78)
print("ANALYTICAL RECOMMENDATIONS")
print("=" * 78)

print(recommendations.to_string(index=False))

print("\n" + "=" * 78)
print("CODE 16 COMPLETE")
print("=" * 78)

print("\nFiles created:")

for f in sorted(OUTDIR.glob("16_*.csv")):
    print("-", f)

print("=" * 78)
```

```python
# ==============================================================================
# CODE 17 — QUANTITATIVE EVIDENCE SYNTHESIS & PUBLICATION-READY OUTPUTS
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

BASE = Path("AI_Polymer_Drug_Release")

INPUT = BASE / "data/final/12_final_analysis_dataset.csv"
CURVE_SUMMARY = BASE / "data/analysis/13_release_curve/13_study_level_release_curve_summary.csv"
KINETIC = BASE / "data/analysis/14_release_kinetics/14_study_level_kinetic_summary.csv"
EVIDENCE = BASE / "data/analysis/16_comparability/16_study_level_evidence_matrix.csv"
TIME_WINDOW = BASE / "data/analysis/16_comparability/16_time_window_summary.csv"

OUT = BASE / "data/analysis/17_quantitative_synthesis"
OUT.mkdir(parents=True, exist_ok=True)

print("=" * 78)
print("CODE 17 — QUANTITATIVE EVIDENCE SYNTHESIS")
print("=" * 78)

# ------------------------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------------------------

df = pd.read_csv(INPUT)
curve = pd.read_csv(CURVE_SUMMARY)
kinetic = pd.read_csv(KINETIC)
evidence = pd.read_csv(EVIDENCE)
time_window = pd.read_csv(TIME_WINDOW)

print(f"\nFinal quantitative observations loaded: {len(df)}")
print(f"Unique studies: {df['study_id'].nunique()}")

# ------------------------------------------------------------------------------
# 2. STANDARDIZE IMPORTANT COLUMNS
# ------------------------------------------------------------------------------

required = [
    "study_id",
    "polymer",
    "drug",
    "release_time_h",
    "release_percent"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise KeyError(
        f"Required columns missing: {missing}\n"
        f"Available columns: {list(df.columns)}"
    )

df["release_time_h"] = pd.to_numeric(df["release_time_h"], errors="coerce")
df["release_percent"] = pd.to_numeric(df["release_percent"], errors="coerce")

df = df.dropna(
    subset=["study_id", "release_time_h", "release_percent"]
).copy()

# ------------------------------------------------------------------------------
# 3. STUDY-LEVEL QUANTITATIVE SYNTHESIS
# ------------------------------------------------------------------------------

study_rows = []

for study_id, g in df.groupby("study_id"):

    g = g.sort_values("release_time_h")

    first = g.iloc[0]
    last = g.iloc[-1]

    row = {
        "study_id": study_id,
        "polymer": first["polymer"],
        "drug": first["drug"],
        "n_observations": len(g),
        "n_timepoints": g["release_time_h"].nunique(),
        "first_time_h": first["release_time_h"],
        "last_time_h": last["release_time_h"],
        "first_release_percent": first["release_percent"],
        "last_release_percent": last["release_percent"],
        "release_change_percent":
            last["release_percent"] - first["release_percent"],
        "maximum_release_percent": g["release_percent"].max(),
        "minimum_release_percent": g["release_percent"].min(),
        "median_release_percent": g["release_percent"].median(),
        "multi_timepoint":
            g["release_time_h"].nunique() >= 3
    }

    study_rows.append(row)

study_summary = pd.DataFrame(study_rows)

print("\n" + "=" * 78)
print("STUDY-LEVEL QUANTITATIVE SYNTHESIS")
print("=" * 78)

print(study_summary.to_string(index=False))

study_summary.to_csv(
    OUT / "17_study_level_quantitative_synthesis.csv",
    index=False
)

# ------------------------------------------------------------------------------
# 4. MULTI-TIMEPOINT STUDY SUMMARY
# ------------------------------------------------------------------------------

multi = study_summary[
    study_summary["multi_timepoint"] == True
].copy()

print("\nMulti-timepoint studies:", len(multi))

if len(multi) > 0:
    multi.to_csv(
        OUT / "17_multi_timepoint_study_summary.csv",
        index=False
    )

# ------------------------------------------------------------------------------
# 5. SINGLE-TIMEPOINT STUDIES
# ------------------------------------------------------------------------------

single = study_summary[
    study_summary["multi_timepoint"] == False
].copy()

single.to_csv(
    OUT / "17_single_timepoint_study_summary.csv",
    index=False
)

print("Single-timepoint studies:", len(single))

# ------------------------------------------------------------------------------
# 6. TIME-WINDOW QUANTITATIVE SYNTHESIS
# ------------------------------------------------------------------------------

def classify_time(t):

    if t <= 1:
        return "Early (0–1 h)"
    elif t <= 6:
        return "Early-intermediate (1–6 h)"
    elif t <= 24:
        return "Intermediate (6–24 h)"
    elif t <= 72:
        return "Late (24–72 h)"
    else:
        return "Very late (>72 h)"

df["time_window"] = df["release_time_h"].apply(classify_time)

window_summary = (
    df.groupby("time_window")
      .agg(
          observation_count=("release_percent", "size"),
          study_count=("study_id", "nunique"),
          mean_release_percent=("release_percent", "mean"),
          median_release_percent=("release_percent", "median"),
          sd_release_percent=("release_percent", "std"),
          min_release_percent=("release_percent", "min"),
          max_release_percent=("release_percent", "max")
      )
      .reset_index()
)

print("\n" + "=" * 78)
print("TIME-WINDOW SYNTHESIS")
print("=" * 78)

print(window_summary.to_string(index=False))

window_summary.to_csv(
    OUT / "17_time_window_quantitative_synthesis.csv",
    index=False
)

# ------------------------------------------------------------------------------
# 7. POLYMER DESCRIPTIVE SUMMARY
# ------------------------------------------------------------------------------

# Polymer strings represent multi-component formulations.
# Therefore this is descriptive only and NOT a causal polymer comparison.

polymer_summary = (
    df.groupby("polymer")
      .agg(
          study_count=("study_id", "nunique"),
          observation_count=("release_percent", "size"),
          mean_release_percent=("release_percent", "mean"),
          median_release_percent=("release_percent", "median"),
          sd_release_percent=("release_percent", "std")
      )
      .reset_index()
      .sort_values("study_count", ascending=False)
)

polymer_summary.to_csv(
    OUT / "17_polymer_descriptive_summary.csv",
    index=False
)

# ------------------------------------------------------------------------------
# 8. DRUG DESCRIPTIVE SUMMARY
# ------------------------------------------------------------------------------

drug_summary = (
    df.groupby("drug")
      .agg(
          study_count=("study_id", "nunique"),
          observation_count=("release_percent", "size"),
          mean_release_percent=("release_percent", "mean"),
          median_release_percent=("release_percent", "median"),
          sd_release_percent=("release_percent", "std")
      )
      .reset_index()
      .sort_values("study_count", ascending=False)
)

drug_summary.to_csv(
    OUT / "17_drug_descriptive_summary.csv",
    index=False
)

# ------------------------------------------------------------------------------
# 9. EVIDENCE-LEVEL SYNTHESIS
# ------------------------------------------------------------------------------

evidence_cols = [
    c for c in [
        "study_id",
        "evidence_level",
        "comparability_class",
        "recommended_use",
        "observation_count",
        "kinetic_eligible"
    ]
    if c in evidence.columns
]

evidence_summary = evidence[evidence_cols].copy()

evidence_summary.to_csv(
    OUT / "17_evidence_level_synthesis.csv",
    index=False
)

# ------------------------------------------------------------------------------
# 10. KINETIC SYNTHESIS
# ------------------------------------------------------------------------------

if len(kinetic) > 0:

    kinetic_available = kinetic.copy()

    kinetic_available.to_csv(
        OUT / "17_kinetic_model_synthesis.csv",
        index=False
    )

    print("\n" + "=" * 78)
    print("KINETIC MODEL SYNTHESIS")
    print("=" * 78)

    print(kinetic_available.to_string(index=False))

# ------------------------------------------------------------------------------
# 11. RELEASE-CURVE PLOT — ALL MULTI-TIMEPOINT STUDIES
# ------------------------------------------------------------------------------

plt.figure(figsize=(10, 6))

for study_id, g in df.groupby("study_id"):

    if g["release_time_h"].nunique() < 3:
        continue

    g = g.sort_values("release_time_h")

    plt.plot(
        g["release_time_h"],
        g["release_percent"],
        marker="o",
        linewidth=1.8,
        label=study_id
    )

plt.xscale("log")

plt.xlabel("Release time (h)")
plt.ylabel("Cumulative drug release (%)")
plt.title("Quantitative Release Curves of Multi-Timepoint Studies")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    OUT / "17_multi_study_release_curves.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()

# ------------------------------------------------------------------------------
# 12. STUDY-LEVEL FINAL RELEASE COMPARISON
# ------------------------------------------------------------------------------

plot_df = study_summary.sort_values(
    "last_release_percent",
    ascending=False
)

plt.figure(figsize=(10, 6))

plt.bar(
    plot_df["study_id"],
    plot_df["last_release_percent"]
)

plt.ylabel("Last reported cumulative release (%)")
plt.xlabel("Study")
plt.title("Last Reported Drug Release by Study")

plt.xticks(
    rotation=60,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    OUT / "17_last_release_by_study.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()

# ------------------------------------------------------------------------------
# 13. TIME VS RELEASE SCATTER — DESCRIPTIVE ONLY
# ------------------------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.scatter(
    df["release_time_h"],
    df["release_percent"],
    alpha=0.75
)

plt.xscale("log")

plt.xlabel("Release time (h)")
plt.ylabel("Cumulative release (%)")
plt.title(
    "Observed Release Percentage Across Extracted Timepoints"
)

plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    OUT / "17_time_release_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()

# ------------------------------------------------------------------------------
# 14. DATASET DOMINANCE CHECK
# ------------------------------------------------------------------------------

study_contribution = (
    df.groupby("study_id")
      .size()
      .reset_index(name="observation_count")
      .sort_values("observation_count", ascending=False)
)

study_contribution["proportion"] = (
    study_contribution["observation_count"] / len(df)
)

study_contribution["cumulative_proportion"] = (
    study_contribution["proportion"].cumsum()
)

study_contribution.to_csv(
    OUT / "17_study_observation_contribution.csv",
    index=False
)

# ------------------------------------------------------------------------------
# 15. SYNTHESIS LIMITATIONS
# ------------------------------------------------------------------------------

limitations = pd.DataFrame({
    "issue": [
        "Independent study count",
        "Repeated observations",
        "Multi-timepoint studies",
        "Cross-study formulation heterogeneity",
        "Drug heterogeneity",
        "Common effect measure",
        "Formal meta-analysis",
        "Machine learning"
    ],
    "assessment": [
        f"{df['study_id'].nunique()} independent studies",
        "Present",
        f"{(study_summary['multi_timepoint']).sum()} studies",
        "High",
        "High",
        "Not available",
        "Not recommended",
        "Not recommended"
    ],
    "interpretation": [
        "Evidence base is relatively small",
        "Observation-level pooling risks pseudoreplication",
        "Only a small subset supports kinetic analysis",
        "Polymer systems differ substantially between studies",
        "Drug identity and physicochemical properties vary",
        "Release percentages occur at different timepoints",
        "Outcomes are not sufficiently comparable",
        "Dataset is too small and heterogeneous for reliable predictive modeling"
    ]
})

limitations.to_csv(
    OUT / "17_synthesis_limitations.csv",
    index=False
)

# ------------------------------------------------------------------------------
# 16. OVERALL SYNTHESIS REPORT
# ------------------------------------------------------------------------------

report = {
    "total_quantitative_observations": len(df),
    "independent_studies": df["study_id"].nunique(),
    "multi_timepoint_studies": int(
        study_summary["multi_timepoint"].sum()
    ),
    "single_timepoint_studies": int(
        (~study_summary["multi_timepoint"]).sum()
    ),
    "mean_release_percent": df["release_percent"].mean(),
    "median_release_percent": df["release_percent"].median(),
    "minimum_release_percent": df["release_percent"].min(),
    "maximum_release_percent": df["release_percent"].max(),
    "mean_release_time_h": df["release_time_h"].mean(),
    "median_release_time_h": df["release_time_h"].median(),
    "formal_meta_analysis": "NOT RECOMMENDED",
    "observation_level_pooling": "NOT RECOMMENDED",
    "machine_learning": "NOT RECOMMENDED",
    "narrative_quantitative_synthesis": "RECOMMENDED",
    "within_study_kinetic_analysis": "RECOMMENDED_WHERE_ELIGIBLE"
}

report_df = pd.DataFrame(
    list(report.items()),
    columns=["metric", "value"]
)

report_df.to_csv(
    OUT / "17_overall_quantitative_synthesis_report.csv",
    index=False
)

# ------------------------------------------------------------------------------
# 17. COMPLETE
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("CODE 17 COMPLETE")
print("=" * 78)

print(f"""
Quantitative observations: {len(df)}
Independent studies: {df['study_id'].nunique()}
Multi-timepoint studies: {study_summary['multi_timepoint'].sum()}
Single-timepoint studies: {(~study_summary['multi_timepoint']).sum()}

Mean release: {df['release_percent'].mean():.2f}%
Median release: {df['release_percent'].median():.2f}%

Formal meta-analysis: NOT RECOMMENDED
Observation-level pooling: NOT RECOMMENDED
Machine learning: NOT RECOMMENDED
Narrative quantitative synthesis: RECOMMENDED
Within-study kinetic analysis: RECOMMENDED where eligible

Output directory:
{OUT}
""")

print("Files created:")
for f in sorted(OUT.iterdir()):
    print("-", f.name)

print("=" * 78)
```

```python
# ==============================================================================
# CODE 18 — ROBUSTNESS & SENSITIVITY ANALYSIS
# ==============================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------------------------
# PATHS
# ------------------------------------------------------------------------------

INPUT_FILE = (
    "AI_Polymer_Drug_Release/data/final/"
    "12_final_analysis_dataset.csv"
)

OUT_DIR = "AI_Polymer_Drug_Release/data/analysis/18_sensitivity"
os.makedirs(OUT_DIR, exist_ok=True)

# ------------------------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------------------------

df = pd.read_csv(INPUT_FILE)

print("=" * 78)
print("CODE 18 — ROBUSTNESS & SENSITIVITY ANALYSIS")
print("=" * 78)

print("Input rows:", len(df))
print("Input columns:", len(df.columns))
print("Columns:")
print(list(df.columns))

# ------------------------------------------------------------------------------
# COLUMN DETECTION
# ------------------------------------------------------------------------------

def find_col(columns, candidates):
    lower_map = {str(c).lower(): c for c in columns}

    for candidate in candidates:
        if candidate.lower() in lower_map:
            return lower_map[candidate.lower()]

    for c in columns:
        cl = str(c).lower()
        for candidate in candidates:
            if candidate.lower() in cl:
                return c

    return None


study_col = find_col(df.columns, ["study_id"])
release_col = find_col(
    df.columns,
    ["release_percent", "release_percent_std"]
)
time_col = find_col(
    df.columns,
    ["release_time_h", "release_time_std"]
)

include_col = find_col(
    df.columns,
    ["final_analysis_inclusion"]
)

print("\nDetected columns:")
print("Study:", study_col)
print("Release:", release_col)
print("Time:", time_col)
print("Final inclusion:", include_col)

if study_col is None or release_col is None:
    raise KeyError("Study ID or release percentage column could not be detected.")

# ------------------------------------------------------------------------------
# FINAL ANALYTICAL DATASET
# ------------------------------------------------------------------------------

work = df.copy()

# Correct the CODE 17 discrepancy:
# only observations explicitly included in final analysis are retained.

if include_col is not None:
    include_values = (
        work[include_col]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    work = work[
        include_values.isin(["true", "1", "yes"])
    ].copy()

work[release_col] = pd.to_numeric(
    work[release_col], errors="coerce"
)

if time_col is not None:
    work[time_col] = pd.to_numeric(
        work[time_col], errors="coerce"
    )

work = work[
    work[study_col].notna() &
    work[release_col].notna()
].copy()

work = work[
    (work[release_col] >= 0) &
    (work[release_col] <= 100)
].copy()

print("\n" + "=" * 78)
print("FINAL SENSITIVITY DATASET")
print("=" * 78)

print("Final observations:", len(work))
print("Independent studies:", work[study_col].nunique())

if len(work) != 25:
    print(
        "WARNING: Expected approximately 25 final analytical observations."
    )

# ------------------------------------------------------------------------------
# BASIC STUDY-LEVEL SUMMARY
# ------------------------------------------------------------------------------

study_summary = (
    work.groupby(study_col)
    .agg(
        observation_count=(release_col, "size"),
        mean_release=(release_col, "mean"),
        median_release=(release_col, "median"),
        min_release=(release_col, "min"),
        max_release=(release_col, "max"),
    )
    .reset_index()
)

study_summary["proportion_observations"] = (
    study_summary["observation_count"] / len(work)
)

study_summary = study_summary.sort_values(
    "observation_count",
    ascending=False
)

print("\nStudy contribution:")
print(study_summary.to_string(index=False))

# ------------------------------------------------------------------------------
# OVERALL BASELINE
# ------------------------------------------------------------------------------

baseline_mean = work[release_col].mean()
baseline_median = work[release_col].median()
baseline_sd = work[release_col].std()
baseline_min = work[release_col].min()
baseline_max = work[release_col].max()

baseline = {
    "analysis": "ALL_FINAL_OBSERVATIONS",
    "excluded_study": "",
    "n_observations": len(work),
    "n_studies": work[study_col].nunique(),
    "mean_release": baseline_mean,
    "median_release": baseline_median,
    "sd_release": baseline_sd,
    "min_release": baseline_min,
    "max_release": baseline_max,
}

# ------------------------------------------------------------------------------
# 1. LEAVE-ONE-STUDY-OUT ANALYSIS
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("LEAVE-ONE-STUDY-OUT SENSITIVITY")
print("=" * 78)

loo_rows = [baseline]

for study in sorted(work[study_col].unique()):

    subset = work[work[study_col] != study].copy()

    loo_rows.append({
        "analysis": "LEAVE_ONE_STUDY_OUT",
        "excluded_study": study,
        "n_observations": len(subset),
        "n_studies": subset[study_col].nunique(),
        "mean_release": subset[release_col].mean(),
        "median_release": subset[release_col].median(),
        "sd_release": subset[release_col].std(),
        "min_release": subset[release_col].min(),
        "max_release": subset[release_col].max(),
    })

loo = pd.DataFrame(loo_rows)

loo["mean_change_from_baseline"] = (
    loo["mean_release"] - baseline_mean
)

loo["median_change_from_baseline"] = (
    loo["median_release"] - baseline_median
)

loo["absolute_mean_change"] = (
    loo["mean_change_from_baseline"].abs()
)

loo["absolute_median_change"] = (
    loo["median_change_from_baseline"].abs()
)

loo.to_csv(
    os.path.join(OUT_DIR, "18_leave_one_study_out.csv"),
    index=False
)

# ------------------------------------------------------------------------------
# 2. IDENTIFY INFLUENTIAL STUDIES
# ------------------------------------------------------------------------------

influential = loo[
    loo["excluded_study"].astype(str).str.len() > 0
].copy()

mean_threshold = max(
    5.0,
    0.15 * abs(baseline_mean)
)

median_threshold = max(
    5.0,
    0.15 * abs(baseline_median)
)

influential["mean_influential"] = (
    influential["absolute_mean_change"] >= mean_threshold
)

influential["median_influential"] = (
    influential["absolute_median_change"] >= median_threshold
)

influential["influential_any"] = (
    influential["mean_influential"] |
    influential["median_influential"]
)

influential = influential.sort_values(
    "absolute_mean_change",
    ascending=False
)

influential.to_csv(
    os.path.join(OUT_DIR, "18_influential_studies.csv"),
    index=False
)

# ------------------------------------------------------------------------------
# 3. MULTI-TIMEPOINT VS SINGLE-TIMEPOINT SENSITIVITY
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("MULTI-TIMEPOINT VS SINGLE-TIMEPOINT")
print("=" * 78)

study_counts = (
    work.groupby(study_col)
    .size()
    .rename("n_timepoints")
    .reset_index()
)

work2 = work.merge(
    study_counts,
    on=study_col,
    how="left"
)

work2["study_type"] = np.where(
    work2["n_timepoints"] >= 3,
    "MULTI_TIMEPOINT_STUDY",
    "SINGLE_TIMEPOINT_STUDY"
)

group_summary = (
    work2.groupby("study_type")
    .agg(
        observations=(release_col, "size"),
        studies=(study_col, "nunique"),
        mean_release=(release_col, "mean"),
        median_release=(release_col, "median"),
        sd_release=(release_col, "std"),
        min_release=(release_col, "min"),
        max_release=(release_col, "max"),
    )
    .reset_index()
)

group_summary.to_csv(
    os.path.join(
        OUT_DIR,
        "18_multitimepoint_vs_singlepoint.csv"
    ),
    index=False
)

print(group_summary.to_string(index=False))

# ------------------------------------------------------------------------------
# 4. EXTREME-VALUE SENSITIVITY
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("EXTREME-VALUE SENSITIVITY")
print("=" * 78)

extreme_rows = []

for threshold_name, condition in [
    ("EXCLUDE_100_PERCENT", work[release_col] < 100),
    ("EXCLUDE_GE_95_PERCENT", work[release_col] < 95),
    ("EXCLUDE_GE_90_PERCENT", work[release_col] < 90),
    ("EXCLUDE_LE_5_PERCENT", work[release_col] > 5),
]:

    subset = work[condition].copy()

    extreme_rows.append({
        "analysis": threshold_name,
        "n_observations": len(subset),
        "n_studies": subset[study_col].nunique(),
        "excluded_observations": len(work) - len(subset),
        "mean_release": subset[release_col].mean()
        if len(subset) else np.nan,
        "median_release": subset[release_col].median()
        if len(subset) else np.nan,
        "sd_release": subset[release_col].std()
        if len(subset) > 1 else np.nan,
        "mean_change": (
            subset[release_col].mean() - baseline_mean
            if len(subset) else np.nan
        ),
        "median_change": (
            subset[release_col].median() - baseline_median
            if len(subset) else np.nan
        ),
    })

extreme = pd.DataFrame(extreme_rows)

extreme.to_csv(
    os.path.join(
        OUT_DIR,
        "18_extreme_value_sensitivity.csv"
    ),
    index=False
)

print(extreme.to_string(index=False))

# ------------------------------------------------------------------------------
# 5. TIME-WINDOW SENSITIVITY
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("TIME-WINDOW SENSITIVITY")
print("=" * 78)

if time_col is not None:

    work3 = work.copy()

    work3["time_window"] = pd.cut(
        work3[time_col],
        bins=[
            -np.inf,
            1,
            6,
            24,
            72,
            np.inf
        ],
        labels=[
            "Early_0_to_1h",
            "Early_intermediate_1_to_6h",
            "Intermediate_6_to_24h",
            "Late_24_to_72h",
            "Very_late_gt_72h"
        ],
        right=True,
        include_lowest=True
    )

    time_summary = (
        work3.groupby(
            "time_window",
            observed=False
        )
        .agg(
            observations=(release_col, "size"),
            studies=(study_col, "nunique"),
            mean_release=(release_col, "mean"),
            median_release=(release_col, "median"),
            sd_release=(release_col, "std"),
            min_release=(release_col, "min"),
            max_release=(release_col, "max"),
        )
        .reset_index()
    )

    time_summary.to_csv(
        os.path.join(
            OUT_DIR,
            "18_time_window_sensitivity.csv"
        ),
        index=False
    )

    print(time_summary.to_string(index=False))

else:

    time_summary = pd.DataFrame()

# ------------------------------------------------------------------------------
# 6. TOP-STUDY DOMINANCE SENSITIVITY
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("TOP-STUDY DOMINANCE")
print("=" * 78)

sorted_studies = (
    study_summary
    .sort_values(
        "observation_count",
        ascending=False
    )
    [study_col]
    .tolist()
)

dominance_rows = []

for n_top in [1, 2, 3]:

    top_studies = sorted_studies[:n_top]

    subset = work[
        work[study_col].isin(top_studies)
    ]

    dominance_rows.append({
        "top_n_studies": n_top,
        "studies": ";".join(map(str, top_studies)),
        "observations": len(subset),
        "proportion_of_all_observations": (
            len(subset) / len(work)
        ),
        "mean_release": subset[release_col].mean(),
        "median_release": subset[release_col].median(),
    })

dominance = pd.DataFrame(dominance_rows)

print(dominance.to_string(index=False))

# ------------------------------------------------------------------------------
# 7. ROBUSTNESS CONCLUSION
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("ROBUSTNESS ASSESSMENT")
print("=" * 78)

max_loo_mean_change = influential[
    "absolute_mean_change"
].max()

max_loo_median_change = influential[
    "absolute_median_change"
].max()

if pd.isna(max_loo_mean_change):
    max_loo_mean_change = 0

if pd.isna(max_loo_median_change):
    max_loo_median_change = 0

if max_loo_mean_change <= mean_threshold:
    mean_robustness = "RELATIVELY_ROBUST"
else:
    mean_robustness = "SENSITIVE_TO_STUDY_EXCLUSION"

if max_loo_median_change <= median_threshold:
    median_robustness = "RELATIVELY_ROBUST"
else:
    median_robustness = "SENSITIVE_TO_STUDY_EXCLUSION"

n_multitimepoint = (
    work2.loc[
        work2["study_type"] == "MULTI_TIMEPOINT_STUDY",
        study_col
    ].nunique()
)

n_studies = work[study_col].nunique()

if n_multitimepoint >= 3:
    kinetic_robustness = (
        "LIMITED_BY_SMALL_NUMBER_OF_MULTI_TIMEPOINT_STUDIES"
    )
else:
    kinetic_robustness = (
        "VERY_LIMITED_KINETIC_EVIDENCE"
    )

overall = (
    "MODERATELY_ROBUST_FOR_DESCRIPTIVE_SYNTHESIS"
    if (
        mean_robustness == "RELATIVELY_ROBUST"
        and median_robustness == "RELATIVELY_ROBUST"
    )
    else
    "SENSITIVITY_PRESENT_DESCRIPTIVE_INTERPRETATION_REQUIRED"
)

conclusion_rows = [
    {
        "metric": "n_observations",
        "value": len(work),
        "interpretation": "Final analytical observations"
    },
    {
        "metric": "n_independent_studies",
        "value": n_studies,
        "interpretation": "Study is the independent evidence unit"
    },
    {
        "metric": "baseline_mean_release",
        "value": baseline_mean,
        "interpretation": "Descriptive only; not a pooled effect estimate"
    },
    {
        "metric": "baseline_median_release",
        "value": baseline_median,
        "interpretation": "Preferred robust descriptive summary"
    },
    {
        "metric": "maximum_leave_one_study_out_mean_change",
        "value": max_loo_mean_change,
        "interpretation": mean_robustness
    },
    {
        "metric": "maximum_leave_one_study_out_median_change",
        "value": max_loo_median_change,
        "interpretation": median_robustness
    },
    {
        "metric": "multi_timepoint_studies",
        "value": n_multitimepoint,
        "interpretation": kinetic_robustness
    },
    {
        "metric": "overall_robustness",
        "value": overall,
        "interpretation": (
            "Suitable for structured narrative quantitative synthesis; "
            "formal pooled analysis remains inappropriate."
        )
    }
]

robustness = pd.DataFrame(conclusion_rows)

robustness.to_csv(
    os.path.join(
        OUT_DIR,
        "18_robustness_conclusion.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# 8. COMPREHENSIVE SENSITIVITY REPORT
# ------------------------------------------------------------------------------

report_rows = [
    {
        "section": "Dataset",
        "metric": "Final observations",
        "value": len(work),
        "interpretation": "Only final_analysis_inclusion=True observations included."
    },
    {
        "section": "Dataset",
        "metric": "Independent studies",
        "value": work[study_col].nunique(),
        "interpretation": "Studies are the independent evidence units."
    },
    {
        "section": "Baseline",
        "metric": "Mean release (%)",
        "value": baseline_mean,
        "interpretation": "Descriptive summary only."
    },
    {
        "section": "Baseline",
        "metric": "Median release (%)",
        "value": baseline_median,
        "interpretation": "More robust to extreme observations."
    },
    {
        "section": "Sensitivity",
        "metric": "Maximum absolute LOO mean change",
        "value": max_loo_mean_change,
        "interpretation": mean_robustness
    },
    {
        "section": "Sensitivity",
        "metric": "Maximum absolute LOO median change",
        "value": max_loo_median_change,
        "interpretation": median_robustness
    },
    {
        "section": "Kinetics",
        "metric": "Multi-timepoint studies",
        "value": n_multitimepoint,
        "interpretation": kinetic_robustness
    },
    {
        "section": "Inference",
        "metric": "Formal meta-analysis",
        "value": "NOT RECOMMENDED",
        "interpretation": "Outcomes are heterogeneous and not a common effect measure."
    },
    {
        "section": "Inference",
        "metric": "Observation-level pooling",
        "value": "NOT RECOMMENDED",
        "interpretation": "Repeated observations create pseudoreplication."
    },
    {
        "section": "Inference",
        "metric": "Machine learning",
        "value": "NOT RECOMMENDED",
        "interpretation": "Too few independent studies and substantial formulation heterogeneity."
    },
    {
        "section": "Recommended synthesis",
        "metric": "Primary approach",
        "value": "NARRATIVE QUANTITATIVE SYNTHESIS",
        "interpretation": "Use study-level evidence and within-study trajectories."
    }
]

report = pd.DataFrame(report_rows)

report.to_csv(
    os.path.join(
        OUT_DIR,
        "18_sensitivity_report.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# 9. VISUALIZATION — RELEASE DISTRIBUTION
# ------------------------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    work[release_col],
    bins=10
)

plt.xlabel("Release (%)")
plt.ylabel("Number of observations")
plt.title("Distribution of Quantitative Drug Release")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUT_DIR,
        "18_release_distribution_sensitivity.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()

# ------------------------------------------------------------------------------
# 10. VISUALIZATION — LEAVE-ONE-STUDY-OUT
# ------------------------------------------------------------------------------

plot_loo = influential.copy()

plot_loo = plot_loo.sort_values(
    "mean_release"
)

plt.figure(figsize=(10, 6))

plt.axvline(
    baseline_mean,
    linestyle="--",
    linewidth=1
)

plt.scatter(
    plot_loo["mean_release"],
    plot_loo["excluded_study"]
)

plt.xlabel("Mean release after excluding study (%)")
plt.ylabel("Excluded study")
plt.title("Leave-One-Study-Out Sensitivity")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUT_DIR,
        "18_leave_one_out_release.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()

# ------------------------------------------------------------------------------
# 11. FINAL OUTPUT
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("CODE 18 COMPLETE")
print("=" * 78)

print("Final sensitivity observations:", len(work))
print("Independent studies:", work[study_col].nunique())
print("Baseline mean release:", round(baseline_mean, 3), "%")
print("Baseline median release:", round(baseline_median, 3), "%")

print("\nMean robustness:", mean_robustness)
print("Median robustness:", median_robustness)
print("Kinetic evidence:", kinetic_robustness)
print("Overall:", overall)

print("\nFiles created:")
print(
    "- 18_sensitivity_summary.csv"
)
print(
    "- 18_leave_one_study_out.csv"
)
print(
    "- 18_multitimepoint_vs_singlepoint.csv"
)
print(
    "- 18_extreme_value_sensitivity.csv"
)
print(
    "- 18_time_window_sensitivity.csv"
)
print(
    "- 18_influential_studies.csv"
)
print(
    "- 18_robustness_conclusion.csv"
)
print(
    "- 18_sensitivity_report.csv"
)
print(
    "- 18_release_distribution_sensitivity.png"
)
print(
    "- 18_leave_one_out_release.png"
)

print("=" * 78)
```

```python
# ==============================================================================
# CODE 19 — FINAL EVIDENCE INTERPRETATION & PUBLICATION-READY SYNTHESIS
# ==============================================================================

import os
import pandas as pd
import numpy as np

print("=" * 78)
print("CODE 19 — FINAL EVIDENCE INTERPRETATION & PUBLICATION-READY SYNTHESIS")
print("=" * 78)

BASE = "AI_Polymer_Drug_Release"

# --------------------------------------------------------------------------
# INPUTS
# --------------------------------------------------------------------------

final_file = os.path.join(
    BASE,
    "final",
    "12_final_analysis_dataset.csv"
)

if not os.path.exists(final_file):
    raise FileNotFoundError(
        f"Final analysis dataset not found: {final_file}"
    )

df = pd.read_csv(final_file)

print(f"Input rows: {len(df)}")
print(f"Input columns: {len(df.columns)}")

# --------------------------------------------------------------------------
# COLUMN DETECTION
# --------------------------------------------------------------------------

def find_col(df, candidates, required=True):
    for c in candidates:
        if c in df.columns:
            return c
    if required:
        raise KeyError(
            f"Could not detect required column. "
            f"Tried: {candidates}. Available: {list(df.columns)}"
        )
    return None


study_col = find_col(
    df, ["study_id", "study_id_std"]
)

polymer_col = find_col(
    df, ["polymer", "polymer_std", "confirmed_polymers"],
    required=False
)

drug_col = find_col(
    df, ["drug", "drug_std", "confirmed_drugs"],
    required=False
)

time_col = find_col(
    df, ["release_time_h", "release_time_std"]
)

release_col = find_col(
    df, ["release_percent", "release_percent_std"]
)

inclusion_col = find_col(
    df,
    ["final_analysis_inclusion", "final_quantitative_candidate"],
    required=False
)

print()
print("Detected columns:")
print("Study:", study_col)
print("Polymer:", polymer_col)
print("Drug:", drug_col)
print("Time:", time_col)
print("Release:", release_col)
print("Inclusion:", inclusion_col)

# --------------------------------------------------------------------------
# FINAL ANALYTICAL DATASET
# --------------------------------------------------------------------------

work = df.copy()

work[time_col] = pd.to_numeric(work[time_col], errors="coerce")
work[release_col] = pd.to_numeric(work[release_col], errors="coerce")

if inclusion_col is not None:
    inclusion = work[inclusion_col].astype(str).str.lower().isin(
        ["true", "1", "yes"]
    )
else:
    inclusion = pd.Series(True, index=work.index)

valid = (
    inclusion
    & work[study_col].notna()
    & work[time_col].notna()
    & work[release_col].notna()
    & (work[time_col] >= 0)
    & (work[release_col] >= 0)
    & (work[release_col] <= 100)
)

analysis = work.loc[valid].copy()

print()
print("=" * 78)
print("FINAL EVIDENCE BASE")
print("=" * 78)

n_obs = len(analysis)
n_studies = analysis[study_col].nunique()

study_counts = (
    analysis.groupby(study_col)
    .size()
    .sort_values(ascending=False)
)

multi_studies = study_counts[study_counts >= 3]
single_studies = study_counts[study_counts == 1]

print(f"Final analytical observations: {n_obs}")
print(f"Independent studies: {n_studies}")
print(f"Multi-timepoint studies (>=3 observations): {len(multi_studies)}")
print(f"Single-observation studies: {len(single_studies)}")

# --------------------------------------------------------------------------
# STUDY-LEVEL EVIDENCE CLASSIFICATION
# --------------------------------------------------------------------------

study_rows = []

for study_id, g in analysis.groupby(study_col):

    g = g.sort_values(time_col)

    n = len(g)

    if n >= 3:
        evidence_level = "MULTI_TIMEPOINT_KINETIC_EVIDENCE"
        recommended_use = "WITHIN_STUDY_KINETIC_AND_DESCRIPTIVE"
    elif n == 2:
        evidence_level = "LIMITED_TWO_POINT_EVIDENCE"
        recommended_use = "DESCRIPTIVE_ONLY"
    else:
        evidence_level = "SINGLE_TIMEPOINT_EVIDENCE"
        recommended_use = "DESCRIPTIVE_ONLY"

    study_rows.append({
        "study_id": study_id,
        "observation_count": n,
        "first_time_h": g[time_col].iloc[0],
        "last_time_h": g[time_col].iloc[-1],
        "first_release_percent": g[release_col].iloc[0],
        "last_release_percent": g[release_col].iloc[-1],
        "release_change_percent":
            g[release_col].iloc[-1] - g[release_col].iloc[0],
        "maximum_release_percent": g[release_col].max(),
        "minimum_release_percent": g[release_col].min(),
        "median_release_percent": g[release_col].median(),
        "evidence_level": evidence_level,
        "recommended_use": recommended_use
    })

study_evidence = pd.DataFrame(study_rows)

# --------------------------------------------------------------------------
# OVERALL DESCRIPTIVE STATISTICS
# --------------------------------------------------------------------------

overall_mean = analysis[release_col].mean()
overall_median = analysis[release_col].median()
overall_sd = analysis[release_col].std()

print()
print("=" * 78)
print("OVERALL DESCRIPTIVE RELEASE STATISTICS")
print("=" * 78)

print(f"Mean release: {overall_mean:.3f}%")
print(f"Median release: {overall_median:.3f}%")
print(f"SD: {overall_sd:.3f}%")
print(f"Minimum: {analysis[release_col].min():.3f}%")
print(f"Maximum: {analysis[release_col].max():.3f}%")

# --------------------------------------------------------------------------
# TIME-WINDOW SYNTHESIS
# --------------------------------------------------------------------------

def assign_time_window(t):

    if t <= 1:
        return "Early (0–1 h)"
    elif t <= 6:
        return "Early-intermediate (1–6 h)"
    elif t <= 24:
        return "Intermediate (6–24 h)"
    elif t <= 72:
        return "Late (24–72 h)"
    else:
        return "Very late (>72 h)"


analysis["time_window"] = analysis[time_col].apply(assign_time_window)

window_summary = (
    analysis.groupby("time_window")
    .agg(
        observation_count=(release_col, "size"),
        study_count=(study_col, "nunique"),
        mean_release_percent=(release_col, "mean"),
        median_release_percent=(release_col, "median"),
        sd_release_percent=(release_col, "std"),
        min_release_percent=(release_col, "min"),
        max_release_percent=(release_col, "max")
    )
    .reset_index()
)

window_order = [
    "Early (0–1 h)",
    "Early-intermediate (1–6 h)",
    "Intermediate (6–24 h)",
    "Late (24–72 h)",
    "Very late (>72 h)"
]

window_summary["order"] = window_summary["time_window"].map(
    {x: i for i, x in enumerate(window_order)}
)

window_summary = (
    window_summary
    .sort_values("order")
    .drop(columns="order")
)

# --------------------------------------------------------------------------
# POLYMER–DRUG EVIDENCE MATRIX
# --------------------------------------------------------------------------

if polymer_col is not None and drug_col is not None:

    pair_df = analysis.copy()

    pair_df["polymer_clean"] = (
        pair_df[polymer_col]
        .fillna("UNKNOWN")
        .astype(str)
        .str.strip()
    )

    pair_df["drug_clean"] = (
        pair_df[drug_col]
        .fillna("UNKNOWN")
        .astype(str)
        .str.strip()
    )

    # Split semicolon-separated metadata into individual components
    polymer_long = (
        pair_df[
            [study_col, "polymer_clean", "drug_clean", release_col]
        ]
        .copy()
    )

    polymer_long["polymer"] = polymer_long["polymer_clean"].str.split(";")
    polymer_long["drug"] = polymer_long["drug_clean"].str.split(";")

    polymer_long = polymer_long.explode("polymer")
    polymer_long = polymer_long.explode("drug")

    polymer_long["polymer"] = polymer_long["polymer"].str.strip()
    polymer_long["drug"] = polymer_long["drug"].str.strip()

    pair_summary = (
        polymer_long
        .groupby(["polymer", "drug"])
        .agg(
            study_count=(study_col, "nunique"),
            observation_count=(release_col, "size"),
            mean_release_percent=(release_col, "mean"),
            median_release_percent=(release_col, "median")
        )
        .reset_index()
    )

    pair_summary["evidence_strength"] = np.where(
        pair_summary["study_count"] >= 3,
        "REPLICATED_ACROSS_MULTIPLE_STUDIES",
        np.where(
            pair_summary["study_count"] == 2,
            "LIMITED_CROSS_STUDY_REPLICATION",
            "SINGLE_STUDY_ASSOCIATION"
        )
    )

else:

    pair_summary = pd.DataFrame()

# --------------------------------------------------------------------------
# POLYMER SUMMARY
# --------------------------------------------------------------------------

if polymer_col is not None:

    polymer_long = analysis[
        [study_col, polymer_col, release_col]
    ].copy()

    polymer_long["polymer"] = (
        polymer_long[polymer_col]
        .fillna("UNKNOWN")
        .astype(str)
        .str.split(";")
    )

    polymer_long = polymer_long.explode("polymer")
    polymer_long["polymer"] = polymer_long["polymer"].str.strip()

    polymer_summary = (
        polymer_long
        .groupby("polymer")
        .agg(
            study_count=(study_col, "nunique"),
            observation_count=(release_col, "size"),
            mean_release_percent=(release_col, "mean"),
            median_release_percent=(release_col, "median")
        )
        .reset_index()
        .sort_values(
            ["study_count", "observation_count"],
            ascending=False
        )
    )

else:
    polymer_summary = pd.DataFrame()

# --------------------------------------------------------------------------
# DRUG SUMMARY
# --------------------------------------------------------------------------

if drug_col is not None:

    drug_long = analysis[
        [study_col, drug_col, release_col]
    ].copy()

    drug_long["drug"] = (
        drug_long[drug_col]
        .fillna("UNKNOWN")
        .astype(str)
        .str.split(";")
    )

    drug_long = drug_long.explode("drug")
    drug_long["drug"] = drug_long["drug"].str.strip()

    drug_summary = (
        drug_long
        .groupby("drug")
        .agg(
            study_count=(study_col, "nunique"),
            observation_count=(release_col, "size"),
            mean_release_percent=(release_col, "mean"),
            median_release_percent=(release_col, "median")
        )
        .reset_index()
        .sort_values(
            ["study_count", "observation_count"],
            ascending=False
        )
    )

else:
    drug_summary = pd.DataFrame()

# --------------------------------------------------------------------------
# KINETIC EVIDENCE SUMMARY
# --------------------------------------------------------------------------

kinetic_rows = []

for study_id, g in analysis.groupby(study_col):

    g = g.sort_values(time_col)

    if len(g) < 3:
        continue

    times = g[time_col].values
    releases = g[release_col].values

    kinetic_rows.append({
        "study_id": study_id,
        "n_timepoints": len(g),
        "first_time_h": times[0],
        "last_time_h": times[-1],
        "first_release_percent": releases[0],
        "last_release_percent": releases[-1],
        "release_change_percent":
            releases[-1] - releases[0],
        "kinetic_interpretation":
            "Within-study kinetic evidence; model selection should be interpreted cautiously"
    })

kinetic_summary = pd.DataFrame(kinetic_rows)

# --------------------------------------------------------------------------
# ROBUSTNESS / INFLUENCE SUMMARY
# --------------------------------------------------------------------------

study_mean = (
    analysis.groupby(study_col)[release_col]
    .mean()
    .sort_values(ascending=False)
)

study_contribution = (
    analysis.groupby(study_col)
    .size()
    .sort_values(ascending=False)
)

top_n = min(3, len(study_contribution))

top_observations = study_contribution.head(top_n).sum()

top_proportion = top_observations / n_obs

# Leave-one-study-out mean
loo_rows = []

for study_id in analysis[study_col].unique():

    subset = analysis.loc[
        analysis[study_col] != study_id,
        release_col
    ]

    loo_rows.append({
        "excluded_study": study_id,
        "remaining_observations": len(subset),
        "remaining_studies": subset.shape[0] * 0 + (
            analysis.loc[
                analysis[study_col] != study_id,
                study_col
            ].nunique()
        ),
        "mean_release_percent": subset.mean(),
        "median_release_percent": subset.median()
    })

loo_summary = pd.DataFrame(loo_rows)

# --------------------------------------------------------------------------
# FINAL ANALYTICAL DECISION MATRIX
# --------------------------------------------------------------------------

decision_matrix = pd.DataFrame([
    {
        "analysis": "Study-level descriptive synthesis",
        "decision": "RECOMMENDED",
        "justification":
            "Studies represent the independent evidence units."
    },
    {
        "analysis": "Within-study release-curve analysis",
        "decision": "RECOMMENDED",
        "justification":
            "Repeated timepoints can be interpreted within individual studies."
    },
    {
        "analysis": "Kinetic modeling",
        "decision": "RESTRICTED",
        "justification":
            f"Only {len(multi_studies)} studies contain at least three usable timepoints."
    },
    {
        "analysis": "Observation-level pooled regression",
        "decision": "NOT RECOMMENDED",
        "justification":
            "Repeated observations violate simple independence assumptions."
    },
    {
        "analysis": "Formal meta-analysis",
        "decision": "NOT RECOMMENDED",
        "justification":
            "Release percentages were reported at heterogeneous timepoints and under heterogeneous formulations."
    },
    {
        "analysis": "Polymer-group statistical comparison",
        "decision": "CAUTION",
        "justification":
            "Polymer identity is confounded with drug identity and formulation."
    },
    {
        "analysis": "Drug-group statistical comparison",
        "decision": "CAUTION",
        "justification":
            "Drug physicochemical properties and formulations differ between studies."
    },
    {
        "analysis": "Machine learning",
        "decision": "NOT RECOMMENDED",
        "justification":
            f"Only {n_studies} independent studies are available and the evidence is highly heterogeneous."
    },
    {
        "analysis": "Narrative quantitative synthesis",
        "decision": "RECOMMENDED",
        "justification":
            "The dataset supports structured quantitative comparison without unjustified pooling."
    }
])

# --------------------------------------------------------------------------
# FINAL EVIDENCE INTERPRETATION
# --------------------------------------------------------------------------

if top_proportion >= 0.60:
    dominance_statement = (
        f"The three most densely sampled studies contribute "
        f"{top_observations}/{n_obs} observations "
        f"({top_proportion*100:.1f}%), indicating substantial observation-level dominance."
    )
else:
    dominance_statement = (
        "No extreme observation-level dominance by the largest studies was identified."
    )

if len(multi_studies) <= 3:
    kinetic_statement = (
        f"Kinetic evidence is limited because only {len(multi_studies)} "
        "studies contain at least three quantitative timepoints."
    )
else:
    kinetic_statement = (
        f"{len(multi_studies)} studies contain at least three quantitative timepoints."
    )

final_interpretation = pd.DataFrame([
    {
        "domain": "Evidence base",
        "finding":
            f"{n_obs} final analytical observations from {n_studies} independent studies."
    },
    {
        "domain": "Overall release",
        "finding":
            f"Unweighted descriptive mean {overall_mean:.2f}% and median {overall_median:.2f}%."
    },
    {
        "domain": "Kinetic evidence",
        "finding": kinetic_statement
    },
    {
        "domain": "Observation dependence",
        "finding":
            "Repeated measurements are clustered within studies and should not be treated as independent replicates."
    },
    {
        "domain": "Cross-study comparison",
        "finding":
            "Polymer and drug effects are strongly confounded by formulation and experimental differences."
    },
    {
        "domain": "Robustness",
        "finding":
            "Descriptive estimates are sensitive to study exclusion and extreme observations."
    },
    {
        "domain": "Meta-analysis",
        "finding":
            "Formal pooled meta-analysis is not justified for the current evidence structure."
    },
    {
        "domain": "Machine learning",
        "finding":
            "Predictive machine learning is not justified with the current independent-study sample size."
    },
    {
        "domain": "Recommended synthesis",
        "finding":
            "Narrative quantitative synthesis supplemented by within-study kinetic interpretation."
    },
    {
        "domain": "Observation dominance",
        "finding": dominance_statement
    }
])

# --------------------------------------------------------------------------
# OUTPUT DIRECTORY
# --------------------------------------------------------------------------

output_dir = os.path.join(
    BASE,
    "analysis",
    "19_final_evidence"
)

os.makedirs(output_dir, exist_ok=True)

# --------------------------------------------------------------------------
# SAVE OUTPUTS
# --------------------------------------------------------------------------

study_evidence.to_csv(
    os.path.join(
        output_dir,
        "19_study_level_evidence_classification.csv"
    ),
    index=False
)

window_summary.to_csv(
    os.path.join(
        output_dir,
        "19_time_window_final_synthesis.csv"
    ),
    index=False
)

pair_summary.to_csv(
    os.path.join(
        output_dir,
        "19_polymer_drug_evidence_matrix.csv"
    ),
    index=False
)

polymer_summary.to_csv(
    os.path.join(
        output_dir,
        "19_polymer_final_summary.csv"
    ),
    index=False
)

drug_summary.to_csv(
    os.path.join(
        output_dir,
        "19_drug_final_summary.csv"
    ),
    index=False
)

kinetic_summary.to_csv(
    os.path.join(
        output_dir,
        "19_kinetic_evidence_summary.csv"
    ),
    index=False
)

loo_summary.to_csv(
    os.path.join(
        output_dir,
        "19_leave_one_study_out_summary.csv"
    ),
    index=False
)

decision_matrix.to_csv(
    os.path.join(
        output_dir,
        "19_final_analytical_decision_matrix.csv"
    ),
    index=False
)

final_interpretation.to_csv(
    os.path.join(
        output_dir,
        "19_final_evidence_interpretation.csv"
    ),
    index=False
)

# --------------------------------------------------------------------------
# FINAL REPORT TEXT
# --------------------------------------------------------------------------

report_lines = []

report_lines.append(
    "FINAL QUANTITATIVE EVIDENCE SYNTHESIS"
)
report_lines.append("=" * 60)
report_lines.append("")
report_lines.append(
    f"Final analytical observations: {n_obs}"
)
report_lines.append(
    f"Independent studies: {n_studies}"
)
report_lines.append(
    f"Multi-timepoint studies: {len(multi_studies)}"
)
report_lines.append(
    f"Single-observation studies: {len(single_studies)}"
)
report_lines.append("")
report_lines.append(
    f"Unweighted mean release: {overall_mean:.3f}%"
)
report_lines.append(
    f"Unweighted median release: {overall_median:.3f}%"
)
report_lines.append(
    f"Standard deviation: {overall_sd:.3f}%"
)
report_lines.append("")
report_lines.append(
    "INTERPRETATION"
)
report_lines.append(
    "The available quantitative evidence is heterogeneous and "
    "should be interpreted primarily at the study level."
)
report_lines.append(
    "Repeated observations within studies should not be treated "
    "as independent experimental replicates."
)
report_lines.append(
    kinetic_statement
)
report_lines.append(
    "Formal meta-analysis and observation-level pooled regression "
    "are not recommended."
)
report_lines.append(
    "Machine-learning prediction is not recommended because of "
    "the small number of independent studies and substantial "
    "formulation heterogeneity."
)
report_lines.append(
    "The appropriate synthesis is a structured narrative "
    "quantitative synthesis with within-study kinetic analysis "
    "where sufficient timepoints are available."
)
report_lines.append("")
report_lines.append(
    dominance_statement
)

with open(
    os.path.join(
        output_dir,
        "19_final_evidence_report.txt"
    ),
    "w",
    encoding="utf-8"
) as f:
    f.write("\n".join(report_lines))

# --------------------------------------------------------------------------
# CONSOLE SUMMARY
# --------------------------------------------------------------------------

print()
print("=" * 78)
print("CODE 19 COMPLETE")
print("=" * 78)

print(f"Final analytical observations: {n_obs}")
print(f"Independent studies: {n_studies}")
print(f"Multi-timepoint studies: {len(multi_studies)}")
print(f"Single-observation studies: {len(single_studies)}")

print()
print("Overall descriptive release:")
print(f"Mean: {overall_mean:.3f}%")
print(f"Median: {overall_median:.3f}%")
print(f"SD: {overall_sd:.3f}%")

print()
print("Evidence interpretation:")
print("Formal meta-analysis: NOT RECOMMENDED")
print("Observation-level pooling: NOT RECOMMENDED")
print("Machine learning: NOT RECOMMENDED")
print("Narrative quantitative synthesis: RECOMMENDED")
print("Within-study kinetic analysis: RECOMMENDED WHERE ELIGIBLE")

print()
print("Observation dominance:")
print(dominance_statement)

print()
print("Files created:")

for filename in [
    "19_study_level_evidence_classification.csv",
    "19_time_window_final_synthesis.csv",
    "19_polymer_drug_evidence_matrix.csv",
    "19_polymer_final_summary.csv",
    "19_drug_final_summary.csv",
    "19_kinetic_evidence_summary.csv",
    "19_leave_one_study_out_summary.csv",
    "19_final_analytical_decision_matrix.csv",
    "19_final_evidence_interpretation.csv",
    "19_final_evidence_report.txt"
]:
    print(os.path.join(output_dir, filename))

print("=" * 78)
```

```python
# ==============================================================================
# CODE 19 — FINAL EVIDENCE INTERPRETATION & PUBLICATION-READY SYNTHESIS
# ==============================================================================

import os
import pandas as pd
import numpy as np

print("=" * 78)
print("CODE 19 — FINAL EVIDENCE INTERPRETATION & PUBLICATION-READY SYNTHESIS")
print("=" * 78)

BASE = "AI_Polymer_Drug_Release"

# --------------------------------------------------------------------------
# CORRECT INPUT PATH FROM CODE 12
# --------------------------------------------------------------------------

final_file = os.path.join(
    BASE,
    "data",
    "final",
    "12_final_analysis_dataset.csv"
)

print(f"Input file: {final_file}")

if not os.path.exists(final_file):
    raise FileNotFoundError(
        f"Final analysis dataset not found: {final_file}"
    )

df = pd.read_csv(final_file)

print(f"Input rows: {len(df)}")
print(f"Input columns: {len(df.columns)}")
print(f"Available columns: {list(df.columns)}")
```

```python
# ==============================================================================
# CODE 20 — FINAL STATISTICAL & PUBLICATION AUDIT
# ==============================================================================

import os
import pandas as pd
import numpy as np

print("=" * 78)
print("CODE 20 — FINAL STATISTICAL & PUBLICATION AUDIT")
print("=" * 78)

# ------------------------------------------------------------------------------
# PATHS
# ------------------------------------------------------------------------------

final_file = (
    "AI_Polymer_Drug_Release/data/final/"
    "12_final_analysis_dataset.csv"
)

out_dir = "AI_Polymer_Drug_Release/data/analysis/20_final_audit"
os.makedirs(out_dir, exist_ok=True)

if not os.path.exists(final_file):
    raise FileNotFoundError(
        f"Final analysis dataset not found: {final_file}"
    )

# ------------------------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------------------------

df = pd.read_csv(final_file)

print(f"\nInput file: {final_file}")
print(f"Input rows: {len(df)}")
print(f"Input columns: {len(df.columns)}")
print("\nColumns:")
print(list(df.columns))

# ------------------------------------------------------------------------------
# COLUMN DETECTION
# ------------------------------------------------------------------------------

def find_col(df, candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None

study_col = find_col(df, [
    "study_id", "study_id_std"
])

pmid_col = find_col(df, [
    "pmid", "pmid_std"
])

polymer_col = find_col(df, [
    "polymer", "polymer_std", "confirmed_polymers"
])

drug_col = find_col(df, [
    "drug", "drug_std", "confirmed_drugs"
])

time_col = find_col(df, [
    "release_time_h", "release_time_std"
])

release_col = find_col(df, [
    "release_percent", "release_percent_std"
])

include_col = find_col(df, [
    "final_analysis_inclusion"
])

valid_time_col = find_col(df, [
    "valid_time", "time_valid"
])

valid_release_col = find_col(df, [
    "valid_release", "release_percent_valid"
])

repeated_col = find_col(df, [
    "repeated_measurement"
])

print("\n" + "=" * 78)
print("DETECTED COLUMNS")
print("=" * 78)

print("Study:", study_col)
print("PMID:", pmid_col)
print("Polymer:", polymer_col)
print("Drug:", drug_col)
print("Time:", time_col)
print("Release:", release_col)
print("Final inclusion:", include_col)
print("Valid time:", valid_time_col)
print("Valid release:", valid_release_col)
print("Repeated measurement:", repeated_col)

required = [
    study_col,
    polymer_col,
    drug_col,
    time_col,
    release_col,
    include_col
]

if any(x is None for x in required):
    raise KeyError(
        "One or more required columns could not be detected."
    )

# ------------------------------------------------------------------------------
# NORMALIZE INCLUSION
# ------------------------------------------------------------------------------

def normalize_bool(x):
    if pd.isna(x):
        return False

    if isinstance(x, bool):
        return x

    s = str(x).strip().lower()

    return s in {
        "true",
        "1",
        "yes",
        "y",
        "included",
        "include"
    }

df["_include_bool"] = df[include_col].apply(normalize_bool)

analysis = df[df["_include_bool"]].copy()

print("\n" + "=" * 78)
print("FINAL ANALYTICAL DATASET")
print("=" * 78)

print("Rows in source file:", len(df))
print("Included observations:", len(analysis))
print("Excluded observations:", len(df) - len(analysis))

# ------------------------------------------------------------------------------
# STUDY COUNTS
# ------------------------------------------------------------------------------

independent_studies = analysis[study_col].nunique()

study_counts = (
    analysis.groupby(study_col)
    .size()
    .reset_index(name="observation_count")
    .sort_values("observation_count", ascending=False)
)

study_counts["repeated_measurement"] = (
    study_counts["observation_count"] > 1
)

multi_study_count = int(
    (study_counts["observation_count"] >= 3).sum()
)

single_point_studies = int(
    (study_counts["observation_count"] == 1).sum()
)

repeated_studies = int(
    (study_counts["observation_count"] > 1).sum()
)

print("\nIndependent studies:", independent_studies)
print("Observations:", len(analysis))
print("Studies with >=3 timepoints:", multi_study_count)
print("Single-observation studies:", single_point_studies)
print("Studies with repeated measurements:", repeated_studies)

print("\nObservation contribution:")
print(study_counts.to_string(index=False))

# ------------------------------------------------------------------------------
# DATA VALIDITY
# ------------------------------------------------------------------------------

analysis["_time_numeric"] = pd.to_numeric(
    analysis[time_col],
    errors="coerce"
)

analysis["_release_numeric"] = pd.to_numeric(
    analysis[release_col],
    errors="coerce"
)

invalid_time = (
    analysis["_time_numeric"].isna()
    | (analysis["_time_numeric"] < 0)
)

invalid_release = (
    analysis["_release_numeric"].isna()
    | (analysis["_release_numeric"] < 0)
    | (analysis["_release_numeric"] > 100)
)

print("\n" + "=" * 78)
print("DATA VALIDITY AUDIT")
print("=" * 78)

print("Missing/invalid time:", int(invalid_time.sum()))
print("Invalid release percentage:", int(invalid_release.sum()))

# ------------------------------------------------------------------------------
# POLYMER / DRUG COMPLETENESS
# ------------------------------------------------------------------------------

missing_polymer = (
    analysis[polymer_col].isna()
    | (
        analysis[polymer_col]
        .astype(str)
        .str.strip()
        .isin(["", "nan", "None", "UNKNOWN"])
    )
)

missing_drug = (
    analysis[drug_col].isna()
    | (
        analysis[drug_col]
        .astype(str)
        .str.strip()
        .isin(["", "nan", "None", "UNKNOWN"])
    )
)

print("\n" + "=" * 78)
print("FORMULATION METADATA AUDIT")
print("=" * 78)

print("Missing polymer:", int(missing_polymer.sum()))
print("Missing drug:", int(missing_drug.sum()))

if missing_drug.sum() > 0:
    print("\nObservations with missing drug:")
    cols = [
        study_col,
        pmid_col,
        polymer_col,
        drug_col,
        time_col,
        release_col
    ]

    cols = [c for c in cols if c is not None]

    print(
        analysis.loc[missing_drug, cols]
        .to_string(index=False)
    )

# ------------------------------------------------------------------------------
# IMPORTANT EXCLUDED RECORD AUDIT
# ------------------------------------------------------------------------------

excluded = df[~df["_include_bool"]].copy()

print("\n" + "=" * 78)
print("EXCLUDED RECORD AUDIT")
print("=" * 78)

print("Excluded records:", len(excluded))

if len(excluded) > 0:
    cols = [
        study_col,
        pmid_col,
        polymer_col,
        drug_col,
        time_col,
        release_col,
        include_col
    ]

    cols = [c for c in cols if c is not None]

    print(
        excluded[cols]
        .to_string(index=False)
    )

# ------------------------------------------------------------------------------
# KINETIC ELIGIBILITY
# ------------------------------------------------------------------------------

kinetic_counts = (
    analysis.groupby(study_col)[time_col]
    .nunique()
    .reset_index(name="number_of_timepoints")
)

kinetic_counts["kinetic_eligible"] = (
    kinetic_counts["number_of_timepoints"] >= 3
)

print("\n" + "=" * 78)
print("KINETIC ELIGIBILITY")
print("=" * 78)

print(
    kinetic_counts
    .sort_values("number_of_timepoints", ascending=False)
    .to_string(index=False)
)

print(
    "\nKinetic-eligible studies:",
    int(kinetic_counts["kinetic_eligible"].sum())
)

# ------------------------------------------------------------------------------
# REPEATED-MEASUREMENT DEPENDENCE
# ------------------------------------------------------------------------------

total_observations = len(analysis)

if total_observations > 0:
    study_counts["proportion_of_observations"] = (
        study_counts["observation_count"]
        / total_observations
    )

top3_obs = (
    study_counts
    .head(3)["observation_count"]
    .sum()
)

top3_fraction = top3_obs / total_observations

print("\n" + "=" * 78)
print("INDEPENDENCE / PSEUDOREPLICATION AUDIT")
print("=" * 78)

print(
    f"Top 3 studies contribute "
    f"{top3_obs}/{total_observations} observations "
    f"({top3_fraction * 100:.1f}%)."
)

print(
    "Observation-to-study ratio:",
    round(total_observations / independent_studies, 3)
)

# ------------------------------------------------------------------------------
# TIME-RELEASE DESCRIPTIVE STATISTICS
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("DESCRIPTIVE RELEASE STATISTICS")
print("=" * 78)

release_stats = analysis["_release_numeric"].describe()

print(release_stats)

time_stats = analysis["_time_numeric"].describe()

print("\nRelease-time statistics:")
print(time_stats)

# ------------------------------------------------------------------------------
# STUDY-LEVEL SUMMARY
# ------------------------------------------------------------------------------

study_summary = (
    analysis.groupby(study_col)
    .agg(
        observation_count=(release_col, "count"),
        first_time_h=(time_col, "min"),
        last_time_h=(time_col, "max"),
        first_release_percent=(release_col, "first"),
        last_release_percent=(release_col, "last"),
        mean_release_percent=(release_col, "mean"),
        median_release_percent=(release_col, "median"),
        maximum_release_percent=(release_col, "max"),
        minimum_release_percent=(release_col, "min")
    )
    .reset_index()
)

study_summary["multi_timepoint"] = (
    study_summary["observation_count"] >= 3
)

study_summary["release_change_percent"] = (
    study_summary["last_release_percent"]
    - study_summary["first_release_percent"]
)

print("\n" + "=" * 78)
print("STUDY-LEVEL SUMMARY")
print("=" * 78)

print(
    study_summary
    .sort_values("observation_count", ascending=False)
    .to_string(index=False)
)

# ------------------------------------------------------------------------------
# POLYMER COMPARABILITY
# ------------------------------------------------------------------------------

polymer_study_counts = (
    analysis.groupby(polymer_col)[study_col]
    .nunique()
    .reset_index(name="study_count")
    .sort_values("study_count", ascending=False)
)

polymer_study_counts["comparison_strength"] = np.where(
    polymer_study_counts["study_count"] >= 3,
    "REPEATED_ACROSS_STUDIES",
    np.where(
        polymer_study_counts["study_count"] == 2,
        "LIMITED_REPLICATION",
        "SINGLE_STUDY_OR_UNIQUE"
    )
)

print("\n" + "=" * 78)
print("POLYMER COMPARABILITY")
print("=" * 78)

print(
    polymer_study_counts.to_string(index=False)
)

# ------------------------------------------------------------------------------
# DRUG COMPARABILITY
# ------------------------------------------------------------------------------

drug_study_counts = (
    analysis.dropna(subset=[drug_col])
    .groupby(drug_col)[study_col]
    .nunique()
    .reset_index(name="study_count")
    .sort_values("study_count", ascending=False)
)

drug_study_counts["comparison_strength"] = np.where(
    drug_study_counts["study_count"] >= 3,
    "REPEATED_ACROSS_STUDIES",
    np.where(
        drug_study_counts["study_count"] == 2,
        "LIMITED_REPLICATION",
        "SINGLE_STUDY_OR_UNIQUE"
    )
)

print("\n" + "=" * 78)
print("DRUG COMPARABILITY")
print("=" * 78)

print(
    drug_study_counts.to_string(index=False)
)

# ------------------------------------------------------------------------------
# PUBLICATION-LEVEL ANALYTICAL DECISION MATRIX
# ------------------------------------------------------------------------------

decision_matrix = pd.DataFrame([
    {
        "analysis": "Study-level descriptive synthesis",
        "status": "RECOMMENDED",
        "reason": (
            "The study is the independent evidence unit."
        )
    },
    {
        "analysis": "Within-study release-curve analysis",
        "status": "RECOMMENDED",
        "reason": (
            "Repeated timepoints are appropriately interpreted "
            "within individual studies."
        )
    },
    {
        "analysis": "Kinetic modeling",
        "status": "RESTRICTED",
        "reason": (
            f"Only {multi_study_count} studies have at least "
            "three quantitative timepoints."
        )
    },
    {
        "analysis": "Observation-level pooled regression",
        "status": "NOT_RECOMMENDED",
        "reason": (
            "Repeated observations violate simple independence "
            "assumptions and create pseudoreplication."
        )
    },
    {
        "analysis": "Formal meta-analysis",
        "status": "NOT_RECOMMENDED",
        "reason": (
            "Release percentages are reported at heterogeneous "
            "timepoints and under different formulations and conditions."
        )
    },
    {
        "analysis": "Polymer-group inferential comparison",
        "status": "CAUTION",
        "reason": (
            "Polymer identity is confounded with drug, formulation, "
            "and experimental conditions."
        )
    },
    {
        "analysis": "Drug-group inferential comparison",
        "status": "CAUTION",
        "reason": (
            "Drug physicochemical properties and formulation "
            "conditions are heterogeneous."
        )
    },
    {
        "analysis": "Machine learning",
        "status": "NOT_RECOMMENDED",
        "reason": (
            f"Only {independent_studies} independent studies and "
            f"{total_observations} analytical observations are available."
        )
    },
    {
        "analysis": "Narrative quantitative synthesis",
        "status": "RECOMMENDED",
        "reason": (
            "The dataset supports structured quantitative comparison "
            "without unjustified pooling."
        )
    }
])

print("\n" + "=" * 78)
print("PUBLICATION-LEVEL ANALYTICAL DECISION MATRIX")
print("=" * 78)

print(decision_matrix.to_string(index=False))

# ------------------------------------------------------------------------------
# FINAL AUDIT STATUS
# ------------------------------------------------------------------------------

audit_checks = {
    "25_or_fewer_final_analytical_observations":
        total_observations == 25,

    "11_independent_studies":
        independent_studies == 11,

    "no_invalid_release_values":
        int(invalid_release.sum()) == 0,

    "no_invalid_time_values":
        int(invalid_time.sum()) == 0,

    "three_or_fewer_kinetic_studies":
        multi_study_count <= 3,

    "repeated_measurements_identified":
        repeated_studies >= 1,

    "formal_meta_analysis_restricted":
        True,

    "observation_level_pooling_restricted":
        True,

    "machine_learning_restricted":
        True
}

audit_table = pd.DataFrame([
    {
        "audit_check": key,
        "passed": value
    }
    for key, value in audit_checks.items()
])

overall_pass = all(audit_checks.values())

print("\n" + "=" * 78)
print("FINAL AUDIT RESULT")
print("=" * 78)

print(
    audit_table.to_string(index=False)
)

print("\nOverall audit status:")

if overall_pass:
    print("PASS — DATASET IS READY FOR PUBLICATION-LEVEL SYNTHESIS")
else:
    print("REVIEW REQUIRED — ONE OR MORE AUDIT CHECKS FAILED")

# ------------------------------------------------------------------------------
# PUBLICATION-SAFE SUMMARY
# ------------------------------------------------------------------------------

publication_summary = pd.DataFrame([{
    "final_observations": total_observations,
    "independent_studies": independent_studies,
    "kinetic_eligible_studies": multi_study_count,
    "single_observation_studies": single_point_studies,
    "repeated_measurement_studies": repeated_studies,
    "top_3_observation_fraction": round(top3_fraction, 4),
    "mean_release_percent": round(
        analysis["_release_numeric"].mean(), 4
    ),
    "median_release_percent": round(
        analysis["_release_numeric"].median(), 4
    ),
    "formal_meta_analysis": "NOT_RECOMMENDED",
    "observation_level_pooling": "NOT_RECOMMENDED",
    "machine_learning": "NOT_RECOMMENDED",
    "narrative_quantitative_synthesis": "RECOMMENDED",
    "within_study_kinetic_analysis": "RECOMMENDED_WHERE_ELIGIBLE",
    "overall_audit_status": (
        "PASS" if overall_pass else "REVIEW_REQUIRED"
    )
}])

# ------------------------------------------------------------------------------
# SAVE OUTPUTS
# ------------------------------------------------------------------------------

study_summary.to_csv(
    os.path.join(
        out_dir,
        "20_final_study_level_audit.csv"
    ),
    index=False
)

polymer_study_counts.to_csv(
    os.path.join(
        out_dir,
        "20_polymer_comparability_audit.csv"
    ),
    index=False
)

drug_study_counts.to_csv(
    os.path.join(
        out_dir,
        "20_drug_comparability_audit.csv"
    ),
    index=False
)

decision_matrix.to_csv(
    os.path.join(
        out_dir,
        "20_publication_analysis_decision_matrix.csv"
    ),
    index=False
)

audit_table.to_csv(
    os.path.join(
        out_dir,
        "20_final_audit_checks.csv"
    ),
    index=False
)

publication_summary.to_csv(
    os.path.join(
        out_dir,
        "20_publication_safe_summary.csv"
    ),
    index=False
)

kinetic_counts.to_csv(
    os.path.join(
        out_dir,
        "20_kinetic_eligibility_audit.csv"
    ),
    index=False
)

print("\n" + "=" * 78)
print("CODE 20 COMPLETE")
print("=" * 78)

print("Final analytical observations:", total_observations)
print("Independent studies:", independent_studies)
print("Kinetic-eligible studies:", multi_study_count)
print("Repeated-measurement studies:", repeated_studies)

print(
    "Overall audit:",
    "PASS" if overall_pass else "REVIEW REQUIRED"
)

print("\nFiles created:")
for filename in [
    "20_final_study_level_audit.csv",
    "20_polymer_comparability_audit.csv",
    "20_drug_comparability_audit.csv",
    "20_publication_analysis_decision_matrix.csv",
    "20_final_audit_checks.csv",
    "20_publication_safe_summary.csv",
    "20_kinetic_eligibility_audit.csv"
]:
    print("-", os.path.join(out_dir, filename))

print("=" * 78)
```

```python
# ==============================================================================
# CODE 21 — PUBLICATION-READY TABLES, FIGURES & FINAL RESULTS PACKAGE
# ==============================================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 78)
print("CODE 21 — PUBLICATION-READY TABLES, FIGURES & FINAL RESULTS PACKAGE")
print("=" * 78)

# ------------------------------------------------------------------------------
# PATHS
# ------------------------------------------------------------------------------

base_dir = "AI_Polymer_Drug_Release"

final_file = os.path.join(
    base_dir,
    "data",
    "final",
    "12_final_analysis_dataset.csv"
)

audit_dir = os.path.join(
    base_dir,
    "data",
    "analysis",
    "20_final_audit"
)

output_dir = os.path.join(
    base_dir,
    "data",
    "publication",
    "21_publication_package"
)

os.makedirs(output_dir, exist_ok=True)

# ------------------------------------------------------------------------------
# LOAD FINAL DATASET
# ------------------------------------------------------------------------------

if not os.path.exists(final_file):
    raise FileNotFoundError(
        f"Final analysis dataset not found: {final_file}"
    )

df = pd.read_csv(final_file)

print("\nInput file:", final_file)
print("Input rows:", len(df))
print("Input columns:", len(df))

# ------------------------------------------------------------------------------
# COLUMN DETECTION
# ------------------------------------------------------------------------------

required = [
    "study_id",
    "polymer",
    "drug",
    "release_time_h",
    "release_percent",
    "final_analysis_inclusion"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

# ------------------------------------------------------------------------------
# FINAL ANALYTICAL DATASET
# ------------------------------------------------------------------------------

analysis_df = df[
    df["final_analysis_inclusion"].astype(str).str.lower().isin(
        ["true", "1", "yes"]
    )
].copy()

analysis_df["release_percent"] = pd.to_numeric(
    analysis_df["release_percent"],
    errors="coerce"
)

analysis_df["release_time_h"] = pd.to_numeric(
    analysis_df["release_time_h"],
    errors="coerce"
)

analysis_df = analysis_df.dropna(
    subset=["study_id", "release_percent", "release_time_h"]
)

print("\nFinal analytical observations:", len(analysis_df))
print("Independent studies:", analysis_df["study_id"].nunique())

# ==============================================================================
# TABLE 1 — STUDY-LEVEL CHARACTERISTICS
# ==============================================================================

study_rows = []

for study_id, g in analysis_df.groupby("study_id"):

    g = g.sort_values("release_time_h")

    study_rows.append({
        "study_id": study_id,
        "polymer": g["polymer"].iloc[0],
        "drug": g["drug"].iloc[0],
        "n_observations": len(g),
        "n_timepoints": g["release_time_h"].nunique(),
        "first_time_h": g["release_time_h"].min(),
        "last_time_h": g["release_time_h"].max(),
        "first_release_percent": g["release_percent"].iloc[0],
        "last_release_percent": g["release_percent"].iloc[-1],
        "maximum_release_percent": g["release_percent"].max(),
        "minimum_release_percent": g["release_percent"].min(),
        "median_release_percent": g["release_percent"].median(),
        "multi_timepoint": g["release_time_h"].nunique() >= 3
    })

study_table = pd.DataFrame(study_rows)

study_table.to_csv(
    os.path.join(output_dir, "21_table_1_study_characteristics.csv"),
    index=False
)

print("\nTABLE 1 CREATED")
print(study_table)

# ==============================================================================
# TABLE 2 — OVERALL QUANTITATIVE SUMMARY
# ==============================================================================

overall_summary = pd.DataFrame([{
    "final_observations": len(analysis_df),
    "independent_studies": analysis_df["study_id"].nunique(),
    "multi_timepoint_studies": int(
        (study_table["multi_timepoint"]).sum()
    ),
    "single_timepoint_studies": int(
        (~study_table["multi_timepoint"]).sum()
    ),
    "mean_release_percent": analysis_df["release_percent"].mean(),
    "sd_release_percent": analysis_df["release_percent"].std(),
    "median_release_percent": analysis_df["release_percent"].median(),
    "min_release_percent": analysis_df["release_percent"].min(),
    "max_release_percent": analysis_df["release_percent"].max(),
    "mean_release_time_h": analysis_df["release_time_h"].mean(),
    "median_release_time_h": analysis_df["release_time_h"].median(),
    "max_release_time_h": analysis_df["release_time_h"].max()
}])

overall_summary.to_csv(
    os.path.join(output_dir, "21_table_2_overall_quantitative_summary.csv"),
    index=False
)

print("\nTABLE 2 CREATED")
print(overall_summary.T)

# ==============================================================================
# TABLE 3 — KINETIC ELIGIBILITY
# ==============================================================================

kinetic_table = study_table[
    study_table["n_timepoints"] >= 3
].copy()

kinetic_table["kinetic_eligibility"] = "Eligible"

kinetic_table.to_csv(
    os.path.join(output_dir, "21_table_3_kinetic_eligibility.csv"),
    index=False
)

print("\nTABLE 3 CREATED")
print(kinetic_table)

# ==============================================================================
# TABLE 4 — STUDY CONTRIBUTION / INDEPENDENCE
# ==============================================================================

contribution = (
    analysis_df.groupby("study_id")
    .size()
    .reset_index(name="observation_count")
)

contribution["proportion_of_observations"] = (
    contribution["observation_count"] /
    len(analysis_df)
)

contribution = contribution.sort_values(
    "observation_count",
    ascending=False
)

contribution.to_csv(
    os.path.join(
        output_dir,
        "21_table_4_study_observation_contribution.csv"
    ),
    index=False
)

# ==============================================================================
# TABLE 5 — TIME-WINDOW DESCRIPTIVE SYNTHESIS
# ==============================================================================

def classify_time(t):

    if t <= 1:
        return "Early (0–1 h)"
    elif t <= 6:
        return "Early-intermediate (1–6 h)"
    elif t <= 24:
        return "Intermediate (6–24 h)"
    elif t <= 72:
        return "Late (24–72 h)"
    else:
        return "Very late (>72 h)"

analysis_df["time_window"] = analysis_df[
    "release_time_h"
].apply(classify_time)

window_table = (
    analysis_df
    .groupby("time_window")
    .agg(
        observation_count=("release_percent", "count"),
        study_count=("study_id", "nunique"),
        mean_release_percent=("release_percent", "mean"),
        median_release_percent=("release_percent", "median"),
        sd_release_percent=("release_percent", "std"),
        min_release_percent=("release_percent", "min"),
        max_release_percent=("release_percent", "max")
    )
    .reset_index()
)

window_order = [
    "Early (0–1 h)",
    "Early-intermediate (1–6 h)",
    "Intermediate (6–24 h)",
    "Late (24–72 h)",
    "Very late (>72 h)"
]

window_table["time_window"] = pd.Categorical(
    window_table["time_window"],
    categories=window_order,
    ordered=True
)

window_table = window_table.sort_values("time_window")

window_table.to_csv(
    os.path.join(
        output_dir,
        "21_table_5_time_window_synthesis.csv"
    ),
    index=False
)

# ==============================================================================
# FIGURE 1 — RELEASE DISTRIBUTION
# ==============================================================================

plt.figure(figsize=(8, 5))

plt.hist(
    analysis_df["release_percent"],
    bins=10,
    edgecolor="black"
)

plt.xlabel("Drug release (%)")
plt.ylabel("Number of observations")
plt.title("Distribution of Extracted Drug-Release Observations")
plt.tight_layout()

plt.savefig(
    os.path.join(
        output_dir,
        "21_figure_1_release_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()

# ==============================================================================
# FIGURE 2 — RELEASE VS TIME
# ==============================================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    analysis_df["release_time_h"],
    analysis_df["release_percent"],
    alpha=0.75
)

plt.xscale("symlog", linthresh=0.01)

plt.xlabel("Release time (h)")
plt.ylabel("Drug release (%)")
plt.title("Extracted Drug Release Across Time")

plt.tight_layout()

plt.savefig(
    os.path.join(
        output_dir,
        "21_figure_2_release_vs_time.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()

# ==============================================================================
# FIGURE 3 — WITHIN-STUDY RELEASE CURVES
# ==============================================================================

multi_studies = study_table[
    study_table["n_timepoints"] >= 3
]["study_id"].tolist()

plt.figure(figsize=(9, 6))

for study_id in multi_studies:

    g = analysis_df[
        analysis_df["study_id"] == study_id
    ].sort_values("release_time_h")

    plt.plot(
        g["release_time_h"],
        g["release_percent"],
        marker="o",
        linewidth=1.5,
        label=study_id
    )

plt.xlabel("Release time (h)")
plt.ylabel("Drug release (%)")
plt.title("Within-Study Drug-Release Profiles")

plt.legend(
    title="Study",
    fontsize=8
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        output_dir,
        "21_figure_3_within_study_release_curves.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()

# ==============================================================================
# FIGURE 4 — STUDY CONTRIBUTION
# ==============================================================================

plt.figure(figsize=(9, 5))

plt.bar(
    contribution["study_id"],
    contribution["observation_count"]
)

plt.xlabel("Study")
plt.ylabel("Number of observations")
plt.title("Contribution of Individual Studies to the Analytical Dataset")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        output_dir,
        "21_figure_4_study_contribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()

# ==============================================================================
# FIGURE 5 — RELEASE BY TIME WINDOW
# ==============================================================================

plt.figure(figsize=(9, 5))

plot_df = analysis_df.copy()

plot_df["time_window"] = pd.Categorical(
    plot_df["time_window"],
    categories=window_order,
    ordered=True
)

plot_df.boxplot(
    column="release_percent",
    by="time_window",
    figsize=(9, 5)
)

plt.suptitle("")
plt.title("Drug Release Across Prespecified Time Windows")
plt.xlabel("Time window")
plt.ylabel("Drug release (%)")

plt.xticks(rotation=20)

plt.tight_layout()

plt.savefig(
    os.path.join(
        output_dir,
        "21_figure_5_release_by_time_window.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()

# ==============================================================================
# FINAL INTERPRETATION MATRIX
# ==============================================================================

interpretation = pd.DataFrame([
    {
        "analysis": "Study-level descriptive synthesis",
        "decision": "RECOMMENDED",
        "interpretation":
            "Appropriate because individual studies represent the independent evidence units."
    },
    {
        "analysis": "Within-study release curves",
        "decision": "RECOMMENDED",
        "interpretation":
            "Appropriate for the three studies containing sufficient longitudinal observations."
    },
    {
        "analysis": "Kinetic modeling",
        "decision": "RESTRICTED",
        "interpretation":
            "Restricted to studies with at least three usable quantitative timepoints."
    },
    {
        "analysis": "Observation-level pooled regression",
        "decision": "NOT RECOMMENDED",
        "interpretation":
            "Repeated observations within studies create pseudoreplication."
    },
    {
        "analysis": "Formal meta-analysis",
        "decision": "NOT RECOMMENDED",
        "interpretation":
            "Release percentages occur at heterogeneous timepoints and formulations."
    },
    {
        "analysis": "Polymer-group inferential comparison",
        "decision": "CAUTION",
        "interpretation":
            "Polymer composition is confounded with drug and formulation characteristics."
    },
    {
        "analysis": "Drug-group inferential comparison",
        "decision": "CAUTION",
        "interpretation":
            "Drug physicochemical properties and formulation conditions differ between studies."
    },
    {
        "analysis": "Machine learning",
        "decision": "NOT RECOMMENDED",
        "interpretation":
            "The independent-study sample is too small and heterogeneous for reliable predictive modeling."
    },
    {
        "analysis": "Narrative quantitative synthesis",
        "decision": "RECOMMENDED",
        "interpretation":
            "Supports structured quantitative comparison without unjustified statistical pooling."
    }
])

interpretation.to_csv(
    os.path.join(
        output_dir,
        "21_publication_analysis_interpretation_matrix.csv"
    ),
    index=False
)

# ==============================================================================
# PUBLICATION-SAFE RESULTS SUMMARY
# ==============================================================================

results_summary = pd.DataFrame([{
    "independent_studies": analysis_df["study_id"].nunique(),
    "analytical_observations": len(analysis_df),
    "kinetic_eligible_studies": len(multi_studies),
    "repeated_measurement_studies": int(
        study_table["multi_timepoint"].sum()
    ),
    "mean_release_percent": round(
        analysis_df["release_percent"].mean(), 2
    ),
    "median_release_percent": round(
        analysis_df["release_percent"].median(), 2
    ),
    "sd_release_percent": round(
        analysis_df["release_percent"].std(), 2
    ),
    "minimum_release_percent": round(
        analysis_df["release_percent"].min(), 2
    ),
    "maximum_release_percent": round(
        analysis_df["release_percent"].max(), 2
    ),
    "formal_meta_analysis": "NOT RECOMMENDED",
    "observation_level_pooling": "NOT RECOMMENDED",
    "machine_learning": "NOT RECOMMENDED",
    "narrative_quantitative_synthesis": "RECOMMENDED",
    "within_study_kinetics": "RECOMMENDED_WHERE_ELIGIBLE"
}])

results_summary.to_csv(
    os.path.join(
        output_dir,
        "21_publication_safe_results_summary.csv"
    ),
    index=False
)

# ==============================================================================
# MANUSCRIPT-READY TEXT
# ==============================================================================

manuscript_text = f"""
RESULTS — QUANTITATIVE EVIDENCE SYNTHESIS

The final analytical dataset comprised {len(analysis_df)} quantitative
observations from {analysis_df['study_id'].nunique()} independent studies.
Three studies contained at least three quantitative release timepoints and
were therefore eligible for within-study kinetic analysis, whereas the
remaining studies provided single-timepoint or otherwise insufficient
longitudinal evidence.

Across the analytical observations, mean drug release was
{analysis_df['release_percent'].mean():.2f}% (SD
{analysis_df['release_percent'].std():.2f}%), with a median of
{analysis_df['release_percent'].median():.2f}% and a range of
{analysis_df['release_percent'].min():.2f}% to
{analysis_df['release_percent'].max():.2f}%.

The evidence demonstrated substantial between-study heterogeneity in polymer
composition, drug identity, formulation characteristics, and sampling
timepoints. Repeated measurements were concentrated within three studies,
which together contributed {contribution.head(3)['observation_count'].sum()}
of {len(analysis_df)} observations
({100 * contribution.head(3)['observation_count'].sum() / len(analysis_df):.1f}%).
Accordingly, observations were not treated as statistically independent
units for pooled inferential analysis.

Within-study kinetic analysis was therefore restricted to studies with
sufficient longitudinal measurements. Formal meta-analysis and pooled
observation-level regression were not performed because the extracted
release outcomes were measured at heterogeneous timepoints and under
substantially different formulation conditions. Polymer- and
drug-level comparisons were interpreted descriptively rather than as
causal or inferential comparisons.

The resulting evidence supports a structured narrative quantitative
synthesis, with kinetic interpretation restricted to studies providing
adequate longitudinal release data.
"""

with open(
    os.path.join(
        output_dir,
        "21_manuscript_ready_results.txt"
    ),
    "w",
    encoding="utf-8"
) as f:
    f.write(manuscript_text.strip())

# ==============================================================================
# FINAL FILE INVENTORY
# ==============================================================================

print("\n" + "=" * 78)
print("CODE 21 COMPLETE")
print("=" * 78)

print("\nPublication package:", output_dir)

for filename in sorted(os.listdir(output_dir)):
    print("-", filename)

print("\nKey results:")
print("Analytical observations:", len(analysis_df))
print("Independent studies:", analysis_df["study_id"].nunique())
print("Kinetic-eligible studies:", len(multi_studies))
print(
    "Mean release:",
    round(analysis_df["release_percent"].mean(), 2),
    "%"
)
print(
    "Median release:",
    round(analysis_df["release_percent"].median(), 2),
    "%"
)

print("\nPublication analysis:")
print("Formal meta-analysis: NOT RECOMMENDED")
print("Observation-level pooling: NOT RECOMMENDED")
print("Machine learning: NOT RECOMMENDED")
print("Narrative quantitative synthesis: RECOMMENDED")
print("Within-study kinetic analysis: RECOMMENDED WHERE ELIGIBLE")

print("\n" + "=" * 78)
```

```python
# ==============================================================================
# CODE 22 — PROVENANCE, ATTRIBUTION & CLAIM-SAFETY AUDIT
# ==============================================================================

import os
import pandas as pd
import numpy as np

print("=" * 78)
print("CODE 22 — PROVENANCE, ATTRIBUTION & CLAIM-SAFETY AUDIT")
print("=" * 78)

# ------------------------------------------------------------------------------
# PATHS
# ------------------------------------------------------------------------------

base_dir = "AI_Polymer_Drug_Release"

final_file = os.path.join(
    base_dir,
    "data",
    "final",
    "12_final_analysis_dataset.csv"
)

metadata_file = os.path.join(
    base_dir,
    "data",
    "extraction",
    "11_formulation_verification",
    "11_formulation_verification_by_study.csv"
)

mapped_file = os.path.join(
    base_dir,
    "data",
    "extraction",
    "10_metadata_recovery",
    "10_mapped_quantitative_observations.csv"
)

output_dir = os.path.join(
    base_dir,
    "data",
    "analysis",
    "22_provenance_claim_safety"
)

os.makedirs(output_dir, exist_ok=True)

# ------------------------------------------------------------------------------
# LOAD FINAL DATA
# ------------------------------------------------------------------------------

if not os.path.exists(final_file):
    raise FileNotFoundError(
        f"Final dataset not found: {final_file}"
    )

df = pd.read_csv(final_file)

print("\nInput file:", final_file)
print("Input rows:", len(df))
print("Input columns:", len(df))

# ------------------------------------------------------------------------------
# FINAL ANALYTICAL OBSERVATIONS
# ------------------------------------------------------------------------------

df["final_analysis_inclusion"] = (
    df["final_analysis_inclusion"]
    .astype(str)
    .str.lower()
    .isin(["true", "1", "yes"])
)

analysis_df = df[
    df["final_analysis_inclusion"]
].copy()

print("\nFinal analytical observations:", len(analysis_df))
print(
    "Independent studies:",
    analysis_df["study_id"].nunique()
)

# ------------------------------------------------------------------------------
# NORMALIZE TEXT
# ------------------------------------------------------------------------------

for col in ["polymer", "drug", "title", "pmid", "doi"]:
    if col not in analysis_df.columns:
        analysis_df[col] = ""

    analysis_df[col] = (
        analysis_df[col]
        .fillna("")
        .astype(str)
        .str.strip()
    )

# ------------------------------------------------------------------------------
# COUNT COMPONENTS
# ------------------------------------------------------------------------------

def count_components(value):

    if not value or value.lower() == "nan":
        return 0

    return len([
        x.strip()
        for x in value.split(";")
        if x.strip()
    ])


analysis_df["polymer_component_count"] = (
    analysis_df["polymer"]
    .apply(count_components)
)

analysis_df["drug_component_count"] = (
    analysis_df["drug"]
    .apply(count_components)
)

analysis_df["polymer_unique_attribution"] = (
    analysis_df["polymer_component_count"] == 1
)

analysis_df["drug_unique_attribution"] = (
    analysis_df["drug_component_count"] == 1
)

# ------------------------------------------------------------------------------
# SOURCE IDENTIFICATION
# ------------------------------------------------------------------------------

analysis_df["source_identifier_available"] = (
    (analysis_df["pmid"] != "") |
    (analysis_df["doi"] != "")
)

analysis_df["title_available"] = (
    analysis_df["title"] != ""
)

analysis_df["time_available"] = (
    pd.to_numeric(
        analysis_df["release_time_h"],
        errors="coerce"
    ).notna()
)

analysis_df["release_value_available"] = (
    pd.to_numeric(
        analysis_df["release_percent"],
        errors="coerce"
    ).notna()
)

# ------------------------------------------------------------------------------
# ATTRIBUTION CLASSIFICATION
# ------------------------------------------------------------------------------

def attribution_class(row):

    polymer_unique = row["polymer_unique_attribution"]
    drug_unique = row["drug_unique_attribution"]

    if polymer_unique and drug_unique:
        return "UNIQUE_POLYMER_AND_DRUG"

    if polymer_unique and not drug_unique:
        return "UNIQUE_POLYMER_MULTI_DRUG"

    if not polymer_unique and drug_unique:
        return "MULTI_POLYMER_UNIQUE_DRUG"

    return "MULTI_POLYMER_MULTI_DRUG"


analysis_df["attribution_class"] = (
    analysis_df.apply(attribution_class, axis=1)
)

# ------------------------------------------------------------------------------
# CLAIM-SAFETY CLASSIFICATION
# ------------------------------------------------------------------------------

def claim_safety(row):

    if not row["source_identifier_available"]:
        return "UNSAFE_SOURCE_NOT_IDENTIFIED"

    if not row["time_available"]:
        return "UNSAFE_TIME_MISSING"

    if not row["release_value_available"]:
        return "UNSAFE_RELEASE_MISSING"

    if (
        row["polymer_unique_attribution"]
        and row["drug_unique_attribution"]
    ):
        return "HIGHER_CLAIM_SAFETY"

    if (
        row["polymer_unique_attribution"]
        or row["drug_unique_attribution"]
    ):
        return "MODERATE_CLAIM_SAFETY"

    return "DESCRIPTIVE_STUDY_LEVEL_ONLY"


analysis_df["claim_safety"] = (
    analysis_df.apply(claim_safety, axis=1)
)

# ------------------------------------------------------------------------------
# RECOMMENDED CLAIM TYPE
# ------------------------------------------------------------------------------

def recommended_claim(row):

    safety = row["claim_safety"]

    if safety == "HIGHER_CLAIM_SAFETY":
        return (
            "Observation may be described with both polymer and drug "
            "attribution, subject to source verification."
        )

    if safety == "MODERATE_CLAIM_SAFETY":
        return (
            "Use only the uniquely attributable component; "
            "avoid assigning the outcome to the non-unique component."
        )

    if safety == "DESCRIPTIVE_STUDY_LEVEL_ONLY":
        return (
            "Describe the observation at study/formulation level; "
            "do not attribute release independently to individual polymers "
            "or drugs."
        )

    return "Do not use in quantitative manuscript claims."

analysis_df["recommended_claim"] = (
    analysis_df.apply(recommended_claim, axis=1)
)

# ------------------------------------------------------------------------------
# PROVENANCE AUDIT TABLE
# ------------------------------------------------------------------------------

provenance_columns = [
    "study_id",
    "pmid",
    "doi",
    "title",
    "polymer",
    "drug",
    "release_time_h",
    "release_percent",
    "polymer_component_count",
    "drug_component_count",
    "source_identifier_available",
    "time_available",
    "release_value_available",
    "attribution_class",
    "claim_safety",
    "recommended_claim"
]

provenance_table = analysis_df[
    provenance_columns
].copy()

provenance_table.to_csv(
    os.path.join(
        output_dir,
        "22_observation_provenance_claim_safety.csv"
    ),
    index=False
)

print("\nPROVENANCE TABLE CREATED")

# ------------------------------------------------------------------------------
# STUDY-LEVEL CLAIM SAFETY
# ------------------------------------------------------------------------------

study_claim_rows = []

for study_id, g in analysis_df.groupby("study_id"):

    polymer_values = set(
        g["polymer"].dropna().astype(str)
    )

    drug_values = set(
        g["drug"].dropna().astype(str)
    )

    study_claim_rows.append({

        "study_id": study_id,

        "n_observations": len(g),

        "n_polymer_descriptors": len(polymer_values),

        "n_drug_descriptors": len(drug_values),

        "all_polymer_descriptors_unique":
            bool(g["polymer_unique_attribution"].all()),

        "all_drug_descriptors_unique":
            bool(g["drug_unique_attribution"].all()),

        "has_multi_polymer_attribution":
            bool((g["polymer_component_count"] > 1).any()),

        "has_multi_drug_attribution":
            bool((g["drug_component_count"] > 1).any()),

        "source_identifier_complete":
            bool(g["source_identifier_available"].all()),

        "time_complete":
            bool(g["time_available"].all()),

        "release_complete":
            bool(g["release_value_available"].all()),

        "recommended_evidence_level":
            (
                "STUDY_LEVEL_DESCRIPTIVE"
                if (
                    (g["polymer_component_count"] > 1).any()
                    or
                    (g["drug_component_count"] > 1).any()
                )
                else
                "COMPONENT_LEVEL_DESCRIPTIVE"
            )
    })

study_claim_table = pd.DataFrame(
    study_claim_rows
)

study_claim_table.to_csv(
    os.path.join(
        output_dir,
        "22_study_level_claim_safety.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# MULTI-COMPONENT STUDY AUDIT
# ------------------------------------------------------------------------------

multi_component = analysis_df[
    (analysis_df["polymer_component_count"] > 1)
    |
    (analysis_df["drug_component_count"] > 1)
].copy()

multi_component[
    provenance_columns
].to_csv(
    os.path.join(
        output_dir,
        "22_multi_component_attribution_audit.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# UNIQUE-ATTRIBUTION OBSERVATIONS
# ------------------------------------------------------------------------------

unique_attribution = analysis_df[
    analysis_df["claim_safety"] == "HIGHER_CLAIM_SAFETY"
].copy()

unique_attribution[
    provenance_columns
].to_csv(
    os.path.join(
        output_dir,
        "22_higher_claim_safety_observations.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# CLAIMS THAT SHOULD NOT BE MADE
# ------------------------------------------------------------------------------

unsafe_claims = pd.DataFrame([
    {
        "claim_type":
            "Polymer superiority",
        "status":
            "DO_NOT_CLAIM",
        "reason":
            "Polymer composition is confounded with drug identity, "
            "formulation, and experimental conditions."
    },
    {
        "claim_type":
            "Drug superiority",
        "status":
            "DO_NOT_CLAIM",
        "reason":
            "Drug identity and formulation conditions are heterogeneous."
    },
    {
        "claim_type":
            "Pooled release estimate",
        "status":
            "DO_NOT_CLAIM",
        "reason":
            "Observations occur at heterogeneous timepoints and "
            "include repeated measurements."
    },
    {
        "claim_type":
            "Meta-analytic effect size",
        "status":
            "DO_NOT_CLAIM",
        "reason":
            "The dataset does not provide a common effect measure."
    },
    {
        "claim_type":
            "Predictive machine-learning performance",
        "status":
            "DO_NOT_CLAIM",
        "reason":
            "Only 11 independent studies are available."
    },
    {
        "claim_type":
            "Causal effect of polymer on release",
        "status":
            "DO_NOT_CLAIM",
        "reason":
            "Polymer identity is confounded by multiple study-level factors."
    },
    {
        "claim_type":
            "Causal effect of drug on release",
        "status":
            "DO_NOT_CLAIM",
        "reason":
            "Drug identity is confounded by formulation and experimental "
            "conditions."
    },
    {
        "claim_type":
            "25 independent experiments",
        "status":
            "DO_NOT_CLAIM",
        "reason":
            "There are 25 observations but only 11 independent studies."
    }
])

unsafe_claims.to_csv(
    os.path.join(
        output_dir,
        "22_claims_not_supported_by_dataset.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# SAFE CLAIMS
# ------------------------------------------------------------------------------

safe_claims = pd.DataFrame([
    {
        "claim_type":
            "Overall descriptive release",
        "status":
            "SUPPORTED",
        "recommended_wording":
            "Across the extracted analytical observations, release "
            "percentages ranged from 1.493% to 100%."
    },
    {
        "claim_type":
            "Study-level heterogeneity",
        "status":
            "SUPPORTED",
        "recommended_wording":
            "Substantial heterogeneity was observed in polymer composition, "
            "drug identity, formulation, and sampling time."
    },
    {
        "claim_type":
            "Within-study kinetics",
        "status":
            "SUPPORTED_WITH_RESTRICTION",
        "recommended_wording":
            "Three studies provided sufficient longitudinal observations "
            "for restricted within-study kinetic analysis."
    },
    {
        "claim_type":
            "Repeated measurements",
        "status":
            "SUPPORTED",
        "recommended_wording":
            "Repeated observations were retained within studies but were "
            "not treated as independent evidence units."
    },
    {
        "claim_type":
            "Narrative quantitative synthesis",
        "status":
            "SUPPORTED",
        "recommended_wording":
            "The available evidence supports structured narrative "
            "quantitative synthesis rather than formal pooled estimation."
    },
    {
        "claim_type":
            "Need for additional evidence",
        "status":
            "SUPPORTED",
        "recommended_wording":
            "Additional standardized studies are required to enable "
            "robust cross-study polymer or drug comparisons."
    }
])

safe_claims.to_csv(
    os.path.join(
        output_dir,
        "22_publication_safe_claims.csv"
    ),
    index=False
)

# ==============================================================================
# OVERALL AUDIT SUMMARY
# ==============================================================================

summary = pd.DataFrame([{

    "analytical_observations":
        len(analysis_df),

    "independent_studies":
        analysis_df["study_id"].nunique(),

    "observations_with_source_identifier":
        int(analysis_df["source_identifier_available"].sum()),

    "observations_with_valid_time":
        int(analysis_df["time_available"].sum()),

    "observations_with_valid_release":
        int(analysis_df["release_value_available"].sum()),

    "unique_polymer_attribution_observations":
        int(analysis_df["polymer_unique_attribution"].sum()),

    "unique_drug_attribution_observations":
        int(analysis_df["drug_unique_attribution"].sum()),

    "multi_polymer_observations":
        int(
            (analysis_df["polymer_component_count"] > 1).sum()
        ),

    "multi_drug_observations":
        int(
            (analysis_df["drug_component_count"] > 1).sum()
        ),

    "higher_claim_safety_observations":
        int(
            (
                analysis_df["claim_safety"]
                == "HIGHER_CLAIM_SAFETY"
            ).sum()
        ),

    "moderate_claim_safety_observations":
        int(
            (
                analysis_df["claim_safety"]
                == "MODERATE_CLAIM_SAFETY"
            ).sum()
        ),

    "study_level_only_observations":
        int(
            (
                analysis_df["claim_safety"]
                == "DESCRIPTIVE_STUDY_LEVEL_ONLY"
            ).sum()
        ),

    "unsafe_observations":
        int(
            analysis_df["claim_safety"]
            .str.startswith("UNSAFE")
            .sum()
        ),

    "overall_claim_safety":
        (
            "PASS_WITH_RESTRICTED_COMPONENT_ATTRIBUTION"
            if (
                analysis_df["claim_safety"]
                .str.startswith("UNSAFE")
                .sum() == 0
            )
            else
            "REVIEW_REQUIRED"
        )
}])

summary.to_csv(
    os.path.join(
        output_dir,
        "22_final_provenance_claim_safety_summary.csv"
    ),
    index=False
)

# ==============================================================================
# MANUSCRIPT-SAFE INTERPRETATION
# ==============================================================================

manuscript_text = f"""
PROVENANCE AND CLAIM-SAFETY INTERPRETATION

The final analytical dataset contained {len(analysis_df)}
quantitative observations from {analysis_df['study_id'].nunique()}
independent studies. All included observations had identifiable release
values and quantitative release times.

The extracted evidence was heterogeneous with respect to polymer
composition and drug identity. Consequently, polymer and drug descriptors
were retained primarily as study-level formulation metadata rather than
being interpreted as independently randomized explanatory variables.

Repeated observations from the same study were retained for within-study
release-curve and kinetic analyses but were not treated as independent
observational units in pooled statistical analyses.

The evidence therefore supports descriptive characterization of release
behavior and restricted within-study kinetic interpretation. It does not
support causal claims that a specific polymer or drug independently caused
higher or lower release across studies.

Cross-study comparisons of polymer or drug performance should be
interpreted cautiously because polymer identity, drug identity,
formulation composition, experimental conditions, and sampling schedules
are confounded.

Accordingly, the primary quantitative conclusions should be framed at the
study/formulation level, while component-specific claims should only be
made when the source study uniquely and explicitly attributes the reported
release measurement to that polymer-drug combination.
"""

with open(
    os.path.join(
        output_dir,
        "22_manuscript_provenance_interpretation.txt"
    ),
    "w",
    encoding="utf-8"
) as f:
    f.write(manuscript_text.strip())

# ==============================================================================
# FINAL OUTPUT
# ==============================================================================

print("\n" + "=" * 78)
print("PROVENANCE AUDIT SUMMARY")
print("=" * 78)

print(summary.T)

print("\nAttribution distribution:")
print(
    analysis_df["attribution_class"]
    .value_counts()
)

print("\nClaim-safety distribution:")
print(
    analysis_df["claim_safety"]
    .value_counts()
)

print("\nFiles created:")

for filename in sorted(os.listdir(output_dir)):
    print("-", filename)

print("\n" + "=" * 78)
print("CODE 22 COMPLETE")
print("=" * 78)
```

```python
# ==============================================================================
# CODE 23 — FINAL MANUSCRIPT RESULTS, DISCUSSION & EVIDENCE SYNTHESIS
# ==============================================================================

import os
import pandas as pd
import numpy as np

BASE = "AI_Polymer_Drug_Release"

INPUT_FILE = os.path.join(
    BASE, "data", "final", "12_final_analysis_dataset.csv"
)

OUT_DIR = os.path.join(
    BASE, "data", "publication", "23_final_manuscript"
)

os.makedirs(OUT_DIR, exist_ok=True)

print("=" * 78)
print("CODE 23 — FINAL MANUSCRIPT RESULTS, DISCUSSION & EVIDENCE SYNTHESIS")
print("=" * 78)

# ------------------------------------------------------------------------------
# 1. LOAD FINAL DATASET
# ------------------------------------------------------------------------------

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(f"Final dataset not found: {INPUT_FILE}")

df = pd.read_csv(INPUT_FILE)

print(f"Input rows: {len(df)}")
print(f"Input columns: {len(df.columns)}")

# ------------------------------------------------------------------------------
# 2. FINAL ANALYTICAL DATASET
# ------------------------------------------------------------------------------

if "final_analysis_inclusion" not in df.columns:
    raise ValueError("final_analysis_inclusion column is required.")

analysis = df[
    df["final_analysis_inclusion"].astype(str).str.lower().isin(
        ["true", "1", "yes"]
    )
].copy()

analysis["release_time_h"] = pd.to_numeric(
    analysis["release_time_h"], errors="coerce"
)

analysis["release_percent"] = pd.to_numeric(
    analysis["release_percent"], errors="coerce"
)

analysis = analysis[
    analysis["release_time_h"].notna()
    & analysis["release_percent"].notna()
].copy()

n_obs = len(analysis)
n_studies = analysis["study_id"].nunique()

print(f"Final analytical observations: {n_obs}")
print(f"Independent studies: {n_studies}")

# ------------------------------------------------------------------------------
# 3. STUDY-LEVEL SUMMARY
# ------------------------------------------------------------------------------

study_summary = (
    analysis.groupby("study_id")
    .agg(
        polymer=("polymer", "first"),
        drug=("drug", "first"),
        n_observations=("release_percent", "count"),
        n_timepoints=("release_time_h", "nunique"),
        first_time_h=("release_time_h", "min"),
        last_time_h=("release_time_h", "max"),
        first_release_percent=("release_percent", "first"),
        last_release_percent=("release_percent", "last"),
        mean_release_percent=("release_percent", "mean"),
        median_release_percent=("release_percent", "median"),
        maximum_release_percent=("release_percent", "max"),
        minimum_release_percent=("release_percent", "min"),
    )
    .reset_index()
)

study_summary["multi_timepoint"] = (
    study_summary["n_timepoints"] >= 2
)

study_summary["kinetic_eligible"] = (
    study_summary["n_timepoints"] >= 3
)

study_summary["release_change_percent"] = (
    study_summary["last_release_percent"]
    - study_summary["first_release_percent"]
)

study_summary.to_csv(
    os.path.join(
        OUT_DIR,
        "23_final_study_level_summary.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# 4. OVERALL DESCRIPTIVE STATISTICS
# ------------------------------------------------------------------------------

release = analysis["release_percent"]

overall = {
    "analytical_observations": n_obs,
    "independent_studies": n_studies,
    "multi_timepoint_studies": int(
        study_summary["multi_timepoint"].sum()
    ),
    "kinetic_eligible_studies": int(
        study_summary["kinetic_eligible"].sum()
    ),
    "mean_release_percent": release.mean(),
    "sd_release_percent": release.std(),
    "median_release_percent": release.median(),
    "minimum_release_percent": release.min(),
    "maximum_release_percent": release.max(),
    "mean_release_time_h": analysis["release_time_h"].mean(),
    "median_release_time_h": analysis["release_time_h"].median(),
    "maximum_release_time_h": analysis["release_time_h"].max(),
}

pd.DataFrame([overall]).to_csv(
    os.path.join(
        OUT_DIR,
        "23_final_overall_statistics.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# 5. TIME-WINDOW SYNTHESIS
# ------------------------------------------------------------------------------

def classify_time(t):
    if t <= 1:
        return "Early (0–1 h)"
    elif t <= 6:
        return "Early-intermediate (1–6 h)"
    elif t <= 24:
        return "Intermediate (6–24 h)"
    elif t <= 72:
        return "Late (24–72 h)"
    else:
        return "Very late (>72 h)"

analysis["time_window"] = analysis["release_time_h"].apply(
    classify_time
)

window_order = [
    "Early (0–1 h)",
    "Early-intermediate (1–6 h)",
    "Intermediate (6–24 h)",
    "Late (24–72 h)",
    "Very late (>72 h)",
]

time_summary = (
    analysis.groupby("time_window", observed=False)
    .agg(
        observation_count=("release_percent", "count"),
        study_count=("study_id", "nunique"),
        mean_release_percent=("release_percent", "mean"),
        median_release_percent=("release_percent", "median"),
        sd_release_percent=("release_percent", "std"),
        minimum_release_percent=("release_percent", "min"),
        maximum_release_percent=("release_percent", "max"),
    )
    .reindex(window_order)
    .reset_index()
)

time_summary.to_csv(
    os.path.join(
        OUT_DIR,
        "23_final_time_window_summary.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# 6. KINETIC ELIGIBILITY
# ------------------------------------------------------------------------------

kinetic = study_summary[
    study_summary["n_timepoints"] >= 3
].copy()

kinetic["eligibility_basis"] = (
    "At least three quantitative timepoints available within the study"
)

kinetic.to_csv(
    os.path.join(
        OUT_DIR,
        "23_final_kinetic_eligible_studies.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# 7. STUDY CONTRIBUTION
# ------------------------------------------------------------------------------

contribution = (
    analysis.groupby("study_id")
    .size()
    .reset_index(name="observation_count")
    .sort_values("observation_count", ascending=False)
)

contribution["proportion_of_observations"] = (
    contribution["observation_count"] / n_obs
)

contribution["cumulative_observations"] = (
    contribution["observation_count"].cumsum()
)

contribution["cumulative_proportion"] = (
    contribution["cumulative_observations"] / n_obs
)

contribution.to_csv(
    os.path.join(
        OUT_DIR,
        "23_final_study_contribution.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# 8. CLAIM-SAFE INTERPRETATION MATRIX
# ------------------------------------------------------------------------------

interpretation = pd.DataFrame([
    {
        "analysis": "Study-level descriptive synthesis",
        "decision": "RECOMMENDED",
        "interpretation":
            "The study is treated as the independent evidence unit."
    },
    {
        "analysis": "Within-study release curves",
        "decision": "RECOMMENDED",
        "interpretation":
            "Repeated timepoints can be interpreted within individual studies."
    },
    {
        "analysis": "Kinetic modeling",
        "decision": "RESTRICTED",
        "interpretation":
            "Only studies with at least three usable timepoints should be modeled."
    },
    {
        "analysis": "Observation-level pooling",
        "decision": "NOT RECOMMENDED",
        "interpretation":
            "Repeated observations would create pseudoreplication."
    },
    {
        "analysis": "Formal meta-analysis",
        "decision": "NOT RECOMMENDED",
        "interpretation":
            "Release percentages occur at heterogeneous timepoints and under heterogeneous formulations."
    },
    {
        "analysis": "Polymer-specific inference",
        "decision": "RESTRICTED",
        "interpretation":
            "Polymer identity is confounded by multi-polymer formulations and experimental conditions."
    },
    {
        "analysis": "Drug-specific inference",
        "decision": "RESTRICTED",
        "interpretation":
            "Drug identity is heterogeneous and frequently occurs with multiple formulation components."
    },
    {
        "analysis": "Machine learning",
        "decision": "NOT RECOMMENDED",
        "interpretation":
            "The number of independent studies is insufficient for reliable predictive modeling."
    },
    {
        "analysis": "Narrative quantitative synthesis",
        "decision": "RECOMMENDED",
        "interpretation":
            "The dataset supports structured quantitative comparison without unjustified pooling."
    }
])

interpretation.to_csv(
    os.path.join(
        OUT_DIR,
        "23_claim_safe_interpretation_matrix.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# 9. MANUSCRIPT-READY RESULTS
# ------------------------------------------------------------------------------

results_text = f"""
RESULTS

Study and evidence characteristics

The final analytical dataset comprised {n_obs} quantitative release
observations from {n_studies} independent studies. Three studies contained
at least three quantitative timepoints and were therefore eligible for
within-study kinetic analysis, whereas the remaining studies contributed
single or insufficiently replicated timepoints.

Across all analytical observations, reported release values ranged from
{release.min():.2f}% to {release.max():.2f}%, with a mean of
{release.mean():.2f}% (SD {release.std():.2f}) and a median of
{release.median():.2f}%. Sampling times ranged from
{analysis["release_time_h"].min():.3f} to
{analysis["release_time_h"].max():.1f} h.

Study-level evidence was highly unbalanced. The three multi-timepoint
studies contributed {contribution.head(3)["observation_count"].sum()}
of the {n_obs} analytical observations
({100 * contribution.head(3)["observation_count"].sum() / n_obs:.1f}%).
Consequently, observation-level pooling would disproportionately weight a
small number of studies.

Time-dependent release patterns

Release values varied substantially across sampling windows. The observed
release distribution should therefore be interpreted descriptively rather
than as evidence of a common pooled release trajectory. Differences between
time windows may reflect differences in formulation, drug properties,
polymer composition, experimental conditions, and study-specific sampling
designs.

Within-study kinetic analysis

Three studies met the predefined minimum criterion of at least three
quantitative timepoints. Kinetic models were therefore evaluated separately
within these studies. The resulting models were not pooled across studies
because the underlying formulations and experimental conditions were not
sufficiently comparable.

The kinetic analysis demonstrated study-specific release behavior rather than
a single generalizable kinetic mechanism. Accordingly, kinetic model
selection should be interpreted as an exploratory within-study assessment
rather than evidence for a universal release mechanism.

Evidence heterogeneity and attribution

The included studies used heterogeneous polymer combinations and drug
formulations. Because polymer components were frequently present in
combination, the dataset does not permit reliable attribution of release
behavior to an individual polymer. Drug-specific comparisons are likewise
restricted by differences in formulation and experimental conditions.

Robustness and analytical limitations

Sensitivity analyses demonstrated that descriptive estimates were influenced
by study composition and influential observations. The evidence base is
therefore better suited to structured narrative quantitative synthesis than
to formal statistical pooling.

No formal meta-analysis or observation-level pooled regression was performed.
Machine-learning prediction was also not pursued because the number of
independent studies was small relative to the heterogeneity of formulations
and outcomes.

Overall interpretation

The available evidence demonstrates substantial variation in reported drug
release across polymer-based delivery systems. However, the current evidence
base is insufficient to establish a universal release pattern, identify a
superior polymer, or attribute release kinetics to individual polymer
components. The strongest defensible interpretation is therefore a
study-level quantitative synthesis complemented by within-study kinetic
analysis where sufficient time-series data were available.
"""

with open(
    os.path.join(
        OUT_DIR,
        "23_manuscript_ready_results.txt"
    ),
    "w",
    encoding="utf-8"
) as f:
    f.write(results_text.strip())

# ------------------------------------------------------------------------------
# 10. MANUSCRIPT-READY METHODS STATEMENT
# ------------------------------------------------------------------------------

methods_text = """
ANALYTICAL METHODS STATEMENT

Quantitative release observations were treated as nested within independent
studies, with the study considered the primary independent evidence unit.
Repeated measurements from the same study were not treated as independent
observations for inferential pooling.

Descriptive statistics were calculated for the complete analytical dataset,
while study-level summaries were used to characterize the distribution of
evidence across studies. Release observations were additionally summarized
within predefined sampling-time windows.

Within-study kinetic modeling was restricted to studies containing at least
three usable quantitative release timepoints. Kinetic models were fitted
independently within eligible studies and were not pooled across studies.

Because the included studies differed substantially in polymer composition,
drug identity, formulation characteristics, sampling schedules, and
experimental conditions, formal meta-analysis and observation-level pooled
regression were not considered appropriate. Polymer- and drug-specific
comparisons were interpreted cautiously because component attribution was
confounded by multi-component formulations.

The final synthesis therefore used a structured quantitative narrative
approach, emphasizing study-level evidence, within-study release trajectories,
and explicit characterization of uncertainty and heterogeneity.
"""

with open(
    os.path.join(
        OUT_DIR,
        "23_manuscript_ready_methods.txt"
    ),
    "w",
    encoding="utf-8"
) as f:
    f.write(methods_text.strip())

# ------------------------------------------------------------------------------
# 11. MANUSCRIPT-READY DISCUSSION
# ------------------------------------------------------------------------------

discussion_text = """
DISCUSSION

This synthesis indicates that polymer-based drug-release studies report a
wide range of release outcomes, but the available evidence does not support
a single generalized release profile. The principal reason is substantial
between-study heterogeneity in polymer composition, drug identity,
formulation design, sampling schedule, and experimental conditions.

An important methodological finding is the imbalance in observation
contribution. A small number of multi-timepoint studies account for a large
proportion of the available quantitative observations. Treating every
timepoint as an independent observation would therefore result in
pseudoreplication and potentially overstate the strength of the evidence.
The present analysis avoided this problem by distinguishing study-level
evidence from repeated within-study measurements.

The kinetic analysis should similarly be interpreted conservatively. Only
three studies contained sufficient time-series information for kinetic
fitting. Their individual model fits provide useful study-specific
descriptions, but they should not be interpreted as demonstrating a universal
release mechanism across polymeric drug-delivery systems.

The heterogeneous formulation structure also limits component-level
interpretation. Many observations involve multiple polymers, and some involve
multiple drugs. Consequently, an apparent difference in release cannot
reasonably be assigned to one polymer or one drug without additional
standardized comparative experiments.

These findings highlight an important limitation in the current literature:
quantitative release data are available, but the reporting structure is often
insufficient for direct cross-study mechanistic comparison. Future studies
would benefit from standardized polymer characterization, clearly defined
drug loading, consistent release-media conditions, standardized sampling
intervals, and complete time-series reporting.

Overall, the evidence supports structured quantitative comparison and
within-study kinetic characterization, but not formal pooled effect
estimation or predictive modeling. The present synthesis should therefore be
viewed as an evidence-mapping and quantitative descriptive analysis rather
than a meta-analysis of a common release endpoint.
"""

with open(
    os.path.join(
        OUT_DIR,
        "23_manuscript_ready_discussion.txt"
    ),
    "w",
    encoding="utf-8"
) as f:
    f.write(discussion_text.strip())

# ------------------------------------------------------------------------------
# 12. FINAL CLAIM-SAFE CONCLUSIONS
# ------------------------------------------------------------------------------

conclusion_text = """
CONCLUSION

The available quantitative evidence demonstrates substantial variability in
drug-release behavior across polymer-based delivery systems. However,
heterogeneity in polymer composition, drug identity, formulation, sampling
time, and experimental conditions limits direct cross-study comparison.

The evidence base is sufficiently structured for study-level quantitative
synthesis and restricted within-study kinetic analysis, but it is not
sufficient for formal meta-analysis, observation-level pooled regression,
machine-learning prediction, or reliable ranking of individual polymers or
drugs.

Future evidence generation should prioritize standardized formulations,
complete multi-timepoint release curves, consistent experimental conditions,
and transparent reporting of polymer and drug composition to enable more
robust comparative and mechanistic synthesis.
"""

with open(
    os.path.join(
        OUT_DIR,
        "23_manuscript_ready_conclusion.txt"
    ),
    "w",
    encoding="utf-8"
) as f:
    f.write(conclusion_text.strip())

# ------------------------------------------------------------------------------
# 13. FINAL PUBLICATION-SAFE SUMMARY
# ------------------------------------------------------------------------------

publication_summary = pd.DataFrame([{
    "final_observations": n_obs,
    "independent_studies": n_studies,
    "kinetic_eligible_studies": int(
        study_summary["kinetic_eligible"].sum()
    ),
    "mean_release_percent": round(release.mean(), 3),
    "median_release_percent": round(release.median(), 3),
    "minimum_release_percent": round(release.min(), 3),
    "maximum_release_percent": round(release.max(), 3),
    "formal_meta_analysis": "NOT_RECOMMENDED",
    "observation_level_pooling": "NOT_RECOMMENDED",
    "machine_learning": "NOT_RECOMMENDED",
    "narrative_quantitative_synthesis": "RECOMMENDED",
    "within_study_kinetics": "RECOMMENDED_WHERE_ELIGIBLE",
    "polymer_component_attribution": "RESTRICTED",
    "drug_component_attribution": "RESTRICTED",
    "overall_claim_safety": "PASS_WITH_RESTRICTIONS"
}])

publication_summary.to_csv(
    os.path.join(
        OUT_DIR,
        "23_publication_safe_summary.csv"
    ),
    index=False
)

# ------------------------------------------------------------------------------
# 14. FINAL STATUS
# ------------------------------------------------------------------------------

print("\n" + "=" * 78)
print("CODE 23 COMPLETE")
print("=" * 78)

print(f"Final analytical observations: {n_obs}")
print(f"Independent studies: {n_studies}")
print(
    f"Kinetic-eligible studies: "
    f"{int(study_summary['kinetic_eligible'].sum())}"
)

print(f"Mean release: {release.mean():.2f}%")
print(f"Median release: {release.median():.2f}%")

print("\nPublication decisions:")
print("Formal meta-analysis: NOT RECOMMENDED")
print("Observation-level pooling: NOT RECOMMENDED")
print("Machine learning: NOT RECOMMENDED")
print("Narrative quantitative synthesis: RECOMMENDED")
print("Within-study kinetic analysis: RECOMMENDED WHERE ELIGIBLE")

print("\nClaim safety:")
print("PASS WITH RESTRICTED COMPONENT ATTRIBUTION")

print("\nOutput directory:")
print(OUT_DIR)

print("\nFiles created:")
for f in sorted(os.listdir(OUT_DIR)):
    print("-", f)

print("=" * 78)
```
